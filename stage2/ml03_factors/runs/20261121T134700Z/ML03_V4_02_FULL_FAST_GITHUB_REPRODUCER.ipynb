```python
# 02 · ENVIRONMENT, MINIMAL COLAB DEPENDENCIES, AUDIT LOG
import sys, os, shutil, platform, subprocess, importlib.util, json, time, hashlib, re, math, itertools, warnings, zipfile, io, base64, traceback, statistics, datetime, collections, urllib.parse, ast, zlib, pprint
from pathlib import Path
WORK=Path(CFG['WORKDIR']);WORK.mkdir(parents=True,exist_ok=True)
for d in ('cache','evidence','charts','models'): (WORK/d).mkdir(exist_ok=True)
usage=shutil.disk_usage(str(WORK))
# Среда и свободное место пишутся в технический журнал, а не в вывод ноутбука.
ensure_disk=usage.free/2**30>=CFG['MIN_DISK_FREE_GB']
if not ensure_disk:raise RuntimeError('INSUFFICIENT_COLAB_DISK_CAPACITY')
NEEDED={'duckdb':'duckdb>=1.1,<2','pyarrow':'pyarrow>=17,<24','pandas':'pandas>=2.1,<3',
        'numpy':'numpy>=1.26,<3','scipy':'scipy>=1.11,<2','sklearn':'scikit-learn>=1.4,<2',
        'matplotlib':'matplotlib>=3.8,<4','openpyxl':'openpyxl>=3.1,<4',
        'nbformat':'nbformat>=5.9,<6','requests':'requests>=2.31,<3',
        'ipywidgets':'ipywidgets>=8,<9','joblib':'joblib>=1.3,<2'}
OPTIONAL={'catboost':'catboost>=1.2,<2','lightgbm':'lightgbm>=4,<5',
          'linearmodels':'linearmodels>=6,<9','shap':'shap>=0.46,<1','xgboost':'xgboost>=2.0,<4'}
missing=[v for k,v in NEEDED.items() if importlib.util.find_spec(k) is None]
if missing:
    result=subprocess.run([sys.executable,'-m','pip','-q','install',*missing],capture_output=True,text=True)
    if result.returncode:raise RuntimeError('MANDATORY_PIP_DEPENDENCY_FAILURE')
optional_missing=[v for k,v in OPTIONAL.items() if importlib.util.find_spec(k) is None]
if optional_missing:
    result=subprocess.run([sys.executable,'-m','pip','-q','install',*optional_missing],capture_output=True,text=True)
    if result.returncode:
        if CFG['MODE']=='FULL' and CFG['ADVANCED_ENABLE'] and CFG['ADVANCED_REQUIRE_ALL_FAMILIES']:
            raise RuntimeError('ADVANCED_FULL_DEPENDENCY_INSTALL_FAILED: '+result.stderr[-1000:])
        pass  # Уточнение о недоступности — в статусе экспериментов, без засорения интерфейса.
import numpy as np, pandas as pd, duckdb, requests, pyarrow as pa, pyarrow.parquet as pq
import matplotlib;matplotlib.use('Agg')
import matplotlib.pyplot as plt
from scipy import stats
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge,ElasticNet
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.metrics import mean_absolute_error,mean_squared_error
from openpyxl import Workbook,load_workbook
from openpyxl.styles import Font,PatternFill,Alignment
from openpyxl.utils import get_column_letter
import nbformat,joblib
pd.options.mode.copy_on_write=True
STATUS={};WARNINGS=[];TRACE=WORK/'ml03_technical.log';START=time.monotonic()
if TRACE.exists():TRACE.unlink()
def log(msg,severity='INFO'):
    text=str(msg)
    text=re.sub(r'(?i)(?:gh'+'p_|github'+'_pat_)[a-z0-9_]+','[REDACTED]',text)
    with open(TRACE,'a',encoding='utf-8') as f:
        f.write(datetime.datetime.now(datetime.timezone.utc).isoformat()+f' [{severity}] '+text+'\n')
def mark(name,status,reason=''):
    STATUS[name]={'status':status,'reason':str(reason)[:480]}
    log(name+': '+status+': '+str(reason)[:480])
    if status not in ('PASS','NOT_FEASIBLE','CAUSAL_NOT_IDENTIFIED'):
        WARNINGS.append({'step':name,'status':status,'reason':str(reason)[:480]})
def ensure(ok,msg):
    if not ok:raise AssertionError(msg)
def digest(path):
    h=hashlib.sha256()
    with open(path,'rb') as f:
        for b in iter(lambda:f.read(4*1024*1024),b''):h.update(b)
    return h.hexdigest()
log('Environment OK: Python='+sys.version.split()[0]+' CPU='+str(os.cpu_count())+' free_disk_gib='+str(round(usage.free/2**30,1)))
# Fail before expensive work if the required runtime is not available.
ensure(pd.__version__.split('.')[0] in ('2',), 'PANDAS_MAJOR_VERSION_NOT_TESTED') if 'ensure' in globals() else None



# ML-03 V4.02: РОВНО ОДНО визуальное окно прогресса (ipywidgets.HTML).
# Виджет отображается один раз в cell 02. Межъячеечные обновления меняют его value,
# не создавая новых FloatProgress, tqdm, display() или многократных выводов Colab.
from IPython.display import display as _single_display
try:
    import ipywidgets as _widgets
except ImportError:
    _widgets=None
_PHASE_BOUNDS={'ENV': (0, 26), 'AUTH': (26, 47), 'INPUT': (47, 187), 'BDMO': (187, 210), 'INVENTORY': (210, 238), 'SCAN': (238, 301), 'FACTORS': (301, 310), 'PANEL': (310, 357), 'TASKS': (357, 383), 'DEV': (383, 472), 'TRAIN': (472, 582), 'ADVANCED': (582, 684), 'STATS': (684, 723), 'SHAP': (723, 786), 'SENSITIVITY': (786, 820), 'EXCEL': (820, 831), 'EVIDENCE': (831, 847), 'FINAL': (847, 850), 'CAUSAL_LOAD': (850, 870), 'CAUSAL_REG': (870, 884), 'CAUSAL_PANEL': (884, 898), 'CAUSAL_EST': (898, 927), 'CAUSAL_EXPORT': (927, 940), 'ARCHIVE_RESEARCH': (940, 962), 'ARCHIVE_TECH': (962, 995), 'DELIVERY': (995, 1000)}
_PHASE_TITLES={'ENV': 'Среда', 'AUTH': 'GitHub и каталог', 'INPUT': '52 источника', 'BDMO': 'Дополнительные БДМО', 'INVENTORY': 'Инвентаризация', 'SCAN': 'Предметный отбор', 'FACTORS': 'Готовность факторов', 'PANEL': 'Панель', 'TASKS': 'План эксперимента', 'DEV': 'Подбор моделей DEV', 'TRAIN': 'Обучение и тест', 'ADVANCED': 'Ансамбли', 'STATS': 'Статистика и абляции', 'SHAP': 'Объяснения SHAP', 'SENSITIVITY': 'Утечки и устойчивость', 'EXCEL': 'Книга Excel', 'EVIDENCE': 'Evidence и контроль', 'FINAL': 'Исходные результаты', 'CAUSAL_LOAD': 'Причинный анализ: вход', 'CAUSAL_REG': 'Реестр воздействий', 'CAUSAL_PANEL': 'Дизайн причинного исследования', 'CAUSAL_EST': 'Причинные оценки', 'CAUSAL_EXPORT': 'Протокол причинного блока', 'ARCHIVE_RESEARCH': 'Архив таблиц и графиков', 'ARCHIVE_TECH': 'GitHub: тяжёлые результаты', 'DELIVERY': 'Проверка и выдача'}
class _OneProgress:
    def __init__(self):
        self.n=0;self.total=1000;self.desc='Инициализация';self.postfix='';self._last=0.;self._last_percent=-1
        self.widget=_widgets.HTML(value=self._html()) if _widgets else None
        if self.widget is not None:_single_display(self.widget)
    def _html(self):
        import html
        pct=int(round(100*self.n/self.total));name=html.escape(self.desc)
        sub=html.escape(self.postfix)
        return ('<div style="max-width:920px;padding:15px 17px;border:1px solid #d4dce5;'
          'border-radius:13px;background:#0b1c31;color:#f5f9ff;font:14px Arial,sans-serif">'
          '<div style="display:flex;justify-content:space-between;gap:14px;margin-bottom:10px">'
          '<b>ML-03 V4.02 · Объединённый запуск · '+name+'</b><b>'+str(pct)+'%</b></div>'
          '<div role="progressbar" aria-valuenow="'+str(pct)+'" aria-valuemin="0" aria-valuemax="100"'
          ' style="height:15px;border-radius:9px;background:#35445b;overflow:hidden">'
          '<div style="width:'+str(pct)+'%;height:100%;background:linear-gradient(90deg,#21a5af,#61d8b0);'
          'border-radius:9px"></div></div><div style="margin-top:8px;color:#a8c5df;font-size:12px">'
          +sub+'</div></div>')
    def update(self,increment):
        self.n=min(self.total,max(0,self.n+int(increment)));self.refresh()
    def set_description_str(self,s,refresh=False):self.desc=str(s);self.refresh(force=refresh)
    def set_postfix_str(self,s,refresh=False):self.postfix=str(s);self.refresh(force=refresh)
    def refresh(self,force=False):
        now=time.monotonic();pct=int(100*self.n/self.total)
        if not force and now-self._last<.3 and pct==self._last_percent:return
        self._last=now;self._last_percent=pct
        if self.widget is not None:self.widget.value=self._html()
        elif force or pct%10==0:log(f'PROGRESS {pct}% {self.desc} {self.postfix}')
    def close(self):self.refresh(force=True)
_PROGRESS_BAR=_OneProgress();_PROGRESS_LAST=None
PROGRESS_LEDGER=[]
def progress(phase,done=1,total=1,detail=''):
    global _PROGRESS_LAST
    if phase not in _PHASE_BOUNDS:raise KeyError('PROGRESS_UNKNOWN_PHASE_'+str(phase))
    start,end=_PHASE_BOUNDS[phase];total=max(1,int(total));done=max(0,min(int(done),total))
    goal=start+int((end-start)*done/total)
    goal=max(goal,_PROGRESS_BAR.n)  # монотонная шкала, определяется фактическим выполнением этапов
    detail=str(detail).replace('\n',' ')[:85]
    state=(_PHASE_TITLES[phase],done,total,detail)
    if state!=_PROGRESS_LAST:
        _PROGRESS_BAR.desc=state[0];_PROGRESS_BAR.postfix=f'{done}/{total} · {detail}'
        _PROGRESS_LAST=state
    if goal>_PROGRESS_BAR.n:_PROGRESS_BAR.update(goal-_PROGRESS_BAR.n)
    else:_PROGRESS_BAR.refresh()
    if done==total:PROGRESS_LEDGER.append({'phase':phase,'completed':done,'total':total,'bar_value':_PROGRESS_BAR.n})
    return _PROGRESS_BAR.n

def quiet_print(*args,**kwargs):
    # Отладочные сообщения доступны в ml03_technical.log, но не создают сотни строк в Colab.
    try:log(' '.join(str(a) for a in args))
    except Exception:pass

# Every warning goes to the technical log, not the live notebook output. Exceptions remain visible.
def _ml03_warning_to_log(message, category, filename, lineno, file=None, line=None):
    log(f'{category.__name__}: {message} [{Path(filename).name}:{lineno}]','WARN')
warnings.showwarning=_ml03_warning_to_log

progress('ENV',1,1,'Среда и зависимости подготовлены')

# Не маскируем научные ошибки: подавляется только многословный pandas PerformanceWarning.
warnings.filterwarnings('ignore', message='DataFrame is highly fragmented.*', category=pd.errors.PerformanceWarning)
```

```python
# 03 · SECURE AUTH / READ-ONLY PREFLIGHT
from getpass import getpass
# Colab secret name is TRIAD-RU-Colab; never write its VALUE to logs, ZIP or notebook.
_TOKEN_KEY=str(CFG.get('GITHUB_COLAB_SECRET','TRIAD-RU-Colab'))
TOKEN=''
try:
    from google.colab import userdata
    for _secret in (_TOKEN_KEY,'GITHUB_TOKEN'):
        try:TOKEN=str(userdata.get(_secret) or '').strip()
        except (KeyError,ValueError,Exception):TOKEN=''
        if TOKEN:break
except ImportError:pass
if not TOKEN:
    TOKEN=(os.environ.get('TRIAD_RU_COLAB') or os.environ.get('GITHUB_TOKEN') or '').strip()
if not TOKEN:TOKEN=getpass("GitHub token (скрытый ввод, если секрет Colab не найден): ").strip()
if not TOKEN: raise RuntimeError("BLOCKED_AUTH: требуется токен приватного GitHub")
API=f"https://api.github.com/repos/{CFG['REPO']}"
SESSION=requests.Session()
SESSION.headers.update({"Authorization":"Bearer "+TOKEN,"Accept":"application/vnd.github+json",
                        "X-GitHub-Api-Version":"2022-11-28","User-Agent":"Demograf-sovetnik-ML03"})
# В трассировки не передаются объекты HTTP-ошибок с заголовками/URL авторизации.
def api(method,endpoint,*,expected=(200,),**kwargs):
    url=endpoint if endpoint.startswith('https://') else API+endpoint
    r=SESSION.request(method,url,timeout=kwargs.pop('timeout',(20,90)),**kwargs)
    if r.status_code not in expected:
        code="ACCESS_DENIED" if r.status_code in (401,403) else "NOT_FOUND_OR_NO_RIGHTS" if r.status_code==404 else "HTTP_FAILURE"
        raise RuntimeError(f"{code}: GitHub HTTP {r.status_code}; resource={endpoint.split('?')[0][:130]}")
    return r
repo_obj=api('GET','').json()
ensure(repo_obj.get('private') is True,"Ожидался приватный репозиторий; настройки не меняем")
source_sha=api('GET',f"/commits/{CFG['SOURCE_REF']}").json()['sha']
SOURCE_COMMIT_SHA=source_sha  # CFG не изменяется после ячейки 01: воспроизводимый источник

def repo_bytes(path):
    obj=api('GET',f"/contents/{urllib.parse.quote(path,safe='/')}?ref={source_sha}").json()
    if obj.get('content'):
        return base64.b64decode(obj['content'])
    # Для крупных git blobs получаем Git blob, не доверяя download_url с временными параметрами.
    blob=api('GET','/git/blobs/'+obj['sha']).json()
    if blob.get('encoding') != 'base64': raise RuntimeError("UNKNOWN_GIT_BLOB_ENCODING")
    return base64.b64decode(blob['content'])
RECEIPT=json.loads(repo_bytes(CFG['HANDOFF']+'/publication_receipt.json'))
raw_index=repo_bytes(CFG['HANDOFF']+'/dataset_index.json')
receipt_index=next((x for x in RECEIPT['files'] if x['path'].endswith('/dataset_index.json')),None)
ensure(receipt_index is not None and hashlib.sha256(raw_index).hexdigest()==receipt_index['sha256'],
       'HANDOFF_INDEX_SHA_MISMATCH')
INDEX=json.loads(raw_index); ASSETS=INDEX['assets']
ensure(len(ASSETS)==52 and len({a['asset_id'] for a in ASSETS})==len(ASSETS),
       'Нарушен контракт индекса: ожидаются 52 уникальных массива')
ensure(all(a['sha256_verified'] and a['locator']['type'] in ('git','release') for a in ASSETS),
       'Индекс содержит неверифицированные/неизвестные источники')
mark('PREFLIGHT_CATALOG','PASS',f"52 актива; immutable source SHA {source_sha[:12]}")
quiet_print(f'GitHub READ: OK; handoff={INDEX["version"]}; sources={len(ASSETS)}; SHA={source_sha[:12]}')
progress('AUTH',1,1,'Каталог и commit подтверждены')
```

## 3. Научный входной контроль и состав Stage I

Хранилище Stage I содержит 52 файла, но их совокупность не образует автоматически муниципальную панель. Для каждой группы фиксируются первичный код, размерности, исходная единица и год. Источники образовательных организаций и выпускников загружаются для полного локального контроля состава (режим FULL), но **не присоединяются** к муниципальным данным без подтверждённого соответствия уровней. Проверка миграционного коэффициента независима от последующих ML-моделей.

```python
# 04 · V3.00 · VERIFIED 52 SOURCE INVENTORY, MUNICIPAL/REGIONAL/ORGANIZATION GRAIN
import tempfile
source_receipts=[]; DOWNLOAD_QA=[]; LOCAL_ASSETS={}
def safe_remote_download(asset):
    """SHA-verified cache; GitHub git/release; no untrusted URL in logs."""
    name=hashlib.sha256(asset['asset_id'].encode()).hexdigest()[:22]+'.parquet'
    dst=WORK/'cache'/name
    if dst.exists() and digest(dst)==asset['sha256']:
        if asset['asset_id'] not in {x['asset_id'] for x in source_receipts}:
            source_receipts.append({'asset_id':asset['asset_id'],'sha256':asset['sha256'],'local_bytes':dst.stat().st_size,'source_ref':source_sha,'cache_hit':True})
        LOCAL_ASSETS[asset['asset_id']]=str(dst)
        return dst
    dst.unlink(missing_ok=True); tmp=dst.with_suffix('.part');tmp.unlink(missing_ok=True)
    loc=asset['locator']; total=0
    try:
        if loc['type']=='git':
            path=urllib.parse.quote(loc['path'],safe='/')
            obj=api('GET',f'/contents/{path}?ref={source_sha}').json()
            blob=api('GET','/git/blobs/'+obj['sha'],timeout=(30,240)).json()
            ensure(blob.get('encoding')=='base64','GIT_BLOB_ENCODING_UNSUPPORTED')
            buf=base64.b64decode(blob['content'])
            with open(tmp,'wb') as f:f.write(buf)
            total=len(buf)
        else:
            u=API+'/releases/assets/'+str(int(loc['asset_id']))
            r=SESSION.get(u,headers={'Accept':'application/octet-stream'},stream=True,timeout=(30,1200))
            if r.status_code!=200:
                r.close();raise RuntimeError('RELEASE_ACCESS_HTTP_'+str(r.status_code))
            if 'json' in r.headers.get('Content-Type','').lower():
                j=r.json();r.close();download=j.get('browser_download_url','')
                host=urllib.parse.urlparse(download)
                ensure(host.scheme=='https' and host.hostname in ('github.com','api.github.com'), 'INVALID_RELEASE_HOST')
                r=SESSION.get(download,stream=True,timeout=(30,1200))
                if r.status_code!=200 or 'json' in r.headers.get('Content-Type','').lower():
                    code=r.status_code;r.close();raise RuntimeError('RELEASE_BINARY_HTTP_'+str(code))
            try:
                with open(tmp,'wb') as f:
                    for block in r.iter_content(4*1024*1024):
                        if block:
                            total+=len(block)
                            ensure(total<=int(CFG['MAX_SINGLE_ASSET_GB']*2**30),'ASSET_LIMIT_EXCEEDED')
                            f.write(block)
            finally:r.close()
        ensure(digest(tmp)==asset['sha256'],'INPUT_SHA_MISMATCH_'+asset['asset_id'])
        tmp.replace(dst);LOCAL_ASSETS[asset['asset_id']]=str(dst)
        source_receipts.append({'asset_id':asset['asset_id'],'sha256':asset['sha256'],'local_bytes':dst.stat().st_size,'source_ref':source_sha,'cache_hit':False})
        return dst
    except Exception:tmp.unlink(missing_ok=True);raise

def get_asset(aid):
    a=next((v for v in ASSETS if v['asset_id']==aid),None)
    if a is None:raise KeyError('UNKNOWN_HANDOFF_ASSET '+aid)
    return a

SOURCE_CLASSES={
 'G1':['02::labor','02::socio1','02::socio2','02::socio3','02::5ndfl','02::7ndfl'],
 'G2':['02::housing','02::construction','06::BDMO6::BDMO6_member00::full'],
 'G3':['02::education','02::social'],
 'G4':['02::budget'],
 'G5':['06::TRANSPORT','06::D04::D04_member00::spatial_reference::derived']}
def source_group(a):
    return next((g for g,pp in SOURCE_CLASSES.items() if any(a['asset_id'].startswith(p) for p in pp)),'CONTEXT_ONLY')
BASE_ASSET='01::0'
# Дополнительный обход всего реестра Stage I: длинные муниципальные наборы независимо от
# папки/источника; запрещено превращать региональные и организационные записи в ОКТМО.
SOURCE_ADAPTER_AUDIT=[]
ECONOMY_MANIFEST=json.loads(repo_bytes('manifests/02_economy_full_v1.json'))
ECONOMY_SOURCE_MANIFEST_QA=pd.DataFrame([{'source_key':k,'source':v.get('source'),
 'rows_in_manifest':v.get('rows'),'column_count':len(v.get('columns') or []),
 'has_object_level':('object_level' in (v.get('columns') or [])),
 'has_municipal_key':bool({'oktmo','object_oktmo'}&set(v.get('columns') or [])),
 'report_grain':'REQUIRES_CONTENT_VALIDATION' if 'object_level' in (v.get('columns') or []) else 'MUNICIPAL_LONG_CANDIDATE' if 'oktmo' in (v.get('columns') or []) else 'UNKNOWN'}
 for k,v in ECONOMY_MANIFEST.get('files',{}).items()])

def eligible_municipal_long_schema(a):
    cc=set(a.get('columns_manifest') or [])
    ok=('oktmo' in cc or 'object_oktmo' in cc) and {'indicator_code','indicator_name','indicator_value','year'}.issubset(cc)
    return bool(ok)

def classify_asset_for_scan(a):
    declared=source_group(a)
    if declared!='CONTEXT_ONLY':return declared
    if eligible_municipal_long_schema(a):return 'DISCOVERY_MULTIDOMAIN'
    return 'CONTEXT_ONLY'

FACTOR_ASSETS=[a for a in ASSETS if a['asset_id']!='01::0' and
               classify_asset_for_scan(a)!='CONTEXT_ONLY']
# An education organization's OKTMO is a LOCATION OF ORGANIZATION, not proof of
# municipal total or municipal population coverage. Keep a full exclusion receipt.
ORGANIZATION_GRAIN_PREFIXES=('04::E02::','05::E04::','05::E05::')
ORGANIZATION_GRAIN_EXCLUSIONS=[{'asset_id':a['asset_id'],
    'reason':'ORGANIZATION_LEVEL_NOT_MUNICIPAL_AGGREGATE; NO_VALIDATED_ORGANIZATION_TO_MUNICIPAL_CONCORDANCE',
    'status':'EXPLICITLY_EXCLUDED_FROM_MUNICIPAL_MODEL'} for a in ASSETS
    if a['asset_id'].startswith(ORGANIZATION_GRAIN_PREFIXES)]
UNADAPTED_MUNICIPAL_WIDE=[a['asset_id'] for a in ASSETS
 if {'oktmo','year'}.issubset(set(a.get('columns_manifest') or [])) and
 not eligible_municipal_long_schema(a) and a['asset_id']!='01::0'
 and not a['asset_id'].startswith(ORGANIZATION_GRAIN_PREFIXES)]
SOURCE_ADAPTER_AUDIT=[{'asset_id':a['asset_id'],'declared_group':source_group(a),
 'scan_group':classify_asset_for_scan(a),'schema_type':('LONG_MUNICIPAL' if eligible_municipal_long_schema(a)
 else 'NON_MUNICIPAL_OR_WIDE_UNVERIFIED'),
 'has_oktmo':int(bool({'oktmo','object_oktmo'}&set(a.get('columns_manifest') or []))),
 'decision':('SCAN' if a in FACTOR_ASSETS else 'INDEX_ONLY_NO_VALIDATED_MUNICIPAL_GRAIN')}
 for a in ASSETS]

DOWNLOAD_ASSETS=(ASSETS if (CFG['MODE']=='FULL' and CFG['DOWNLOAD_ALL_STAGE1']) else
    [get_asset(BASE_ASSET)]+FACTOR_ASSETS)
if CFG['MODE']=='SMOKE':
    allow={'01::0','02::5ndfl::00::00','02::7ndfl::00::00',
           '02::housing::00::00','02::education::00::00','02::budget::00::00',
           '02::labor::00::00','02::social::00::00','02::construction::00::00',
           '02::socio1::00::00','02::socio2::00::00','02::socio3::00::00',
           '06::D04::D04_member00::spatial_reference::derived'}
    DOWNLOAD_ASSETS=[a for a in DOWNLOAD_ASSETS if a['asset_id'] in allow]
progress('INPUT',0,len(DOWNLOAD_ASSETS),'Сверка файлов с SHA-256')
input_map=pd.DataFrame([{'asset_id':a['asset_id'],'source':a['source_id'],
 'rows_manifest':a['rows_actual'],'group':source_group(a),
 'storage':a['locator']['type'],'status':'INDEXED','sha256':a['sha256']} for a in ASSETS])

# All 52 assets are indexed, including 03-05 (not misinterpreted as municipal observations).
# FULL downloads every asset by default, while only validated relevant sources enter ML-03 JOINs.
for ai,a in enumerate(DOWNLOAD_ASSETS,1):
    try:
        p=safe_remote_download(a)
        DOWNLOAD_QA.append({'asset_id':a['asset_id'],'source_group':source_group(a),'status':'SHA_PASS',
                             'bytes':p.stat().st_size,'error':''})
    except Exception as ex:
        DOWNLOAD_QA.append({'asset_id':a['asset_id'],'source_group':source_group(a),'status':'ERROR',
                             'bytes':0,'error':type(ex).__name__+': '+str(ex)[:120]})
        log('download '+a['asset_id']+': '+type(ex).__name__,'WARN')
    progress('INPUT',ai,len(DOWNLOAD_ASSETS),f'SHA: {sum(v["status"]=="SHA_PASS" for v in DOWNLOAD_QA)} из {len(DOWNLOAD_ASSETS)}')
DOWNLOAD_QA=pd.DataFrame(DOWNLOAD_QA)
required=['01::0']+[a['asset_id'] for a in FACTOR_ASSETS if a in DOWNLOAD_ASSETS]
missing=[aid for aid in required if aid not in LOCAL_ASSETS]
ensure('01::0' in LOCAL_ASSETS,'BLOCKED_INPUT_D01_NOT_AVAILABLE')
mark('PASS_INPUT','PASS' if not missing else 'FAIL',
     f'stage1 available={len(LOCAL_ASSETS)}/{len(DOWNLOAD_ASSETS)}; required missing={len(missing)}')
mark('ALL_STAGE1_DOWNLOAD','PASS' if CFG['MODE']=='FULL' and CFG['DOWNLOAD_ALL_STAGE1'] and len(LOCAL_ASSETS)==len(ASSETS) else 'PARTIAL',
     f'SHA downloaded={len(LOCAL_ASSETS)}/{len(ASSETS)}; includes independent education sources without ML05 JOIN')
if CFG['MODE']=='FULL' and CFG['DOWNLOAD_ALL_STAGE1']:
    ensure(len(LOCAL_ASSETS)==len(ASSETS),
           'FULL_STAGE1_INCOMPLETE_DOWNLOADS; review DOWNLOAD_QA and auth/release assets')

input_map['status']=input_map['asset_id'].map({r['asset_id']:'SHA_VERIFIED' for r in source_receipts}).fillna('INDEX_ONLY_OR_UNAVAILABLE')
STAGE1_PHYSICAL_AUDIT=[]
for a in ASSETS:
    aid=a['asset_id'];p=LOCAL_ASSETS.get(aid)
    if not p:
        STAGE1_PHYSICAL_AUDIT.append({'asset_id':aid,'status':'INDEXED_NOT_DOWNLOADED',
          'expected_rows':a.get('rows_actual'),'actual_rows':None,'metadata_columns':None,'sha_verified':False})
        continue
    try:
        pf=pq.ParquetFile(p);nr=pf.metadata.num_rows;nc=len(pf.schema.names)
        schema_columns=set(pf.schema.names);manifest_columns=set(a.get('columns_manifest') or [])
        ok=nr==a.get('rows_actual') and (not manifest_columns or manifest_columns==schema_columns)
        STAGE1_PHYSICAL_AUDIT.append({'asset_id':aid,'status':'PASS' if ok else 'SCHEMA_OR_ROWS_MISMATCH',
          'expected_rows':a.get('rows_actual'),'actual_rows':nr,'metadata_columns':nc,'sha_verified':True,
          'schema_columns_missing':','.join(sorted(manifest_columns-schema_columns))})
    except Exception as exc:
        STAGE1_PHYSICAL_AUDIT.append({'asset_id':aid,'status':'PARQUET_METADATA_ERROR',
          'expected_rows':a.get('rows_actual'),'actual_rows':None,'metadata_columns':None,
          'sha_verified':True,'exception_type':type(exc).__name__})
STAGE1_PHYSICAL_AUDIT=pd.DataFrame(STAGE1_PHYSICAL_AUDIT)
physical_errors=STAGE1_PHYSICAL_AUDIT.status.isin(['SCHEMA_OR_ROWS_MISMATCH','PARQUET_METADATA_ERROR']).sum()
mark('STAGE1_PHYSICAL_AUDIT','PASS' if physical_errors==0 else 'FAIL',
  f'{len(STAGE1_PHYSICAL_AUDIT)} registered assets; {len(LOCAL_ASSETS)} physical reads; metadata mismatches={int(physical_errors)}')
if physical_errors and CFG['MODE']=='FULL':raise RuntimeError('STAGE1_PHYSICAL_AUDIT_FAILED')

con=duckdb.connect(str(WORK/'ml03.duckdb'))
con.execute(f"SET memory_limit='{CFG['DUCKDB_MEMORY']}'");con.execute(f"SET threads={int(CFG['DUCKDB_THREADS'])}")
base=pd.read_parquet(LOCAL_ASSETS['01::0'],columns=['oktmo','region','mun_type','municipality','year',
    'population','average_population','migration','migration_rate','births','deaths'])
for c in ('oktmo','region','mun_type','municipality'):
    base[c]=base[c].astype('string').str.strip()
base=base.rename(columns={'oktmo':'entity_id','region':'region_name','year':'observation_year'})
for c in ('population','average_population','migration','migration_rate','births','deaths'):
    base[c]=pd.to_numeric(base[c],errors='coerce')
base['observation_year']=pd.to_numeric(base['observation_year'],errors='coerce')
base=base[base.entity_id.notna()&base.entity_id.ne('')&base.observation_year.between(1990,2026)].copy()
base['observation_year']=base.observation_year.astype(int)
base['entity_id']=base.entity_id.astype(str)
repeats=base[base.duplicated(['entity_id','observation_year'],keep=False)].copy()
if len(repeats):
    valuecols=['population','average_population','migration','migration_rate','births','deaths','region_name','mun_type']
    diversity=repeats.groupby(['entity_id','observation_year'])[valuecols].nunique(dropna=False)
    ensure(diversity.le(1).all().all(),'D01_CONFLICTING_ENTITY_YEAR')
    base=base.drop_duplicates(['entity_id','observation_year'])
base.loc[~base.population.gt(0),'population']=np.nan
base=base.sort_values(['entity_id','observation_year']).reset_index(drop=True)
YEARS=sorted(base.observation_year.unique().tolist())

# Independent target audit: no forced equivalence between reported migration_rate and net-migration counts.
MIGRATION_AUDIT=[];record=base.migration_rate
for denom in ['average_population','population']:
    for numerator_scale in [1.,10.,100.,1000.]:
        candidate=base.migration/base[denom]*numerator_scale
        good=record.notna() & candidate.notna() & base[denom].gt(0)
        if good.sum()<100:continue
        delta=(record[good]-candidate[good]).abs()
        consistent=(delta<=np.maximum(1e-5,0.05*record[good].abs())).mean()
        MIGRATION_AUDIT.append({'denominator':denom,'numerator_scale':numerator_scale,
           'n_pairs':int(good.sum()),'agreement_5pct':float(consistent),'median_abs_error':float(delta.median()),
           'median_observed_rate':float(record[good].median())})
MIGRATION_QA=pd.DataFrame(MIGRATION_AUDIT).sort_values(['agreement_5pct','n_pairs','median_abs_error'],ascending=[False,False,True])
if len(MIGRATION_QA):
    best=MIGRATION_QA.iloc[0];valid_rate=bool(best.agreement_5pct>=0.90 and best.n_pairs>=300)
else:valid_rate=False
if valid_rate:
    denom=best['denominator'];scale=best['numerator_scale']
    base['migration_per_1000']=1000.0*base.migration/base[denom]  # persons per 1000 residents; source migration_rate is a fraction when scale=1
    mark('MIGRATION_TARGET_SEMANTICS','PASS',f'{denom}, source_ratio_scale={scale}, converted_to_per_1000=True, n={best.n_pairs}, agreement={best.agreement_5pct:.3f}')
else:
    base['migration_per_1000']=np.nan
    mark('MIGRATION_TARGET_SEMANTICS','NOT_FEASIBLE','Reported migration rate inconsistent with all inspected denominators/scales; source-definition audit required')
base['boundary_uncertainty']='UNVERIFIED_CROSS_YEAR_BOUNDARIES'
base['source_vintage']=source_sha
base_keys=base[['entity_id','observation_year','region_name']].drop_duplicates()
con.register('base_keys',base_keys)
quiet_print('Демография D01:',len(base),'наблюдений;',base.entity_id.nunique(),'ОКТМО;',min(YEARS),'-',max(YEARS))

# Test windows are declared in CONFIG; D01-only presence checks prevent selecting
# a future year with no migration labels. Selection never examines outcome magnitudes.
TARGET_TEST_WINDOWS={t:list(map(int,CFG['PRIMARY_TEST_WINDOWS'][t])) for t in CFG['TARGETS']}
TARGET_CALENDAR_PREFLIGHT=[]
# D01 does not contain population_change_pct: construct availability from real
# population at target year and at origin=target-h, without evaluating magnitude.
_POP_SOURCE=base.loc[base.population.gt(0),['entity_id','observation_year','population']].copy()
_POP_SOURCE= _POP_SOURCE.drop_duplicates(['entity_id','observation_year'])
for target,years in TARGET_TEST_WINDOWS.items():
    ensure(target in ('migration_per_1000','population_change_pct'), 'UNKNOWN_D01_TARGET_'+str(target))
    for year in years:
        by_h={}
        for h in CFG['HORIZONS']:
            if target=='migration_per_1000':
                observed=base.loc[base.observation_year.eq(year),'migration_per_1000'].notna()
                n=int(observed.sum())
            else:
                current=_POP_SOURCE.loc[_POP_SOURCE.observation_year.eq(year),['entity_id']]
                origin=_POP_SOURCE.loc[_POP_SOURCE.observation_year.eq(year-int(h)),['entity_id']]
                n=int(current.merge(origin,on='entity_id',how='inner',validate='one_to_one').shape[0])
            by_h[int(h)]=n
        # Record each horizon separately; no silent reuse of h1 counts for h3/h5.
        for h,n in by_h.items():
            TARGET_CALENDAR_PREFLIGHT.append({'target':target,'target_year':int(year),'horizon':int(h),
              'available_labels':int(n),'role':'PREREGISTERED_TWO_YEAR_PRIMARY',
              'status':'OBSERVED' if n>0 else 'MISSING',
              'availability_derived_from':'D01_MIGRATION_PER_1000' if target=='migration_per_1000' else 'D01_POPULATION_ORIGIN_AND_TARGET'})
TARGET_CALENDAR_PREFLIGHT=pd.DataFrame(TARGET_CALENDAR_PREFLIGHT)
if TARGET_CALENDAR_PREFLIGHT.loc[TARGET_CALENDAR_PREFLIGHT.horizon.eq(1),'status'].eq('MISSING').any():
    mark('PRIMARY_TARGET_D01_GATE','FAIL','Preregistered target years are unobserved; revise protocol explicitly, never silently shift')
    if CFG['MODE']=='FULL':raise RuntimeError('PRIMARY_TARGET_D01_GATE_FAILURE: '+TARGET_CALENDAR_PREFLIGHT.to_json(orient='records'))
else:mark('PRIMARY_TARGET_D01_GATE','PASS',f'Observed outcome labels in primary h=1 windows: {TARGET_TEST_WINDOWS}')
```

## 4. Предметная инвентаризация источников
Проводится проверка официальных БДМО. Если раздел слишком велик или имеет нестандартный формат, он остаётся в отчёте проблем, без реконструкции данных. Организационные таблицы E02/E04/E05 без подтверждённой муниципальной агрегации не выдаются за территориальные итоги.


```python
# 05 · V4.00 · SUBJECT-VALIDATED FACTOR INVENTORY · SQL/LEVEL CHECKED
# Дополнительные муниципальные источники БДМО: только проверяемые фактические данные.
# Разделы основного каталога Stage I (52) остаются неизменными и отдельно аудируются.
import html as _html
from html.parser import HTMLParser as _HTMLParser
BDMO_EXTERNAL_QA_ROWS=[];BDMO_EXTERNAL_ASSETS=[];BDMO_EXTERNAL_RECEIPTS=[]
BDMO_OFFICIAL_CATALOGUE='https://tochno.st/datasets/bdmo'
BDMO_EXTRA_SECTIONS={1:'G1',2:'G1',3:'G3',6:'G5',8:'G2',9:'G1',19:'G3',32:'G1',33:'G3',39:'G1',42:'G4',70:'G5'}
BDMO_EXTERNAL_DESIGN=pd.DataFrame([{'section':s,'assigned_group':g,'source':'BDMO/Rosstat via tochno.st',
    'data_grain':'municipality_year_long, validated at query time',
    'selection':'all codes catalogued, only preregistered semantic domains retained'}
    for s,g in BDMO_EXTRA_SECTIONS.items()])

class _LinkCollector(_HTMLParser):
    def __init__(self):super().__init__();self.hrefs=[]
    def handle_starttag(self,tag,attrs):
        if tag=='a':
            d=dict(attrs)
            if d.get('href'):self.hrefs.append(_html.unescape(d['href']))

def _supplementary_urls():
    r=SESSION.get(BDMO_OFFICIAL_CATALOGUE,timeout=(15,60));r.raise_for_status()
    parser=_LinkCollector();parser.feed(r.text)
    u=set(parser.hrefs)|set(re.findall(r'https?://[^\s\"<>]+?(?:\.zip)',_html.unescape(r.text)))
    result={}
    for sec in BDMO_EXTRA_SECTIONS:
        pattern=rf'/by_(?:section|indicator)/data_section{sec}_.*?\.zip'
        matches=[v for v in u if re.search(pattern,v) and urllib.parse.urlparse(v).hostname=='storage.yandexcloud.net']
        # Fallback is the same dated catalogue release already audited in Stage I,
        # not a claim that the remote file exists: HTTP response must validate it.
        if not matches:
            base_url=next((x.get('source_url') for x in ASSETS if x.get('source_id')=='BDMO' and '/by_section/' in str(x.get('source_url'))),None)
            if base_url and sec!=32:
                mm=re.search(r'data_section\d+_',base_url)
                if mm:matches=[base_url[:mm.start()]+'data_section'+str(sec)+'_'+base_url[mm.end():].replace('data_section13_',f'data_section{sec}_')]
        if matches:result[sec]=max(matches,key=lambda s:re.search(r'v(20\d{6})',s).group(1) if re.search(r'v(20\d{6})',s) else '')
    return result

def _supplement_one_section(sec,url,group):
    pu=urllib.parse.urlparse(url)
    ensure(pu.scheme=='https' and pu.hostname=='storage.yandexcloud.net' and '/tochno-st-catalog/Rosstat/' in pu.path,'BDMO_EXTERNAL_URL_UNTRUSTED')
    zpath=WORK/'cache'/f'bdmo_external_{sec}.zip';part=zpath.with_suffix('.part')
    side=zpath.with_suffix('.source_url.txt')
    if zpath.exists() and (not side.exists() or side.read_text(encoding='utf-8')!=url):
        zpath.unlink()
    if not zpath.exists():
        part.unlink(missing_ok=True);nr=0;h=hashlib.sha256()
        try:
            rr=SESSION.get(url,timeout=(30,180),stream=True);rr.raise_for_status()
            with open(part,'wb') as fd:
                for buf in rr.iter_content(4*1024*1024):
                    if not buf:continue
                    nr+=len(buf)
                    ensure(nr<=CFG['BDMO_EXTERNAL_MAX_SOURCE_GB']*2**30,'BDMO_EXTERNAL_ZIP_SIZE_LIMIT')
                    if nr%(512*1024*1024)<len(buf):
                        ensure(shutil.disk_usage(WORK).free>CFG['MIN_DISK_FREE_GB']*2**30,
                               'BDMO_EXTERNAL_DISK_SAFETY_STOP')
                    h.update(buf);fd.write(buf)
            rr.close();ensure(zipfile.is_zipfile(part),'BDMO_EXTERNAL_NOT_ZIP')
            part.replace(zpath);side.write_text(url,encoding='utf-8')
        except Exception:part.unlink(missing_ok=True);raise
    ensure(zipfile.is_zipfile(zpath),'BDMO_EXTERNAL_CACHE_NOT_ZIP')
    source_sha=digest(zpath);indexed=0
    with zipfile.ZipFile(zpath) as z:
        ensure(z.testzip() is None,'BDMO_EXTERNAL_ZIP_CRC')
        infos=[i for i in z.infolist() if not i.is_dir() and i.filename.lower().endswith('.parquet')]
        infos.sort(key=lambda x:x.filename)
        if not infos:
            BDMO_EXTERNAL_QA_ROWS.append({'section':sec,'status':'SOURCE_FORMAT_NOT_PARQUET',
                'reason':'The validated archive contains no direct Parquet member; ' +
                         'nested/CSV sources require separate schema and grain validation',
                'members_sample':json.dumps([i.filename for i in z.infolist()[:20]],ensure_ascii=False)})
            return 0
        ensure(len(infos)<=CFG['BDMO_EXTERNAL_MAX_MEMBERS_PER_SECTION'],'BDMO_EXTERNAL_TOO_MANY_MEMBERS')
        for idx,member in enumerate(infos):
            ensure(member.file_size<CFG['BDMO_EXTERNAL_MAX_UNPACKED_GB']*2**30,'BDMO_EXTERNAL_UNPACKED_MEMBER_LIMIT')
            dest=WORK/'cache'/f'bdmo_external_{sec}_{idx:03d}.parquet'
            if not dest.exists() or dest.stat().st_size!=member.file_size:
                tmp=dest.with_suffix('.part')
                try:
                    with z.open(member) as src,open(tmp,'wb') as dst:shutil.copyfileobj(src,dst,length=4*1024*1024)
                    ensure(tmp.stat().st_size==member.file_size,'BDMO_EXTERNAL_MEMBER_BYTES')
                    tmp.replace(dest)
                except Exception:tmp.unlink(missing_ok=True);raise
            info=pq.ParquetFile(dest);cols=info.schema.names
            needed={'year','indicator_code','indicator_name','indicator_value','oktmo'}
            if not needed.issubset(cols):
                BDMO_EXTERNAL_QA_ROWS.append({'section':sec,'member':member.filename,'status':'SCHEMA_UNSUPPORTED','columns':','.join(cols[:35])});continue
            aid=f'SUPP::BDMO::{sec}::{idx:03d}'
            entry={'asset_id':aid,'source_id':'BDMO','role':'municipal_long_supplement',
                'segment':'BDMO_EXTERNAL','columns_manifest':cols,'rows_actual':info.metadata.num_rows,
                'source_url':url,'source_sha256':source_sha,'sha256':digest(dest),
                'locator':{'type':'verified_external_cache','path':str(dest)}}
            BDMO_EXTERNAL_ASSETS.append(entry);LOCAL_ASSETS[aid]=str(dest)
            SOURCE_CLASSES[group].append(aid)
            if entry not in FACTOR_ASSETS:FACTOR_ASSETS.append(entry)
            BDMO_EXTERNAL_RECEIPTS.append({'asset_id':aid,'group':group,'source_url':url,'source_sha256':source_sha,
                'parquet_sha256':entry['sha256'],'source_member':member.filename,'rows':entry['rows_actual'],
                'columns':len(cols),'status':'SHA_VERIFIED_AND_SCHEMA_CHECKED'})
            BDMO_EXTERNAL_QA_ROWS.append({'section':sec,'member':member.filename,'status':'ADAPTED',
                'asset_id':aid,'rows':entry['rows_actual'],'group':group})
            indexed+=1
    return indexed

if CFG['ENABLE_EXTERNAL_BDMO'] and CFG['MODE']=='FULL':
    try:
        urls=_supplementary_urls();BDMO_EXTERNAL_QA_ROWS.append({'section':0,'status':'CATALOGUE_OPENED','n_urls_found':len(urls)})
        for _bi,(section,group) in enumerate(BDMO_EXTRA_SECTIONS.items(),1):
            progress('BDMO',_bi-1,len(BDMO_EXTRA_SECTIONS),f'Раздел {section}')
            if section not in urls:
                BDMO_EXTERNAL_QA_ROWS.append({'section':section,'group':group,'status':'URL_NOT_DISCOVERED'});continue
            try:
                n=_supplement_one_section(section,urls[section],group)
                if n==0:log(f'BDMO section {section}: no verified Parquet member; review format audit','WARN')
                quiet_print(f'Дополнительный БДМО раздел {section}: {n} Parquet; SHA проверен')
            except Exception as ex:
                BDMO_EXTERNAL_QA_ROWS.append({'section':section,'group':group,'status':'SOURCE_LOAD_FAILED',
                    'reason':type(ex).__name__+': '+str(ex)[:240]})
                log(f'External BDMO section {section} rejected: {type(ex).__name__}: {str(ex)[:180]}','WARN')
    except Exception as ex:
        BDMO_EXTERNAL_QA_ROWS.append({'section':0,'status':'SOURCE_CATALOGUE_UNAVAILABLE',
            'reason':type(ex).__name__+': '+str(ex)[:240]})
        log('BDMO external catalogue unavailable: '+type(ex).__name__,'WARN')
progress('BDMO',1,1,f'Проверено разделов: {len(BDMO_EXTRA_SECTIONS)}')
BDMO_EXTERNAL_RECOVERY=pd.DataFrame([{
    'section':int(s),'group':g,'status':next((x.get('status') for x in reversed(BDMO_EXTERNAL_QA_ROWS)
        if x.get('section')==s),'NOT_ATTEMPTED'),
    'source_url':urls.get(s,'') if 'urls' in locals() else '',
    'must_not_claim_loaded_without_sha':True}
    for s,g in BDMO_EXTRA_SECTIONS.items()])
BDMO_EXTERNAL_QA=pd.DataFrame(BDMO_EXTERNAL_QA_ROWS)
mark('EXTERNAL_BDMO_SOURCE_AUDIT','PASS' if len(BDMO_EXTERNAL_ASSETS)>=3 else 'PARTIAL',
     f'External verified Parquet={len(BDMO_EXTERNAL_ASSETS)}, sections={sorted(set(int(s.split("::")[2]) for s in LOCAL_ASSETS if s.startswith("SUPP::BDMO::")))}; not part of original 52')
_ORIGINAL_GET_ASSET=get_asset
def get_asset(aid):
    for _a in BDMO_EXTERNAL_ASSETS:
        if _a['asset_id']==aid:return _a
    return _ORIGINAL_GET_ASSET(aid)

PATTERNS = {
 'G1':'зарплат|заработ|оплат.{0,12}труда|доход|инвестиц|занятост|занятых|безработ|предпринимател|предприяти|оборот|производств|налог|уровен.{0,8}жизни|средн.{0,15}душ',
 'G2':'жил|жилищ|квартир|строительств|коммунал|водоснабж|водоотвед|теплоснабж|газификац|электроснабж|канализац|благоустрой',
 'G3':'образован|школ|дошколь|детск.{0,8}сад|медицин|врач|фельдшер|больниц|поликлиник|амбулатор|социал|койко|здравоохран|спорт|культур',
 'G4':'бюджет|расход|дотаци|субсид|субвенци|межбюджет|налогов.{0,10}доход|казн',
 'G5':'расстояни|транспорт|доступност|пространств|территори|площад.{0,18}(?:земл|муницип)|автомобил.{0,12}дорог|покрыти.{0,12}дорог|связь|телефон|интернет'}
SEMANTIC_DOMAINS = {
 'G1':[('wages','заработн.{0,10}плат|средн.{0,12}зарплат|оплат.{0,7}труда|фонд.{0,10}оплат'),('employment','численност.{0,10}(?:занят|работающ|работник)|число.{0,10}(?:занят|работающ|работник)|уровень.{0,10}безработ|число.{0,12}безработ|рабоч.{0,8}мест|занят.{0,10}насел'),('business','мал.{0,12}предприят|средн.{0,12}предприят|предпринимател|субъект.{0,12}мсп|оборот.{0,12}организац|зарегистрированн.{0,15}организац'),('investment','инвестиц|капитальн.{0,10}вложен'),('retail','розничн.{0,14}торгов|оборот.{0,15}торгов|общественн.{0,15}питан'),('paid_services','платн.{0,12}услуг|услуг.{0,12}населен'),('income','денежн.{0,12}доход|сумм.{0,12}доход|совокупн.{0,10}доход|налогооблагаем.{0,10}доход|среднедушев.{0,12}доход|доход.{0,12}населени')],
 'G2':[('housing_stock','жил.{0,8}фонд|жил.{0,9}площад|обеспеченн.{0,12}жиль|площад.{0,12}жил'),('housing_construction','ввод.{0,30}(?:жил|жиль|квартир)|введен.{0,20}(?:жил|квартир)|строительств.{0,14}жил|построенн.{0,12}квартир'),('housing_access','улучшивш.{0,12}жилищ|нуждающ.{0,12}жилищ|семе.{0,12}жилищ'),('utilities','водоснабж|водоотвед|теплоснабж|газификац|канализац|коммунал|благоустроенн.{0,12}жил|электроснабж')],
 'G3':[('health','врач|фельдшер|поликлиник|больниц|больнич|здравоохран|медицин|амбулатор|койко|медработ'),('education','образован|школ|дошколь|детск.{0,8}сад|обучающ|учащ|педагог'),('social','социальн.{0,14}(учрежден|обслужив|защит|услуг)|дом.{0,12}культур|спортивн.{0,10}сооруж')],
 'G5':[('territorial_area','площад.{0,18}(?:земл|муницип)|площад.{0,8}территори'),('roads','автомобил.{0,12}дорог|протяж[её]нност.{0,12}дорог|дорог.{0,12}покрыти'),('connectivity','телефонн|интернет|связ[ьи]|доступност.{0,12}сети')],
 'G4':[('municipal_spending','расход|содержание.{0,12}работник|затрат'),('municipal_revenue','доход.{0,12}бюджет|налогов.{0,12}доход|собственн.{0,10}доход'),('intergovernmental','дотаци|субвенци|субсид|межбюджет|трансферт'),('budget_balance','дефицит|профицит|долг|сбалансированност|бюджет')]}
EXCLUDED_INCOME='ценн.{0,8}бумаг|дивиденд|вычет|ставк.{0,12}налог|исчисленн.{0,12}налог|сумм.{0,12}налог|полученн.{0,8}сведен|количеств.{0,12}сведен|числ.{0,9}сведен|операц.{0,14}бумаг|код.{0,8}доход'
REJECT_NONMEASUREMENT='методик|справочн|код.{0,7}показател|инструкция|наименование|описание|классификатор'

def classify_indicator(group,asset_id,name,unit):
    title=re.sub(r'\s+',' ',str(name).casefold()).strip();un=str(unit).casefold()
    if not title or re.search(REJECT_NONMEASUREMENT,title):return None,'NON_MEASUREMENT'
    if group not in SEMANTIC_DOMAINS:return None,'NO_REGISTERED_DOMAIN'
    if group=='G1' and re.search(EXCLUDED_INCOME,title):return None,'NARROW_TAX_OR_DEDUCTION_PROXY'
    if group=='G1' and 'ndfl' in asset_id.lower():
        if not re.search(r'денежн.{0,12}доход|сумм.{0,12}доход|совокупн.{0,10}доход|налогооблагаем.{0,10}доход',title):return None,'NDFL_NOT_INCOME_AMOUNT'
        if re.search(r'штук|единиц|количеств|сведен|запис',un):return None,'NDFL_COUNT_NOT_INCOME'
    # 'Продано товаров (предприятия со штатом >15)' measures SALES, not workers.
    if group=='G1' and re.search(r'продано товар|продаж.{0,25}товар|отгружено товар|отгрузк.{0,20}товар',title):
        return 'business','SALES_VOLUME_NOT_EMPLOYMENT'
    # A count of medical social workers is not automatically hospital infrastructure.
    for dom,pattern in SEMANTIC_DOMAINS[group]:
        if re.search(pattern,title):return dom,''
    return None,'NO_SUBSTANTIVE_MATCH'

TOTAL_DIM_VALUES={'всего','итого','все','все категории','total','all','общий','общая','общие'}
NDFL_RISK_DIMENSIONS=('individual_type','income','deduction','tax_rate','report_type','index_form','tax_authority')

def eligible_slice(asset_id,dims):
    if 'ndfl' not in asset_id.lower():return True,'CATEGORY_SPECIFIED_NO_COLLAPSE'
    for k in NDFL_RISK_DIMENSIONS:
        if k in dims and str(dims[k]).casefold().strip() not in TOTAL_DIM_VALUES:
            return False,'NON_VERIFIED_TOTAL_NDFL_DIMENSION_'+k
    return True,'NDFL_ALL_RISK_DIMENSIONS_EXPLICIT_TOTAL_OR_ABSENT'

ensure(classify_indicator('G1','02::5ndfl::00::00','Количество сведений о доходах','Штука')[0] is None,'SEMANTIC_CANARY_TAX_COUNTS')
ensure(classify_indicator('G1','02::labor','Средняя начисленная заработная плата','рублей')[0]=='wages','SEMANTIC_CANARY_WAGES')
ensure(classify_indicator('G3','02::social','Количество врачей','человек')[0]=='health','SEMANTIC_CANARY_HEALTH')
ensure(classify_indicator('G1','SUPP::BDMO','Продано товаров несобственного производства (без субъектов малого предпринимательства), средняя численность работников которых превышает 15 человек','рублей')[0]=='business','SEMANTIC_CANARY_SALES_NOT_EMPLOYMENT')
ensure(not eligible_slice('02::5ndfl::00::00',{'income':'2204','tax_rate':'9'})[0],'SEMANTIC_CANARY_NDFL_SLICES')
mark('SEMANTIC_CANARY','PASS','Tax counts, wage concept, medicine and narrow NDFL slice guards')

PRIMARY_TARGET_YEARS=sorted(set(sum(TARGET_TEST_WINDOWS.values(),[])))
FACTOR_CUTOFF_BY_H={int(h):int(min(min(yy) for yy in TARGET_TEST_WINDOWS.values())-h-CFG['FEATURE_PUBLICATION_LAG']) for h in CFG['HORIZONS']}
TRAIN_CATALOG_MAX_YEAR=max(FACTOR_CUTOFF_BY_H.values());TRAIN_CATALOG_MIN_YEAR=TRAIN_CATALOG_MAX_YEAR-CFG['CATALOG_RECENT_WINDOW']
CATALOG_ROWS=[];FEATURE_PARTS=[];conflict_rows=[];SOURCE_QA=[];FEATURE_CANDIDATES=[];SELECTED=[];CANDIDATE_AUDIT=[];G5_DIAGNOSTICS=[];SCAN_AUDIT=[];FACTOR_ELIGIBILITY=[];DIMENSION_AUDIT=[]
LAST_SLICE_REASON='NOT_STARTED'

def qstr(v):return "'"+str(v).replace("'","''")+"'"
def qident(s):
    v=str(s);ensure(bool(v) and '\x00' not in v,'INVALID_SOURCE_COLUMN')
    return '"'+v.replace('"','""')+'"'

SQL_FAILURES=[];SQL_EXECUTION_AUDIT=[];REAL_SQL_PREFLIGHT=[]

def sql_run(sql,params=None,*,asset_id='',code='',stage='',preview=False):
    args=list(params or []);t0=time.monotonic()
    try:
        result=con.execute(sql,args);data=result.fetchone() if preview else result.df()
        SQL_EXECUTION_AUDIT.append({'asset_id':asset_id,'code':str(code),'stage':stage,'status':'PASS','elapsed_ms':int((time.monotonic()-t0)*1000),'sql_sha256':hashlib.sha256(sql.encode()).hexdigest()})
        return data
    except Exception as exc:
        safe_sql=re.sub(r'(?i)(ghp_|github_pat_)[a-z0-9_]+','[REDACTED]',sql)
        detail={'asset_id':asset_id,'code':str(code),'stage':stage,'status':'SCAN_ERROR','exception_type':type(exc).__name__,'message':str(exc)[:2000],'sql':safe_sql[:12000],'sql_sha256':hashlib.sha256(sql.encode()).hexdigest(),'parameters_types':[type(v).__name__ for v in args],'elapsed_ms':int((time.monotonic()-t0)*1000)}
        SQL_FAILURES.append(detail);log('SQL_SCAN_ERROR '+json.dumps({k:v for k,v in detail.items() if k!='sql'},ensure_ascii=False),'WARN')
        raise

def column_key(a):
    cc=a.get('columns_manifest') or []
    return 'oktmo' if 'oktmo' in cc else 'object_oktmo' if 'object_oktmo' in cc else None

def dimensions_for(a):
    cc=a.get('columns_manifest') or [];aid=a['asset_id']
    if 'ndfl' in aid.lower():return [k for k in NDFL_RISK_DIMENSIONS if k in cc]
    omit={'indicator_section_code','indicator_section','indicator_code','indicator_name','indicator_unit','indicator_period','indicator_value','year','region_id','region_name','region','oktmo','object_oktmo','oktmo_stable','oktmo_history','oktmo_year_from','oktmo_year_to','mun_level','mun_district','municipality','mun_type','mun_type_oktmo','object_name','object_level','object_okato','source','comment','version_date','report_date','report_region_id','report_region_name','null_value_reason'}
    return [c for c in cc if c not in omit]

def val_sql(alias='a'):
    return f"TRY_CAST(REPLACE(REPLACE(REPLACE(TRIM(CAST({alias}.indicator_value AS VARCHAR)), ' ', ''), chr(160), ''), ',', '.') AS DOUBLE)"

def _grp_eligible_from_source(a):
    g=classify_asset_for_scan(a)
    return ['G1','G2','G3','G4'] if g=='DISCOVERY_MULTIDOMAIN' else [g] if g in SEMANTIC_DOMAINS else []

def inventory_asset(a,path,group):
    key=column_key(a);cc=set(a.get('columns_manifest') or [])
    if key is None or not {'indicator_code','indicator_name','indicator_value','year'}.issubset(cc):return pd.DataFrame()
    unit=f"COALESCE(TRIM(CAST(a.{qident('indicator_unit')} AS VARCHAR)), '')" if 'indicator_unit' in cc else "''"
    k=qident(key);yr='TRY_CAST(a.'+qident('year')+' AS INTEGER)'
    level_predicate=("regexp_matches(lower(coalesce(cast(a."+qident('object_level')+" AS VARCHAR),'')), 'муниципал|городск.{0,8}округ')" if 'object_level' in cc else 'TRUE')
    sql=f"""SELECT TRIM(CAST(a.{qident('indicator_code')} AS VARCHAR)) AS indicator_code,
      TRIM(CAST(a.{qident('indicator_name')} AS VARCHAR)) AS indicator_name,{unit} AS indicator_unit,
      COUNT(DISTINCT TRIM(CAST(a.{k} AS VARCHAR))) AS n_municipalities,
      COUNT(DISTINCT {yr}) AS n_years,
      COUNT(DISTINCT CASE WHEN {yr} BETWEEN {TRAIN_CATALOG_MIN_YEAR} AND {TRAIN_CATALOG_MAX_YEAR}
      THEN TRIM(CAST(a.{k} AS VARCHAR))||':'||CAST({yr} AS VARCHAR) END) AS recent_keys,
      MIN({yr}) AS min_year,MAX({yr}) AS max_year
      FROM read_parquet(?) a JOIN base_keys b ON TRIM(CAST(a.{k} AS VARCHAR))=b.entity_id
         AND {yr}=b.observation_year
      WHERE {yr}<={TRAIN_CATALOG_MAX_YEAR} AND {level_predicate}
       AND regexp_matches(lower(coalesce(cast(a.{qident('indicator_name')} AS VARCHAR),'')),?)
       AND a.{qident('indicator_code')} IS NOT NULL
      GROUP BY 1,2,3 HAVING COUNT(DISTINCT a.{k})>={CFG['MIN_INDICATOR_MUNICIPALITIES']}
      ORDER BY indicator_code,indicator_name,indicator_unit"""
    return sql_run(sql,[str(path),PATTERNS[group]],asset_id=a['asset_id'],stage='INVENTORY')

def _slice_dimensions_supported(a,dims):
    if 'ndfl' not in a['asset_id'].lower():return True,'NON_NDFL_DIMENSIONAL_SLICE'
    ok,reason=eligible_slice(a['asset_id'],dims)
    if not ok:return False,reason
    if any(k in dims for k in ('income','deduction','tax_rate','individual_type','report_type','index_form')):
        return True,'NDFL_EXPLICIT_TOTAL_VALUES_NO_COLLAPSE'
    return True,'NDFL_UNSTRATIFIED_NO_AGGREGATION'

def factor_raw_sql(a):
    key=column_key(a);cc=set(a.get('columns_manifest') or []);dims=dimensions_for(a)
    ensure(key is not None and {'indicator_code','indicator_name','indicator_value','year'}.issubset(cc),'INVALID_FACTOR_SCHEMA')
    d_expr=','.join(f"COALESCE(NULLIF(TRIM(CAST(a.{qident(d)} AS VARCHAR)),''),'∅') AS {qident('dim_'+str(i).zfill(2))}" for i,d in enumerate(dims))
    d_expr=','+d_expr if d_expr else ''
    unitfilter=f" AND COALESCE(TRIM(CAST(a.{qident('indicator_unit')} AS VARCHAR)),'')=?" if 'indicator_unit' in cc else ''
    level_filter=(" AND regexp_matches(lower(coalesce(cast(a."+qident('object_level')+" AS VARCHAR),'')), 'муниципал|городск.{0,8}округ')" if 'object_level' in cc else '')
    raw=f"""SELECT TRIM(CAST(a.{qident(key)} AS VARCHAR)) AS entity_id,
       TRY_CAST(a.{qident('year')} AS INTEGER) AS observation_year,{val_sql()} AS val{d_expr}
       FROM read_parquet(?) a JOIN base_keys b
         ON TRIM(CAST(a.{qident(key)} AS VARCHAR))=b.entity_id
         AND TRY_CAST(a.{qident('year')} AS INTEGER)=b.observation_year
       WHERE TRIM(CAST(a.{qident('indicator_code')} AS VARCHAR))=?
         AND TRIM(CAST(a.{qident('indicator_name')} AS VARCHAR))=? {unitfilter} {level_filter}"""
    return raw,dims,bool(unitfilter)

def select_dimensional_slice(a,path,code,name,unit,group):
    global LAST_SLICE_REASON
    LAST_SLICE_REASON='NO_VALID_CANDIDATE'
    raw,dims,with_unit=factor_raw_sql(a)
    baseparams=[str(path),str(code),str(name)]+([str(unit)] if with_unit else [])
    d_alias=[qident('dim_'+str(i).zfill(2)) for i in range(len(dims))]
    d_select=','.join(d_alias)+',' if dims else ''
    pos=' GROUP BY '+','.join(map(str,range(1,len(dims)+1))) if dims else ''
    summary_sql=f"""WITH v AS ({raw}) SELECT {d_select}
       COUNT(DISTINCT CASE WHEN observation_year<={TRAIN_CATALOG_MAX_YEAR} AND val IS NOT NULL
           THEN entity_id||':'||CAST(observation_year AS VARCHAR) END) AS n_train_keys,
       COUNT(DISTINCT CASE WHEN observation_year BETWEEN {TRAIN_CATALOG_MIN_YEAR} AND {TRAIN_CATALOG_MAX_YEAR}
         AND val IS NOT NULL THEN entity_id||':'||CAST(observation_year AS VARCHAR) END) AS n_recent_keys,
       COUNT(DISTINCT CASE WHEN observation_year<={TRAIN_CATALOG_MAX_YEAR} AND val IS NOT NULL
         THEN observation_year END) AS n_train_years FROM v {pos}"""
    if 'ndfl' in a['asset_id'].lower():
        filter_terms=[f'lower(trim({d_alias[i]})) IN ('+','.join(qstr(s) for s in sorted(TOTAL_DIM_VALUES))+')' for i,d in enumerate(dims) if d in NDFL_RISK_DIMENSIONS]
        predicate=' AND '.join(filter_terms) if filter_terms else 'TRUE'
        budget=max(1,int(CFG['MAX_NDFL_SLICE_VALUES']))
    else:
        predicate='TRUE';budget=max(1,int(CFG['MAX_DIMENSIONAL_SLICES_PER_CODE']))
    total_sql='SELECT COUNT(*) FROM ('+summary_sql+') total_slices'
    total=int(sql_run(total_sql,baseparams,asset_id=a['asset_id'],code=code,stage='SLICE_COUNT',preview=True)[0])
    valid_sql='SELECT COUNT(*) FROM ('+summary_sql+') grouped WHERE '+predicate
    eligible=int(sql_run(valid_sql,baseparams,asset_id=a['asset_id'],code=code,stage='SLICE_ELIGIBLE_COUNT',preview=True)[0])
    take=min(eligible,budget)
    DIMENSION_AUDIT.append({'asset_id':a['asset_id'],'indicator_code':code,'slices_found':total,'semantically_eligible':eligible,'inspected':take,'remaining':max(eligible-take,0),'rejected_semantically':total-eligible,'status':'NO_VALID_TOTAL_SLICE' if 'ndfl' in a['asset_id'].lower() and eligible==0 else 'CANDIDATE_BUDGET_PARTIAL' if eligible>budget else 'EXHAUSTIVE_SQL_FILTERED'})
    if not eligible:
        LAST_SLICE_REASON='NO_VALID_TOTAL_SLICE' if 'ndfl' in a['asset_id'].lower() else 'NO_DIMENSION_SLICES'
        return None,None,None
    top_sql='SELECT * FROM ('+summary_sql+') grouped WHERE '+predicate+' ORDER BY n_recent_keys DESC,n_train_keys DESC LIMIT ?'
    top=sql_run(top_sql,baseparams+[take],asset_id=a['asset_id'],code=code,stage='SLICE_RANKING')
    if top.empty:return None,None,None
    for row in top.to_dict('records'):
        sl={d:str(row['dim_'+str(i).zfill(2)]) for i,d in enumerate(dims)}
        ok,why=_slice_dimensions_supported(a,sl)
        if not ok:
            CANDIDATE_AUDIT.append({'asset_id':a['asset_id'],'code':code,'name':name,'decision':why,'slice':json.dumps(sl,ensure_ascii=False)})
            continue
        if int(row['n_train_keys'])<CFG['MIN_INDICATOR_OBS_TRAIN'] or int(row['n_train_years'])<CFG['MIN_INDICATOR_YEARS_TRAIN']:continue
        pred=' AND '.join(f'{d_alias[i]}=?' for i in range(len(dims))) if dims else 'TRUE'

        # FIX: DuckDB requires the output alias "value" to be quoted.
        audit_sql=f"""WITH v AS ({raw}), sliced AS
          (SELECT entity_id,observation_year,val FROM v WHERE {pred})
          SELECT entity_id,observation_year,
            MIN(val) AS "value",
            COUNT(DISTINCT val) AS n_different,
            COUNT(*) AS n_raw
          FROM sliced GROUP BY entity_id,observation_year"""

        audit=sql_run(audit_sql,baseparams+[sl[d] for d in dims],asset_id=a['asset_id'],code=code,stage='EXTRACT_SLICE')
        if audit.empty:continue
        conflict=audit.n_different.gt(1);train_mask=audit.observation_year.le(TRAIN_CATALOG_MAX_YEAR)
        train_conf=conflict&train_mask;share=float(train_conf.sum()/max(1,train_mask.sum()))
        if conflict.any():
            conflict_rows.append({'asset':a['asset_id'],'code':code,'n_collisions':int(conflict.sum()),'share':share,'slice':json.dumps(sl,ensure_ascii=False),'decision':'REJECT_NONUNIQUE_SLICED_KEY' if share>CFG['MAX_CONFLICT_SHARE'] else 'EXCLUDE_CONFLICTED_ENTITY_YEARS'})
        if share>CFG['MAX_CONFLICT_SHARE']:continue
        audit=audit.loc[~conflict,['entity_id','observation_year','value']].copy()
        train_values=audit.observation_year.le(TRAIN_CATALOG_MAX_YEAR)&audit['value'].notna()
        if int(train_values.sum())<CFG['MIN_INDICATOR_OBS_TRAIN']:continue
        ensure(not audit.duplicated(['entity_id','observation_year']).any(),'NON_UNIQUE_FEATURE_OUTPUT')
        LAST_SLICE_REASON='ELIGIBLE'
        return audit,sl,row
    LAST_SLICE_REASON='PARTIAL_CANDIDATES_INSPECTED' if eligible>budget else 'NO_VALID_SLICE_AFTER_SUPPORT_CHECK'
    return None,None,None

def run_duckdb_sql_contracts():
    tmp=WORK/'cache'/'ml03_stage3_sql_contract.parquet';tmp_ndfl=WORK/'cache'/'ml03_stage3_ndfl_contract.parquet'
    audit0=len(SQL_EXECUTION_AUDIT);dim0=len(DIMENSION_AUDIT);conf0=len(conflict_rows)
    selected=base_keys.loc[base_keys.observation_year.between(TRAIN_CATALOG_MIN_YEAR,TRAIN_CATALOG_MAX_YEAR),['entity_id','observation_year']].drop_duplicates()
    eligible_ids=selected.groupby('entity_id').observation_year.nunique()
    eligible_ids=eligible_ids[eligible_ids>=2].index[:100]
    selected=selected[selected.entity_id.isin(eligible_ids)].groupby('entity_id').head(2)
    ensure(len(selected)>=100 and selected.observation_year.nunique()>=2 and selected.entity_id.nunique()>=CFG['MIN_INDICATOR_MUNICIPALITIES'],'SQL_UNIT_FIXTURE_INSUFFICIENT_BASE_KEYS')
    fixture=selected.rename(columns={'entity_id':'oktmo','observation_year':'year'}).copy()
    fixture['indicator_code']='Y48010001';fixture['indicator_name']='Площадь жилищного фонда';fixture['indicator_unit']='кв. м'
    fixture['indicator_value']='1 234,5';fixture['group']='Все';fixture['order']='Вся территория';fixture['index"quoted']='Данные'
    con.register('ml03_fixture_df',fixture)
    try:
        con.execute('COPY (SELECT * FROM ml03_fixture_df) TO '+qstr(str(tmp))+' (FORMAT PARQUET)')
        toy={'asset_id':'02::housing::SYNTHETIC_SQL_CONTRACT','columns_manifest':list(fixture.columns)}
        inv=inventory_asset(toy,tmp,'G2')
        ensure(not inv.empty and inv.indicator_code.eq('Y48010001').any(),'SQL_UNIT_INVENTORY_EMPTY')
        dat,sl,support=select_dimensional_slice(toy,tmp,'Y48010001','Площадь жилищного фонда','кв. м','G2')
        ensure(dat is not None and len(dat)>=100,'SQL_UNIT_EXTRACT_NO_ROWS')
        ensure(dat['value'].notna().all() and np.allclose(dat['value'].to_numpy(),1234.5),'SQL_UNIT_LOCALE_DECIMAL')
        ensure(not dat.duplicated(['entity_id','observation_year']).any(),'SQL_UNIT_DUPLICATE_KEYS')
        ensure(sl['group']=='Все' and sl['order']=='Вся территория' and sl['index"quoted']=='Данные','SQL_UNIT_DIMENSION_ROUNDTRIP')

        # Municipal-level contract: a D05-like region row with a matching OKTMO
        # must still be excluded, even if its identifier happens to coincide.
        lev_fixture=fixture.iloc[:2].copy().drop(columns=['group','order','index"quoted'])
        lev_fixture['object_oktmo']=lev_fixture.pop('oktmo')
        lev_fixture['object_level']=['Регион','Муниципальный район']
        con.register('ml03_level_fixture_df',lev_fixture)
        level_path=WORK/'cache'/'ml03_level_contract.parquet'
        con.execute('COPY (SELECT * FROM ml03_level_fixture_df) TO '+qstr(str(level_path))+' (FORMAT PARQUET)')
        try:
            lev_asset={'asset_id':'02::labor::SYNTHETIC_LEVEL','columns_manifest':list(lev_fixture.columns)}
            lvi=inventory_asset(lev_asset,level_path,'G2')
            ensure(len(lvi)<=1,'LEVEL_INVENTORY_UNEXPECTED')
            raw_level,_,hasunit=factor_raw_sql(lev_asset)
            pp=[str(level_path),'Y48010001','Площадь жилищного фонда']+(['кв. м'] if hasunit else [])
            lv=sql_run('SELECT COUNT(*) AS n FROM ('+raw_level+') filtered',pp,asset_id=lev_asset['asset_id'],stage='OBJECT_LEVEL_UNIT',preview=True)
            ensure(int(lv[0])==1,'REGIONAL_OBJECT_ROW_LEAKED_IN_MUNICIPAL_JOIN')
        finally:
            con.unregister('ml03_level_fixture_df');level_path.unlink(missing_ok=True)
        ndfl=fixture.iloc[:1].copy().drop(columns=['group','order','index"quoted'])
        ndfl['indicator_code']='SYNTHETIC_NDFL_TOTAL';ndfl['indicator_name']='Общая сумма денежных доходов'
        ndfl['indicator_unit']='рублей';ndfl['income']='9999';ndfl['tax_rate']='13'
        sparse=pd.concat([ndfl.assign(income=str(10000+i)) for i in range(2101)],ignore_index=True)
        total=fixture.drop(columns=['group','order','index"quoted']).copy()
        total['indicator_code']='SYNTHETIC_NDFL_TOTAL';total['indicator_name']='Общая сумма денежных доходов'
        total['indicator_unit']='рублей';total['income']='всего';total['tax_rate']='всего'
        ndfl=pd.concat([sparse,total],ignore_index=True)
        con.register('ml03_ndfl_fixture_df',ndfl)
        con.execute('COPY (SELECT * FROM ml03_ndfl_fixture_df) TO '+qstr(str(tmp_ndfl))+' (FORMAT PARQUET)')
        toy_ndfl={'asset_id':'02::5ndfl::SYNTHETIC_SQL_CONTRACT','columns_manifest':list(ndfl.columns)}
        a,sl2,z=select_dimensional_slice(toy_ndfl,tmp_ndfl,'SYNTHETIC_NDFL_TOTAL','Общая сумма денежных доходов','рублей','G1')
        ensure(a is not None and len(a)>=100,'SQL_UNIT_NDFL_NO_TOTAL_SLICE')
        ensure(sl2=={'income':'всего','tax_rate':'всего'},'SQL_UNIT_NDFL_NARROW_SLICE_INCLUDED')
        record=DIMENSION_AUDIT[-1]
        ensure(record['slices_found']>2000 and record['semantically_eligible']==1 and record['inspected']==1 and record['remaining']==0,'SQL_UNIT_NDFL_LIMIT_ORDER')
        mark('SQL_DUCKDB_UNIT_CONTRACT','PASS',f'real DuckDB queries on Parquet; quoted columns; numeric parsing; >{record["slices_found"]} NDFL slices -> 1 eligible total')
        return {'status':'PASS','housing_rows':len(dat),'ndfl_total_rows':len(a),'ndfl_slices_found':record['slices_found'],'ndfl_semantically_eligible':1,'real_duckdb':True,'synthetic_only':True}
    finally:
        try:con.unregister('ml03_fixture_df');con.unregister('ml03_ndfl_fixture_df')
        except Exception:pass
        tmp.unlink(missing_ok=True);tmp_ndfl.unlink(missing_ok=True)
        del SQL_EXECUTION_AUDIT[audit0:];del DIMENSION_AUDIT[dim0:];del conflict_rows[conf0:]

SQL_DUCKDB_CONTRACT_REPORT=run_duckdb_sql_contracts()
REQUIRED_SQL_PREFLIGHT={'02::housing::00::00':['Y48010001','Y48011010'],'02::education::00::00':['Y48014001','Y48014003'],'02::budget::00::00':['Y48013001','Y48013002']}

def execute_real_sql_preflight():
    for aid,codes in REQUIRED_SQL_PREFLIGHT.items():
        if aid not in LOCAL_ASSETS:
            for code in codes:REAL_SQL_PREFLIGHT.append({'asset_id':aid,'code':code,'status':'NOT_DOWNLOADED','n_rows':0})
            continue
        a=get_asset(aid)
        for code in codes:
            try:
                row=sql_run('SELECT COUNT(*) FROM read_parquet(?) WHERE TRIM(CAST('+qident('indicator_code')+' AS VARCHAR))=?',[str(LOCAL_ASSETS[aid]),code],asset_id=aid,code=code,stage='REAL_PREFLIGHT_COUNT',preview=True)
                n=int(row[0]);REAL_SQL_PREFLIGHT.append({'asset_id':aid,'code':code,'status':'PASS' if n>0 else 'NO_CODE_ROWS','n_rows':n})
                raw,_,with_unit=factor_raw_sql(a)
                header_cols='indicator_name, indicator_unit' if 'indicator_unit' in a['columns_manifest'] else 'indicator_name'
                hdr=sql_run('SELECT '+header_cols+' FROM read_parquet(?) WHERE TRIM(CAST('+qident('indicator_code')+' AS VARCHAR))=? LIMIT 1',[str(LOCAL_ASSETS[aid]),code],asset_id=aid,code=code,stage='REAL_PREFLIGHT_HEADER')
                if len(hdr):
                    name=str(hdr.iloc[0]['indicator_name']).strip()
                    unit=str(hdr.iloc[0]['indicator_unit']).strip() if with_unit and pd.notna(hdr.iloc[0]['indicator_unit']) else ''
                    p=[str(LOCAL_ASSETS[aid]),code,name]+([unit] if with_unit else [])
                    observed=sql_run('SELECT COUNT(*) FROM ('+raw+') source_example',p,asset_id=aid,code=code,stage='REAL_PREFLIGHT_RAW_JOINED',preview=True)
                    joined=int(observed[0]);REAL_SQL_PREFLIGHT[-1]['n_joined']=joined
                    if joined==0:REAL_SQL_PREFLIGHT[-1]['status']='NO_MATCHED_MUNICIPAL_YEARS'
                    else:
                        dat,chosen_dims,validation=select_dimensional_slice(a,LOCAL_ASSETS[aid],code,name,unit,source_group(a))
                        if dat is None or dat.empty:
                            REAL_SQL_PREFLIGHT[-1]['status']='NO_VALIDATED_DIMENSIONAL_SLICE'
                            REAL_SQL_PREFLIGHT[-1]['slice_rejection_reason']=LAST_SLICE_REASON
                        elif dat.duplicated(['entity_id','observation_year']).any():
                            REAL_SQL_PREFLIGHT[-1]['status']='DUPLICATE_ENTITY_YEAR'
                        else:
                            REAL_SQL_PREFLIGHT[-1]['selected_keys']=int(len(dat))
                            REAL_SQL_PREFLIGHT[-1]['selected_nonmissing_values']=int(dat['value'].notna().sum())
                            REAL_SQL_PREFLIGHT[-1]['selected_dimensions_json']=json.dumps(chosen_dims,ensure_ascii=False)
                            if not dat['value'].notna().any():REAL_SQL_PREFLIGHT[-1]['status']='NO_NUMERIC_VALUES'
                else:REAL_SQL_PREFLIGHT[-1]['status']='NO_HEADER'
            except Exception as ex:
                REAL_SQL_PREFLIGHT.append({'asset_id':aid,'code':code,'status':'SCAN_ERROR','n_rows':0,'exception_type':type(ex).__name__,'message':str(ex)[:300]})
    return pd.DataFrame(REAL_SQL_PREFLIGHT)

REAL_SQL_PREFLIGHT_DF=execute_real_sql_preflight()
if CFG['MODE']=='FULL' and CFG['SQL_PREFLIGHT_REAL_CODES']:
    failed=REAL_SQL_PREFLIGHT_DF[~REAL_SQL_PREFLIGHT_DF.status.eq('PASS')]
    if len(failed):
        out=WORK/'ml03_sql_preflight_failures.json';out.write_text(json.dumps(REAL_SQL_PREFLIGHT,ensure_ascii=False,indent=2),encoding='utf-8')
        mark('REAL_STAGE1_SQL_PREFLIGHT','FAIL',f'Failed real SQL tests: {len(failed)}/6. Details: {out.name}')
        raise RuntimeError('FAIL_PIPELINE_SQL_PREFLIGHT: '+str(failed[['asset_id','code','status']].to_dict('records')))
mark('REAL_STAGE1_SQL_PREFLIGHT','PASS' if REAL_SQL_PREFLIGHT_DF.status.eq('PASS').all() else 'PARTIAL',f"real source code checks={len(REAL_SQL_PREFLIGHT_DF)}, passed={int(REAL_SQL_PREFLIGHT_DF.status.eq('PASS').sum())}; DuckDB fixture={SQL_DUCKDB_CONTRACT_REPORT['status']}")

progress('INVENTORY',0,len(FACTOR_ASSETS),'Аудит схем и индикаторов')
for _fi,a in enumerate(FACTOR_ASSETS,1):
    aid=a['asset_id'];groups=_grp_eligible_from_source(a)
    progress('INVENTORY',_fi,len(FACTOR_ASSETS),f'{_fi} из {len(FACTOR_ASSETS)} массивов')
    if aid not in LOCAL_ASSETS:
        SOURCE_QA.append({'asset_id':aid,'group':source_group(a),'status':'NOT_DOWNLOADED'});continue
    if not eligible_municipal_long_schema(a):
        SOURCE_QA.append({'asset_id':aid,'group':source_group(a),'status':'SCHEMA_NOT_ADAPTED','reason':'WIDE_OR_NON_MUNICIPAL_UNVALIDATED'});continue
    for g in groups:
        try:
            inv=inventory_asset(a,LOCAL_ASSETS[aid],g)
            SOURCE_QA.append({'asset_id':aid,'group':g,'status':'INVENTORIED_EXHAUSTIVE','n_candidates':len(inv),'cutoff_year':TRAIN_CATALOG_MAX_YEAR})
            if len(inv):
                inv['asset_id']=aid;inv['group']=g;CATALOG_ROWS.append(inv)
        except Exception as ex:
            SOURCE_QA.append({'asset_id':aid,'group':g,'status':'INVENTORY_ERROR','error_type':type(ex).__name__,'error_message':str(ex)[:1000]})
            log('INVENTORY_ERROR '+aid+': '+type(ex).__name__+' '+str(ex)[:400],'WARN')

progress('INVENTORY',1,1,'Инвентаризация завершена')
CATALOG=pd.concat(CATALOG_ROWS,ignore_index=True) if CATALOG_ROWS else pd.DataFrame()
if len(CATALOG):
    result=[classify_indicator(x['group'],x['asset_id'],x['indicator_name'],x['indicator_unit']) for x in CATALOG.to_dict('records')]
    CATALOG['domain']=[v[0] for v in result];CATALOG['semantic_rejection_reason']=[v[1] for v in result]
    for x in CATALOG.loc[CATALOG.domain.isna()].to_dict('records'):
        CANDIDATE_AUDIT.append({'asset_id':x['asset_id'],'code':x['indicator_code'],'name':x['indicator_name'],'decision':x['semantic_rejection_reason'],'slice':''})
    inspection_by_group=collections.Counter()
    _scan_total=int(CATALOG.loc[CATALOG.group.isin(('G1','G2','G3','G4'))&CATALOG.domain.notna()].shape[0])
    _scan_done=0
    progress('SCAN',0,_scan_total,'Размерные разрезы показателей')
    for g in ('G1','G2','G3','G4'):
        ranked=CATALOG.loc[CATALOG.group.eq(g)&CATALOG.domain.notna()].sort_values(['recent_keys','n_municipalities','n_years'],ascending=False)
        for idx,row in enumerate(ranked.to_dict('records')):
            _scan_done+=1
            if _scan_done%5==0 or _scan_done==_scan_total:
                progress('SCAN',_scan_done,_scan_total,f'{g}: {_scan_done}/{_scan_total}')
            if inspection_by_group[g]>=CFG['MAX_DIMENSIONAL_SCANS']:
                SCAN_AUDIT.append({'group':g,'asset_id':row['asset_id'],'code':row['indicator_code'],'status':'NOT_INSPECTED_CAP','reason':'RESOURCE_BUDGET','rank':idx})
                continue
            inspection_by_group[g]+=1;a=get_asset(row['asset_id'])
            try:
                dat,dims,support=select_dimensional_slice(a,LOCAL_ASSETS[a['asset_id']],row['indicator_code'],row['indicator_name'],row['indicator_unit'],g)
                reason='ELIGIBLE' if dat is not None else LAST_SLICE_REASON
                SCAN_AUDIT.append({'group':g,'asset_id':a['asset_id'],'code':row['indicator_code'],'status':reason,'rank':idx})
                if dat is None:continue
                feature_score=(int(support['n_recent_keys']),int(support['n_train_keys']))
                FEATURE_CANDIDATES.append({'group':g,'domain':row['domain'],'asset_id':a['asset_id'],'code':row['indicator_code'],'name':row['indicator_name'],'unit':row['indicator_unit'],'dims':dims,'support':support,'data':dat,'score':feature_score})
            except Exception as ex:
                SCAN_AUDIT.append({'group':g,'asset_id':a['asset_id'],'code':row['indicator_code'],'status':'SCAN_ERROR','reason':type(ex).__name__+': '+str(ex)[:400],'rank':idx})
                log('DIMENSION_SCAN '+a['asset_id']+': '+type(ex).__name__+' '+str(ex)[:400],'WARN')
    incomplete=any(r['status'] in ('NOT_INSPECTED_CAP','SCAN_ERROR') for r in SCAN_AUDIT)
else:incomplete=False
progress('SCAN',1,1,'Предметный отбор завершён')

SCAN_AUDIT=pd.DataFrame(SCAN_AUDIT);DIMENSION_AUDIT=pd.DataFrame(DIMENSION_AUDIT)
if len(CATALOG):
    g1=CATALOG.loc[CATALOG.group.eq('G1')].copy()
    if g1.empty:
        G1_DISCOVERY=pd.DataFrame([{'asset_id':a['asset_id'],'domain':'UNRESOLVED','discovered_codes':0,'discovered_rows':0,'reason_code':'UNADAPTED_MUNICIPAL_WIDE' if a['asset_id'] in UNADAPTED_MUNICIPAL_WIDE else 'NO_MATCHING_G1_INDICATORS_AT_VALIDATED_MUNICIPAL_GRAIN'} for a in ASSETS if a['asset_id']!='01::0'])
    g1['scan_key']=g1.asset_id.astype(str)+'|'+g1.indicator_code.astype(str)+'|'+g1.indicator_name.astype(str)
    insp=SCAN_AUDIT.loc[SCAN_AUDIT.group.eq('G1')].copy() if len(SCAN_AUDIT) else pd.DataFrame()
    if not g1.empty:
        G1_DISCOVERY=g1.groupby(['asset_id','domain'],dropna=False).agg(discovered_codes=('indicator_code','nunique'),discovered_rows=('indicator_name','size'),earliest_year=('min_year','min'),latest_year=('max_year','max'),max_municipalities=('n_municipalities','max')).reset_index()
    checked=insp.groupby('asset_id',dropna=False).status.value_counts().unstack(fill_value=0) if len(insp) else pd.DataFrame()
    if len(checked):
        for label in ('ELIGIBLE','NO_VALID_TOTAL_SLICE','NO_VALID_SLICE_AFTER_SUPPORT_CHECK','SCAN_ERROR','NOT_INSPECTED_CAP'):
            G1_DISCOVERY['n_'+label.lower()]=G1_DISCOVERY.asset_id.map(checked.get(label,pd.Series(dtype=int))).fillna(0).astype(int)
    G1_DISCOVERY['data_grain']='MUNICIPALITY_YEAR_EXACT_OKTMO_OR_NOT_VALIDATED'
    if not g1.empty:G1_DISCOVERY['reason_code']=np.where(G1_DISCOVERY.domain.isna(),'NO_SEMANTIC_MATCH','AUDIT_SLICES_IN_SCAN_AUDIT')
else:
    G1_DISCOVERY=pd.DataFrame([{'asset_id':a['asset_id'],'domain':None,'discovered_codes':0,'discovered_rows':0,'reason_code':'NO_MATCHED_MUNICIPAL_INDICATORS'} for a in ASSETS])

G1_EMPTY_CAUSE='NOT_FULLY_INSPECTED' if incomplete else 'UNADAPTED_MUNICIPAL_WIDE' if UNADAPTED_MUNICIPAL_WIDE else 'NO_VALIDATED_ECONOMIC_FEATURE_ON_STAGE_I'
DISCOVERY_EXHAUSTIVE=not incomplete and not UNADAPTED_MUNICIPAL_WIDE and (DIMENSION_AUDIT.empty or not DIMENSION_AUDIT.status.eq('CANDIDATE_BUDGET_PARTIAL').any()) and not any(r.get('status')=='INVENTORY_ERROR' for r in SOURCE_QA) and not SQL_FAILURES
mark('DISCOVERY_EXHAUSTIVENESS','PASS' if DISCOVERY_EXHAUSTIVE else 'PARTIAL',f'inspected={int(len(SCAN_AUDIT))}; unadapted_municipal_wide={len(UNADAPTED_MUNICIPAL_WIDE)}; incomplete={not DISCOVERY_EXHAUSTIVE}')

SOURCE_GRAIN_AUDIT=[]
for a in ASSETS:
    aid=a['asset_id'];records=[r for r in SOURCE_QA if r.get('asset_id')==aid]
    rawstage=next((r.get('status') for r in reversed(records) if r.get('status') in ('INVENTORY_ERROR','SCHEMA_NOT_ADAPTED','INVENTORIED_EXHAUSTIVE','NOT_DOWNLOADED')),None)
    stage=('NOT_DOWNLOADED' if aid not in LOCAL_ASSETS else 'TARGET_D01' if aid=='01::0' else 'SCANNED' if any(r.get('asset_id')==aid and r.get('status')=='ELIGIBLE' for r in SCAN_AUDIT.to_dict('records')) else 'SCAN_ERROR' if any(r.get('asset_id')==aid and r.get('status')=='SCAN_ERROR' for r in SCAN_AUDIT.to_dict('records')) else 'INVENTORIED_NO_MODEL_SLICE' if rawstage=='INVENTORIED_EXHAUSTIVE' else 'SOURCE_SCHEMA_ERROR' if rawstage=='INVENTORY_ERROR' else 'NOT_APPLICABLE_GRAIN' if not eligible_municipal_long_schema(a) else 'NOT_SEMANTICALLY_REVIEWED')
    SOURCE_GRAIN_AUDIT.append({'asset_id':aid,'downloaded_sha':aid in LOCAL_ASSETS,'columns_manifest_n':len(a.get('columns_manifest') or []),'source_grain':'MUNICIPAL_LONG_SCHEMA_AND_LEVEL_FILTER' if eligible_municipal_long_schema(a) and 'object_level' in (a.get('columns_manifest') or []) else 'MUNICIPAL_LONG_SCHEMA_NO_EXPLICIT_LEVEL' if eligible_municipal_long_schema(a) else 'NOT_VALIDATED_MUNICIPAL_LONG','stage':stage,'reason':rawstage or source_group(a)})

SOURCE_GRAIN_AUDIT=pd.DataFrame(SOURCE_GRAIN_AUDIT);SCAN_SQL_ERRORS=pd.DataFrame(SQL_FAILURES)
# The 52-source inventory is exhaustive at file/metadata level, not necessarily at validated content grain.
SOURCE_GRAIN_AUDIT['scientific_join_role']=SOURCE_GRAIN_AUDIT.stage.map(lambda s:'TARGET' if s=='TARGET_D01' else 'MUNICIPAL_LONG_CANDIDATE' if s in ('SCANNED','INVENTORIED_NO_MODEL_SLICE') else 'INDEX_ONLY_NOT_JOINED')
SCAN_SQL_AUDIT=pd.DataFrame(SQL_EXECUTION_AUDIT);REAL_SQL_PREFLIGHT_QA=pd.DataFrame(REAL_SQL_PREFLIGHT)
DATA_SMOKE_REPORT={'version':CFG['VERSION'],'asset_count':len(ASSETS),'scan_error_count':len(SQL_FAILURES),'inventory_errors':sum(r.get('status')=='INVENTORY_ERROR' for r in SOURCE_QA),'scanner_errors':int(SCAN_AUDIT.status.eq('SCAN_ERROR').sum()) if len(SCAN_AUDIT) else 0,'real_sql_preflight':REAL_SQL_PREFLIGHT,'dimension_audit':DIMENSION_AUDIT.to_dict('records') if len(DIMENSION_AUDIT) else []}
(WORK/'ml03_data_smoke.json').write_text(json.dumps(DATA_SMOKE_REPORT,ensure_ascii=False,indent=2,default=str),encoding='utf-8')
fatal_sql=len(SQL_FAILURES)>0 or DATA_SMOKE_REPORT['inventory_errors']>0 or DATA_SMOKE_REPORT['scanner_errors']>0
mark('DATA_SMOKE','FAIL' if fatal_sql else 'PASS',f"SQL runtime errors={len(SQL_FAILURES)}, inventory errors={DATA_SMOKE_REPORT['inventory_errors']}, scanner errors={DATA_SMOKE_REPORT['scanner_errors']}")
if fatal_sql and CFG['MODE']=='FULL' and CFG['FAIL_FAST_SQL']:
    raise RuntimeError('FAIL_PIPELINE_SQL_SCAN: see '+str(WORK/'ml03_data_smoke.json')+' and SCAN_SQL_ERRORS')

selected_by_group={g:[] for g in PATTERNS};SELECTION_BY_H=[]
for g in ('G1','G2','G3','G4'):
    selected_candidates={}
    for h in sorted(FACTOR_CUTOFF_BY_H):
        cutoff=int(FACTOR_CUTOFF_BY_H[h]);candidates=[]
        for c in FEATURE_CANDIDATES:
            if c['group']!=g:continue
            z=c['data'].loc[c['data'].observation_year.le(cutoff)&c['data']['value'].notna()]
            years=int(z.observation_year.nunique());num=int(len(z));mun=int(z.entity_id.nunique())
            recent=int(z.observation_year.ge(cutoff-CFG['CATALOG_RECENT_WINDOW']).sum())
            if num<CFG['MIN_INDICATOR_OBS_TRAIN'] or years<CFG['MIN_INDICATOR_YEARS_TRAIN'] or mun<CFG['MIN_INDICATOR_MUNICIPALITIES']:continue
            candidates.append({**c,'_h_score':(recent,num,mun),'_h_rows':num})
        candidates.sort(key=lambda c:c['_h_score'],reverse=True)
        source_counts=collections.Counter();domain_counts=collections.Counter();h_selected=[];pool=candidates.copy()
        for diversified in (True,False):
            while pool and len(h_selected)<CFG['MAX_FEATURES_PER_GROUP']:
                opts=[x for x in pool if source_counts[x['asset_id']]<CFG['MAX_FEATURES_FROM_ONE_ASSET'] and (not diversified or domain_counts[x['domain']]==0)]
                if not opts:break
                chosen=max(opts,key=lambda z:(z['_h_score'][0]/(1+source_counts[z['asset_id']]),z['_h_score'][1],-domain_counts[z['domain']]))
                pool.pop(next(i for i,x in enumerate(pool) if x is chosen))
                source_counts[chosen['asset_id']]+=1;domain_counts[chosen['domain']]+=1
                key=(chosen['asset_id'],str(chosen['code']),chosen['name'],chosen['unit'],json.dumps(chosen['dims'],ensure_ascii=False,sort_keys=True))
                if key not in selected_candidates:
                    feature=f'{g}_{len(selected_by_group[g])+1:02d}';data=chosen['data'].rename(columns={'value':feature})
                    ensure(not data.duplicated(['entity_id','observation_year']).any(),'FEATURE_UNIQUE_KEY_'+feature)
                    meta={'feature':feature,'group':g,'domain':chosen['domain'],'asset_id':chosen['asset_id'],'indicator_code':chosen['code'],'indicator_name':chosen['name'],'unit':chosen['unit'],'dimensions_json':key[-1],'observed_grain':'MUNICIPAL_OKTMO_YEAR_DIMENSION_SLICE','report_available_year':'NOT_VERIFIED','availability_provenance':'CONSERVATIVE_OBSERVATION_PLUS_LAG_NOT_OFFICIAL_PUBLICATION_DATE','effective_available_year_rule':f'observation_year+{CFG["FEATURE_PUBLICATION_LAG"]}','n_observed':int(data[feature].notna().sum()),'n_recent_train_keys':int(chosen['_h_score'][0]),'first_year':int(data.observation_year.min()),'last_year':int(data.observation_year.max()),'transform':'RAW_PRESERVED; UNIT_QA_REQUIRED','selected_horizons':[]}
                    selected_candidates[key]=meta;selected_by_group[g].append(meta);FEATURE_PARTS.append(data)
                meta=selected_candidates[key];meta['selected_horizons'].append(h);h_selected.append(meta['feature'])
            if not diversified:break
        SELECTION_BY_H.append({'group':g,'horizon':h,'max_observation_year':cutoff,'n_eligible_candidates':len(candidates),'n_selected':len(h_selected),'selected_features_json':json.dumps(h_selected,ensure_ascii=False),'domain_coverage_json':json.dumps(sorted(domain_counts),ensure_ascii=False),'status':'COMPLETE_SELECTION' if h_selected else 'NO_VALIDATED_CANDIDATES'})
    for meta in selected_by_group[g]:meta['selected_horizons_json']=json.dumps(sorted(set(meta.pop('selected_horizons'))))
SELECTION_BY_H=pd.DataFrame(SELECTION_BY_H)

from sklearn.neighbors import BallTree
SPATIAL_GEOMETRY_QA=[];SPATIAL_FEATURE_AUDIT=[];SPATIAL_FUNDAMENTAL_GATE='NOT_FEASIBLE'

def _is_municipal_grain(v):
    x=str(v).casefold().strip().replace('ё','е')
    return bool(re.search(r'(^municipality$|^municipal$|муниципальн.{0,15}(?:район|округ|образован)|городск.{0,8}округ)',x))

def _true_flag(v):return str(v).strip().casefold() in ('true','1','да','yes','истина')

def _spatial_snapshot_candidates(source,demography):
    audit={'n_source':int(len(source)),'n_matched_municipal':0,'n_verified':0,'n_distinct':0,'n_conflicting_codes':0,'n_ambiguous_regions':0,'status':'NOT_FEASIBLE'}
    required={'source_oktmo','object_level','region_name','latitude','longitude','coordinates_valid','census_year'}
    if not required.issubset(source.columns):
        audit['reason']='SPATIAL_COLUMN_CONTRACT_MISSING';return pd.DataFrame(),audit
    z=source[list(required)+(['population_original'] if 'population_original' in source else [])].copy()
    z=z.rename(columns={'source_oktmo':'entity_id','census_year':'snapshot_year'})
    z['entity_id']=z.entity_id.astype('string').str.strip()
    z=z[z.entity_id.notna()&z.entity_id.str.fullmatch(r'\d{5,11}',na=False)].copy()
    z=z[z.object_level.map(_is_municipal_grain)&z.coordinates_valid.map(_true_flag)].copy()
    for c in ['latitude','longitude','snapshot_year']:z[c]=pd.to_numeric(z[c],errors='coerce')
    z=z[z.latitude.between(-89.999,89.999)&z.longitude.between(-179.999,179.999)&z.snapshot_year.between(1990,max(YEARS))].copy()
    audit['n_matched_municipal']=int(len(z))
    if z.empty:
        audit['reason']='NO_VERIFIED_MUNICIPAL_GEOMETRY_OR_VINTAGE';return pd.DataFrame(),audit
    z['snapshot_year']=z.snapshot_year.astype(int)
    d=demography[['entity_id','region_name']].drop_duplicates()
    region_conflicts=d.groupby('entity_id').region_name.nunique(dropna=False)
    ambiguous=set(region_conflicts[region_conflicts.ne(1)].index);audit['n_ambiguous_regions']=len(ambiguous)
    d=d[~d.entity_id.isin(ambiguous)]
    z=z.merge(d,on='entity_id',how='inner',validate='many_to_one',suffixes=('_ref','_d01'))
    def norm_region(x):return re.sub(r'\s+',' ',str(x).casefold().replace('ё','е')).strip()
    z=z[z.region_name_ref.map(norm_region).eq(z.region_name_d01.map(norm_region))].copy()
    if z.empty:
        audit['reason']='NO_EXACT_REGION_AND_OKTMO_MATCH';return pd.DataFrame(),audit
    counts=z.groupby(['entity_id','snapshot_year']).agg(n_lat=('latitude','nunique'),n_lon=('longitude','nunique'))
    conflicted=set(counts.index[counts.n_lat.gt(1)|counts.n_lon.gt(1)].tolist())
    audit['n_conflicting_codes']=len(conflicted)
    if conflicted:z=z[~pd.MultiIndex.from_frame(z[['entity_id','snapshot_year']]).isin(conflicted)]
    z=z.drop_duplicates(['entity_id','snapshot_year','latitude','longitude']).sort_values(['entity_id','snapshot_year'])
    audit['n_verified']=len(z);audit['n_distinct']=z.entity_id.nunique()
    if z.entity_id.nunique()<max(30,int(CFG['MIN_INDICATOR_MUNICIPALITIES'])):
        audit['reason']='TOO_FEW_EXACT_VERIFIED_MUNICIPALITY_COORDS';return pd.DataFrame(),audit
    audit['status']='PASS';audit['reason']='TIME_STAMPED_VERIFIED_GEO_EXACT_MUNICIPAL_OKTMO_REGION'
    return z[['entity_id','snapshot_year','latitude','longitude','region_name_d01']],audit

def _snapshot_knn(z):
    out=[];earth=6371.0088
    for year in sorted(z.snapshot_year.unique()):
        available=z[z.snapshot_year.le(year)].sort_values('snapshot_year').drop_duplicates('entity_id',keep='last')
        this=z[z.snapshot_year.eq(year)].copy()
        if len(available)<5:continue
        latlon=np.radians(available[['latitude','longitude']].to_numpy(dtype=float))
        query=np.radians(this[['latitude','longitude']].to_numpy(dtype=float))
        tree=BallTree(latlon,metric='haversine');dist,idx=tree.query(query,k=min(5,len(available)))
        for i,v in enumerate(this.itertuples(index=False)):
            neighbours=[(float(dd)*earth,available.iloc[int(j)].entity_id) for dd,j in zip(dist[i],idx[i]) if available.iloc[int(j)].entity_id!=v.entity_id]
            if not neighbours:continue
            nearby=tree.query_radius(query[i:i+1],r=50/earth)[0]
            n_within_50=sum(available.iloc[int(j)].entity_id!=v.entity_id for j in nearby)
            out.append({'entity_id':v.entity_id,'snapshot_year':int(year),'nearest_municipality_km':neighbours[0][0],'municipalities_50km_n':int(n_within_50),'known_neighbour_universe_n':len(available)})
    return pd.DataFrame(out)

_SPATIAL_SOURCE_ID='06::D04::D04_member00::spatial_reference::derived'
if _SPATIAL_SOURCE_ID in LOCAL_ASSETS:
    try:
        spatial_source=pd.read_parquet(LOCAL_ASSETS[_SPATIAL_SOURCE_ID],columns=['source_oktmo','object_level','region_name','latitude','longitude','coordinates_valid','census_year'])
        geo,qa=_spatial_snapshot_candidates(spatial_source,base)
        SPATIAL_GEOMETRY_QA.append({'asset_id':_SPATIAL_SOURCE_ID,**qa})
        if not geo.empty:
            near=_snapshot_knn(geo)
            if not near.empty:
                pubyear=CFG.get('SPATIAL_PUBLICATION_YEAR')
                pubref=str(CFG.get('SPATIAL_PUBLICATION_EVIDENCE','')).strip()
                if pubyear is None or not pubref or not CFG.get('SPATIAL_BOUNDARY_CROSSWALK_VERIFIED',False) or not str(CFG.get('SPATIAL_BOUNDARY_CROSSWALK_EVIDENCE','')).strip():
                    SPATIAL_GEOMETRY_QA.append({'asset_id':_SPATIAL_SOURCE_ID,'status':'NOT_FEASIBLE','reason':'PUBLICATION_YEAR_OR_BOUNDARY_CROSSWALK_UNVERIFIED','n_verified_geometry':int(len(near))})
                    near=pd.DataFrame()
            if not near.empty and int(pubyear)>max(YEARS):
                SPATIAL_GEOMETRY_QA.append({'asset_id':_SPATIAL_SOURCE_ID,'status':'NOT_FEASIBLE','reason':'VERIFIED_PUBLICATION_AFTER_LAST_TARGET_YEAR'})
                near=pd.DataFrame()
            if not near.empty:
                yrs=pd.DataFrame({'observation_year':YEARS})
                snapshots=near.loc[near.snapshot_year.le(max(YEARS))].copy().merge(yrs,how='cross')
                snapshots=snapshots.loc[snapshots.observation_year.ge(np.maximum(snapshots.snapshot_year,int(pubyear)))]
                snapshots=snapshots.sort_values('snapshot_year').drop_duplicates(['entity_id','observation_year'],keep='last')
                for f,un,domain,col in (('G5_01_km','км','spatial_distance','nearest_municipality_km'),('G5_02_neighbors','число муниципалитетов','spatial_density','municipalities_50km_n')):
                    data=snapshots[['entity_id','observation_year',col]].rename(columns={col:f})
                    data[f]=pd.to_numeric(data[f],errors='coerce')
                    ensure(not data.duplicated(['entity_id','observation_year']).any(),'SPATIAL_TIME_KEY_DUPLICATE')
                    FEATURE_PARTS.append(data)
                    selected_by_group['G5'].append({'feature':f,'group':'G5','domain':domain,'asset_id':_SPATIAL_SOURCE_ID,'indicator_code':'D04_VERIFIED_GEO_'+domain.upper(),'indicator_name':'Подтверждённое пространственное расстояние' if domain=='spatial_distance' else 'Муниципалитеты в радиусе 50 км','unit':un,'dimensions_json':'{}','observed_grain':'MUNICIPAL_GEOMETRY_VINTAGE','report_available_year':f'max(snapshot_year,{pubyear})+publication_lag','effective_available_year_rule':f'max(census_year,{pubyear})+{CFG["FEATURE_PUBLICATION_LAG"]}<=origin_year','n_observed':int(data[f].notna().sum()),'n_recent_train_keys':int(data[data.observation_year.le(TRAIN_CATALOG_MAX_YEAR)][f].notna().sum()),'first_year':int(data.observation_year.min()),'last_year':int(data.observation_year.max()),'transform':'SOURCE_VINTAGE_GEOMETRY; NO FUTURE_COORDINATES'})
                    SPATIAL_FEATURE_AUDIT.append({'feature':f,'n_rows':len(data),'n_municipalities':data.entity_id.nunique(),'snapshot_min':int(near.snapshot_year.min()),'snapshot_max':int(near.snapshot_year.max()),'model_eligible':'PRELIMINARY_TIMING_GATE'})
                SPATIAL_FUNDAMENTAL_GATE='PASS'
    except Exception as ex:
        SPATIAL_GEOMETRY_QA.append({'asset_id':_SPATIAL_SOURCE_ID,'status':'NOT_FEASIBLE','reason':'SPATIAL_REFERENCE_REJECTED_'+type(ex).__name__})
        log('SPATIAL_DATA_GATE: '+type(ex).__name__+': '+str(ex)[:250],'WARN')
else:
    SPATIAL_GEOMETRY_QA.append({'asset_id':_SPATIAL_SOURCE_ID,'status':'NOT_FEASIBLE','reason':'SPATIAL_VINTAGE_FILE_NOT_DOWNLOADED'})

SPATIAL_GEOMETRY_QA.append({'asset_id':'06::TRANSPORT::TRANSPORT_member00::full','status':'NOT_FEASIBLE','reason':'TRANSPORT_OBSERVATION_YEAR_NOT_PROVIDED'})
SPATIAL_FEATURE_AUDIT=pd.DataFrame(SPATIAL_FEATURE_AUDIT);SPATIAL_GEOMETRY_QA=pd.DataFrame(SPATIAL_GEOMETRY_QA)
G5_DIAGNOSTICS.extend(SPATIAL_GEOMETRY_QA.to_dict('records'))

# V3.00: independent region-level economic CONTEXT. Never claim these are
# municipality-specific wages/employment. Strict object-level and exact region/year.
REGIONAL_CONTEXT_PARTS=[];REGIONAL_CONTEXT_META=[];REGIONAL_CONTEXT_AUDIT=[]
D05_ECONOMIC_SOURCES=('02::labor::','02::socio1::','02::socio2::','02::socio3::','02::social::')
def _region_norm(s):return re.sub(r'\s+',' ',str(s).casefold().replace('ё','е')).strip()
def _region_context_collect():
    if not CFG.get('ENABLE_REGIONAL_ECONOMIC_CONTEXT'):return
    for asset in ASSETS:
        aid=asset['asset_id']
        if aid not in LOCAL_ASSETS or not aid.startswith(D05_ECONOMIC_SOURCES):continue
        cols=set(asset.get('columns_manifest') or [])
        if not {'object_level','object_name','indicator_name','indicator_value','year','indicator_code','indicator_unit'}.issubset(cols):continue
        sql="""SELECT CAST(object_name AS VARCHAR) region, TRY_CAST(year AS INTEGER) yr,
             CAST(indicator_code AS VARCHAR) code, CAST(indicator_name AS VARCHAR) title,
             CAST(indicator_unit AS VARCHAR) unit, CAST(indicator_value AS VARCHAR) val,
             CAST(object_level AS VARCHAR) object_level
             FROM read_parquet(?) WHERE TRY_CAST(year AS INTEGER) BETWEEN ? AND ?
             AND regexp_matches(lower(coalesce(cast(indicator_name AS VARCHAR),'')),?)"""
        try:
            x=sql_run(sql,[str(LOCAL_ASSETS[aid]),min(YEARS),max(YEARS),PATTERNS['G1']],asset_id=aid,stage='REGION_CONTEXT_DISCOVERY')
        except Exception as exc:
            REGIONAL_CONTEXT_AUDIT.append({'asset_id':aid,'status':'SQL_FAILURE','reason':type(exc).__name__});continue
        if x.empty:
            REGIONAL_CONTEXT_AUDIT.append({'asset_id':aid,'status':'NO_G1_CANDIDATES'});continue
        lev=x.object_level.astype('string').str.casefold().fillna('')
        x=x[lev.str.contains(r'регион|субъект',regex=True,na=False)].copy()
        if x.empty:
            REGIONAL_CONTEXT_AUDIT.append({'asset_id':aid,'status':'NO_VALID_REGION_ROWS'});continue
        x['region_key']=x.region.map(_region_norm);x['yr']=pd.to_numeric(x.yr,errors='coerce')
        x=x[x.yr.notna()].copy();x['yr']=x.yr.astype(int)
        regions={_region_norm(v) for v in base.region_name.dropna().unique()}
        x=x[x.region_key.isin(regions)].copy()
        if x.empty:
            REGIONAL_CONTEXT_AUDIT.append({'asset_id':aid,'status':'REGION_NAMES_NOT_EXACT'});continue
        x['value']=pd.to_numeric(x.val.astype('string').str.replace(' ','',regex=False).str.replace(',','.',regex=False),errors='coerce')
        x=x[x.value.notna()];candidates=0
        for (code,title,unit),s in x.groupby(['code','title','unit'],dropna=False,sort=False):
            domain,_=classify_indicator('G1',aid,title,unit)
            if domain is None:continue
            # Only intensive regional quantities: an explicit rate, percentage,
            # average wage or average per-capita income; no regional totals.
            text=(str(title)+' '+str(unit)).casefold()
            intensive=bool(re.search(r'средн|на душу|на одного|в расчете на|уровень безработ|коэффициент|процент|%|дол[яи]',text))
            if not intensive:continue
            key=['region_key','yr'];span=s.groupby(key)['value'].agg(['size','nunique','first']).reset_index()
            clashes=int(span['nunique'].gt(1).sum())
            if clashes:
                REGIONAL_CONTEXT_AUDIT.append({'asset_id':aid,'code':str(code),'status':'CONFLICTING_REGION_YEAR','n_conflicts':clashes});continue
            z=span[['region_key','yr','first']].rename(columns={'yr':'observation_year'})
            before=z[z.observation_year.le(TRAIN_CATALOG_MAX_YEAR)]
            if before.region_key.nunique()<CFG['REGIONAL_ECONOMIC_MIN_REGIONS'] or before.observation_year.nunique()<2:continue
            f=f'G1R_{len(REGIONAL_CONTEXT_META)+1:02d}'
            z=z.rename(columns={'first':f})
            REGIONAL_CONTEXT_META.append({'feature':f,'group':'G1R','domain':domain,
                 'source_grain':'REGION_YEAR_ONLY','source_id':aid,'indicator_code':str(code),
                 'indicator_name':str(title),'unit':str(unit),'n_regions':int(z.region_key.nunique()),
                 'n_years':int(z.observation_year.nunique()),'n_conflicts':clashes,
                 'note':'NOT_MUNICIPAL_G1; no region-to-municipality economic reconstruction'})
            REGIONAL_CONTEXT_PARTS.append(z)
            candidates+=1
            if len(REGIONAL_CONTEXT_META)>=CFG['REGIONAL_ECONOMIC_MAX_FEATURES']:break
        REGIONAL_CONTEXT_AUDIT.append({'asset_id':aid,'status':'AUDITED_REGION_CONTEXT',
            'candidate_rows':int(len(x)),'accepted_features_from_source':candidates})
        if len(REGIONAL_CONTEXT_META)>=CFG['REGIONAL_ECONOMIC_MAX_FEATURES']:break
_region_context_collect()
REGIONAL_CONTEXT_AUDIT=pd.DataFrame(REGIONAL_CONTEXT_AUDIT)
REGIONAL_CONTEXT_METADATA=pd.DataFrame(REGIONAL_CONTEXT_META)
mark('G1_REGIONAL_CONTEXT_DISCOVERY','PASS' if REGIONAL_CONTEXT_META else 'NOT_FEASIBLE',
     f'{len(REGIONAL_CONTEXT_META)} exact-match region-year economic indicators, strictly separate from municipal G1')

FEATURE_DICT=pd.DataFrame(sum(selected_by_group.values(),[]))
SOURCE_QA=pd.DataFrame(SOURCE_QA);CONFLICT_QA=pd.DataFrame(conflict_rows)
SPATIAL_SOURCE_QA=pd.DataFrame(G5_DIAGNOSTICS);SEMANTIC_QA=pd.DataFrame(CANDIDATE_AUDIT)
progress('FACTORS',1,1,'Срезы и факторы рассчитаны')
DOMAIN_COVERAGE=pd.DataFrame([{'group':g,'domain':d,'n_selected':sum(x['domain']==d for x in selected_by_group[g])} for g in SEMANTIC_DOMAINS for d,_ in SEMANTIC_DOMAINS[g]])
for g in PATTERNS:
    domains={r['domain'] for r in selected_by_group[g]}
    required=2 if g=='G3' else 1
    status='PASS' if len(domains)>=required else 'PARTIAL' if domains else 'NOT_FEASIBLE'
    if g=='G1' and not DISCOVERY_EXHAUSTIVE:status='PARTIAL'
    mark('FEATURE_'+g,status,f'{len(selected_by_group[g])} valid indicators; domains={sorted(domains)}')
```

## 5. Формирование панели и контроль научных единиц измерения

Сохраняются исходные значения в отдельных столбцах. Производные признаки создаются только при установленной размерности, с датированной историей коэффициентов перевода. Неизвестная неизменность муниципальных границ маркируется как неопределённость: отсутствие записи о переименовании не является подтверждением территориальной сопоставимости. Эксперимент по-прежнему использует информацию, доступную до года прогноза.


```python
# 06 · V3.01 · RESEARCH PANEL — VERIFIED G1R, UNITS, TIME AND LEVELS
PANEL=base[['entity_id','observation_year','region_name','mun_type','municipality','population','average_population','migration','migration_rate','births','deaths','migration_per_1000','boundary_uncertainty','source_vintage']].copy()
merge_audit=[]
for data in FEATURE_PARTS:
    f=data.columns[-1]; n=len(PANEL)
    ensure(not data.duplicated(['entity_id','observation_year']).any(),'SOURCE_JOIN_MANY_TO_MANY_'+f)
    PANEL=PANEL.merge(data,on=['entity_id','observation_year'],how='left',validate='one_to_one')
    ensure(len(PANEL)==n,'PANEL_CARDINALITY_CHANGED_'+f)
    merge_audit.append(dict(feature=f,before_rows=n,after_rows=len(PANEL),nonmissing=int(PANEL[f].notna().sum()),status='ONE_TO_ONE'))
PANEL=PANEL.sort_values(['entity_id','observation_year']).reset_index(drop=True); QA_KEYS=pd.DataFrame(merge_audit)

HISTORY_ID='06::BDMO6::BDMO6_member00::municipality_history_observed::derived'
PANEL['history_observed']=False; PANEL['observed_name_change']=False; PANEL['observed_type_change']=False; PANEL['boundary_change_not_verified']=True
if HISTORY_ID in LOCAL_ASSETS:
    try:
        hh=pd.read_parquet(LOCAL_ASSETS[HISTORY_ID],columns=['source_oktmo','observation_year','observed_name_change','observed_type_change','boundary_change_not_verified'])
        hh['entity_id']=hh.source_oktmo.astype('string').str.strip().astype(str); hh['observation_year']=pd.to_numeric(hh.observation_year,errors='coerce')
        hh=hh[hh.observation_year.notna()].copy(); hh['observation_year']=hh.observation_year.astype(int)
        flags=['observed_name_change','observed_type_change','boundary_change_not_verified']
        for c in flags: hh[c]=hh[c].astype('string').str.lower().isin(['true','1','да','yes'])
        hh=hh.groupby(['entity_id','observation_year'],as_index=False)[flags].max(); hh['history_observed']=True
        PANEL=PANEL.drop(columns=['history_observed']+flags).merge(hh,on=['entity_id','observation_year'],how='left',validate='one_to_one')
        for c in ['history_observed','observed_name_change','observed_type_change']: PANEL[c]=PANEL[c].astype('boolean').fillna(False).astype(bool)
        PANEL['boundary_change_not_verified']=PANEL['boundary_change_not_verified'].astype('boolean').fillna(True).astype(bool)
    except Exception as exc:
        PANEL['history_observed']=False; PANEL['observed_name_change']=False; PANEL['observed_type_change']=False; PANEL['boundary_change_not_verified']=True
        mark('HISTORY_BOUNDARY_QA','PARTIAL','Audit failed: '+type(exc).__name__)
PANEL['boundary_year_at_risk']=~PANEL.history_observed | PANEL.boundary_change_not_verified | PANEL.observed_name_change | PANEL.observed_type_change
PANEL['boundary_evidence_status']=np.select([~PANEL.history_observed,PANEL.observed_name_change|PANEL.observed_type_change,PANEL.boundary_change_not_verified],['NO_HISTORY_RECORD','OBSERVED_NAME_OR_TYPE_CHANGE','BOUNDARY_NOT_VERIFIED'],default='NO_OBSERVED_CHANGE_NOT_PROOF_OF_STABILITY')
BOUNDARY_QA=pd.DataFrame([{'metric':c,'n_yearly_observations_flagged':int(PANEL[c].sum())} for c in ['history_observed','boundary_change_not_verified','boundary_year_at_risk','observed_name_change','observed_type_change']])
mark('PASS_BOUNDARY_AUDIT','PASS',f'Unverified entity-years={int(PANEL.boundary_year_at_risk.sum())}/{len(PANEL)}; NEVER coerced to stable')

def mark_boundary_forecast_window(t,reference):
    z=reference[['entity_id','observation_year','boundary_year_at_risk']].sort_values(['entity_id','observation_year']).copy()
    z['cum_obs']=z.groupby('entity_id').cumcount()+1; z['cum_bad']=z.boundary_year_at_risk.astype(int).groupby(z.entity_id).cumsum()
    a=z.rename(columns={'observation_year':'origin_year','cum_obs':'start_obs','cum_bad':'start_bad'}); b=z.rename(columns={'observation_year':'target_year','cum_obs':'end_obs','cum_bad':'end_bad'})
    out=t.merge(a[['entity_id','origin_year','start_obs','start_bad']],on=['entity_id','origin_year'],how='left',validate='many_to_one')
    out=out.merge(b[['entity_id','target_year','end_obs','end_bad']],on=['entity_id','target_year'],how='left',validate='many_to_one')
    ensure(len(out)==len(t),'BOUNDARY_JOIN_CARDINALITY_CHANGE')
    miss=out[['start_obs','start_bad','end_obs','end_bad']].isna().any(axis=1)
    out['boundary_period_uncertain']=(miss | (out.end_obs-out.start_obs).ne(out.horizon) | (out.end_bad-out.start_bad).gt(0)).astype('boolean').fillna(True).astype(bool)
    return out.drop(columns=['start_obs','start_bad','end_obs','end_bad'])

RATIO_REGEX=r'на\s*(?:одного|одну|одно|1\b|1000\b|10\s*тыс|душу|жител|человек|учащ|предприяти)|в\s+расч[её]те\s+на|в\s+среднем\s+на|удельн|на\s+единиц'
RATE_REGEX=r'(?:%|процент|коэффициент|индекс|дол[яи]\s|отношение|промилле)'
def unit_policy(indicator_name,unit,group,year=2020):
    title=str(indicator_name).casefold().replace('ё','е'); raw=str(unit).casefold().replace('ё','е'); both=title+' '+raw
    m=re.search(r'с\s*(\d{4})\s*г?\.?\s*[-–—:, ]*',raw); eff=(raw[m.end():] if year>=int(m.group(1)) else raw[:m.start()]) if m and re.search(r'рубл|руб',raw) else raw
    rub=bool(re.search(r'рубл|руб\b',eff)); scale=1e6 if re.search(r'млн\.?\s*руб|миллион.{0,4}руб',eff) else 1e3 if re.search(r'тыс\.?\s*руб|тысяч.{0,5}руб',eff) else 1.
    def result(action,mult=1.,den='NONE',family='UNCLASSIFIED',status='PASS'):
        return dict(action=action,multiplier=mult,denominator=den,effective_unit=eff,unit_family=family,status=status)
    if group=='G5' and re.search(r'км|километр|число муниципалитетов|муниципалитет',raw): return result('KEEP_SPATIAL_MEASUREMENT',family='SPATIAL_REFERENCE')
    wage=bool(re.search(r'(средн.{0,20}(?:заработн|начисленн.{0,9}зарплат)|заработн.{0,9}плат.{0,8}одн.{0,9}работник)',both))
    if wage and rub:return result('KEEP_RAW_AVERAGE',scale,family='AVERAGE_WAGE_PER_WORKER')
    if re.search(RATIO_REGEX,both):return result('KEEP_ALREADY_NORMALIZED',scale,family='RATIO_OR_PER_ENTITY')
    if re.search(RATE_REGEX,both):return result('KEEP_RATE',family='RATE')
    if rub:
        action='PER_RESIDENT' if group in ['G1','G2','G3','G4'] else 'KEEP_RAW'
        return result(action,scale,'D01_population_same_year' if action=='PER_RESIDENT' else 'NONE','MONEY_NOMINAL_RUB')
    if re.search(r'человек|чел\b|единиц|ед\.|штук|штука|кв\.?\s*м|м2|квадрат|метр',eff) and group in ['G1','G2','G3']:
        if group=='G1' and re.search(r'средн|уровень|размер',title):return result('KEEP_RAW_AVERAGE',family='AVERAGE')
        return result('PER_1000_RESIDENTS',1000.,'D01_population_same_year','COUNT_OR_AREA')
    return result('KEEP_RAW_UNCLASSIFIED',family='UNCLASSIFIED',status='REVIEW')

ensure(unit_policy('Расходы ОМСУ на одного жителя','2008 г. — тысяча рублей, с 2009 г. — рубль','G4',2020)['action']=='KEEP_ALREADY_NORMALIZED','UNIT_CANARY_DOUBLE_DENOMINATOR')
ensure(unit_policy('Расходы местного бюджета','2008 г. — тыс. руб.; с 2009 г. — рубли','G4',2020)['multiplier']==1,'UNIT_CANARY_HISTORICAL_RUBLES')
ensure(unit_policy('Средняя начисленная заработная плата','тыс. рублей','G1',2020)['multiplier']==1000,'UNIT_CANARY_AVERAGE_WAGE_THOUSANDS')
ensure(unit_policy('Средняя начисленная заработная плата','рублей','G1',2020)['action']=='KEEP_RAW_AVERAGE','UNIT_CANARY_WAGE_NOT_PER_RESIDENT')
ensure(unit_policy('Расходы на одно малое предприятие','рублей','G4',2020)['denominator']=='NONE','UNIT_CANARY_PER_ENTERPRISE')
mark('UNIT_SEMANTICS_CANARY','PASS','No double ratio, mixed-year rubles correctly resolved')


# V3.00: add region-level G1R context under an explicit separate namespace.
# Exact region_key joins only; no organizational aggregation or apportionment to municipalities.
PANEL['region_key']=PANEL.region_name.map(_region_norm)
for part in REGIONAL_CONTEXT_PARTS:
    rows_before=len(PANEL)
    PANEL=PANEL.merge(part,on=['region_key','observation_year'],how='left',validate='many_to_one')
    ensure(len(PANEL)==rows_before,'REGIONAL_CONTEXT_ROW_MULTIPLICATION')
GROUP_FEATURES={g:[r['feature'] for r in selected_by_group[g]] for g in PATTERNS}
GROUP_FEATURES['G1R']=[r['feature'] for r in REGIONAL_CONTEXT_META]
# Regional quantities are not normalized by local municipal population.
# Unit conversion happens in the actual source observation year; noncomparable
# units are removed instead of filling unknown levels with guessed municipal values.
G1R_UNIT_QA=[]
for meta in REGIONAL_CONTEXT_META:
    feature=meta['feature'];unit=meta['unit'];name=meta['indicator_name']
    policies={int(y):unit_policy(name,unit,'G1R',int(y)) for y in PANEL.observation_year.unique()}
    accepted=all(p['status']=='PASS' for p in policies.values()) and len({p['action'] for p in policies.values()})==1
    action=next(iter(policies.values()))['action']
    # A regional total can never masquerade as a municipal income rate.
    accepted=bool(accepted and action in ('KEEP_RAW_AVERAGE','KEEP_ALREADY_NORMALIZED','KEEP_RATE'))
    if accepted:
        scale=PANEL.observation_year.map({y:p['multiplier'] for y,p in policies.items()})
        PANEL[feature]=pd.to_numeric(PANEL[feature],errors='coerce')*scale
    else:
        GROUP_FEATURES['G1R']=[c for c in GROUP_FEATURES['G1R'] if c!=feature]
    G1R_UNIT_QA.append({'feature':feature,'source_id':meta['source_id'],'unit':unit,
        'action':action,'decision':'ACCEPT_REGIONAL_CONTEXT' if accepted else 'EXCLUDE_REGIONAL_UNIT_AMBIGUITY',
        'source_grain':'REGION_YEAR_ONLY','n_values':int(PANEL[feature].notna().sum())})
G1R_UNIT_QA=pd.DataFrame(G1R_UNIT_QA)
audit=[]
for row in FEATURE_DICT.to_dict('records'):
    f,g,unit,title=row['feature'],row['group'],row['unit'],row['indicator_name']
    years=pd.to_numeric(PANEL.observation_year,errors='coerce'); policies={int(y):unit_policy(title,unit,g,int(y)) for y in years.dropna().unique()}
    pol=policies[int(years.max())]; ok={p['status'] for p in policies.values()}=={'PASS'} and len({p['action'] for p in policies.values()})==1
    suffix='_per_resident' if pol['action']=='PER_RESIDENT' else '_per_1000' if pol['action']=='PER_1000_RESIDENTS' else '_scaled_rub' if any(p['multiplier']!=1 for p in policies.values()) else ''
    derived=f+suffix
    if ok and suffix:
        value=pd.to_numeric(PANEL[f],errors='coerce')*years.map({y:p['multiplier'] for y,p in policies.items()})
        if pol['action'] in ['PER_RESIDENT','PER_1000_RESIDENTS']: value=value/pd.to_numeric(PANEL.population,errors='coerce').where(PANEL.population.gt(0))
        PANEL[derived]=value
    if ok: GROUP_FEATURES[g]=[derived if x==f else x for x in GROUP_FEATURES[g]]
    else: GROUP_FEATURES[g]=[x for x in GROUP_FEATURES[g] if x!=f]
    examples={int(y):float(pd.to_numeric(PANEL.loc[PANEL.observation_year.eq(y),f],errors='coerce').dropna().iloc[0]) for y in sorted(policies) if PANEL.loc[PANEL.observation_year.eq(y),f].notna().any()}
    audit.append(dict(parent_feature=f,model_feature=derived if ok else None,group=g,domain=row['domain'],indicator_name=title,asset_id=row['asset_id'],action=pol['action'],ratio_denominator=pol['denominator'],source_units=unit,effective_unit_recent=pol['effective_unit'],effective_unit_by_year_json=json.dumps({y:p['effective_unit'] for y,p in policies.items()},ensure_ascii=False),multiplier_by_year_json=json.dumps({y:p['multiplier'] for y,p in policies.items()}),yearwise_raw_value_exemplar=json.dumps(examples),unit_family=pol['unit_family'],decision='ACCEPT_DIMENSIONAL_CONVERSION' if ok else 'EXCLUDE_UNIT_AMBIGUITY',notes='Raw field preserved; per-resident conversion only where explicitly required.'))
TRANSFORM_QA=pd.DataFrame(audit,columns=['parent_feature','model_feature','group','domain','indicator_name','asset_id','action','ratio_denominator','source_units','effective_unit_recent','effective_unit_by_year_json','multiplier_by_year_json','yearwise_raw_value_exemplar','unit_family','decision','notes'])
for c in ['feature','group','domain','model_feature','unit_status']:
    if c not in FEATURE_DICT.columns: FEATURE_DICT[c]=pd.Series(dtype='object')
if len(FEATURE_DICT):
    lookup=TRANSFORM_QA.set_index('parent_feature'); FEATURE_DICT['model_feature']=FEATURE_DICT.feature.map(lookup.model_feature); FEATURE_DICT['unit_status']=FEATURE_DICT.feature.map(lookup.decision)
MODEL_DOMAIN_COVERAGE=pd.DataFrame([dict(group=g,domain=d,n_model_features=int(((TRANSFORM_QA.group==g)&(TRANSFORM_QA.domain==d)&TRANSFORM_QA.model_feature.notna()).sum())) for g in SEMANTIC_DOMAINS for d,_ in SEMANTIC_DOMAINS[g]])
for g in ['G1','G2','G3','G4']:
    valid=TRANSFORM_QA.loc[(TRANSFORM_QA.group==g)&TRANSFORM_QA.model_feature.notna(),'domain'].dropna().unique().tolist(); threshold=2 if g=='G3' else 1
    mark('FEATURE_'+g,'PASS' if len(valid)>=threshold else 'PARTIAL' if valid else 'NOT_FEASIBLE',f'Usable domains after strict unit QA: {sorted(valid)}; features={len(GROUP_FEATURES[g])}')
actual=set(MODEL_DOMAIN_COVERAGE.loc[(MODEL_DOMAIN_COVERAGE.group=='G1')&MODEL_DOMAIN_COVERAGE.n_model_features.gt(0),'domain'])
# Distinguish nominal domains from valid labour/income proxies; do not call sales 'employment'.
G1_DOMAIN_QA=pd.DataFrame([{'domain':d,'n_model_features':int(n),'adequate_for_claim':int(n)>0,
  'claim':'DESCRIPTIVE_ONLY' if d in ('business','retail','investment') else 'REQUIRES_SEMANTICALLY_VALID_DIRECT_INDICATOR'}
  for d,n in zip(MODEL_DOMAIN_COVERAGE.loc[MODEL_DOMAIN_COVERAGE.group.eq('G1'),'domain'],
                 MODEL_DOMAIN_COVERAGE.loc[MODEL_DOMAIN_COVERAGE.group.eq('G1'),'n_model_features'])])
econ_core=bool(actual & {'wages','employment','income'})
mark('G1_DOMAIN_COMPLETENESS','PASS' if len(actual)>=CFG['MIN_ECON_DOMAINS'] and econ_core else 'PARTIAL' if actual else 'NOT_FEASIBLE',
   f'observed={sorted(actual)}; direct wages/employment/income present={econ_core}; sales not employment')
HEALTH_DOMAIN_QA=pd.DataFrame([{
   'feature':r.model_feature,'name':r.indicator_name,'domain':r.domain,
   'hospital_specific':bool(re.search(r'больниц|больнич|стационар|койк|поликлиник|амбулатор',str(r.indicator_name).casefold())),
   'source_grain':'VALIDATED_MUNICIPAL' if pd.notna(r.model_feature) else 'NOT_SELECTED',
   'n_observed':int(PANEL[r.parent_feature].notna().sum()) if r.parent_feature in PANEL else 0}
    for r in TRANSFORM_QA.loc[(TRANSFORM_QA.group=='G3')&(TRANSFORM_QA.domain=='health')].itertuples()])
health=int(((HEALTH_DOMAIN_QA.hospital_specific)&(HEALTH_DOMAIN_QA.source_grain=='VALIDATED_MUNICIPAL')).sum()) if len(HEALTH_DOMAIN_QA) else 0
HEALTH_AVAILABLE=bool(health)
mark('HEALTH_INFRASTRUCTURE_GATE','PASS' if HEALTH_AVAILABLE else 'PARTIAL' if len(HEALTH_DOMAIN_QA) else 'NOT_FEASIBLE',
    f'Hospital/ambulatory indicators selected={health}, broad health candidates={len(HEALTH_DOMAIN_QA)}; horizon checks below')
mark('PASS_UNIT_AUDIT','PASS' if not TRANSFORM_QA.decision.eq('EXCLUDE_UNIT_AMBIGUITY').any() else 'PARTIAL',f"Excluded ambiguous units={int(TRANSFORM_QA.decision.eq('EXCLUDE_UNIT_AMBIGUITY').sum())}")

FACTOR_ELIGIBILITY=[]; GROUP_FEATURES_BY_H={}
for h,cutoff in FACTOR_CUTOFF_BY_H.items():
    GROUP_FEATURES_BY_H[h]={}
    for g,ff in GROUP_FEATURES.items():
        eligible=[]
        for f in ff:
            parent=next((r['parent_feature'] for r in audit if r['model_feature']==f),f)
            valid=PANEL.loc[PANEL.observation_year.le(cutoff)&PANEL[f].notna(),['entity_id','observation_year']]
            n,ny,nm=len(valid),valid.observation_year.nunique(),valid.entity_id.nunique()
            meta=next((r for r in selected_by_group.get(g,[]) if r['feature']==parent),None)
            selected=True if g in ('G5','G1R') else bool(meta is not None and h in json.loads(meta.get('selected_horizons_json','[]')))
            ok=bool(selected and n>=CFG['MIN_INDICATOR_OBS_TRAIN'] and ny>=CFG['MIN_INDICATOR_YEARS_TRAIN'] and nm>=CFG['MIN_INDICATOR_MUNICIPALITIES'])
            if ok:eligible.append(f)
            FACTOR_ELIGIBILITY.append(dict(horizon=h,group=g,feature=f,parent_feature=parent,observation_cutoff_year=cutoff,source_published_year='NOT_VERIFIED',effective_available_year_rule=f'observation_year+{CFG["FEATURE_PUBLICATION_LAG"]}',n_keys_before_cutoff=n,n_years_before_cutoff=int(ny),n_municipalities_before_cutoff=int(nm),eligible=ok,reason='TRAIN_SUPPORT_VERIFIED' if ok else 'INSUFFICIENT_PRE_ORIGIN_SUPPORT'))
        GROUP_FEATURES_BY_H[h][g]=eligible
FACTOR_ELIGIBILITY=pd.DataFrame(FACTOR_ELIGIBILITY,columns=['horizon','group','feature','parent_feature','observation_cutoff_year','source_published_year','effective_available_year_rule','n_keys_before_cutoff','n_years_before_cutoff','n_municipalities_before_cutoff','eligible','reason'])

ORIGINS=PANEL[['entity_id','region_name','mun_type','municipality','observation_year','boundary_uncertainty','source_vintage']].rename(columns={'observation_year':'origin_year'}).copy()
prior=PANEL[['entity_id','observation_year','population','migration_per_1000']].rename(columns={'observation_year':'known_year'})
lag0=CFG['FEATURE_PUBLICATION_LAG']
for lag in range(lag0,lag0+4):
    tmp=prior.rename(columns={'population':f'hist_population_l{lag}','migration_per_1000':f'hist_migration_l{lag}'})
    ORIGINS['known_year']=ORIGINS.origin_year-lag; ORIGINS=ORIGINS.merge(tmp,on=['entity_id','known_year'],how='left',validate='many_to_one').drop(columns='known_year')
G0_FEATURES=[c for c in ORIGINS if c.startswith('hist_')]+['origin_year']
ORIGINS['hist_growth_lagged_pct']=100*(ORIGINS[f'hist_population_l{lag0}']/ORIGINS[f'hist_population_l{lag0+2}']-1)
ORIGINS['hist_growth_lagged_pct']=ORIGINS.hist_growth_lagged_pct.replace([np.inf,-np.inf],np.nan); G0_FEATURES.append('hist_growth_lagged_pct')
all_factors=list(dict.fromkeys(sum(GROUP_FEATURES.values(),[])))
if all_factors:
    raw=PANEL[['entity_id','observation_year']+all_factors].copy()
    # last observed value carries forward at most MAX_FEATURE_STALENESS_YEARS,
    # with a separate age ledger and with the required publication lag.
    STALENESS_QA=[]
    for f in all_factors:
        observed=pd.to_numeric(raw[f],errors='coerce')
        lastyear=raw['observation_year'].where(observed.notna()).groupby(raw.entity_id).ffill()
        lag_age=raw.observation_year-lastyear
        stale=lag_age.gt(CFG['MAX_FEATURE_STALENESS_YEARS']) | lastyear.isna()
        filled=observed.groupby(raw.entity_id).ffill()
        raw[f]=filled.where(~stale)
        STALENESS_QA.append({'feature':f,'n_observed':int(observed.notna().sum()),
             'n_carried_valid':int((raw[f].notna()&observed.isna()).sum()),
             'max_age_years':CFG['MAX_FEATURE_STALENESS_YEARS'],'publication_lag':lag0})
    STALENESS_QA=pd.DataFrame(STALENESS_QA)
    ensure(STALENESS_QA.n_carried_valid.ge(0).all(),'STALE_QA_INVALID')
    raw=raw.rename(columns={'observation_year':'known_year'})
    ORIGINS['known_year']=ORIGINS.origin_year-lag0
    ORIGINS=ORIGINS.merge(raw,on=['entity_id','known_year'],how='left',validate='many_to_one').drop(columns='known_year')
else:STALENESS_QA=pd.DataFrame(columns=['feature','n_observed','n_carried_valid','max_age_years','publication_lag'])
LABELS=PANEL[['entity_id','observation_year','population','migration_per_1000']].rename(columns={'observation_year':'target_year','population':'future_population','migration_per_1000':'future_migration_per_1000'})
CURRENT=PANEL[['entity_id','observation_year','population']].rename(columns={'observation_year':'origin_year','population':'population_at_origin'})
ORIGINS=ORIGINS.merge(CURRENT,on=['entity_id','origin_year'],how='left',validate='one_to_one'); frames=[]
for h in CFG['HORIZONS']:
    t=ORIGINS.copy(); t['horizon']=h; t['target_year']=t.origin_year+h
    t=t.merge(LABELS,on=['entity_id','target_year'],how='left',validate='many_to_one')
    t['migration_per_1000']=t.future_migration_per_1000
    t['population_change_pct']=np.where(t.population_at_origin.gt(0)&t.future_population.gt(0),100*(t.future_population/t.population_at_origin-1),np.nan)
    t['information_cutoff']=t.origin_year; t['effective_available_year']=t.origin_year-lag0
    t['target_origin_unit']='ENTITY_YEAR'; t['source_vintage']=source_sha
    frames.append(mark_boundary_forecast_window(t,PANEL).drop(columns=['future_population','future_migration_per_1000']))
TASK_PANEL=pd.concat(frames,ignore_index=True)
ensure(not TASK_PANEL.duplicated(['entity_id','origin_year','horizon']).any(),'FORECAST_KEY_COLLISION')
ensure((TASK_PANEL.effective_available_year<TASK_PANEL.origin_year).all(),'FUTURE_COVARIATE_LEAK')
if CFG['MODE']=='SMOKE': TASK_PANEL=TASK_PANEL[TASK_PANEL.entity_id.isin(sorted(TASK_PANEL.entity_id.unique())[:CFG['SMOKE_MUNICIPALITIES']])].copy()

TARGET_AVAILABILITY=[]
for target in CFG['TARGETS']:
    for y in YEARS:
        src=base.loc[base.observation_year.eq(y)]
        valid_raw=(src.migration.notna()&src.average_population.gt(0)) if target=='migration_per_1000' else src.population.gt(0)
        for h in CFG['HORIZONS']:
            tp=TASK_PANEL.loc[TASK_PANEL.horizon.eq(h)&TASK_PANEL.target_year.eq(y)]
            available=tp[target].notna(); g0=tp[[c for c in G0_FEATURES if c!='origin_year']].notna().any(axis=1)
            g1=tp[GROUP_FEATURES['G1']].notna().any(axis=1) if GROUP_FEATURES['G1'] else pd.Series(False,index=tp.index)
            TARGET_AVAILABILITY.append(dict(target=target,target_year=int(y),horizon=int(h),raw_d01_rows=len(src),raw_d01_target_available=int(valid_raw.sum()),label_rows=int(available.sum()),g0_ready=int((available&g0).sum()),factor_rows_g1=int((available&g1).sum()),reason='D01_MIGRATION_UNOBSERVED' if target=='migration_per_1000' and not valid_raw.any() else 'OBSERVED_OR_OTHER_SUPPORT'))
TARGET_AVAILABILITY=pd.DataFrame(TARGET_AVAILABILITY)
GROUP_SUPPORT={g:TASK_PANEL[ff].notna().any(axis=1) if ff else pd.Series(False,index=TASK_PANEL.index) for g,ff in GROUP_FEATURES.items()}
G0_SUPPORT=TASK_PANEL[[c for c in G0_FEATURES if c!='origin_year']].notna().any(axis=1)
TASK_PANEL['common_support']=G0_SUPPORT.copy()
for g in ['G1','G2','G3','G4','G5']:
    if GROUP_FEATURES[g]: TASK_PANEL['common_support'] &= GROUP_SUPPORT[g]
TASK_PANEL['available_factor_groups']=sum(v.astype(int) for v in GROUP_SUPPORT.values())
coverage=pd.DataFrame([dict(group=g,n_features=len(ff),raw_panel_coverage=float(PANEL[ff].notna().any(axis=1).mean()) if ff else 0.,forecast_coverage=float(GROUP_SUPPORT[g].mean()),status='PRESENT' if ff else 'NOT_FEASIBLE') for g,ff in GROUP_FEATURES.items()])
mark('G1R_NONMUNICIPAL_GRAIN','PASS','Regional economic context is labeled G1R, NOT municipal G1; zero municipal claims')
mark('PASS_DATA',('PASS' if sum(map(len,GROUP_FEATURES.values()))>0 and not fatal_sql else 'PARTIAL' if len(PANEL) and not PANEL[['entity_id','observation_year']].duplicated().any() else 'FAIL'),f'panel={len(PANEL)}, features={sum(map(len,GROUP_FEATURES.values()))}, common_support={int(TASK_PANEL.common_support.sum())}')
quiet_print('Панель:',len(PANEL),'строк | label candidates:',len(TASK_PANEL),'| complete support:',int(TASK_PANEL.common_support.sum()))
def audited_stability_share(sample):
    if sample.empty:return {'n':0,'n_unverified':0,'n_verified':0,'verified_share':0.}
    flag=sample.boundary_period_uncertain.astype('boolean').fillna(True).astype(bool)
    return dict(n=len(sample),n_unverified=int(flag.sum()),n_verified=int((~flag).sum()),verified_share=float((~flag).mean()))
BOUNDARY_WINDOW_QA=TASK_PANEL.groupby(['horizon','target_year'],as_index=False).agg(n=('entity_id','size'),uncertain=('boundary_period_uncertain','sum'))

progress('PANEL',1,1,f'Наблюдений: {len(PANEL)}; прогнозных строк: {len(TASK_PANEL)}')
```

## 6. Регистрация эксперимента: временные окна и пространственная переносимость (этап II)

Каноническая последовательность M0–M5 сохраняется, но G5 допускается в M5 только при доказанной доступности исходной геометрии на **каждый целевой год двухлетнего периода**. Недатированная транспортная сеть не превращается в историческое наблюдение. Помимо общего rolling-time проверяются отдельно два календарных эксперимента `YEAR_2022` и `YEAR_2023`, если для каждого соблюдён собственный информационный cutoff. Отсутствующие годы не заменяются более ранними. Пространственные, временные и смешанные holdout раздельны; связанные территории остаются внутри региона. Это проверки прогностической переносимости, а не причинных воздействий.


**v2.3:** основной holdout 2022–2023, вторичный отдельный миграционный observed-window, `FEASIBLE_PARTIAL_YEARS` отличается от полного; историческая географическая проверка не называется тестом 2022–2023.

```python
# 07 · PREREGISTERED ABLATIONS, COMMON-SUPPORT STUDIES, SPATIAL/TEMPORAL SPLITS
GROUP_ORDER=('G1','G2','G3','G4','G5')
PRIMARY_TEST_YEARS=sorted(set(sum(TARGET_TEST_WINDOWS.values(),[])))
SPATIAL_YEAR_COUNTS={int(y):int((TASK_PANEL.horizon.eq(1)&TASK_PANEL.target_year.eq(y)&
    GROUP_SUPPORT['G5']).sum()) for y in PRIMARY_TEST_YEARS}
G5_CANONICAL_ELIGIBLE=bool(GROUP_FEATURES.get('G5')) and all(
    SPATIAL_YEAR_COUNTS[y]>=CFG['MIN_SPATIAL_CANONICAL_TEST_ROWS'] for y in PRIMARY_TEST_YEARS)
mark('SPATIAL_CANONICAL_GATE','PASS' if G5_CANONICAL_ELIGIBLE else 'NOT_FEASIBLE',
     f'vintage-gated support={SPATIAL_YEAR_COUNTS}; undated transport excluded')
CANONICAL=[{'arm':'M0','groups':[]}];prefix=[]
for j,g in enumerate(GROUP_ORDER,1):
    if not GROUP_FEATURES[g] or (g=='G5' and not G5_CANONICAL_ELIGIBLE):
        for k in range(j,6):mark('ARM_M'+str(k),'NOT_FEASIBLE','Missing or time-ineligible group '+g)
        break
    prefix=prefix+[g];CANONICAL.append({'arm':'M'+str(j),'groups':prefix.copy()})
AVAILABLE_GROUPS=[g for g in GROUP_ORDER if GROUP_FEATURES[g] and
                  (g!='G5' or G5_CANONICAL_ELIGIBLE)]
STUDIES=[]
if len(CANONICAL)>1:
    STUDIES.append({'id':'CANONICAL','required_groups':prefix.copy(),'arms':CANONICAL,
        'protocols':['ROLLING_TIME','TERRITORY_ONLY','CROSSED'],'confirmatory':len(CANONICAL)>=5})
else:
    STUDIES.append({'id':'CANONICAL','required_groups':[],'arms':CANONICAL,
        'protocols':['ROLLING_TIME'],'confirmatory':False})
# Separate, self-contained temporal experiments for the two nominal test years.
if CFG['MODE']=='FULL' and len(CANONICAL)>1:
    for target,years in TARGET_TEST_WINDOWS.items():
        for year in years:
            STUDIES.append({'id':f'YEAR_{target}_{year}','test_year':int(year),
               'allowed_targets':[target],'required_groups':prefix.copy(),'arms':CANONICAL,
               'protocols':['ROLLING_TIME'],'confirmatory':False})
for g in AVAILABLE_GROUPS:
    STUDIES.append({'id':'AUX_'+g,'required_groups':[g],
      'arms':[{'arm':'M0','groups':[]},{'arm':'ADD_'+g,'groups':[g]}],
      'protocols':['ROLLING_TIME'],'confirmatory':False})
if GROUP_FEATURES.get('G5') and not G5_CANONICAL_ELIGIBLE:
    STUDIES.append({'id':'AUX_G5_VINTAGED','required_groups':['G5'],
      'arms':[{'arm':'M0','groups':[]},{'arm':'ADD_G5','groups':['G5']}],
      'protocols':['ROLLING_TIME'],'confirmatory':False})
# Secondary observed migration window only if it differs from the predeclared migration calendar.
if CFG['MODE']=='FULL' and CFG.get('EXPLORE_SECONDARY_MIGRATION_WINDOW'):
    mig_years=sorted(base.loc[base.migration_per_1000.notna(),'observation_year'].unique())
    secondary=mig_years[-CFG['TEST_YEARS']:] if len(mig_years)>=CFG['TEST_YEARS'] else []
    if secondary and list(map(int,secondary))!=TARGET_TEST_WINDOWS['migration_per_1000']:
        STUDIES.append({'id':'MIGRATION_OBSERVED_LAST_TWO_YEARS','test_years':list(map(int,secondary)),
          'allowed_targets':['migration_per_1000'],'required_groups':[],
          'arms':[{'arm':'M0','groups':[]}], 'protocols':['ROLLING_TIME'],
          'confirmatory':False,'label':'SECONDARY_OBSERVED_WINDOW_NOT_PRIMARY'})
if GROUP_FEATURES.get('G1R'):
    STUDIES.append({'id':'AUX_G1_REGION_CONTEXT','required_groups':['G1R'],
       'arms':[{'arm':'M0','groups':[]},{'arm':'ADD_G1_REGION','groups':['G1R']}],
       'protocols':['ROLLING_TIME','TERRITORY_ONLY','CROSSED'],
       'confirmatory':False,'label':'REGION_CONTEXT_NOT_MUNICIPAL_G1'})
if len(AVAILABLE_GROUPS)>=2:
    all_g=AVAILABLE_GROUPS.copy()
    specs=[{'arm':'M0','groups':[]},{'arm':'FULL_AVAILABLE','groups':all_g}]
    specs += [{'arm':'WITHOUT_'+g,'groups':[x for x in all_g if x!=g]} for g in all_g]
    STUDIES.append({'id':'MAX_AVAILABLE','required_groups':all_g,'arms':specs,
       'protocols':['ROLLING_TIME','TERRITORY_ONLY','CROSSED'],'confirmatory':False})
for study in STUDIES:
    for arm in study['arms']:
        arm['features']=list(dict.fromkeys(G0_FEATURES+sum((GROUP_FEATURES[g] for g in arm['groups']),[])))
        arm['n_factors']=sum(len(GROUP_FEATURES[g]) for g in arm['groups'])
# ПРЕРЕГИСТРАЦИЯ до обучения: H3.1–H3.8 из научного протокола, без просмотра TEST.
_HYPOTHESES=[
 ('H3.1','CANONICAL','M1 → M3','Последовательный вклад жилищной и социальной инфраструктуры'),
 ('H3.2','INTERACTION','G1 × G2','Экономические условия × жилищная среда'),
 ('H3.3','AUX_G3','M0 → ADD_G3','Дополнительная ценность социальной инфраструктуры'),
 ('H3.4','AUX_G4','M0 → ADD_G4','Дополнительная ценность муниципальных финансов'),
 ('H3.5','SENSITIVITY','LAG_AND_PROXY','Лаги публикации и коррелированные прокси'),
 ('H3.6','HEALTH','HEALTH_VS_EDUCATION','Независимая идентифицируемость медицинских факторов'),
 ('H3.7','BOUNDARIES','STABLE_VS_UNVERIFIED','Устойчивость к изменениям муниципальных границ'),
 ('H3.8','SIZE','SIZE_STRATA','Неоднородность эффекта по размеру муниципалитетов')]
HYPOTHESIS_REGISTRY=pd.DataFrame([{
    'hypothesis_id':h,'study_id':study,'contrast':contrast,'definition':definition,
    'test_family':('CONFIRMATORY_BY_TARGET_HORIZON_PROTOCOL' if h=='H3.1' else
                   'EXPLORATORY_BY_TARGET_HORIZON_PROTOCOL'),
    'sample_contract':'PAIRED_IDENTICAL_ENTITY_ORIGIN_TARGET_YEAR',
    'primary_metric':'MAE','multiplicity_method':('HOLM' if h=='H3.1' else 'BH_FDR'),
    'preregistered_before_test':True,'status':'PLANNED_CONDITIONAL',
    'version':CFG['VERSION'],
    'reason':'Requires verified source domain, sufficient paired support and temporal labels'}
    for h,study,contrast,definition in _HYPOTHESES])
STUDY_HYPOTHESIS_FAMILIES=pd.DataFrame([{
    'study_id':s['id'],'confirmatory_design':bool(s['confirmatory']),
    'family_scope':'TARGET_HORIZON_PROTOCOL',
    'multiplicity_method':'HOLM' if s['confirmatory'] else 'BH_FDR',
    'preregistered_before_test':True,'status':'DESIGN_FROZEN'} for s in STUDIES])
TASKS=[];TASK_RECORDS=[]
ALL_REGIONS=sorted(base.region_name.dropna().astype(str).unique())
rng=np.random.default_rng(CFG['SEED']);HELD_REGIONS=set(rng.choice(
   np.asarray(ALL_REGIONS,dtype=object),size=max(2,int(0.2*len(ALL_REGIONS))),replace=False)) if len(ALL_REGIONS)>8 else set()

def splits_for(d,protocol,h,test_year=None,test_years=None):
    # Real-time forecasting: all training AND DEV outcome labels must have been observed
    # before the earliest issuance (origin year) of the held-out test forecasts.
    yrs=[y for y in YEARS if y>=min(YEARS)+h]
    nt=1 if CFG['MODE']=='SMOKE' else CFG['TEST_YEARS']
    nd=1 if CFG['MODE']=='SMOKE' else CFG['DEV_YEARS']
    if len(yrs)<nt+nd+CFG['MIN_TRAIN_YEARS']:return None,'TOO_FEW_TARGET_YEARS'
    test_yrs=([int(test_year)] if test_year is not None else
              sorted(int(v) for v in test_years) if test_years else yrs[-nt:])
    first_test_origin=test_yrs[0]-h
    latest_known_label=first_test_origin-int(CFG['TARGET_LABEL_AVAILABILITY_LAG'])
    potential_dev=[y for y in yrs if y<=latest_known_label]
    if len(potential_dev)<CFG['MIN_TRAIN_YEARS']+nd:
        return None,'LONG_HORIZON_TARGET_LABELS_NOT_YET_KNOWN_AT_FORECAST_ORIGIN'
    dv=potential_dev[-nd:]
    tr=[y for y in potential_dev if y<min(dv)]
    if len(tr)<CFG['MIN_TRAIN_YEARS']:return None,'TOO_FEW_REALTIME_TRAIN_YEARS'
    held=d.region_name.astype(str).isin(HELD_REGIONS)
    if protocol=='ROLLING_TIME':
        return {'train':d.target_year.isin(tr),'dev':d.target_year.isin(dv),
                'test':d.target_year.isin(test_yrs)},''
    if not HELD_REGIONS:return None,'INSUFFICIENT_REGIONS_FOR_GROUP_HOLDOUT'
    if protocol=='TERRITORY_ONLY':
        # Pure geography generalization (historical evaluation on held regions).
        return {'train':(~held)&d.target_year.isin(tr),
                'dev':(~held)&d.target_year.isin(dv),
                'test':held&d.target_year.isin(dv)},''
    if protocol=='CROSSED':
        return {'train':(~held)&d.target_year.isin(tr),
                'dev':(~held)&d.target_year.isin(dv),
                'test':held&d.target_year.isin(test_yrs)},''
    raise ValueError('UNKNOWN_SPLIT_PROTOCOL')

# A missing late test year must NOT move the test window backwards.
_test_stub=pd.DataFrame({'target_year':[y for y in YEARS if y<max(YEARS)-2],
                         'region_name':['unit_test']*len([y for y in YEARS if y<max(YEARS)-2])})
_stub_masks,_=splits_for(_test_stub,'ROLLING_TIME',min(CFG['HORIZONS']))
if _stub_masks is not None:
    ensure(int(_stub_masks['test'].sum())==0,'UNIT_TEST_CALENDAR_SLIPPAGE')
mark('UNIT_SPLIT_CONTRACT','PASS','D01-frozen target years and issued-forecast cutoff; source gaps do not shift test years')

for st in STUDIES:
    for target in CFG['TARGETS']:
        if target not in st.get('allowed_targets',CFG['TARGETS']):continue
        if target=='migration_per_1000' and not valid_rate:
            for h in (CFG['HORIZONS'] if CFG['MODE']=='FULL' else [1]):
                for protocol in st['protocols']:
                    TASK_RECORDS.append({'study_id':st['id'],'target':target,'horizon':h,
                      'protocol':protocol,'status':'NOT_FEASIBLE','scientific_status':'NOT_FEASIBLE',
                      'reason':'MIGRATION_TARGET_SEMANTICS_FAILED','n_support':0,'n_train':0,'n_dev':0,'n_test':0,
                      'requested_test_years':','.join(map(str,TARGET_TEST_WINDOWS[target])),
                      'actual_test_year_counts_json':'{}','missing_test_years':','.join(map(str,TARGET_TEST_WINDOWS[target]))})
            continue
        for h in (CFG['HORIZONS'] if CFG['MODE']=='FULL' else [1]):
            support=G0_SUPPORT.copy()
            missing_required=[g for g in st['required_groups'] if not GROUP_FEATURES_BY_H[h].get(g)]
            for g in st['required_groups']:
                if GROUP_FEATURES_BY_H[h].get(g):
                    support &= TASK_PANEL[GROUP_FEATURES_BY_H[h][g]].notna().any(axis=1)
            d=TASK_PANEL[TASK_PANEL.horizon.eq(h)&TASK_PANEL[target].notna()&support].copy()
            local_st=dict(st)
            local_st['arms']=[{**a,'features':list(dict.fromkeys(
                G0_FEATURES+sum((GROUP_FEATURES_BY_H[h][g] for g in a['groups']),[]))),
                'n_factors':sum(len(GROUP_FEATURES_BY_H[h][g]) for g in a['groups'])}
               for a in st['arms']]
            for protocol in st['protocols']:
                mm,issue=splits_for(d,protocol,h,st.get('test_year'),st.get('test_years',TARGET_TEST_WINDOWS[target]))
                if missing_required:issue='GROUP_NOT_ELIGIBLE_AT_HORIZON_'+','.join(missing_required);mm=None
                counts={k:int(v.sum()) for k,v in mm.items()} if mm is not None else {}
                if mm is not None and (counts['train']<CFG['MIN_TRAIN_ROWS'] or
                      min(counts['dev'],counts['test'])<CFG['MIN_TEST_ROWS']):
                    issue='INSUFFICIENT_TRAIN_DEV_TEST';mm=None
                requested=([int(st['test_year'])] if 'test_year' in st else
                           st['test_years'] if 'test_years' in st else
                           TARGET_TEST_WINDOWS[target])
                test_by_year=(d.loc[mm['test']].groupby('target_year').size().to_dict()
                              if mm is not None else {})
                missing_test_years=[y for y in requested if test_by_year.get(y,0)==0] if protocol!='TERRITORY_ONLY' else []
                task_status=('FEASIBLE_COMPLETE' if mm is not None and not missing_test_years else
                             'FEASIBLE_PARTIAL_YEARS' if mm is not None else 'NOT_FEASIBLE')
                TASK_RECORDS.append({'study_id':st['id'],'target':target,'horizon':h,'protocol':protocol,
                  'status':task_status,'scientific_status':('COMPLETE_YEARS' if task_status=='FEASIBLE_COMPLETE' else 'PARTIAL_YEARS' if task_status=='FEASIBLE_PARTIAL_YEARS' else 'NOT_FEASIBLE'),
                  'reason':issue,'n_support':len(d),
                  'n_train':counts.get('train'),'n_dev':counts.get('dev'),'n_test':counts.get('test'),
                  'requested_test_years':','.join(map(str,requested)),
                  'actual_test_year_counts_json':json.dumps({int(k):int(v) for k,v in test_by_year.items()}),
                  'missing_test_years':','.join(map(str,missing_test_years)),
                  'evaluation_window':('HISTORICAL_DEV_YEARS_GEO_ONLY' if protocol=='TERRITORY_ONLY' else
                     'PRIMARY_TARGET_YEARS' if 'test_years' not in st else 'SECONDARY_OBSERVED_YEARS')})
                if mm is not None:
                    TASKS.append({'study':local_st,'target':target,'horizon':h,'protocol':protocol,
                                 'data':d,'masks':mm})
TASK_STATUSES=pd.DataFrame(TASK_RECORDS)
ensure({'status','scientific_status','study_id','target','horizon','protocol','n_test'}.
       issubset(TASK_STATUSES.columns),'TASK_STATUS_SCHEMA_INCOMPLETE')
ensure((TASK_STATUSES.loc[TASK_STATUSES.status.eq('FEASIBLE_PARTIAL_YEARS'),
       'scientific_status']=='PARTIAL_YEARS').all(),'INCOMPLETE_TEST_YEARS_MISLABELLED')

def _study_target_years(st,target):
    if 'test_years' in st:return list(map(int,st['test_years']))
    if 'test_year' in st:return [int(st['test_year'])]
    return TARGET_TEST_WINDOWS[target]

# Audit target availability per study with explicit source/target coverage.
TEST_YEAR_AUDIT_ROWS=[]
for rec in TASK_RECORDS:
    if rec['protocol']=='TERRITORY_ONLY':continue  # historical geo-only protocol, not the primary time calendar
    st=next(s for s in STUDIES if s['id']==rec['study_id'])
    required=list(st['required_groups'])
    for year in _study_target_years(st,rec['target']):
        mask=TASK_PANEL.target_year.eq(year)&TASK_PANEL.horizon.eq(rec['horizon'])&TASK_PANEL[rec['target']].notna()
        supported=mask&G0_SUPPORT
        for g in required:
            if GROUP_FEATURES_BY_H[int(rec['horizon'])].get(g):
                supported=supported&TASK_PANEL[GROUP_FEATURES_BY_H[int(rec['horizon'])][g]].notna().any(axis=1)
            else:supported=supported&False
        by_year=json.loads(rec.get('actual_test_year_counts_json','{}') or '{}')
        TEST_YEAR_AUDIT_ROWS.append({'study_id':rec['study_id'],'target':rec['target'],
            'horizon':rec['horizon'],'protocol':rec['protocol'],'target_year':int(year),
            'n_target_observations':int(mask.sum()),'n_with_study_factor_support':int(supported.sum()),
            'n_test':int(by_year.get(str(year),0)),
            'status':rec['status'],'reason':rec['reason']})
TEST_YEAR_AUDIT=pd.DataFrame(TEST_YEAR_AUDIT_ROWS)
CANONICAL_YEAR_QA=TASK_STATUSES.loc[TASK_STATUSES.study_id.eq('CANONICAL') &
    TASK_STATUSES.protocol.isin(['ROLLING_TIME','CROSSED'])].copy()
if CFG['MODE']=='FULL':
    primary=CANONICAL_YEAR_QA.loc[CANONICAL_YEAR_QA.horizon.eq(1) &
        CANONICAL_YEAR_QA.status.str.startswith('FEASIBLE')]
    missing_any=primary.missing_test_years.ne('').any() if len(primary) else True
    mark('TWO_YEAR_CALENDAR_QA','PARTIAL' if missing_any else 'PASS',
         'Two-year target-specific observed primary windows: migration 2021–2022, population 2022–2023; no silent sliding')
else: mark('TWO_YEAR_CALENDAR_QA','SMOKE_ONLY','Year coverage in FULL checked separately')
H5_CALENDAR_GAPS=TASK_STATUSES.loc[TASK_STATUSES.horizon.eq(5),
    ['study_id','target','protocol','horizon','status','reason','requested_test_years','actual_test_year_counts_json']].copy()
if len(H5_CALENDAR_GAPS):
    mark('H5_LONG_HORIZON','PASS' if H5_CALENDAR_GAPS.status.eq('FEASIBLE_COMPLETE').all() else 'NOT_FEASIBLE',
         f'{int(H5_CALENDAR_GAPS.status.eq("FEASIBLE_COMPLETE").sum())}/{len(H5_CALENDAR_GAPS)}; unmet label availability is never bypassed')
# A historical time-window redefinition is NOT substituted for preregistered h5 tests.
ARM_SPECS=CANONICAL
if not TASKS:mark('PASS_SPLITS','NOT_FEASIBLE','No feasible task after paired common-support gating')
else:mark('PASS_SPLITS','PASS',f'{len(TASKS)} dataset/protocol tasks, {len(STUDIES)} studies')
# One fixed work budget before the first model; never grow denominator during run.
RUN_COMPLETION_STATUS='IN_PROGRESS'
_TICK_PHASE=None;_TICK_DONE=0;_TICK_TOTAL=0
def set_tick(phase,total):
    global _TICK_PHASE,_TICK_TOTAL,_TICK_DONE
    _TICK_PHASE=phase;_TICK_TOTAL=max(0,int(total));_TICK_DONE=0
    if _TICK_TOTAL==0:progress(phase,1,1,'Нет доступных единиц работы')
    else:progress(phase,0,_TICK_TOTAL,'Выполнение')
def tick(stage,n=1,detail=''):
    global _TICK_DONE
    ensure(_TICK_PHASE is not None,'PROGRESS_PHASE_NOT_SET')
    _TICK_DONE+=int(n)
    ensure(_TICK_DONE<=_TICK_TOTAL,'PROGRESS_WORK_COUNT_OVERFLOW')
    progress(_TICK_PHASE,_TICK_DONE,_TICK_TOTAL,detail)

quiet_print('Studies:',[(s['id'],len(s['arms'])) for s in STUDIES], '| feasible tasks:',len(TASKS),
      '| frozen regions held out:',len(HELD_REGIONS))

progress('TASKS',1,1,f'Задач в плане: {len(TASKS)}')
```

## 7. Обучение и внешние проверки

Ridge, HistGradientBoosting, CatBoost и LightGBM (если доступны) сопоставляются на train→development. Выбранная конфигурация фиксируется до окончательного теста, затем измеряется перенос на будущие годы, новые территории и их пересечение. Простой исторический прогноз обязателен как независимый baseline. Стохастические сиды сохраняются без статистического удвоения объёма тестовой выборки.

```python
# 08 · DEV-LOCKED MODEL BANK; TRUE PAIRED TEST PREDICTIONS, SEEDS, NO OUTCOME-BASED SELECTION
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
PRED_ROWS=[];METRIC_ROWS=[];DEV_ROWS=[];SPLIT_ROWS=[];IMPORTANCE_ROWS=[];MODEL_REGISTRY={}
try:
    from catboost import CatBoostRegressor
    CB_AVAILABLE=True
except ImportError:CB_AVAILABLE=False
try:
    from lightgbm import LGBMRegressor
    LGBM_AVAILABLE=True
except ImportError:LGBM_AVAILABLE=False
BANK=[f for f in CFG['MODEL_FAMILIES'] if f in ('RIDGE','HGB','ELASTICNET') or (f=='CATBOOST' and CB_AVAILABLE) or
      (f=='LIGHTGBM' and LGBM_AVAILABLE)]
mark('MODEL_BANK','PASS' if 'RIDGE' in BANK and 'HGB' in BANK else 'FAIL',
      f'available={BANK}; CatBoost={CB_AVAILABLE}; LightGBM={LGBM_AVAILABLE}')

def design(train,test,features,keep_reference=None):
    refs=sorted(train.region_name.dropna().astype(str).unique()) if keep_reference is None else list(keep_reference)
    usable=[f for f in dict.fromkeys(features) if f in train and
            pd.to_numeric(train[f],errors='coerce').notna().any()]
    ensure('origin_year' in usable,'ORIGIN_YEAR_MISSING')
    def mx(df):
        xx=df[usable].apply(pd.to_numeric,errors='coerce').replace([np.inf,-np.inf],np.nan).reset_index(drop=True)
        for f in usable:
            if f.startswith(('G1_','G2_','G3_','G4_','G5_')):
                xx[f]=np.sign(xx[f])*np.log1p(np.abs(xx[f])) # fixed transform; no fitted test statistics
        reg=df.region_name.astype(str).reset_index(drop=True)
        # One-shot matrix construction: no repeated __setitem__ / pandas frame fragmentation.
        region_columns={'region_'+hashlib.sha256(name.encode()).hexdigest()[:10]:
                        reg.eq(name).to_numpy(dtype=float) for name in refs}
        if region_columns:
            region_matrix=pd.DataFrame(region_columns,index=xx.index,dtype=float)
            xx=pd.concat([xx,region_matrix],axis=1,copy=False)
        ensure(not xx.columns.duplicated().any(),'DUPLICATE_REGION_DUMMY_COLUMNS')
        return xx.astype(float)
    return mx(train),mx(test),usable,refs

def estimator(family,seed,params):
    if family=='RIDGE':
        return make_pipeline(SimpleImputer(strategy='median',add_indicator=True),StandardScaler(),
                             Ridge(alpha=float(params['alpha'])))
    if family=='ELASTICNET':
        return make_pipeline(SimpleImputer(strategy='median',add_indicator=True),StandardScaler(),
                ElasticNet(alpha=float(params['alpha']),l1_ratio=float(params['l1_ratio']),
                           max_iter=5000,random_state=int(seed)))
    if family=='HGB':
        return HistGradientBoostingRegressor(max_iter=int(params['iterations']),
            max_leaf_nodes=int(params['leaves']),learning_rate=0.045,min_samples_leaf=18,
            l2_regularization=2.0,early_stopping=False,random_state=int(seed))
    if family=='CATBOOST':
        return CatBoostRegressor(iterations=int(params['iterations']),depth=int(params['depth']),
             learning_rate=0.05,l2_leaf_reg=4,loss_function='RMSE',verbose=False,
             thread_count=int(CFG['DUCKDB_THREADS']),random_seed=int(seed),allow_writing_files=False)
    if family=='LIGHTGBM':
        return LGBMRegressor(n_estimators=int(params['iterations']),num_leaves=int(params['leaves']),
             max_depth=6,learning_rate=0.04,reg_lambda=3.0,min_child_samples=18,
             n_jobs=int(CFG['DUCKDB_THREADS']),verbosity=-1,random_state=int(seed))
    raise ValueError('UNKNOWN_MODEL_FAMILY')
PARAM_GRID={
 'RIDGE':[{'alpha':2},{'alpha':20}],
 'ELASTICNET':[{'alpha':0.2,'l1_ratio':0.25},{'alpha':2.,'l1_ratio':0.55}],
 'HGB':[{'iterations':125,'leaves':15},{'iterations':175,'leaves':23}],
 'CATBOOST':[{'iterations':220,'depth':4},{'iterations':300,'depth':6}],
 'LIGHTGBM':[{'iterations':150,'leaves':15},{'iterations':200,'leaves':23}]}
def fit_predict(family,train,test,target,features,seed=1729,params=None):
    Xtr,Xte,used,refs=design(train,test,features)
    model=estimator(family,seed,params or PARAM_GRID[family][0]);model.fit(Xtr,train[target].to_numpy(float))
    pred=np.asarray(model.predict(Xte),dtype=float)
    ensure(len(pred)==len(test) and np.isfinite(pred).all(),'PREDICTION_NOT_FINITE')
    return pred,model,used,refs

def baseline_predict(train,test,target,h):
    la=CFG['FEATURE_PUBLICATION_LAG'];hist=test.get('hist_migration_l'+str(la))
    if target=='migration_per_1000':
        return pd.to_numeric(hist,errors='coerce').fillna(float(train[target].median())).to_numpy(float)
    p1=pd.to_numeric(test[f'hist_population_l{la}'],errors='coerce')
    p3=pd.to_numeric(test[f'hist_population_l{la+2}'],errors='coerce')
    recent_growth=np.power(p1/p3,0.5*h)
    series=100*(recent_growth-1)
    return series.replace([np.inf,-np.inf],np.nan).fillna(float(train[target].median())).to_numpy(float)

def regional_baseline_predict(train,test,target):
    """Train-only regional tendency; unknown region falls back to overall median."""
    med=train.groupby('region_name')[target].median()
    glob=float(train[target].median())
    return test.region_name.map(med).fillna(glob).to_numpy(float)

def deterministic_family(family):
    # HGB and LightGBM used here have neither row nor column sampling, and no early stopping.
    return family in ('RIDGE','ELASTICNET','HGB','LIGHTGBM')

def family_seeds(family):
    return [CFG['SEED']] if deterministic_family(family) else list(CFG['SEEDS_FINAL'])

def metrics(y,p,w):
    y=np.asarray(y,float);p=np.asarray(p,float);w=np.asarray(w,float)
    ensure(len(y)==len(p) and np.isfinite(y).all() and np.isfinite(p).all(),'INVALID_EVALUATION_VECTORS')
    w=np.where(np.isfinite(w)&(w>0),w,0)
    err=y-p
    return {'mae':float(np.mean(np.abs(err))), 'rmse':float(np.sqrt(np.mean(err**2))),
       'weighted_mae':float(np.average(np.abs(err),weights=w)) if w.sum()>0 else np.nan,
       'n_test':int(len(y))}

def add_result(task,arm,model,seed,train,test,pred,n_feat):
    target,h,protocol=task['target'],task['horizon'],task['protocol'];study=task['study']['id']
    y=test[target].to_numpy(float)
    w=test[f'hist_population_l{CFG["FEATURE_PUBLICATION_LAG"]}'].to_numpy(float)
    met=metrics(y,pred,w)
    METRIC_ROWS.append({'study_id':study,'target':target,'horizon':h,'protocol':protocol,
         'arm':arm,'model':model,'seed':seed,'n_train':len(train),'n_features_used':n_feat,**met})
    rows=test[['entity_id','origin_year','target_year','region_name','mun_type','information_cutoff',
       'source_vintage','horizon','population_at_origin']].copy()
    rows['target']=target;rows['study_id']=study;rows['split_id']=protocol
    rows['arm']=arm;rows['model']=model;rows['seed']=seed;rows['y_true']=y;rows['y_pred']=pred
    rows['population_weight']=w
    rows['absolute_error']=np.abs(y-pred);rows['squared_error']=(y-pred)**2
    PRED_ROWS.append(rows)

# CPU-only, isolated contract test (synthetic values never added to evidence or metrics).
def _model_unit_check():
    rng=np.random.default_rng(CFG['SEED']);n=100
    fake=pd.DataFrame({'region_name':np.repeat(['R1','R2'],n//2),
       'origin_year':np.tile(np.arange(2010,2020),n//10),
       'G1_01_per_resident':rng.lognormal(size=n),'G2_01_per_1000':rng.random(n)*20,
       '_test_target':rng.normal(size=n),'hist_population_l1':rng.integers(100,10000,n)})
    for fam in ('RIDGE','HGB','ELASTICNET'):
        yp,_,_,_=fit_predict(fam,fake.iloc[:80],fake.iloc[80:],'_test_target',
            ['origin_year','G1_01_per_resident','G2_01_per_1000'],CFG['SEED'],PARAM_GRID[fam][0])
        ensure(np.isfinite(yp).all() and len(yp)==20,'MODEL_UNIT_TEST_FAIL_'+fam)
_model_unit_check()
mark('UNIT_MODEL_CONTRACT','PASS','Ridge and HGB synthetic function-only smoke; excluded from scientific results')

# DEV count is known after TASKS and exact parameter grids are registered.
DEV_RUNS=[(t['study']['id'],t['target'],t['horizon']) for t in TASKS
     if t['protocol']=='ROLLING_TIME']
DEV_WORK=len(set(DEV_RUNS))

DEV_TOTAL=len(set(DEV_RUNS))*sum(len(PARAM_GRID[f]) for f in BANK
                  if f in PARAM_GRID and (CFG['MODE']=='FULL'))
if CFG['MODE']!='FULL':DEV_TOTAL=len(set(DEV_RUNS))*len(BANK)
set_tick('DEV',DEV_TOTAL)
# Tune on a held-out historical DEV block, prior to FINAL test. No use of final target values.
SELECTED_FAMILY={}
for st in STUDIES:
    for target in CFG['TARGETS']:
        for h in CFG['HORIZONS'] if CFG['MODE']=='FULL' else [1]:
            runs=[t for t in TASKS if t['study']['id']==st['id'] and t['target']==target and
                  t['horizon']==h and t['protocol']=='ROLLING_TIME']
            if not runs:continue
            task=runs[0];df=task['data'];tr=df[task['masks']['train']];dv=df[task['masks']['dev']]
            maxarm=max(task['study']['arms'],key=lambda a:len(a['features']))
            for family in BANK:
                for params in PARAM_GRID[family] if CFG['MODE']=='FULL' else PARAM_GRID[family][:1]:
                    try:
                        yp,_,used,_=fit_predict(family,tr,dv,target,maxarm['features'],CFG['SEED'],params)
                        sc=metrics(dv[target],yp,dv[f'hist_population_l{CFG["FEATURE_PUBLICATION_LAG"]}'])
                        DEV_ROWS.append({'study_id':st['id'],'target':target,'horizon':h,
                            'family':family,'params_json':json.dumps(params,sort_keys=True),
                            'dev_mae':sc['mae'],'dev_rmse':sc['rmse'],'n_dev':len(dv),
                            'features_used':len(used)})
                    except Exception as ex:
                        log(f'DEV model failure {st["id"]} h{h} {family}: {type(ex).__name__}', 'WARN')
                    finally:tick('DEV model',1,f'{st["id"]} h{h} {family}')
            v=[r for r in DEV_ROWS if (r['study_id'],r['target'],r['horizon'])==(st['id'],target,h)]
            if v:
                choice=min(v,key=lambda r:(r['dev_mae'],r['dev_rmse']))
                SELECTED_FAMILY[(st['id'],target,h)]=(choice['family'],json.loads(choice['params_json']))
            # Fits, not tasks, are the exact DEV work units.

ensure(_TICK_DONE==_TICK_TOTAL,'DEV_PROGRESS_MISMATCH')
TRAIN_TOTAL=sum(2+(len(t['study']['arms'])*len(family_seeds(SELECTED_FAMILY[(t['study']['id'],t['target'],t['horizon'])][0]))
                   if (t['study']['id'],t['target'],t['horizon']) in SELECTED_FAMILY else 0)
                for t in TASKS)
set_tick('TRAIN',TRAIN_TOTAL)
for task in TASKS:
    study=task['study'];d=task['data'];m=task['masks'];protocol=task['protocol']
    tr=d[m['train']|m['dev']].copy();te=d[m['test']].copy()
    if protocol in ('ROLLING_TIME','CROSSED'):
        ensure(tr.target_year.max()<te.origin_year.min(),'TRAIN_LABEL_NOT_KNOWN_AT_EARLIEST_TEST_ORIGIN')
    if protocol in ('TERRITORY_ONLY','CROSSED'):
        ensure(set(tr.region_name.astype(str)).isdisjoint(set(te.region_name.astype(str))),'REGIONAL_CROSS_FOLD_LEAK')
        ensure(set(tr.entity_id).isdisjoint(set(te.entity_id)),'ENTITY_CROSS_FOLD_LEAK')
    SPLIT_ROWS.append(pd.concat([tr[['entity_id','origin_year','target_year','region_name']].assign(
           study_id=study['id'],target=task['target'],horizon=task['horizon'],protocol=protocol,part='train'),
           te[['entity_id','origin_year','target_year','region_name']].assign(
           study_id=study['id'],target=task['target'],horizon=task['horizon'],protocol=protocol,part='test')]))
    bp=baseline_predict(tr,te,task['target'],task['horizon'])
    add_result(task,'M0','HISTORICAL',CFG['SEED'],tr,te,bp,1)
    rb=regional_baseline_predict(tr,te,task['target'])
    add_result(task,'M0','REGIONAL_MEDIAN',CFG['SEED'],tr,te,rb,1)
    tick('historical baselines',2,study['id'])
    choice=SELECTED_FAMILY.get((study['id'],task['target'],task['horizon']))
    if choice is None:continue
    family,params=choice
    for arm in study['arms']:
        seeds=family_seeds(family)
        for si,seed in enumerate(seeds):
            try:
                yp,model,used,refs=fit_predict(family,tr,te,task['target'],arm['features'],seed,params)
                add_result(task,arm['arm'],family,seed,tr,te,yp,len(used))
                if si==0 and protocol=='ROLLING_TIME' and arm==max(study['arms'],key=lambda a:len(a['features'])):
                    key=(study['id'],task['target'],task['horizon'])
                    MODEL_REGISTRY[key]={'model':model,'features':arm['features'],
                       'used':used,'refs':refs,'train':tr,'test':te,'pred':yp,
                       'family':family,'params':params,'arm':arm['arm']}
                    name=f'{study["id"]}_{task["target"]}_h{task["horizon"]}_{family}.joblib'
                    joblib.dump({'model':model,'features':used,'refs':refs,'target':task['target'],
                        'horizon':task['horizon'],'study':study['id'],'source_commit':source_sha,
                        'model_family':family,'params':params},WORK/'models'/name,compress=3)
            except Exception as ex:
                mark(f'FIT_{study["id"]}_{task["target"]}_{task["horizon"]}_{protocol}_{arm["arm"]}_{seed}',
                     'FAIL',type(ex).__name__+': '+str(ex)[:100])
                log(f'FIT {study["id"]} {arm["arm"]} {family}: {type(ex).__name__}','ERROR')
            tick('training',1,
                 f'{study["id"]} {arm["arm"]} h={task["horizon"]} s={seed}')
ensure(_TICK_DONE==_TICK_TOTAL,'TRAIN_PROGRESS_MISMATCH')
ensure(DEV_WORK==len(set(DEV_RUNS)),'DEV_WORK_COUNT_CONTRACT')
# Единственный _PROGRESS_BAR сохраняется на весь прогон.
RUN_COMPLETION_STATUS='MODELS_COMPLETED_EXPORT_PENDING'
METRICS=pd.DataFrame(METRIC_ROWS);DEV_METRICS=pd.DataFrame(DEV_ROWS)
PREDICTIONS=pd.concat(PRED_ROWS,ignore_index=True) if PRED_ROWS else pd.DataFrame()
SPLITS=pd.concat(SPLIT_ROWS,ignore_index=True) if SPLIT_ROWS else pd.DataFrame()
SEED_QA_ROWS=[]
if len(PREDICTIONS):
    for (sid,targ,h,proto,arm,model),g in PREDICTIONS[PREDICTIONS.model.isin(BANK)].groupby(
          ['study_id','target','horizon','split_id','arm','model']):
        hashes=[]
        for seed,z in g.groupby('seed'):
            arr=z.sort_values(['entity_id','origin_year','target_year']).y_pred.to_numpy(float)
            hashes.append(hashlib.sha256(np.asarray(arr,dtype='<f8').tobytes()).hexdigest())
        SEED_QA_ROWS.append({'study_id':sid,'target':targ,'horizon':h,'protocol':proto,
            'arm':arm,'model':model,'n_requested_seeds':len(family_seeds(model)),
            'n_configured_seeds':len(CFG['SEEDS_FINAL']),
            'n_executed_seeds':len(hashes),'n_distinct_prediction_hashes':len(set(hashes)),
            'n_effectively_distinct_runs':len(set(hashes)),
            'seeds_are_independent_samples':False})
SEED_QA=pd.DataFrame(SEED_QA_ROWS)
if len(PREDICTIONS):
    cols=['entity_id','origin_year','target_year','horizon','target','study_id','split_id','arm','model','seed']
    ensure(not PREDICTIONS.duplicated(cols).any(),'DUPLICATE_TEST_PREDICTIONS')
    for keys,p in PREDICTIONS[PREDICTIONS.model.ne('HISTORICAL')].groupby(
           ['study_id','target','horizon','split_id','model','seed']):
        ensure(p.groupby('arm').size().nunique()==1,'ABLA_SAMPLE_SIZE_MISMATCH')
        key_hashes=[hashlib.sha256(a[['entity_id','origin_year','target_year']].sort_values(['entity_id','origin_year']).to_csv(index=False).encode()).hexdigest() for _,a in p.groupby('arm')]
        ensure(len(set(key_hashes))==1,'ABLA_TEST_KEYS_MISMATCH')
mark('PASS_TRAINING','PASS' if len(METRICS) and len(SELECTED_FAMILY)==DEV_WORK and DEV_WORK>0 and not any(v['status']=='FAIL' for k,v in STATUS.items() if k.startswith('FIT_')) else 'NOT_FEASIBLE' if not DEV_WORK else 'PARTIAL',
     f'{len(METRICS)} model evaluations; {len(PREDICTIONS)} test predictions; dev-selected families={len(SELECTED_FAMILY)}')
quiet_print('Bank:',BANK,'| evaluated models:',len(METRICS),'| test forecasts:',len(PREDICTIONS))


# ADVANCED · V4.00 · DEV-LOCKED NON-NEGATIVE SUPER LEARNER, NO TEST-TUNING
# Additional experiment: all feasible TASKS; legacy preregistered comparisons remain intact.
from scipy.optimize import minimize as _minimize
from sklearn.ensemble import ExtraTreesRegressor as _ExtraTreesRegressor
try:
    from xgboost import XGBRegressor as _XGBRegressor
    XGB_AVAILABLE=True
except Exception:
    XGB_AVAILABLE=False

def _advanced_estimator(family,seed,params=None):
    z=CFG['ADVANCED_N_JOBS'];p=params or {}
    if family=='XGBOOST':
        ensure(XGB_AVAILABLE,'XGBOOST_NOT_INSTALLED')
        return _XGBRegressor(n_estimators=p.get('n_estimators',260),max_depth=p.get('max_depth',5),
            learning_rate=p.get('learning_rate',0.045),min_child_weight=6,subsample=1.,colsample_bytree=1.,
            reg_lambda=7.,reg_alpha=0.3,tree_method='hist',n_jobs=z,random_state=seed,objective='reg:squarederror')
    if family=='EXTRA_TREES':
        return make_pipeline(SimpleImputer(strategy='median',add_indicator=True),
          _ExtraTreesRegressor(n_estimators=p.get('n_estimators',240),max_features=0.84,
             min_samples_leaf=8,max_depth=16,n_jobs=z,random_state=seed))
    if family=='RIDGE':return estimator('RIDGE',seed,{'alpha':20.})
    if family=='HGB':return estimator('HGB',seed,{'iterations':175,'leaves':23})
    if family=='CATBOOST':return estimator('CATBOOST',seed,{'iterations':320,'depth':6})
    if family=='LIGHTGBM':return estimator('LIGHTGBM',seed,{'iterations':250,'leaves':23})
    raise ValueError('ADVANCED_FAMILY_NOT_REGISTERED '+str(family))

_ADV_REQUEST=['XGBOOST','CATBOOST','LIGHTGBM','HGB','EXTRA_TREES','RIDGE']
_ADV_READY={'XGBOOST':XGB_AVAILABLE,'CATBOOST':CB_AVAILABLE,'LIGHTGBM':LGBM_AVAILABLE,
            'HGB':True,'EXTRA_TREES':True,'RIDGE':True}
ADV_BANK=[x for x in _ADV_REQUEST if _ADV_READY[x]]
if CFG['MODE']=='FULL' and CFG['ADVANCED_REQUIRE_ALL_FAMILIES']:
    ensure(all(_ADV_READY.values()),'ADVANCED_FULL_MODEL_DEPENDENCIES_INCOMPLETE')
ensure(len(ADV_BANK)>=3,'ADVANCED_ENSEMBLE_REQUIRES_THREE_MODEL_FAMILIES')
ADVANCED_DEV_ROWS=[];ADVANCED_TEST_ROWS=[];ADVANCED_INTERVAL_ROWS=[];ADVANCED_ENSEMBLE_REGISTRY={};ADVANCED_MODEL_REGISTRY={}

def _advance_design_model(family,tr,te,target,features,seed=1729):
    xtr,xte,used,refs=design(tr,te,features)
    # Missing predictors are supported by all configured estimators through native missing
    # handling (XGBoost, HGB, LGBM, CatBoost) or train-only median imputation.
    mod=_advanced_estimator(family,int(seed));mod.fit(xtr,tr[target].to_numpy(float))
    yp=np.asarray(mod.predict(xte),float).reshape(-1)
    ensure(len(yp)==len(te) and np.isfinite(yp).all(),'ADVANCED_PREDICTION_NONFINITE_'+family)
    return {'family':family,'model':mod,'prediction':yp,'used':used,'refs':refs,'cols':list(xte.columns)}

def _simplex_weights(y,prediction_columns):
    P=np.column_stack(prediction_columns).astype(float);v=np.asarray(y,float)
    ensure(np.isfinite(P).all() and np.isfinite(v).all(),'ENSEMBLE_DEV_NONFINITE')
    n=P.shape[1];baseline=int(np.argmin(np.mean(np.abs(P-v[:,None]),axis=0)))
    if n==1:return np.ones(1,dtype=float)
    # Smoothed absolute loss with weak shrinkage toward equal weights; DEV only.
    fun=lambda w:float(np.mean(np.sqrt((P@w-v)**2+0.03**2))+CFG['ADVANCED_WEIGHT_RIDGE']*np.sum(w**2))
    res=_minimize(fun,x0=np.full(n,1./n),method='SLSQP',bounds=[(0.,1.)]*n,
        constraints={'type':'eq','fun':lambda w:float(np.sum(w)-1.)},options={'maxiter':200,'ftol':1e-9})
    w=np.maximum(0,np.asarray(res.x if res.success else np.eye(n)[baseline],float))
    if not np.isfinite(w).all() or w.sum()<=0:w=np.eye(n)[baseline]
    w=w/w.sum()
    if mean_absolute_error(v,P@w)>mean_absolute_error(v,P[:,baseline])+1e-9:
        w=np.eye(n)[baseline]  # no regressive weights chosen on DEV
    ensure(np.all(w>=0) and abs(w.sum()-1)<1e-9,'ENSEMBLE_WEIGHT_SIMPLEX_BROKEN')
    return w

def _adv_fit_one_task(task):
    sid=task['study']['id'];target=task['target'];h=int(task['horizon']);proto=task['protocol']
    d=task['data'];m=task['masks'];tr=d[m['train']].copy();dv=d[m['dev']].copy();te=d[m['test']].copy()
    ensure(len(tr)>=CFG['MIN_TRAIN_ROWS'] and len(dv)>=CFG['MIN_TEST_ROWS'],'ADVANCED_DEV_TOO_SMALL')
    ensure(len(te)>=CFG['MIN_TEST_ROWS'],'ADVANCED_TEST_TOO_SMALL')
    if proto in ('ROLLING_TIME','CROSSED'):
        ensure(tr.target_year.max()<te.origin_year.min(),'ADVANCED_ORIGIN_CAUSAL_ORDER_BROKEN')
    if proto in ('TERRITORY_ONLY','CROSSED'):
        ensure(set(tr.region_name.astype(str)).isdisjoint(set(te.region_name.astype(str))),'ADVANCED_REGIONAL_LEAK')
        ensure(set(dv.entity_id).isdisjoint(set(te.entity_id)),'ADVANCED_GEOGRAPHICAL_DEV_LEAK')
    arms=task['study']['arms'];full_arm=max(arms,key=lambda a:len(a['features']));features=full_arm['features']
    supported=[f for f in features if f in tr and pd.to_numeric(tr[f],errors='coerce').notna().any()]
    if len(supported)<2:return {'status':'NOT_FEASIBLE','reason':'TOO_FEW_TRAIN_PREDICTORS'}
    seed=int(CFG['SEED']);devmods=[]
    for fam in ADV_BANK[:CFG['ADVANCED_MAX_CANDIDATE_FAMILIES']]:
        try:
            r=_advance_design_model(fam,tr,dv,target,features,seed)
            loss=float(mean_absolute_error(dv[target],r['prediction']))
            devmods.append((loss,fam,r))
            ADVANCED_DEV_ROWS.append({'study_id':sid,'target':target,'horizon':h,'protocol':proto,
                'family':fam,'dev_mae':loss,'n_train':len(tr),'n_dev':len(dv),'n_columns':len(r['cols']),
                'selection':'PRE_TEST_HISTORICAL_DEV_ONLY','status':'PASS'})
        except Exception as ex:
            ADVANCED_DEV_ROWS.append({'study_id':sid,'target':target,'horizon':h,'protocol':proto,
                'family':fam,'dev_mae':np.nan,'n_train':len(tr),'n_dev':len(dv),
                'selection':'PRE_TEST_HISTORICAL_DEV_ONLY','status':'FIT_FAILED_'+type(ex).__name__})
            log(f'ADV DEV {sid}/{h}/{fam}: {type(ex).__name__}: {str(ex)[:120]}','WARN')
    if len(devmods)<2:return {'status':'NOT_FEASIBLE','reason':'LESS_THAN_TWO_DEV_FAMILIES_FIT'}
    devmods.sort(key=lambda z:(z[0],z[1]));top=devmods[:min(CFG['ADVANCED_ENSEMBLE_K'],len(devmods))]
    weights=_simplex_weights(dv[target], [v[2]['prediction'] for v in top]);dev_mix=sum(float(w)*entry[2]['prediction'] for w,entry in zip(weights,top))
    dev_abs=np.abs(dv[target].to_numpy(float)-dev_mix)
    q90=float(np.quantile(dev_abs,min(1.,math.ceil((len(dv)+1)*0.9)/len(dv)),method='higher'))
    full=pd.concat([tr,dv],ignore_index=True);members=[];testpred=[]
    for (dev_mae,fam,_),weight in zip(top,weights):
        r=_advance_design_model(fam,full,te,target,features,seed)
        members.append({'family':fam,'weight':float(weight),'model':r['model'],'used':r['used'],
                        'refs':r['refs'],'columns':r['cols'],'dev_mae':dev_mae})
        testpred.append(r['prediction'])
    best_single_family=str(top[0][1]);best_single_prediction=np.asarray(testpred[0],float)
    pred=sum(member['weight']*v for member,v in zip(members,testpred))
    ensure(np.isfinite(pred).all(),'ADVANCED_FINAL_NONFINITE')
    arm='SUPERLEARNER_DEV';add_result(task,arm,'SUPERLEARNER',seed,full,te,pred,len(supported))
    for member in members:
        r=next(v for v,memb in zip(testpred,members) if memb is member)
        if CFG['ADVANCED_EXPORT_MEMBERS']:add_result(task,arm,'SL_'+member['family'],seed,full,te,r,len(member['used']))
    n_test=len(te);covers=np.abs(te[target].to_numpy(float)-pred)<=q90
    ADVANCED_INTERVAL_ROWS.append({'study_id':sid,'target':target,'horizon':h,'protocol':proto,
        'n_dev':len(dv),'n_test':n_test,'q90_abs_dev':q90,'empirical_test_coverage':float(covers.mean()),
        'avg_width':2*q90,'method':'DEV_RESIDUAL_QUANTILE_AFTER_REFIT_HEURISTIC','nominal':0.9,
        'formal_coverage_guarantee':False,'reason':'Temporal shift and refitting after DEV; not formal split-conformal coverage'})
    result={'study_id':sid,'target':target,'horizon':h,'protocol':proto,
        'n_train':len(tr),'n_dev':len(dv),'n_test':len(te),'arm':arm,
        'members':[m['family'] for m in members], 'weights':[round(m['weight'],8) for m in members],
        'dev_mae':float(mean_absolute_error(dv[target],dev_mix)),
        'test_mae':float(mean_absolute_error(te[target],pred)),
        'best_single_family_dev_locked':best_single_family,
        'best_single_dev_mae':float(top[0][0]),
        'best_single_test_mae':float(mean_absolute_error(te[target],best_single_prediction)),
        'ensemble_minus_single_test_mae':float(mean_absolute_error(te[target],pred)-mean_absolute_error(te[target],best_single_prediction)),
        'status':'PASS','selection':'DEV_ONLY','n_factors':len(supported)}
    if proto=='ROLLING_TIME':
        key=(sid,target,h)
        ADVANCED_ENSEMBLE_REGISTRY[key]={'members':members,'study_id':sid,'target':target,'horizon':h,
            'train':full,'test':te,'features':features,'pred':pred,'dev_predictions':dev_mix,
            'dev_target':dv[target].to_numpy(float),'q90_dev':q90,
            'best_single_family':best_single_family,'best_single_pred':best_single_prediction}
        base=WORK/'models'/f'SUPERLEARNER_{sid}_{target}_h{h}.joblib'
        joblib.dump({'members':members,'source_sha':source_sha,'model_stage':'DEV_FROZEN_FULL_REFIT',
                     'target':target,'horizon':h,'features':features},base,compress=3)
    return result

if CFG['ADVANCED_ENABLE']:
    progress('ADVANCED',0,len(TASKS),'Ансамбль: проверка по DEV')
    _adv_task_done=0
    for t in TASKS:
        sid=t['study']['id'];h=t['horizon'];target=t['target'];proto=t['protocol']
        try:ADVANCED_TEST_ROWS.append(_adv_fit_one_task(t))
        except Exception as ex:
            ADVANCED_TEST_ROWS.append({'study_id':sid,'target':target,'horizon':h,'protocol':proto,
                'status':'FAIL','reason':type(ex).__name__+': '+str(ex)[:250]})
            log('ADVANCED ENSEMBLE '+sid+'/'+target+': '+type(ex).__name__+': '+str(ex)[:200],'ERROR')
        _adv_task_done+=1
        progress('ADVANCED',_adv_task_done,len(TASKS),f'Задач: {_adv_task_done}/{len(TASKS)}')
    progress('ADVANCED',1,1,'DEV-ансамбль завершён')
if not CFG['ADVANCED_ENABLE']:progress('ADVANCED',1,1,'Ансамбль отключён настройкой')
ADVANCED_DEV=pd.DataFrame(ADVANCED_DEV_ROWS);ADVANCED_ENSEMBLES=pd.DataFrame(ADVANCED_TEST_ROWS)
ADVANCED_INTERVALS=pd.DataFrame(ADVANCED_INTERVAL_ROWS)
if CFG['ADVANCED_ENABLE']:
    allpass=bool(len(ADVANCED_ENSEMBLES)==len(TASKS) and ADVANCED_ENSEMBLES.status.eq('PASS').all()) if len(TASKS) else False
    mark('ADVANCED_ENSEMBLE','PASS' if allpass else 'PARTIAL',
        f'completed={int(ADVANCED_ENSEMBLES.status.eq("PASS").sum()) if len(ADVANCED_ENSEMBLES) else 0}/{len(TASKS)}; ' +
        f'families={ADV_BANK}; dev-only weight calibration')
else:mark('ADVANCED_ENSEMBLE','NOT_FEASIBLE','Explicitly disabled')
# Rebuild canonical output structures; the next cell computes all independent metrics/CI from combined predictions.
METRICS=pd.DataFrame(METRIC_ROWS)
PREDICTIONS=pd.concat(PRED_ROWS,ignore_index=True) if PRED_ROWS else pd.DataFrame()
if len(PREDICTIONS):
    _id_cols=['entity_id','origin_year','target_year','horizon','target','study_id','split_id','arm','model','seed']
    ensure(not PREDICTIONS.duplicated(_id_cols).any(),'ADVANCED_OUTPUT_DUPLICATE_TEST_KEYS')
quiet_print('Advanced:',len(ADVANCED_ENSEMBLES),'tasks | pass',int(ADVANCED_ENSEMBLES.status.eq('PASS').sum()) if len(ADVANCED_ENSEMBLES) else 0,
      '| full prediction rows:',len(PREDICTIONS))
```

## 8. Парные сравнения, неопределённость и интерпретация

Для каждой пары вариантов используют ровно одинаковые муниципалитеты и годы теста. Доверительные интервалы строятся региональным блочным bootstrap, подтверждающие сравнения корректируются по Холму, исследовательские — методом Бенджамини—Хохберга. Актуализированные проверки: leave-one-group-out, ALE, перестановочная важность, G1×G2 и пространственно-временная переносимость. Все результаты имеют **прогнозную**, не причинную интерпретацию.
Дополнительно оцениваются неоднородность по размеру муниципалитета и чувствительность к документально подтверждённой стабильности территориальных единиц. Панельные фиксированные эффекты оцениваются только на обучающих данных при доказанной идентифицируемости и с регистрацией причин отказа.


**v2.3:** реестр гипотез формируется до TEST; CI по региональным блокам, Holm для подтверждающих и BH-FDR для разведочных. Дубли сидов не увеличивают `n`.

```python
# 09 · COMMON-KEY ABLATIONS, REGION-BLOCK BOOTSTRAP, HOLM/FDR, ALE, INTERACTIONS
progress('STATS',0,1,'Парные сравнения и статистика')
ABLA_ROWS=[];LOO_ROWS=[];CORR_ROWS=[];INTERACTION_ROWS=[];TRANSFER_ROWS=[]
GEO_ROWS=[];IMP_ROWS=[];ALE_ROWS=[];SHAP_ROWS=[];FE_ROWS=[];SENSITIVITY_ROWS=[];ERROR_YEAR_ROWS=[]
def holm_adjust(p):
    arr=np.asarray(p,float);order=np.argsort(arr);out=np.empty(len(arr));running=0.
    for rank,j in enumerate(order):
        running=max(running,min(1.0,float(arr[j])*(len(arr)-rank)));out[j]=running
    return out
def bh_adjust(p):
    arr=np.asarray(p,float);ordr=np.argsort(arr);out=np.empty(len(arr));running=1.
    for rank in reversed(range(len(arr))):
        j=ordr[rank];running=min(running,float(arr[j])*len(arr)/(rank+1));out[j]=running
    return out
def block_delta(z,old,new,boots,seed):
    v=z[['region_name','target_year',old,new]].dropna()
    groups=[g for _,g in v.groupby('region_name',dropna=False)]
    if len(groups)<CFG['MIN_BOOTSTRAP_REGIONS']:return None
    d=np.array([(g[old]-g[new]).sum() for g in groups]);n=np.array([len(g) for g in groups])
    point=float(d.sum()/n.sum())
    rng=np.random.default_rng(seed);ids=rng.integers(0,len(groups),(boots,len(groups)))
    vals=(d[ids].sum(axis=1)/n[ids].sum(axis=1))
    ci=np.quantile(vals,[0.025,0.975])
    # Monte Carlo finite-sample p with +1 correction; no p=0 due to bootstrap discretization.
    # Random region-level sign-flips approximate a paired null test; CI remains a block bootstrap.
    signs=rng.choice(np.array([-1.,1.]),size=(boots,len(groups)),replace=True)
    null_dist=(signs*d).sum(axis=1)/n.sum()
    p=(1+int((np.abs(null_dist)>=abs(point)).sum()))/(boots+1)
    return {'delta_mae':point,'ci_low':float(ci[0]),'ci_high':float(ci[1]),
         'p_cluster_signflip':float(p),'n_regions':len(groups),'n_common_keys':len(v),
         'bootstrap_unit':'region','n_bootstraps':boots}

# Isolated, inexpensive sign-flip / regional block unit test.
_check=pd.DataFrame({'region_name':list('ABCDEFGH')*2,'target_year':[2022]*16,
                      '_baseline_error':[2.0]*16,'_improved_error':[1.0]*16})
_st=block_delta(_check,'_baseline_error','_improved_error',99,CFG['SEED'])
ensure(_st is not None and abs(_st['delta_mae']-1.0)<1e-12 and _st['n_regions']==8,
       'UNIT_CLUSTER_INFERENCE_FAILURE')
mark('UNIT_STAT_CONTRACT','PASS','Synthetic sign-flip/bootstrap arithmetic; excluded from scientific findings')

if len(PREDICTIONS):
    selected=PREDICTIONS[~PREDICTIONS.model.isin(['HISTORICAL','REGIONAL_MEDIAN'])].copy()
    keys=['entity_id','origin_year','target_year','region_name']
    for (sid,target,h,proto,model),p in selected.groupby(['study_id','target','horizon','split_id','model']):
        # Average losses over stochastic seeds BEFORE block resampling; seeds are not independent municipalities.
        grp=p.groupby(keys+['arm'],dropna=False).agg(absolute_error=('absolute_error','mean')).reset_index()
        wide=grp.pivot(index=keys,columns='arm',values='absolute_error').reset_index()
        st=next(s for s in STUDIES if s['id']==sid)
        if st['id']=='CANONICAL' or st['id'].startswith('YEAR_'):
            pairs=[(a['arm'],b['arm']) for a,b in zip(st['arms'][:-1],st['arms'][1:])]
            pairs+=([('M1','M3')] if 'M3' in wide and 'M1' in wide else [])
        elif st['id'].startswith('AUX_'):
            pairs=[('M0',st['arms'][-1]['arm'])]
        else:
            pairs=[('M0','FULL_AVAILABLE')]
            pairs += [('WITHOUT_'+g,'FULL_AVAILABLE') for g in AVAILABLE_GROUPS if 'WITHOUT_'+g in wide]
        for old,new in pairs:
            if old not in wide or new not in wide:continue
            stat=block_delta(wide,old,new,CFG['BOOTSTRAPS'],CFG['SEED']+h)
            if stat:
                row={'study_id':sid,'target':target,'horizon':h,'protocol':proto,
                     'model':model,'old_arm':old,'new_arm':new,
                     'comparison':old+' → '+new,'confirmatory':(st['confirmatory'] and
                     len(TASK_STATUSES.loc[(TASK_STATUSES.study_id==sid)&
                      (TASK_STATUSES.target==target)&(TASK_STATUSES.horizon==h)&
                      (TASK_STATUSES.protocol==proto)&
                      (TASK_STATUSES.scientific_status=='COMPLETE_YEARS')])>0)}
                ABLA_ROWS.append({**row,**stat})
        best=p[p.arm.eq(st['arms'][-1]['arm'])].drop_duplicates(
            ['entity_id','origin_year','target_year','arm'],keep='first')
        if len(best):
            for reg,v in best.groupby('region_name'):
                if len(v)>5:TRANSFER_ROWS.append({'study_id':sid,'target':target,'horizon':h,
                    'protocol':proto,'region':reg,'n':len(v),'mae':float(v.absolute_error.mean())})
            for (reg,mt),v in best.groupby(['region_name','mun_type'],dropna=False):
                if len(v)>5:GEO_ROWS.append({'study_id':sid,'target':target,'horizon':h,
                    'protocol':proto,'region':reg,'mun_type_original':mt,'n':len(v),
                    'mae':float(v.absolute_error.mean())})
            for year,v in best.groupby('target_year'):
                ERROR_YEAR_ROWS.append({'study_id':sid,'target':target,'horizon':h,'protocol':proto,
                   'year':int(year),'n':len(v),'mae':float(v.absolute_error.mean())})
ABLA=pd.DataFrame(ABLA_ROWS)
if len(ABLA):
    ABLA['p_holm']=np.nan;ABLA['q_fdr_bh']=np.nan
    for keys,inds in ABLA.groupby(['target','horizon','protocol']).groups.items():
        for is_confirm in (True,False):
            ix=[i for i in inds if bool(ABLA.loc[i,'confirmatory'])==is_confirm]
            if ix:
                p=ABLA.loc[ix,'p_cluster_signflip'].tolist()
                if is_confirm:ABLA.loc[ix,'p_holm']=holm_adjust(p)
                else:ABLA.loc[ix,'q_fdr_bh']=bh_adjust(p)
    ABLA['finding']=np.where((ABLA.ci_low>0)&(
          np.where(ABLA.confirmatory,ABLA.p_holm,ABLA.q_fdr_bh)<0.05),
          'PREDICTIVE_IMPROVEMENT',np.where(ABLA.ci_high<0,'DEGRADATION_OR_NEGATIVE','NOT_CONFIRMED'))
    mark('PASS_VALIDATION','PASS',f'{len(ABLA)} paired group contrasts; region-block CI; Holm confirmatory/FDR exploratory')
else:mark('PASS_VALIDATION','NOT_FEASIBLE','Not enough regions/pairs for block bootstrap')
LOO=ABLA[ABLA.study_id.eq('MAX_AVAILABLE')&ABLA.old_arm.str.startswith('WITHOUT_',na=False)].copy() if len(ABLA) else pd.DataFrame()
REGIONAL_ERRORS=pd.DataFrame(TRANSFER_ROWS);MUNICIPAL_TYPES=pd.DataFrame(GEO_ROWS)
ERROR_YEARS=pd.DataFrame(ERROR_YEAR_ROWS)

# Непосредственно из сохранённых test predictions: независимый пересчёт каждой MAE/RMSE.
PREDICTION_METRIC_QA=pd.DataFrame()
if len(PREDICTIONS) and len(METRICS):
    groups=['study_id','target','horizon','split_id','arm','model','seed']
    pp=PREDICTIONS.copy()
    pp['weight_eff']=pd.to_numeric(pp.population_weight,errors='coerce').fillna(0).clip(lower=0)
    pp['weighted_loss']=pp.absolute_error*pp.weight_eff
    calc=pp.groupby(groups,dropna=False).agg(
      independent_mae=('absolute_error','mean'),
      independent_rmse=('squared_error',lambda z:float(np.sqrt(np.mean(z)))),
      independent_n=('entity_id','size'),sum_weight=('weight_eff','sum'),
      sum_weighted_loss=('weighted_loss','sum')).reset_index().rename(columns={'split_id':'protocol'})
    calc['independent_weighted_mae']=np.where(calc.sum_weight.gt(0),
         calc.sum_weighted_loss/calc.sum_weight,np.nan)
    cmp=METRICS.merge(calc,on=['study_id','target','horizon','protocol','arm','model','seed'],
        how='outer',validate='one_to_one',indicator=True)
    cmp['mae_abs_difference']=(cmp.mae-cmp.independent_mae).abs()
    cmp['rmse_abs_difference']=(cmp.rmse-cmp.independent_rmse).abs()
    cmp['n_difference']=(cmp.n_test-cmp.independent_n).abs()
    cmp['weighted_mae_abs_difference']=(cmp.weighted_mae-cmp.independent_weighted_mae).abs()
    PREDICTION_METRIC_QA=cmp[['study_id','target','horizon','protocol','arm','model','seed',
       '_merge','mae_abs_difference','rmse_abs_difference',
       'weighted_mae_abs_difference','n_difference']].copy()
    passed=(cmp['_merge'].eq('both').all() and
       cmp[['mae_abs_difference','rmse_abs_difference','weighted_mae_abs_difference']].max().max()<1e-10 and
       cmp.n_difference.eq(0).all())
    mark('INDEPENDENT_METRIC_RECONCILIATION','PASS' if passed else 'FAIL',
         f'rows={len(cmp)}; max MAE diff={cmp.mae_abs_difference.max()}')
    ensure(passed,'INDEPENDENT_METRIC_RECONCILIATION_FAILED')

# Correlation diagnostics use TRAIN ONLY and the ORIGINAL training measurements (no post-hoc test selection).
for task in [t for t in TASKS if t['protocol']=='ROLLING_TIME' and t['study']['id']=='MAX_AVAILABLE']:
    tr=task['data'][task['masks']['train']]
    ff=list(dict.fromkeys(G0_FEATURES+sum(GROUP_FEATURES.values(),[])))
    for i,one in enumerate(ff):
        for two in ff[i+1:]:
            if one=='origin_year' or two=='origin_year':continue
            sample=tr[[one,two]].apply(pd.to_numeric,errors='coerce').dropna()
            if len(sample)<50 or min(sample[one].nunique(),sample[two].nunique())<5:continue
            r=sample[one].corr(sample[two],method='spearman')
            if np.isfinite(r) and abs(r)>=0.7:
                CORR_ROWS.append({'target':task['target'],'horizon':task['horizon'],
                  'feature_A':one,'feature_B':two,'n_train_pairs':len(sample),'spearman_r':float(r)})
CORRELATIONS=pd.DataFrame(CORR_ROWS)

# Interpretation of the fitted predictive function, not causal effects.
for (sid,target,h),registry in MODEL_REGISTRY.items():
    if sid not in ('MAX_AVAILABLE','CANONICAL'):continue
    tr=registry['train'];te=registry['test'];model=registry['model']
    Xtr,Xte,used,_=design(tr,te,registry['features'],registry['refs'])
    n=min(len(te),CFG['MAX_IMPORTANCE_ROWS'])
    if n<40:continue
    indices=np.sort(np.random.default_rng(CFG['SEED']).choice(len(te),n,replace=False))
    x=Xte.iloc[indices].copy();y=te[target].to_numpy(float)[indices]
    try:
        benchmark=mean_absolute_error(y,model.predict(x))
        for col in [u for u in used if u!='origin_year'][:CFG['MAX_IMPORTANCE_FEATURES']]:
            losses=[]
            for s in (11,37,71):
                z=x.copy();z[col]=np.random.default_rng(s).permutation(z[col].to_numpy())
                losses.append(float(mean_absolute_error(y,model.predict(z))-benchmark))
            IMP_ROWS.append({'study_id':sid,'target':target,'horizon':h,'feature':col,
                'mae_increase':float(np.mean(losses)),'sd_permutation':float(np.std(losses)),
                'n_test_sample':n,'scope':'predictive_unconditional_permutation'})
        # First-order ALE on observed TRAIN support; percentiles computed without final test outcomes.
        for col in [u for u in used if u.startswith(('G1_','G2_','G3_','G4_'))][:4]:
            observed=Xtr[col].dropna()
            if observed.nunique()<12:continue
            edges=np.unique(np.quantile(observed,np.linspace(0,1,11)))
            if len(edges)<5:continue
            accumulated=0.;raw_rows=[]
            for k in range(len(edges)-1):
                lo,hi=edges[k:k+2]
                mask=Xtr[col].between(lo,hi,inclusive='both')
                if mask.sum()<15:continue
                xl=Xtr.loc[mask].copy();xh=xl.copy();xl[col]=lo;xh[col]=hi
                inc=float(np.mean(model.predict(xh)-model.predict(xl)))
                accumulated+=inc
                raw_rows.append((k,(lo+hi)/2.,inc,accumulated,int(mask.sum())))
            if raw_rows:
                center=float(np.average([v[3] for v in raw_rows],weights=[v[4] for v in raw_rows]))
                for k,middle,inc,total,nn in raw_rows:
                    ALE_ROWS.append({'study_id':sid,'target':target,'horizon':h,'feature':col,
                      'bin':k,'feature_mid_logscale':middle,'ale_centered':total-center,'n_train':nn,
                      'scope':'train_support_first_order_ALE_not_causal'})
    except Exception as ex:log('importance/ALE '+sid+': '+type(ex).__name__,'WARN')
IMPORTANCE=pd.DataFrame(IMP_ROWS);ALE=pd.DataFrame(ALE_ROWS)

# Additional interpretable evidence: SHAP for best available tree model, on held-out data.
try:
    import shap
    SHAP_AVAILABLE=True
except Exception as ex:
    SHAP_AVAILABLE=False;log('SHAP import unavailable: '+type(ex).__name__,'WARN')
if SHAP_AVAILABLE:
    for (sid,target,h),r in MODEL_REGISTRY.items():
        if sid!='MAX_AVAILABLE' or r['family'] not in ('HGB','CATBOOST','LIGHTGBM'):continue
        try:
            xtr,xtest,used,_=design(r['train'],r['test'],r['features'],r['refs'])
            take=min(len(xtest),CFG['MAX_SHAP_ROWS']);xtest=xtest.iloc[:take]
            tree=shap.TreeExplainer(r['model'])
            values=tree.shap_values(xtest)
            if isinstance(values,list):values=values[0]
            values=np.asarray(values)
            ensure(values.shape==(len(xtest),xtest.shape[1]),'SHAP_SHAPE_MISMATCH')
            for j,name in enumerate(xtest.columns):
                SHAP_ROWS.append({'study_id':sid,'target':target,'horizon':h,'model':r['family'],
                  'feature':name,'mean_abs_shap':float(np.abs(values[:,j]).mean()),
                  'n_test_sample':take,'scope':'explanation_of_prediction_not_causal'})
        except Exception as ex:log('SHAP computation '+sid+': '+type(ex).__name__+' '+str(ex)[:120],'WARN')
SHAP_IMPORTANCE=pd.DataFrame(SHAP_ROWS)
mark('SHAP_QA','PASS' if len(SHAP_IMPORTANCE) else 'NOT_FEASIBLE',
     f'tree SHAP cells={len(SHAP_IMPORTANCE)}; optional where compatible')

# Train-only two-way fixed effects: identifiability / cluster diagnostics, never causal.
FE_QA_ROWS=[]
try:
    from linearmodels.panel import PanelOLS
    PANEL_OLS_AVAILABLE=True
except ImportError:
    PANEL_OLS_AVAILABLE=False

def _fit_fe_train_only(task):
    tr=task['data'].loc[task['masks']['train']].copy()
    report={'target':task['target'],'horizon':task['horizon'],
      'n_initial':int(len(tr)),'n_complete':0,'n_entities':0,'n_regions':0,
      'n_years':0,'n_regressors':0,'status':'NOT_FEASIBLE','reason':''}
    if not PANEL_OLS_AVAILABLE:
        report['reason']='LINEARMODELS_NOT_INSTALLED';return pd.DataFrame(),report
    available=[]
    for g in ('G1','G2','G3','G4','G5'):
        ff=[f for f in GROUP_FEATURES.get(g,[]) if f in tr]
        ff.sort(key=lambda f:tr[f].notna().sum(),reverse=True)
        if ff:available.append(ff[0])
    if not available:
        report['reason']='NO_FE_COVARIATES';return pd.DataFrame(),report
    columns=['entity_id','origin_year','region_name',task['target']]+available
    w=tr[columns].copy()
    for f in available:w[f]=pd.to_numeric(w[f],errors='coerce')
    w=w.dropna()
    counts=w.groupby('entity_id').origin_year.nunique()
    w=w[w.entity_id.isin(counts[counts.ge(CFG['MIN_FE_ENTITY_OBSERVATIONS'])].index)].copy()
    report.update(n_complete=int(len(w)),n_entities=int(w.entity_id.nunique()),
        n_regions=int(w.region_name.nunique()),n_years=int(w.origin_year.nunique()))
    if len(w)<400 or w.entity_id.nunique()<20 or w.region_name.nunique()<8 or w.origin_year.nunique()<4:
        report['reason']='INSUFFICIENT_CONNECTED_FE_PANEL';return pd.DataFrame(),report
    # Drop regressors absorbed by entity or time effects: residual variance of two-way demeaned x.
    keys=['entity_id','origin_year']
    X=pd.DataFrame(index=w.index)
    for f in available:
        x=np.sign(w[f])*np.log1p(np.abs(w[f]))
        residual=x-x.groupby(w.entity_id).transform('mean')-x.groupby(w.origin_year).transform('mean')+x.mean()
        if residual.std()>1e-8:X[f]=x
    if X.empty:
        report['reason']='ALL_COVARIATES_ABSORBED_BY_FIXED_EFFECTS';return pd.DataFrame(),report
    residualized=pd.DataFrame(index=w.index)
    for f in X:
        a=X[f]
        residualized[f]=a-a.groupby(w.entity_id).transform('mean')-a.groupby(w.origin_year).transform('mean')+a.mean()
    rank=np.linalg.matrix_rank(residualized.to_numpy(float),tol=1e-9)
    if rank<len(residualized.columns):
        selected=[]
        for f in residualized.columns:
            candidate=selected+[f]
            if np.linalg.matrix_rank(residualized[candidate].to_numpy(float),tol=1e-9)==len(candidate):
                selected.append(f)
        X=X[selected]
    report['n_regressors']=len(X.columns)
    if X.empty:
        report['reason']='COLLINEAR_WITHIN_FE';return pd.DataFrame(),report
    aligned=w.set_index(keys).sort_index()
    x=X.loc[aligned.reset_index().index] if False else pd.DataFrame(
       X.to_numpy(float),index=pd.MultiIndex.from_frame(w[keys]),columns=X.columns).sort_index()
    y=aligned[task['target']].astype(float)
    groups=pd.DataFrame({'region_cluster':pd.factorize(aligned.region_name.astype(str))[0]},index=aligned.index)
    try:
        model=PanelOLS(y,x,entity_effects=True,time_effects=True,drop_absorbed=True,check_rank=False)
        fit=model.fit(cov_type='clustered',clusters=groups)
        rows=pd.DataFrame([{'target':task['target'],'horizon':task['horizon'],'parameter':f,
           'coefficient':float(fit.params[f]),'std_error_region_cluster':float(fit.std_errors[f]),
           'p_value':float(fit.pvalues[f]),'n_train':int(len(w)),
           'n_municipalities':int(w.entity_id.nunique()),'n_clusters':int(w.region_name.nunique()),
           'interpretation':'train-only two-way FE; predictive association, NOT causal'}
          for f in fit.params.index])
        report.update(status='PASS',reason='TWO_WAY_FE_TRAIN_ONLY_REGION_CLUSTERED')
        return rows,report
    except Exception as ex:
        report['reason']='FIT_EXCEPTION_'+type(ex).__name__+': '+str(ex)[:200]
        return pd.DataFrame(),report

if PANEL_OLS_AVAILABLE:
    for task in TASKS:
        if task['study']['id']!='MAX_AVAILABLE' or task['protocol']!='ROLLING_TIME' or task['horizon']!=1:
            continue
        rows,qa=_fit_fe_train_only(task)
        FE_QA_ROWS.append(qa)
        if not rows.empty:FE_ROWS.extend(rows.to_dict('records'))
else:
    FE_QA_ROWS.append({'status':'NOT_FEASIBLE','reason':'LINEARMODELS_UNAVAILABLE'})
FE_QA=pd.DataFrame(FE_QA_ROWS)
FE_DIAGNOSTICS=pd.DataFrame(FE_ROWS)
mark('FE_DIAGNOSTICS','PASS' if len(FE_DIAGNOSTICS) else 'NOT_FEASIBLE',
     f'train-only FE parameter rows={len(FE_DIAGNOSTICS)}, audited attempts={len(FE_QA)}')
# Exploratory interactions (H3.2), identical test labels/keys, no post-hoc causal claim.
# Non-arbitrary G6: semantically identified wage/income by housing access or stock.
def _feature_for_domain(group,domains):
    rows=FEATURE_DICT[FEATURE_DICT.group.eq(group)&
       FEATURE_DICT.domain.isin(domains)&FEATURE_DICT.model_feature.notna()]
    for record in rows.to_dict('records'):
        if record['model_feature'] in GROUP_FEATURES[group]:return record['model_feature']
    return None
G6_ECON_FEATURE=_feature_for_domain('G1',('income','wages'))
G6_HOUSING_FEATURE=_feature_for_domain('G2',('housing_access','housing_stock','housing_construction'))
if G6_ECON_FEATURE and G6_HOUSING_FEATURE:
    g1,g2=G6_ECON_FEATURE,G6_HOUSING_FEATURE
    for (sid,target,h),reg in MODEL_REGISTRY.items():
        if sid!='MAX_AVAILABLE':continue
        tr=reg['train'].copy();te=reg['test'].copy()
        inter='G6_income_housing_interaction'
        signed_log=lambda x:np.sign(pd.to_numeric(x,errors='coerce'))*np.log1p(np.abs(pd.to_numeric(x,errors='coerce')))
        tr[inter]=signed_log(tr[g1])*signed_log(tr[g2]);te[inter]=signed_log(te[g1])*signed_log(te[g2])
        if tr[inter].notna().sum()<CFG['MIN_TRAIN_ROWS']:continue
        try:
            pred,_,_,_=fit_predict(reg['family'],tr,te,target,reg['features']+[inter],CFG['SEED'],reg['params'])
            ref=metrics(te[target],reg['pred'],te[f'hist_population_l{CFG["FEATURE_PUBLICATION_LAG"]}'])
            cur=metrics(te[target],pred,te[f'hist_population_l{CFG["FEATURE_PUBLICATION_LAG"]}'])
            INTERACTION_ROWS.append({'target':target,'horizon':h,'study_id':sid,
              'factor_A':g1,'factor_B':g2,'base_mae':ref['mae'],'interaction_mae':cur['mae'],
              'delta_mae':ref['mae']-cur['mae'],'n_test':len(te),
              'status':'EXPLORATORY_PREDICTIVE_NOT_CAUSAL'})
        except Exception as ex:log('interaction '+str(h)+': '+type(ex).__name__,'WARN')
INTERACTIONS=pd.DataFrame(INTERACTION_ROWS)
mark('H3.2','PASS' if len(INTERACTIONS) else 'NOT_FEASIBLE',
     f'{len(INTERACTIONS)} semantically anchored income/wages × housing comparisons')
mark('CAUSAL_MODULE','CAUSAL_NOT_IDENTIFIED','No validated treatment register with dates, exposure, and controls')


# ADVANCED · V4.00 · ENSEMBLE SHAP, ADDITIVITY, GROUP PERMUTATION, ALE/PDP/ICE, REGION STABILITY
# All explanations refer to frozen predictors trained without TEST labels. Predictive, never causal.
ADVANCED_SHAP_ROWS=[];ADVANCED_SHAP_LOCAL_ROWS=[];ADVANCED_SHAP_QA_ROWS=[]
ADVANCED_GROUP_PERM_ROWS=[];ADVANCED_PDP_ROWS=[];ADVANCED_ICE_ROWS=[];ADVANCED_SHAP_CI_ROWS=[];ADVANCED_SHAP_STABILITY_ROWS=[];ADVANCED_SHAP_INTERACTION_ROWS=[];ADVANCED_LOO_ROWS=[]
try:import shap as _shap
except Exception as _e:
    _shap=None;log('Advanced SHAP import failed: '+str(_e)[:150],'WARN')

def _explain_one(m,x,bg):
    model=m['model'];family=m['family'];cols=list(x.columns)
    if family=='RIDGE':
        prep=model[:-1];head=model[-1];tx=np.asarray(prep.transform(x),float);tb=np.asarray(prep.transform(bg),float)
        mean=tb.mean(axis=0);co=np.asarray(head.coef_,float)
        v=(tx-mean)*co;bias=np.full(len(x),float(head.intercept_+mean@co))
        names=list(cols)+[f'missing_indicator_{int(k)}' for k in prep[0].indicator_.features_] if v.shape[1]>len(cols) else list(cols)
        return v,bias,names,'ANALYTICAL_LINEAR'
    if family=='EXTRA_TREES':
        prep=model[:-1];tree=model[-1];xx=np.asarray(prep.transform(x),float)
        names=list(cols)+[f'missing_indicator_{int(k)}' for k in prep[0].indicator_.features_] if xx.shape[1]>len(cols) else list(cols)
        if _shap is None:raise RuntimeError('SHAP_UNAVAILABLE_FOR_EXTRA_TREES')
        expl=_shap.TreeExplainer(tree,feature_perturbation='tree_path_dependent')
        values=np.asarray(expl.shap_values(xx),float)
        bias=np.broadcast_to(np.asarray(expl.expected_value,float).reshape(-1)[0],len(x))
        return values,bias,names,'TREE_PATH_DEPENDENT'
    if family=='XGBOOST':
        import xgboost
        dm=xgboost.DMatrix(x,feature_names=cols)
        raw=np.asarray(model.get_booster().predict(dm,pred_contribs=True),float)
        ensure(raw.shape==(len(x),len(cols)+1),'XGB_SHAP_CONTRIBS_WRONG_SHAPE')
        return raw[:,:-1],raw[:,-1],cols,'NATIVE_TREE_SHAP'
    if family=='LIGHTGBM':
        raw=np.asarray(model.predict(x,pred_contrib=True),float)
        ensure(raw.shape==(len(x),len(cols)+1),'LGB_SHAP_CONTRIBS_WRONG_SHAPE')
        return raw[:,:-1],raw[:,-1],cols,'NATIVE_TREE_SHAP'
    if family=='CATBOOST':
        from catboost import Pool
        raw=np.asarray(model.get_feature_importance(Pool(x),type='ShapValues'),float)
        ensure(raw.shape==(len(x),len(cols)+1),'CATBOOST_SHAP_CONTRIBS_WRONG_SHAPE')
        return raw[:,:-1],raw[:,-1],cols,'NATIVE_TREE_SHAP'
    if _shap is None:raise RuntimeError('SHAP_NOT_INSTALLED_FOR_'+family)
    expl=_shap.TreeExplainer(model,feature_perturbation='tree_path_dependent')
    values=np.asarray(expl.shap_values(x),float)
    return values,np.broadcast_to(float(np.asarray(expl.expected_value).reshape(-1)[0]),len(x)),cols,'TREE_PATH_DEPENDENT'

def _adv_predict(reg,x):
    pp=[]
    for m in reg['members']:
        pp.append(m['weight']*np.asarray(m['model'].predict(x),float))
    return np.sum(pp,axis=0)

def _adv_feature_group(feature):
    if feature.startswith('G1R_'):return 'G1R_REGIONAL_NOT_MUNICIPAL'
    for g in ('G1','G2','G3','G4','G5'):
        if feature.startswith(g+'_'):return g
    if feature.startswith('region_'):return 'REGION_FIXED_DUMMIES'
    if feature.startswith('hist_'):return 'G0_DEMOGRAPHIC_HISTORY'
    if feature=='origin_year':return 'G0_TIME'
    return 'OTHER'

def _within_strata_permute(X,source,cols,seed):
    rng=np.random.default_rng(seed);res=X.copy();groups=source[['region_name','target_year']].reset_index(drop=True)
    ensure(len(groups)==len(X),'SHUFFLE_STRATA_ALIGNMENT_INVALID')
    # Joint permutation preserves correlations WITHIN each named feature block.
    # Within region-year strata when >=2; isolated observations remain unchanged.
    for _,group in groups.groupby(['region_name','target_year'],dropna=False).groups.items():
        ids=np.asarray(list(group),dtype=int)
        if len(ids)<2:continue
        shuffle=rng.permutation(ids)
        res.iloc[ids,res.columns.get_indexer(cols)]=X.iloc[shuffle,X.columns.get_indexer(cols)].to_numpy()
    return res

def _adv_explain_registry(reg):
    sid=reg['study_id'];target=reg['target'];h=reg['horizon'];tr=reg['train'];te=reg['test']
    cols=reg['members'][0]['columns']
    # Design by the exact fitted train population; no fitted statistics from target TEST.
    _,XX,_,_=design(tr,te,reg['features'],reg['members'][0]['refs'])
    indices=np.sort(np.random.default_rng(CFG['SEED']+h).choice(len(XX),
                  size=min(CFG['ADVANCED_SHAP_ROWS'],len(XX)),replace=False))
    x=XX.iloc[indices].reset_index(drop=True);meta=te.iloc[indices].reset_index(drop=True)
    background=tr.sample(n=min(CFG['ADVANCED_BACKGROUND_ROWS'],len(tr)),random_state=CFG['SEED'])
    # Design shares the same fitted train population for test and background.
    _,BACK,_,_=design(tr,background,reg['features'],reg['members'][0]['refs'])
    ensure(list(x.columns)==list(BACK.columns),'SHAP_BACKGROUND_COLUMN_DIFFERENCE')
    contributions=np.zeros((len(x),len(cols)),float);bline=np.zeros(len(x),float);ok_members=0
    fam_rows=[]
    for member in reg['members']:
        fam=member['family'];weight=member['weight']
        try:
            XX_m=X if False else x
            vals,baseval,names,method=_explain_one(member,XX_m,BACK)
            ensure(vals.shape==(len(x),len(names)),'SHAP_MEMBER_MATRIX_SHAPE')
            reconstructed=np.asarray(baseval,float)+vals.sum(axis=1)
            direct=np.asarray(member['model'].predict(XX_m),float)
            discrepancy=np.max(np.abs(reconstructed-direct))
            ensure(discrepancy<CFG['ADVANCED_SHAP_TOLERANCE'],'SHAP_MEMBER_ADDITIVITY_'+fam)
            for j,feature in enumerate(names):
                fam_rows.append({'study_id':sid,'target':target,'horizon':h,'method':'SHAP_MEMBER',
                    'model':fam,'feature':feature,'group':_adv_feature_group(feature),
                    'mean_abs_shap':float(np.abs(vals[:,j]).mean()),
                    'mean_signed_shap':float(vals[:,j].mean()),'ensemble_weight':weight,
                    'n_explained':len(x),'mode':method,'scope':'PREDICTIVE_NOT_CAUSAL'})
                if feature in cols:contributions[:,cols.index(feature)]+=weight*vals[:,j]
                else:
                    # Imputer missing indicators contribute jointly to their original input.
                    if feature.startswith('missing_indicator_'):
                        nmissing=int(feature.split('_')[-1]);
                        if nmissing<len(cols):contributions[:,nmissing]+=weight*vals[:,j]
                    else:raise RuntimeError('UNRECOGNIZED_SHAP_INTERNAL_FEATURE_'+feature)
            bline+=weight*np.asarray(baseval,float)
            ADVANCED_SHAP_QA_ROWS.append({'study_id':sid,'target':target,'horizon':h,'member':fam,
                'max_additivity_abs_error':float(discrepancy),'n_explained':len(x),'status':'PASS','mode':method})
            ok_members+=1
        except Exception as ex:
            ADVANCED_SHAP_QA_ROWS.append({'study_id':sid,'target':target,'horizon':h,'member':fam,
                'max_additivity_abs_error':np.nan,'n_explained':len(x),'status':'FAIL',
                'reason':type(ex).__name__+': '+str(ex)[:160]})
            log('SHAP member '+sid+'/'+fam+': '+str(ex)[:160],'ERROR')
    if ok_members!=len(reg['members']):return 0
    direct=_adv_predict(reg,x)
    diff=float(np.max(np.abs(bline+contributions.sum(axis=1)-direct)))
    ensure(diff<CFG['ADVANCED_SHAP_TOLERANCE'],'SHAP_ENSEMBLE_ADDITIVITY_FAILED')
    ADVANCED_SHAP_QA_ROWS.append({'study_id':sid,'target':target,'horizon':h,'member':'WEIGHTED_ENSEMBLE',
        'max_additivity_abs_error':diff,'n_explained':len(x),'status':'PASS','mode':'WEIGHTED_MEMBER_SUM'})
    for j,feature in enumerate(cols):
        rho=np.nan
        if x[feature].nunique(dropna=True)>5:
            rho=float(pd.Series(x[feature]).corr(pd.Series(contributions[:,j]),method='spearman'))
        fam_rows.append({'study_id':sid,'target':target,'horizon':h,'method':'SHAP_ENSEMBLE',
            'model':'SUPERLEARNER','feature':feature,'group':_adv_feature_group(feature),
            'mean_abs_shap':float(np.abs(contributions[:,j]).mean()),
            'mean_signed_shap':float(contributions[:,j].mean()),'shap_value_spearman_r':rho,
            'ensemble_weight':1.,'n_explained':len(x),'mode':'WEIGHTED_MEMBER_SUM',
            'scope':'PREDICTIVE_NOT_CAUSAL'})
        for i in range(len(x)):
            if abs(contributions[i,j])>1e-9:
                ADVANCED_SHAP_LOCAL_ROWS.append({'study_id':sid,'target':target,'horizon':h,
                    'entity_id':meta.entity_id.iloc[i],'target_year':int(meta.target_year.iloc[i]),
                    'region_name':meta.region_name.iloc[i],'feature':feature,
                    'feature_value':float(x[feature].iloc[i]) if pd.notna(x[feature].iloc[i]) else np.nan,
                    'shap_value':float(contributions[i,j]),'baseline':float(bline[i]),
                    'prediction':float(direct[i]),'method':'WEIGHTED_MEMBER_SUM'})
    ADVANCED_SHAP_ROWS.extend(fam_rows)
    # Region-clustered confidence intervals for mean |SHAP|; seeds never treated as independent units.
    rng=np.random.default_rng(CFG['SEED']+h+19)
    region=meta.region_name.fillna('UNKNOWN').astype(str).to_numpy()
    ids=[np.flatnonzero(region==s) for s in np.unique(region)]
    if len(ids)>=CFG['MIN_BOOTSTRAP_REGIONS']:
        boot=np.empty((CFG['ADVANCED_EXPLAIN_BOOTSTRAPS'],len(cols)))
        for j in range(len(boot)):
            draw=rng.integers(0,len(ids),len(ids));part=np.concatenate([ids[k] for k in draw]);boot[j]=np.mean(np.abs(contributions[part]),axis=0)
        center=np.abs(contributions).mean(axis=0);top_k=min(5,len(center));global_top=set(np.argsort(center)[-top_k:].tolist())
        for j,feature in enumerate(cols):
            probability=float(np.mean([j in np.argsort(boot[b])[-top_k:] for b in range(len(boot))]))
            ADVANCED_SHAP_STABILITY_ROWS.append({'study_id':sid,'target':target,'horizon':h,
                'feature':feature,'top_k':top_k,'bootstrap_in_top_k_share':probability,
                'base_rank':int(len(center)-np.argsort(center).tolist().index(j)),
                'n_region_resamples':len(boot),'n_regions':len(ids),'scope':'SHAP_RANK_STABILITY_NONCAUSAL'})
        for j,feature in enumerate(cols):
            ADVANCED_SHAP_CI_ROWS.append({'study_id':sid,'target':target,'horizon':h,'feature':feature,
                'mean_abs_shap':float(np.abs(contributions[:,j]).mean()),
                'lo95':float(np.quantile(boot[:,j],0.025)),'hi95':float(np.quantile(boot[:,j],0.975)),
                'n_regions':len(ids),'n_bootstrap':len(boot),'unit':'REGION'})
    # Model-native interactions. XGBoost returns true pairwise SHAP interactions;
    # CatBoost's Interaction metric is a different, structural split score (marked explicitly).
    for member in reg['members']:
        try:
            if member['family']=='XGBOOST':
                import xgboost
                matrix=xgboost.DMatrix(x.iloc[:min(45,len(x))],feature_names=cols)
                shap_pairs=np.asarray(member['model'].get_booster().predict(matrix,pred_interactions=True),float)
                ensure(shap_pairs.shape[1:]==(len(cols)+1,len(cols)+1),'XGB_INTERACTION_SHAPE')
                for jj in range(len(cols)):
                    for kk in range(jj+1,len(cols)):
                        strength=float(2*np.mean(np.abs(shap_pairs[:,jj,kk])))
                        if strength>0:
                            ADVANCED_SHAP_INTERACTION_ROWS.append({'study_id':sid,'target':target,'horizon':h,
                                'family':'XGBOOST','feature_a':cols[jj],'feature_b':cols[kk],
                                'interaction_strength':strength,'definition':'TWO_TIMES_MEAN_ABS_PAIRWISE_SHAP',
                                'n_sample':len(shap_pairs),'predictive_only':True})
            elif member['family']=='CATBOOST':
                rank=np.asarray(member['model'].get_feature_importance(type='Interaction'),float)
                if rank.ndim==2 and rank.shape[1]>=3:
                    for pair in rank[:100]:
                        jj,kk=int(pair[0]),int(pair[1])
                        if 0<=jj<len(cols) and 0<=kk<len(cols) and jj!=kk:
                            ADVANCED_SHAP_INTERACTION_ROWS.append({'study_id':sid,'target':target,'horizon':h,
                                'family':'CATBOOST','feature_a':cols[jj],'feature_b':cols[kk],
                                'interaction_strength':float(pair[2]),
                                'definition':'CATBOOST_NATIVE_SPLIT_INTERACTION_NOT_SHAP',
                                'n_sample':len(x),'predictive_only':True})
        except Exception as ex:log('Interaction audit '+sid+'/'+member['family']+': '+type(ex).__name__,'WARN')
    # Grouped conditional permutation on held-out TEST; no re-training and no test selection.
    n=min(len(te),CFG['ADVANCED_PERM_ROWS']);take=np.sort(np.random.default_rng(CFG['SEED']+37).choice(len(te),n,replace=False))
    sample=te.iloc[take].reset_index(drop=True)
    _,mx,_,_=design(tr,sample,reg['features'],reg['members'][0]['refs'])
    before=np.abs(sample[target].to_numpy(float)-_adv_predict(reg,mx))
    groups={g:[f for f in mx.columns if _adv_feature_group(f)==g] for g in
            ('G1','G2','G3','G4','G5','G1R_REGIONAL_NOT_MUNICIPAL','G0_DEMOGRAPHIC_HISTORY')}
    for g,ff in groups.items():
        if not ff:continue
        altered=[];effective_counts=[]
        for rep in range(CFG['ADVANCED_PERM_REPEATS']):
            perm=_within_strata_permute(mx,sample,ff,CFG['SEED']+h*31+rep)
            changed=int((perm[ff].fillna(-999998)!=mx[ff].fillna(-999998)).any(axis=1).sum())
            effective_counts.append(changed)
            if changed:altered.append(np.abs(sample[target].to_numpy(float)-_adv_predict(reg,perm)))
        if not altered:
            ADVANCED_GROUP_PERM_ROWS.append({'study_id':sid,'target':target,'horizon':h,'group':g,
                'status':'NOT_FEASIBLE','reason':'NO_EXCHANGEABLE_REGION_YEAR_STRATA','n_permuted':0});continue
        pert=np.mean(np.column_stack(altered),axis=1)
        z=pd.DataFrame({'region_name':sample.region_name,'target_year':sample.target_year,
                        'perturbed_error':pert,'base_error':before})
        delta=float(np.mean(pert-before))
        stat=block_delta(z,'perturbed_error','base_error',CFG['ADVANCED_EXPLAIN_BOOTSTRAPS'],CFG['SEED']+h)
        ADVANCED_GROUP_PERM_ROWS.append({'study_id':sid,'target':target,'horizon':h,'group':g,
            'n_test_sample':len(sample),'n_permuted':int(np.mean(effective_counts)),
            'n_features':len(ff),'delta_mae':delta,'ci_low':stat['ci_low'] if stat else np.nan,
            'ci_high':stat['ci_high'] if stat else np.nan,'n_regions':stat['n_regions'] if stat else 0,
            'repetitions':len(altered),'status':'PASS','method':'JOINT_REGION_YEAR_STRATIFIED_PERMUTATION',
            'interpretation':'Predictive reliance, not causal importance'})
    # PDP/ICE on training domain only to minimize post-test interpretation bias.
    strongest=np.argsort(np.abs(contributions).mean(axis=0))[::-1]
    train_sample=tr.sample(n=min(CFG['ADVANCED_PDP_TRAIN_ROWS'],len(tr)),random_state=CFG['SEED'])
    _,xt,_,_=design(tr,train_sample,reg['features'],reg['members'][0]['refs'])
    for j in strongest[:CFG['ADVANCED_PDP_TOP']]:
        feature=cols[j]
        if feature.startswith('region_') or not xt[feature].notna().any() or xt[feature].nunique()<8:continue
        knots=np.unique(np.quantile(xt[feature].dropna().to_numpy(float),np.linspace(.05,.95,CFG['ADVANCED_PDP_BINS'])))
        chosen=np.sort(np.random.default_rng(CFG['SEED']).choice(len(xt),min(12,len(xt)),replace=False))
        for k,v in enumerate(knots):
            xx=xt.copy();xx[feature]=float(v);yp=_adv_predict(reg,xx)
            ADVANCED_PDP_ROWS.append({'study_id':sid,'target':target,'horizon':h,'feature':feature,
                'bin':k,'feature_value_transformed':float(v),'mean_prediction':float(yp.mean()),
                'n_train_background':len(xt),'method':'TRAIN_BACKGROUND_PDP_MARGINAL_NOT_CAUSAL'})
            for ix in chosen:
                ADVANCED_ICE_ROWS.append({'study_id':sid,'target':target,'horizon':h,'feature':feature,
                    'train_sample_index':int(ix),'feature_value_transformed':float(v),
                    'prediction':float(yp[ix]),'method':'ICE_TRAIN_BACKGROUND_NONCAUSAL'})
    return len(x)

progress('STATS',1,1,'Статистика и факторные вклады рассчитаны')
if CFG['ADVANCED_EXPLAIN_ENABLE']:
    keys=list(ADVANCED_ENSEMBLE_REGISTRY)
    if CFG['MODE']=='SMOKE':keys=keys[:2]
    else:keys=[k for k in keys if k[0] in ('MAX_AVAILABLE','CANONICAL') or k[0].startswith('AUX_')]
    progress('SHAP',0,len(keys),'SHAP, permutation, PDP/ICE')
    _shap_done=0
    for key in keys:
        try:_adv_explain_registry(ADVANCED_ENSEMBLE_REGISTRY[key])
        except Exception as ex:
            ADVANCED_SHAP_QA_ROWS.append({'study_id':key[0],'target':key[1],'horizon':key[2],
                'member':'ENSEMBLE_OVERALL','status':'FAIL','reason':type(ex).__name__+': '+str(ex)[:200]})
            log('ADV EXPLANATION '+str(key)+': '+type(ex).__name__+': '+str(ex)[:200],'ERROR')
        _shap_done+=1
        progress('SHAP',_shap_done,len(keys),f'Моделей объяснено: {_shap_done}/{len(keys)}')
    progress('SHAP',1,1,'Объяснения сохранены')
if not CFG['ADVANCED_EXPLAIN_ENABLE']:progress('SHAP',1,1,'Объяснения отключены настройкой')
ADVANCED_SHAP=pd.DataFrame(ADVANCED_SHAP_ROWS);ADVANCED_SHAP_LOCAL=pd.DataFrame(ADVANCED_SHAP_LOCAL_ROWS)
ADVANCED_SHAP_QA=pd.DataFrame(ADVANCED_SHAP_QA_ROWS);ADVANCED_GROUP_PERM=pd.DataFrame(ADVANCED_GROUP_PERM_ROWS)
ADVANCED_PDP=pd.DataFrame(ADVANCED_PDP_ROWS);ADVANCED_ICE=pd.DataFrame(ADVANCED_ICE_ROWS)
ADVANCED_SHAP_CI=pd.DataFrame(ADVANCED_SHAP_CI_ROWS)
ADVANCED_SHAP_STABILITY=pd.DataFrame(ADVANCED_SHAP_STABILITY_ROWS)
ADVANCED_SHAP_INTERACTIONS=pd.DataFrame(ADVANCED_SHAP_INTERACTION_ROWS)
# V4.02: independent, paired test against the strongest SINGLE family selected on DEV.
# Only predictions with the same municipality, target year and task enter inference.
ADVANCED_SINGLE_DUEL_ROWS=[]
if len(PREDICTIONS) and len(ADVANCED_ENSEMBLES):
    for row in ADVANCED_ENSEMBLES.loc[ADVANCED_ENSEMBLES.status.eq('PASS')].itertuples():
        family=getattr(row,'best_single_family_dev_locked',None)
        if not family:continue
        subset=PREDICTIONS.loc[
          PREDICTIONS.study_id.eq(row.study_id)&PREDICTIONS.target.eq(row.target)&
          PREDICTIONS.horizon.eq(row.horizon)&PREDICTIONS.split_id.eq(row.protocol)&
          PREDICTIONS.model.isin(['SUPERLEARNER','SL_'+family])]
        if subset.empty:continue
        z=subset.pivot_table(index=['entity_id','origin_year','target_year','region_name'],
              columns='model',values='absolute_error',aggfunc='first').reset_index()
        single='SL_'+family
        if single not in z.columns or 'SUPERLEARNER' not in z.columns:continue
        stat=block_delta(z,single,'SUPERLEARNER',CFG['BOOTSTRAPS'],
                         CFG['SEED']+int(row.horizon)+sum(ord(c) for c in str(row.study_id)))
        if stat:ADVANCED_SINGLE_DUEL_ROWS.append({
            'study_id':row.study_id,'target':row.target,'horizon':int(row.horizon),
            'protocol':row.protocol,'best_single_family_dev_locked':family,
            'comparison':'DEV_BEST_SINGLE → SUPERLEARNER','confirmatory':False,
            'model_selected_without_test_labels':True,**stat})
ADVANCED_SINGLE_DUEL=pd.DataFrame(ADVANCED_SINGLE_DUEL_ROWS)
if len(ADVANCED_SINGLE_DUEL):
    ADVANCED_SINGLE_DUEL['q_fdr_bh']=np.nan
    for _,inds in ADVANCED_SINGLE_DUEL.groupby(['target','horizon','protocol']).groups.items():
        ADVANCED_SINGLE_DUEL.loc[list(inds),'q_fdr_bh']=bh_adjust(
             ADVANCED_SINGLE_DUEL.loc[list(inds),'p_cluster_signflip'])
mark('ADVANCED_VS_BEST_SINGLE','PASS' if len(ADVANCED_SINGLE_DUEL)==int(ADVANCED_ENSEMBLES.status.eq('PASS').sum()) else 'PARTIAL',
     f'paired={len(ADVANCED_SINGLE_DUEL)}/{int(ADVANCED_ENSEMBLES.status.eq("PASS").sum())}; regional bootstrap; DEV locked')

# Independently test SuperLearner vs historical M0 on identical keys/regions; exploratory only.
if len(PREDICTIONS):
    subset=PREDICTIONS[PREDICTIONS.model.isin(['HISTORICAL','SUPERLEARNER'])]
    for (sid,targ,h,proto),grp in subset.groupby(['study_id','target','horizon','split_id']):
        z=grp.pivot_table(index=['entity_id','origin_year','target_year','region_name'],columns='model',
                          values='absolute_error',aggfunc='first').reset_index()
        if 'HISTORICAL' not in z or 'SUPERLEARNER' not in z:continue
        stat=block_delta(z,'HISTORICAL','SUPERLEARNER',CFG['BOOTSTRAPS'],CFG['SEED']+h)
        if stat:ADVANCED_LOO_ROWS.append({'study_id':sid,'target':targ,'horizon':h,'protocol':proto,
            'comparison':'M0_HISTORICAL → SUPERLEARNER','confirmatory':False,**stat})
ADVANCED_COMPARISONS=pd.DataFrame(ADVANCED_LOO_ROWS)
if len(ADVANCED_COMPARISONS):
    ADVANCED_COMPARISONS['q_fdr_bh']=np.nan
    for _,inds in ADVANCED_COMPARISONS.groupby(['target','horizon','protocol']).groups.items():
        ADVANCED_COMPARISONS.loc[list(inds),'q_fdr_bh']=bh_adjust(ADVANCED_COMPARISONS.loc[list(inds),'p_cluster_signflip'])
    ADVANCED_COMPARISONS['significant_improvement']=(ADVANCED_COMPARISONS.delta_mae.gt(0)&
        ADVANCED_COMPARISONS.ci_low.gt(0)&ADVANCED_COMPARISONS.q_fdr_bh.lt(0.05))
mark('ADVANCED_SHAP_QA','PASS' if len(ADVANCED_SHAP_QA) and ADVANCED_SHAP_QA.status.eq('PASS').all() else 'PARTIAL',
     f'additive explainers={int(ADVANCED_SHAP_QA.status.eq("PASS").sum()) if len(ADVANCED_SHAP_QA) else 0}; '
     f'failures={int(ADVANCED_SHAP_QA.status.eq("FAIL").sum()) if len(ADVANCED_SHAP_QA) else 0}; '
     f'region-block CI={len(ADVANCED_SHAP_CI)}; interactions={len(ADVANCED_SHAP_INTERACTIONS)}; perm={len(ADVANCED_GROUP_PERM)}')
quiet_print('Advanced explanations: SHAP global',len(ADVANCED_SHAP),'local',len(ADVANCED_SHAP_LOCAL),
      'perm group',len(ADVANCED_GROUP_PERM),'PDP',len(ADVANCED_PDP),'ICE',len(ADVANCED_ICE))
```

## 9. Дополнительные проверки устойчивости

Повторная оценка при увеличении лага публикации на год и удалении сильно коррелированных прокси выполняется с теми же тестовыми ключами. Настоящий `future-mutation canary` изменяет будущие значения признаков и проверяет неизменность доступных на дату прогноза данных. Неподтверждённые эффекты сохраняются, а не исключаются из отчётности.
Проверка future-mutation повторно формирует признаки из изменённой копии панели и сравнивает **конечный прогноз уже обученной модели**, а не только лаговое соединение.


```python
# 10 · V3.02 · REPLAY-CONSISTENT PUBLICATION LAG / PROXY SENSITIVITY / FUTURE CANARY

# Functions must exist BEFORE the sensitivity loop; previous V3.00 swallowed NameError.
def shifted_predictors(frame,year_lag,features,panel_override=None):
    """Rebuild the exact factor availability transform used in cell 06.

    Origin t uses the factor value at t-publication_lag. A missing value at that
    source year is carried forward from the latest *past* observation only if
    its age <= MAX_FEATURE_STALENESS_YEARS. Future data are never backfilled.
    Unlike the former direct-year join, this mirrors cell 06's ffill+age guard.
    """
    source=PANEL if panel_override is None else panel_override
    ff=list(dict.fromkeys(features))
    if not ff:return frame.copy()
    ensure(set(ff).issubset(source.columns),'CANARY_FEATURE_NOT_IN_SOURCE_PANEL')
    ensure(not source.duplicated(['entity_id','observation_year']).any(),'CANARY_SOURCE_DUPLICATE_KEYS')
    known=source[['entity_id','observation_year']+ff].sort_values(['entity_id','observation_year']).copy()
    max_age=int(CFG['MAX_FEATURE_STALENESS_YEARS'])
    for f in ff:
        observed=pd.to_numeric(known[f],errors='coerce')
        source_year=known.observation_year.where(observed.notna())
        last=source_year.groupby(known.entity_id,sort=False).ffill()
        filled=observed.groupby(known.entity_id,sort=False).ffill()
        age=known.observation_year-last
        known[f]=filled.where(last.notna()&age.le(max_age))
    known=known.rename(columns={'observation_year':'available_year'})
    result=frame.drop(columns=ff,errors='ignore').copy()
    result['available_year']=result.origin_year.astype(int)-int(year_lag)
    result=result.merge(known,on=['entity_id','available_year'],how='left',validate='many_to_one',sort=False)
    ensure(len(result)==len(frame),'CANARY_JOIN_CARDINALITY_CHANGED')
    return result.drop(columns='available_year')

def _temporal_replay_unit_test():
    """Small real-pandas contract: bounded carry, future isolation, past sensitivity."""
    yrs=[2014,2018,2019,2020,2021]
    toy=pd.DataFrame({'entity_id':['TEST_A']*len(yrs),'observation_year':yrs,
                      'G2_TEST':[1.,10.,np.nan,333.,444.]})
    frame=pd.DataFrame({'entity_id':['TEST_A'],'origin_year':[2020],'G2_TEST':[10.]})
    f=['G2_TEST'];replay=shifted_predictors(frame,1,f,toy)
    ensure(float(replay.G2_TEST.iloc[0])==10.,'UNIT_CANARY_CARRY_FORWARD_MISSING')
    future=toy.copy();future.loc[future.observation_year.ge(2020),'G2_TEST']=1e12
    after=shifted_predictors(frame,1,f,future)
    ensure(float(after.G2_TEST.iloc[0])==10.,'UNIT_CANARY_FUTURE_MUTATION_LEAK')
    past=toy.copy();past.loc[past.observation_year.eq(2018),'G2_TEST']=12.
    changed=shifted_predictors(frame,1,f,past)
    ensure(float(changed.G2_TEST.iloc[0])==12.,'UNIT_CANARY_PAST_MUTATION_NOT_OBSERVED')
    stale=toy.copy();stale.loc[stale.observation_year.eq(2018),'G2_TEST']=np.nan
    ensure(pd.isna(shifted_predictors(frame,1,f,stale).G2_TEST.iloc[0]),'UNIT_CANARY_STALE_VALUE_NOT_EXCLUDED')
    mark('UNIT_TEMPORAL_REPLAY','PASS','Bounded ffill, lag, future invariance, past mutation and stale guard')
_temporal_replay_unit_test()

def rebuild_historical_g0(sample,source):
    """Reconstruct demographic lags from source observations, never from target labels."""
    out=sample.copy();lag=CFG['FEATURE_PUBLICATION_LAG']
    history=source[['entity_id','observation_year','population','migration_per_1000']].copy()
    for k in range(lag,lag+4):
        p='hist_population_l'+str(k);m='hist_migration_l'+str(k)
        values=history.rename(columns={'observation_year':'lookback_year','population':p,
                                        'migration_per_1000':m})
        out=out.drop(columns=[p,m],errors='ignore')
        out['lookback_year']=out.origin_year-k
        out=out.merge(values,on=['entity_id','lookback_year'],how='left',validate='many_to_one')
        out=out.drop(columns='lookback_year')
    out['hist_growth_lagged_pct']=100*(out['hist_population_l'+str(lag)]/
                                         out['hist_population_l'+str(lag+2)]-1)
    out['hist_growth_lagged_pct']=out.hist_growth_lagged_pct.replace([np.inf,-np.inf],np.nan)
    ensure(len(out)==len(sample),'G0_HISTORY_REBUILD_CHANGED_ROWS')
    return out



for (sid,target,h),r in MODEL_REGISTRY.items():
    if sid!='MAX_AVAILABLE' or not AVAILABLE_GROUPS:continue
    features=list(dict.fromkeys(sum((GROUP_FEATURES[g] for g in AVAILABLE_GROUPS),[])))
    try:
        alt_train=shifted_predictors(r['train'],CFG['FEATURE_PUBLICATION_LAG']+1,features)
        alt_test=shifted_predictors(r['test'],CFG['FEATURE_PUBLICATION_LAG']+1,features)
        pred,_,_,_=fit_predict(r['family'],alt_train,alt_test,target,r['features'],CFG['SEED'],r['params'])
        alt=metrics(r['test'][target],pred,r['test'][f'hist_population_l{CFG["FEATURE_PUBLICATION_LAG"]}'])
        original=metrics(r['test'][target],r['pred'],r['test'][f'hist_population_l{CFG["FEATURE_PUBLICATION_LAG"]}'])
        SENSITIVITY_ROWS.append({'study_id':sid,'target':target,'horizon':h,'test_keys':len(r['test']),
           'check':'publication_lag_plus_1_year','original_mae':original['mae'],
           'alternative_mae':alt['mae'],'delta_mae':original['mae']-alt['mae'],
           'interpretation':'different information timing; test keys held fixed'})
        # Remove correlated proxies, preferentially retaining more complete training measurements.
        ff=[f for f in r['features'] if f.startswith(('G1_','G2_','G3_','G4_','G5_'))]
        tr=r['train'];removed=[];retained=[]
        for f in sorted(ff,key=lambda c:tr[c].notna().sum(),reverse=True):
            redundant=False
            for stay in retained:
                sample=tr[[stay,f]].dropna()
                if len(sample)>50 and abs(sample[stay].corr(sample[f],method='spearman'))>=CFG['CORRELATED_PROXY_CUTOFF']:
                    redundant=True;break
            if redundant:removed.append(f)
            else:retained.append(f)
        if removed:
            feats=[f for f in r['features'] if f not in removed]
            pred,_,_,_=fit_predict(r['family'],tr,r['test'],target,feats,CFG['SEED'],r['params'])
            teststat=metrics(r['test'][target],pred,r['test'][f'hist_population_l{CFG["FEATURE_PUBLICATION_LAG"]}'])
            SENSITIVITY_ROWS.append({'study_id':sid,'target':target,'horizon':h,
               'test_keys':len(r['test']),'check':'remove_correlated_proxies_on_train',
               'original_mae':original['mae'],'alternative_mae':teststat['mae'],
               'delta_mae':original['mae']-teststat['mae'],
               'excluded_features':','.join(removed),'interpretation':'exploratory; no causal attribution'})
    except Exception as ex:
        mark('SENSITIVITY_'+sid+'_'+target+'_h'+str(h),'FAIL',type(ex).__name__+': '+str(ex)[:180])
SENSITIVITY=pd.DataFrame(SENSITIVITY_ROWS)

# Two independent canary contracts:
# (1) SOURCE_REPLAY reproduces *exactly* the model's actual feature matrix.
# (2) FUTURE_INVARIANCE changes all source rows t >= issue origin and reruns
#     the SAME transformation. A breach is genuine only after replay parity.
CANARY_RECORDS=[]

def _canary_model_matrix(registry,frame):
    return design(registry['train'],frame,registry['features'],registry['refs'])[1]

def _canary_comparison(before,after,rtol=0.,atol=1e-10):
    same_shape=(list(before.columns)==list(after.columns) and before.shape==after.shape)
    if not same_shape:
        return False,['SHAPE_OR_COLUMN_DIFFERENCE'],float('inf')
    lhs=before.to_numpy(dtype=float);rhs=after.to_numpy(dtype=float)
    matches=np.isclose(lhs,rhs,equal_nan=True,rtol=rtol,atol=atol)
    changed=[str(c) for c in before.columns[np.any(~matches,axis=0)].tolist()]
    different=np.abs(lhs-rhs);good=np.isfinite(different)
    max_delta=float(np.max(different[good])) if np.any(good) else 0.
    return bool(matches.all()),changed,max_delta

for (sid,target,h),r in MODEL_REGISTRY.items():
    if sid not in ('MAX_AVAILABLE','CANONICAL','AUX_G1_REGION_CONTEXT'):continue
    covariates=[f for f in dict.fromkeys(r['features']) if f in PANEL.columns]
    sample=r['test'].drop_duplicates('entity_id').head(CFG['MAX_FUTURE_MUTATION_ROWS']).copy()
    if sample.empty:continue
    record={'study_id':sid,'target':target,'horizon':h,'n_test_rows':len(sample),
            'n_model_source_factors':len(covariates),'status':'NOT_RUN'}
    try:
        train=r['train'];refs=r['refs']
        x_original=_canary_model_matrix(r,sample)
        # Fidelity contract: replay without modifying a single source value.
        replay=shifted_predictors(sample,CFG['FEATURE_PUBLICATION_LAG'],covariates)
        replay=rebuild_historical_g0(replay,PANEL)
        x_replay=_canary_model_matrix(r,replay)
        faithful,replay_cols,replay_delta=_canary_comparison(x_original,x_replay)
        record.update(source_replay_exact=faithful,
          replay_mismatch_columns_json=json.dumps(replay_cols,ensure_ascii=False),
          replay_max_finite_delta=replay_delta)
        if not faithful:
            record['status']='FAIL_SOURCE_REPLAY_NOT_IDENTICAL'
            raise AssertionError('CANARY_SOURCE_REPLAY_MISMATCH: '+','.join(replay_cols[:8]))
        y_original=np.asarray(r['model'].predict(x_original),float)
        y_replay=np.asarray(r['model'].predict(x_replay),float)
        ensure(np.allclose(y_original,y_replay,rtol=0,atol=1e-9,equal_nan=True),
               'CANARY_UNMUTATED_MODEL_REPLAY_DIFFERENCE')
        mutated=PANEL.copy(deep=True)
        origins=dict(zip(sample.entity_id,sample.origin_year))
        at_or_after=mutated.observation_year.ge(mutated.entity_id.map(origins))
        mask=mutated.entity_id.isin(origins)&at_or_after.fillna(False)
        mutated_source_rows=int(mask.sum())
        ensure(mutated_source_rows>0,'CANARY_HAS_NO_FUTURE_SOURCE_ROWS')
        for f in ('population','migration_per_1000'):
            if f in mutated.columns:mutated.loc[mask,f]=1e12
        for f in covariates:mutated.loc[mask,f]=1e12
        after=shifted_predictors(sample,CFG['FEATURE_PUBLICATION_LAG'],covariates,mutated)
        after=rebuild_historical_g0(after,mutated)
        x_after=_canary_model_matrix(r,after)
        invariant_x,changed_columns,changed_max=_canary_comparison(x_replay,x_after)
        y_after=np.asarray(r['model'].predict(x_after),float)
        invariant_y=bool(np.allclose(y_replay,y_after,rtol=0,atol=1e-9,equal_nan=True))
        record.update(n_future_source_rows_mutated=mutated_source_rows,
          n_model_features=len(x_original.columns),invariant_features=invariant_x,
          invariant_predictions=invariant_y,
          mutated_mismatch_columns_json=json.dumps(changed_columns,ensure_ascii=False),
          mutated_max_finite_delta=changed_max,
          status='PASS' if invariant_x and invariant_y else 'FAIL_FUTURE_DEPENDENCE')
        if not invariant_x or not invariant_y:
            raise AssertionError('FUTURE_MUTATION_CHANGES_FINAL_FORECAST: '+','.join(changed_columns[:8]))
    except Exception as ex:
        if record['status']=='NOT_RUN':record['status']='FAIL_CANARY_RUNTIME'
        record['exception_type']=type(ex).__name__;record['reason']=str(ex)[:400]
        CANARY_RECORDS.append(record)
        CANARY_QA=pd.DataFrame(CANARY_RECORDS)
        (WORK/'ml03_canary_failure.json').write_text(json.dumps(CANARY_RECORDS,ensure_ascii=False,indent=2,default=str),encoding='utf-8')
        mark('TEMPORAL_CANARY','FAIL',record['status']+' '+record['reason'])
        raise
    CANARY_RECORDS.append(record)
CANARY_QA=pd.DataFrame(CANARY_RECORDS)
mark('TEMPORAL_CANARY','PASS' if len(CANARY_QA) and CANARY_QA.status.eq('PASS').all() else 'NOT_FEASIBLE',
     f'Source replay + future mutation; passing models={int(CANARY_QA.status.eq("PASS").sum()) if len(CANARY_QA) else 0}/{len(CANARY_QA)}')

# Same fitted models, objectively restricted boundary and municipality-size subsets.
BOUNDARY_ROWS=[];SIZE_ROWS=[]
for (sid,target,h),r in MODEL_REGISTRY.items():
    if sid not in ('MAX_AVAILABLE','CANONICAL'):continue
    test=r['test'];y_true=test[target].to_numpy(float)
    err=np.abs(y_true-np.asarray(r['pred'],float))
    if 'boundary_period_uncertain' in test:
        safe=~test.boundary_period_uncertain.fillna(True).astype(bool)
        verified=test.loc[safe]
        if safe.sum()>=CFG['MIN_TEST_ROWS'] and verified.region_name.nunique()>=CFG['MIN_BOUNDARY_SENSITIVITY_REGIONS']:
            BOUNDARY_ROWS.append({'study_id':sid,'target':target,'horizon':h,
                'n_test_all':len(test),'n_boundary_verified':int(safe.sum()),
                'n_regions_verified':int(verified.region_name.nunique()),
                'mae_all':float(err.mean()),'mae_verified_boundary':float(err[safe.to_numpy()].mean()),
                'status':'EXPLORATORY_FIXED_MODEL_STRATUM_NONCAUSAL'})
    pop='hist_population_l'+str(CFG['FEATURE_PUBLICATION_LAG'])
    train_pop=pd.to_numeric(r['train'][pop],errors='coerce')
    train_pop=train_pop[train_pop.gt(0)].dropna()
    thresholds=np.quantile(train_pop,[1/3,2/3]) if len(train_pop)>100 else None
    if thresholds is not None and thresholds[0]<thresholds[1]:
        values=pd.to_numeric(test[pop],errors='coerce').to_numpy(float)
        bins=np.digitize(values,thresholds)
        for group in range(3):
            flag=np.isfinite(values)&(bins==group)
            if flag.sum()>=15:
                SIZE_ROWS.append({'study_id':sid,'target':target,'horizon':h,
                    'size_group':('small','medium','large')[group],
                    'n_test':int(flag.sum()),'mae':float(err[flag].mean()),
                    'train_size_thresholds_json':json.dumps(thresholds.tolist()),
                    'scope':'train-defined lagged-population strata; prediction not causal'})
BOUNDARY_SENSITIVITY=pd.DataFrame(BOUNDARY_ROWS)
SIZE_SENSITIVITY=pd.DataFrame(SIZE_ROWS)
mark('BOUNDARY_SENSITIVITY','PASS' if len(BOUNDARY_SENSITIVITY) else 'NOT_FEASIBLE',
    f'Boundary-confirmed test strata={len(BOUNDARY_SENSITIVITY)}; undocumented does not imply stable')
mark('SIZE_HETEROGENEITY','PASS' if len(SIZE_SENSITIVITY) else 'NOT_FEASIBLE',
    f'Train-derived municipality-size strata={len(SIZE_SENSITIVITY)}')
# H3.5 can only be marked tested when publication lag AND correlated-proxy comparisons are present.
lag_exists=(len(SENSITIVITY)>0 and SENSITIVITY.check.eq('publication_lag_plus_1_year').any())
proxy_exists=(len(SENSITIVITY)>0 and SENSITIVITY.check.eq('remove_correlated_proxies_on_train').any())
mark('H3.5','PASS' if lag_exists and proxy_exists else 'NOT_FEASIBLE',
     f'lag sensitivity={lag_exists}; correlated-proxy sensitivity={proxy_exists}')
# Scientific interpretation is predictive only; lack of causal data is not, by itself, a fatal gate.
if len(ABLA) and len(SENSITIVITY):
    mark('PASS_INTERPRETATION','PASS',
       f'factor tests={len(ABLA)}; improved={int(ABLA.finding.eq("PREDICTIVE_IMPROVEMENT").sum())}; uncertainty and negative findings retained')
else:
    mark('PASS_INTERPRETATION','NOT_FEASIBLE','No paired factor tests or inadequate sensitivity checks')
quiet_print('Independent factor comparisons:',len(ABLA),'| ALE rows:',len(ALE),
      '| sensitivity:',len(SENSITIVITY),'| future-mutation tests:',len(CANARY_QA))

CELL10_COMPLETE=True
mark('CELL10_PIPELINE','PASS','Sensitivity, temporal canary, boundary and size tables constructed')


# V4 · ADVANCED temporal anti-leakage replay for the actual fitted SuperLearner.
# Validates source replay, model matrix and future-mutation invariance, fails closed.
ADVANCED_TEMPORAL_CANARY_ROWS=[]
for (sid,target,h),r in ADVANCED_ENSEMBLE_REGISTRY.items():
    if sid not in ('MAX_AVAILABLE','CANONICAL','AUX_G1_REGION_CONTEXT'):continue
    source_cols=[f for f in dict.fromkeys(r['features']) if f in PANEL.columns]
    small=r['test'].drop_duplicates('entity_id').head(CFG['MAX_FUTURE_MUTATION_ROWS']).copy()
    if small.empty:continue
    rec={'study_id':sid,'target':target,'horizon':h,'n_test':len(small),'n_source_features':len(source_cols)}
    try:
        orig=design(r['train'],small,r['features'],r['members'][0]['refs'])[1]
        replay=shifted_predictors(small,CFG['FEATURE_PUBLICATION_LAG'],source_cols)
        replay=rebuild_historical_g0(replay,PANEL)
        rebuilt=design(r['train'],replay,r['features'],r['members'][0]['refs'])[1]
        faithful,missing,delta=_canary_comparison(orig,rebuilt)
        ensure(faithful,'ADVANCED_SOURCE_REPLAY_MISMATCH: '+','.join(missing[:8]))
        mutated=PANEL.copy(deep=True)
        first=dict(zip(small.entity_id,small.origin_year))
        mask=mutated.entity_id.isin(first)&mutated.observation_year.ge(mutated.entity_id.map(first)).fillna(False)
        ensure(mask.sum()>0,'ADVANCED_FUTURE_MUTATION_NO_ROWS')
        for f in ['population','migration_per_1000']+source_cols:mutated.loc[mask,f]=1e12
        altered=shifted_predictors(small,CFG['FEATURE_PUBLICATION_LAG'],source_cols,mutated)
        altered=rebuild_historical_g0(altered,mutated)
        after=design(r['train'],altered,r['features'],r['members'][0]['refs'])[1]
        stable,changed,change=_canary_comparison(rebuilt,after)
        y0=_adv_predict(r,rebuilt);y1=_adv_predict(r,after)
        ok=bool(stable and np.allclose(y0,y1,rtol=0,atol=1e-9))
        rec.update({'n_mutated_source_rows':int(mask.sum()),'source_replay_exact':True,
            'prediction_invariant':ok,'matrix_invariant':stable,
            'max_predictive_diff':float(np.abs(y0-y1).max()),'status':'PASS' if ok else 'FAIL'})
        ensure(ok,'ADVANCED_ENSEMBLE_FUTURE_DEPENDENCE')
    except Exception as ex:
        rec.update({'status':'FAIL','exception':type(ex).__name__+': '+str(ex)[:250]})
        ADVANCED_TEMPORAL_CANARY_ROWS.append(rec)
        mark('ADVANCED_TEMPORAL_CANARY','FAIL',rec['exception'])
        raise
    ADVANCED_TEMPORAL_CANARY_ROWS.append(rec)
ADVANCED_TEMPORAL_CANARY=pd.DataFrame(ADVANCED_TEMPORAL_CANARY_ROWS)
mark('ADVANCED_TEMPORAL_CANARY', 'PASS' if len(ADVANCED_TEMPORAL_CANARY) and ADVANCED_TEMPORAL_CANARY.status.eq('PASS').all() else 'NOT_FEASIBLE',
     f'passing weighted-ensemble replay/future tests={len(ADVANCED_TEMPORAL_CANARY)}')

PUBLICATION_AVAILABILITY_QA=pd.DataFrame([{
  'feature':r.get('feature'),'indicator_name':r.get('indicator_name'),'group':r.get('group'),
  'documented_publication_date':r.get('report_available_year','NOT_VERIFIED'),
  'availability_provenance':r.get('availability_provenance','UNKNOWN'),
  'lag_assumption_years':CFG['FEATURE_PUBLICATION_LAG'],
  'status':'DOCUMENTED_SOURCE_DATE' if str(r.get('report_available_year','')).isdigit() and str(r.get('availability_provenance','')).startswith('OFFICIAL') else 'LAG_ASSUMPTION_ONLY'}
  for r in FEATURE_DICT.to_dict('records')])
if len(PUBLICATION_AVAILABILITY_QA):
    real=int(PUBLICATION_AVAILABILITY_QA.status.eq('DOCUMENTED_SOURCE_DATE').sum())
    mark('HISTORICAL_PUBLICATION_DATES','PASS' if real==len(PUBLICATION_AVAILABILITY_QA) else 'PARTIAL',
         f'documented={real}/{len(PUBLICATION_AVAILABILITY_QA)}; future-mutation and lag sensitivity do not prove original release dates')

progress('SENSITIVITY',1,1,'Canary и чувствительность завершены')
```

## 10. Графики и аналитическая книга

```python
# 11 · V3.01 · ANALYTICAL WORKBOOK + OBSERVED FIGURES · STAGE CONTRACTS
ensure(globals().get('CELL10_COMPLETE',False),
       'CELL10_NOT_COMPLETE: execute cell 10 before workbook; do not manufacture missing diagnostics')
_REQUIRED_11=('SENSITIVITY','BOUNDARY_SENSITIVITY','SIZE_SENSITIVITY','CANARY_QA','MODEL_REGISTRY','REGIONAL_CONTEXT_METADATA')
_MISSING_11=[k for k in _REQUIRED_11 if k not in globals()]
ensure(not _MISSING_11,'CELL11_INPUTS_MISSING: '+','.join(_MISSING_11))
# H3.5 tests are derived from the actual sensitivity ledger, never stale variables.
_SENS_CHECKS=set(SENSITIVITY['check'].dropna().astype(str)) if isinstance(SENSITIVITY,pd.DataFrame) and 'check' in SENSITIVITY.columns else set()
lag_exists='publication_lag_plus_1_year' in _SENS_CHECKS
proxy_exists='remove_correlated_proxies_on_train' in _SENS_CHECKS
CHARTS=[]
plt.rcParams.update({'figure.dpi':125,'savefig.dpi':160,'font.size':9,
    'axes.spines.top':False,'axes.spines.right':False})
def figure_file(name):
    path=WORK/'charts'/(name+'.png');plt.savefig(path,bbox_inches='tight',dpi=160)
    plt.close();CHARTS.append(path)
def safe_plot(label,fn):
    try:fn()
    except Exception as ex:log('PLOT '+label+': '+type(ex).__name__+' '+str(ex)[:150],'WARN');plt.close('all')

def chart_ablation():
    if ABLA.empty:return
    p=ABLA[(ABLA.protocol=='ROLLING_TIME')&(ABLA.target=='population_change_pct')].head(16)
    if p.empty:return
    plt.figure(figsize=(11,5));x=np.arange(len(p));plt.bar(x,p.delta_mae,yerr=[p.delta_mae-p.ci_low,p.ci_high-p.delta_mae],capsize=3)
    plt.axhline(0,color='black',lw=.7);plt.xticks(x,[f"h{r.horizon}:{r.comparison}" for r in p.itertuples()],rotation=45,ha='right')
    plt.ylabel('Δ MAE, п.п. (положительное = улучшение)');plt.title('Парный прирост качества: общие тестовые наблюдения и региональные интервалы')
    figure_file('01_ablation_region_CI')
safe_plot('ablation',chart_ablation)
def chart_loo():
    if LOO.empty:return
    p=LOO[(LOO.target=='population_change_pct')&(LOO.protocol=='ROLLING_TIME')].head(12)
    if p.empty:return
    plt.figure(figsize=(9,4));plt.barh(np.arange(len(p)),p.delta_mae)
    plt.yticks(np.arange(len(p)),[f'{r.horizon} г. {r.old_arm}' for r in p.itertuples()])
    plt.xlabel('Вклад исключённой группы в MAE, п.п.');plt.title('Leave-one-group-out: прогнозный вклад группы')
    figure_file('02_leave_one_group_out')
safe_plot('loo',chart_loo)
def chart_importance():
    if IMPORTANCE.empty:return
    p=IMPORTANCE.sort_values('mae_increase',ascending=False).head(15)
    plt.figure(figsize=(9,5));plt.barh(p.feature[::-1],p.mae_increase[::-1])
    plt.xlabel('Ухудшение MAE при перестановке');plt.title('Важность для прогностической функции, не причинный эффект')
    figure_file('03_permutation_importance')
safe_plot('importance',chart_importance)
def chart_ale():
    if ALE.empty:return
    p=ALE[ALE.feature.eq(ALE.feature.iloc[0])];plt.figure(figsize=(8,4))
    for h,a in p.groupby('horizon'):
        plt.plot(a.feature_mid_logscale,a.ale_centered,marker='o',label=f'{h} год(лет)')
    plt.legend();plt.ylabel('Центрированная ALE');plt.xlabel('Признак в логарифмической шкале')
    plt.title('Условная нелинейность на обучающей области наблюдений')
    figure_file('04_train_ALE')
safe_plot('ale',chart_ale)
def chart_heterogeneity():
    if REGIONAL_ERRORS.empty:return
    p=REGIONAL_ERRORS[(REGIONAL_ERRORS.protocol=='ROLLING_TIME')&(REGIONAL_ERRORS.horizon==1)]
    p=p.groupby('region',as_index=False).agg(mae=('mae','mean')).sort_values('mae').head(18)
    if p.empty:return
    plt.figure(figsize=(10,5));plt.barh(p.region,p.mae)
    plt.xlabel('MAE');plt.title('Территориальная неоднородность качества')
    figure_file('05_regional_errors')
safe_plot('regional',chart_heterogeneity)
def chart_times():
    if ERROR_YEARS.empty:return
    p=ERROR_YEARS[(ERROR_YEARS.study_id=='MAX_AVAILABLE')]
    if p.empty:p=ERROR_YEARS
    p=p.groupby(['target','horizon','year'],as_index=False).agg(mae=('mae','mean'))
    plt.figure(figsize=(9,4))
    for (target,h),x in p.groupby(['target','horizon']):plt.plot(x.year,x.mae,marker='o',label=target[:10]+' h'+str(h))
    plt.legend(fontsize=7);plt.xlabel('Год целевого наблюдения');plt.ylabel('MAE')
    plt.title('Стабильность ошибки по временным отложениям')
    figure_file('06_out_of_time_stability')
safe_plot('time',chart_times)
def chart_coverage():
    if coverage.empty:return
    plt.figure(figsize=(7,3));plt.bar(coverage.group,coverage.forecast_coverage)
    plt.ylim(0,1);plt.ylabel('Доля прогнозных наблюдений с доступными факторами')
    plt.title('Наблюдаемость блоков после публикационного лага')
    figure_file('07_factor_availability')
safe_plot('coverage',chart_coverage)
def chart_sensitivity():
    if SENSITIVITY.empty:return
    p=SENSITIVITY.groupby('check',as_index=False).agg(delta_mae=('delta_mae','mean'))
    plt.figure(figsize=(8,3));plt.barh(p.check,p.delta_mae)
    plt.axvline(0,color='black',lw=.8);plt.xlabel('Средняя разница MAE (оригинал − вариант)')
    plt.title('Проверка задержки публикации и коррелированных показателей')
    figure_file('08_sensitivity')
safe_plot('sensitivity',chart_sensitivity)
def chart_migration_gate():
    if MIGRATION_QA.empty:return
    p=MIGRATION_QA.head(8);plt.figure(figsize=(8,3))
    plt.barh([f'{r.denominator}, ×{r.numerator_scale:g}' for r in p.itertuples()][::-1],
             p.agreement_5pct.iloc[::-1]);plt.xlim(0,1)
    plt.xlabel('Доля согласованных значений');plt.title('Аудит семантики миграционного показателя D01')
    figure_file('09_migration_semantics')
safe_plot('migration',chart_migration_gate)
def chart_interactions():
    if INTERACTIONS.empty:return
    p=INTERACTIONS.head(8);plt.figure(figsize=(8,3))
    plt.bar([str(h) for h in p.horizon],p.delta_mae)
    plt.axhline(0,color='black',lw=.8);plt.ylabel('ΔMAE взаимодействия, п.п.')
    plt.title('G1 × G2: исследовательская проверка добавочной прогностической ценности')
    figure_file('10_income_housing_interaction')
safe_plot('interactions',chart_interactions)

# Advanced figures from real fitted ensembles/explanations; no decorative plots.
# Scientific visualization labels from source dictionary and transformations (never generic ID alone).
_VIS_META={str(r.model_feature):str(r.indicator_name) for r in TRANSFORM_QA.itertuples()
           if pd.notna(r.model_feature)}
_VIS_META.update({str(r.feature):str(r.indicator_name) for r in FEATURE_DICT.itertuples()})
def feature_label(feature):
    f=str(feature);label=_VIS_META.get(f)
    if not label and f.startswith('hist_'):
        lab=re.sub(r'_l(\d+)$',r' (лаг \1 г.)',f[5:]).replace('_',' ')
        label='История: '+lab
    if not label and f.startswith('G1R_'):label='Региональный контекст: '+f
    if not label:label=f.replace('_',' ')
    return str(label).strip()[:74]
def _uniquify_axis_labels(features):
    labs=[feature_label(f) for f in features]
    if len(labs)!=len(set(labs)):labs=[f'{lab} [{f}]' for lab,f in zip(labs,features)]
    return labs

def chart_adv_global():
    if ADVANCED_SHAP.empty:return
    p=ADVANCED_SHAP[(ADVANCED_SHAP.method=='SHAP_ENSEMBLE') & ADVANCED_SHAP.study_id.eq('MAX_AVAILABLE')]
    if p.empty:return
    ss=p.groupby(['feature','group'],as_index=False).mean_abs_shap.mean().sort_values('mean_abs_shap').tail(18)
    plt.figure(figsize=(11,8));plt.barh(_uniquify_axis_labels(ss.feature),ss.mean_abs_shap)
    plt.xlabel('Среднее абсолютное SHAP, единицы целевого прогноза');plt.title('SuperLearner: объяснение предсказаний на тестовых наблюдениях')
    figure_file('11_ensemble_shap_global')
safe_plot('advanced_shap_global',chart_adv_global)

def chart_adv_permutation():
    if ADVANCED_GROUP_PERM.empty:return
    p=ADVANCED_GROUP_PERM[(ADVANCED_GROUP_PERM.status=='PASS')&ADVANCED_GROUP_PERM.study_id.eq('MAX_AVAILABLE')]
    if p.empty:return
    p=p.groupby('group',as_index=False).delta_mae.mean().sort_values('delta_mae')
    plt.figure(figsize=(9,4));plt.barh(p.group,p.delta_mae);plt.axvline(0,color='black',lw=.8)
    plt.xlabel('Рост MAE после совместной перестановки блока');plt.title('Зависимость прогноза от групп факторов (не причинный эффект)')
    figure_file('12_conditional_group_permutation')
safe_plot('advanced_group_permutation',chart_adv_permutation)

def chart_adv_comparison():
    if ADVANCED_COMPARISONS.empty:return
    p=ADVANCED_COMPARISONS[(ADVANCED_COMPARISONS.protocol=='ROLLING_TIME') & ADVANCED_COMPARISONS.study_id.eq('MAX_AVAILABLE')]
    if p.empty:return
    lab=[('миграция' if 'migration' in x.target else 'население')+f' {int(x.horizon)} г.' for x in p.itertuples()]
    plt.figure(figsize=(9,4));xx=np.arange(len(p));plt.bar(xx,p.delta_mae)
    plt.axhline(0,color='black',lw=.8);plt.xticks(xx,lab,rotation=25,ha='right')
    plt.ylabel('ΔMAE (M0 − SuperLearner); положительное = улучшение');plt.title('Независимая проверка ансамбля по регионам')
    figure_file('13_superlearner_vs_historical')
safe_plot('advanced_superlearner',chart_adv_comparison)

def chart_adv_pdp():
    if ADVANCED_PDP.empty:return
    p=ADVANCED_PDP[ADVANCED_PDP.study_id.eq('MAX_AVAILABLE')]
    if p.empty:return
    row=p.groupby(['target','horizon','feature']).size().sort_values(ascending=False).index[0]
    p=p[(p.target==row[0])&(p.horizon==row[1])&(p.feature==row[2])]
    plt.figure(figsize=(8,4));plt.plot(p.feature_value_transformed,p.mean_prediction,marker='o')
    plt.xlabel(feature_label(row[2])+' (преобразованная шкала)');plt.ylabel('Прогноз модели')
    plt.title('PDP на обучающей области, не причинная зависимость')
    figure_file('14_ensemble_train_PDP')
safe_plot('advanced_pdp',chart_adv_pdp)

def chart_adv_best_single():
    if ADVANCED_SINGLE_DUEL.empty:return
    d=ADVANCED_SINGLE_DUEL[(ADVANCED_SINGLE_DUEL.protocol=='ROLLING_TIME') &
                          ADVANCED_SINGLE_DUEL.study_id.eq('CANONICAL')].copy()
    if d.empty:return
    labels=[('Миграция' if 'migration' in v.target else 'Численность')+f' · {int(v.horizon)} г.'
            for v in d.itertuples()]
    plt.figure(figsize=(10,5));x=np.arange(len(d));plt.bar(x,d.delta_mae)
    plt.errorbar(x,d.delta_mae,yerr=[d.delta_mae-d.ci_low,d.ci_high-d.delta_mae],
        fmt='none',ecolor='black',capsize=4)
    plt.axhline(0,color='black',lw=.8);plt.xticks(x,labels,rotation=20,ha='right')
    plt.ylabel('MAE одиночной − MAE ансамбля (плюс означает выигрыш ансамбля)')
    plt.title('Сравнение с лучшим одиночным методом по DEV · один тестовый набор')
    figure_file('15_superlearner_vs_best_single')
safe_plot('advanced_vs_best_single',chart_adv_best_single)

PASSPORT=pd.DataFrame([
 {'Параметр':'Исследование','Значение':'ML-03: экономика, инфраструктура и демографическая устойчивость'},
 {'Параметр':'Версия','Значение':CFG['VERSION']},{'Параметр':'Режим','Значение':CFG['MODE']},
 {'Параметр':'Источник Stage I','Значение':CFG['REPO']+'@'+source_sha},
 {'Параметр':'SHA-256 handoff','Значение':hashlib.sha256(raw_index).hexdigest()},
 {'Параметр':'Верифицированных Stage I файлов','Значение':sum(a['asset_id'] in LOCAL_ASSETS for a in ASSETS)},
 {'Параметр':'Муниципалитетов D01','Значение':int(base.entity_id.nunique())},
 {'Параметр':'Муниципалитет×год','Значение':len(PANEL)},
 {'Параметр':'Диапазон лет','Значение':str(min(YEARS))+'–'+str(max(YEARS))},
 {'Параметр':'Дополнительные БДМО Parquet','Значение':len(BDMO_EXTERNAL_ASSETS)},
 {'Параметр':'Advanced SuperLearner experiments','Значение':int(ADVANCED_ENSEMBLES.status.eq('PASS').sum()) if len(ADVANCED_ENSEMBLES) else 0},
 {'Параметр':'SHAP additivity checks','Значение':int(ADVANCED_SHAP_QA.status.eq('PASS').sum()) if len(ADVANCED_SHAP_QA) else 0},
 {'Параметр':'Факторных признаков','Значение':sum(len(x) for x in GROUP_FEATURES.values())},
 {'Параметр':'Тестовых прогнозов','Значение':len(PREDICTIONS)},
 {'Параметр':'Модельных оценок','Значение':len(METRICS)},
 {'Параметр':'Вариантов факторных исследований','Значение':len(STUDIES)},
 {'Параметр':'Миграционная цель проходит проверку','Значение':bool(valid_rate)},
 {'Параметр':'Причинная идентификация','Значение':'НЕ ПРОВЕДЕНА — нет реестра вмешательств'},
 {'Параметр':'Прогнозные признаки доступны','Значение':f'не позднее origin_year−{CFG["FEATURE_PUBLICATION_LAG"]}'},
 {'Параметр':'Разбиение','Значение':'rolling-time / territory-only / crossed; раздельное DEV и TEST'},
 {'Параметр':'Семантика оценок','Значение':'ассоциации и прогнозы; НЕ причинные эффекты'}])
HYPOTHESES=pd.DataFrame([
 {'Гипотеза':'H3.1','Критерий':'M1→M3; region-block CI, Holm; same test keys',
    'Статус':'TESTED' if len(ABLA) and ABLA.comparison.eq('M1 → M3').any() else 'NOT_FEASIBLE'},
 {'Гипотеза':'H3.2','Критерий':'G1×G2 interaction; exploratory',
    'Статус':'PRELIMINARY_PENDING_SEMANTICS' if len(INTERACTIONS) else 'NOT_FEASIBLE'},
 {'Гипотеза':'H3.3','Критерий':'G3 incremental on common support',
    'Статус':'PRELIMINARY_PENDING_HEALTH_COVERAGE' if len(ABLA) and ABLA.study_id.eq('AUX_G3').any() else 'NOT_FEASIBLE'},
 {'Гипотеза':'H3.4','Критерий':'G4/G5 availability and transfer',
    'Статус':'EXPLORATORY_PARTIAL' if len(ABLA) and ABLA.study_id.eq('AUX_G4').any() else 'NOT_FEASIBLE'},
 {'Гипотеза':'H3.5','Критерий':'publication lag + correlated-proxy sensitivity',
    'Статус':'EXPLORATORY_TESTED' if lag_exists and proxy_exists else 'NOT_FEASIBLE'},
 {'Гипотеза':'H3.6','Критерий':'health separately identified from education',
    'Статус':'EXPLORATORY_ELIGIBLE' if HEALTH_AVAILABLE else 'NOT_FEASIBLE'},
 {'Гипотеза':'H3.7','Критерий':'municipality boundary-consistency strata',
    'Статус':'EXPLORATORY_TESTED' if len(BOUNDARY_SENSITIVITY) else 'NOT_FEASIBLE'},
 {'Гипотеза':'H3.8','Критерий':'municipality size strata using train thresholds',
    'Статус':'EXPLORATORY_TESTED' if len(SIZE_SENSITIVITY) else 'NOT_FEASIBLE'}])
LIMITATIONS=pd.DataFrame([
 {'Ограничение':'Неполнота предметного поиска','Следствие':'Если сохранены неразобранные коды, SCIENTIFIC_PASS запрещен'},
 {'Ограничение':'Муниципальные границы','Следствие':'ОКТМО не означает неизменных границ; spatial history requires validated crosswalk'},
 {'Ограничение':'Отбор источников по семантике','Следствие':'Владение всеми 52 активами не означает применимость всех; широкие таблицы без проверенной единицы остаются в аудите'},
 {'Ограничение':'Региональные и организационные источники','Следствие':'Indexed and downloaded but not joined without verified grain/crosswalk'},
 {'Ограничение':'Срок публикации','Следствие':'Отсутствует конкретный available_at; применяется conservative publication lag'},
 {'Ограничение':'Миграционная цель','Следствие':'Defined only when independent numerical audit validates scale and denominator'},
 {'Ограничение':'Финансы','Следствие':'Local budget spending is not treated as a treatment assignment'},
 {'Ограничение':'Инфраструктура','Следствие':'Counts are different from access; age-cohort denominator not inferred'},
 {'Ограничение':'Денежные показатели','Следствие':'Nominal source-year amounts; no unreliable municipal deflator imputed'},
 {'Ограничение':'Персональные траектории','Следствие':'Population change is municipal stock variation, not individual retention'},
 {'Ограничение':'Коррелированные признаки','Следствие':'Permutation and ALE explain prediction, not structural/causal mechanisms'},
 {'Ограничение':'Пространственные признаки','Следствие':'Static transport/settlement crosswalk withheld if source-vintage unavailable'},
 {'Ограничение':'Множественная проверка','Следствие':'Holm for preregistered, BH FDR for exploratory; regional cluster bootstrap'},
 {'Ограничение':'Устойчивость модели','Следствие':'Stochastic seeds not counted as independent test municipalities'}])
SHEETS={
 'Паспорт':PASSPORT,'Гипотезы':HYPOTHESES,'Источники':input_map,'Загрузки_52':DOWNLOAD_QA,
 'Аудит_миграции':MIGRATION_QA,'Окна_целей_V3':TARGET_CALENDAR_PREFLIGHT,
 'Реестр_физ_52':STAGE1_PHYSICAL_AUDIT,'Манифест_экономики':ECONOMY_SOURCE_MANIFEST_QA,
 'Региональный_G1R':REGIONAL_CONTEXT_METADATA,'G1R_аудит':REGIONAL_CONTEXT_AUDIT,
 'Единицы_G1R':G1R_UNIT_QA,
 'Давность_факторов':STALENESS_QA,'Наличие_целей_по_годам':TARGET_AVAILABILITY,'Факторы':FEATURE_DICT,'Преобразования':TRANSFORM_QA,
 'Каталог_индикаторов':CATALOG,'Поиск_Г1_все':SCAN_AUDIT,'Паспорт_поиска_Г1':G1_DISCOVERY,
 'Измерения_кодов':DIMENSION_AUDIT,'Уровни_источников':pd.DataFrame(SOURCE_ADAPTER_AUDIT),
 'Паспорт_52_источников':SOURCE_GRAIN_AUDIT,'SQL_ошибки':SCAN_SQL_ERRORS,'SQL_запросы':SCAN_SQL_AUDIT,'SQL_предпроверка':REAL_SQL_PREFLIGHT_QA,
 'Факторы_по_горизонтам':FACTOR_ELIGIBILITY,'Отбор_по_горизонтам':SELECTION_BY_H,'Семантика_отбора':SEMANTIC_QA,'Домены':DOMAIN_COVERAGE,'Домены_модели':MODEL_DOMAIN_COVERAGE,'Коллизии_разрезов':CONFLICT_QA,'QA_источников':SOURCE_QA,
 'Дата_публикации':PUBLICATION_AVAILABILITY_QA,'Экономика_домены':G1_DOMAIN_QA,'Медицина_покрытие':HEALTH_DOMAIN_QA,'Пятилетний_аудит':H5_CALENDAR_GAPS,'Исключения_организаций':pd.DataFrame(ORGANIZATION_GRAIN_EXCLUSIONS),
 'Пространственные_данные':SPATIAL_SOURCE_QA,'Геометрия_G5':SPATIAL_GEOMETRY_QA,'Геопризнаки_G5':SPATIAL_FEATURE_AUDIT,'История_муниципалитетов':BOUNDARY_QA,'Неопределенность_границ':BOUNDARY_WINDOW_QA,'Покрытие':coverage,'JOIN_аудит':QA_KEYS,
 'План_задач':TASK_STATUSES,'Годы_раздельно':TEST_YEAR_AUDIT,'Годы_CANONICAL':CANONICAL_YEAR_QA,'Модели_DEV':DEV_METRICS,'Метрики_TEST':METRICS,
 'Абляции_CI':ABLA,'Вклад_LOO':LOO,'Реальные_сиды':SEED_QA,
 'Сверка_метрик':PREDICTION_METRIC_QA,'Реестр_гипотез':HYPOTHESIS_REGISTRY,'Семейства_тестов':STUDY_HYPOTHESIS_FAMILIES,'Корреляции':CORRELATIONS,'Взаимодействия':INTERACTIONS,
 'Региональные_ошибки':REGIONAL_ERRORS,'Типы_муниципалитетов':MUNICIPAL_TYPES,
 'ALE':ALE,'SHAP':SHAP_IMPORTANCE,'FE_Панель':FE_DIAGNOSTICS,'FE_QA':FE_QA,'Важность':IMPORTANCE,'Сенситивити':SENSITIVITY,'Ошибки_по_годам':ERROR_YEARS,
 'Canary_контроль':CANARY_QA,'Границы_сенситивити':BOUNDARY_SENSITIVITY,'Размер_территорий':SIZE_SENSITIVITY,'Приемка':pd.DataFrame([{'Этап':k,**v} for k,v in STATUS.items()]),
 'БДМО_доп_план':BDMO_EXTERNAL_DESIGN,'БДМО_доп_загрузки':BDMO_EXTERNAL_QA,'БДМО_неуспехи':BDMO_EXTERNAL_RECOVERY,
 'Ансамбль_DEV':ADVANCED_DEV,'Ансамбль_TEST':ADVANCED_ENSEMBLES,'Ансамбль_интервалы':ADVANCED_INTERVALS,
 'Ансамбль_сравнения':ADVANCED_COMPARISONS,'Ансамбль_лучший_один':ADVANCED_SINGLE_DUEL,'SHAP_ансамбли':ADVANCED_SHAP,
 'SHAP_локальные':ADVANCED_SHAP_LOCAL,'SHAP_аддитивность':ADVANCED_SHAP_QA,'SHAP_регионы_CI':ADVANCED_SHAP_CI,
 'SHAP_устойчивость':ADVANCED_SHAP_STABILITY,'SHAP_взаимодействия':ADVANCED_SHAP_INTERACTIONS,
 'Canary_ансамбль':ADVANCED_TEMPORAL_CANARY,'Важность_групп':ADVANCED_GROUP_PERM,'Ансамбль_PDP':ADVANCED_PDP,'Ансамбль_ICE':ADVANCED_ICE,
 'Ограничения':LIMITATIONS}

def write_xlsx(path,tables):
    # One-pass streaming workbook. No head() truncation and no multi-gigabyte cell cache.
    from openpyxl.cell import WriteOnlyCell
    wb=Workbook(write_only=True);count=0;sheet_index=[];used=set()
    for name,df in tables.items():
        if df is None or len(df)==0:continue
        maxrows=min(int(CFG['EXCEL_MAX_ROWS_PER_SHEET']),1048575)
        chunks=max(1,math.ceil(len(df)/maxrows))
        for chunk in range(chunks):
            offset=chunk*maxrows
            d=df.iloc[offset:offset+maxrows].copy().replace([np.inf,-np.inf],np.nan)
            sheetname=(name[:25]+f'_{chunk+1:02d}') if chunks>1 else name[:31]
            ensure(sheetname not in used,'XLSX_DUPLICATE_SHEET_NAME_'+sheetname);used.add(sheetname)
            ws=wb.create_sheet(sheetname);ws.freeze_panes='A2';ws.sheet_view.showGridLines=False
            cols=list(d.columns)
            for j,col in enumerate(cols,1):
                sample=d[col].dropna().head(90)
                widths=[len(str(v)) for v in sample]
                w=max(13,len(str(col))+2,int(np.percentile(widths,80))+2 if widths else 13)
                ws.column_dimensions[get_column_letter(j)].width=min(48,w)
            hdr=[]
            for c in cols:
                cell=WriteOnlyCell(ws,value=str(c));cell.fill=PatternFill('solid',fgColor='173653')
                cell.font=Font(color='FFFFFF',bold=True,size=10)
                cell.alignment=Alignment(vertical='center',wrap_text=True);hdr.append(cell)
            ws.append(hdr)
            for row in d.itertuples(index=False,name=None):
                vals=[]
                for v in row:
                    if isinstance(v,(list,dict,tuple)):v=json.dumps(v,ensure_ascii=False,default=str)
                    elif isinstance(v,np.integer):v=int(v)
                    elif isinstance(v,np.floating):v=float(v)
                    elif isinstance(v,(pd.Timestamp,datetime.datetime)):v=str(v)
                    elif v is pd.NA or v is pd.NaT:v=None
                    elif isinstance(v,float) and not np.isfinite(v):v=None
                    if isinstance(v,str) and v.startswith(('=','+','@')):v="'"+v
                    vals.append(v)
                ws.append(vals)
            ws.auto_filter.ref=f'A1:{get_column_letter(len(cols))}{len(d)+1}'
            sheet_index.append({'source':name,'sheet':sheetname,'rows':len(d),
                                'offset':offset,'full_source_rows':len(df)})
            count+=1
    ensure(count>5,'XLSX_HAS_TOO_FEW_REAL_SHEETS')
    index=wb.create_sheet('Индекс_листов')
    index.append(['Источник','Лист','Строк','Смещение','Всего строк'])
    for r in sheet_index:index.append([r['source'],r['sheet'],r['rows'],r['offset'],r['full_source_rows']])
    wb.save(path)
    exported=sum(r['rows'] for r in sheet_index if r['source']=='SHAP_локальные')
    ensure(exported==len(ADVANCED_SHAP_LOCAL),'XLSX_SHAP_LOCAL_TRUNCATED')
    return sheet_index
# Единый сквозной бар создан в ячейке 02. Второй tqdm запрещён.
XLSX=WORK/'ML3_FACTORS_RESULTS.xlsx';EXCEL_SHEET_INDEX=write_xlsx(XLSX,SHEETS)
progress('EXCEL',1,1,'Книга записана');quiet_print('Excel:',XLSX.name,'sheets=',sum(len(v)>0 for v in SHEETS.values() if v is not None),
                   'graphs=',len(CHARTS))
```

## 12. Проверки основного исследования и подготовка удалённого Evidence

```python
# 13 · V4.00 · SELF-EXPORT, FINAL SCIENTIFIC QA, RUN MANIFEST, SAFE GITHUB PUBLICATION
# The compressed, secret-free notebook snapshot is embedded at build time.
NB_SNAPSHOT_B85 = 'c-qXpdvhDfl`r`J{S;+&%oHGjP(6VH*k}huLK3Gbk|wEcw<#`MPZAag&;Y2GTEh)L#$$W!JH9hJH)3Mq&SUS!{x7Rrwz}P|j@VB?eTDrb`#UGIs<H|NO6}>{SkJTw6tePh^5i+cbMoZLr~gH}7RIB|>{?el{4bjJ6#t1QgZY!Q7*DUA`qPKe<m2(R2LCpU=%E|gomSfk<-^l>?nnOI$4^|tLNR)HA<E7s=hHBj?cw`7+P$q6!`41<nuewQ-+$CD|LyYC<)1FUdi|Tr&$Y`Jm!Dm}Ouu|}`Rw)2U;i@w`t|Fdpy;Rg?<;=s?_dAs^-o{_3{RiE{+)LD9Hl<P$6sDP)9~!Y<sV=F{PIt)fBX7pc=$8C_56Vxm3A%vbosBMMeXvBm(S3|&tL!a@(b<GWO6);we`u!54xIbX_qh2)GtuUXVjb5ztB+l3)K4Z@;RE@pZmwLw%O3|^Q+6R#ene2&(I6>fCumb9lrd>%l~otuPF8<*MuK0UpA<!uh8l<JpH|RvVV7Nx1rJCxa%+Z)62^*E}u078D3(P=rmL4cN&l287TAB<*U3X`b>p?E-Ixbum2V={!RnQes}qY*ME<$@kqYpMqW_0sO3*YDNs|peqA(0BgF5h`|?Ze&TxMBJXm4+(m4OfqkP5XUS7Z6)Gq%UCL(?E$E?l&=O_P&@!+K|=+)Oh6Qo5OKRnpmT-#XLyT8Iue#mwF3-$gBY7kxc?d!i2-6KMA-Ll6-gcmej7%2aF##BLfsRa<?cc`Ds(;T2Wjj7I){S~_Md-2QbU*YfP+7C8&uU|(!MYj8r>Rx0fnk$;QPnr4tMv#{a3!;Njx#2IE75JB?%IPn(U4Q!H^LWnXUtRv58+<`CDp*Bf5E|R7*T0~a@ayl08ay^kk(do_FCNAIEati}uo5xy7yiUOdL~$s$nmwL%^zvb1s9;ui_0&Yl}qK{Xi^33K9ihA?|eZW7Cib&@}`^?;;zrA+v%cFN&A8Z{;3$2c!~d_pJ~-;$HA`8MTcMUr2UcW$>}AjTgmF^$}iIT^G;Q={8yq&Y3<BKf<~XA@N+Ky2c8w#&*#L^R2>cer%bueR7!$fvKBu@OJ6frfS1tQ|IRPGOczw;LKQs{-1IBrBgllmLBBqym3swS;O|ekFE7M0%S8!^w=xy;)$+spyIVVJ8-v?xn_K;jAChDeefZ(-+W!6zi7AK^j{b|3?iK1IoJ6tI67l0WJf9CI<3Tt%AJ6gY_<S@{)rue6R?u<$H>j1Yn4(rD=kv4kIqR{bWs1fyNlo0_lG?_gZ@2GnZSkCHp!b)|{=bwA_;-R#Qpvm&tMCN`<^Kli60^O?8JE`eS3>oW8hHJie3h)<zN7WDr}+c=N58kT(bu~=HLo|civPW%>HGI~-tX(QV*W()`1?$~y~?$p;6jpkH#JdPR(^pA1)G*SvU~s5*5>}*0lhn*Y1O;8{n0FLWc?^gOyXItRj}R{1-Z-GSXtHb0TcOqYo(U=`nx+I!P<B}8E;HJIf;f3Cy#D!#HW+vsh<pHlSlD<Jbb7(O7HFO+}~U85BB=E@wV$ZZ9D07T*omY-w&;h8~V28+W62(Ov4Nk-*jSsp_aRA+Z#K#sfrVS98HqsrqLcevKCwG-?`V{K0vt{DRt}S>Bz7Li671<)7e7l5Bhuin>*WC>6O!wz48dm*0ffV;n*K7yz$}A-uoMydwAn!I2q65@%-u*_WS(}y=%6uPWhLOoe#HpxvpU*_Yc<Y^i92cFg=%3`HqxNgkFFrRvyLEVKR)P6s(Y-C~5>oS)m`E#H-q0?QCu@asmV)V<0SuUkkA&@fyDTZJFcOzCYOC+`hBbAArmH2ZKAe^sd)5%EP#~xjop}+<$*?d#~T;LamBIsI33N=0<;eU6e7K3-q|Vv$y$!ooz7mq1mt-?oqkG!P?%P{sH~2pANxuESt{a>A*A$LvQG3le6=Ycp8F9j^n{uIM)}dx!qqoxWCsQ@Jd==JHY%8w$|>TPYW~D|6uS@e{By<S__@&?;kAvk!W(Tw}yXZ343{oy`2wHz_cm}%dLnKPGx~xJ39yasAw1K>&_-q&#NN79Q6KRx4$>Iv$u19m)g9t<n6tkdxM>AIsJOqtt!2aWx2VrcF-TJLqdw`WqHe}e&JwkYv;~l@$Jc&IRBj-0=@nTIGUC5>)-NI25TfFza&M>I`S)3qJO5!@w`Q}G}y;ftL+UocJ6`tve^p4_SeA{qR_G#-rrf@T-$m>ArNS18*H{Ucq7N`?}8ao@ZQ?N`rSTp%>o6}A}a4vmDS^@DxI!E2rzwF`CB{pw>Q@IJ{qj=?d<P=xVH8F0D^M!_9lrkp>5v&&8r$zM_nS+Ql=*sz0V8y-u(kHCQ;U2vOr{y_WRrWn+Ka8Y#w}c)mp(TxBCNV5GuEGYrntu0XMR25pRP3LE4=?bO+tv++Ev3D>QqSdv$qKE<^xUqG4QJmf0pPUCBIq{dII&?ZihZv$Yp0PZyzl=|eQM^Wl;ezOi|aD7UkXnc2rAk+5>nUtX#&BnoD9^MXux0Nodi^~qw~-Z>cbzkhe_{yqrbZ%9B~bPEuMc~i6b;$3GPW*M>r4B-AqI1ir`NkEqPOK#|I$*?V;F9mDwe{XArtv|L5xyYC71JD@Bf62!2vnq0DbY>zC_V4fR?(9K|wTx=51r2knzcpC<0MLBx)+SF+wbG%%Z*Sa|qaJ*)wspVHD?yUH!sa4*`Afo&WMc@MiM<FI>(9v=6t)|?HpqZr2ZMMh=qC(hqE9+Suh=ZaPyA6L5djui&|W?(D^-jOb_W(NXewb$Sw?&hk_aepYxB-stnWUpc~Sz^rTgE9ErUhAPlA{i@V(7Ddt$w({95(i?%w7-D$PPf6hsS8s}Mu?&@wEufq$%{W$1y+SX9P7y0}7J5`f$4Z}<22m#X~!_S&tjzLY%}JPmB;-sU=}uzk?~{z3J47K^J&x?;^%yuji?N<7n8`hx7oU~_*!3pZ!+J7#s?B=kGj++N2zCr-^ZQl*k(ZM+Q*KzU%=JuGDdhTTHJ^_{)F{uWTh1~_Kt`%owM4;cI#&Ca5b0uvGM-NUNtGdkSc+_=*RAiN8t*&tyDl(c_<kIZ^^Zr#I2RF96YZv%Q9Rbh7EpG^>jz$jsNvu)e>$7$mqmp--WlSiLA_|-E^{A1ETV3ex*Fax`NC>7RMeIYC=`Cf1d0KoeDy8!VA`%6@AhWto4bUIbE-n>WctP$qUP^R&Qf;nedmGeigO<9|@DD2n-UeJW@VyheX*Wcf`HMrNmx3l*Vph1hKs=Ca<UC4?J8s(ylSnqG~dY}%me%Qa;CopN!0xviEj<(M}1}z>>hvCU7Y`3oFN6~OTd=zUiDTm?EA8EHX?(Jx^I3!Up(<bNh*)WQ=`AMt|PfyS1elUtN8_czur9R!+Jh*%R7T6qOo`lyPbw@8cNp&H~NeKV0;8kV+FbiPSMF#o4DxH|D=bp1}l?@$$;M?UDeV=gW{e2Pw-v?DTg+*M&ZYi$IanfBZXv_(~$L{?D25so;<cz2R;+*;OlZGaIH8-Ec{&c3Dp3ml50P2jR_$<ah<N4@`7RJ*#+pEhOR6*$n;B(l&`%4-{&BM@#dCe_qXXn9a7_LmE*K`_(@$hUuTh>qtFl8f4+Bg}`&;XhlX~*$|++Z`!A4l}%G@ef}py^~b``8~n<hd(2@x}HAyL^S0xny}i7Yg;WbT>i34087W9FKpGcfPoMQRX8TV&VWOdbht$;AtQfBB8;LGB*J_@_S+nY235F3%$v=TE1dG;WNbvN$yt3UV^nQzbxC~E|>?p4BRkSyT7q{pf)F566`{L&UMqM|BynjuYs)yT>OcdSs)YMml69qF9itsf?apiQ;AVhJXY;EQW<5TKcBIA_?wE2687a2X)2n`{A<#!ojSjN$$kDpA|XjydY{nJPlcbkT3$nM_V-AMlJuxvYV=E9j%BGBAp1kCMW@N)3-`Bo*Vf+$Ptz-w@iwnO!rtodN#RJWMBXj{_yyUx7tgS`!jl!?B*?q?@!7~9j#=8p<5_%)4_%1G5d>o#K?00otnvWs;$%7*PezktUOOB856Tj8ZQ}#dEE@wss%qk7h4%X2yT1uMMNY9?r(gNX_TaBzOI7HHtVGdMkyqPlEbr?+2#1M%zXHkSrM^#kGo|qxy7^&W0AB_tgnO*3>fIvW&|6VV(a69dON-qC=aA!Pdt<rFsaWU1&Q^bKjZJquUUA;oTNRO5i-;sE4_3S<`_MLwHxvVET07XkY7%xgb_=dF$2G1lyR&;0uTYU&?9VanMZ1}m>Hz@B0qInR)%#d&LY<@@_)Mu}KHK>;_U8mRpskLfX`oNm_dlqvJ=4Thl~>jtOwPxVk2-}CtEzi#Z~ZQ`2x-{$t@|4QgliwB@@IVwXfxYJct@N1A7=oE-O2cOOd1{yKG8=0^f)fp4tOphPwLfHgmh6E%57n}7f~H4@C;hv7qt7iTG;FsC0{JrEUpG(2mSTC!mT8CPSTFOBdcbmOjY+E0q2bY4{wyVZ_B#8u&d6mrW?3b+v*D)1Gw(fbPVL22r+$&9azHepZOjZw`P8@xw~wGiS+EJZ1Vq+Hkg*Ry0t^Z;03<Fc5j!&i&+8CCHWXdeI$LRcxkYCJ9T0#6HNQ-o8&&b4a*ZoTz_q!h562h{Q^%A-u;!_qb+RPeXbq*^SH2aMO4jpy|W#v)pYw@M1HUm|Lp?O5^*J8;OpN>IIXxYF7l85YzD-tt>3<LC^q|!06Vps+|+A`y}J5Bd4DvjoxEo-R2<HR<JsH?3)M~<!|}YXC6lRkLiKR9*#^f^y}I3HI}=Cm_00SYFvvf!AM2-LlUW8h{DGK8lW_~bXUEPT)ZhUQeUWAG`>(}E!mpo6&?Fd^u_^6Ai>*jTXTQd$&o#M?`Z=N1x7kas3`KH$BQY$;eZ!t{R_v-!q<1v4sa^gba<g0n%n&a;XJ6P$-X>sgS7kKBKVG@*sue;c1T=i5t?%Anm7y<NPU_CYiX$oV75ef&6=9Wb{mWhGpQqIn!bg+<fPDM^zrOw(!sWuL$rJL!bhGh?87`F}P8(AguN*z`6Tu1W&*O~~7L+e|c#MV@Hw$SdU;j#zAfCe|NT|?$3-y=xTX@lQSrUO8{$7Ip&(qEIyoFbpDKw2|=ch5v#b^BMZ!@Fz8Ha-WiK$gJDc7$*vYK}H`t>(NV!Z;wr?pIBQvH!Xn-9Zn5sbsk@Tu&MI@SOJdNiaxrlN@83M=-;UkduYq+KP@_aB7eBmCUy0{df;W70SvW!0e%Q2qeY&2KmK!s`TaV#V>&WfT&VW?!{yH2lH&bVx6s%;#sb?#-K{2_XB)WH#@5hG*QAlYrmj;qeKz>U7da(>RHzq#dKd$HVA2p3ky%h+h+@9R#-DG6KtYlF;&6e%o*Z&u=?^Y(`#Mvv#fBZYPe_PU3bHIF{uGk<n@gEz30xFSG(DFKOGBNi{@X;5&}r30%YP_`YMuQ4(88l9#l?BxtwX&`ykw<3&NoXmz|+<hFdzupKkbOPWC^j$_kl`9{l$Fz8l0vE0~bTj;PC8F|UT4?^2WY%6pe$8bGA=>%TO3ENS-6$G6qFBuw^jc&C&juktG*G6Z}pzYgH(2ni6)yhjoVQ2(xbRFY%S{<{~w%dW<PE0$D+8AkG(nl%JLkB}YZ~{N}6SEz*yiO-fT#zeJ#@=y5&+dR$k>PYAuVdPQ??y(a6WMkgy2{v{D6pJP=p~@M>4zQ1h-}MFtR(QvxYfBP3Lf#Fi-szUi_De=?`-mKamzETAa1QVuIaBho^7mnpz=y=`W@FbOe5@k^S~btEbVW9u(`LhO*p%u-P_#WytlTcG023Kx!d2ST}4=N4UL@w+SbmUT#gJ+&nD2SvnR8LHo<?h6D+4uLxW8_hf+Splk?zgItk+$eifxh!=Ncj{b)8BH?;Zi6vE=fpPk^B283EeJ7w?Ja1M<<nT+t}$NqFY93SJ;C&RO3I6|4>q`|VL<uqvJhjHMC57FM-pASh+p)6KvENaIp9Z64PLpz_2s1IlUlyEqN8fxXH&(5ZR6Zz<p>Ex6>arjks9tvfCl9!+;l^#8+F|bI3R*vec^s#yRFd9y4Xb2i*)?-i>KSu8clZX6G!6yV^A_0Zi7625AB~Zr4@Wh|aXZUam;We7+^{!T<7H&p$*~qeb&u1_&dSYIh(Qx)~z)xzhjktz-y?}TMN*<#4xeS8)j1{B^^%71TjiKPLSbwDowIYg;G#Dx6p9nRP$I5Z=EbC9yUiQ*GQ?yn*o}EwQ0rkDdeQzeyIKF9Jziu1v^#qVC$Hp8fEShNJ$y`%v=|WGyh}eU~2XpkN)5)}^!xFuJ8(5k!b!uW!jq7W>YwMc_AE}shyWijFZ}gt((RuhV3Sc*i4}gzNv+=G)F)~kl)cY}hlArKf8$UVp8$hf-qo4T0dp(QZvdeyX2KnRj)3YZi%73YTtHnk2Svce$_%AmsTb@0HjWivLpB@h9EBuv8I9#Bj5iGZ}(S#Nz-paqc*Rz{m<6Y|V<SZVaJ$XFB@99VS)#P8Rx*gE!_vh4e`SHD;+w3&nZK1Yl3`{ng6A99fL<D;&6);4p4e?6;<-H!d*Fl9pngpWr;sXuVrok$9$Ik8n@4jKyLVq4ipp8*J{m7IR^c#^*J`PUtYx?oM9!Td_^$dL@tm9ARy6hIdzSnDs{>@JOGnx?md9P<Q9XvMi`0;Vt5BZ6xU=U4}j-L)^uy>AohmY9IdDwVFQbMp`6C(0-R;z1Z@}h8UCc|+wn4QI8?IC`iX%x3oTotn7ZgEtEDM#~OF1DM~^KtDE622+W5_u{c`pPLW^6(7*uKXDPuthc+>5c2M+M`D3pUr8>2~8xV-#mUiFH0!;2i*}3H>WXP8WTpVT&DNdwl~-}KiJ*e9c23HBaX#|O{Xr?)PXaoCSl4E{~;ol+xWJrS+4(EAnPI*qnKik_j)p5mTYZ)UiK^>Rs2-xPd@5aEg6udVrH|NIO0~BSxDvsQ7c7TcQyTn9Mfzb#nb8G3hhC5kLrs}ox$j5N6-J6nU-lBhX}{b;`5<&zRpjD@V=jj56_~($x8XJFkOC%f+8M1-RCH5Vvy<eQ?~E3==YagiW~`KZGKKCSegq6isd`7h&Db04q?4O-_9bS7}hNWjD)Df??R<;0YCdnfQD^ddi>*s7IL}1ntyDb&tj1N_*h*x3okU!p3o<*cvQwmtbkH<MCfZ)E+7<A6a3AFp_bMEFdn22m%M|b=kx51ef%B8oBV9Ltd(gjml{?}_Tro<I*s;Whb?_mNbUiP?)1eyQZ$X+@?G+3x^a`fb{AOe&eV^Fz<sw!u%j<~@iF*zHd)p@?S2BaX4HxO@xTvelMx0xh-r;9_}T2o=fIPZcvPw@)y-)oA5NwZp;#U^MiU<frTn(!C9J0>qd4op?a6rF0JH_E$G3;Dir1hH$K+dI_CATs>~-j87}cZm)A3*w&kOUFvb+Qa4WYF2-_N3^5PP#GO-3`EoIM%9M*nyUNzGy|FALFdaDTt|^kVhHwY}}l?K}Iu!=u%My|wjzj{&mIQRDMCJQ>r*S#vZw)>rY?-a(Iq4_W{^oZz=wy&%}RiYB9K=!UNNdOik|JXA|YaiXEh+Ue}LF^g%ZVE&{BFnN0i7`-TwSV$A>M%jABPt&*wJyDzLweJq=weNP1PxKr5+2F5_hx3#30G|giMF#alf8`Tnr879X0Y!1R*WXxMKj?29>8u1w^?r<3d46i#|Ay}CjhM6mX+!;do~(F!otAY{A*hok6TFsMQO)ebc=B=nl>YeydJ)qf)|<oGM9gZfek0Mf!>1`xE{-6$sf#!C2M@-2iG<|Btc`t`Y9v9<XN_s>!#M2e3se*nen@4HfaOGSy(?eS#e4L9S`(k^hh4`rjxLJT(40{nH*``c;_dWv>CHu*DvGmN!3G;24{7KW%yqlJw!cX_vXO2x(Nl`dJrHrd^$JR-tlRX@D6peO!-@eInIPX_h$W&5lc$E|PaM}>7zXu!l3GJ+>>tOX6Y99gqf}coJcjnEkvUwh>!c^G_a+do<+f@|*^t9p^l5Mf*8!jfL{_r?Ya{<Oi2QB>{NhJ7=elWF;-7jWAl^7>o}UpX)`EH&coE*wPMRn2W7%`11kJwQA3qvSCoIvlo%g%i?vwcmvD^(pXJFbHh|!##jfQyj(7<vduXIl^$Yj<G&(4L{uvV|*H?m3Dx*Qx216gWH`-Qc#&Dt>QMWg$+KOAX6jFEFV0H%%Nq)1xREKxudss15^zBUz*iLfZmp8p815`<hZmunEBgMl2yU?9h#H+33{;93|v^@axK1)|;`+*|vroxOoPr^8H3k)6b_)CpbGs?(dtqe<Y82r$FMi5U}DxtOay6%I9bwab6M{MXBWxcryPf7S5eAH|2~6ios=mTyGyX~Qq%NldYS#oJos4u7WAa(l3O_u$@Ey;<7Ac?pmwo$jSgo9sYecHUKZn+f}o%UCn4rgr&1mK|;*&&XHonf_Cb_5>>A&}U7C@;>rMh26XS=*25`-9O_!PIh9y;?VA2Xt$xT=eyGhLxzSn|8aB*AUMP<`cFW34*hvajmeEV4s4VAq!ehAIac_@CU2xPOKN*l@Ox9%n(FQJE0*VgAk>JtnjYov&!<ny(nP$RPX)g(KP$ZXIE>Hc+9sD~K~t(HeWh3As=@AE+TP?HdnkLozx@GZb<JpKR*QIe?f$`CdSJC0n$xByGWblZIqimKdi+d;3~4od#fzpP%6$OQ%&}g0-J}|AkA9#Ke12wo4b3)8dUhMCc8B6mv>Kk^v8V$psK3Pz>3!Sf#tzo@-xnoav}SuYJ>2MXzwwRwX>mDm)_^){H8j`bhp8<@&+(|CwLFJAb-*ziS`FVDTHE4EX^WYj;VYhbOc3P|2LI%NFrLM|<zP;Fi0^o4^T_hUBY19OXr3UWjLpSEqNC?>4N|d*=mY%ZQax?;r@j6i{-sU7h_1_DGFo{w0<YsKzd_#cFC9?4W3|;^zAplKwHlt|M~53r_w~~)cmCZ$lbx2O{!XC@{96l@>~vf)6I=L6OgkQdM_@$OFBVwtVDn(BpR=5vJ4p4M@APzo{G}#0c*R-iPj!V8^)x$z8UV+&NWRn|0pFYyK`$|_e)&%*{ao6j&x+{;^qi4(-no1sf*13E5R!<$Cwo;d@V9>Xze{2HJT;4HuQJ`e6Z_ZCb%ps+KX1@+pLJyc&<}L9|Ep#8Gt)q<44seue<pgQMu*V0I2BPAj1HBNrvqS{{$+M1LoY5O{qnyRQ%C5^Vxl=+$RELT`HTufmUEP&o-Z}gG--sNrY;3K<U&6#tUaBb52Q7|5Lpl8$&go<pUc1+NnKr8oBHKH^O!!RZCA+5kzQP*dKN<?SS~$e{VFAXx^BT8YCeuM$fJ;?;q`Cyg+(p6r4)GkOt|og!MdbkxLUou>UC)K7h)D@4xdXU$DR*P5AgExbA3@D5cf0ZUVct_CyL2C<oW3(Aq09^ge>;Hu$BE%uaGR_9VHvZl0N)C-QL%$B@kFG4XV;u5HP>J{#%|YH4TPbKP4e0MM<|9kF|W-6gehGkR5#&Jgj7UQN7CHh<sy%opGF^jFKiM$Uhhi$HVzxP@Bc0q{?ui=go1?Smh7%$=o0HNRW!}Q9KKKLO#9ZH6)oQlTM=8*<?0PhL2(EiSGmQw)Bjq@{P_Cg}~{(6>~`^1a9aJR3unq;MEh=oXk&0K<Wf$(?V7FRdI<b^`>~6%63bzsHT*wSJ03fUPeHPo2X2M`!byOhOjS$83DRp7xccFQ@dVYB_pv%6*NI^|17R$w6E}p&d!6yvT_0{ic2f1roS5vA8G6#>OHu2>OWq=c;_eGj%A!ZUOl6;9pmG!>7G5-%r>3`lW7!ByC%NRCZl1by%RZ67`v-_<ra!4zCsSO^I6xlsRloMc+3OuzGDPtXq&6yWCXnUPU0q=Bw0-+<9XL~(8Ahu=#LsRe>_{6QKaUBYwuMzQy7Gty>2p!AFuxCd^R5@PgZ2`Pj`0ahw(}f&p(di@#?XE#x1R$`qSg#cqN$3=abW}iLozlD|oM*zln85mbmHfbGfP}R6;@hhHy;0doy?s)d*7(yZ|fnFZkI_jN_^yPA5FVGr6z>fBN8>=7UNrydcJtkMXNoZQ<wB;TS&~#b^FwJOc;fy~=hovWk!vW`)(_w026fYP*i(2CEa;sL5#Zara~xL8ZOj<O%L8P-Zo4*)aY>ZAW-@IoO`Ov9*p7#m9|zEYo-Wr14J6jJ&|8zZGM4c)kVs<-f&c!Q3&QoOL~nRkEw^g>EF+b!9gEB<`Bla<;jF^+^+-x1d*u6zTGywc8ku!)Z(|j4Q-5X0(=zj0PqXmXrAH2FaJa4u}(eK8+J_&PlDJ@mV|{kZ_p}*+n~`Dd3)v&yL^fae#BZo9h5BQ>r?dK%)2Lw<YaJ!Stp!QU&toG?SL&>P2B$^&&ZG;go&GtErH#7SEN6g)|8_#oK78btTty?>23IO4arj6<}9L<Rw!rG$$&oQk|&r8>u<vd2>;bE>;zfP?;FkWJ>-23)ATJj4t`w5`A}X=MGTPzV;NqT>ORhG~<p7`B~25MHR<~qsc6mO0;5nOFN%~w6VdhwY{E-aWG%`k907#<Y&l#>u=oY@AVFkiayX(q}I+({8`+H@N&;=NZiwi;<-N@Rrsuh$4o0Wz2J`JI#Z`|!Q&xawC~4H<X&l3(f#fBw|73=mZhaYq<ZQlD$eHqbl!-^QLj+*As0Sc6+<A&U{XYNEd>`mLLo3niqKE=sNRqTO5>cM5^jm*N)64e@Ynivdh2G@n&5Y+1^Un^OkHza=iS>cImT2#^E8#;>tBfwTRzG8N`{MySkvESZoSHA(=P=kPVMX$>FGQQ2mTocxGx+&lM_qC6(8$hK%j>(uJL9@L)Y~BVb^ny7KlENd$ocqK~kJ~M^I7nP4zZI{GDErZ7Sqak!Vul;?+arXtnrTDk&5FDYbub^C|b>f?&r}(TxlJss?z?aODfVrTKfWG=JUVcd59j<Vty_i}evQFTK1(C!u9#(jJebTUx>SjtY}q?}~-1hZEXKp2v~i<%!VwZM`cdL<eLVu*}m-ofuzK?=zH2Rr$V2e*E}+NNSCu3~Sc`7iNv?*B^dNAIc&FZ2zxd$3O9z&ODr*ea7Ng!HZJ-ya-{SBbv{(#XV0x@BFIlcxEp($1CtD()@tP7g~ken0G3tPgtUZ<C{O45U{|a@WlxXp9B{VraYU%?c3)+01aIV%&6WU==0ep8SV+5if7t<!h0aua5Nqc$w_)VqP-G)iD_qB6G1V}yr~&x=+sBQ9t~&1;zkKuXG|&C++aQ#@Lnz?ynjqlqu|!#$@ED>qmWbHjnfEEjwj+tPvlafNBproi9+=~O{-}R22{sjaM9JC%0@1<!>1fSnzm4H(!Uqor=p>YBS;t8A5_(18V%c=olHKKUwTzTsiJYF!`P+SF!6u-+nnu4&pv;eIR^hEPC=_ZD{ON7i-;F}CX+kLBq@u@Eg{QaUjClSbkjzt{emPfoDIviJ<_wn(vp&KGrO=TXb*ZE#vclDG%IKag3se=`n9GHkH?c~tT!|{Gd+F7pZm9`zSwR%fpsx@q9s%Rn4oeTHLvSAhxE>(CI@89nv1R1|J*Y?v1zoEBxt|k3{O_<>N7mFm-Uaitk5WZFFTu*vpWAE58a5wG~&b?xi7W-7&gLuCZmGWeJk->Azv-!&6K`fp6w->YyUmXheMOkwWxBe6b}6Q=XB(U-EJ?)Irtgt?mvhxFNH&!4!?ZO+jo>``1*C)Fs9=zU!|w=izf*y6Z3w)nw|NpJm$V|<n8M$KN|J(MJWxdaNFhMlu)B7SH^P@UM4wRkg}YQLX-Z;7h)XG)w5wQzx6z$wr97TBt7!+pLN*9b}EiO6q_y{eXty(6S^I1#d8yX#Ys$arDOZXiWNs4$BH_(7k~4k4>n<*{r>v>y}m}<IND8((x0sCY;S#}W$BYjd#133IHoX<aVl!EcjWS!@oLe=Qg##<z^XQ;m>7+tYqTL273ZEJJqM9JYw+1qsMA!bDZ)!+AsM{i|A@ip`t3VS^r2Rn45d*{hN5a^t_PONReQQ(#C-a#$rGOr`+OEpi7!gU(dB_0IK>y_eQoG!4wcFX7;0VgiN=*Sz=M-@)`=uAt(qt>JgYQ71<I(UW>Vt=LU7_^h6n21#i1+4F#p&MAUqykxocgVc4^a#my7j^)+k;i7U3ph!gy*1d-uhZ={&e{nIh4a4!_X2XioBF?ZGv<F*%<+j4>r?vParHAjji#5hKvZ^UGDF9?x>gLFjN!=XzIraP3VUsS5FZaP8LC4yV<i1;<+)3!?L9xs)#_Rj2J2sj#jo&93ck_L2vg`0>w%O%X8%raiq0J8d$%`IKXoIkE21#RCPU>;tlI?ritc*xKek1&&U}>bPW?<_R=P3<WQ(n5Pe}t)0(LCez_3oXzaPwJru2`%^6K2iI;e85$@P67%qnuti3cUk8P6J{m`|n>Xkb)Prk{>PTtze-EyGe?_LjUOOAEd?12*xN1tCwgS0mc~sN2`&d9LYsVC;Nx!DKgIChr!7G&P`N6e|;@O{n(Bx56K4&WTlD<fjrOSQW7pdKl2F?hlHYa!1?j9WMu4I6Y%m{)5d|;kcwy2QPyx+XPx0NNTS9h(2UQ;kSW)zSdl#a-MaR#Gf-nibNKq<<Q(yLLft@=_!>U=utrLWNnPd_)=F)^diw3?nWk1k`Hz*9HUf+c&em$HLonY3$Yr$MX^<<k>Afwg^B)8#Y0fqphRMqTOYVyu2s#P|(JYp0%|bf6VdlZy4gV(aT<d2aN#H~Sk8t}WI?;@C0FhGW=rfDf*5up(zE+1VRx?+o@x-R)Pr-Rn7qlfFe0oF27M*U=S${aBk%sU8?TQW{Yu+S5WK7prtofwgq+X+A|FXzjZkw077vZ3^SOR&9YR@x`>}2b17OJ>new4(mclPFiQ`x|V@*X=x#JVVgg~>el(Vhse4>S6}>rqsS;E?WY1AzBQiwir14&9-*uz87Ibhq0}J>A`E6HzS4cr@Ft|$DZql1WlFMdrimQgA)PT&Prs$9M|SSr+vLO&y}SYPGEf$F(V>?jzo+O_y1bR-5DM_r4700=$ORdYM=w8@QFLDl<x&cDS>_U9@&`}m@vJhmv^b4|4!lPPLa66Bn?F7WToUo6Ge3d#hQ)mG-D#Zkp5`M~D7CQoG8T{xRN@rLY~tK$GiW(cOzd03%ZJ&E<%r>Tv@{387i<K7Eu-`x|FvK=31;a96F(QX42bJP_@^{Y-DgTt)!%7kC~`zUD8Lf-IkVrF>O~J!=QB1xgn9H6dAmljYt)9iOm3o2H@QzY^&31W=BK4ZqoRUpX;?W{&Ubp6E-0ig!2zk;8W3jyOmJ(^-(KI@*xbIeY;`XXoEomCIOW>i+&$<~kDElLSxw<Fksw6oxIem~-xNviL_SiP;f@DefXh?=;{m1Jd)&LKDuS%EZN(*JC8|{7%4I#{`0>0}d(5c$F$rkd?L(a$K{LvVl4y?)bz*4!2*5Css;aV;dV}3h6|zlfQ{CO+J*HC7b~ebm8Z_PO6)ItdK<R~Kp32<;lux<O87v3)HupKpwn_;8wJ<$d!>bx6z}fF{nM0kDw#;X$Jd*<(#p9YNiVnIK4~L#T_2o;0VFVgxUP)2*0)>?~D@H&!>TssKUkMCay1ydL<MM(7*`9SZ&gb#60Ey5WtZ^CMmj>{qcd3XB9#9<m;$%^Z8$YG*AWb?Y58L4fGOXv5seW{*lXaga&cTZ1Bek3vNw1g3Tsg{r%10sGIiJZ(TVyVz7cz+GCB-UnuudVB-{p&&Tu4ZLaUNVc^U48^FV%taDGxo*O)S%6qx<3n1;jfANh&u+ttz=dg&3zWJ53SkfCW|w6{ueJB#q4^<G!oyyuYf+do6lT1(Q9v_9z9?M}Q%s=2`El!YFu5nscF5b!@p?m4ia1U-h5WXi##V2`#?lwjA=M{&}9lRx5OjAc@}KDa?vp?J2ay)lzDvu;*zIp~!(%3OP*Of~iL<O>0(}59t3&(`s{I+^@;GD3hQqup+G`d6ADy*~(FXzYry#%Y>6(D<P`s<wP=Xj4&;)x)2L+<>y&|+Sl2N{rT&^V-1={v^H}c&SwWc6Lk1gD)g`Eo*|Cw6Q)z1YcsW*o^z{NIv_T?i&@oEM|-1iiJCYWpj;m!ee`5deTH8BQ+dH@<*YesATh}y!ls|f8Axre&$C!T8Hki8=u{J)6%t*go>IAyD&r$dcW33f%yV}6U4d8V2#1nbi%$R6<y2q4p5N^BlBYb4p2||olZ*F@Ws~?Dk!3!WT3;5kxT>p9gk#C>icL28^Uu<#RqFQ7r5q%t_*^D~Dg<19NkO}8L;r1>sY#&$C7ennY8w13#BFiM<(bR|pP}z8Js9!+y{(mEWc0V3@+j)G{LnP*H^`E#*f;gOI^=o%z?K(HrRjtD`W-e_v@E>6p=BAaZ>HJPZ|>~fS=-+H0cUN~?ocrJ!pYKke9B2;7Obi?g)yBG)v|kmzPHBTMZ!J_K%QO{%@ndJ@cJo?j6n*&Ym79?W&+G&{9oZFA$eDj)I<carMg}PAk#1l)AJc6dnK0$6gPQ+o1&Ux)gsihh}yUhD(Go45|3;30*pKCS{5*BdPcnD{fuUFd0JNFp1DaaE$*|==GlQ*1|_m`JO{9n9ZBI~b((>N-lkrTr;ita8^hw##9b^&C{QYvZ~JF;b{Cp%q<!s{y53-@H4-pi@6xczRZ%0K=YV4P@MhXxqIYvt&=BMsoB)z>3dV~HH?_)UhYOuL5>Y0z`SNI7{J1cC>B(f7B_hiMw5rWd&w5mVtcStuJV}O+Ys>~Pa98o;vNy{+*<m`t84N|;95rMtMhH~SJzh+aYp=(!b;)Fs&C{ze?zsYC5al+0brmJIfv$KBsDeeufL?9lrvH0OJ2~?J)@-9;IY#~JS|#wWGNM5tX+rbvNzqCk+~1ZLq0srvB|Sb*df&imR$tAJ>w-TT`p4O6-Sk8_>R$4y;0S`w)nz!Yr~1|N9(i5$R5Z?R3ghpl2x=oN&-0%zmHLuC<StsG>3Ku??R!sk;fvS1y6TCag!6d5GNa7Fr}{-he2{5Xl-Z;!9hS9NcA0*shr-?E4*h9U1T@xHs}iAP$y|k7mtC&%Jh9ry6`j%9D?s9Nn(bOJf31s|Upb&v$WGDG<l}f+t6$OCAN8h^iC6P}rz)@~eP+RQ^6@O54ph$+H?X{3+9EEla(v+rwX~Ki-j+*;DrwFjbD<b}!i_KT*6<JXkJy8ihO*ohLeGlZM7lf*y0*2sktNf;OLv)E(S;(PrG5Uh7-%UI@vh&uIh)zGL7}bHx8ulLo7;SIMb42|F=DkMc%zWHy6>DG0D|D5pa&%h7)W_mNDt6`^LCnWUxp9gqG%@CZV1^%uhrk~HOZ*__PZP@E>bO3-1>NQo$~|K8@hP3wMjmEIul5_zVx?jFcZ>Obds;!96jk~(_SH|MZYD)!yGL_yoNa2T5$A%FBULOMePJU(hyn#JF9;_D*FE+^y3o*7vBKEEd)qj0nEe-UACP<X?J)`HGdeDSAj3x&LR!kqufbIYvGYfAj~yhFME|!5TB+(TZl_f{ragRm}}R-62qw4ukTllBi*ywqSs-HDR1g@Mb@EiSl#Z(4}d}&{CPGBhZ9r%VyRzj`NbVa$w+=~^XGyWTfbwam7h$`iRiK{1Pu7^ob!j$qgJ;|XSUNngVXplh^J8b-2}FDxl%i=G>*;%=1}_PFBEsu;=wtcy(s_e^3Q1cU~g@EpAWB!9&H#7|D*I~2CRz~23b<!ZWr2P_=vp9W&2ieG98*atNHbA<!lM=IcDxTZk(MF0~IGr&86p$pVSKOcy-JDOgO5!inFsiWL|nJ8U<jfvhEgRA!4MSZkXM!p%;P{mY<pbTzJBMA@YmnH|w#pNuE!>e3m*NyX??X4-7B!DamLUKBG?I){Fj|SM0wjbn|OIS@6Z>%bSZi09Ml@fE0G~Yk93>cGDv{lYT8PXI+wXn~%O%`t@Gr!^x8Ov!{=aBmVQ{|D=op&4rb>wz0N*(3iIo6%N6OyY4<3+*{k;yxrelw#%lU`r~0j0W~)bYXIA0JUM+bpv5tGWG-&6rM1fD)5^&)a~3tTFx=YV(^UFCjGyS;hp8~uyN?7Sl3xpZ9y&qO!EiiC`;?WYA5d18Y0B6irgurfKT3Ck>4%vJzB)WAR(s;lC~P1O=YvuFC?4tEnqK(MsB=k8i&dSTkB8y#46BvL7EC51=&;Gd`RRo6uZoJr*NeX-`TkZl)0|A_Sl0fKGQ!Fvjr-yX!tH|rBVKl;>1;T>^#DXbh@RhFxU~bjG`ssp*2dj^i)y6aP!Fi9nL5i{C+ZR#rH!W-^;GH{)t9Nocr-kw@PCEOqY0EMn-2b>zzf43>(D~4a}G+oUy4<D*sCdH4ns!og{c$Ih4<{Kp53>`Ae{4lrVq$$DLfFzQ5rhk%%)#70Se9o>S7o!*)b;unmk-e5l#RwOr8j|FaWhd41c^xiYP|WX&m)RDq_(+C<}h4SCr-56!pm87hZhx<j5NIjm`b_9ZGjFxOac+U{l^LT{-?@AF8MzvYn(3-$TE=bILkQ)nDI96%b9|QZT-puw}v=sTyK!tY!EBd~*y|6LE$5Oxt<?;NA{_E}y-J?D^igt*M@HaUkYwI+-NeBvC8Qh$<`?;>r9`ei<vQ!bkCxGL-?(#b;=W#5Ez)(K+porq}-|+CpuAz@B)%cXspp0Btp$;7Ffv4B-=(K0C^1{G%%QimG7uY~3gG8go+s^93MZx)UeOji(nX-0FYO-zwd;x_0NzUjGh6)vC6=lghz9MZ+mC9_%P@VD8p;_BQyUu5!C+jy}Fl7v&GOJ`w}ipoF$`!Fyg0-=nHul-X?&W-r4|SJ1t#@YMZnp~p9rwm;n5=!4a-3X)LzfllNqs4RHiMn2X{D}@uWziG9n@eUOWPpL{Aou2=PsZ<SlmNT55iZLMwGaE1!iv{&d3qLKlTrN;DcN#ly*Aa|hU7Agr{#G(lX;$de?H%>@YzR7<`@#L~?1uRYIKxxjB`3w3@y8qIzF0rVHY7Ph<KvW6(vlo&1vxA`FCvIYN)8IEQ@Jad(?HeBh(TfGqbi}FZTd;EU(02E?aCb3`HNdybP4sHzNt(6Qe3%*xe1<|F7=}ug<%%M=8FxBo3P5J`)D-zxc5}91I3)hSA+gj&D#7~rIu1H+|uir3MxGpw&>f`%Vv4$ZDHt^9;!z5+f*!U{EC_<R@FoMjdgxwo4J%iw#-T@f++pAAkO)uEKTWARRobp8P_lxWQ&xls?%vrCFCHBk9oxj%PCN4g<M%GhmFt9=UAX;3)a+9A(z@!6ptq4TUpx*vnsc#P?Dec;ryJP8v5JFM<8234t{$fye-dsE5rdrmg2X|eRd0)^k~U})HUE4iWTI2B@Hs1uTp*-M-6-c%%Pn(&9GMtSF4d~Ii-u{=JE7wN+-q1JDmfk33TA}5igZlG2<x7jVwZUrp0`qncJjWTmilP^U=JaeS8A*MVpLAPbfd=kdBl@8tr{O^2hTu3QePvrc;?T^{;j|w`a}bjWPbP!2^%S=chn#^SI0|4Raw4b){SWtXCc3P~pT>;wCDctyP?~Oj+hrOipAG1^9MuZcUdy+Oq`@Eg!_C%RDdMa1NcOG=ALu|7wW+y}g~iZ#+ID8xSpv8inVk=g9Mf)p$xFiNzeyX-#}yKb=zY#;V;Hbu3i9c<7wDq%5Mqf0Z&Sq514;PiN<+wMT~!t_d1ExCZ^ku`xW=53c2YXoceS3vr?J)2f;aF_&rtPeHV)wW$%<B8zhGP^!_R8_S?@5tF1yrPP_Dic4DjVI*o9@?}n0;%2I17^TrJ>TbtkaC2&9mI^lX^!S|EcG)mYwrTQfLKj1A@us|oUB=j^`G;hOCETJDJqwJP5a%ApW}anQ1jVX-s$bl!o~G4&=qYWup!`ytN730FSml@7D!u;Z=q!s0FZ8$&0xrkDyS7JH*H@5$qK(s~1I}4WfPCCSmC!v3P%b1oD8G1a+NYA8Y?R6Q9I_qihD5u56()R};g;l>5-GoY{7bMAU%?{hViQ&`U0-|9&!%OcX7|xB{#aeM!fZRApQI5SQqfkYhsZF=VJaVwddf4!q{H{x)2aGMD2vRF;q^AH3KpwVl84dQuc`I~mNhKnsx>1AynFYfeZI(e(fwTl8fA<=jCuumvU=82NzVq;;#Qbkk!KZpsIo_u8fh3U!X=gt=?oEis?2eho-Z7*JG|@XzVQ^xL5GP?vIu8A(bKrlLF&_$&jHKw!wbu|LTS@}I6jZx3Y;XpvmZC*Rj#*zn`&qE)$z1<mNe5|Gy#CPd#mG+863X}+nDpI%&Inyh0!g$&;H)>ORFiqEtL9ZUSa7OMt6HvwB=<Z;AW{U_SBWz?-hGp|JHaWle3eDQYh}P-|gR9qxC}fz2`@HzAZKyPpg<{9F|$A;_hqNW|0=@6<IoY8D(;!$rN3xm8rN=B4YjRfMVGF7OB1Wz5D%x!M*;$+Q!<!+TaRo_-(JyYEmweT#wV?OUIN1iD8ZsFWx9U3$U!Rh<b+R<jIUOJc+xFX1!|S3L-YiIIA5l=h9^~K7yqHRJs5^DjQP)NV#`CV<~z{bOr_0<Gq0LJRZ(q*s)PbI&xL302Wz;8s)M<nRGr0QJwm8@bIknl$>>?o)y8>V$X9NQN$OO7w%P#LEgnyt=b{kd*aTrCO(eAy4hu;;x?wHyf#za0!B%BBVJ4=#wl#%6t?K}$zU`*9nO2YIQG18fBpT9TbR;&l>PqbLVs|rzFP6_{Df&H$fLe_aF-5((be$9Gh-AgOUz2I4P=mfVLmdOp^=K>!^}HM^T|;ir{{95@JHb)7|x%h-t+RVLB9J!5s6~^Wl(r{Iy{~(d>)XYj|ahUIzORHVWZfm4@afzWJ3}6_}0BFdkTmqhv89A{%!g*mH;|s{-79TN&&K%281a||A3YxC2A?XQg3os2O>s$Bo?z?b>)Y2<CE#t)fG3(Kx{x%kMUck<T6TtLFSVI`&x!!O+IZ%c>q8VbM!%x7Ke+2sH~`>*c)mV6T_s=DB1*)0N}0vW%)7YO_xv04K-z=>zdi=7!Av?TJ<BobglL_L#eJ(@(PyULLK?+%a`+B%D2E%1+)DOvocdg8=Z^rZK(Fc0y9;R@L?RE$&|i}gQ8Oj7`JSpJnOvW4(_F;{w*_Dqr~pnk}PEqjfQ+Ub^fF$``Bc^Qt;%B1Fs)u-9G|29uFbWYSDCZHufbsD$T{L!6s0)S#Q#zWchDHCy&GW&i3uC&GiF1hCr8^9egzSsK2&X_7vq(fZ7~nGd6p3$BX>Zq^SS@l|SX=ZXVBT6ep`pz)^2}2Gl;T5<~<8MF7QHG!L<r9j|ACmCwV_aQAx>T2^zvBzGN)d^DuvS4%w7n~IBI?oTNNvwt29=UqA%#<`6njoc{cJKy#p7pf-0df2z3GMd;Fjpr-bV~vBFRIl;9%{zNyPc$`eS79$trja03vBJ{z<nefNO7ixwdIhMn(OpaqCSBBG-*d!4`G;n+VH)(G_{V5gof``=EOwBWx_FZ%0{80Zda3rsT8<}^s3(S+3Q6rR^SYI1O0|N~DvvW!eF90oi=KC@?O72(Gn)IonrQlvUOQUJnib!%!h%_?+De`gR2<KHHLmAfV)Ik~@$eKaoAHX<Fq(#YU9DWyRj)Vcy3ndhODvFED?H}mn73iN{H39n=n4a^_+W_#X$%lGoX*5%a!x8TRF0=HXPY0mXW?A$CUn(VEgxu|43}sqZa*DJ*S9Fd6ljWHYxQN$EBUrqlw>uL2$nSz=#*8Yf{L6tWv6n9Z1s{@DUGX<k)UEcM-6`#g68ZpW@+3I|B|RVT2xSKrxvgY4BMVspgAOwX&kNcz7h|=$J_Z-vRUjV-A3LqSZ?sIvd{1JY(r&xKDMP)$DJ+oGXHTQd<<;Vqsk8p{B^X->?40EbJ<Y^{!+B+Sv(z(xPa7WsbO4SE*U8Q4$99c7cl<vMPHO`iY`zh%S)z}C`nQdwo0ZxZ%=C<pFrq}{+2GF5w>Ga{)4r>JFq$TVUt2RuJ7l$E%c|{pNmH3rljV?czaLz4}QZsfM_LSANVjGrZ`NbjeAekNkZvtGUYGD$?Z1nq+kkGoW8q)vSoexmcuRGg1Wb$sx(31alYimVIY7EJe>(&ho<n#>?Ttuo2!*5PUvDHcFd5v%=Q`<v3Ee1fcP*Y&!Io3b06a93C5=9aWI4Ry`BAi1~DQj{@Na0U0$w8PTC_>Ks(k)N*=9ZQHtM{VD$}HqX+_zjuxO&$v#fq1~olvM{i^wmn1-;d$nG0zJToVsn@^LfaJP5Z{5-R6jSh(yzT7oL@E-_UM6Z+t%wxZH1s@v9L4Bv0ER{QVMXT~M#R@8xaR=Oib)?wlaIwld>T|mpnwqO=ABjQ;|lMOpAjM>_!FMQ;lmkye?(j}i$~xU$WXOvid&om$!bJm!C1gjIrEMD0L!4M?TeQuW0{RLe(Ylu@r+W4hm%tX1pgR3MmsqQdXoH3SECL-+}z&S`EbAYblzo<M;jHCn~>cNyL+4W=&YPtQa?JJQ%oFzA>Ib$lA_A}!ifyo%=+3^pZIZ*rOYYpM!Z8L(jv-{PX1#;Ae8Amhhc;8#2*t%gDlXpc$;)TZ%9VJ$nrXk{gGO)5*F%b$dn1nn>=%EG94Zd$35}O%1I;J&WcGN`ZOsy0o4SyyF0rw%)ghR`)Y9i#>Jz;%|+8rZAnwU{738&o;V(IgMOneKCOZy(s1<Z4LM0oc3V`)U3FZ_{P79Z^mrB7VX|GZ{V(b)AH<~Y@}!|H?r#OQ4xS873dt_igO(>5liuCg+x)@K_I_nv2MLHpCwjfAQC)GyGR0{neps-n;vd)OZ@p2ety1{Dyqz&0zD!$_aCFJJ{N<<A8I+$-r=)XLuA+Xc-<Nm4-~c(gf-Yi-{W*=S&`3|}Z{L=njT{$nF($Gm+HcT}*vbMqnS9*ShvRXK6_~~?_4LU&CcJ__<P|*e&*<)vWr+#EafhOVz!-~V-qH3Lq-(Jso@kt3m$uF0GoNt7=n3acnbB<tr1#Hfu{KGx6H{8cf^Y4cCvI~IPdCQI)zph6>oon}QX!lhOoHY0t`rLLyAT{~`ACkHqY%$d>Ptg_^eoh3AoP@YV~i^3Qa9Q-+Urv$oc#m**%%yrxFdjzR5#0mi}J0fom>0;y$@*LoHm}{OE)p@ZSK=~Wc`hOW3@4mF_8mOAE=ZhQjQ9D`+H(pzDX>|w&`uK_5CgJI%LhAO;}mm8&b*@yuQnIW5HTmTC8YYEuUmsY|DR@GfFGY%wxQc=zB&=SoPy43+$C9!IStvv_KS~!w3-nVmC?OoyN*Gk)nbHU^*YCl1#*6fQ)C;$$Sz{=*$Kwx8Z#Bq#?nLST2B4vy)*m*B1j2-gY-=Nw?FKz!XZhoMw;^oJazsBk`#VV|-TF2V&9tCK9C<)s7@VhS|iT;WMSvA!>P|o%GBUXm+)y70UXew2Ad+rQZuf+i=>J_lA^NS+T3r?>REFm~vUO$O89r`AyG@nHLMUSY?^YI7M%knXa6^>{XVXtdO}<UOkO&{r5CayvQ%dDabVOjBn(6#n)?bVzQT%mrdSNFD?XmA(KnW%x_<D+RUHiDU|<)kFauzKEBlJD=p0u_xkVCE<fj^wXc6kPtw$IulS}87&(nJ!5b4#B%R}9zI=&@yu>5R2lx6HO}b?1szbBt1-`F;qxB8zX5Vn|pZgZ6<UU_=gjoE<DfRd?Z+5-NSK?NX-xM>f|B-^cf5y3gf1>1;E}s77<c6RACtL~QxVF*u!stKuN)S>8>BB7WK1%QD%@zlKwF5KuI*D($+`w^N%ZoZrr(;El9XL_D6{K}**L*jKV&4qfUKBY&;(MlV*rw5Tg0N!+iJO;<W5-Kk%Qt+#;{>kld6t7J{kUbdUBfi;lBQwDZX6_@;f9XodS1u3+$fA~FAjV!vGbB)7}<6w39TTEoR-t}0y_@dp>Mcm5}>0>>q%r=2Kwo=ZKvg$UMr4l$MxMXa$P5#DsCMm+|Ug@(~KM^u)I##0pU795Jiz?C?(rL<e+&YGLyvdgCuY)+v)^y?7N-7HI<S^V4JaRxuzXkPTUH;c4C?BjsrRyk>@J?1R>0J&<^~>ahxc$K`Tfz*Yn#gJd7iy_1JAEPRDE6wvEZLJC^CTBfn)@F=}@ddbT@J$4%O?=b$#%v3xV>fWRG$F9-u$X+5&SByQW7d*8%}Y|l1qH|cnh-^P@N%7`343EHhrr<Jrl#|o^D5jaMev^=9@wCtABPp{)RR&1N5)ixcw<N0RWHajs05IIz>Qm5(Qu@#^a%sX+CXLT@HUTC1Pwo+$kgkI2!5<j#t9>ehbmfJ!vUDRqgzLl2*<t)=ms1noZpheqGYzu@l4KD^U)DeMjfn~S-4(jw=-$Gwnk<qpsGclZ4rIPP-lBm@RlF%`UIPI1l$3fcyD|+hK1CTxO!Z_@>k<+n#KeXCy&j8aqAy!EzFWE8!!-_*6jYW=|1RYGh>2}(n8EDW}T2Cy`X<4q{vTO&fcRVwU+8&07ne?EU6)Gj55J(aQe%ry)KpE2teY+h;uH}24vWk<iMXZU&!NeWY^IAbCLJus<4Ba?XO2)1OE+!h;j?p&!sBL>u&`x4>1>;gU*)?piT>x6OvA|kB1_;iITLE~+YAIv)!cGhl`gVvpY7qmtZsK)p$BqmGQ>?TeL23kT*KIoq_!#TR4?@F04=tzNiWRwJN038F+%l{HBQn5EZQqSd&$1HV@>EWC1FWLhh(VAR)+R`q_&#w!%W~Yftxzcf!(a#ySf&e7xn2t;+ewRLWy??{gX3b2x8lU~I{|p!>ewzQ9|w^iMW|lkc_R#*c8DqT;x@VwdsZv%#9;`r(~iTSqw;4v!JJ^N`GFHVfd`>UYZyYv4_c1G^8v^aIgsB8x`|pn&q*MCJP*<dGSye=>?E<<f;<2lIYvUu+>4U7W3{8UX$HQ^A-)r0iUKb-uy$MzOTA@y2_&L}p?T_xOA?F{OpG~l60p8$dJrBE_kq`HD`N*sfFW8agOX7Qd6?K?3`q?M6}DBb3oHZ6vJEYXAq8N5qt$L%QOkpd>G-iiC5Tx_wT>M@3Oi0{_;%tY=x;kTf)4l~FKM;0RME{i3PC0aa5oAdJ267&kib$_F;of7c-(QYI>IP`W=K5C^gvm3-E~!27q?7E84NUVf|k*7JDo6!uo58KI&D*3n^=rVVznTQ9Lx*^bA-X$P!Dzj<*$s$j}mA!sFk+k!~uE{VqId~V3`=YN0IYp*ov$q#uOVEyGcqsj?vM$4W3e09GKn5T4;4_NH)x35TgU32@Tl}KnP_vI~~`Lp@G56P#<9k;RVKb+h!bDk!!ZqRUG*a1de07vEx9*J7&vln^2qWuw$xo<N!<<L1M&-Vc>JrL0L#ND=^}g*H+hN?6l|tai|K+5kz|@#@g*ftqv_v2xyf*ZLBui1nW2s1d`u^9`&F(oybP5s`L&Z(=ce%5<7<J0=!_-Vi&Xv?J!Wr4uKi_0Sa2+B|DBfVd#QM{V4Q&=rM&uu*@KU+AVOg>6jo8Sc!<6xL~jBYC)!F+%bGmBXnKp6dUsygl23-3^<_2)Y$~|a-e5H5^$Z}!D5Ll=(#vH+fY;rhnRtvbXpx?hr|hC`M8j<KxEK1j^U^n%nBjxqR_`E3@3<MeiDO+lD3(E`@j-P>xK;m?Lc?Cgxf+7y+F%GJ2s(zB1O_V=$#EH5E+5#gjNSs#VYE!)M@CGmdf+cW=2FKwH&_{V#QJWHbgwusi)NGI)c&xNd_$P0OA}3g8+8}Xj5h0<G?VXf+8b?ydm}LqrVUhkP(;&bws9RV16+hpr`{@4X{$f5NpW}tQN$BB9nc`2a89TXK=C$aSLo=TF|aoBd)582|AK!fKq_wYTLdSwn%nI?bv|;Qpautpt4~=3@1)tcF^sh6*~!52_UwqjK~UnA7B+2jqcqufdJ#E?E||eAg%8!RDvoe2?sR=yay(RG)}<uF5qftVl}BHp)hO+<Ou6KF|d?va6a^mWrt2|Dv}lu7Tt`|7<6m|Y!kGB!aLw*2s@~GwN3*n6Y9(cqPA?nWRi`YB*0P*U3GPY7Jy&d1T;*5X23iF6m!_}tT=#{>nQz9AT6P*puQ49fLM)C+Zdz=GPsGa%6VV}AF>ym6p%deB0I5wJpDG1a-hn(7SOx}#RkCdLV^LSKy61I1N;uH-Bwmd3!{ZL0X~7&#8L`?*aNarNQ+^L6)Ht72+lU7k>fk<1dGb`p`E}^tr&<y)g3_4b`pUskg%590RX{L3cN7H1jH&{Lhm7OpbI>3KJ-EW=$WANW&+yUmaU>n2MRDW>^AgQ2Qn%O0T&|}H&{_%M1?<-4)7Ong=a&8wPK)JP~VGP+lKJ>Lv@ZoyQG!0fPsK211bp+21N+96@@ASG$3o-zz)D=fc!?s4zYG)bfeRP6jkIWxQ)y<U`QX9lmjW>4ly-u3wjvyq-q2^@FN(T#E#ID=(X##I}qEp(Xv8yHUZ=S2EgqeRw*PhkPl(o5KBFz^`-O^__XD~79kX5c|NdXlEie$r2!OVE0VTl06;;6xfWIj@KPtTEK)S2OyUC12Y~Q^2hb=0Ns$o+Hl%plzzZf8l!`9GHYO=BJ!tqAYK%L8;53UJ!)vu{RepMqT(G;0B!vDzZGHqW42HK&n30gwN<Uiwvj(v+)L$FD#uP<f2tI_kHzTEFkhCKkz%C{ZHDRdGf+7nT(9CRGfhy}@1A9gbodW6GNnm(3Mh3`?B^s-`1H-j|=^<;|7>XajGIj&d31*Gifi72c2O5k(6C0K|X3F<qIXh$nSfB#{r#eRl01|{Y0J@X3EFa{BpaaOj+VEqA=V23KS-2k5r{@5Ic8ri#x^IN|+e%bi7XWpE#UNx12TKPW2RQ=SXN35iC>jB}*$N>xU?0Os@k78H2UaH(39wD1Vul1L+h|2*<ip;K2~q+olaY-7cWo8V`z^q9({Dj8SR`(}#7KP4@}PNQ{9>!95{3W-C!}}F0TAi5?GW}m^j-^`7b<mH3FZW(_d#apO&4rM1|(R7Sx{MV0faQDkw97lT|l!2kqHHXSqDFwv9EG65D>VMY<GY)(6VDW9n-;-CKm9j!XdU#Mw1u7%z~n$b<_y}#M)S)Hf$S3Cc8il;2U5*h-Rp7$FN;=DegpW3*A@hWVi*L??D&0p#ws%1KtK)0@MXER`n6ozXMwe0vNcSj7OkrpiIMT+0Zo#*TK?{Ojwx^f<*nmBiR8C02j9mM`;~{ae?>%fMA`)ZCXZ7+YU?^4H49XGVfsD1Pjq}A(}Bywr2re7+~Vi21HOt1bGB83-t!;8JrBu9d@$Oh6J~uN!5ND5GjEf0PWl0k{ASlO&qm>G1`!xk*i_`0#HDcR2gKR57Jsr3#KM)b+4tAgis?b<HCGEMX*pj2o#fiBWA~q)sob<!4kphK+;2fbigA4(7y{c-%)wK&ENuZA!=bZVFM(v;z`0|6~hu%_|vq&GFW5;x?ISE$O8^QBVcdXk*aOj0hA&87JYF1PA3QhpbyBkb{qkQs&)s!E^tLeX1NoD;HC~n)`Hr_A_l%tk-rO;g_+^CI+#)$V3lMYtOhWS>4vHw3<Si?Lu+B=0l79k0&q|~0NEXN9l?MDsvyY)1nT-7fCa!ZA9xK^3l$A(IpmQ59025jhK)MV1VBVaf*Ep?SVh!s*!E&zy%;!>gpm!?J2pD#qJ#feI0PaarYUeuj1`9&^xD2@#%)MXpNvISdP5z88=_WZJH$po0FX?q3M2}BsPq#^8a+24y`YAn<uDwWZ_wcoE|}j)waUTbK=V)^A^VrCw$TQ%i7W^g*tUv&WOQJ>14#R1z_|d{u*yAZ8k)OfsazKtSZ5?>00i2$>%-FY$oy%yJAPuSUQM%;w1Bu#QOEJHDlws9hi+a6^FjG290LB~sq_sp%wh0=q%8xM3JkkgZ5@goa>E2X16pueM$l=8FwH<$ENAFZRfgDZ;wI#O24pqtpkuc^pe(y(K~P7k-v-Dg@Bt#DBtc2QWT?1^gs10{@uAwAj*Asy+K}tuN5FY<S>Zc64Y{M}O%rI#C36wj4t5+AL<nvL5p356_%&4>7eZ@6y}OWg7M28zDibE8(I#`*Q;}8MhNOT}g1zm8ehX3_)(Jpe3m807b%*I1<nM&_4yg#<0~E%n62MlI)|85u0O+9d6Ij9~1`QJ|fXXo}VCpEbqJrM^!w$F!dL{-=iUHI@SQp?;ShGf=Vg}O)S^@b#z;)0e&@i2V<{tnQazvH$u*<O00&+z;c0@jH2uNs92<F%h3*Kzsz`BMqg8B!DhxUlEexVW}#vHXICJc5Z%to@fJJ5zuudV?j0hSO5tWpVh8<U96V(baDfJv@X0*aW+*jDipkP;wQjG-sM0v0)7A;fKR!?!{e2>_3SjcgOr0mKH7gC>Adg7SiJ@qJYzfGjYV$=HXa4;<KjzS&BG7_bNeRQ0;M;5Ih^CxfJhPu|xyZL~xlfm~P7+Au{yyACNU8|(-%g5?1r=9*+#w?cIfrGrJ-0#q{`U>iFCa1Gox<fj>fM<P`tKu2P@=vqXBfGG<#3tk1Af)y1G@ggt;U=r+O!b=zzj1QC~X9BDiQ^htwNYKxK-XZKSh`CnFja=Xopzt_S?GDp(J+l+F*pKT#Ai=H!z60L|5I=5V#{r@kN-MNUZ$bqFJ)o-q%1~f!h3B#UeVA{i3Hjg<S~MaUYLFmM@v))WaRCwC0jZ(O9ooK%e3(EHc-jHex~er{*-)|p5Uo!h2Z&Kfdq8KPW-IQfI}U9J7$N{F1nhM@;APq;@!AldWY~0U6)!~zEH)2n&2?;mNjvm`u0vu56TnQ7-dKh(NQ@Ac3)Tc^MQ|0!*B}>sV5>OUu|p_M7@FW?AnVX=Lrxn}D}b@9%H+t!icL&d5tic+yuk{9SfbVD0k){t8M$?0VmWA#HjH?nP8SwsCvF3Uc!gadyX7TVjW%#{>~(-AO>)S#pio<}>IVxWmmoH{-m#(UY^WB`3$e0cCMWH7VP~WR@Bwc2Z6IsNc<5C!&kTTd(^2jB5d7enNdj#O2?lU!dQf`?m=-gVsC#?>K9-quFd=9i7&eGvXnVZiwAvP|U{xkVqQz~G5D&&41W+Il@@*J^0GsM64mvOefB<2(+D;fao&^BqwM?iJ@SUnRedw1q5Lg>B5Eu+d18)Uh$HioU#tP+Q=s#FaK(tVw9&mr#Y?CO%ifxlFS4;XP(3Ne%%y10I4THR!kmoLhGhmO(^U$d-xlTNfJtqbX4it|d9{nJ8RBP8KF>AErutQh`SRQ63tciBi!lG06<6uXiDew(#6vE&nH48(+aAC2zs_)mM{k#zCHD>Qthy?-749N?z1O!{$FDC!73ELis3w8(Uwb>O%;KXdHG6Z4)@~`8;7zUmLNQOxb><Tn!18b`|1hS7-C$t}IF2GO_H~`gPd;0AtRQD#Z(7ez_56JFlb;z{9oF&A@z8PE-1&{d8MMI5gDFz6nSHNxZZ?qCU)9I`PfYU4B<7g#}J!=K*VnEjs5yNkMeVog;#yRr(xc&RL{;I!zu#%;Sl&PEYOXRrJ_qJ|w5?F2hZl6*py=4mdtFyqrT)cWwM#^+G4#=RgU!=L*vkROlQbk0U{7rh*X?`E2xa>K*XSI-7?(*OBpc_g&PnQFJCGR85E*Vtr8_kp3Yc<!Z=g~A-Y)N{FYw6}_PA2~|+0buAI{Igz!)L{-Et?A$@14v~N0cjLKp&L2X^JtYN8+NNsV0lxJ-D~EE1u9x`Bx?T|Hi$YLI3*){k`qAt-<%!ILiYaS}gvMreG{Tm$xeK7vko*$j<HCo9mmD)>m9Yd%rL559;2$IiG|l<4JQiPcK`Z-3+4BiN4say|uwU*pIUmKQ+5_Jw?M3f9x(@{m^K2>CT3R*QMJf8XfVQ*%3e5X<1p=mcKjn+|e6tgWtYbMvsmD{^p(S#U%OSWP{$FH7Eng@OYf3t3A$>JL*)~&ArKtq8GJCLto1VzL2Vb^N6Hr<khSJ9PG-uXd6l*L)7=<b4q+jv4e}*lGC#nbQRZV#1|6Hv=B{<CdcQLk1;*>p@cxr;!}S-ABI{qIYl>Sl;|6O7xQvbl<%x@Oi_Rp({y#Ncuvx*4Yr2khwGCO^>Q+;DU7+896iwn13HW}7}PMVq~4vKpV1vm&Fpb~m40ZROygu$;KQ^6x{!Dj4>)%*)+^W3z=wuEpHFA?>f7}2d#|Va%OdNeUNj8nHBrOTko`!clA*5Y=y}%7G{TwJar1l}&qDt!u0@BU=uv%XTKNHne>UPXXng8wI$8{{E?%6qWHsh@XPIuJWz^~RF^)(TnUgu*wjAy+q)V?fef3rGyqQ1;)~3yQ{J4CN(7DL|B#JQw>VHj7rg1YFjw3q$J{8j8yAThD|I331vj-0zTzmJuqZ{9?eYgAILGzR0SzT|G=q1~#FJ$PSosZ@Vr!JUp<8TSzR;9(l?8eh^Z<^>ggC_%YL@xZl&aMrpvjiK+=TGtXV$i()UD08EvC1^9%%O5c&^cdC7B^>D=;29Cw%T|kQdd>q;P$AVUA_h`m1$m^PkcI5AC4yHl#pVsA5}&F-v+e<pDy^CY4ej<o6!YsLaJmELd#OjjVNKu6LAbs93PV^uTfPuCk&?y{lin#O-Gwfs4mKRs?l{4nz-JytI-u_G*4%fF<Lo2pUpGjrVZ!K$`vVccrpE!EJ_(0Ig-~fJytI*;T!Yn`NgMs>d9iUVZ@h15{HO{fIRndf^TxtNIDTzQFp!aJ}31yG}%3_7BzmS2mQ4C(8%d&FXxJ>t}@z#2hojzJe!CLufFx}^t79?UHZmh_w-aGbBCn9QK(y-c*2+Z4Sk^B$lkyUB(mz=qwHp}!h5DYpqTN+^`O^}D(5quOR-!JQTC`uCmZGS#>4nYZ{(i_k*`sqoDz>}mH`SpZCyYl%<3X#N6oAk*U>^bc^E>kzRyGS;w{83ox({6NMve&I1TPXSq`6_FS&nlamwV`wX^fe%NFtYEmphaR)H~AX7d%foKplMm=|@?4Xq8mh;;7n;lJ&Jz5DwWcb<O2H^vq&FivzrYy+$_$SQv{;HhSzN7`WC<5wy##VuA%$HXfO=PRo@e_VB>LcCPCFPYQ)QH}XylKJ$2^G=brsHJzn_Vn|4vf}CW@6aq&p#af>ye6(XsbsMel{#d){PN)Ov}dfIly97_y3TLuRgcrsZLHwM!ljs=QCDxwgqDA^nqRy^l18E2AsJmEwfF+b^MsT0{`vPuO7fzjD-wScPS{OvTyZUK#f6dM>AO8Xl~Y^?d~c+<vLe4Rvb<P+u(><f-~2&e+#va;iz8_))4!;>rlpMk-L$iBaXY5$@$BS$J{&cp;q2kyd<F@?Dc|bNWE#isi33KP+k=hG{r3mA_xgqYzhz2nm<}cThbr4&yWKzdXs~~<v-_r@Vzr!+4(E}rCea)3GoI!bqsmc2gb-3_4rc@Tk-4v4o&!pQfO%NfBVKo@p5j#sGl2vF_ee+rS}Y6FpyH;^^5v0NuIbY4k;?p4j7)T_Jhb(-^}GFL1I*JR^yJ;4a&$7+e)+m^Uc_nBKNz0f&fBH6{7HA|)uf*QUC-v9Q1-d>LdvRzu)eomMYZ8LnE-GMSs)Ba9DLHGXChxyy1<7-jc7QP0(?jbg*mHTdetsr4sqWM>kmpBSdAyBInEeUQ*guM?&GY+`chEI?=D$~mHv_ZfNJguh(k5mzK|fdclPeB9q>%bWSLiGr_V1}Isj@`Y<-}a;iEXsKS0R{G>{A}Ycgf8Ca%|5)pYH~TP$@Q_7f_;xxW8FmM&Xn_{h!y%j~^8E9{m?JH)A|t{8;q%4`6$;*3tz@&$&``RUoLc9>6ZHCG&VE#v5@A^30L&%)tQ=2E*@CO-stGnO(jiCLcC`ciR!uYd1We{Y{p)k<rSZXGBVJlNTx>+wEHOINWxk12ZG5F{vG8zK78z4D%V@gCENZ(4YFNr&%m@2+v4`sErbD<ZmkwW5kB!_k}3-QXz(*EQ_uB3p?|O&wZRRBZu>@q^-J65r{S2YrP;oS&ZcxQ;i&%D2BKJC~77hB~4@EE{Lju(5c9b~KxJr7EVJ7}W7e@S_?P#!r;<<rrXD#X4_=N9ac4B5isLv?<TBoZMR<9rX9#q?!viLdTUiJug*w-{7|70v~|q%ZbIg=~YhpNIw_Sfmhcuj$^(VK&E*8vo3X%_iA0quA&{~;z#uHDs=XZG=rE)y{pV9byigwBG2;u?fv_^bn7tR#*-zM&N6OOuMWGmdjyfVdNnkrboq%+)rotQyIo$!_>-<<eVI7IlIblb*UlUy!n3VsNfXL<re(JBm6~D}A9I;*k*o7zE4Lrfup5Px++Aj$h4l1SHkdnWx`h4o982+?N9t|k=`)G3m~F4BJH1`0`&8Vn)Ya2$-?<bO!nL6j_(E%R<2qG)aq%|LLtb$p`HQEhe5?B~lJ~eTU*5&rqz~C`ag<iNcy)rZM7YKDewD)YE#1PmVh`VDWa%E#(yZ?3&I<i$e3MzcV!osX%iiMds`u2(YL+fskqBPCR=k<(ie|F~O}K<5OzDcIENdEN=a4c%?y2e*^=|>AzF88<TP?3#yQLYrf*6I_TQ&k|yWA*e|C8`e{#<#_iOj9h-=^C}aw@QZI$D~E{cUc=puLq|<$lZR+fv+6$eQiE40d+=+brJ4T;Cv}9D0H!=zAsDC2TnehK-u6TY4OeAoGgtxW#+QGD|tTb4mzxw6bDy>bi>R=;kSPn_PP;`*Wc$7ag-S5u6YnG{$d7uQDOZ;vuCsc{KTg7I~*JFQ7N5%&Vc|xOascnEOIQJCt9J<P=@$n;-XjMwjjnOPKl@sWP&1rs7nUEZBSUmLM?4sV`F->(w`lssB^c7bf}@;TfN20T;!DBuej|jxW+)t<v3y#iS=MF27pVk+;4p?%U=IsRH>{C3(%l8~kILVg9BI;w-27f;jb}xJ2(uJfKz2I41Q+oT3xNU1*Rfp5I63p}8gV{nq6Db3|E+A~K1lUih?xnEzoS@-2HzvJ_A$`<3V8EVI{JUbj~kdnW!5i{amj+<W<|!1?r3SKK91B^(PvZ~^4-#p>l)l7+|(Np4!31P+v_5=s5y-llk{>}yi)&gm?UYLydP?F36Jla1GVJ!+cozB@SR-(xgTUkEb!f2n)d?lz7rLGZ6+sivJsF$t4;_yIcY1Cfx#F&{D?memjzoy3Dg2?S^Wlt_`R<40AwYO1<DHQjr5&g@Kg*Up|j`(@6`vSrJ%EuXXd2gturzhv(tA~GT}G6Bk}ni+SO2p}UPZrr$W<HmiOa@dgi03qEc<(`i}k->?Q)1U@GnBVMv(D+-2qoGb_<2o0DtBZV>D4%h;oS*E3V&lVHiE-HClv##&A?*aILN+>i=)!yOc651cK|IABY^EO_W%i3Xk9qBe(UC};k5^bCqT+C`+Zh$^UMQ9BUI?m}YD_FcyZj-Y)OvyMyBGAkBfJ}6`8_J725m4aAg3^lvFfW%`69s_7Ll%OaPN^S+HBfYlUmMr`EVq)ugCfl>$7&d?bM9|aZv=_mV^{pkDbD$16u1Qlzh5{`(bTm-HSYZ=b}$p$*25QAHh-j9Xbgfi?5j9>N$RWE}s;SX2RJ^Cn9gwSErAT;h*Pczs5vY$1(>Qc@jQ7`wevihw<m+3Hh<;mw&(%TgRLyOCRV*k=pD>%478dB?o)N=eqTfdVP9K3BEAvm3sX-c`Z^lGXmuCG375NugBxFPbH5${Q>9l#>wf|)E<U9hUPo~J(|6G0zG_s_FIF{(Nho5g)ad_Uvm<(v)>(`K7&p<W7*TQkB&LD7{Vfj5(x|CK@*;%kFXEUyT<;Sh=;G$PvmHcpni2sz2F)c?Y<=L{uC3lk$2~Bj?ey{vYkES(e<{^$Z_-$C10atX2+Dg>=F8g!Yn1c-&(W1jI#wCi<1OV2mKt2gl^yQ@{2ELf55L#&;j)H8DkedJwnR(p76?^vOmBCDc$sE#{fA?Y+peiaBLw=&xKzp5fICEIJ@rzJv=*x9=`{GK4pr;#Rqi5ACa1lQT#3bi}dytLH?Mq`D^<812+E=-$Rmlh8@B3e1fE5P5uf_{tHTb{WZ?(n85y6lKWS*K$M~l7yMIb`Z2)pmG~mk@!?aN|JO!W2~i?aiQr~V^v(2)5dMP&(p3OSK*0{ipEK!x!c)^{_7!hB0Ody%26zMh$aImkC#p>6kUJLC_B&iZg6@rz!aF5^2or__k`3SI02k#k`89t}I})~qU=;+$N3{89HJ`yazsFsYeh$DthWUI0bi}&`fch0?+#}AyFj9`dPjq#t%r9SGSY0kJy7^nR;6I4BTUc+wWNl)r1Lh^`vzPz?Xt73v7rw5uLIzf#3%XP-K>#WEt=+wD|By9{{@G{fAx?Ys2e~-nHO8`FlgApU=$rjJ{Ri3(!LjP=`3v`;amBMgNbqguY>jit(L#RwOz_t4`7qG<zU&;cR(_0YOt{(FA7XxWTA%N6U&s+sEZ?y{={C8?h}fzBj@hK_<a^vX#%Us-Jh!tz%F_|^Gd>kt>8U^<Ip@pMBlRZDwbQ%b8jb`Vq>oJ8_CL}-6!8QSV~>eje9iY#=!aaKyGTnoG@W1=Z9vVhoQQo)%q2aotLf*k(;14#Op|gF5*8kG52TAaCRXh<csB_OSF3?X3Bdg+pS9wnM5S7EH3_WrKz0vNSpd1@4?1UHW=LxWr>5(PNL~7j`J981>WuceI`kePBTKhV$P53);~)M$PLG>>bWZN{sUOr*as#j-xEPQPYB=gQ?%}+@mb?-UOTj;mlONwa(fKeh<P&1@PcSY-Ez~8JW9SIB8Mk~!2$$oljQ?Do2~hZ0PUbRm5M9K<ZVdq^IHotA@Ur1smyEZp##!Q!AXfR-kLqZ8ZqCi<&X`BsXfWTPWBk*z&tSox!ZGq&W9lrT{Qzyg770{;mwryG2OKN)ga$;TX=l`7s(-06n@coHNr^65Exr7i@?VM?FEcf^;_hgNW|S$BIYlr_Ow+%9mL+xBLSR~q(<L`*oSZFugvv_k2=PxBO0fr^7<nx6$KT<MpHQ3?GK?>2;9<YfiHX;0f~I25S568<c^?t`q8-7VGyB7pdCZ5#?{EWXj4>ck6TW5$qnenCJt*ErEblo<9qDzwO6v;O@Ihw;SKPQ;j|Z(-a_V!)b5`>=PSP?WA4C1PNqb@<0g%ri5c0mRLT`VpS<;W$T76U}y)Hk}=%R>h=RQ|bHN{4U;HYHGDy-hqno~%YC)vl8UOUamuMmn8yYX<3t99{FoZJs+oxT#OpIHi!yu}v5m^w%%7o?taPX%rE#TjXxIb$vq>67w{)Df?re&R?5>e=q!Zw-26IqJ%h&8v=fyD{!Qj;oG#+4vRF&?iFZ!wk;#t+*Ta8d3(3Nyd}AiuS`-#ER&m@C8klZ#;(ZB`-P}0I-7EeWCKf{L<z^Wud;fGQYZ9iG1i?q2rAcVo4~25&`1Adn#m;ukhabdkV)SG4vDSaL=^!Unm(!Dl`QAbQ@cu(cUJ8OWa5U3+YxKx6CKMlcAyNAiZR^k|3}gVRrDGBnTw8-^epqoA)zGV{hohk)Uv`f^oI=mDQEyN_l;At+HHM*@%3Rke<@W15G3QlE(l_rdq$Ri@i+jTj_V6+9}!3K4+?aDuPhuBEv@FDLf@sE`E4MC&|Y|5b|^-PUUAsMo>gwr+rvgUN;4{xJWHrNakhOiX&UUH|mVKEsAm3-zp6JyLX3EK*64m4(kPS*(^5VVXNKmHVZ)@D<RMA_sGkEdej;?{qt!E^gT=BLY#;XsbL^hh3xAJP3l@3KQ#hPRGKyIJG9+bU8^g?O)@%+BD`;t={kYh+Hr1p%WG?^x9c0_53pQ2G~S%&av^K0AKV!~nBQx*yQH$0ODCOhOPQ{-dbrL1AlDDyjhlt*Yz}7p)yvDPo2exg6VoSk_FFpG;B+CmOA2c#=5cKDhHu0T^kW|=-v!D%4zB@ZLGp7?nxZ&Z{W%A`lLU`{y@E@Vd-gfjih#DirGthf3GeOMtAX$BwqkTM9D3{JrOJ9pi}#>ExYrsCi{~IiSmhrxz&h5Rd`-6*z9Gw}c=&zE5|^~BIiG&R#{t}hlILno)9WQrtE)d~iuLzbmo}H#jlQycYiae4V-Nq_D|1~Vv}Rxx(MH_$SWuPNLDzQ*Yuq38cR_H57TrT{r$5~5z`2e-j)RzQfq>p(+JMpQ?}j3z)5Rr(3sz|iA<CdoleDpK-*;Ly6xUhbytcl9L%v_BFPAswZ#p^o#_9%zdo0)AFE4FY)}tf5&U{SzndhWq5fw0=^F;@qs1GtL0<|S&(NlOf>a*Ab=#n>$-(k_i<jbS8&-EuybcH1Cku{7y0hlT)>y!YXK%SlVJI(!=t8I|4r%--0@nR$HHKX`I76#`^&PnTL>i{F4c10^#Rjf1wf^TNTb*tOi?qJ2JVYkz071XiT?ChreL(%ZbTvQwiDNZVuhYF6OOHk#;>e`+9`mM@5yM4im&95#km2a)vECGq>9`@<NjhlX`Il{>ir|238U#j<N`nf63RSB|93Z-1ndoHU>N2PLI;~of11!3Kvy5}TR#soDkFQw*HuY7Q8X>opWLv)I6hS%23jjR<nVpnR1NJ3d3{FzIo(p6laQfX!e|A+MVpKMl+oJnKmNVFbi=ujsNQ0}ks&HtN^hSNuWP@SnIF*0h8&zD!Yz|sfI?+qvINr3IfTYdC-KK+x_r~#v=ydsY*?e`;5%Eq`A(T`HQ_=xUSbnU>Gz9pc6$xEfa4fXBvjmpFTUlstXOTH7WQbR(+lLBA(RG_*=!lU#lHb6Hk<)w|A=Rsj&zeb31&r`0OWZ&*&+83uy^z;#&!lE9Mp%3Q{zfWWqFHe~+^?0X`g3wD;r?k*66UW&fJp1a)13sf4p3~j_*K$a_ZT%J{rL#}G*{hy~ayLcjN2#?2e6(6;YI;SGl~{b!wfT-D`!4yHF3nugId{o_(?3CY(l8LOSO>0Pk|%p;SdCRNGk+sK5G4#l4@n-wsn9nn<Z~Y;OiZ#D4+i}Q#A7`I+}qxdsfe1>AHXc3Mm2dO%>e_8V7dyJilXL0Sgje^D}*QY{N~2$_3QO(K=aWNmoy}YAS?rn|2sTnACAMrtO}`efM9xOn*3j{SIQf3kE!3<yat%#I)F>%8@})caaAJK>*cP2tRnrrC<Y0qO40ghkdwjZtoPtEz31+Z0hPa^SNMvYke<mxH4ur4)Xd6joS_I7whb#cqIdxOLHOysSJs;YeJoH$D|%<WvILjH$^w2?PuQ~uf4XrM&ZP2<mDP1D!c8wgtK_*rLS<?3#^SZbrNxaq_-k_b$%e@+mddbNM8q)D^7`sZME+JFs!`}imN4=m9^&_v7;*Q4fA_)^|E8jQ=*JNLg+IX*eX}~D((Hwl^m3o>)k77EOGWKP8WOmjA7DXAAtc-n{lyi^2`0Uu=26^U$w;mMAox@Fle`h|HAJv3_`wu@aZPvqy(L-pkTDM5D<4#V&~%K>0rr3!z%zFZXLIJm54-ijey;$3LyU9W9Y%CyUsT$P$oE5xE5g4+%yfbu_Snx>#3EW)iwE1oh~gKyN7OdJww@c!L>P%*-0cI-`n^uWR#CW}jYaYrtVhn(<5~nQjuOmas7u5xitW~@u>-(!Fs%TQ-OEIaX4@6n@4%xo@<PhFK80CPxC(yafQ4=P1^#GWp@Px2y7Aty)vWJ|;;IE33)7>S;38I}!>(XKWC(_p1_2%wDLT7_%AcLqfe6KMy2r~45#VPMKiGx?@x!<cXVr(@L?KmprxO&mclPSX+nv$Q{#G4sVfCOIPd}WQezRUfWmR>pvQS3GQ3G=M(3Oa#Nhap#r=ga(O=WFubqx_L=fbxd=}_Cyy<K<V_!d14=oyCA&fsYZP{OZbuGP|P^7FYD0wbBwd2)pDz6b??S|w=iC$J=Wj*x_F=c11jH}y)cAjy_d9L$l2Fg8fg<@r-u0P@kj66amI7ZODLqI4Fe8`IAs!qAB;2szuc-U@=WG^7;sQ@__bEW|eRXf&djoo`byZvitBH$UZoM!Ijxk3Hy2c$fCK(1X0LH!^=o+9LbQ0vs-jfCcm_#VhlR06&K@h=xaI)_oc2BYjP--pjxCs!ML8VfEbK?Tn%$EfiEZ$fkovrPuTMB1=*GZ2Yj-8g%+iXY{EET*X{_0036`mY+qp{7uEFhURjKpD|erh|}upP6rbK9jfn?Cy>fT8^4UOgWdgJz1zCq>Wa_J4qo~jBmSy#4T`5iSt38yEvN+0;+yPNfgIC_A_pR}xLx6=gMR5a>87EXF~6BO(ti@8V*OM`Y+VEtoH%z#wLz5A5VQ_-(vY$Xl18S4a3774A9f3Xk9dgr|5C~Ixc~ND5Q^`B6uG`pSgYJxD$iF^e>dQEP@on^)ZIzZq7ARS?!8}Lo4;9J3*hpHrVq?d6?%=GL1Ff_nLvDndBZ<N*j4bt!YYXHl|UEjtXx&wgT1=CeIZ5Ec{S2Icso*eXWpWe6sQsRps5?Je6hw<n|$fNamBBZ${6mAGScc!eR*?fV^J76n9o$VxtU7`Dq3W&33_|=kk+NEct{znoH@5J`G>)9=YUJmOZl8CZ9an6rAh~zAn_c%Q1Cfz{}U~+yrKA;5rSanOau-S0h!IO;+@+dr?_||WO$n+1t&I5!Etqg_^j1ZKLq%$U?DORLjWFvxx+!^D|(NM{*)Y_CI>=14!gLrQMplB3oOMYafmtys5E{nf(r`l>DIwsU1WMIFxPMN<8EsRWZH;_xL$Cxk)s&A2JbZbEHUH+XPQnWd@MB5q_UBoE5iTbQ*=uas4==XPSTCF%}Oc*56FmZyu94Kz}H4^%*qg{SSy2H%oeG*sWH70z2){whSwup?Gy6>{*)SyG?WF{!yAh$_MGmS!_YXrx^{GQTElT29vu!&#EVEYW81^n74P=V$_l;dU8`)|uHYB9P5}U1Sx7w;nuU}cxq;p0&G7gbr0(NmX5Kiw_Ry=WEYL&-tp)~y!UjoC2H+~F8Si#@&LIG#{Td(8ulB3g*H)K3%2y|X)C=#x9o+l3tBWg~LJ|k`w!GE+DMVW!wU01^aHr`|;S;(56lm7pLT^-Z3UdMki?CLq#&1Pg80P-0t)7<OvDHMIe=>sSt^ZEo94B`#F0Okks~g_R=F*b=@{P6C&0F5JJKk(~DZK36EWf|Fa>J%h!gug?bfglmEv_t}Vr_K|PQR7K`Ndo1rNxcKim6EB5mwh0Dr-1`#R06XYi?NPIL=d9S8G)9#>Z*JWVDj7G1d|#har+nJ%`HZ?&oA5MwMt?yBJ_VH0wL#2`ys!&~CFB8k0=xQCTQ_t&^pHFJvJwvW;}c{0J_mdr1FY$*^N_Z>gqn@34^A($tNlFxL8*wpV7_BZC~{wq_%<>|t!8%dr}!d)QoAhdsQZ%#Gz6H`XdQNF$u7c5KI(=^1AAfQ93iCUwmrMx)e4)0sa{2xPc!B$T0A@})^+i9S>4ayj>gefDEEmpzb$WI7&QK0{uzc-<+cHEp>J#d`nGzx_3f^omabBI+Cp-3f{h+nsK=a485*Kun?BVqw$F_dtZ53E>ofq&(#ONkrhL<xCi9)Eb~vL!9#`iJUOC=$-L0r|*;!1t0|h$p18QVoL8FKbSy9MA14ez^PL^9Z9c=;76teexeKD_}M<x!y|dM<+kwch{+;-N>3<(IHU?&Cdr)W1k%Zblbt|-5faP`i0MGdS-_oJIFH`(j*`uGBBdqqHRpH-VxluFB!f)SMdmTh^&gUB8W*0=MZP(cu%X;C-52TYZhtH8dMP`V*~)3gsxu>1b~Hb=H+VXf#*eSL_W5GIfix$~HJW&nsLhqXXl<%cRlcF5a8P4NB}0l`vZkk+&w`VDm4mpq&1of>``Yy;>Wuv4h!7(L<lEY5vkc*6fqVU7<R{{%Kc)KEjrR%wQpkgIb{-mjY8G!}xxEhecXu&$k#=ppy|{7HyN|?KI0Br*bwn_!BME+*CuH8v?9DgkjsScoPxwx*XNo=OcQJb~=e?{~A7RE)_r+}TiM-q^r{Ejb^ndzP^Dd!(#YNk2V$fa@y7=?)a^E`wdUvut(<<P+FE|op5lgJY|By?xeQr4#bao3zVhdI~wG-GaaEKMOz4aj*y_2LE!hgf|g9~5>2Y!wqM%ysUQsGEhoM^;+Lb4PowTV!l`VYEVD%~STgOFRsY>2iszMR5RCO$J_HfD_(lCq)aQZAG<>Rk39sp}f&)Sai)YxZ#DA4*Kg7tDu=^<u#E0ZXKDE<n9}qf244j?N-VcOtIBs{1fGJhJY;WbdalQQU+XQ3Ow7{HJuM)fxK$p|r^7ng<=Xt)>u4{3(TDdp-C{YfMXx_TtlGCzYj1=QoP)wIXynlJJB~09G^0*Tv~kU_fUJUL}Im=jD|t@n{v{M^TyzU9Xj8!!Id_ps(Q`<;;Rt0A9RiV+|_|C@UeM%9hjg6dey_lC8Slw-|6b=~!9dl~luQfwd76a4Nc`s-=u(#0E}$jIM1ip!b$c2t=tiMf0~A)8duN2RF-`>*U8xM)1X@4UyE&soId8IJR`2(01PGF>LU9LIA=wFH@UA9yjVAdn4}|d+Fl2H_?=zsMn0s(OZBkZK%rz^ovKSvAuUv9Az&`c*&l%<)*4gNVRkBytVSmyNfF~9JNefgko!VZ**9i07rNNp^qZ03BEV#*Qpd4kaVL@o{BE#mTq*^BxewKv+5Rdg)5-@aC68r{d?hqox><!Cphe8=iTq$%kV9?Jn$x9COQkDyH|;mnl@XF&X5wS0jPqz7&+;O-SecI&(0p%K{Mruv8&S<9DHv%RK%;-*6Sp2`8CvprCudLDdS!8GL>)Vlt;lWVTl&LE(%5O_}SMQ`%+;^9L~aGiyxG{h5g38g=?~$09P2}vZEdi?A^UUTDZFxfOx&FmiOU)A1K99@r$xq=H6OD_Rwv@Pp;)E2p*E?Q&Kt+lxu}FDvTV6H23O{@6Am!XbFOGNDfw$&uDAO>+&^p*l4#~0~kB^#f$Pz5I=A=qnQensat0Dd~DG?V~G@-R10P>%3-(KtZ^ii)Syq!Hz}*%yg00UHz%QC1^3L}4ALn$;^rDgi`%2ZY%s_BZZ{s@6HRP%6uYg0`;O!+*o3{|pphK;m1Kmu;Z8hgMeS}s9u<<NMRXSrUS-KTY0W<e?%B)Ui)f?BO7UxWL|fc6pOpe!-7axZ;jO2}W1uKe`T+e5@z3N8R_Dxq5|O*PviRO+g_nDw{%++C$spAC+v=Jzzq)cA8UR(mRTyp-N<N|pS@jq?sH|)(Zrovp>CB!zzVjTCV(MpN>Gm7dUrOABRbTBY@{BSG30<Mc5K*g91&Kl?{jekb`RVZ$dZQ|bDZ*jfiwi;Ksilu57Iw^`m4g;71cU)^?(cO;G;0>dAf^^D8M6sc5QB8sU)|iewdt&3;*e+MB_~@7CD=eb31Lr+)oBXpTniciv|!L4`s!zk>`JQU-pbcEu*3}4ci^n<Il-F_>Ce>i4$zSgj@#xIGF_^&Hi#RJI&@e!#VGIM?{;S{*NMkoHr`c*UFP7OI*R@|FXH3$o0fAtVtIxtvpM|E$^6Gpt@|S->TdHRgL~P&IHauvIY*f#Xe%b)QoClWHG0s3UfnvqFk?rJ+tRR=n*+EeC|XYy)l(MgPIDMZu(@>Q)?q>GJxIgc+wXM%xw&2HU#j|-x1&q7A{m;irP-O8S_T5;hpK{2fm8%Wo!tMi&!Fape5VyR;U?iU-{MlrzJs?T07cY66~krE8fIWNO(4_SOyGzSl6N@#xB=Ja4~VP4Hr}M#$QW#Nt-M~rQzunSr`<Wg@a|;Y22?bmAne-Fk%onm=EZAKNl93q11H}XgH&zJNG<a3TzO*#J>qBmIY%qH@Ib56e@XHUo&CW}l5eR%MwuP$^pUe8GSXs4&rTl~z0>dgu>gt_Ch}*!OP8;NSG8_Ab2aw%A1GWrc>=f8)<8bND92P*q_fXW_*w^dFYrZyUi&9bQ*-uv=+<&c2f0JmG%ql>68+ShZ9f#BCkn*TBXIor)mwMeg+eGLtlk4}W7YGgSe*?20}hHRW3Rwf;5<<QE^wV-qrO>hhk_+VdbUz&{m#n9%?diqBKd-ZLPD{#P7cK(37CrFvFq;M>D`ZPw^{%qq)JyeE;-8}Cg(c5$pSYQ?SpS7$rf857Brewr@GpG%GC(NNH^}dhHJ)jItyJwlxfmbP$%Ps=i3QkLhobd4lt!WQ0U4WkC>xrQ<PU1O#HHXdp*C`l&Dt3lIW~F?Cs$fu99gq9rf!y5U~-nv#|M!S9J_iX75UUp)$X?TwWTpp3v<JNoi`dh0R+ORqE!RdJWTV!Wa~1VCLRU7(+g1P|a<|U~vW=J|$hiwRdG<W6i9q>C$E_OXt>UGShof&?$Mbw=i>cy4$&jfp!?1K-xH}As2H|YuL5k>bo1utFDfz*!ghO>2^{5fd!oRo2_Q(VTv*m$VWquY|)tgec<iTb!~{+l19JNYhVpiM{$JKeaVU<x&1pR36(C%O9lxj>co=E5DKJ+Ha3IYL8@kmN-ff!7588qaR^S>3wzYm0IJh}<(#m%J8@_JXWLgtl_sGe_;0CQ&CR{7gFw1lB*%dUiW^XLW=)m8)_WTx{!V_5*udGRNdnB|he~EK8=N=s{Pdwi87K53XQ}djWhwit(bqlayrEI=cT#7Lf+S2Pn8k!GSrWPn-il^JpCHtZl+BgO2jXg$RW&qzzz7fP;Z8hsxu7V(vg!-(u`7zUE<fQ2Lcf*d6e7nJ0au3GN`h0ReoL7jQ5C4nCu+)QJYkD6?DB9g$<r38DZ8A5SFe$dhH;HN)iNCBE*I(;y7$qx7~z6pA~Z4*i%v0V*;1waE@d}Mg!Vf6n>p*OV$F%}VYafXxZfMM_k+1)<+5s9ru(NGvB_U&UmkzO-d)mvvKh&YPbNBl&Tg%40skzsD|s^K+UX-tPtL~qL6USf&rcuw<Drq410!RIf`a^@;pv<3pZrxhLgW45vSWxLaLr*03$b7_j(Ak4;)_L$NZZD;z1S(`B*0BIGu27^VO^S?nGI?oH1ET@5{k3dj(r5=;+=D`vK4jM>*R};YW}iUEj569*;F&RZd$f4`A}az?QzDg&=!%GHKcAPTN}q|P1+^>u*M&P7d>9B-W+!|YI2?n8>`1s2dGGKbKW}}5q{1Aik8$%;ltc}5J42mxrA;K3}2ActW-fE27+eG`LXk!6sK-UA}yogc93^_xD-WN(jX&6ot8RC`~4&}7$?bOjqKtIXVrAZ^piO~vJ{im>DgK)eEdq)bYiK!jVTsH9!2ro>CXZem|3G);?<IRl=!#6qZC5PrJ+u45U!btOiA#{!sh(D3)lEfjYc_>M;lAFb)`iYq}xHKHT2-Rqw4)JdlU%{hI1ad40IY;9|{lBIq&Tw0r1@mI!^9hsGVpjmWJN++up2FXEG1-v$K30rE-qQO!i!dR9?wLa$c#kD_HU<mJdu{BlqITrfv_U*^qD~)bL!WEcAzmz0pn!TG#tMbSiR{FUu5RP-=IIFoy8rY+<uZ*G95Sd+!1JasZK2@LTei$)%<yT%R4<|0){?O0_v`VDB@u42VaD0ia5Tfl@6iWso3gNzL`O3R+vDY-5R@lKH*OMNH3YrV~0cwQ(YOS*dD>`}7%vkr^ZzY_Ia{r{1`ty#b#u%Qx6`lxluiYYv|;8S#Wxs!Bcw>qL_{O%TxmbK5(kf~&R~mhMGyg}h$flb5+s-1#M;0~yw*SCY#R%qbrO3`c=8IBY4MyD75<wBaL|IOT2B7gletU@cEkP?xv)%=sO7;>fH;=Pto(!v7*gWC6hJE1NU$#tvRdg759v6%NDK0#1Am2Thd;4`|(WKMX+%hW7<6k6Ie1zi3{uT<O#V*S@)*y3(q)V(;yl#JGx{#jIUTmcEx)fE7be>XXNv=>MT<tA&W}4kJr`54}{Jn>+R`uBOF5>AqlH3332w4pIc<Pif!*6})UZ5>Aq{FO{BdR>~OyipFB_z=jzIrTh}Th%>q=LR-F=&(zY_ES2jvzcV6@oX+@4l_btJ!i-Q#gS!yx@$)-y$LrYt2}M%|&MAxw)koOpL=jK+f2`j}3ve!1VjU7{k3`|u)<L|xhxv&1&bRDq<=dp$L54jcFTwxNff0a2rXnHE;4CPjqbDW50Rl3A$nPEyCq{^7N?Cmhc2pyOX-cLI=VhX=Vn3t&aWbMT?EldE)!69|Tj;2^I}FM1pTvnMC%7*~BX*>pSUS>n0JjUq1A<<lZ1s|faL`!9-ge{P#{VWlCb`L!(PwL1%>TKEA@5{%_eT?!$Qo6P#zZ!{7mv~L0>-kJhOETYrg^vdlfXU^;pGA$ydQ*PGwnrr*9u-z*S&sEUMScfO^o9-lCrDXOBz{oVE+YEh~~wp??s0@<76NRw|BR_mxyGdy|S{oTv=P3mw{Z_W7CpP)TKCzGnDv~mXgm#_4$N7Bip4C=`#~w=H$H$swKEGrOL^>vTvn<b{dG&&FTx+qr8h(@-D1DJ1J2>Ton1&H<y;!oK3%*-O@-X;ja63d5Cu0>B33ngr)jR=g&pPE_r5<Fdd{*G2&k4u3hY^zdsUoP<8^;XIaPBu)Ii-w$2>1Ej}2)-LXExYW3PiN(1NEPnh`zxN3AM2r2O>aD&L-AGN37$W~`@#~bAvm07cYpdm~ciu$bUG&zsdTF}NDfdD=%c}LpBPhNek=q*rQP<(<DLE1XW%=+oA?e{Q)X=O0Lyz@9z{np~G$`VL8CLkq`LvDDas%B*q`#^1hTFnvF6QpvT=S^8%m?oqK;c#MAr*C7bHY$mNas%If8#}GWy&+elcXp2D=uxc%6iUg<Cxw8y`N^ErI7)wpqinTK!p2VSUU2T$y+MB)=yF)l(i?>{6hJQ$y|B0xg#PKjry9gh&pz>}{@jz(@4eHf9_8!*O0R`Q`fzPDs#D+Ygs~YnW%M7;rt(JH;SX~pmJOr4{Eb$^0N3OY636t~{t-Mm!RdU%6mK>-p^61u`irU@Ja)vXO|I}!IoC5;Kbzc2mPned-8mtxr$c*V1w1Bc>FV7+R*40I-Hp>Zi(V3xoF|A%B9s@(w>C%*C4|h9e|vGEf<;THl$g3GY_2GIH+x!0Rg^B1$)a=~2M?EfpX++CogukQjF)W<6OslouT&DON=JAD&N#fkw1f&CU;`Y&k=bvzg==g^`hg4p#c758<Om}0Sl?iHJ)NW9qzl^ioL)^pP8NF|uR6~agMJbM?<A?d@leT@cKC5x=0T-&Eemy&CEFSbP9^|q%Eu!rHO`bT4le5Z3Z)CT6Jd{GxU=wo{^@_@I^9%F^br+8`G(73irO_)#QP(v7X1Sj)k)QS(2UIbx;aeDNFaI>KG?F-tqN)qFQha6Qbzu&qs+{KBBk)a@R~VDE1%5YqzVXk`2iK*;K3aSza^-NhE>1WM;jn|&7o^el4qUat<9;1(s<>%1;e|u>37qd$bb<wTJkePQS1!y8dIAr8hL{F-UC-q+CkdULWcXg5U$0T9pqv`#jW}2XnMB6V(YC#=Kq-KuFJm-i}P4g1|#9k*C>f{JxNR9!W?zRu@wbJTIC>k#lo?;E7mYE|6VJds(}Y>_Is^pCQDGjyA=vR8h{}>U1AZ@@43b=I?u9q<u_D9=$o_Od+Ac6Treo($a2odmESP#-2*W(8#^E#Zs*_uQz?pSGiVc6vHo=EnCWThClOX1!d-*w4QJlp$<L^H5Q;$iM5<$uUGjOxQIM?4R5Y7;nkB9N<|?L%K#?+8&PZFXob%3J=A;K>y+IV*K7q?k^7!P{Bdx)SpMB~&_Cs~MCZlicBrpI)`u1v^ld6nP!UU3&F<}Q39G_+g^mCde7$T$!)C1w-mM)vERm3UHDLC*EtgT;yWzVxAWt$K)_yynxR(?)5Fj~_IUc``yDq_+{jSZ2oB|m-WTIj~1^{e2h)~|esWlE8Jcv61WBp_4MC@cRUI-Jj;kDZii;uoM926#eRFty{Qe9|Dc(u&DDV+c&66_I1CSU$8^=p3vTgw|6dsYDJ?=?YgaED7m>(IG{N7eyYQ!ET!IiAqXt(!oq5fOMi`U(6K*S^fY|L0W^SK<PeV%r6-T%vzdrYb^hg1ZAo-CHHYi*`>6YJrNO-K(D3Wf#Fb|*FpUWl07rEMjvf`qgI4^>OrD*;ltj=X0=(flsll(4+_zG(LmHwYh%!v?7C6#sm7Oye-VWexn4SlgirxQtc&=Oai2K*<LM9X%ZgUT&A1zAV5X|L84c~Q<w@@p+q3yX4|!L*(*s}?joC~K+>P0~D#KxdTFptHBbkA}xq_$S()*;vXm((i-N!%yB5GE`Yd<Q*JQKCZ;{|;PMTn;HKs^Yteq&&yh^*F$mj@YI7gui7VQwn6cfDTT(5!~ZyEiC7!3jG!;Q)N<rN;hQ<j>5ON;9d`cY7Ad8)YHjw5TGOEvh2E@sPF<?3iH6KlZ2KnHBh$+X6S8-XeX|h%`oobfJB9lZI4;Xat*o)zuVclTs11lfmAG&RQKB=u6*=g(zK&x3>#6<wS*ao^BDk?FXSnQc)qD!+Y3yh%G^U?}98pB0(P7$-)X)tq4E~!Y=BVlE<RL0zQrp;4%JSUJIa=<EB8Zv7#bP7S#$i4oM46{s_apBo;owj)@bY?B=>fT!pj{6!&}Zd*dEvd=xc}qM0C3Z4LJ(TT9B3z~DF&W5P!0##^l}^J*oCh3=$p;UrlrLi12d^Z%^NkWN@A)A&;anoK8Gi}VJ`#>4!#Vv3Ydo?Z$8FYDx#TxjB+!i#nQtA^-=`{7I=_Zt3gf@6Y#QU-|XtW5E>qycPPuU9@mivtC8vMyKSJbwPxG6&N6WG(9L6(8&X)fbF88~PNk;4aH0arySpr+qFnY%cPN&zs-3Pu1*cpZ{>+;?^&CC^#Kv5^kPm*fr;Y4KtvKaUinI^$OQaMJwtZD<4ZDO^9<O_fQ&t1|=s=Wqx&Kbs1Je<)v9+8(e)+qlKgtR-y@y*m9x(qu_4%DRXpeU`rSx?p_q6vL9+jwag5Y!q*^B<C93obsB_N7erA?8Li21Qn#ISp;u+Jt&}cQtW?NtUl}P~ZBn+#Bp!n8iq^S@mc#vC+>H0on<Fs-J?J!BY^}^fG`^{cpGS<@G(;yELL}pZ1P@20UyH8CS^<SZDXI<}C`V+8wNkIZYN0VFm1LdW?|Nu}x7Ro6dC<bLe66O4UEz$wc56B~($)#5=_}I;WSR4VfRY>8`FsfGg55aB_9*icauMu>nOWI5k^-J+fuyn``$|aOLL*1nq;^B1zHnS+T*^RgP`QC#WL*9I^Np;p12W&$g^HfnRnbS3wPJDxOnPFilc_j5P^?0-O`wZg<gJG3hFLJ5*jlS~A`iea^7*yO`-_#^>5#4TF3dE&HTw^G-G1D}?7urP{b*BFI!>?N6VXFsD%!}>mqMT@CGtdlX%#Lp0+GO-MKCod@-?!EWg~@3ai_|Vj0H*{(+vtxhpH;X(f*Uck}Qzx9)=_Mc0psQ{gA|i;8~{ilSz53d0|gI54c1-#T_A4u3Il?Czdwi&%KSEmUs2iwD5RQMs(EiwGMXT{o$x{AI|O(2GO@(-EECx6dyTRL`bCFYc<d!JLq%|v7K<;r8v3X2zv8~GKzVV!y9OaQ-Czcj~T=&iV$>AIU$lH<ZisX)r`GiNpM3>o(LOMXO2!vNJ@3kE6ngE!wmCyNy2UxDj$F>1UnQdLO6H_mZ}A@?pi%pYB-16(0k=*a-ThzPPx9my!viM#(*8k610Aa%oT9FqwJKZJ?Un~+gy1Ckt%=<?b>65d@x*G`dU!SA*wKUUr?yxA$zTX>I;P(jpu7Zo~~pQ{!^J6ff*62XgKcq4*yV6%CZTeP9b`J$(A$I$BrD<R*1<U1TAK*pm`C})oZxhzt>{V!63tJFtwO@UV~w?)sE3*k4lz255jgVYY?Q?^ldW1rG~Ahm%5hO)LKVqn39dvOVUe`3O-1%ln6}WxOXwW5wkv`yP?OOvS4dKeiCY2OSYP5BJidFIqyAtR;K5j0lxBV&GL99Z^`Rq_<3(BWhk7?zY+<(<PBRbq07#rt|=bdIvLzzy>bj!cK@U8Qni*Cxn+69lQ|BUL88BQr+zaf{ZDKXS3@TLQqj>#+Qz=4l_?Qke84+buV1I}UAh-`wIWx9^b6IUS}>;#S*>LRKjN@y>ryGxSlm$VF+Yp8$5_W6Mh(`)P;r4qHTmXBe9Q9*UQw-59`!Z%FysM1czBqTaWa_e?eD7ihKKMSq#*H@Bl%=jnjtdhWev71XE9A%L3t}x<}nrJZ8%U@Z&Q?6PLKl%)BC#<B$%gZk4+|zEn%W>O^`}+%xWgoq@?hR7aO5p->C~jUV*U_0{kJ28PGEk!nwcevUh2HBJE>Qxk!jSjS>)o>bn+-TWwZcnGiWy5$ks6nPaV2ew&@TN}pP-&F%I3-ADs6D`gDjLhq0$((bhA>QSJ?0jw;}*iH|2K;HCl43$nJ8D6GqE`K1^Qe1k6vb^UyVD$H3cGZK#$BRZlAyCIr)o{nEIYz@6K@k5ce18H`BIOl@yJ%Cznd^APex(@!6Io>KTD@ol09@JvN#<UTiv~sOP>CXJ9%_{GuvB=clS*}__G)2v%AD;(W41wPzA;-1r_EVB)Mk-qnMd4Gc05wbbVRw01Jl6>8<9A5fOig_N?xZX=-l3bADvH3cx596bf#yLEbdxJ_sjM7#LukK4np}|*#6~LVeL-^R8`sjUvo)w<lKec4!n!=y)dvgtWXX)4$(1jZ^EY_BEP8mk)Ehncj5`AW=k`dnkW7o9yyVk0iRN-jKF*zNr#1#F)?ChHqQ8k(s9b8StL$>w{oZMPl@rz43GC%fLSzFP=^kO5u12){|E<_r0YBxB$KMwTe3(<xQ|$r<lE7DC7=KdY?BCrrAsSd*n~qi9jjbXQkglvA9uPjDn3NYW>Roe)~H@GG!xG|U0jj9K_7(ZxJSOD^Q$ZCm9_WFRH6yaU;Ydu+*;aPuP>Ev(9L+2vQ1+|^yW1R<ir+q^$z`3yN!p;{gwqqeXx(&X4<L6pQ7^PcQ445*ZYJ4G0zV9JxZMLT@nebDpuGIrVJA|B+cK5OFxJuFqkL2(C!R`$u;RdE$(0mG3>q@Ptq7TH#P<kdP6FYUGgz*9PX%y*2==1M>&h$E7xV-64W)xO}o<{bRJU4ifZj7pYR>rq6$;4MYtp8Tx>uJG}s`RQ<yKWZ74r0oK((+qs^8z@Eks^I_9DV-q8gQ9h$1zzqK5B(L9o~j#BeV<MN6Ufcd2zvX|G>+0I$KMT%0&L^}{9`A(E4XT=XS`9@n@F-N)YwKFAYpHXr|9)d9Yc)OMEKwPPe!!mI9rM$c5SC?-sRW{%*DltZP_yiW05iz+ur5w&tCcwosjxvq>N*VOw2wE1?L5EX(iSzQN2`ex{3$44Xsi!lZId9DvuI3Hzb%6kS#a^qky|dLH486`SGFR_f-0f}*TH2*y9d6Oc4R2Idmn&HD`MomA((-4yy`)(!d9b;%PzDx`jte&ck-`*QDX&{)Kg<kC_0CXJXwwN%NQDp`Q1XxaL9r1JTkU?gSx~t?;F#a*#*J3NKl>x5(tF~&LyM_W^Pp860v;MWg@IrAUo<C^!+&#}eEm1a>4M~#<QS`v&t5$#yi-at9Umt}$?@YC?4^&vdYJb*7*oeQ@WX>3vuHH%`~6nE-Ho>qBVrx`5pwnkLRh7NuK*e--z}$7KP3Mdn9Ij!pHSrFBR_k@;U2IH)D`vOz2Q!Oq_Kwri?w01wcCef+S@ry#ZOZCog?HL)RfN?65i-LpJiM+Ub6?2Z$hsQEP(i*_}c6Y;WpbCiLWx_7^Wdeyc~LUxI=Wd_xt<9I%odo-wk(jMS>?bVNgZ1CO|AC_M}XNX3jx51bdc_PhG&!XrCOW(EK7lH~RfSv(rON)ycqszagU<GVeWcx}``Mik;zbe+&5ngQ5~omNKdwN!JxhY$|r({OZ!?@=6ksyS%u*j@1q4T+#JGNd1`IKa6<zDrLZwzy+8K8CcbRm(LPlQ`?v=kJBPMpRhWN3^p?7joWwfVe~L9s2j^sGK6g%r6pgHsi@*%^{vb4wMbVK2|pwsw^47~r~TwkMFap=4HXj$XRqkL;mhs)ZkI?KSj*kJ&7-T~>@0|S%Hu3^`c>3KCG<viBL`E&5mPkAjTM{77({6bt69Wr;;~kGSZ&nspy7fBm_f3DKZQaG4aDBU{q3BL975IgnSSHV;+t>28S)>ThZ;I*Gy6I>5`VFk0hDSv`{tW7=rvd%Jz0=3WwA$G+@_{$5mTG;VWtOWYTKdwAZ{VLx2rrhSJ&$AFRq{;(s?#u)n1FpMnt9yT{B8pbJe_}>Ktr*VI<L7vfN>mQ58w+RcT%gFH37pB&&`n;KmxHYX1PTN?|q-g_43fwW4W`Un66Y`aO-7>Mv8Zlaki8G(!T4--fA#7$&2kai{+P6V3H7AGNe0M<8kU4(qT{`aRf5!~J%<b3jh40A3qVUvFZ)6&t1&vj4%Zm_p$|9q%Zi>~9ed8Dd3r&nK<Ufg$>&MV=s!d1>o}KoQf>Mi0<%X|@8GQ?0U{-_N&a*msjN&Xdyx>j2o5a`K%b)W8olPiURutzE=U(7ie^Cs-Pd3no6#J@lr)hZ<j0PlHrZBO+3&(uE}$5f)L>OMIcRxRT{@@)O(y=qC)zr873jVq9r~r)Jb%Yc<mgkvuEm%Ks9uxV+yTbrun@BIT`U6Gk6abFl^JqD`WW<=7=Q7s=}~Lu^80iV-oIw}cspQgS2X=b^?blniMe{u#yFn3B`2-on%7@(cxkO<;!!CX(4UR#!nVy)AY_`W!)5=lRvuwT1O_><KYSl1Pf|)N|><7ndvOBD;JGhE!w2$tA$C^>e&g-IWpZNhwoYwzv(3Jz5j8Zzi(edp%ra3&)rW3GSjW<Y*^)?egoh#hIBm-bjl<2$W{3wa->%K6EXiQ~(lBHQ(#D3i#IwfVOZEq;`ysxO=VEUPQV$XLkFY&d@O~Fm5q!mT5~kfhgdEx31<G3SqI33HJ7igSgp=d&A_#YKHJw)5>14(?1o_by){6VfmrnInL@O^g9}~T9H&_6nGBdZVL{?M&$3rq*;T*8Eab(M`5RV5aAm|0;fR#hW8@WhF%R#Tm@NS=;3|SLWC2+#LR;cEG{f8=e=MdD9d~(R5AM?Pge_Ec(~aNE)ofc3I<MAimc}su*!;F*k8EW41a-uJcN@U=T`01aC-2E%vIw$`B`*7`BKZ(PbR>QF*El7{;fj|!F62h{UPJI+9`FWW^3Ufx;pbJ%`~XaWQ^RaW8TzGuYPqVLQVL&W<%&9+%P~H+E+{_rWC1K!APwqo1>9VgEk$A2k-|SQFTR9{obx$O3;OQW-*A<3}}Km-P=X~RqRZgDD-)+*M9(fQFee~xnwz4-mZ!|C&{4)6Lczj>QZMQyxbNSB7f$!QfXo43jI^xZS8Kg2AKF<s)Rst#OmCQR;kp4BNuHKhLcgc%P~o*+`;rp14TLjyo4HAla09f^pKceRaTJsD=2kbTK(--Km0I`@>OZZB&j;PqbG+Z*5yH+#zJ`U;)iiYb!`fUYV~sy#v)xu=7|fh1Y<Y^o=0P{(C+PTQ6M0QQS6B+LD$V>ls+}c?F49%G{U+4RNh})sH~7BG4q0CO+>(_C)dTf@rd$UYS3#~!LSVLGrzXFzJ9yB^sZ)}<U>Hd>jc{Iz8m&2R`VC9mkfoemYTy}AXINS;ND!(46})BUI+5Q1Nksf2_wA(!8jv4QDR=SWhNDhLa3vHSkvI`L}`?lci=CGLnUeJ$mC#7O0^-J!ioO#5cA?l$(i^e$RS3bM~-UAguXn`B+nQ|G$<a_^dTiKJ2ZGg?E-+Xrlm3H4~PClgQ-t6V)D)9;s^344BpuVks@hmz!&Od6Xbb7xga>#4Pcxn$(U<_E|j*z{a(n1Qal9798~acT&>T{qD(?Hb+MkVJbO@JBtMC4IVK3XgG+T~H^iz_pP?}cYbknZI~mo5#rzLkp-K&gt|6mCKT8k_#_=|C@KH#vj=>uupqBOkf)AzkIF`e{7`FOb{5nPh{+I9}lAMMv?dU<I@{Nxh{Y5+>3=a%t<OV88yd89>B;}5TEfS9;L55=(rtOF1qu^<%Bm5~5e7z4t4+=rrHwmkHqH*;TPAzi!HGKXf7956OpLnM~qU{wepJbQ)fo!0Z@K7|spE1KKrIGy{Ng1mjV-g`6H;U5RMR*US>)=Y#$K-5}Ib9p(uKg4%pLthjJhA`SdkXey+GEN;I`MPerPQe;Een}WQf>4ADVS{SNbt0iU<y7FAtJqX<Lya2ov=Sm_SJhE@ALFl#HK;16>>X3ox}BsUU0iA!Rg{oT^iI}L;RJTr3BrJg;Ogz=dG-I*EiXLAN75UD^4erEIGw+$8hh^o>_z~4jifYN;DCN@>9G1Vm=SPu%=IRiJqi!e30E(pl26$eOIg9=HFVYU|E|Llqomx!wvEz$wfxC`w1yO%@-$5L-m&w1LZqwmFtx?6!PKIzvV<7j*L*Qxl?{A*$o8*;fym9N*PJ4E?sqHlyW4d_-i0Q?uF}%OO<*mx3n|pbTX(&l>t0pWgP>gw2zuv1MT(x1i&Wm>JFrI1;STut*zo!%>>x(D`h%vxtauXF6S%fds+i+U=Y?87RxtQR@XNc=hus^gAt;}g*8W*x(ney_uju;1ax6Ntrp@Lqh(ojGEJHk4_@!}cRLLa-A*bWY!tl}4E69D-MF*sjdogaSb}t^cd3h+)0jX4gKE78@pfzYDqK^#{X;kxM@4TPHMk>>6*poyyw$Z1;)aCcRpKamH3(73T7bjC584J&PXHHX=Yw>{h=rM}NtzKR_4TNrmuL66V&Jgf==5jJub0fPmyNG+7rq9TtwZXj;N&O_$C-wQ=Qj=OgHQ<c0Y=(VJWWc;AnuS#X;E3B+yM${3M~SAzJ%9N-4HiuX5I~zd&4exex5>yQqK~TS57<_4N<pRaIIgEqsgY4a+IPSp%@d&pQ&a$vUQH)tI#>7QfehXaHInI?!^N?OlRt{vNoP1?qdjf-pnwx_U;AzKUq@@3(skQ>2>DB9t%BW8q|=qW6%OPcjZO}S{x3-jI){An~)oQQ;8Y7RSE(SNV=^FhoBN``pZ;d%Wg%uHJZUMxg+SSXuPOD$mk%q?q3#CuT?f~S1K#sJ6?HZ!Fwm8YkUR^)H>J`UQWXTC+%wV<8EsROaS=8C<nQuTMXX`m>$kk&EP<lrh~tk6$nu%WF3#@uocQ6-71FP)~pJJsz4oihe(F3Qvdi9R{>~bK<;MtnTbh~l%YvLcRFWiLl@WPvNs=wkmV}kUKF~ORZPWb_^{`I0rVOjm`;aIe@N|E5gFWk*8&_w2UX~WUx7j4qBew6in4)z372E^lBTleU!46r`03f%Cw@qDOJQW=JS6|pc;m!+g5aofX)f*#s?u`A$tVwm-xPadY6o{P2Tc#rfzMPV9Tb6Oj@Kd~Qn6{0TAh)jdZmY%zJQ4v-!M8d<ca%1Qd<D+PKFmBd+$*qM85u=WI>i7fn9~fVE;W9QLNnx-2T={CMPGBvCAl;(^Ja>zq_}C60JP6$_Nrx=_Ix1A^cy_Y%uYP6I5!^^&9{*ZeN4~ouhnA;&~T@lq}{z^NSNg1GSL~C^JVWg?2`{(au7pSC1S;Hz<(;E{X<5rbU7~nlMB6&%M=N_Ymzks42VO5=0$$y$|=Xb_6OfN^mIvj*psEIRDWfBlPwV%RNeQl+|&}F-LUoKuJ!1hgYW?@m_}lej5GV)*J|P$+(n{e^|`O4-b-bgqdUugo!|<b0(DNf_hf2dNI!^6|Mptl+!E!aEvh=-sx9opPv2p>{IdWIX;1ZzdU_@EVPgC#%H98f8t@AV`xu>vZ&t6AH35qsFDP#DKSCwC&#DH;LqQm{p##jLaRx2VUC3k^*7?{-yFk8&riRro_&7$wC0C6<sjewq-Hf{u`#UKV6P#uy1C(A+>3^apy5f7B$q(H&MIf4P5d>5Ru8r4IT>S>v=0;w(i8cG<5*SIPmR4b$7OsF9pn~ZJQr_ceK>`4!zVMiZRPsX;`|2c#suRdd!Zg91MLJvxjaw=eBgxBSXUTy#8Dq&=tat6k)@>D0*W8BM4ZFqs@hCsbAS2)lz3{AduIA-iVmCn>pG{@5Pf8)8+q%}7+P&dZQy!qby0NBp;}PE=~<X}1imoNvu#w)fyUTnX4le&nH=4+oh(lyGD;@qt8{vbT5UT{LbrM2*p{4xYf2l5Vw`LT5aC^%yv4Qr`Qyo2Go`VqX54!zzw`d6C6tJowPsd3I;KbcX<hM^D9%s=*Mqt4Q#KQS;<Y!-dJltfX<zbc-DDH7s3YUpQ(CSOzm^rW{Gu`)HLEDOT<VrG*1%V@6s8X8WO9Iu88`4|sK+U56&zuC6^2f-{bye)axztKIciLY3fbKzuM-3s%>`rnp}E<K(cpfWYM9*1jq;|B?zGgDiIUyIcH7TkrYEfRohD6(+)|`CQTs?SlwpW|R<L<NavZkyViXBI3WHV~(8HAdE`_($$1lJcYoZBdfRtstC>%Y4uxjL)QON_HNip9m2XpcuPqmSphe+9BRLJPsRQ(5YDo459&?@y7xXOBPMzVET+2c=1)v>1@h=-5x5|44s@Qbpbe@6CiG_wC*NA1GuG9B<pC*>QJRFd;_I+LcEcL4FOB_)cYW|A7onH1(ENeHKqz)-BS*K}BSJ%d+kfg??v)K^*39m4~M_8@^RNgn9}EGzu@5)}}MB1}TZ`(!qif*H7Clx#7w<ns8ygJefmdhX<|HFWL1bO{rLVAfX|#9jHeO9?1Eh}jEs3b}>+sqK*4hhl4IEEjp!YTI`%L9UViVHlkF!37+y2LGJkz(@G)LZj8~QccbaC1zKcPnZF_&}rh2muF_$TUTeVh#zH=Sa@_K9GJ~!_~Qw;<#Pg_R)ziQ=ib$#clv))T>tk#^dHG+f4nkz<O}ix)#H0g#fqPy0*o9`o?{vTx{JY8>@(abXTSF7PWO~ONvIUCDEjw=e)`<SqfY+>WB*!sw0wT{577C?Sg`Xc6-lO>AXN;6zHqG1WMAl(LIrugLhvx70YSjtF97csaLIdg_G_5jzYw$t%wxU3A4Q4KACLneEBF|meuH@i6cF4M^!*Io!{`MMU0|3s5T?FU{x{!>qK2~0A-n3FJ`JH~kGlQDz%RceQW-oaV4mO)&WkI)g{#5H<as=Tx&0UT#xr|D{UC}y_h`uk_IV`o8v29hP6feB_a(0ZFWvW4LG>8|3~xM<i-IK@pK<rk2_H`c`v0g8Q=)}IEh+trPD)RC1s}nn=xzs;^gNl)Qvqi}c~5B!4{qgvVfa@8@T2V?#pFQ6w8>%VBSfS^sF)zH4zBNyVpSdyOyA2j0!;olI0ilzAPIoZKA{1p#Z14Zc>@M%(Fhdm{ToIQ74n8f<Bi01Xa6C@g=egM3NQQs82JbK$wS=+uA%Vig0DXi`8`5q%s8%q_YtxATL-Pi{)h#z{oV+k_V&A7om*U)iR1RAOWE84B(``<ef2J&BDKCh=-|scqtV{5^y;hKek1Pg^oOI;8#8aryec*?JV#eb9QE=Q`sF0lIgmNDccH&O+S?y3(zDm&tIcNn^-I%RS6_Q``pRpKH>R6cuDm|oZd`c-UV5#0wY8;lE`E22tgDv6%e<M_y#MzbGV;&j2h8S#)*(3cRw_&0pFjF1Pdem*$g~zxlBf01LuW)XKv8;ft-=Jo7a=~z`e<4q7$BQ*U1h^Z6x9()f(gp@#i@dfA9p)&U9~^gQSO5uwmO5+4mz$jLH?%?Nt3<SppLbfu^A9UGaek)_j`@j0Ie7_xVY@z?~pbT^&z%`o7|Q;FLNr<EM+3sa<tN&&5|VA5jkH8EdZuWgKHzjvJx__OUX-dDP<87Q-Z@)W#vwNW0n8<Q*CQFUQha0hQ-6E7#Tm-b(x&FyRG&}kCyNES~6JPJ+WloQPjh01$?Z}m&q}XqhfECS8kB6lntx45X##XeOy#F!I1SnT+^x-#_bU-KYd9?4SW6GZf7{e{M2}jq+cC=gHjU044Z&H_+z-x=y=NxadxeduCPQMJtte7^WH0?_pg;SUz@V?I&*Vzosyf&_^xXU%d7H_<_u*xe(!XKl!RJ(Xu9IP7%RTr6vT5fBFuTBf%0V~>u6&q?%^KFYJ}TAtC8BC{Jaj!p@)7Wl8kV)WacjvhPJblA9|K_pBr+P;(1Plgl-gsV<sKAz&3N<?CcbEm~!*p=!Nj+sHZe}TGle?-flC2r3-Kiy*u7b89TF}aJ`#r3&rd>xIixx{W&SQX}A(X-#}pDUpMAOku^4<Rx^@!)v{2Sk@~{HisNBj4@s^Pd^L8&Wh6Uxvbm+?$U0*~_EboyKi$U^4J0Fbc6R;}=Z2NxC<Vm%_FSzb7!E9){GiqjjdHCRYa$4L2w7l}pX1{o17l+u*hH<eA)!Me2aw6VWc~+d?J8v_=^7*(R-uURl}Q&VDwO$UJQ=-q-08wOmiR|&4m&~WP*dPg_(MFZ4?4qp0LL$x=xxM!$2^Z>-ZWGr7uM1bHEd;r*Sg<nVmx+T<@@OE6<NhnsAd2gW<znD4MRA7g0--U-ZXMO)++O>YZwE{PSaSWro3FipwW#xw<<!0gU4#!jGVqTRWAhlR9)~}((QWndL90MeWQGB5#>UiEojt`(l>Hm2*1<_g+-`)s3%GtQX;i&8JD71084!*V>nGNC5IW5bxsCVOR45jFPQ8AE;|{@8YTW_PZ1$A-f6D5aHQr}biyU4{OVC69G}d2D{$dk^Z0OYdZRvSCoy^>Gp8iNSXU#U{Ecpj^#^bu^&b>Qp~@>@d!9Uu)T8lt$&5`~JE!pN%+Xi%8~eMs%#=sSnF<XWkTRhVOwG>8H@0F<JLpEQCa<VIwfsZWCoO@M<Huz;PDkqcX1tZ0K*=YsX!;x~Amy+4`2+-ltqI3~m$qA@^cZmSN{xWOasTPeJtH9?#P&KHbI>7A;H}jhLWTBThCHEXZF4|`i9r~qEej@d&dE&W{9N$p@0=eV?P84SJR-WRqETFJdAi7_@5pvXc&){mRjV1ebk9drz9r>Be9#$2vuT8B$I>2fn|?FVqQDj?;K*>wA^b%##58_ZU<8=$4}TFkkORFnfT6X9ZvndH%ml%Ui$fx$0(-TD*mGdU$@11PKnP8xl@FqUkJ{zCcZU}X@03gzE!L`kdMtlEp1pf_>ljs0{1;!sP|~xHj}uR?uQ|l@^t)qH>VGG^xt<-9J>k*W?~av*nKbWbpQENWI6i%hAtTZ!ta|$7?2k3<2!AElFwsD{n4!{EJv{p~F&G?+2~Fu{3*VT1MNJ+i{WrGzSeBC+cJ&o`2Y+lhn88@bWV1k3`ERA_TJ(zM7r?@Ir%wXkJao`)q`uehb{dD413Y;rhQx3~%0+T%=F*IP^pGD4Wf(DE9usOahc!4O<+d`e@m7Dd6VXuQCkXFa`!zBh0Ev(=x9k;QqVW0j%XjYHExv;vtJ8n}=pX<5(La?!uXfQ7p)2UW)ow=x_*mV=ZBz(Kl&=>j{Wht#cKO^X%*qm_*tWFG`wVX3ltgNm>!9j@pnc;DBtrVYqp1$8eYU^laAh<*0D|{JfHeS+hm9BzKihhZ3Yu!~z*2vA`h%P<zj&<HpPbAwx%QP42^4lHtf{Tbc5l41tls!X*fXDDPgh+%71L=v2dyCps)d-Gcfws1T4#$OzxGgeT>%-l9q)F!hmpUzGQR{OnMjUe*mPY!w}L~F0+TzBv@tq+-A-!|;YS>+zl%wV`hy5Qhe;$iQegyNiRSA95dI?p;6Le1h+JPS!m$1n0bn3IZse!LfxahG(?`b>r-eAqe$2E~@>Ikd;nsh*Qn{rfO_wX>_06>khWaMRO~usOXCHd`rYW(IusmGzOEZ|<9ZgD!Tdd%DYIbBD#Kv!)h=xDHuaC|?0XA#BC|Zf<D);+W@`>ISF!3z#Lu3h#ti<G_Ov|vpRdOK*OL3Ih!vbgvH<3b(__=-?|G^SeSdOof%CKbY5bY3Skeq{-m)0uf1-!ehErZ<gw?x;**xV1?`fyxWFzcS?8n-fb*flE6q76}T^$Cb_#24~L@)4Z1$~xv);+m{P?tHCF7X_>d{|Sj{KXbR5%z$y>FU-tp)~fmt?|38%QXC^gWP>CfHjsh%%kT%WjFIs`;+xlUaX9d8;efuPSfL-<^aVKJF^CpuYJH|YpgiylB?9J-xYxhGL_c9h3z8x}IhNaAut%iLEF8Luo#`BAHqBu)XLr+U{aAMx@HyTf7{4Q|z2FDqR*;qhr7^W8=nMtli#h8V`tu->Zy>#wVP|9%qDVn$9t`=%1=rekU$(_PK^J}e(|;wIkGw+Aee0i3|MTgePXDWy3L5Yd;qvTrn8~G?nK!)CFN<UXPEJO6OP=27dO?mF(Dcousd#$!2i`j&iC0yh>6|K#X7Uc7DhSP&DgVR5>L$h+1L}$HH&zXfHOXARR}LOxhF_fh15Eg9I)@&)=U5aID9&Zh?DVNQv*b9&+0B}>V0{=CO2>V>R9PQ8i9boo2Xue#rH>x=R<TYR{!tt>G&gfPyjlqNY!=)=+a~D7xv_C`b#uMEvhde~KND`7l8lXR%zz*#ZK2Lv&q+o4Mj<C3;;h0M%DQQRSf7SGp%hI}Bk%=#*FhHNWM%Ev+Tyywr&!fTZKdbpx_C<|`}~!@*X-|abx{%rEQN68KWH`CPj-O654}czFldo$EHI*e7xk7-ZsN!R5jUG=?uQ$*Yo2goh8tJ2H83Kvl<O_toQ8XF`zh+3R<tY*=*RD2RrR6Q?{yE!ZN5L)j(eSl#AM^H7jJJvTlz-PT6>uG2)<yd2vkFw32llgwYd?s$!bqM@-Y>@Y5H-%OeoiY+X>&UlH7kVtc$RhEH{p{Uj+`qsv{#7XHo;%WC|qdS)qX#`l(5+t=_ILZ!T>t-jaEo^e(A{sd8_p8q5Q&&{@6{tgBjcR@YSuUa3VH-8&t{YdVRorGo`A7YQdtghdAAJ#00-twWETwKVte<a{@1b`vuU0L6={_1+G-nNYyxXs3n3+&;W43g2ieN0)A-Y78jET>)EB?-w}`gEnaG_U}^(+To}bH@$w_LzPyiw+)RbH?T*0Z|GrS3vYYBH5@|o1aek&Yfxj9k|@oWa!-<F&77#T-IRnEjaAofjxL$}BS)6b;f4~wSV2CP&762{rzLC=he1i_#EG(oW0<%JM=u{T{Ur7(%?}_WbXi7$L~*xU*wd2aOTTK`=3H-=p;Es6fvWF&w3nPeIWL^R4x6=yCpn+X%wnzL^eEcdX1@D#uWZGiVpxyYqdYsi@eosM$ITX+bYqPYIC_-=Vv8BLpbN45MzdrHxR#jXc+HmGlB=2H{wjxX<$ETLPsUmzS3Wu{?U~XFlR#+V9+0|AsL!%!%kr9*3yb+M<-Kx|;IsH?I$ak2m%+RmiOta2f&xf?7D))qeznc(3iFdSVYP@nQH)UiP)jln!d6N{EorVKF0CM?G&)22#V^m#S8i>j;(mO6NySyQ^1=Ml=7N?Nh@dVnUt7GfiMM{&$m1god%|+38)Gxdr0NT=vph3MCV8MWce`%F5vhA_VnC21YnII<!)@kStU-o8G*%hXR}Od)UqmcEea`2VY_xUI?syNZxqL>O1nmRxDK1Onj!g^!T}5(`s?3l-6q){Nb6)>mbfhueJl+{aN6e8=UIN?YdQ6c|j=mu9qQ}ID=oj8L?h-EjOX41!^@gP)kna)GXx2gG=DpH8ELc43XVGl#iNhWFw!WGCFr)Sq&nv}e`}YWE{M5mssDC#Dbg(|ThA)E$+xTNWk%9Ekh0$6`08RhC>Z1QdR!dEx_+Y06lCRxPm=Ci~3W4StkLf^i$>!wSz~{wVl;<Rh+Ps1XNvvAxm73#Jadn*62aN`Yhuzmfyo-fM)PCpsA};J<!t9LG7puzzp@T9Kc0i>#W@;M05D{CFCdh^~@kolNlcV~Hqq>pd*bS02=7vrDyE&t%{I0DCYz$ftOgX~Fzi3+|%Vs96YNwlAn9XxZjw))a)lwH?ok7|m-M#C&S%#h)*Z)u)ekA!v#C%P=jYxdUubPkGZL95;@GZeK!g{@7-naJKxqH`HTshhKV}_cXMUe6JL1);i58?-W%-357nEo*y=>1Ak4h$m}VMY6aT*g>#iKN(~3d*zO0y6BntOYH}L5VH4w1V;csBVoP!1b%uZF+m?mXFR@bCe)P9wqu^5H%MbF&w1>AuWrj@M_4#lvVUP$j>Peg{WL}DK}}{msxX@?<`NK@L5M*K}D~jq|wk%(P$FtLL_Nqgx2UW{ku*O6uaXj6_Ly^W5GpMBDt8F5E@cy1=9&y^TEqOgN-K|Knft~B$o}DPSOzg#5$CWUqx}$bzm=WSaWeIk=s$%lZZfF{Y_h3?DqTjfEDS~P)wpf?6n6$&H%%jVl6U27eCYa&Eph}_Nm?s1dZ)Wu8=n2<<*7CQawN5oA|cHfSX~jZh*$AcnSprFT!559lT=xE;F?^gIA2-jUMH}B5w=9%Z(9tS`DqWCRLre)OaFhO>+EP6s_0N;<+j>Ip>Xb0HHhm?gWA6>G*9k=A5Pwyd9}79U<o&%<W3|(DpZnsQ6M*=@fq@QZ<kY%zLj~@?^|64|X!=DWV`p)RYK>LE+k6j#O}zyVXgKjC1!R%_14R;%;ISDoc{dN(8!AZ-L5BZtr)7J0A1du9`cgqhMTN2I2j|t3q9~H|QS#TfrE3;z)3l=z2p;=;`@Ma%A67@JTQo?#@P11rv6WpB*!#G~tlK>?v)`;e&~r<+HbEjElQ39{LNF_4&2MTbL7o@9LCofYnA@`&e#oIHU;AQQYa$nRLH391-O5ZnIW-Z*virP?<BzOG|e|I>rTfFu#!$*)*%C^9Jq7%6Hl8oU_a*a?(O4YnPYc=ckU5!hJ2!X$5nb)Ej6o5E?nsBfsJu`IHqK58Y(x5t~RVhAXW~(xM<)zO_`@sH{}TFl20F?p|iE!Q@z&xJL3g5Zoxqwgkzs#|3<2bGo#Z3%knXAjXDm-Rz(pDwXy-*#{tSf&6=<0D*USAK<po@S$PhW~IEeaWi{67|<npw1LTBVYw(H)u3p(4yxTWD7q49`8)j~<_+xDWm0p$lyW&jlcxx=Coc^B>u1LtX-|$(&&e$M5qV6WetY`tn8WAMQHcH}m!ad+N2lMPesT7xFr7X}uLyKNv|-ns5$!}-wjz%9V0l9?N6QM3rZ{>md<7aMa8_y52%AIL1wZ(vJt&$Yj;+)q&lanVQvLI&Gbiap<76J1?4-rBFiWEFcUp0GwButD@}#z>v2nSD5LayKE2iiHMoKl(kq=|KkRRyyz#VLNB357~wG+zkmlu~XVLLMx1~<hvMB~Mk>ucrpjkV2rI{&HkhQ>PVE**>^2e#bYPi_CEK-H`9?iOlk`h!C)BaAX7Mn^n_lhE55^y8+-U$r!6EG49X0OlcF%e}@<t8s7WZMC}n2fjO)hOdKqAdSprX%dQ%fr-&)TqT7r!$MZrzI&l^(CF?r(e5H^(X$<pl{POhkWhr*z2MGlMLC`7m8HcSiyShA#Sbi%QvKSU`pxL*#FTM&!p8on-)__Oh`X8J+*rMSotn!#y(RBDI<D^2jK1Z5-43~L^OcN#Sy#q2V+^IeQ4EcuaN8LfIhmobF|II3-LW+rGu_6C3#D&cQ9B6R8EN-e*dMF+>b44u+29pB&p<dXSShP)ZrD2vd%IC0ud|atPQpXcU~GYO6?zF$okU)awV%o+Y-3_Hfj|v)jPW2ueO$&UYRh9R2i;s)yre>7XgL#miWnOkvRO2s%AF{}?BENeO6mfu7*Zpy<)B~{JYgtx6xEs^WF>m--(w~!U*=3%niQ4PgkQb3UZ=#SLYV3uI-5~Oe6!)Mvtg>nd}b!c{=Jgi98y)9MWbnkU?^7wYN8#bFHQ8Umg1GahB}2YSref}{8?%|sr1-z2arz>eIxQS%q>F}wNv|Gg4$z#$c4j%zl#7Zf{nr%L9O5E;&T{;c8x4ov)v$6tf+r4<gx&f&pB<^H*ZnhAT>Ed^9Yjm=Jo5as+AQCXT{i9SO@i%@1jCCOGP7rgd&&S8^Un<xu$d!Yp<!*5<OZ<A(^r4i9*p_@v*h$@V1m*FJH#%x_-}FDI8Ch7*mxS+_77{1nVLe-(<j(CM(K|O&yn6=)7^AC+UFIy+Nlxa8AGhm^+73V)dL5BraK=N&?-Dw`Z7{Wo%C0vbe(w$gu|)=Ul+IL;PpzN-8>QcQ2xOrlHejRBBU%w(hFkJ%SET+;f%9lSiVpZOs%!Qq`wji^Q)*ZF#2Q^*PyGHDG5ym_j;CRPS&CCqFZCQGZ)^g_ub^^>kMh0S;-t^-4*M1i11{Rn0Kk)Gpxwcs%MKjCSfAG+*Cqi~t#c(2E7xEMpmFPW;-d<JwGJf>zUdyy^JCC8FI8R3Kipw+C0`b~w#{Ya#sY#k~L|;sj;#K+GCVS3Gt>HjEp1TZu^Ziic<s)qKdyD0{3W-WsW6O}l~CneAH|KbUNH*QT^i+*LV#?qM}cPbIS8%1DiPfOZWel?U+vq@0nr+lOXaPGiivutcgQ7GXlWU-C_i6AxevU+T7++vFkK<j3eRsZy6}^+=HoSpI6saF!3FA?{}*jifiJi877g81gt#SzkmI2`u=A1Kq;aKn{}_9As=iyr)O^x6!S$;2dbXj=^f!-X3OJLgDbE8=)Vacp25wEJkK88k~XNv6_|$@^;*&08GG|n%8-!!w8PJ8-b)>60nZO`a)+FGd1-{Hoj$gwhUb^W_Bm~Zq|vbawmccWFN1nP8lg<JKc>B5bH9*{CLd6(Q6gpEfGr$a_HVnB&|JhdwUzGv}TQE&}V`odctBhKx_%E;J!i1r&Qb$upC_NpCAgH``nyyzH1$#0MuXS5GB`~?8Qjr=0qs&>1knXq6;mGNt1&vpDvZJRhHKO$}*jq=0X>%{PsT1u|94f)zI-wr;Yi|wKcc{y}<llKw`G0q6r-FsLl-y;M{~KKz^abBYGmUIj@ZbG)(!PlIDWg$W0jeZ>esHEd(o=WUjnXa&N0eJM`U1JQ9ucT{g92_a`%+FulF(EG7VriyC3nOfpGzpe>h3u*4!Ch;DqajEb%mWh3n?Y9!kcFeG`^@%UADe;|-bX)y&Nh>{psqFsh<#M8oL-GJ@J4qlint7ykgU6If6)m#pU2C*-gk@#V_6I&t_QK39=$0cmy7$Q4wii;~`GXSGl!bW-h-8u(E0bgzOd%y?_3{%+I?%`FQvlmiUmsI#3spVA{ISFwMKA^UW%8Gk^b*+N+I5Ev?eSURmX>r|(d=di|w9&E_@-3~6Hn2K$Rmu1F%WI1m^}bXoziW99u^h3wP$75Y_2t!fD}G6vYLtAe>L4jSE~S+ehF50;Rr03FLJN(N(l*u{RsHnUTf0%&kY*uiSTOk+6Rcut{7RaqDJN1AlU%_fg9hc6CY?HcaJ!P$oUYHECc2(Q1?Z)yXi^D%_-i`3v~U&1dooV~R>DEgtklCU>{s3Pi3u3Iswuaptuswr`zojyXT1}=qCL=<btX#4sa8^LG}^23OG*x&^R{R3XCs6qxg3kY74y(DU3i`8qU3+uv)Fk$CMKg39M094QJ?tZaX~|77$dyf!-otA9HeWgz{QI5pHM?-#BSB0)RE)YodA=U)SEN2NlIxTh~!uCeu8m)OOd6#W5A<UmwUkHs<nK+JyXZ5#D{EiR*zrVo>A8nz64VrbwIM+P}Q?j`fAK*u)KI<jWu(dD>CDvKbGuev5;gMy0uefZbXyF#?8u_Q@ibAAclPHXpjm=6&+(Vs>T)D?MzZ?xAO=`+eu776WA$F46U04NQ34j?jx;NztP{_#XJ=I6cP<*MKXGdZRdI{hG<)JxW=Yj?({<_J-+u!<Qa<sZ@3{(j!Z$aq+CGnQzX#)dWyuPXsyw85;2pE(C9N4&+|O(C9bBpPup$Yrh9P@HlON%qb71m31sHDCp<GGNzD{PB{f-n*?5t!C;6bT0)t3OBl|O5Cnh1qn!__}t-i9lq7|njtE0*sphBfpIV~C3-x+$ulUhx#00F=lb+qQZhpmAYfA5ix;4s6p0?Uv2VIgs<z?CMq;#M!rP>=?}+cS#Ov}1Cp`LLVee!|kJYCf5+w5CgqrpE>r{q}*U`a~O%o(whATvI2S2E1Wz7UAlVFdOr+^ytuW`Qd&C4w1b9zUZI+i!3&D_J@)mYE=w;s-S|IbDoiw4;D~!k`Zl_f#HQ?gl(_e8i5?c!<RVddK`FxnR<_0CR$DOB;V>#K>@gLfYdysIF2CAUig714K0%79QklPpFz#ZzDi#!elWhuFFXePRJ7eo-~tti5i(O${^UH8#9O##hnQxlLrzC9LLv{|WEOln{}LanW7(XP;2g4zU>L4!$F~<(7FKVgdP-ZeBu9K=)HQGNR&sc%sN0JQ#+IS_x)_2ZFcuYZ5`IGS{axdTch#69@VLTU+dus;lnmxO%Fx3}9=|PlX69k5>`zj>ipg9w*Jzse%?s5Fn|brK*Be*goEcLuOf}1|7xr4wJN;Lf)|Xtmic)C(hAMKM{k^M((&?j;clwy}%ZPGRTz?B@$@yrW3S`LMpaB<+(o*dFsLF^v$MR9Ka?{x#yu#VvQ$432y~Rc6BAxyxQLKxcQZbDZW(5<)q(my8uekP<NE!79tzy>l%%4B{$K|V3Jt`?vCJR!99%a(Gag~ZX!8AVQ-jjk=k3D?ZQAdnkct-iZv}#~RwJ2Gg$|#1$?Y^W6Tq13mESUAkyLgeh{2H_UV9BfBQ0_D?81}8GIfa8f20(v}Pbgo}6V50l*Ny5@UA!o&i@6GyJ)+uZa;*rDAF0eo@WT@>Hugi3S?Dp5$?r_e=(W$($~+epzO4=P1Pi%|B&pvcw9ik!fc_ZtL7$R}OpiLdExEI(LYQ7q&ZrfKJ$!}p;hJWg&gm=4<0Q6X1}%GkM<M-}W-eX&J0xB8_43~VF+3r9qQ(7A)=T@ERvN*i3S?qhSTRlfpi?_7Te^}MQ7=DaJse=;pUo0$SXvAaLWS18O!7U6vTF>#h7I!6h994JqU_rbG|Q*5UY>h1;?517yP4|eAwB+<d!t`5s;0j`l7;cKN`07-47&MTQa?3!6mY{ral(yUf&Q54sC~lK_&#Dz#17_I#tmOJXn>0Q>Fky_l}2#iX7U%T-iI`zC>dXS$^c`{yvL$s;FHsDRiV3Yi}@L-E?&HUsd%|`@ghx4l)d9^E%Ala1#AbYim;vR5kVkU-9!8dH^!He8Ef<+EjQDw@(jJdlfrjhNSkcO`#bLSI~qpYrq4wcJj__jbUf$YxWrPKFEJdyO7UA`Q6yS%uYUueKGsGhO5jS4-|N^YzTDDy@#}B28gE`{|92>jXAW(a#+!Zf%{SxMFTFPXMx!x1ePyN@Pq!|=Haor5eEqep*Dk%@x^lI7?$UU#V?{i;kYQqy^$qy5;FYf}k&<LR^yXKWmseM&#a+%@-`HGOtibOg7454Vm1THPUV<N%7B|*yg~jUHLS-!~WL?AcvLtJZ%P8J$RMt0`@FJnp#V9wF8HIv`8n6(U?Oh@><q{L7e3k1eO908DZtdc1zIO2ym+!Z{=8{@Lg)k)fn^N=AH7Hjxx!vjtrA}lgMH$3kU9ZTY7cs_{KGd&UR`m3YtA{ns&sH`_1|*Qy0F5$=-3@Q*ikt4q4x(_*YqILt!j3b&jki=!-5?#)17R6ub8I#m#JwRl#A-hjZlIHv0wt;*`SAcF`<7><*XPhLRcj};xxGOPbEw!l{a=KnNNSNM4UARL=sKKr-1G1{o=#u^Jr%aKMAOvDg(O!T0K#!E@frL@_*zinl_;+**O&b%428M}ms=CH{xTQ%!@*Gin(p+Zi>SzA>rZVvbjFS%_X0WmX9^}Xry$p=+%skwoL>Y9e)%<_rPJ-sD#)fz7+a63+eyT-hG?gH4)K^1%Lw@X8RHu|bfa2ZYQfix<rk_J^<wb$Y{~Xqp|=tgpsS6qULb#3cKxb5aS<ZU(3M8M;mRKccgay7&Ko3De+><GD>v#`nH0JQcD=SprHDi-srj|l^>wN(<YR^P4jO6ugG0Y$%_O>dVtao4<f*|K^V8;&0-CPsCq}gKx^T7&F><0kjp{qNj3WzZ?E(Wp)!qSK)iGw0y!*%lZ}gzAVv|R$;Yj3r%CSRU-&|VqRcfXj8b}O(l<R}CB)U$j>@Xi<y_maH=Ap_kkvW&7?Gr~lVh~R1N7w^8|Kc=-Z-~rP6=cW$`f#B1A7BwC>Ic_Gn8{uS4Zn<~SA4-{8WRz2dGiB~^iNamb~+?lOeyoU2<3zF!UA`z^>*Uu@25;W#rqOjApa<t7%}kn`->|Z<r|dZFpVLSY<?P&D4fac@MYr(c>^>-WnJRztE_n*-*Py84WP6$@|vA?yEPC=3Xn?xT{l~eZagqv#3X0f>tx&le&_Y6d~j=N6}=)c>YNmpiS3bjLMp8ujLEKYQt~yO0MeGpBu9qBQ&)LCSh?aPk|H%+b(;MZl*H%PxQGg}KGjM_((BDwq_cgMQQNJSf><beilf~CI8mq`1;NiE2d8v<P?cP6v@U{9u&nH|OX%*haeGxjt?@>nvn<%0Y|OzXozS{ZM|Y3;yQmiC?+Rz1j-3+vWF3JK5`t6+5-N2rIY{EaKM4oI=)}pH%&f?osq$;OSq&x&P0#9l%tj7n@*M2>#MSo;N?7TRa`A4?o{+8CDQ9hr<KDB6I@p4jEQDh?z;<LebOu6Tc>XXNk3wW-N&LIkh{igpIePV`B2RUEdlB>KZ&(MTmfw&15jp+fyi7!XdQh8$qR{VL2QfuS)RmM?2*#JU@RNut!_j_|?$ZK(htvqLQA9DetTLo7HU|%s%QZR;IKpl;H;XCMKcv$KX*kC2qz6JuZ(<DElYu6}Rl%K@D64DzWIp|u(|<kvU+I7Uc={ht|8)9)z~_H|`hUUa|A%ZFthWD@inyRT=JS$w^Kx<a&ma9G{CNWv-`{YR1#F^9U9ffN6S9|}9`aF$`s*jyRv3;*UrYLG56>9w46on3a|^w7*74S-A)w&nz_{MpU1xdL`}1%9gSU*jE3`IbS%+BT`j~Q;JVJ}g*Hk|0>Dlk3edZ|}j8GN(NLn0}W?$5<`-F-sJVR>_eEPj_oJ9m5D$*9QfI@kGgZbQzS?~Y*-#pT=o&GmM#}8~2QJPjX0FW7qYv!O`4&E>zWGjuR`0&UvF3z8rIeu84A*2w;CB*y_TDpWn_^D<xqi%mAtaj3V_A=0LMS^q10_V!P!AV!^;`Ky}&#y?c&n7KEodXb*Zr3pl<Hq9qlv&YVD&MG=R~E2@?+5t(^nZsAzGQ=gD6m4>@87U3<Ecj_13x0%e`i|N*p9+Fax^_XbpY@h48hqz)Q9@}>-EaQrl@QX`ed|2tIA_4RPvNeNS`ChzY;dAr?OljA^wSGOC^)eR{}dfFgY=19^nF!!rzUV*OT(c=w(ZBICxIIkLgE}fO!A(zY4Yfw`53poDBVpcE2zx3GQ(EecpnxnL$)-OoQ|WLb3R_74fIOzOhE`I*cJ<OQi{YrPC{yb9_SceMFWIPcPGS_UV{{aK5f@lFVmuz0P{~wL7_4oIHQ$aIX)SlwoH`3D9@2ZYLZZ^o2gru{0XQ!;xQ-pD@K_yVL7%AV~7rIty7K7HKI<1s`OT>x*m4WiC2?r!J&);lztHy|FsK%HBvjaQz{UNq2!fg`6kuL}R<PWpAIoP3k<W2F%aDRbE^J!Ijb{^QD)=0z`TN(K7`J*q!YSI=kr5AxmU~EK$)idN$A-1fK2mn}~{=t4qs{d3ye}oAv7pYiU^F5{W_d!%+*x5fP)0`1MQZW)sTn;4p`!^2!Pp{04D+af33rrhpY$S4Fg)OzvvSo-*DagiMx$hMua1Ey{F=OkSKcyV1d<u6r>Br3w{0UDT8kjfB>r@5o}a6Ed|a5=*EGWdQA1)kOf5g6`Bcs=r=dUR*+rYd+O$G+y;N&%Z8w)tj&(wtFH0*|2I#X0_DfFp1ZPARe~>TUi8XzviTkgfvp)nk4e#6tV!Qy0KEfzP9?el@*@_<)J22)+}FH)L+k$0gEfsoBHhX(vr9rts4pky+Dvku_>IvhND5?(j*Le+Zc~D0HW#d7DXNVI{eB3x$6}al!LjOm8FGbOt>^ci(+G^-)Xc=^A+?J597fgJ}hWphMaYllg<Vqx@Se0Z2uB@?9CJ}U8Gx5@<KrF-W2*S{Jc~$F}>B>Z-__+fGD$~`w$j<G{nfDLNk;%g>WZS%7ci~Y$v(|`fgu{u{=-QWd{`LF5BsCm+-EGW-Hh_y=@O91#hA9K2~a`w1B+*yxslb$lHnUx4f-ZtEaSD_6STuq{0F?x*b$-cZS3LxCa}*zz)}>Sq`w!4mR6qb*b1iE5jvRLn|Qt5av`pwBl8$xbEB0Zl_mZ>)_Ombz|#|qFFB^u!r%U(j!?NG$-l+5axbgtT=!lyww{`^_o-42+BEYq>@YcHPSlz8>_2uueq(C9qU$Q0$yvtN`b0BsHG1`Tc*AmDVp+5MX?JhLPGQXPDm;+`W^D*4^d;+8<p(*NWg%gA+_c;k0iX+ncB1!2-%J4Ax8oexkah>rx7qMl&MZJlSYIBCpZc0G8rTEs*Jn9p6?qJbcps5pr5y*^!V;YH2MnOaxWw-advHCUfrctxY|obuTN*4G%<JJ>$@wfw^!=rjk*evV`;)JPjNpofP%GXmMq{$17q4odqx26pXU9*J!xRNZ=Hu^*HGYJT&@^x&7N_`M@@s*>nxsP=yYzKu5Kh*AZ$fLRg#VQBY(uXHk*bk1&sxRqHucO`4AA3lcv_^v(nY6p!eV;w2O)e%e+2C6PHG&D$xQdfi-GCI!g<yn;U)}S}c{312I4MZeecE?N)z#5by0A!lzzqfCbvv)=<C<EU@{oOj`?mTSO(m9<84!k2MU6<56DtCB6~7l7adh?9!s-ygt$5;M*B5JJ&|TqSUjM{->Q}4jBU6z`@@~ucZnmW5HUN%y-CtO1lHOYgCrKUE$P$3^cK2v#3BpEB*T{3eYoQ_1cOX_b|<ke!?@S2LXG_l);Y-$N7?50C)?<Fh2jMFXiTxRyfmBi<Lwke!^poq_|1OF`P#T^!jc*yf+N%ktVGr6ALY+E;3y!9Mwumq;;yF;w5MD5j-BX7phDaHQ@plMY7Z^NjmeGS7V{PR$pIQymiYe&#ad{B|1PW0V?;B(4~c$+3EJ6{}ALr5ogRI0bbis*lRVXr5q<*f#>AiZ#&)_%0%nKolbkC`9Qb{jWyW$>gx%koy7(htwr^T;fPI0wmB^{y9!8=(dq5Ca{8PwW^>faaH~*($~I|u8D?I#dCt~?Qa0IB(JR5zc*0f9+E6RyZy5awml!^DeUH2hGYb5Y@IX<~FQuKads4xJ4jtlllUzM5SF$0g3^2KKJ)_Ck;gpUdvCs-REUEO1VQ8cdOad)A97-^J`b@l=e<e0NGs)O+fgNJJmTvlmaf(=um(0H)28Awbv~Yt7g7Kqp*7sxxb}trLS4;j8=j0>y{S)9g$q4h#-YPGh!lhScP6)1%NmtBS8tDu-Kcr>bdRKEhQp!}IoN_lS)<3`JoSm04b6ew$*|buR_DZD34QD1ZJx2<GW=0B(th5MrhJ#lGXtmr`B6oGz-4TWP4M$^|NsDDv{r{<Z_wF{1D_``hbnLa<XtD``lw?bw+Ur0hBy-G%LXfsRFjfypG)Rbe1r3s*70$|IGP(C;&g5j)+%^B)bMHET-2a_SJaL>v`Qv;6>MP7Ax%*L7UDaLP0A)EdC$mN-x~uE4t7_M-{rK&dOG|A(u~1em2FiTO%`shS1k>z{83SgKtd>|Yt9B!%)`n+q<T|%{vg69d#sn!?mTz_<3<f?O@IjtNN}i^P%b%Bh28U}QTPfTZ=^b3<4!KN7JP4%I8Q!1eHQQZII9sX$Z+$QzeOZ39%shjaXXL+lp*z-}`J^Wq3=}8<b#jnVyf`nfi^ny{z-gDb-KMx>w9*%G;V(a$%sUQH*pzkQk=625%_Ju3SNy~ge;$_+gg0)h+%Z&#VMXqI(yHT7D6()ZCX1)V>MZREuh;noo&=?_sV}@#b@8q1D_Xeu6;OjBJ6lldkR#n-Aj#iSF~i1v$19_jnNmw}5?pVg9$J@^Sb+sKWTgQ^Z=~i)sR6}sXm4Nn6})7$^!xMoGFk|`i=IIzxV#9dE}*<v90VqF>$o`^1%G7wzxbpAYqkfRxp|~aAw=Z2@>b+$n|xd}FeN){^t1}jXkTU0YU9->A+Sw?fdeuB!&E^V{fb8;#dTshhCq%dXlKsXN@8c8C(a^z<S7|me}@meM5Z@B!zXilyO4mdip>${=ZNxpLQv$NXB<>^VVq9JeoYdB7B;)k?I2rd-e?k4qQJK!%l1|SxC9x`l_e#7THi}Do#m@G7un^miX_twDw(4|X#~U?wYaD9m)^B^i?e)z8+MvCz{(b~?OWvw8P#KZ+NMl(q%?A?gTZwg@~CkDl4&Dt9M`#87OwB5<Ucl^qdAMHvH2QQh8mLEf!A3bGjf6>G?^NGl*vP3^J>q-#>Pi*s}Hw!6$i^h7*rK0&^Psjdr#?rau}Os;ai4JdCbl4Btl`*K`r7j+}Uc7aR`+Wxd!5pX{t#NG3-wPpk|2-%t}Dg0%7`~r-UcZE2t?|kzJazg1HpPR5_|=N)jaWcmmUkkZ&+mu<&;!$a9$%Sg0TGh)*O&l(w)G#W@QViNKi%%t>_2F;&Ux&(anOCD$pM@3b*Ssz=ZWqgh!_qR|S)gs8ntEQG^B$We%5Csh<k%K2HOSBf*P8Nx3FrHmk;-l-kf%Ao5AwFoT(tJcD^AV`Xcv7A*%O^Tgvy;LCg2?~Op(X9QRUD-f&C&;3@=D}!IDIhEmmn=-cps^qCC5j}w_Uc4R!*uz~O$%hZ*W->zb$9`VPSQX1*L!ocI<|&xr{|$COk2z3%3=xB6?Ypg`4e<NXx57a7$0?!9tOtJAK9918`CC8rxtc{ZW`+#=XgvGRg-M`37gv2CaBZgz(zdG1c-1FA0;xv;<}SiR4&Rz(^IjP1WKQhcZsLUxhMw7SMn=<qu<pqB$|6#$^Teh5Cm2E=u*ji__56FW5^8VY^9_I%T(H8<GGAP)haxcdAjLXPMqXDCxwn+ekJFSEu@)f3l&^p0pYbPsjKM-bz;^*W;`^4>{A2uA>DQSU-lCrjlfhIsbrE1RVy9N$8tLC`#^NBoR!_n!>Ht{{E5=*g^LSxKJ*H-rLdMEZHaC1`!zBcCRl-5AJh#J&54#3=pI~-R_IjWJ+mUyc_z`6@<ilIL|oOPB-biytBl36<~s>Gu#VBP4kxP<_xk928uxo0=vOBFK=H^JxLL&Ge1=_>LIve)gk`~LFVU-2#EJK>jLf<BlNz7ao;Bt!c5Rv5Pw(5NRjAfP=|sL(Vz}78zWu`L7qjh_b#G82(A2mxk)@mf^?s<kK^a+3Qx#Z6??jYm>3I=A4+iz1TXI1gl-bT~<nwn_s?R$wtz#toj{!)QjQePaq<9mIU}?Y^Ic~?Xi3fsi6q*P~*cif50waJEcjlNpD)81b?M=^7564=3&41T9y-zoN-LnEwIU3LE8;yh;88arK+#6L5GL-Ri+zg1v?g;TWV?ymfU605oFPyg-s&Ylzo=)0R+pe-T!+ZdyM#2dKK^?rC!VT|^GE_{-y@D|Zg=7oGaNYN!*re&+Fmk%R8FcD$G&ysQ&O0deYrd)EU+7SN(RDF9m}|=EL=4R^#f0bQ6LfBwoDVg6?jNd5-)J@F&51g2beDO$kGa1!w~&La4G_~lsx|7Q1z+a}n|r(Updx-8>>bp$gG$QZ7sP~V@A#}VJ_Q>1)6uluPw%6X{^$r+jMj#))#xWpnV;!>qS}pqpo>d|Jes*p&**aIs-u4zRH_PUlDmySlN88;=v)S9Acw|K7!L=jaQMovDM0QQX(^D_O{(sSF<AHJpI-k$1y~6%J~suEN(7A&(8nJJt9~b%!X^z|w~(+|0t$tJMn_%O1B+R#1AZ5BitdB-n3CF;IkBZz&<YlN6c__mt$ez-|1pCQg?pdW_n&NIFn?Me5v0qm6|C8yBvZwzXE1PZ8@%nRuHP)la0mipm8M5w9y3A&e8>Hv$w72n+(uRy#p3K|B-{<z_i%JR72g?dqI2L+sSO2x`Y#w1dGqD#pA*RApD9QZmX3yL+W0nbrfZ$(;l1why*Y5Es#r0c=?4g<s0XDonjvBGA)29hk^&U8kz>CHtk8Epf+OSUq}@U7)Z@{J;sAVSqs>3C?rj8yd-{0CDd~Wb#vb+<8}iQ*{SaFIgAU(_V4tLC|M$QDrw;Jg{%DL)I4~C6t-n^7j$bkg<tqY|M1a7b>UgdAh(ZlH5>Qqy0(JSbpPR6#3}h+c`7jdqccQ=_RRAC2ZYH%E(2Cr1B=^m4p)V$0C8Md{{H6d5lCV&JrYa?@(Vy!Op%S}N#F<)f3cn`2M;(2aqmqBAMnY(y4Bq(-ffSnYlob7oAvOKpFoHJ!CpGai4rWb9;^G$w#cATl>NuPj$1ET&e<C`pgHsZ{e@bAbuVh>#clf`){uKsy>R^ci=M^dVYZ^B6;w@t)KB%l69k)MtFlYCvVz<R9e1JHGoApmtw)ZwZu5UVc9o%*v*LFW%cMkRs>kh{RlI{ZW3^%tn1co6-BQ+4x-Pt=l070U@@))@BCOqG+QyER+h*&W3Ssz62VR_se(^NXW0ZCWS#P6Kq2rNE|Rp|}Viyt5|9VSbPCOXz(eN!qs*LUg%`&%38C)#4<-|P!2e{CVs;z4b9qn><=G)N*yb#I61<SOm7iJd)ePmhS*$X*KevmX)1PP;X1JWgO10`ndk&5on-)Y+nX$eFonlujwFa{ALpmUnVAkp0;P_LtKae07`Nq{^rIg9|5Uc9X=KO2@FRf)m+uq01TB_qR5mqCo!9)5q{{9YJNbHg-|UAauVjx~|s_9`Efn4g?gCMh(y6FhZYfZ$16!06NN_m|fNt3M|5N_ps~w<xO-AyNR!;ChDeoqJjKp&iihXpdxsQrQu#BS4Hqr9|XLRCsgp#?CN#CnS?_$OLQnvw4BVa9{XoeG!7?G7cNr6<H#C62X7S|5>;`Fa7Cnz*#uqbls^IK+sy(gDegQbvgpgibZnDD@zg@0PUd41nt55Nuq`vIp)^h_mV}&Yq*B4AxwpP6f}lx(va{7_Z0$Zxi>Ej-gITH5wJE`8v!K&$a&0YC`}VI&M!&P&otJMI3gG>U5co?tYw0X!y0*WY3?^UK-_@0H2u2hV!YuuJm;Sv+|K1N>uSnBBn_=Qs=q9aF`sFRb&;ICSbzH)@)T3BD_MCUc4R<QJI_b2!y(xi^qEkKUG+<fvj$>Q+$_WrQHwJ4_646zAB|b&}oAz|lo1yD3)Np=`=W`W%1k(`_Flrb=o#=cJhRz*l49l_|b0R$a_XI^l&f;b5Eg$zX>7;nzF2iN;wBI`_$EWSJ`|p=%Ah3BrSimf@r_rp_I{{jhyk>Q^GSAmtG=*9qtjH6~Y<yMX;9trM!S@<`&xW9Gd~J;gF#DX-vM9u~GnM7$_aaNuY{uD<e7!^Vs%JATw_kHM^NqGzKAV4h)Z&Tw<ZL}!zQ5T_TGuU^TYPu(@EIkVuS;aD&BL7?O6qsGv)i!Qb%gB~>J<ymHW*O|*aNXA&&R)DK0RMB7^Ht5w})~)3WG0kti2ycQO8GJ>>#$-#JLcL6w(ovAM-V6pGBb*$V!bK0z$xHe?E;SxO;#wCs%<xh&sLY(DmD$4sIkQ*l5E0$Zt$>^-nqtH04f8-l9$$bbBkc`{UELNK(S1flD_&bF%<S`nWVX;XldX>X;0>cf1=-3kT)*S3}b6r{2rQS#*V*|6vkf;6Rh$ex*|J88LtyGdMLIu8%?rq9Bt`I@_gM$&V$~Hnt>3fDeQvFn0|2AxQPza0py2Ha>^P7ZH6*Vn2B`<6tV~%Dwv-Viq!7NlX<4s}FLT=&ym;adO^oPek-*uvV`47{Gr8<Qa{}P%|@$Jfh{(ObMd-gv$v2rGzd8<m!NweQLa?Ck3k^L#6vZ44sIJ2jV@yKZ@h9dp@M&3pl|}1H|l$XmSLXDM5gos;8%zB#>{UfeKdNJlyzr^Kp3a(SE(Qi7wTOev!yTMC9`+_$<7f^l;6D-5yMtl_~3lRpNR+VT~J1X&5SwSVUGh@j_l$IcqJ40=Q0jO5V3DM?8MQetU4#X$N;<apB@TKJE28a*5`zPG=at7#+n$ROKn^O|OEL)y0-*e)Ow$@A}Nh3Q*gx?SxO^bO>~47<pVE<{<H_>Kfq}ioYu+%%j1ClTRw;H9uiz`18JAN!fgP|9+}c;!9AQHj&lUwfnxDwCXBVXT1JEd%boyRbjGeQ?;(GReZ6{s`t`$-mN4r-%r(;tk!ft?^lZYxmKx|{k%30c@TI*1j|MLcyY{Y*E``escw0+?N&cndnm*J>pk!JbmBjsMj&E<xImY_{Ci_+Rgpz0C=xvIf}X6TCKqTmD=lxGJ}$$6O4tLwkIC>UC5f_>l+p0Rur%Bnu1S&w#^MzSj6%p`TR&`+fU>HQl*P{VaD9r&5E*d+>93e5w#dryWNUZpAma-!I1zs6Q@0Yb=E-eRZK>upJE5oPIqSC>5m@*^h0hKjlRS!@BV3Smw6YQk2|Fd>Pc2V&pHl_&+m@rOD7H7B_nNnBY%Y6M6*=!%sxc>=B`nN4m!A97)UO7qdpr!W+2!$RxzzhF`|WEtd%I87c6agGme=OjH;Ilx7((B{OE?!`C5=ayAQO&P?~L#Iz{r+PGmy}j88)E&@W#h0tM-M(AI!h>e@jxED|8}qcZTzqYI>o5{to`%0TyFtPgxh-+2Lk4JibC#0>YvZui(V*_hZ6MaHb=NyiEFihkIS-`chm7xB)AuR9umcc;ID5o;835g@m%s1nn(dp@3e*1LPNi_E_)2B<@6T2!N143Jie<DA;Kh*9y)*)>%rU74jJ8G@3*VSv8D$Q~2eeiGPR*f$1j8c`(WH#X&>_S#{+|iMMPWCilVA!)k}RRLZLJE_Z|y01nQEIA>(im0ANbBQ<alaWs9R`mBM>xwdUM@0y)q1j4v<<&XW#bT_W_v*pU?iCZ7*Psj4<rT+BNpL5_`iPL7BJ!-C`XHVt~(j4YLscmm<hIRPQA@llA>Kg}p`x%7;_39EAL`kpe(tEFT`C-NLFa31hc;IWn)ymj9A1~cOJ94YMMW8;)5GvGj@EryLJMLr<>t*5}Op>xtXlZH98Z7KyQcA9<6Pj(b*||(_#^<7OaLM{*k<t9a3NHcXe3lk;$55UGJ*qgBTNW<(Am*I+F*kdB=C>yUag38pqp>?W;W^85di}^lTLlUSSeNFM%RdvZTmGH++OCs#oA`<YWLoTWkY1>CUgZc)=_QeSw&A`ZH!zzh@%cL~IQNR{hTRlsI|?atnJS{0+U#M01f%-xV}BS39jDEud2_2geJ;))w1cylgvc@C7irquBx84o?9{FD1Al1=hL#S10?ryUK~tHZHM@Z6G3^^?5`vx%$=C~j!ofX4087$;5`K(pi>)EMY$%d+nK4AkY_=tWoXurX_B>gN0spbiPZM|Yfc}(^G5DPyOVZowM9Q}U_*Z<!^lt_1b|$LAS%BK%KjO(1%*A;G&#u%1TtL)T8gG#cXrpD00A(+&9Xg8P;G0Gv32iQ!8jzU)pQAW<(WM`pW7-i_NWKP+Wmc+8$8rJV_c>ru8_y0&-0__r^nvUF5<-`a6{YxU2xl~$XAURQ6K;9HfY6=+XY)m`Lv}Mbzs26t!N({HiYOZroD(_UW^x=Yh!r|S0W@VHZl6VIa4G{v3acek2YQqvX#4&eaI)c@J8L#BukQ=|blU4PZ|A&fYn9cCzgAg;qX}wQt=Ot~x*Aqi=_UeO6lsI84j{GkM<+wvKK&Abl;T!?x+aVAr3H&I4N60AY0ZLK<FencilKlH(v@cT<|>zi{TQ7@leh$J`Pg2?A7%n3$+p%W%rj?Ou2)xMKg9nFQa67eXQ<pmzv=3ON=y6spi*u5=74fEP0bX)vXD4K%$Bb$9pAj#)4e|5tfEUhV_n>#Q+W)s5o%jlAC;hYq3Jwym!Wzcfh@c9t&m<^Sr*cZEDO{&%F@YRl)6y9tHguRS+uSa8?xI(665Hwq!8yJ#v!aMDM-=w%~u@uASQP^YCW1EM<7)R%DB$Sq}PcZ;73m4=dA5|@=!V4Z3NBIG$k4MQ(eH&L!lTlOa}5bQuBis{#5*)ndioA9sMsk4Sh5%$?n))5eXH8d}Xucmn0sHzJNqmyV7@}$i^XU{>k?YAz}$v@p1!Nz9$1(3cHGrusd2c%3&Fia9v-0Zg^!MoDU<!k8CHHznbeiE{-GEu4g{RcHj(+9J;me#nRaw;jIo)gS(tNk3%{6u+WwSds4y4#2Q?~5$P>s$tf_aTVh$|8H3M<anPKmmZw;d#@Y@72^+ys>6-i!$f(%jjFkF@2!c~t%|gpQwd~M3j9o7P$dk|>0hR-vxZddsVV)w;TkF#ax_R&;jzw=N)Eg*$VR=h9HmzKf8Hiov&7Qh5uyeA17F`8wjBwf{^tSc9HhBm{;?NS;f7}A|VE0(%YivuY@-18>77cNAb)Ne|)~(I<KRQ222b(!US$XbvE~J{3>8O*&IP8QeeSVwqvAPa*E>t#0=n4FK!wc8~txEUB?+mt3h)Ob9g`l9zWXa-l@N*wM$U^kk687>1TY>e?DuMX8N&1Htffk3XXTXQ_S4%;f5H`xi!u6*vd-yf}RS0!K!VG&m*6BZ}u58EfY=MOca%!oI=LK6$U}yFEh)Dbr=H~NvdL5i+=k(?2weylGdtG&2lBY;C^U5KK-C=m``i}c)ZGYFD(@#1z`d3wTOmVX=g#+ZbnOE&a(45d7r^t7?)NoUmCPp0cL=!_Mx)>4!CAA8bStFm#UKEg&(-;sC{)NulCROZ=TshRfS~7(=YpFxDY3pfF;%1w21GIcC`dhx!H$!fd?>bL<(^$YAaF|zHW)UEP5T}bmxlq8>Px}mzhZe77M$F(8si>Kqt;SC6VB;fwb{V>ir5=>V)|N>in`TZAmVgZPZH8WpOQUp0C|<j`sBR(NqIhHWhYyvCSA1F}GKW>NBl@_$My6DrISL7sRaDdH6d48HddHLj27z(HucwrSiD;6+x9>CuN?Q2gK#B%J+p|86L;-=dW>W%b2t(pZYL5xGT}Z{kn@$Ye3Y0y(N7JaR-6+erK@psWNXQus&~mEK{_C}U^z{}OJpuVYudBJ)0Sh{d5<PS{$NdvA_&dEAk)EbpG|DXs!(orV8eA)};qb}c_GY+UuYK%#^V(OMLR*W;OK3ynRGQO55~Lu{e9Z-NI(b9gB&?ZsOb!6n{UdpOAhIWFY%;Cq7MnCk>$}xbSDIX4o6|s{lv&%IM*sVbhv3sP?CxkBSPZHDH1dn>1gUD$whB%wnqZ5$hVaS_9F6^qVfiR1vhQZQwNN+ICfTWPZq;^eeVH6(`of%7yCusG#jd~RYZH|EP6!04V}Lv#O6If{(l^U`xMCT*kaOh|g0OJ0;ZNQ;=O`Vpfz4QIzmUg0GEHIC5qUARrxAI8#d)sx#rGz-rKJ2;a;{C;<(j@b>Hf$04O&cY!j|Cc?nK-o^iE0Y#h7$)F9<4h&(F)ALND>&Oc!TkUglny=g9l``E3?nacyR(SraO61us+Ptpo!)JqvShCNAHAXue&XaKyXm#=cDT0hAY6H(&x>GR{<C9_H6;zbLv@aWPb;Ix*eX#Wf-?59sQGWR<s<@Fmgm^LLoG^ypMR69ndR;(}9tF*@q^j@;tLxDZ`rOZlX>kWAgKjrpE`J5283>BB`LB<jma8Uqo7q6f{1pYdsXk)~L8UOYS3K5BaJaM{?sgVgsi%o3ezgS&Y!TDQF%ld8V!J^?Xifr+)wtK8o3C$6Q>-+_Pm4f2iax#<x^Y+mN3z3f&U_L76A7*10u+NXTGhjBQmaX4nL=3VYxkZ$+(^KRjMKFZv6c$Y62TxPGNyK)*>2<pUvAe^`OWs@cR!t17}LfH^3ns&alY6uU?ZewP&{=XPVgW{`y!^wmx{yXWw<Jx{Ix;u$|r(pNrP<Zw~Bthq|ZvHeM_JkO5L*e9XhV|zd7aTsT?`~qGUs3_PJETxVH6I9%f{m@Eq;ADCgO)`|wHD;0?9qM?Euay{ud$X5PeApaufs+odVWxXrP1n5ZWTEdCk5uz4ANfs{aO}VQa~qZ!}2VaiBjC_kRUESX%cT7&nH+yt%9cZcbWZx3GR#xHwoalZM!t*Ahas?2R^Z+V~G{o5##fsIIv-Uldy)8%oo`8lf+k%d?NBBiuPm@d{HJHDw}>vi(Z0SUYT7fihXy?p38e=I?gchaf5$_bG7@Z`@ls9$D=C{4^r`mmR9uTVAWB>bQ|jFLjRvUB)>fWpJABbevx?0I$uN3=CxITFi5#(PtGT19TvaC{3;mD!K{F3&*<Ff<2XE40yo}b`85|sjwy;4HPY=y@a;teLsK=PF>3KjsC!{s?EW!1>rQ3xid&^<o$KsoH+h*p<=$=fKW63#7QgN%n^o;!ZBOTHVBr&HkMKs-tj<!jAW-a^B@mQ!SlTD9#RUY-ApUqh>q1~c?IJ&if!&6KnL{s7nT-<$g+x=y8HO^(KL}QRxAu6u7B*^l9)^_B-FEbAx1vxo$P|NVFRa^XwHD-tE%LoSjxS1ld~tl4JzAP3C-f<uOnTt0ToTExl#I*Ms6y-E<N9ZIaY)03BjE|f2Mt#Os}|Bh;v$n$Z$Mj8UZNuM$A=Y_Bmx)dMFMDwMpqPeb*y}rZ6r!Q;Wn5Un{d>-o)vIVTEkT@j6?|;!CL^{$zfd!J6M8{AFneSmChHg*<CK-hES*zU97}WA2=NP<RAmb7>buR5*bpgf#~(PJv^(rKFSBa`$O3asE}_w0&q)q7K@k(n?(V=uU0P@1N(6E6U30&L^z#K?v*Q*6t2$h-tNk7{b}uB>yx_EIDA&$ced-b{oVRLLFnL$S~)nxwPMuT=@4ilCN7<gqe%~;1vtyK0v{tA#X78_llJjd3Purcb?r$nM!c~}gt_V+^f5i2meXiE$G!1Yc?37R0q_C|8-srN^c(Jaq5P49(*qkP8eo6KxAhs^Wez4$6sxdJs>bqsE}YHm#6Z`1{`9d~r~V)n0QYn*)Q-B*R*ODO;oPLWHQSvFObgx#i{j<15qe*6u$Npu;coa>dygAvGYXO@yL7%KJ$4qlnHQ5Rc$qN}Uw%O5nyp>n5otXBRP>iYT{H4_%mHZnGuK~xzvB5x1g$8e9=QJfJi){awG|&44_%+)!96YJmi>f-_k9EuB@rfAE&F(o(~l4&SXgZMEEl&z2J?PUrqEuF->;PK`uJryh^D8bPC%+_*B1f2Uznk}G|Qu-7c7|tXrrhK=VSIgF2XW+4yOd16!m)JI|Vba4d1pK7n$OFVv41{45;<sUT!pXI(YwqwmeqfFCb}AFusUY7UW6xopiyaDba$L%h&VJ5VBQ~7q6(ODB^e_%z$k5>Yc5PyK8y`@<PG$*%s70TMKLVD@CKohLtNO#V0Uv5LR8+5h9K?YQmynJupM40lxVDA)Iz_1(2-E*C?TNSvpZ``@<Kr+sf2g+x#qenJ~i?&r4=u!+4(1&II1dyE)N1!R=&rOFR-w5`Sn68~ewwXfqPdW|Ut-qXfavs!=p&t=2r0b_PN-&JC<<Yrh@>H`jpH>=~BPJ3DExEW~Y8(j_wt%?J<c%{uD9>bo0IV{zsg6q@8{y$^%CT7C81Mtui?s6{^;1bI*oQ^;_hK{ozWhwf7kRc=4y>DDf+#7}D5**9AI3e_)pTLAcHKj=daqi0w3wBal#PCtF}1LyT;_Sk%^CnqfbU1vw6Di?C|Xu?kADwl$DbVUMUqR>b<LX6-Zqw|wfXV}J*N+Q{(+c<><hT)_R3J)lH`RGjV6ADKHu1XMn(<J_vPal-h&aL=YK&_xaR7&tI5I=y>17?ICupo56)i?zlpkM=L2phoPARgFhdmNQAQP_Gc=U71pSfV+Ts7<wi4E%syh7aJthOBj0MP8<@G>KA+O>ulFB7<D69gotSa&A}{q{<qVzkp+s9A-jnk}-Wk8znzat<+~oSAK_wGlF?%zcI0|)-P;<wsv)lM%!n3k+K(x=4vM+Ur@wO3s}S8S()a!x!PKno(2q0jgd(w=)RN+cYU^U(X9F?qqjVU6sy<}l!igDDo<MY5{k#A)dEb4cb&#yG@72mHk3hY4r23gR;SCTedffcaBn@sd<)ajWqShq6-39rBc^DbTn2H2rM6S|d?*dsm)Qtjiqb7T`!AO(7Z~rh41)7riqu^;4~|dkcm{^zTt$pN^ccLPqUnVj2kI7Bg4X?61!uW?4l^_KSIc)0)l5jsjqS$wpSk|g=zQ3TgJ!8xh5{bcZ#h~SXb;#3CZ=^Ma<b}1KY-I78iCcF86)-3TY(C%{n41gl`xWdhT@F?TIo(vpS&8aJiN9=P-Ec%R8#S_bqXUa&lvG8hDMH$K_FHU$k7$7Rj(-*Q_Ht9A)z-hL$k%A25u<=n;!+0Ds>}WX}YJEcf=Q&R&bs}GYqnH;lq4;ALhJ*)vem`mXR`+>xQ*gxQYq?kgf=LZiuysPBd{e#84Pmq<PyLM}PStY$tZhfW0OL1w8l>_Cwp%VmUUpP|A27KHlH^xW1c%#By15Oo!;&0=qLrJ1DuIq}f{B>)7J3*R2j>>kQA8_U>TU&N+UabG$=SZlB8A6Jtf1Z{!229c9xANMiP6%To8)xiE?(S(CK`j>Dk#!p2Sw!>;n1i{{rwPjf4iPN9DR=T+4s0qwd`J8XdPxVOJq-_KcjX7B#P{Z%#bU-;xL;H}!ZxP`Z>YT;>p@9BQ+*+(k1sjVe+u?nnlO6zz^gff%~pUri|T36Zy`K|swwU<?XHyu@2meyc|*=;NZ7kq21V4=oWt$L5vs!Fbt3CGeoXJp1P7uWui`lq3EtsvPEp+2tD2too`&xq&-b^sz`))1}5A<Iixvud98JGJMq$agolHqok;ToY~1iutaDlOXpRWX4<>cc2{GG>Sq3p~rS?X)#ulrY;rd1{w(RNGo&+8$fSEf0Fk_=BBxj1$z#LMS=H@F8KRNFcft(jbGtqVxU|la`Q7F3ta*VfkR!ksnB%-!OoZmr*ahw9;TKrS*&a^Q@3&j`%gk$0U<M=zHga`Xk$ae91y@=&U%IA?h14{1JaS~{}RY^edpyV{67rJ?p{_!=QQchwpMAOznLwTOmHr-OK!)vw$cCU;G-HG?4Rz@XeA2?VVG&|B{w3PbW0kq|Dsy@qFHJAU%=H2nEZ4!4n)yr6%S^UWhue%W096e_B_p!C4x}qhO4m<r+Lv_Yngn6_$H?oAB~y4!X;-$aZ}8fF1;l<5+_&oPcMCXi4}a&8<#|@K2+vq(BUtTy-BE&q+e;h#){}e>iZu&tOV)`1UJMXA#I3S(IQe>E&GG^^t5~&_4=ifSkcR?-cqIf(7PjkCwH_{=qlWo(*4BKHZY=1fTy=C4rA+sXmAuw@F-eGag?wdZO2kcY(wJ1p+tL<Bm+`O{3n~-RCL^6SM!u^maQfFn{Xh5R|v!zan0mko>I$96zYxC#=sN$V<8`4p|ol8A1%_{FcVh{T;o>a4;y=sgjPs3sbet}kAg@csNw5J!YepHHUT7uEr=U%T`!STq-$?vi@0g6xQ!D^t2-z$d7F}ut)&Z^eSXX?2{_+k7JON+H(xJgfORHHYV^p;*8!oh>5{2kSlW?gx$8C@Fr%0-mdLe5m{+E>XL0}GLE#X^@T@DD(pocBE_^Jg_JE*;FTVEpijM_iQKM(Y=1*};c3Ry5mRa9N{l9&%2)ZICOGUgIgPzccWf@-P>0vYYn$(l7Q_@hy-KJ44ccaOC6aYIj>$9uzzaw$MB35xA1pn7}p?k<zxfJ~H-DoiGQRoum@5m&nl40=pS^sABy>r_aC*ftUgBHEDC90A+F&wQV9Mv0Jn};y$?;qB74?tuHYfpd>@EnDRkLri}TZkqECo*St>?45kLt8_HIo4M_3{TDxjA<H0!nsesi$v&C(e69(X|GFI8<+^9H%0Vr0cit$TPLL!SL=|=*$SE$__rd&byi^r3om&4GdEvE=D8G0GYLnrRVQ|wTScvk+zq^^we5@<|HL5$B`2uXEDaJ?=Rfc)pIBN`EPkOoj}!!~%ZXOuoVt>2=rgOeRg^s>ke(qT*x)&reLb&|)F8!aQ)$g@{gy+rpI!DrO8+F(V4uft*dWnx<t_HgylnZ#6+^pD3`P2M>Ex?UqHDW)Djg43wR&s@$_<}j5VB?R&8I3-O>Q1}sNZ6*=h)cA=BATK3r<Tyvy_x40@mo<q>ZJy5(*irH_K}VU(3$ri{g9^s52m^uIvu6?ix3}GR#s*?=9hSsEZHGxUH@!+|Vr%<EU^T7%6vJ5D=FNg3HdLiaIkQ&_sM0rJfZl^vV9-U)OipVSrAm`B7O#{69O**JQQG4^R{LS3YcP7J;mkw2M9-8SjWBzp5<h`V~E=t9$d`F*NDs%bP#m{2DWKR?~?}zr<9TukggoWXVO71d3o`XEcn`35ZS13R~zf+A7k9iqk+x%TRU;&1fO46ZxhIsp-o)R!F(aVoUMfnocOw+IOWCO7t-l5vb&!d7*)8_baU;C5>7tj%1Peb+5DR-f@??&s^i&hAMEo%8w8<m(Oyrnk`~mzUF88TGsOhSDlw;L6I9?PaQmGpHPBqba$qZJZ+0OBAF{Dm;FCp|LW#Xu9wN$p0hi1+evRj2a|2X?)$^*e|-IOY=a^JDZejn&fimF*Wc$k7nxx#8Fi(Xbs}oye4<vfttEM>$@|#2^mw=DUT`@cw}Cw#jYfUxiIwWMaBAUcB~d?Hrp!A96UVVkBCg$@)EtVh$ts?5idIp-S@oSB03muhC!~O)3XtjdBQ)#0L_0`fcRzmzTloAPk8R7U!xlOS5FvA!uRE@@Z0kf4h9K=K*Bq!oPESIY1=qsf5xZ6zo^hD)J)b5iO>2FIkx0EMRNLuMR=AECAG;qNo%i}3Xl*ze5{d>V`E@u^As(0Ogf)eV1_!eaQwRWvsWJJ4Y4KhXdHBx8mdIf{8XWb8QAZERDBmLN^)D3DIzPY+4v+9ZbC8+rp9(Zf?9hh?&khg5RNxr~8s)~i)Py^oh*_~Lb{tcD7CS$5aM!v1+>Ipn{i55*n(H2<cWogNYIjap&6_2quiU+VeD8i|UQ&8ltT-wC159NOC&rH`+8on}Q*1fqq~{diJ|`xBMe^QPaBO0R_FqxXcTOb#51cF=DxgH_T#mAgQ}TD4Tzd1*m}H(4pZ^<Xfu~IA*p1h}z%ceNDGU3TR3(U*lwKXhOUe!ZyVpO(tmcjoKmWuz<SBR_^Qf!5?l*sY{ZkqObOS5D`2%Hm$B_A7P?tqAcrD315aXYfO#5|`jh^dEhWHz-jdHE?LHqibA`YKO^si9z6?H~_{6|c`{s$=cbMX_7swp~tjeQcEp&LIDt^OMcCjTfB)@P>!iS~-p)qi#K4P|iu5z~Wl08j?M`TvkoIDDTI)Bl(f&Ht80_N5$^${hdiKl*2&6;0VIDg?d#6*c!;=naY*NZGHU7*_oHXP8MIn*UcK*MG-O0G)}tAn<%GX5~wPU+`7>p73?mA;)_-LZTIlNkj}`->Gfai56ec>Y>@dsl`O|NHys9FKJ~!m(;@Zou{k+>tCL((QHyPVnV(&aTIP|t-zW<iv3YM_>yTwjd6X@`#;KhVO;+NPrl;mg!M-B!WGR+hp2^1aA`?FEnr`N2hGA<EG_Y#;XZv0-T#`X0k87+$v>8j?eU72JI%RThP2Y5G%W?1%YRJpAHJg5!?FBG{Ux?1E()T`@2KanU<x(|-sHY^K$`q1(2R*_bU{}7jYEs&8z@Iqr5RJ*%q(ijGGM2R+5I)NtksE)5@E#FdnI^)*h8;>=Fl4cntRP`i-w)-(AR$lufM^$%&fxK|LyhP!E4$xIEQvJ(iXt_zQHjNrpBwAU&F-xT<%Hi#&2c4_=Bcckb)@ur5Yzx{#!wrud!=1&p>=yhjPp~>>n$>A=fWW%U8q|lO-gDeQKp#9G1DXbg@>xTU}b>s=wlj9pX)Yrg~suBf*f&f0X@_s|iL!ahdB5h)8nz6frjH)gOK5qfvi=pZ`JdB0FP4x_#|D{%GaN=DsW~*5_}SQ8A-@{WE09&@pTS=|mKz0|B=eu27tRK_4-Z&fg6Grabp(#c*kHf--giJ9M89r|@=&3OMb^J-XZN96$KrZtjr_GEMtP{e%zNaTK0Udie3_bUKc!ckc8@Ahe#2;%W6k<w50+m?(HpQZ$sjzeZ24ef=0Wn|9t|p?8bkJzVYFzjtr#!ODZq-TN!|9<Fv*+8?alU5OqX-@E&ud+(_G!I6Ia^S3dc|E(N>z^OdM=-rLIot?ei6%fB1jQS>t#dp}#Zsjq?^*WFD_6`~c`?Y5<>3iEdcVH?76Q}E3jm4Z~K_*vPjLmMKehKfh#j*$Q4(7bPIdFsDd}&;0F)Uq1y=?F8slPY&_V?8ftyAh5^nSPTq`sepa(P-$s@d9kmIT$H9$5U{_^6h=c%nTADRJvy>k~w}l+Q@H5q?&$S=vIc-BZ{$VY~C<Jf4=uX`vp$Qfx`OMw3p&Zzd<?X#(%z>G_oWm<QwO6<vn$skc5kA7UtJP$_5CZ<F@$%zt5G?NQ}cp9E16`Z9{Yg#$)~JBAH5U$nd>)%prH;Ca+HcoT@vth!7nlFB1_`X+?AHC<M7=)&pY(Yy)Q7f$P<KDLqPMS!h0@za*79O)xzWztC2bqDJ{TV8!rBB8-5gqzv4UHZZw^*jDBy7WoNWQUMcPcMQmn$1+|CgWNts%W*!oylm7&Y-F6f+)DbHS<JZOb~F<g*{WD&s7Mh3cEWg*XKP_{?m%R2>)KR=!?-~oJ!iuoMsyrL9=wygwD5CPMX+%ah0Q9QXjb?$_|u#Lp`t*N-6))Q&_kmb-3O;LvdAHG(izf#(J+42S$<|*0l>F_$8h=?NPdc3m|1s>NKIE7WIIFc2=Q3nMY|%(O#gtP+>)Wvr?|C-S;cyhacQ;WrOZU&`W2dJ?W1e_DNackY;DhfjrBr&hcn6VdEW|%bjshaXRN1)013uP_1(eGbnPCm($Jc6R||L%#}WDxxRxgPb=MiZyY=AF&rQ>jGBv_HeBa>7~F^FG<gM)oWdv%P?Wbr(sVf;{BSa4`ZK6$DOm_Bs~D@(@_fpVWSIwFq$p+GP0A?*U9}OH2#t%-#8SsI=f)P!G7d_s%cz%-u~@1W@52Cp)2j2NN&*-Kb!vGvFSL%B0NZuxQ{O%Ah5gZ`^3~{(H5#9y6P)&V)$T{@j>F^rIihNWI4qnF`O(;81gaA@JL+qge02u{-~osZ#P94brmmVNN_^y21@j==uX+bn(BIc~43aI8kIaYdi*O-B3e{2T$GMYaXywI$O>s^dEuNZkE*0hnqi{pF_IP8n{^aRLuD7&iGR|sqt+KY(T3UU7u4kN^;1V-X0(@Jp(5G(L8^ChEQ2E=c%~gJ-8)QOQX*%QQ?)#Z~v(@<zAL_`(7D&XHvOA$B54o)hUJ5aTK(fMGuwN<^$F#aeW$1mfB3eOUNe9n4tTwg}F?c@Q+64)$zPnK;U*W_3x@*sQg;Jv><yAu^xl)sTC*e8m2{5PuzAngj)kV07;Ym)ytNBV-3EC6r=vRN6gWT6NpsX{1(k;&()<l^`-Vi9f0D~~e=P3~g3C2r~NS)`mhT^hfaJ}o)*j<W3AEKDDAY-E5<I{FbQ;6tn&g1%%z5Tk_r*LHu`F+>LYmE5Dg=r}CVLKo9#%<_D&&VSsn?9M0g9HK#sktoSm<{Qq+&(!enf@(7DV=f=o!8n!*Q1DPAUbLDOBU1RWv>$j&~$m+yBJLcVi%#M2^ZLawdxE*5Tx6gTR$)Wa;uCPhEYq2MOe`)%~Y>r_n?&dKnSd3cW;*#r9<hp_=SO)Mc0>HNJl)9G1|2)>3SsoV7aR8D~C_4oFk)})s@wjul-)Fw(^pvQ2omR2zc%eCT85Z>w1)a3ID>Xk=Vq9@V$D#*}AOx`}!IMjDgk;pU)pB0m5rthmptu+pyKp7`8x#1agspakz!laFzoHzWa3R;G@06gV0?*$$p;*wd!H`bbs&g87&e0TS>AUkWX<VAS^j-yqcTi21<$z%2NOz<A>U#IZ4LQ9%Gzr_>H>N(7~my2Msfu)$*yA+b^uP0JM2Bx(vP}9}EqkS|NABa6<CU#BsSWF#d%joL5l-ZAM=&yUxG={XaSGGJfk4_QizIC9d2mZ8qakYhF@0LZf2Suz(1zH_DiSGxbPim>~Z0-aK(9^|kQW5sB`+_foHAPKx04)-@-<`1IGUzfUBCkPU;#9R>0Wsc;&Y{NlrkSAguUbW+N>XK4x6m%p@xRoc92&>JooH}NJiKucr}XQAtOs4S{+DK7=;qLKph>a!>s2QDlU%<N%3DsV|)AM@BE=}V#$4t2nSc3nVGiUHBeh4<n8s-c>D6A!84Nzd$EdFQC&7~G+{ujo1AFUscmEdAZD3la^Mu)e|6v(TlPmpVHok}OCusj~cw6HE46WaviU{J~VG;hfsPC53Sv+g6>v%!_E4Qq6wH-gr1*z9{0IAOG^c&2mFNlX9w(oG0ryg0Iq)z-LXk;iCeYQ;84T!*zc8gRt9~gh!|9c@FX~h^Hia;#b=0wp<euS4@u!#uA}5CogL+4#nX`r~`yGR=6vAv!L1IPaQ)1>`(+ed4^H{@F44@PD#w-m$a0hysuZ3^I;q$_0N2p(gZ!MNq<nB;w9-xHC3aSB-2=sUl+-oMittO>S}e{k`R|-p8LzwXcCo}fUHe|<L^D|*`rExVU5f_J>gOC-?BU2qe{7w1C8m*d-NyuaBJt;{@y3`o%-&9>!%uGZPud-x?(=vuWfRoQP|lMc~-cbZD7dU{tg0FcsBPv!Y=-#w!OtQC`E&S`K3K{E%lGIh7rSOIzjN<b*a7NtLw(rI@?2aDxjSV0d4!E2`@go$6U4>DSx?Lkfx&{;h70-%N@QG;Mz~=KlL`Y_x5O^Hi>MK$C2+7LL_K6hxe$25L2P##JD4=i-VtN_J!uc(fLiWN1IZA5Eo8)*cmc_x4v@-_tULsQ25}pXK6=NdHQK}CyYH+HpTIC(wyJ?kDIT^QvPFhdouk=(6;|OGa1LPlutk!+quf_8SD-GHM^QnY7fob<>#+|_4+5KL)1>KerJEDQ7_xv3q_WV3^skSyPyKTkqFRLyIJSWa!1&`0>@)rF=o|njmI*a3^D$J=a?uZ<8fI_EX)P7pe5a2zdvl3Dqi^*xNQlTR5S@Hxh+MV5VcU)*f@34F2R$eJ1P4~yMK&d$a9V<W>F%gnVFt%08aUQt&(5#B_@4^PX)lP{8GUSBrstJfhHl*zVJM+xY{AqsoxmT`~f|Q&j%7@xTH@I6gVGPdYY6M)K#BJYKt#Znl1Q+LU2iH#a<>WteOZ7X(<q=qUq|xP;HTNIyt$pWI5?v`D6;q#CF`I3W{FeNj>&i97=a*Fb>4F0<y!Np=N!!#LnAJYJ~RfWa;SX=rV9eh=5GW!(InrgIx+!8DX>vy1E(=$eFHE_Gymt?RC3Aq~S4=tOT#h@MZ=7n_99S*{5t(;i}B33twd_!-x5`4K3AhSZk)DVSX+8TFB{}UL;es{?gitl^0{v-)_1s_L7j6lk3~2%(oKXQb`d#{1naYHVE{gjygoTrtDLTB!{EX^wibxD>JP#Sr+ep@$9u2+DrhPq4lYf*J@cG81}UygF0-&-`1`~{5_C!x?cy0abs(H%Mb@C+LXv4LOD3L%ED2)F8Y6S8VBeP##%tP@oaanW5G+Y;ZlR-v0dlD&2iKI{Qe639XChD*+8>BAFI%t_Q`OBW|-JHhanL{mxDpLBs5@T1!Kk4wjG(O4UrhJLztQ&c1mdZhZ<H+N5?D!sgPc;Y)>-XoTlRckg!I~o;AH?IvfxRnZj#2LTEZMa0r_Q3Sk02n7ZJ01G=uE7*|H4A>MXNPb(;f9~|!28_Q^uCuO>Nq!6Y3X3MjIW%PWsry8zS-EQ9XcX~-^d);ocx4c?y6{y^Y5zs)_^Ev!^>ghbiJ0}sq9LpFp2j63s;qkbCRWc;9tng*`5=!Z*xO(r={R;LDouc>&Hr7f~@c7~>HlI9&&*t=g66n>TOG-(rNkRX_W1z&KJ#_OstK+Xs66j9EldP0KC{O|@m%~Jjl#@F%KQm^L@<_%Ahc&l~Go*e$hI%eX3gauJyP$}eXR7;Z;Ne87m?nlY2|HgG&YtF_LJcd2k8BM=5`ha3Lti_iF0l)Q($He6E+T~OC3fJGG3dB`9%FVF%IX!TID{FW?{i`o#+ouA%X2hHlq*-+q0*OIU$b|5Im-%ohHO=F2!tTj9uf}eIvwBIV_^YLrBs<t{O1z#JT&pnlTWn-x5-0>P0FA=<NlE-AM7><L)_Xy6da6J65yj2jNV80Dmg<Yp)n#L8*-eq`>^SrBhrORc4Vj=r1&b7YJL^~4az;zM{!#U-1RER$IVfN(=%klX*`1S^AykXzy?-v#f}aMKDw8aff`O9I-$<#*E#-;bB7c>|DMTJTLNq{H39NWu!BrW1jwAciYm7vE#K8u|LzC=2dhPj>@R|uq~#Y)=zS~C1<@;yqsc(v)R(wyLUFG!9W3@&^0s`~kTj9u0=lB6C9eVLXZ=PRc(HNtfa@0p!=t3t8y;f}yi&BwYch`xkzeCZsF6r37f(A$v4Up!x*}4)xYZ$42~!i%ZB}x_5SZa;D3e?n4chD-?>bM=JhwuQ+30AG$kU^vIGS9*s$^+RfTOHCV?370!!wpqupM&NggYbpJGI$u5*>-*wbf&0-X74kfhG6pY6wjA=^Fiem;Swn3D#h1+{;`2vh%<bCcFseN)hstL@ao;x;95Fh&m_C(!?ja`>U)EZ}k1%FdhTvDq&#4>Z<1z3o}&he4(U$e0)AQC*MSn9o8r9%aASvMSGr|&Xun@x3p5Y`THaPv=>mx=2`V@d978HojN{CXnqt;FQW)oKh)sE-sqpll*oxMb%mx1tVy7{ug)U`vHw1{OXdjtRPp}VX)x=nAAROt{SiFpKTmrF-54GR<~~j__)~8s^`!3=F*$9<mIJ80z+*&ArUKkO^MQ=Zr#<h^S{X^{YjsiZaYC@7<t6Jam~Y(O(Igpon;q*)&5Pz;4cAlF)RHKZ0Tj4rD^Nrc&Z{`+bvpfshBX~c+kJmH)N7~~n|WpA%y-G%r9UxZD4iv;F?2lak51y_Hu7RI3>SWcOhp7Eq7kj|Zw)DTIGo>{<m@bX3l<!bFGWbqI>fPqAR4m3o8j&`_%VdB!fF37%>7O+0a&r3s$@ZL6IsL*=jBQ3VLVgHd@QCZN6^=jGWL|rkC_jni(Uuy7F7qsqSy$92Tc2--Rn~cOen`$D1AJ+PNRMY1AiU7k(Z0oAd1X~n9{_2K>=8cW;nEh@}z1Yk#(KuPNNVlpAXe279>1o6E~`^&hf#83(mqfo8)%$y|mf`uhkl^WpexBbYiwtV_pH<&c{I)37q0->OM13HzmX@=lc=<rQd6EE;Opq_F1IvHq?f8y-_iDhz3tD1NxV*6jl5-=isAEh`5HznDP(NMxdL~q?BOW=JZGS=HmC1R!64j<*w!X3dA5z(8ewC2R`bOysD^<|I@6r3LDje+xSUNdqOS9WjOgy5a<MCBFU^kAmI`r@t@+{#&WT~@Nub|YAdcYCvTGeKQ(D@#QPHrVg=W_h=v4sH}sfoYNT7m!E2H{>T|>-RQ(rlk<AA$&6n$Oy(OYOTJ_sQHd2nddaCmR=0EQMAV$O@JS$g<j(}xX$JIhC^ek_p*~mK?)Kq0+FzC(qYw2vxsoJawm9rU>=`7%pMg6`^a-6E_dfnj!N0<)^&3K&BcZ!*R^Ay?#Nk+Em<!I%yeFe9j8EWZS$y9alm_p~y_EE3jn_fA09O0*<E5Lnch#{XsL#3TgPwF!7yCJ~m<nF~7oc4q<3(Mp7FzU-A<r)6i+is*RO0`{J|L}O;skTM6P-@ER{3?p`_aeSsHxg<?q?|QjP)dPLB7BY#-D*3UMA+ltEvQ3b529kWhkk}AQVS$p8F5CKS#Gt0KgpuqLqNrf5vNwMRf?J%tt2Oe<p*=K1Gv}LOjwr5ux7bM(lljlElg{=Td(btFwjWlDqa_Hex}cd&vm4tqovSd#ic)C!QkmCUEAr_PM3E4=f2fpvfDKh%%todm@4-KjUHOU8CxVlH<J{hTcs>CD(MVek9-kxZL?m>EU?|G8lqB3buqZnIvI%!x5yOiVw#$8WABsN{#Nav-nb3nr58hy6Sm>{Ywm^Cvg&F&B^uw^v2yz^Q<IYn_7X^&a;Lep<^0KHU&~R(5JW}{ua%>1GDy$JG^JEDnW~$YNuni=isM?F{2y3LB^3t_vrykXKrDs5#|=aYJm6q(uSM4SnODSb%I9PBj3_0XA&@NN5;#V}A?UH*tHkVN5eO735ufQ<s<aOUPZ5Qzr19)2N-n)e_bL@mnrG?Lqc!o)+VG<X{L<Rsqk9W2e_%L=Po8XTY@ua&V{do2zHxwA2kIg5wA<{4cb!dGNKV31DHr7Ah_kiw=*U>v0|8SV(@?h8!Ay`B?MV-%Xfi5^BfS%$um%;(%D0M=Oq*m(*@vU^F2%B%yRouloRQrGuR*xOJj^EvBHZd&iktM*nrxoO8q8K^YFl0EiOS$;9mWVbUlKlIMnP?emZHluA=+A*sBD`w+unz((aMABb7UE_%))%m*cl70A56uZFwd~|xUskY7#7gu&%!5L&k;%Q$rD&ow_>ZZtGzFx&Ku0!b4$nD0gg3)MzAR)Mth~b=)|-apP06FVh{%;V0Co6e^MT_r<2|+#HS^_TN$+O>8KxILdh&g!EYZ?zAXLwf{U99Xl~}j0(Kb{C0kq0E%3NK?68lTta{l>_r&}@XY@@~V=JE!4Ed@`7F|(Qz_XMiEW717EEZX=Wjk$}rsUY7=ASD<f2$3(K|V0(vW$+`bS~aX7VHCCf(gAWCtjDxEIpFO>;cBJ;5mu;;@gzRCjFP<Z|Qnrar4d$>v?9Xc;muXJLmmrZwr5sqfOYQ8ZHrPKW=#{S~?+SLBFp8(c@-Ht}@d!TOFV6&=2B<=vLu1CnLZtzzJRKIVwRCy%r;sMQsequ4H4M0(ng=UHLO#upMzkl3?uPp=my$_&|BW@*{$Upgkmv@|I2An+9wo!PMi&qYH}Y0_l)Pj&6zK&o~eWAC^^%b}fT~%{RF3x#Ko|OZ4vAwJ|8PcDp@_)08S5=y#dEjPaDkJDc#14%3sElvic=8+%(q;b*iX6i3vMvK8o(rM;?QXyO#`H#uUbtGcYp<;a=##(IWNj<iphx{)XKb=^Iu4UFr!*JckLxOX~IXJu<4$uz1cFPMUt$IzrWF2ZU>WDPU9z{SLRpVmG@A(OL%qVJ(Vf0Ae+rYmDoJEuW_^=IXN3j_3L`7Wj(sz2YTKO=9-#jxyiJ2;0?8}@_wx&z&hGB0-mdQ)c8JX`Gs={(NgOFLewGHS1WFIdex@N;bnG)X^b2YGs@>=%j7O`gs_XxoKZyylZVmd{mz$ybhMQ_2>HZ{P$X8!$}q@o3VCjVLe+NJF;qMUqA8`kTYugwDG1mufcIWzv{gZgZ>lbaxLEU^OxySe3QrH))-%`F3HIt>PX}>W<>kPU791s`lD<+UK1fI-azrQ=E~c-1lpyYxBEKJ&A~8%VL{Dr<~G9cgt&@FZ8y@(s2duV9@P56o2aIe#gS8@VMXZ4a#~8yQASsdver+fj}Ei->-tuH)wO6Fes@PMn9SMNpz<-M8_DsqMnY<W6B72d<>Euqoa(D&&p{%s<1Z0whTKXL?YwZAfJC^1Z^UmP@jbuNYK=O@GSxz*69xmh=Ea>+hARaZ0|jz1ZS@ios_yZi(ZiT1<CjJ{Zh@jN8)Q?OOwsGt*oN3N#jviemUP+vCT}}zo+kq=;|&j4r~)Z>dlhNbBR#}Ftg^Chgv?`J8U4AE3Rk_D-uf~<fTAC9sGb+g!UWwAEzatoCv06IMs3TF3xr%9hkT3pR5I`PyOs?^X<&h$=k3r7g~+c7gpH=q+1Vr&Pt=ORcXiWJm}Zr!pL~mLP=7ME4a2KK`3COn}<{@&#F=zV`@KV8P#W2Q#1DqTXND#le*XnlD=Sm^^?_>x0HIc)<SyN%17Dq5tX-}?Xpc%d7hwXvLM9<^EWc|Q-ZM&|1ZD;Te?uuOe%XUdohbO#;=sUe_8ZU2kGH5EHmcX-6v5u5P3TE40HVHgyQ1hao0WV#n7)LJhcB3Tl)MR$^D)kKBkNkD8Fsjo__vL>pHj1<MWBFO(McuZ!C>SYzrC4n!`=Yd7ehY<!dU22>|DizaGP{HQyz0Y66-78dFpjUc0pY#V>&v@?~x_8rWn~*NXOKCLd*PRZ6O=KYzBpw_iKh+y5-og8wOazjoNDZQt64t_@Ad>Zjh~PE;D2b;^w@HGvs13`Li;@TFI`py>57^WC*6yR?VLutd-=mssW<I_8}J^)E_vNdTkbEV?2SHG~u4P(()1Z?ZdLWZBk}ElfY<?v5N8YT7|}n)VdKz8sEI07;OrVDT+TWYG)+GGG8coo)n>fO($1V$Aw?RZ=FzAphiExl+MDRC+P=5b&YQrhwP>S+;PV)%SM}4`|mB1o@q3o6qiSZPX=VxzjkPJ>J@eFO0VOeMfWIh>_#KFs36k-hUAd<;@sP7Ic<rF4TyPFJ#cN^Gw~KbNvQX1{ni1+<dydw;@tsW#8GH4pa#iu#gDZuYnTThg%!<oOU+0a$YD3=DfGTtE&<kD7)!1p}F?qehp0@KiifL(-$=QCP2}U9rt8XAV>G$>$ZFSs3R}xVT8^x_ifHGDVaWEH!+N;E)D#de{_--kyuYRP)=m)BGb5*Pj{UXGEpSqEMFNczJp8m*5*@d)>2rGV^-1;Y~uQ9)b5Di_;Wflu+X8fW^$N$5*vMbR8S1YXH9d?3FO<uV0y$h3xYJRVB%}NHtbOub#q~EEf+y)I-@MI-qP`C{ixR_FE`AY#Qtic16&s|Zk+va89(_i$0M8Mg`8Uum~u-k!$Cdh#X@QnVelbkQ9FZsXX0Xyqi<HokSq$7rQi~hYWqUD%za3v1gxETcG6m|w&%jf(UId^?7Owy+V*D$7;qM{dce(J0`=!0w}I?duQwK12;|zBEQHxY3*j$G#@XzR=m=lZ?nWeH+~8GU!tlxd`4H9K<ozW4(li}Q_+)QC)Cp*n6~_G_v^awzC4G7Q8%GoP?-3mi0lBP&<M#A4R1g8Ss<V!p*qG@!S}+H4KO9X)?apx<gw5%wl&%c%t<ea@6^gphJcBeYt*oNu!RO6XusYc(RbQe=ukcyz;G<AQl@~11=TF&}*mh@Rt~)z98ja#<_I+m%h<D}99kTGj9ZQm(mto%4YG3NKWs?QMyisj1gPw}==nP114@vmw9kF>NpYp<m78)@9XFXr7*yUBv$6<@35}~NK9yh{I_xIrbO}?~6%IWrHv+5JxE&qvsVYMJw5k3_ySE{Cf=*n%ox*%;-%!@&cw?vEW|4>>eZ+4rn8fA5CPia{wr_0f(e~Tq07i1EGQNkLK3~U^00Bt^@ZRCD?TFTM<<q5hVk2U-diOagnJQta8mK^TN!5?qTUKe4l_IWT_UYfloF5&r1HqBl=$zKJPi&<w%s>FQ5nj*!YrG?7z*y^%5V6Tglrb=3Fi>yR6D^?^7BnXb%TGCgz5>~Y!XhAnS;ZTuwJ}8aHzL5J0hbAYO6ul+!%MzY1b<sMi+|yM;7#Ho*PX!T&3s7Oexo^Xx)!vDdWmoIN$wK`WmRYUq!fwglPMe22J6j-&EL0cim~<jkU{UStJ>9Q8`)F$;<n#fb<<^x({Lrc)JDzoh-Mxd4>ihXi7JG)lZ6HWa+agpsDn0i(e;989Tb`kLiEv<7ky8h0h=mUZ*IAgJsMRvOQPAhX<|r(9ua=gCU4oud0L<Z&Cm=2&i1<M*tij2;i@=h=;|O-q&E&!PE6k?iXkpqY_)_?A%=vHnLXW^6I!@F+J|&^LBd3OO88ex0J|*Hx>_{A-!-U4LhoSRfu(l5Dne&^72xz6TK<k8aK0H3fXyJ|}hgIk^NTteZl#4%X5I6;BHy5lGGa6|;hQxZdT=^Y3Eis8rF0eFw?L8`r<Bm_yyWM_7Nqcng0{qbGz9%pyI_MA(%Ud@c>$GIxG@_%o1*?xriU$1W1!Xmi1nVY(mI@i8)Fwz|4`lwCTq3;=n@N<O7w&fqmCn`^;Xt!~%D<H@em!M&N!eR69_n#mo+NOnAy7PxJze}=XOqJF&>D)sF43%gJeBY+)Si>*#3=!kc!xUO9z#lzM8>@*?OAH6hW_*CWMxPLqogrW(p|R365bleWck!UGI{>II1Y@mLt%`fnVw#YPV_}J-dsGC=P5W8J7ZhNoRdKYJ7p0rc+@^V!>u^%1c>2Ep=+g~rTdR-8y}-Cad%T61jW`c^#~6?3hbGKjD-TS5_F9gON@boIctS|SSU_-*wH$gSsd84)bdD5zwt7W=Yy%53s?t&O{!-iU{a&d#@_bf&ThE5^+cpEax;Mvbk(B7qO=mfh$f@Bl#%|GX6^{X#gc-$Hq2WeorPjRK_$iH+XFkI1c{9F%|db`)D!S~O?4^2Kjm{NPTt0X`22Y|2%e`LMQ6ETud$EjsSAV`AgzQz?J`xd)Eysdrj;=_DmJ#=y{LXqPVG*8=W%^M+^NC6>v{f#i`l(0??)Z|)RR5uIa}t2lLT<~+B!YVoM&7?1+~*sI*vF0qRDk(ioG%1Rj;r@#1Iq~ZKKaC+@@DdH?6Wq7Cv{dw_V@I-)QC2NtoDY_=Z>*RxBkWBeIxhrodfWuK70$D5e$VR`pK1bCVQs6f!idp;3rGe3v90w_0%hnOlExnDT9clx1zghd{4T)ubwNVX|Sl^&^3I>Q)8B?VJ{dazi3`hS*d@B)7`+O~RrVx8Y{*Ww<VoOI8hT3PZq$>~5+^+*%H}?@MZ!bFlG?$M`OW*&@WQBedX-QgFu!caNMGcrQSw+7X_2y@@09U_g2JzFgc&&LuFbcrQ$1p>+7-JBm3|^&*yX2`{y1AMn=aSA)KkqNYdcpvugFEOINtE8^P}O^3S(U;1=+>#q&FBHsR`hXt?{G2iw!=oD2Q%4N>>kF7@6H)Jx%!<k_e!ja8@X&|P`1$uxZ0}u%xMVi@!2a`v*Wz#9Y36T&{--LOaERiObgVlw7DgM*bjA6Ow(J2?IS2iI1c<1?gufAVt<xx1nZV)1vfNi)#N1~-7-=(w!Zq_oKOxE`Tj^ffKIi)8`iG@O}uqYeT<+@!BKP!<Kcq18mUKeJR3}Kh4vD(F?2I<|Q%MDr7;_qgqxu4e2pn~y;q8GM>QLy50n77)ONsB^SY8>ul&cXuGh&i!E<#9T)Cg~AtXx0?9C4jm@!9y3u5AWw}Zmd?M`t6h((WHk;c2$?hAiK${^a%kXiv@PqvtB;lmOxiDle#G9HV&WF_qXe{{auv*-lkMnD9u@DG)Jt<73Z>5C30P2772CSUzpImRx*;{YS}N6^Q=obukzaqgDQtp0)#m8ynG1kENy#8TMTJ_SRgs&`CxqqY1ERibUdEhItBtMby}%Z(gYZY1+}NL=-exSk+qs?(WA#ygPB!7r&_rWjF>o$yU6YlKz>KKCEMPU8bBP_bJ|RnvVGyxJhNLXo>{UG9bk))DgkOr#4bGB=-Y(5EW)|7sz4ciBspA0W46y~e-&xT*`f$ckDcLBs20kc_R#qu*8D@63z^*w$r%&lY+@rRGNw3BBfI7sWDITS^6JBkjvKofWLA>Mx&xy7$Gdx<?qU{*BnCEDV~*y6Ci;E#+tcA_h#IewXd#Hb1o${XZRNM=h!xF#@L{05-c@o~VLWVD6<(Mg3u!^YVx^QW6rGT3dR0;#$~wCoe)YoSSJ*->%j%%x0opTgl21UED^+|s7WId>K88CX_7YYFCj1<t5P)`#=2<K8z{0xVVu$TDUl8syodebr2P=eiow4MmkVS3ZdZzQ=bTkgnNRND<Nkd;kpHKRuqjo=p_dy)-!8bV}MZ74|T%m8PNk+Upg5FLFV(&O;oWZ?PxUkI^VoAx~5WS9CNp~2x(l&{?_^3s0vTyecu}CP?Z3Hic{{RD&`QF0bFr;T8>ES6WrKCww0~8p5apFlZvzV}z>>R486Ka?j^ANEG*ANNaBaxke$ZMc;sZF4S24<$U-H$c;?(Xf*5gy)P?O2_mza3)j{B!j!ON&%xfAsLaR+GZJrWKF1_?2>H?LG>Mr@i?_AAWFuvDr1IZx|*S2=%$fa|vdZd2?p^-gS0RKwm+x{ENs41}!_!pFSp!F23`kU;cSS$rvwDnZz#N+6U|DsyEtGs=$poXN3p~;g}!vfoeN42hY@ZSYLiVY4;t{p*gT1VL6pBZUQj{XPAY#H|<|}<!@zr$v2)L#&*gkQ|x-Vs0(I-1&;2QS$9kUsEW+}-hJup)LP6*Gcno&E8^_-1=bC__a^s?ofh>Oo#VNfGGQb=kPgVzmM{Vd&pV5Eryj~sXa`b^CUxwyh%iT6OJ}zR^#0<7-kC*oy|6l?;mPz=oN{YRsbfDGJZ|F`{@Gcp5cgb<_!L!tSXrPSS;m<k+4i>XQmI%s+8I@~FB0hQHbI8R%@;3{yH0o{9s)sb(~KS<CkX`O5P$hD2-5^iNe}=CW`XWj4Y)1Tb`bz8d{*1q|8%P%+%Rv`@=yWJQa&Fwub!``lfc+~w~G5CPbmr4BP%~C@}y1NWEqyKXd7ERX4z9#<F)Sm%_|_Q%EL}?5Cm)F8$b_K5J7Nvf!&2968r0{SzWDES_>>@+9zZ^!OzLlQ~g+NEw+0qkAfF361xfYAD*3^rJlWb@jZC>zts-Y5XAp-TTCb*aSGPv-sV%`soIuQ9@fGoy73GINrRH8k@}XKb9S3OSroIXizkGmp;-{pTX^AERu+Puq7IW44GPU)cWczq`O#I2Jh~>)ipu(jix(c5DgMfH^HU{3i853XQQQgdeQ`&vyC2*(=afupC$wIK5x<b*ROk<8DW@Z@>x~_bjv`=8k-S^W<aG9&S1TFQ%^(xcBRpnbo>X;Y-huLEBKLla`q~SEla_LtwlDTvPinu&E!)p?SN5(g9@~Dmt2w3?f>0o~Pq$m(BHHeg`V0e484wNkpFpe4!!d+)lSp8;HClgGuN~kr%~LkOwh3~)qhaxg<<jl?hIVb~?&>l$od-g|zPxk@m9VmAx7(6>wv~tYzX#s;g2Bx8+$*<-SJFl&GR@=&11TcOBSbWxpV>eZs0xD1@7{1x;{dH<sf2CjR_0yUQ!&tQbY~a#o|B@B?p*Ar-95w$`>3}2v`*oOA`XDE&9kpLW!n{t1iDqZ#xUBvRgF<Hx;~FLHacP+pOpk{gfh?OVD?20q9L0Fc&>DHZps273TV$3I2m0L*`v~vSMh;>oe)o$5!%i~LX=1;?(q8+{=~LYa83eptq<wz(ieWF4@UF;0=iz)IXK(sUvlN8(@q#!J@vJOEJG44y}$F#i^EVIWmI*y>-jpe2mR^0$GxyWy2Kk#d)k!0T6|q>PZu!g;5VnelT)n<`f0C2*C}!N<U#6-N?wD&WTPpGIAUmO$7Id((#-Q$dms{Kw=JPygc{OBuBkHvyqMraz=6{zG;1RPYI0q&D8mGl1qx+K{$fxnhoKxtm`Jxb=zS3}2pUOcW*;c5I3=IKVKo9a9<%4EQ0>lF`g_ePh8d<IOeyn~Odee1V~%(fIw;YtUCcv&HcbZ#XwAQ|Xr|0JiZCLnr`h$rXZCEKGWd$GLajw(&62<|#FfJ}F`H_x8UoY_k8~<vyiD-k3#6hms3R}=p<bH=`_Dl1>WLA{mHU4A;e9_#m&aR#$#3J%AgssH@C{SS&Y%alh*sBp@pDc<MUt-W!c_YcA@any*({h%RsO-mTrJL_3vc~uY;mQU&9ezf^p)@2gpf{o$BrKg_5sfZITVo+DQon;8ZTm5kO`)f@C$MAo$9u7-r`!g1X5zBwomEf7jXk7?~fU||I+$d#Dhb^+I-j4COYU$hwqVm{Ps`-u*fvr+9}~`-0aN^j|3cWhSK?lG2tl4Y6@GV+ITV|^zxG1IKXQU=0Ww{oB#RxCpTZh|9}1Z$N2BfH#c88H(xui|L*3?n{VLBtJgoe`P1uP+<fKS{03hB0UrPM^)DQF`Xzkv<JUijFQEo3lS-0jm&5A21TmOqvy|L{OpeJH(sxh-8D_mF*(qRevrd}M#?Icyb=WCb75>w#rW2)-Z!+hRI7<xd_yB7=g9Bu4K5^vKSkPefwZrGY<x`Dm1Sho<djJqHJ9|g^x?;wc!a!Q3Lx#h!Ga5#j;2`WubR~z7V|%hkC+1TbVcA>WU&?P78bd%6wR`dRs*WIz57t`67FOZ!sfM!mNqxVzjrxj|U+}`-zlii)==}==bZANMlDsIsh0c1p+1oM=<ah;}lWz6lPA{#bxVFZ+?#=&w^XlfeH($Z7{p#lT4*dA_>;L-tm#=?(^Cx)lr<+$*=VkKg^_`dM<Lh*%(k(Lc_<vFre*OBF4*Yoa`e(==_+f@Gpdd7dH!R*7wDl8YORxVCI1#nxz#C*uKY=d1y7@i4gn#ue7;7_?AT{IR<=JrDiatu54D*`IAVXU>i`H;*3C;N~bxXB149q3xV=2aX283KrVR;JYj%{PBXl%miBKK?M0a8?n>4uYAS3b03v9-lOu11)`wW{uT=acCP!(Az|^#swFryL6_?-mYX_HNe&n2kz5g15ZQ`X>f%^&Z~x5@{f2oJAE`t7R0p@@0TEVF;j>p77zxUk5r!ZVktg6_${?6K)>Xx3frn!pTm$?h<e%R*nxdheX4(z|>Dhmv{ta3sZs-%VTEB2!GA@bIkTNMU@UZ4p~*i0v6~5ue`uPp+(<4BBcr3m_bD2_hBcx2vJjn87&a+CSUQ$9~5B{oFB!}G|<RGqp_U`dKr1J;ozW`>bLivUWzADmf>F(5b#vp;z*RAiJxZ8<jJD7FA@Le*Ur#LgfZSsd(Stv6&z5@vtNIiUILpxpR8X5qTsl9F`9;yURq>$DoUuQ>n+Xndij!pOe-Ud*uFp*axr~M|9q;{h3%7*?)mUIa4D&`>zS!=Y+%&5<N}>2Il$UjPG)KTLRoEK$C&iZ6e(JkS#2n{<EQ3~4p>9yxq#JESBv6xM_*>kXl@uyI;CUw;5`=V6myy9==!nS)m(n0<ifo00wpc#Z6z{evWYagG2Q<>g`UcZQPua$JoZxMN4dIC1<4DdC-D66aUH?G_*nS&zyBvEwZX_Md@`U632v2iAB6!fp*&MUfmzEcWX7yzleE9IgjB!wiVuXmlQA_#Ugp5C6-~2lTbZW&gRt9~gh!`t%No?gxHho;V&hcNs?b1g#1cs6-Pe+wEP!ldtsWUYT}81Ia97hhd~^!esTb$)E@wL}RN+`A|Dm{_ZofA+G?t9qtZMqCp)a2_<g|4j0<teIa`Xx-qQI(tR@*<=s%=|fs#)ma93R#u;r1}M&ClP#Mn8W?wtBrTA`jYqM|qR3qmbl?eU_66dMlB~P)klS*Ew9Kx8Z;of;e;$I|#G|YXWnM>{L+nN^*?g4d;%qHmsAQr%X<QO80W1Y?May$%qNf+(VU(<)lSOaqE^XSBFaN;Gg=mrak{;LKbatgeJQ{F-YIGA*4)TH*|jyK&@}_I8GnCi#Xf7t)m&{OQ*6rSPRbF+nu!qb7(nn=UEvu2cv}2UP#AjYB-Z)jQ(2aVA`pK$#@n$mI{uQjlCUUAX|<63zr;V!EvHhl6#zJW#05S(W>|^juXB5g^rVjzMOMz<kJX?_qxFRhEw$4oQy6ag!hLY*c(>%;qu9}RIyzite>&jSbSqGA;(_sLz!~rzL$UO31vwBdB^YhwwiuwIq2H`erf(WCnCWkM@Gg@;=nxEvMr>KwfwHKK^u-eH`6ep*rQc*f!0u$`7K)$aD>F&oQ<vYxz4&8Jf98sd`l@abHzIQOqt~p?Uboj{K0uI!lW}e-UQiSPufuBpbw5XbNx)~uq9vs)>o{D%S!#|J2GIOwA5ExH(|+bK1;oZC-5Bg^X7ASLW>w1X~93&*xXC}_0DnB?^CMHcdER)JEKd{op*X2e0X#my}$bKVOKoF_{$D(O?dTEa7VKkKD_3#45SHfn?fjh*LhfWZvOe^)y<#b|NlVx(Vw91_w_$W1J0jGCHkA2ui(GmppoaNWb}D;^LsSoz>8OG1hNTTH~%kwDh)z6U$XA??@19$HY2PYtHk=D`L9_Y3)Oz@kk#m4xzJB<evNPO8~B>u{t4gwE2^p;uHm;({158u*FT2;&?s(RIjGKk{Ud0K+u-q_rWuC#`e$;8-{3b6boUpM7_WZ`qy7mD4;9o75$2Cr<llfU_)8E1zo6$*rTiuS1iJnY)+s~dzkL0FLVtem{7v^9&44S)vlh5(dvf(RFov7o!B=0&MxC2q<CuWh_#sTg?}&asPjuKsI(|&*>faz8^x2XXq2627j`Scs_dhs5G5+(XFkV3kqUvwaw)gs{va)~0Z{T}=BdJEi`-&*_TNom*kAH)|f5jz`ME~ve-@X1jDD(|f_X|VE4+Qx~rmXK^#ec#fehnoQ3FUjD3~B6N|NQl@prt>_1y|PBnXyQ}bkby|ekKd?6`CsX&p+aF00I!9Op1OjT23bBS4=)MihM&mh8E(Vh0Wp*^aYHPM)(_Ahq%hH>ZK+60*b-VkT}0j5f>LQt}3#&{8UV{XzP#49`p6mQbI0Tdbq;B!4=M%1xfG<-~AZ>17CeDtsXaDl^5VQ#~^+Swi3;vV{-RlXEAs?Jn#2)KJ@;*hn2PN-G_7dkSeC}q0#wtd_Ikl5w&tSle0>hem=QduB_o7`}Jqrwa->I;K%^+v%Y)aXqmN~?b=i4j`M7P@A+p=18$K5HU^$OIi!?&8@1iq{%7`!;S;znup)f`f)X4iAU!#c>reLJJJPC0!&o4_z!h^erVP0kz0rA0-K#k9CCCY*OE`#j+XA^Nl?yUH?R8P}k|Yo+Q9^4UuR~$KeIn!NVt<Soo<k6}CKI@QQXZSM)cE;3_<#G+(Rr`m(eYO)sh88f0I@&>_d3eez&Vd8AC!YL?o{3{+lYbx?9sK&nc@cw*eU9z8<*3bI@6W$`B5KH$SC7k`=spD6eL<w>a?8-GvrBTTSPAg-UmZOEacJTvOVbtKv-xGagAXLmd5RPx}@V9vx8f&o3iOoW2b!*IgbJdZSWEo41wwEyY)tceD)e;m}iRp#YO70;SYvNK(`fina*G(9}fGyvxo_Rt{M}8pu~90i1FsbOBlzXN67#sVcpp2Vuj1tr;~FSS=lyyHqr!?<Cl;!j8!19_(e{2<6Y|J(q)GTTxFCv9rYFxohJ6O-VnlzUU-7gqO@jWl<||%6um*qFxjL~C~?O%SXAC=Y8E(2K-kzj+}}Vzd>Ef=uNr$IYU`YjX?aDRQc9(etOWr}EqviryP~1Z!)L@C>fy)r&oU`r&W0dVC9vYfEi9u`JtdbHIR)l-$wokej5;CMYj$ocSgTfFL{6>UwcH@N3sEO9vsBV3o849>S*1m7L8PGm2lQDvJ*@KOX#|rbo06W0mb&_|YXhx9byeny743Yase}`!9aFOAp~tNmsCK_qWk`Sf@K&(bR%b~fVIVjJ69LOg0=x%xmitjjOs{8MY24-{LUmV3D!X20mrm6cyt~h&vp8(U=lyBG+Y~)CRc+m_nr67Sk0vn9)@8^wG}8^W0)A1}ly(xg1%cXEaM{Yk%HAna0j#}~h^NON4O1$?bgTa9=rVBoQFltVHcBNJxC7YIAuMFzc&<LNxmf<(w7~Kw;v@_Ky|=@S+Wsaw8lnRLhhk)~<Pyks9ECyKbhq=fU0_!0(WKoE$tCqFL|_9lvy|*-cw+#P1GL%huK-u?v;~YPoVQiy2wf>LC=ZUwD_@AT0`{MLzwmg}g_E-FV7Y=7(Oj#n-t*y~2l&rJ_-Czx|75(jM_1|S5DlTK>0;<#SaVxTNaibwHqI_#cs^Hu>poozu`;N1wd}7}%D!;%zkBy?*}r%1UOD4PMYFn4?@YyWwTYu``FfKTOZBi&Wr@~0&t8x#wXOPHABbgNknEeuRg^`Y9B5c+!4k#Mx`(@4=<rGl6xJr5mwWr4!XgcKwi;+5Nv}?x{Qw%3PRaa))P3U549kXnh=cc%!MB3dXm!o3y4^)|pTe<n14YB7N){O{%j2PpnFzP*wU0A?V|apSL&e`TQ2HY?%;*OtT-)X_*HS{4jG-$WC)QsyN|{u{GS0QsIG@!T2gxW=3fg<zsPBK0Wjyh;-8W{I9?*FCuZ0F|Q)7$1!=~xQa8?x#otpA8BQpm@VoLZ)ZTpY~yZZBu?ZeINkXS2AUmnzVp6%_UcN<;cq-NH49}D(IhqdqE9Z2!G;YE8QqGE{2se3%CZK<a$SnKT;@!mXsJ>e@<nArWolZptW5}9oB?p_K!-RVSw(aEGeKJ6XjC`4ilao5q4v&dYqdg2R_G1kVp;gTGPa67RupmvUf$w@AqhT-R7T`ef+`fiej0mGkqFerLEjwa!1rBcBpJ2sI*`9urA6PQW(|K)o&W5+niX3Q3Du-^JWyiA%_?{f1hiZL!ET*2l246l2Z@m9yTi11JRDzY@H4s0&KIG6DMrdZJ&`-4_SYO{XaO!d7*wQ1+v%2|G+ERl&^XBm+!Rq=d@Y|-JRYW6a#yUDW+68+`$6y?ejWKPFuVFFSvm69Ua)q+xJ>(1?`%$ZZ^@|xFLS-n+Xap3lsWf7QELSVz*u<Nhze_KM_$ros$P9e>zsLrR-@cJ3B(#MBe+nZ97<-j?+I13dOXV^~uGW-XksekJEgqDJ#65GntKgW<szH6QD1*s?q8N-$&V<n48BgcX+^%OQ81{$tPY;O_luhG;X3o=c3elr_>X2;cP_op1gWRngiDuI$;%zKwtl@ia(X#;~RbDE%>Oy8;ntPtZM{Sa=e(<Ck}K?U*EKz-RAIoT5<L{f7?z&GPIhzZ#MNinQ|vC_0*$XI^@>++Mw-R#DDiA{a}PR@`&f2VbAZB`~!C}<n|4bAfi-g*Z#CG(p{n34Y-E^ZE(Q~k{GJ9LjvqvJC;$5PtYaNIwS!&P~zN!oCzw*1v}@n?=lr%O@LjIYO^wNXdFU+96~soGYwV)lwNq0b6T58cm4=qG}5-LyC84NunDD~m#_5O2om_GU1HwgVMWatEV}NafiBNpd_H&3bUe;ZEQ-dT_hGLRKikqz4Grhwd!>Xc{Ha(}SpuoEN<tVoP$4E?f>ZFco8~P!z8o)`cJ`A)%zuzjx{1d-U)9(DgiPrlcTYGI00AaS>_>cbxvV$Fx#^P(c!PO5LsxgbVuZ!BMB}99O54=5Z1*OlE(7g7Rk^F|<t{z<bmI!Mr^bN|*d#Oc?YPgBfx!3j#7#Ab1n6G*^DNmBbepqJC*7;D`4sB>roVB3M_85$DAdI9e^}@&X8WV{d=IzFkAWX0jB88;1vbPoB8*P{0z3Al=rFCQ<t=7Yjudu8N72oS~Zw!G@_%1PzM{Q{_};PJeM##&edc^kEY<wek>i=N=`~Ny1m<AwnIv#o<AVA8H_{x&MGu<ZUtX+~rfi;T1FInFtd;779KF!Zsh)1^$q8n0=P7OL7--ZGn`Q%k>wpB$^%f&*7G@;fVkzETi$PB$gnMB6CP+?zIFpCSWlzjz@q{WDbReBTZR1p-;PjQG@^T+`^xyd?5f4SOkI2&HVD9qOUNmdQ)eaUT!(Q$_JMt1Gt^tM3iz^58Hj$DXn^rFnh6*1yQUfqt5wp6gx{~8tz{$ks*`)Yhwo<odZE795Q388bM~IHBZ>iDgW*M7J4oTEgF2^?@J8_LfAV~=fl83n_whFGlpo4Cg;Ogua$N@^rKPRsVAmsns&4yBWpcAMQi6tG&~1}$Od0hE;_8%anKcYsykc@-4y1D+TS$e)Kc1crees^5Ip9R3~u(>Y*4k{q6D&tESua2Sbvs0H(^G^LxC>RT%Fsq#M8FK;6^?T-JeWH{h(6z+xXwrXl1pM@hQhe$tevvdnJFLPGQMvTKSN3Y;r-74v`Dc54){6X{ef;e4d?tn-qsz5BK&hsWDrO`B{R!-%Vd#`=@ardZn;OIAyR%eK~;w+)?ZMv@UWAtd<Z@dhv086vGYsH1?qmKl%Y|0@%+(I@p=Y_z%+=|MBcf<R<QWJ*LT|HeUD9D*d;jZ3(TPfZMmSDVIoKl|X>d!xiYEw|+7jb-3$<TghpZsHnxfoJPuO&ywOztf|GGpQ3yCL^^zyt?T}1H0qbcAlMhy^92oj)!=ueOOTRE5wL<!APOLG3vi~5z1@TQ^8?q*(Jvg2NU;p(mmwyZiUvby@>ptuLgcz2XSihrlvKw%3Ads0Vu(wc#?33AM(vK~0+2F@2oy}KrWGl}8sVatNzCjy=>)gBiV8e2WacKCSS~`Ch$&+`KYIv?PZKKZs9I<**gxELuNTqCur|iH6?J3$&Ac@G!7NlN0Xv%)$cxsi<dYthL}#hQ_$m$?5h;h~X0LliNT)JJQyhP!RxR6DVV((5m=}swJ-$Z7+{*3d3x4U-di89S5?|-giQif?ase^A=Y3HkL|`lVH*+4#F_z4WFB`Uw2Ul$Qe2m&=;Zh{cI3&LWs&U(T6eS0}n0QqvHfi_*dJy10<qjMYaY^quD_g^MeBA2^_kzN@f!i?$7Gu>1UKBFU-LB^64W4wGa+9<Ik~Tv_m}3vB&HC;EvAmp$D0OBnj-v_6QX+Y#=@KQX+u3UD)DAX2Lh)6T#T8|m)d$wpCRfQ0(?Gv1Nr{sw&j<Ftl8bO&Rbt8HRwRAL#&|sO>7I1dm(2(PO36<TQ%8RR_L3yx;VyA#O!g$sFrsFDwr@I=(QXOyf;>A=P>CWdv~Z%;i~}0`7rk*wGW*2Ki63Z&AKO&5r$B;`?fRk%R_6fuE=fB!Xpc*x7q*HA?f5Lnenqj0^0W7})+H9L&3#9sk32;vX*uOMgLZPEYQ>aWzWuPGyna5aHDKe&!&UA8i=D(!rIqHIKnv2%G19Uij*QBeztNnIqw>2p7Yw5l3Fp7X*SSs{w>@oq(WaNzhW*gtdrJPdGqD_V4jp=FW23?oU>eS(;3>Jg`VFkve)5RDFPgS;A@0;XCdA+8@}Q)XR{=@PCN~n@5RJ`cMRSkGVaRS0`pOR_r?Kn`D{qtXTVipG^us#w=S>Euit3YMt2=qGfz)VU*%1(A&*z@v>2<K^Nv;}5ALZIYvVO>?bC84-Y6PGwv^F#mD@ez<U_P7mXZ799yd$C~RkBHy!e*+~TD{GwpT1DAmADZ;A*4l&YS^eX4vgbJMN*pxu&{oO)YK`lCQ#CUHAm!;N}R)8%#q^e?SZBn$qrD^3Vff@%&p6!0=KOdC@tN+O2u+_FcnBr0nD{{xy93fTxKnn?@FKU?SFj79rP=ACA$*wn`K<xa521`;G&#HvuR0N?`Fl8<&J^Ms5@Bm1%N6J*n@({aBCT!1n&8?yYj#)pO7w8xoB}quGD#^r)t@SHIpW;ja}igG#@WW;@groum48yud0;wUFzqYa9<P7JrfMu5ZX_^CoW6ZHR-9gj*e?F+Z=(*hzMO<+j)V&TDL_^E<^1l&#&{4E*b~}*6mHvD9X+I&gkd`gMX7jG)7wsz9uJQR4pTEbmXv=81bVy$U`@ZK1N7Unhgd8TW@Urbv=s^_7_qB<tci|(?Y3Z)~`ZL3m4$8ygURcZZ}c+F8SLPQ$`S;1^j7NN#jbgqTtvYb;9%EG4N<qW#EXg9~RCTx4TjBxA?N0`@NK?NmP&y`Y3eRJmlcpZV!Y4s*?Okp@keME_lNWP|KQ19@VUmqV*vwG}65))M+E#s#qk3xuYQSD*n${S#LQ}rciIe{$>>w{u22J1)=5G7z8RI$6`qiQ=b<Ts<6sIJK}D#8W$E~lEE9_$*NV)wURWyT87=$BJj;hL5F~vz(&c#?8(-1b<-uV`@@|A7-6_H<+^niUpzb;Y&=$HMlvOez$Uz$7toZ{qW)^CzZhsD&;4L}3cT@j)bEhJ;Q#OKTbmoX&NIK~S9GMc8m(!91UVdyz(5-e36h9Nf&##iWe&Fk0u2%#5P;AihZMsqU2NIAPDPGvms4KHmpHZiX{*MPCK_E`m3-ft`3wFf`(DmDeY*Q}0~~4NN=9J!>2p8td9Tl_aM9J*?tWo@$z7aZI#6_4y>ngx36W_v#_A=ALRdE%xRAI}NFRYRs#p692Y;pgRWKbE;rSY!BBE`^$;qa4l!OSC4lv#hF#i)E`_fl5>`02xOu<;8=kHRQblgSHW!;Lw4LC34sOb{blb&A&KbOR*u$<woq3)vp54qa$KJ-Ldj+`A};0`5;==4rc(2W)=w4Hq_F%39R&?6=m;pq5?>jxU&&2ox$0CMbSxN;;u*+q??4Bv&hmZuJ0<lUtv<|v68Fc0kMIu{;#ixVELXy~L`T8jwXFkKN*t}k*Q=(crS<a7hO+yV(U(bQs`S2UHGg(2*6I0KO5Ie~v9!CrYr7~RBJ`_+mCUk=-+=za!Yb%s7uhu+BufnW(=%TA1j#<^K(R_e8l%1*^FmiRo)HjbuP;W4m^MvhP$PojkgNBX?Axs+c<*9ErXANrO^lZz7r|9Noi4=l}>5%-f--dWRp`sUCV=E4fqIsrv)j}X&qFwxjjbc3Y8{eY9X>b-h}9mljsd2+rX&OghSQp6TJ5j_0!i)U+S20x!<d%V^-kCD|?gQfQLPR%B~;6!VIgRq*aM%a#t>5mlETZ^rCquoQLa2FFNN7K!~g%{>Pr0uW*qiEl^hF$60Lt*)RF}l{_1xEGz{jO`ZtS-FiTSw5LB+MR21mzHWSYM2yDI^J}kl}}R>lK-`M?@qgNFd@X5%2-|I|+ayY!4xh7!a!?keA>oaOMLPw}nh60R8S(yz2IDy>_><UD;_yPz)7MKVj)>;2&{<Yf(|4JS+8CCU?QvhZajpwlkDS)cVj+<J23pI;e@9MDZ?Y{|2cDOd6)5tWi6wMFD$+bLQHAXO*`q<SrziZ&eIfOMj=gFG}j`iWSA-U1$`dY$zWarQ_Ag)>dJG@Wr-(dXmmHe%?fx-Rf>-tGZJOW7ZmB&s<B!n+fMEH!4dLu_W3}F})lsmIh84MGr@~6A2X+wt}qaT;FfqM_37=c0e8s-KnHI7mNHls%6Bd>#m6B36$>^tEO2=AiTh+_wfnVBl1vb1!F?(cE^Mq9#hB6t{X^CFY%x5C?eG7Xb=6CJw=dj{LgU|!Iq}^i9Cywi0}`=>Pdu_E+CPNaoPkrGkJWV<w0if(n4;inOJnnI3<18e&OI+ei&5s6vLQ4j&IiwgFl)W#w=~t&pv47Eg?X8+*JPzoOV}qCEeDl>eG-Gr#lsh#?u@Mu4x=6I9JUy&f7b*jMLv!4C7h$YpUOQ;rjM+{?M6b>y~C~W!gq-wVKZO;ohlDw^r{@57@n-?ceKIlSSs-Z{gHr*^@;pDHdgZJJlYYJU{ZwZ*YKE>QcUr{9n`g5l1K0fPJwL%Mzy!+sZ{D{wl4K;eEC%bFY|L)J@C_?nNPCS57e`ODpmvy6`29#jKLh{4ZSF;EpBG;#1M_rSR`58>U?x-bd{Qav1tkG0|zb^{sPH%bkwStWrB7wet&e7X^o0U7bV@PrL?)_cfEOF_cUkOzdBh1DR^Wo!KrLnfrK(DP`51A<NXVIm=Wg|D{>nQq6#A73lUr&Hg&8>F*i(mG(&rK(wjibco=@=&Tg1;8?*JW^nB7H!~(mTj2x0cXW($H|$BERei~!&kO6KE{LXeGsP;X6KTOCb3UZFv2{O^Q6tQPh&LceeDZG!+vy<JIvk7S(V|4`v7?e1i}(CjRPwV!1(>I6WbnAC4`D74g)Iig#4_&PN}clY#_QFw5Iqvv2b5M<)|vJmaZg^|slC1vuMjmQuv(`lBgGoPIi7MKMru%Jf33E<hgyM<n_4S3D>g&Ck8@K-Z99m<M=JZABY;uVN9n`D&-5&Wwe|JtYL$`{@>Yc&d$57`ws&a&wVkauqCVS{z6tpcIbKwBppk)YF`JG0f$bb50oyY+VKVZj)hA!TjBoTqmok~=jQ2vVrf}>dONQSybafkbr8KSoQ&7VBpoiPMbcxjpn|@ZkS1So=oYY-W6`s`@k{nEHt-|9myR_d52=mep>}e@Jb-g%c&s*iNGlaGM5rQ-m@a{)CMs$cK{3)T{BJS{ywG4jGgmArZd+F9ow{QDXkY!b|Y=rB=Lf(4vGs->s1HzCdoL)lheMorF?=tQ*p$`*8_2-0``|*>98p800@cK~#GZ`NtF7tZ=*Oy><A2M3?)Aup2Ahh&xv~7fo{)k|cKadc|A0dn|Vi$kJ*uako-1lRFE&ZAGz3MKE;kN>nnp*t=5vy?k3aOi5&m{)*BdDInL+g@EvKRs|W9z%E@tddqguf<e<A;O-{SOF>{1M^{OYCGqBz_2s{6mB{{?;*tZ@$!7ymi>UHD%>hvFl<A7bK=I{eTN|>kV?Nx6n81PObK;b;FW4UDkSa15rk>WTRQ$s92~qLfra>rjBLik<sRZoWtZdrew6;G`B=qlJXym(nX8s$!9>>!h(n{vmOk`_dyQ0zakI&No#P5T32#B67aj7Uh8Nu3?{vH<h+BR{@Px(4%D$wD8-I?nq@3*-m#T;HDy?cy7IsS)~z8Y!rC=&yV5KJH7!RvA^`5NA9_Ary@QljSE}{Y)E*>e9ZP8SxjdT%ue!6l*KEi)5uJoq+)3K1SR0NOCTRdun8ZTG8o`40J4gOlqIxkb83{J1^wl(@G>6{mW@Yu&MhP)<wK?4vao}3Mb>Y;5K76E8MC}Ar4%fH-p{p}(eE*B8aMB_sD3L>s?k8;CF`CUXEn&1m+AvZqVzU^z(yLnCgu64UDjxMG`Eh%fAOu3`soOh3ugT7+XBP{L%OLi(&U|<%p4?h+?e1_e$)gV5F1{Q)c6RF=mM0kX20mKAM>0uG<oOsGaq&i%wlG?koA<it88R>g4Ubxr;}Sc2wvUn1vgfJX4Ld&?9NEqaj!JYnhlSn4!y!t=6sKxWsA(6*Qi7qxU2Wh~minC)o-ZjvVMVcyqB{^Q<8YEAm^rsQh<6orvE4y#Hkp=k81HV?nsA_~!qlwAk;wREMBjY9T;CBR=kD9+KamQ|NAu>vL89$wQv8$2jkVhxtQwoA5*FBJ%bZNdQ!3f@+SV2xfATVRUf}nv>xQ?Q`x2{s2;$2*U`>4CR#MA8%IX6QmiI~y{)3yblmjZY>^Lju<XtH6@_+#xIXpLqt#R($9fi4c+zWI?+}EDWjYP^Emwp_>Zkf|wAGqz-h}v2*Y@OEmaUb?~ZUIwn4aX(>5UsGb+ka5Xzr<ZQhpXd$0KM4n-h1x&ax9Y}UKKAfa$IV@uI@b-N&=%uF4hAV=9^}*baUA~ZUI@KEp6z}1?{7p{eORMZI>%9a36X2KPx9PaD4sb-=fSWusJ2nmGA{}T+C=<ZUaYbtx0bJXVL~v>7mF#FA^<K(hp?ZEaLHgByx9r^yFcYb1ra}jaX6-X6}%id-@NQf<dI-cx(}e^)Y<&HFJ?+x|GN8<6Y+E=*3SGNUj?7uontylM7cuZ?&p_5guV%7PQ^y>2OR(r#~E;PR&-WmU(F4{eBJ$O*;}VP^r{|B~Q%aH$O2iyEntbahyw>8~wDHb{KEx<Y)U`Tie{grROd9d)V)It?|i~ow~UVZJdi;{NzK-UxK`IB$<hxIOJnWH}WUgls`enTif)}ExhH!{V+Ei_;aK#nFH>&7*EI|_V-6&qgw6R+iub8V*`li$UbczG?23IU@OTj2S1Iwt%o*hW>E~Au8In(CTgoP_nuZvN82Z_SGXH7;J~-tR?zmzJwk%L^F~{CX4y&0`c4|<g)VEqSez>yxCodKw!{y!7W)q%OJRy0MTKxRWlAF<&oMQ5NDGGR^e9HCK7@Dw7R|}?g-q@v7MhdYC;=6B(bFG5!(To5oO2<$B7F|ixgSx^7v36r0-o&su!J@vsbA2RGi6I?1>E=!Fm4cFjg6Q1R`!;F_xo5zEEFnw^gks?@xh8|xo;!kNFvo9yDdJ|GLwKZaU1b!p<)_yOSbBr^g3`5j)FD}%5CN0U$0FG)d>`1eBXy_6;2FolM)b_<13M<Du?;0p1$jtF2cTlgiIXo4tXm+`Cp_-c~@RwGHVrSDPSVRW>P!yw|W92(kgsGyXYZR@B_+r6xl~VW)%W%%Vd}29{mH9|AHD98N5WcBl_=W`sStVQ8I%OY7hQ9mY_#Sp?D`7?$jrrW$(_LEIN2JWb@tFS39xQH*djXAC_ncv+RTHX4;HOe67^$wR$6SW5&wv=i(n{u$=LtV*8a!g*YtCsHE*uW)F}6axPOz|5`^!IUrmmB9!EjOqh2G8Y=xs3b(%nE_NYKO1XDTLo<jbP{<es{1;LgzRfBx+P{LG3ileOzxqmOpFYv|Ta1;<-bBluOdHIZVuO^eBumNt6xlb%oi*v5_(ta>T40fVC~G%q<@R&>N76T^6vx$#?-XqfBh{tvCu81H0RcVZ7MqcGGh`_a9ly?_wAj$A1E;8n@Od74=$_*)OVV2Y6eP|C3{_MRZt2t(l?2q$mtutUXOvfv5*1=f!H-xOYm^4;D9NS`o=!`DFVw@2<J$ADffhf87oQ6iENWcerhJ2s#C859CBYLyk!bwMm)U#%7P58TuuEIiLt8spgi>}s7LvQ3_Ox-gw=wZl>&SSY%Xil#Coj26n*6IOp-thcKT!wTZUtcvjPhjf1cci{>>m9&rONxjz(9+nhw=Le^F1bX{g~qsZqbLxP7yhxJdat+{m)v>4{0^kg8xxy@V{o||JQWMJ~Ab{{8~s4ABTDVrjWXq1Rj&8n*mc6T~4N4c*Ep|Yp-cPP0O!q9kb}2UQXR_th_YR9wpxa1qxv)uR(`y>l+NOD{5?t<k&htOVyR!juDLCv&n_*Dj!(1H9n;y`aZBkC>G?z<@Pc0O>o6ofBx}@D7!rrcivYijG<C##4=oxUFQ>#b(5)vDNX9*>B&!pA?Z`h`Y8zB6n^Zsz7bOPm0rS4>>lIp3LV(c*Wa|kwyPU;wvxV<t^A7=B(_Jam~m$$XPveQYY002V0h9SphLrzYyHPB;St5$k%t2b0mY*Xb%78rQ{t;eu)oJXAy0eDm@$Av8Wugp4if||LlBkJjk11z%$#<7l_Tk6aeX69*iR{cF0f!&2a+k;kJ4cICGK0YVJ65?V)H;Z))orc+vjC%r33#29oyS!+?XmUH&Jwxn6(jF`e~y+LQ6lLt&z|RKHDzJ?K2a?q9^YU!Py*4r9a2h=ufCVCDp@6L=+yYjIp07d;JHYmMN9QY!qO3TN{f-QeN?Wgm)kjDWQW6ZRCH7)1yyF%zH=)p?@URT~qWJ@9QavlUHeE+Y}a1H!}o2=6HCDrcQmG)ASVDPEeW(?O6(bg32q-9S|9P&|gD_SV;e@U|vr>PFTeq^nc+%9_&PrPl=E^>jkH+e(x<A`i6OcBB_If52Xw!iK68OR*ub%D42|)y6T$M>SL~0-SBv~+8`rr>5_eeBA7o;ksn6=wVaTWHS6V_#(J&3jYL!@YDQv)B8W^28E@&z&?jZi<6g|#sfg2&-l>F@PN-}6h^Xr0Y2UCp`ErHjh_H)j%<r0UaH%NA%0&FBC=A}HN`fy+2(){8m>b((KX>n5=Z5oNY<Fy1Ip*uEd-n>ch<0qZyT4pKa8v>r8xsgvwlUi#cG-UNPa@MeO41(`*+O@%j&Bs|*l$_i?~Dcxk@}iU9GEcu>AToeH@zWIUw6t#1l}ws;eij(M{V32<(&QEOE(W<V9A7|(Cc1!2&>EMtEIS&%{;uu5T58}aI}P~;=80kjfMjJ+WAXP#N$u+PnsjF)4qfsSJwVz_v(Nd0QlRbS}{TaONIGb4vYiFGE;1zoimSJAdCuB6^^l4-py8}>LvGc(GNw=gc2b(E9Yn1v0jL^AQckiayPR|!qlT6h$thZTd8ywO)4Rf+4o?&2z*+E#K)9edI`im0O2R8?n<%B-qr;K^A-yNMWtN$9ceC$KU2a1{QC$U&;CgK8QO;46-G^*T)3iT%69f0OX?q@;_JgG7pI5)39IJ+0CD7<Fa>;yYZ#ttRNY3cpGL_T>ic2#EnV>B-AMBy_HLTmR$R1^qHcWSJJU~yX1&NfCY(QNw^HAp?XZO9wIicdBB0>}Hk|W<xy-k9W+M7vJ|~G|qBu(95ecPd=h~K6(RPGuU70Q@XB>>*37b6I&6EyANH1TC-6&Z}qe1O)s7Z+(hbO-zW)$tPk9EaM><p1sO|zUvH9V-D$<1P3bFxpxqSqb*M>AplGh1A>Jv;4qBtLMCdw+(LqQx#{9Tap%EZ3`zX(E#81MR5j)uft4^jk*%;M+Sm$a?L=mWW8QPd^~<ZcW6*oBLe2J_vVi5}V%<beM>!?cC_v7KQzvOHs}t{{iL!&J9X{S<l*ereDzcHy0SCbI&=d%tJSw#hKl7NSC#}K-w$)!f{rtYx?ZOPU}Cz^G_ou)+}bCX5t;vy#7(Bbm3x*x#MJO#S2SbfIaVX&%Luii#(iqQE6Z5hcG-=v)oplQRd;>U7L}YgM#~!ufDw_Gaf#eW-XFD8^|fRiDP$7!cl{+sr`kdjdLHa7U^#U&7wvBh1@8#i{i5aKNC&zv&Dm1PS$O8i_uc5I=G*n607}=%WS2FTC3TZ((kdTY#};!1BE)~wvZ8Hg>ck>udMG6otP0xD0{^nWhs5m7n#-uSpOMK@J5djK5w6zF3s%YbTj1m*h~){nRble1VpZ|&S-{XbBGXroP8l%7J{o~tKzNgRkkuVc&B9gK$ncPYBKJwS#r=C>x`UL4bt);4bh>KeZwKjC|+{Eo_@)l#V)!3O)LI|u8qT0$BXXYKjpjbEOytW=LI|?pp5HYfWoC9egSqOj<KE0`;W}}iGsp|R(mqnksh=i<w4ugJ!lOv_a~E*J;3b8t{=TzrMK<PZ+0QG2|LiP<b~O;Wa6E_7II^#dByvgUd8WAbgf0n^Bs##pkx(zh<-^NnM8XBq`x00K6SDU+iw%)2xU!dL0<Wf$$G7Kk`3C(;=fBTzQ#ZCDqpMI#XToW$6azilR_zQ=h<XwIdRKt^kW)*ij{VaM93Ux3xXi}F~)*wwVqx||Kx`}ei%rR^>-1KyWo;s6l8SU%_Atiy*0P6R0IKpwQ)FmZ1!{60r>}X{P7f{RP#C7u<;3*13zIqbQqNuiG72DJlUVoz3`DD23PLXA$&-#Jns<EF?E}QWban$4TPpTxN^<4pyKi0aL1(Afssd8!Az6-t~eDiKUQOc;7Qf=n(LL_8UkJawJnh|g5%coh8mmYxy9vM)^Tgl8FstI`j1-yP=iCvtp-JN<JJQf+|e!Tywf}K5r}ctStfVB$3)=c23;{mmSaC-&pl&@Ffg{YZ?}TLpCJCv0n+WQ8ivEd9<DSSM9s-rlK-BicfTgjRO%T;5psMbpYo5$G-&DyUNDEPK+ANZq2+#)R{!Id?0RWzl_GBmoo9w7eoj{UKM6kWD~o+O%@aEr=T3Tq9C=_lzzY8P?VqtT62er7r!zJAwUDX4(j_X6Gx$e%{p%;ck)fNZ1H7}gUGpm6ZC2_#YGG$+T$#XhA<y#95s#YkB&V)p0z;!=z($BdohLt0nCG(`y<Mx->BIo*ruL3B-9`}jl)i}%X7AaPf7KL8(%C}y(9JxVl#jy+H43ay6xOdsUI<w;?W2E*t97F)u(V##@enYEwVl~czZ6t>pJH2I3nTB>Pd-m=WxLX>S63UeQ9sRLhG<GeUnd%UY`jcAOYK0TxwlrW%s9lPGX!0n@AJ6D>^}V=b1=wOMRXWX-!t}t@O{n@pU)ekU&9PN`E62qd(3)zGPM67bOcJdk5TFJ3k)m&-IL$I#Qp*($}iy2Ppv<{{d1BK(H{gQ*hkDtAOpc_*&pj30qA02&)DB+@XDk3fq0y7x-)jXYYX|6YLbQ(4f_kY-x*Ru&(P=J5xqx$%|}`M*m&3nM*nCIy(X=B3nZU$50e+q5%BH2)gBMwmhG%a&#+G-+A(B#Vd#jqKTDr}AM?<8IQ@{>#0|bf<sL_z!y=^M=}%TH-V<?U_oIli8<yot4~^M;W4G4atTZZ(Os@B5bUS}dI|r>-MFh4bhTvDi0h~A+IH$nj7XJMHhucdl7Q1859rlOq`_^jJwKj+SlNCxs%K*!iT{|A)%Z|3h_r3$Q;nY^vLWmy)8cX4yq3({=W(5n{wziif9<K?*C%#U`u`Qa%wo&+xUv4Z~nj^}JrDeRHHW;AETh;9<1>z$r6)Gd(+AL#&z!1hr_g#AS5iF&QV9BC5DdYrNeQStG%T7)Z|NC@=aX`;9gw0BMtGVf|)<8liZ&cD2E21?aP|t6^PB&9Vd2@ba+0uPzHAfs&|7H(T&!-?!VP+oG&2K!YW%Hooc-uEJ;hwKEM*Wao51@Lc@!T?x=h$!cC&$odIKnUnAh*2^yI4u|2ETLK7Dlt#2Lt<agZhTU5nEH~(w}(@FPX=nqpT|wVFZ#7V8;AG`$C|uKkI5-2iSh5aonC^92#1@MgDr6mqW{S@ePJE6SjTijzW6F*qYgw)Nh!(Bath{D?L)u7*5AS<xL^fSP`dK4*z|K^YnXWM5Qk#*Y*qIPaj7xc^Ec{IWK82_@j6w*rm@2H|Qs>)mW`!M%eZ0s)u3=EAFX3r1AeMv5u+ZO32Qkj2V9MK67Hmq5S5_uRz2=S(Lf5gjx{?hCH5%WkD8Qs~c|f5gbssNZ1aC398;a?V;o4nLw=WFbnmd)xSUc$bJL8Bt;<$1ezj9MO`HQogbx+^FxlZ4xMQ}5+6UdmKQA{ggho`<4X>6G))FGykCl1G`|F@?dcC!&@RMUozEpEK?G3hk9ZKHLdBynzC7wdBphoW>R3;RR>hJD3~T1O{9P*EfklGX$3n*XR2PgyiP+pfPMcu_lRt;0#w!Cc!TQ5i2UC=tPRO4S@p~|9gh<IP6CqhgNLJ>bVSoIdXW1Azk-pn$Cf}of;W2#qh*2fJrikpfnG}$wI)v<NO;#c+|MUZ5=W#`l$}`X~It)iI=W86|DN<?LaMna77m`ybA4g9^^cT5>5Pbf$)0?Q%7+4hdzzY?IN6g;ce-%+MUqNS53Cq?HTRuGPprI-lu`yYuPMVB;A`30xF!&GTMJ>e9JR0E7**o)3=?e3-uB{?`nE;CFb%7V8{0GQvwAyVy2v)4tkq;O6G4R6r3<*pt_yF_A&D_JE(;ui7mr`>gW+!#S?oiu(;Z#0{(y`S#q5hM*`so0eKZ2|Zlf6KfyM5|!?__j3xlUJqD{KI9I>Gt_79T|@RBD%Z!+b&vIf9EGb`Gux5Cu73G1}<Gq;`N|?VJ+k;RG2#;P+<k^{>)Hqzp=m)P9GKSj^fZr-<PJ=nxe86i|OTAH?an)t?vnPfmOsY%e$ot})vZ)<%H(ac>@7yXOt;b(<v{l;7q9e=_Nlco$y0M0t<>eun@`c{`nf5R2x9(}5y(*CAS8aWJdiZ=>5!TE9^jKq^w0UQ`qWjKW~u*|gTz>hwkvtfevmQ+D)Eu`Dp#!wK@JquG1?2eW}J>ZOr7&qG*b1m^>R#qTgFBP>KFWjgUC<%Y&l_xvEr(Ka?Km1d))!}YSh=pAC%-+ub8U6j^HH^#^V)h&iv8#iGogCP`q133HMNo(Y?FD1K3eu^)><wd)=R(pMCt5#m~UMstiKB#CiFBUommFj_cF3vJyZ9eprtMg~x-6gx&EY~+GO>Y%W5K1lHZoRS&0@-FW+R*Pp-K0nQ&QGah1lCyab~oQ>5V&i3Z>@?m@N2Q0Kj4qNg!vTuS6Qv?)VAM{7_!^to$5Mf!HkbFB|O60DAes@DiESeQ5GQwW+=*?jkNU(ou}g#u$sM{YO-HH7gl(&E4{ckE6onfYNgTO75HaiF2l|INM8rsEFN3W$fZs$ijGg~MvNS|sr2w1mkE@~0}55rMLoVKI*MP>z4nmdOFSC<CxScEMzRVsx>ei2W~9XBJ^5L|V^$hQx?0{*iyt`{o-U)|H#nov=LTt?-iUVQpICGed^YslV=;_7u~ywiF|-!8^anw^psZ&SJnUjLVIA)@*dM#iN}X$Q-1I{YTRf#&Q4<`ns_ISSwJlEv7Z9_8j~RYM6p`ZN*Tj*9h4gBqxEm+Vu!&&<V-a;u6Du0oCgI)^*Aqt+(JosruY$lY^e=LgDM>Et%SL6ZLi$ec&KusQYfzMXME?HBvHSP|D^2+z{G7-DBNqd0+r~+h%E;(zFGA{zjDwo|;#Z_B_=36#w9n%sKy&!1*ub?Nn66ctE&H`{_OcpII*!~kN-xC>L}}>S|B3sJc<h2BhmVJW$3w+`&0;0h{S`Hd{3p2yy~{V7(2(gL{TAVk5$HlFM;;?h(J0^Ma!B7d%Ddtcl9*}XSu;tOpT*8lt-b+-`g=TF-v5>Fu5Rr$Nc$Z-OxZk>N90cY0h7O5M#ZfcE&|@9k9e{KGSdg%#&TG4qf$eA<{Nt17#kKTFDa%ZACW*l*lJl06?ReS^#n1OtvwR^Gvd+b=~@@<+t;f*aG8r`i!ZGNgmCS@slmMJO9)%`)yf-<II%)C$yc5LD0ukvM|gEKUiIMOY3?z%@iSIqc}a5^l*Q{7og~zZ$kL&YWKsR(l46&tL+Pzm?!qCi-1UTMnKv=77p(vp+lumSlzm}$H$F)S$5=!iy{n=7d0VxbU4(yJW&|`r$0N_#Pl$$ADoNhx_aWW%!Y%aVb2zce)Gj*BqtbaC;y4r!7jkgbhJs;MWG{^V2JTApp+g2>Zw&}!`0aXm8)yS9j3x-l89*{suUEEMcS_Tc8f9)5b+gkf9MhxN?GyvYa?yL?G2$6cJqg>u53N>r;nU_DyOn6;2pm>~e_V1Gfhj;=ju3Cqwbv^i-m(mB_Vn+Oj{$a|ys=YjG->;+<JqUjukluh;P%0|waRD(z9L6BRtVC6aYKaY&3LD&VJCSl4Z<lh5C-J;WDqW0#MEB}%#G(oSKE4|q%`!#WD#(s#hdsY-ZxQjE}*<|-McG+=WyE*KM>0&0{0Ew(SNdw_uUs?JiB7lZ<ic}nJUNxe+MyQExergdioR26A>5aeZ;M44fgTIj+-q1X;f~Be@D}$zCpeFd&FoxLU{Xl;d;fy9sV@pB>VSFd6d%-{N*OgF!4!xJsFA`@MOae>9E}`Z#A;J1bZ~%oQvnw6mVr~KJ?&)+gteLYF}&q?0mMosMj#gV#I?g(o_B*k;x$TiAO$(VfCK~?e$|_(20M8D@#B0&>*}+s_~Mm)nuPiy<lVC!seAcD);WAEZP|THkDG!Gahxn*4vJy{7KRyFN|l@J_}<~4J(a78$~^q$WW%bTG4kO(7MRJGpI~(*qY_-DlN$4!59dDcis=q84#Tlr^bUkp0{cPe-4u+q5jNoolwpskrg3t?F=mfX>qK{_;k<?6>)>1b#mID^yZFwN3FvPnOULTY9H$<g~^8;mAJ!^KNww{_w!VYp-rvVv5U1q|6&y$Mn8IRSbAObBF7^^eL5#HTp{&NX<;Qe_R-IlVs-ZqR!#%VuNOt*mk|^QdEs^lNPs6ICI6|63&9)#F!})|ej)Q{r|XnT3n^(oIRVZ&W_v54n<~0RlJhgL%pk~M)EH*a!j0mQj>!h9Tv)hyd-<hh0{}()_;hd|YHFS53hoIk9?V*=pA+4f7cDs{i45s$94Lo;aAl@(QXaf+7iO$Pg<cdDn@e^|oi1Z3@i#^O&kgaB46#wD6#IiwzBlN)bNt@}7ydhF4a`U<7A++%Qi*W=F(=!>jjruoQqOJ_iVKUKE1O(Mc|fle*gnmrmrNN4wM=xXdYJ+P_lj$OcdPMTptQRr*Hw6Fyq)rP#j|gyPaN7<`BtYBR+BOcSpElx`8F9+JSwbxvhpCvcgMc}mhX*P158U$UbOKcZ{xF`{{VgtA8d?!oh{6mz)^nYmMD#GzZXn$9hr^7Nsk};K1R#nuzi8E1=|7>Nfy$!1uaSIWYqUdoqZ^(<<cN}@bZF_{^5gOXL1~r_6g!+GR~bjP8h|+wH{=A&HH}Im7hcNj&ozt#X|*qXylI(8yF)}xFPrU0=~zs!4G(26qd5O7>HT(I_P^HApFn%k&jq?^tA6!CjL10edi!gb<(N6dE0%EQs;4J97(bfUjWSBPHUDi-$)rt5t5jD5V+)(OprQl2PhvrrL3r>-4^z6unxa-b};PsI=0(ATIHaq!b?lHmX~ZNyRkfib(YrQpPUG$TBrZn?%}ZCAzJHFAXZ^vT1~CKuo0HZ@>dRI+zDDbOEzOU+wOyLYvfJ*^NFD2Fl#c^2G){#sAM51pRvh-3MIniG#m$)s0)(xtW6Ko9i*;gBpeT#D`p2OMib4Pxj8(qF50YI>Yn2i&r07%AIZ+?$tZvlOs%|D(C+m_ui+|h!9|NZz{zgT@5dV?$~*Co{4usfA&1XW2W1BZz5pxXOrh7WCI<pc69G^saQ<Dic%x%)UjCq|U|KOl&1ZNyLxnqKi&umHYBiw}T;S_=jt1ojzGi}nV0%HgH-LQv{h6uGc#XJPRNt8n2$i!-wo93l{@SJy-+Ru!_uP%M?7DGF$`f#wP19tlRzJu%OsA6}M)bicfy=sE>e}T(@iIyLkp5DVZs+Qf<3?{3uI!BKn>jJriA#~?)-Dq7Wm6HmsP048=H$3swL449uSoF4jE5~45ZCUW_WRL<yMj}?a?&nO;2&?43b7PD%qq~!SC$2Huvu<+&05V{ue|QDPM%DJanj0kCVuymBLoR8pFe%aBkhV*8JWJ9Gq<0x9is8`fAZI42Ni0H&)F#YrSwpf-t?4Mj~wTOQWlFI7`S4Bf78~k812V+U#=nLdgc|RpES+%ODqQcPAtShyiu4T%+eIkti&0TP$obQ{`4fLnlN^gGDwDflOS6LvDYy!v1+Q63`Zi#*b%EAKJl9My&Yr{*V1s-!tUwlgwoQ;=r*W}|42Bnk#E@_F&2T*c|K(=Ir@`5{eXnM&(L?yT3F0mu%cnEz{#7PlZbO7CPUF=F6_C;m9@RK<gxV!)tQaGt!5*S5~01qf(bLc)t$A<cT1YQh&8yp(H&D(OZr&kvwZR^nb}eFiEMzB3DR~}-aPHWtc=EZ_t=>Hs$aBStUj+11>l5T8J~6Lys{A9vD^n0nPo}Zk5rm&@@%qU9|6O0jMZb+oZHKJ<`kNXOH0cOi-l<^G*z)FDKratD~Zs;;n?wg{F(Jy*(B!EgfdL%7mQ%efaYXuV2n|UCZhaG>?kL*N<2av7g2|wNO&lTeEcvT7b)l#hO&IGy33(Vv4||FbhW<fCMtsGdLm5H`Uqnu8gD)l`Y$o=$04<p{wmn=56PuP!KJ~x{VJ-fl@3wwJbe$w2$iXbUEvOl@Nl|K3cY11a8DYYQxbX766vW(|5o>g{#z@InfgR}Au7L6juw+jHHmdjrb<HY`FnQEX4lsIx0~~`aQo%fQrlmivQ<^F>*oAiL_C3dWw%~i+grt{uqt<}l*`?<@Rqc8%k|gxDoxn8u7$^)e(cn&-Fju4Y;)E|wYj->$9lb9ZC11$eRoZeF8>UJ4uHIeFPj)B;8nhhW`UTK&SB?d=z(AX5>ZQCL5L!orq+&rAYEg<H(LgMw%$;PT10pSh-#pqT1KrN`XHKBxryEeep}=yBXi(Is&3qd0_g78hHjo97)-n0>YW5Jld@hv2XJ`D!(h~#!0;3xy}_ugr>KmixF-M9Y^!&q&1DZASTW85F=~UJfXGi8rS-3d<D=F<0QP$1bqtGwo@4|s{LB{VKw$!V6y@fR1MlqS!eY{nB3`2zMP(5`+m}BNfJue13<va3{5;4ow$Kd~elE35A`se*C9Stt%X<x2YJ7^vrz!HS$4oYuJF@R-3`T4iiMt%w7tBKTpof0Gp1ij`POcjJ7|!rU9bK2UDr*~+x?|q$9skVhwoZEe3(v>fn1g>}Mj0cbmxoSwM$3$`A&THT!y~CB-8GqK!cVXP4J<53vBa29u?CVEv#P`j_0e_OY(`11ISitg$0=!2iGCq67Pfm+Y%0;OCd1Kv`DvJvB-OH#UQq<ydsgd;sv-of@=?K$l!pv28Fm(9Jp_@4kX9NxquJ&XLBERGTEabrLTIXuMfroOYoj&r`{?JngBCcIB84{fSueoGxCFUiaIpySLEfX_jgBt@RDaQ(lHxzfl>Ndx0V3EX6ddKlJhR_qn1V@N6AI-#({D2QA@+0z@5$tcFohHPI?aiav9UDT5ndLfvES`;YT$GqRlD69k1sq}j0df8$Abx$(;7NnM%u&43?0M+-4_ICM=*i|;+?=|g`45HH$vATI2qAxEfT%}`8@Lnl$bB(iYW>xxVs?3|5B9oVWi?_^bRk)Nh`Q-_Ro_Dc!A#+d;%1gIDrYjJTbkI`}?Fd8TZc3KBjulfQt|tztamOD-v<t#3HcwJ!C;KJ;xyG2dXsa^p4gg?!jp|9->XY^GR_aD5S~dGI&#D7*mB}9+!~@u&`p0;~$6`G4p^Yy8<6*yKuVeWvd^4%fkTndIXd+xV9te@x^F3Irf8I;IX$o?DB}`eTqvGhc|0q(~hSr8ok;xJdz?)<>TEBE%gTN$pCo-I#t5U%GP}6;jqAVViI9=tI`E=x!YwXl5iKtJVZq~$=512HR1Uw3UUgBL~*mbXk?&bnFyE(xZPHNcocERmq$Ps1k$|mC5)BKvm*Cc)Dd5lq6sy+n+mupk&{7mO5i=Bv0L-T4v!d6T-c}WNf32)jdYJAKRvwzNta7m>c}7B(q2&RmlT&vSymDyd8W+nmf?jeVozDB(!DBWKsK<b1DNt?0nK%iE?N(x6sd^GIg(KlJb{v+ZHmyz5~-{Z24h~6GQ^!v4+e4t25(8AYQq6t<*FEUC+RV%cMMr(-mlv3kN8hw#5N|A<D(UreJ52eYm!vFFuF##P@0kslOLtw_K~mzcq8&gi+du9f?6_K-3>2F1F=<E3poNZD?0XlPtV|zbPbGL)DMQ<gVqJ?p~%iR_zH#s0ZqPe8&tfAf{;k~I~tiiydhDw5+uDGZ7sej#t^5nz=+PE&2A)_SJX*ST-CIRMAFPu*Tn8AXUCOEge5nvc_ylp6lu}yh={gaPrx1|Tn?{vB}{eATd#tQhF&#Nvm?~Rx@}ZB)^zc<qx?3yiYCRy21I00KJ>7m2Lg-L>|)a$l`41F=E*zchMgY?j5cS5Kb;Ic@hI0Fxw_dGB|{W{5dqYGE0}cR3Ve!sz9@~}%nMcJI{FrLM`qgrm(EL-8YV6yTL;?`s56e+wS@u<-G037YGb?HT-~%TvGZ5)_QUCxhL=R!(KQGyY>3Or_BCxv=pZgG-?EDW0uic9>dEbaW-AljwuL7NJVehDo>}Q#0(<RZg6E|>B7`@@rbgJ#w|w2q#%WK2(xnX(6%fsfVpO^@1`zsc7sKD`HC@%BdjeuFItV+AXGi(elfmoi%j<(N#v0I7V|c*D{XpYM0l~6kNVf()9{5tUnML?{ApxxJHEZkZ%+t?S4fTn{4&;28T_ATeBqp&s1my`vew#F7beZu594Ik?E<e738?o<`VjrGV|9pZ`nBpR}=n}F(40J_g=sbl-`{zJIEf7iC$B5arQ)_C`$Fe^7EQViVpR~>u*8v*?HUl!L`a<Af<zj^5AkosqVkiLzE~w+yX@CF$D69xU2k6#jc@J)!yOsEdaMHctWO&~PE}hVriYP;SjqTd26%P?k>&@6Fyrl!U+*?RDI$qyDL#2VewouSxP%q}R397x?+^)G6DuS&5_8!3qSRif?neAAivl0cG3m~t_a4rH9)9wSiGB~nR)?cx$$V>JRrQI%uFnRKZlJm>^0_Re~)XO)UwR-h?a1kb^8F?{|bjFJbb}4iXwhe2&d0Gt0?n-sH8Q*eBp~PebXRmVC;zQu0v&LX<;2$y2gN~}{THPMPUH}Qe*}+TZfsfK1%p2}-ixa(}Ump>H$*F0=h@Vj{54f2o@HYgIId35lHHQAfnYbWNrxhJF6OR{E51q1D(Lnwn+zl<=UA$2}sfpbAIy+_~H$0)h@!nug>;T$AjOV85%;XW&LbBr&PH0GggOrNbKuj`RnH>B7=4p$(?Odzl2kmihMAJa%7OgG@d=uYxgTLwaCbTmtsb1eYIz`;^iJurH^QtnFmtBlPak{bh5n357Y9f|whw)fMhEDg61}qTA#9t_?OSr5Q5==SXALJvD+pKJ@dCXHLZiobbZ#Qu={`~%jI33;b@GXB}vEF<h3FHYTK9_qC*q=}zMiZLYfVhrA1&*lXu2tV|z`vsq%|Y!acpXFbQ>(s00DqP^?!?zpJbyYsp3MLX9^q>Eyzh**iWMu|LKb$Hj(uPp%@J(b(?fJJY8|5WYYfB)$(kT1Y!C$TeXD#27}M1{dP&Es2$PC<`sGXjDj>VJPqB5RXXtN`VY$*FYS(J3d)qJ`RCBX<45Yo10IkiaP2$&@$OqeQ*x^JlTq*u72`kPuyj-TN9vM0Z%#A6T*BXwP%%G9VM@M^x(HNg1VhtRA-;%m-j0vFA6kk$lq?Kl=D4%*2NdS5_>;XXrCJn9{!Y(6<gYEZu^FaO9@bLR^lmcrqwEm`1+gY*h`~HZD4KndSeh!hWGqy&@Cnr6;4)X~Pbl*q@4odr%wuA<RrP#NTF+?$v(6NN+A)q($X9rRFm-M|*8*gD`G{+}cm8=1t$PceVPBYl&FAol4T8!{)Qsa>aP^-+A!Cedg5_Jb`iX;-J321mrQ0PStWhYxqjRzNz7hSqMx}k#P@W6KA;gE97*r${3+-=)I6pe07hEravIZ>zEH94>Lh_aRzQ@mJ{lFv-t)M|hu0=(~PCC^BR30X5ZK83rBLK7BqRMO=p`U^|I0Jo1xuaie@ptF+J9Ya~;80eIBq@Ph%5plKT*wuk<kpL(T8X|M{i}~*lfwmpmjvnFhz3Q&!)FlsE{;l4~j;%L5Lb)*iz21l<S@@C`Ru_a-nd+{$R$1Rd$2L`fIMvcE7x>QM1E}TJP4?Hujqp}6=|xCUHQ^u)UeUqnVacY14*}r~c2QV`ok3n;rSKIa4vP6O6_l^IputB(w)j?_{Mm9{7uvpI&)YY;`B>nF5a?kFV)E=?DZxU~vyD6U!HbJ8zPPmD+Ug-mjUL5fp(jG6vKI9YOj`}(M2g<Ll}El2bDcU;Kccm`T3_|n;qFQLAPQ22Lqy?Q&dj{ox~b_3RLN9BXIMDfM^|ff$^pJI4m~=Ed@q2N!)Px2|F>xINGk!SLYt9#<+Z(Py~0}$H7)3K6Q3Slh&&O(nWx4`_q6WD^hLd4{toI#s<j+u=Eayf$4OHYaf$?O81%p+ot$ZXpSP+ZquLjU{MO{SHL*x3Lt$;lzSSDHkKr=yD5pD;xak$T-GbC11}_|rlV=%KcJerfH@RUJps7+gm-Y?(aGw=Y2dQ%gA5kh_Z8fz?pr+&+@`6U|HMnyhmtsjk3}y7?haS+va<F4DJ(^_H^J$uJMlgf&$zi8e6rb_>-h8pJu*m;7?xAgGg_Ru3)ZA39lR81idghW+i3cy#c?XH9LbaDvE@n@aWMZ}41L-?ET-LZ5T^)|(Dr!}ej8LwxZdUIm-KEpdV`^imC-K`ljoPf}XR=N%e1;?WphPx8879mA(%4EEr0SB&D2!PRKc0*)iV4qm(r1(|<6vWEj19S(GRK2VIc9(fxk~eYrv{@m_Ca37W&}iINg!FRSDR$xVqfeu65DD|PEj?88vI236mh61Rakl(h6SbQs|WkR;DCa-XpJ!M3;rl!DC{}?TI8s>z&O9^$iaH}<I(_FE((Zo$zgm2*CqbY7y``S>qZLvRR0^LB?J_YxTI_O+Mu{o)F(!=mBGMlTc8TH@^TF-+q)oW(gI*^c-yt?JgR|mdg1-z?E{@*>>@^G|DyOH-|i0sUnh*{-p($GUTv?ryVpe5MD|eZvX_48lAQFTd{NQ8vhI)%c<D09HHo*(%9SoPkNc|?OB`hpd(Cg7#LS7<MYIW_s}W?45sCXghR3LsM$qlyp}=|_a-IrIok=L~I3m1Mkr?gmRA1Yxq%;t^ab>g;Cc@D2b%sq*9hoiVv}QmsNBx+l$Nwwst!5#F`ulyXf4x?(td<*1uU22H)M3W%mba>FQ=0L+U7wYEA#5(zS}JLUw2{qaTQ*c@gqfncvkDZm+GtjGnzlJkM69q375OO;wI8BCyc>xM_<V{tzx~W18?A1)eQVKgPw|6S#ioR8U?k3OBS>uo)%=b~ovqNyjk$6}6b1(id`xi~oNxP~Met9k(D_G!I*qkd#Jw0!?2}(oXvI4i!SN}<USY7pj|4!rL_dyo=(BikA`U1~lm6b*x2?N3^Au(DTa_=IB1E7UuJz)J3LaYk_HpFVdyMpFJ^kRxCm1&I;)^b4I~0KK8Xhx^RRg5r2N;a<BLx!nnJF7M#fCiOo^zU2m00gFM^BM1%(WD1@0VIERfy1ynOZSZA5HnkVbZ3Daz8$KC{~u%GlE7<X<cSa*<(!qnwmg})<8f*ea>L_+|?MsIgbGs`*U8GNIoBIBtOIXxK{kg!s5#d-CIjHQ_l@4an+yJbC_w)J017%<KxL>6cp#@$z%F>7)*+{7j7@ii(L%wNo5Tsmu}LND_1{@(WYfQN|oN-T)2I6so>w5D|Bxj&fQ$Tb$hPU>fD;U*;(`#TZ=C*9^P8Gb}Y*h#<JiESim2R%GUZE`!BfGdUdC~rR4>nD3<!(jwO9@U8_-ESFtP_#nd&QO_08%Y6X#WS@k`KP!LM>3hYs96dYr;KwzP@sDm)=DBhj+`W=g6rNcUR?s$!z@@``jE}T2Jm#va@?>R4gw7XSat!&n|@TTzYJ(mU$Bb+2b;HHftEOWUeLJST-?3uxlRo%S+>3(3{_s1YEtNa4cZxl1xE_6D<D{{gXGp?BUqEoo5=YYYd>;uqf5R*WYyfHCjqY_?-y%9B>b5=x~jDOJ^<)xq5No&wL^2d4h>%*tsxb+|rXWI5h6RSf1fkPm2|5}$<ah*zl8q4@&G8Y2-h#a11*9;*;p&wyPIDSK6;h7LJAg!0iem?L~lFyCp|90<Q@Z$6MPi|qqF!%Do-xc;3=9WObgGc}VsqJF5X8x5bEAlAz(H=q%IQ@98;5Y{*{MH!D${(fe8>ol1_;2|hT)3Xx!-<{K8MaG<LuPeI2n%v=p`_mN<#C`4mA7u@4{tBy!;YVeZS6d1qkWRzL<y;uc3Xibd`uFVf@O~J>m3x^$dq$k#Ly4h1YZ>F&lL{xyscduDTeTPqUZ+Hr4lQ+j!sx{ImuK<$@Ea+=y3b4VJ!9yC^L>vnd24`JD$&ZDec6saE3?p_BsKP5S9NX5w0DX1oN)vmvrm=Wy!_`vuD}x^kbI&-WJo7u0o--e^ei^s=}g&K|mluZC1*-)F~n?{%>%Kmim(6xU8_ORgzPf2a=TI;^y<m<#o|8MT7CYRp+I)4qNw;FrrtDF!8fik86@G3xZ1ETyVpQTrR`Wg*EIl3fDqme&JR=1^Ym5I<7SZ&H*jCC6GSLt4$9sJWR|kt)r{uU6c>j3`%#@L!=B+>?<h&i-}ti4aJ(++fjL;Y!f;bZr0EVjy4=uXczgv%KMt?$Mmu2*<$ROEx`apc!O}Ai)j_*tcCPQSJb7{JO<_pF7C{Vc@@mnns*Uvd&n?gZkC!8zIue=k?8xCV^%<kSrouh3?BxP+fB!;u0VVUeeyM9Hm}h^sT8j|ek4~bTU$b?H7mZ$r-EHTgXq!enB5*uP6J)JEoPec?TXbJAND5W7H}maU_a;ROpM6)j1!DdwuoMzb?I{bW9J2`^xEAcRu#!9G?2j<(JKj}N6tQK<~0l$`*}E?c6>&mCAddBKG5I+<JDm!tzd#m(7?}$-Z=TgVdo--HZyRsT$$3e5a>**8XA=qI!i(y?x0nSv+(769q9k+`i9MH4)$Rn^bvt`knA#ft(joTgoaWL%PEx#2FhUO_wau->`n<g*hf_$YPrr@{jkRUg#*TgNW*wh-OGu(N-!|fwGrnBGY9qz_h3oggLxs4;6>Zc$HXuL%8+87WR{>Dv>In9f=LqWBM-dt;Lu)~oQz6PAdfrG3r@S;-g(aEm$<ReeOk4__EKoTg|A|bz!SxfKz*^dg4PnS7q&*4aQrKm@Fg(Pk!J1H%8m?`)+&>`h~7J|6c(BEk$N2?sVjGCwO1)lv0UA0c*22*N~idEN4QZONfr|+dPk@v?DS3R?YvT2T&DAajy%zImP#l)+$XV53kzD;QG*5yW2ITKos?~+e}0Xf2K+UeLE4H-{j2S*g{4=}-V3|V&QL!CiBS90+>Mk*9c^Gjt4wV0>Q<$^V}i{=vYX{69x|(~5r-YAd^nujSr#yPx2wAsyGiGZFT?N~Td*XaUKFlO@z_vj<O=fGnK_W3+acXh+y_RaPcD>aI&dd2!=iq^Ak-(x)`ZVk%$FZepQ>N<m@nIyoGN@vqCMeB7hW`c;`yLzD0EyV%bUk>r<`jcM&kM6+QQe>-8XjbxKSQ}{ijRma%J3d5&e7v-Sve10{5n=^8K_WIbfQSvx)n_&)NB-wpJmRyY;=TE$MC|GtMUVTld~`a#a<RPHR)TG<U|l^$xC9uyytYPx~}`6CPaMd(Ld=k!Mr0^hj#xTEWF&(mHpoZ9%`79T?vC1t(`AKrlM)U());<3HF5LEi;YjREd1FKBfMPA|bIM1b?q`0FT@3?c&FcZ%WnEk|H)kTczxj4-;&t~5t*i!Q-02L8lR4eTds=6VmZ>(g-+MNKAfMZ;%Ew><R`XQ-e4)pP*H>meG7=93IZCbnYMSJCNCz&pA+`eQi#uztai_h8OULe5NEstre&fJOIh6dwu;-fnfbvQ^zdGGIhnz)fGTZev88h;vz6C`iJAq26nItFNxzVdw&6C@{zmUWGv(@CV0*bM=w%aN9LgBvW7sMyJ3lG<jl{?#fP6#!w^P&(3Oft4a!FaZG_lfQuCVC~v8@WmvK-f}NDV8OA{VMPWgx>4>@Q(o3=+#@cFD>M*LNe3}4l(Vqzb6a7&P+(XtIwK40IDDgqGpJ70?ShX-GyQ}P!7&H|Av8eyCr2la<ksFzMLrbnU2|iK5lk#cv^%_U`Dqsp2_JOYAzbp825=I{!5BumJ(82J1O9c;G?1Icr)-9{+KR}uG%pde9C18kOqY0wQB`yp>Mz3ZNAnff0>(B3hXl>)@w+s9m2fw|@zj5r_OZ@w$-djCg`YQ`WA~ct1H|lwef69*8SE0WMrC67dfiaSfU-bP6xesDrRe8h?Pv&M1Q_wip3sX@*MZ;ki^B^Ju7fNbBC(>OFvwAWVu7jGr9CcNuu4-k%S=U>nQfVvWgmV<DT&ouW5l=8SN_|{P#FEo&C7%k{1Dv^n^Fs2y^bDt|t=G1;(3K8#Yql=G6W)eIU@4J^lkAhj-civgFoS|2`K3@9<OipN-kYa>?i<vk1!_p#EH4zA*dt-4iX_VwM*0(uHf#@jZ|Bt=ps7$DHY01*)@v~Ix(LtUz1b=$ryo)POQbJ{QND<Y1TvzzPMEY>CQRv6II;VGSz@}x@~xT6(sGw$iudX&5Vbp7wbfTaoQml3g<&chM8!kdShk&FRz*_fYHwB>&03xI-Ntf6Tk9m~OdS3#WJ=<1W?JhQt(mF2>u~SJ2zYv*DRPKBV^nB8#p<^v9*~v$A}Z*Wg=guAx><Mv-+*oadmupHBE1mXfY+rELYwepFcWck<wW<O*eLe1bPXDL9k`1I`kHOk-oSt;R!^`kOE3$W{nU1aegN>ZZ@ag8{da+Ffu3_Wr_ZKhR_?-0kHO9i()cZRP%=!|ia>!RU5G*5qhS@IR_da}bW&@kUfF`<Q1O5Vti29{JD?+El2g@lrAEm^cZA}3jywWwK7e?bv}p;F!!L>9hw(*{?~YJ2QeznJNqQBzX~fZ2#8i=jtU0AZy1T1jTH>C&uPitTG8doQV$w{Uqp9<#2N>ufl?g2B8}u^dELF7plTnWoxyySVL(^p>Mqq`}h>;ZZBD-_oS{uu*B}hyU?Gi8~Mh64t@p##ibGCwMjV*~{X~#L3tRax3u?`EMbv8tdq5x%=g3Bc{W3|Q;A0c44JK^lfH1)k`^i}vvI>ZQ7G@2_UrX6dWOu30<iJUe;#sm%UzNAkgY{*N6)=8L}L=YOAa5!SBfKlrLeYG(p=6vLjdz8n%Xrne%D$8M#Qgx6gq)ye2nqfmoz;k~z6h%>2Lr51wO*cF_=}i(y7pXNJ!{j5jLyE~z;70l9Q*4m-WQfO%;V=YVSrkvhR+t&&F%zinjs3RY8%+X3r5g5G2*4o}vPe{}hqz(twm^44bdxbchhpj#!gGKQ_gnO&buOM<8G%h=DZVKh7(Gq$%B44CO`*~ONB(GdC?Y<`RJRr#1!S*7VxRab^s41bb@;(VoPn|56iyxng^TYb;p3jN)G-`P#xNHWAO(dvW0hu1)<hFYc-CpvDNIyt+Kfp0hApJ<LPMA+PabFET`hGDnYGOfI*MRHxH35#v6si4AiLf;Fly?6b#$oA8qWd()IB7`Pr{KlgGL)VDs5t>;vx`a-$bOodGNgzhs431KLbkiY_mND2s#t{*-+i_vJOJ6KwA<b$4q@w0LlVTS(uXwxE7wmPrg&G=ea=ATVug!-(gf|b&ixH%6}3}fEB}P5eO}k_Rj{<l2Qr9M~S@=$5Wqe(q$O^<q_Nj0}U#K@T$6wE5J9x7u_z$!XLR^ib#&hxYAT$>c3<qLOzO_QDsw#Gm^q06k^nQ^l5!>C;E9R7pdaO72yjFSTWxGemZZSqE;Y%LWn!Al*kEnai}29*3>#q%mikd#Bhp|LrR&y*{V~_Z*l*AI5sJbSWyA*e!3ucVBIiISZ@D-Y}#fihG}Z3ru4QKyzQGxlmlw{+8*)UEa2^|SL@p-q&eBGMFk_#h%!<nG|rl<z_glm0ZY`oj>urjk?*xK+AbOxwu5zLN?&Hvx+i2F{o=;GFZzf?nwKPdVQDJ>rHegrC4h<AM`oP-)RPRi>!j7|D{dkk0(3T7Bmt*90Y9L_c+OUX1mJuI>^JLMz;2UsGaD}XY(q+5mkV;A4qHB3y$WVdpv$2jK*-U1M<BzH2ha>(P6a}rr4lhLGcNiyhh)M@pK${vK=9sdNAb^RtLIrE@Zv`oW4FD2@1!@8V#L0QCfOUyo~9R+x;jnoX-&~ci<`m>KB1|0Bbha<OyzT$;(6ewD4sFKGrJ#hj9v&PtCYQ~^;)CxdU@+r1y(`*n^n=A-S9?1s%e4ch7+gYui<fHZyjhl(stYWraI;9UP$wIT<xn|?X7P|SDfAplduw^Hk#lX87RJU+P>d8<b%mVbsLebLze*Q4G@0D(DBnS<6SH2a->keBmO*wUY`yaOunNLJ&-GWg=@OE!+7#*+Gcwlu0<0QUEtVo4=+@oVEs9k952+0`RIW#SdeZ}W-oJ7x&dRJftg7X;m>=BMjt5n`Pn|Ei!ctLJ)?jxk+nJ8DPTnHESVd%EEzJklJBP`qnSmp*NV^7Xab8csneOI*-!yAOQX{h+{~oI;G$VA3XfEf2N}l+gu*z=v4&8ba8Oe)RXKJ*B*YGcV!Fy!a@a-q!?pc>YlIrV0GXbS&?u6J(+My^6BHYA&KwZ9degMiNO=U*LZ-5@cSF!bIMu^;FT{YPupUg&iueF+L~yVO#}C!>A!P1o2t5SZgd+U&aizE9xl%klL#H>P9d}ILo`nA84O)ZidVdOUVrmd;4|1wM8J^+JzS6=UkgI+~53jgL1M-1U@F>~%Xpdc)8lki%ZD|{1%+w%Vq816Gk>h!5qc_<+JtXr5As7Y(VmM$NMOdlR0gSMHj6TZNp^u1K7U~6bOG>MaaRzvL@==?J$Q%n&0z(CPM{p$}m``r(pA09yOIjlLa17rbyCO&o^+!gN37T|ZAf-$D1eirih|s<F9GX3>%jo1@6|KwN*5p_;<IwTCy7ydSAKRi$yC|bJZ`~9jMf5%O&*3$ugP040Ht}py$Khl8Z_;B^8&Og8o}iClJ7}+>pA8ScUuundXb?vI#=IyP%-c6k$9=S!Q0zzk%~KeeoJZXGC3_z31l*z}(Uz-Fcyfyi3-01V0eX)GQalr(>3y3mJNAL-s4E`wG|8vYWoA!$Pv~Ry^%W~isd0-b<AfX5c3~mIkzUR)(5wH7d7e!rMVnj_3f_CpJm|yz@URpvyI9reaEY$NBWB&?V?opt-x?6b)Bx>yn4WoT6IYU2p_!C@(;GF?o|wf9YjUE<TH+T?wky@&z{RcIDX+d70oFpqoy7TYjEg#k%NKzVh*Qf+Gnm?0@l}QOD3y1sH|%-wVKC1;9t;{ipJfH{4OnUbuMTIICK@Vn8L9|z|H?#N;D%|9M)D4uA421kIfgE?uU!7YC{=ie!PQ#IEiJeO*!L-seq$jbQ7B`nga^cylvGU`N?`EDPP!)BWCF(Pcbpe}<xR56%Q)A(82Pa5AZx*$<BxMDa54m;D2>GqSQ$%Hqm|PVzTx@~QCseh!!$lHJ5+D7x=|@|9N-U*v^!0YEFR^9_OVYml!S4?{uZM=Vq6SF0*sunwB&&_K+CYF^V$O^oHhoJ+^cVyt~A+jY^krcbkmUW%&~W=3r=Gd3n@aiW?jtqpjBQa^o=tk#38?S$ZZf1m^kkd%)OMzLD+AHMTNI_7!Rxxs9579DpooSpFF_vff(sUdKd)pmBFzJK1pnI(RxBzKGoevxIeHFLg>1IIpIh-diYv7)IP++_s}W7l<?Wl4~7pyPk#L8E#Ncqr;|4F)#ENc$=Tm|<2xtc=`_Ey`JL_WG`<(Fny;4<LFmp2F$;)W%tc8?7TS5(HgmRf1OB-|y^8JF)=`WxOe@JeL9L<g61e{YoA3Ke5Jb8_A@R{XZeZ+@xCiZ+kH%GmD=4_}Gd95cgZtE|Y;2Q8oxo-}Oa_U<%6VEDl?FZLl$2IpanZ?d(s@Vfq02%K4>G~dob{0FcD+)1Ei?uenv9h^Z<VdixYeCl;wBSpZea0+CYfPz0x0|qmosne@)L5LhX-0-fp`m8Lh|Rd^Ho#Wgmv6HL!R45^pd&IGlfdZ0~ZhR<_b!Y-6F4IwOgQ9!cq>uRQ@LB{GjKOp-r;O{1Let0SiGN(;Nx?wM}XMs4=_9B37abjP}OtDxA%{YGGX~x-@}jeI-P71v)ebq#l|R#EV7T_$D^8cp+9e0Y;Ax({VRXN90o_?<Neg&jC{^fuHzwpj~|PXn+S^Rr8Lz3Ynfu@r%LOuLKihkU%&tymf+{4$m&a5R$w?c$=lJ#tKEQqbNg!pxLgOS2`PXM89t^RPn4V5)V3lZ5F~SV{e!4yv=5F*JJ)J`ANKPnu(GqXIeCaG{GlcY)U%#&WfSfxn2xUwz*~lODiz*wH@Xp-1U?@pXR5hJF&;bK|EJSe8LikKz0FnxuV)C$CHyjJvs<6tUcv<R@8f%s8N-tDyirc{?Tx{MRY3Hq3wQc*GKm|N+I)?Lj0Y-q;)dN+3Xqeto5F!)=S9}$g&msBP#NbYVFQpt9>7zYaz=<TU*2rXmm^@jES3cH&mX5O!+;eqPku!CjX~9geH)kgtd_ek$G#;P3kJFHsj4@I2kAU$&g>0ws>%i2o-dB5aXxa9zueP5H^EmP+T$MTHs3{^@h2MsVGwDxD-}o4|!*E6ND)*nhE8%ZGul&tC!cC$w=;QDK3}8vw2v~6eSBDVX00qcM!il22sxM?%SO2!=Ne=oP4YhMtTwdOVq-kLKwiYL_DNKr$vXkZP~9C(G~UCln#lh6<?L&dr?hkv%CR9PKEIgc5Bt0rq`^g;Dpd5zFOPft~Tv&kMe5q-ipu)Kk&NXmT-M6m=W~#?vbyz8)d5$Gr=+6$WvI7HrVNue&3EwhsV0R*i)k}O{>NvJ1OVx(kx!zDIxq>vQjZA8Ha5qrM7H5L1dczbCUwYv^eodF_(fF(YM<puG8~w*r<U*Y64I9i-n~H6Ar#{j*_+8oJE_?Q8Cffl>w?fN|;qM<~W4t;G#<ee-F7jVl{!Kq6AdEHta==P-!2>gAPb_j$Xg1!91~e;`A66tXauVDY;_46*~f*3I%{n0HPB8_%|UgF994s`I4Z3A3ga3VP_Flmr#YD{+Qrmf3l(wcc1<c3t9;O8pW6W`pK71-%~KSPv1*Khs-oU>4cZSl8*Ye4oaol%L^{d2WMK`sD$s$KsJmVdODzEqh<vTJ*Mvt5ZVk4V2E7Nx7g#$3X}YD2+EyWb5mxuVC*h~EcfPV&&SLGXq3~tr~2f^%t@B5TWH<TgWlF1<6YhN-brb03ymI@uuc|j_r-lYspkwPac;0Uh8)F#8~CG=eed1?foEg8T%DVJxtR}2(Kyhe-Vd#~&MqHBkhA~7LKNHxnlA$7J|}TN;K5D(Z||fx7>?oT@?Tv2Kj0g-ng'

def _executed_cell_sources():
    # The executing IPython kernel history is the only authoritative source for user-edited cells.
    try:
        ip=get_ipython();history=list(ip.history_manager.input_hist_raw)
    except Exception:return {}
    result={}
    for text in history:
        if not isinstance(text,str):continue
        match=re.search(r'^\s*#\s*(0[1-9]|1[0-3])\s*·',text)
        if match:result[int(match.group(1))]=text
    return result

def materialize_own_notebook():
    doc=nbformat.reads(zlib.decompress(base64.b85decode(NB_SNAPSHOT_B85.encode())).decode('utf-8'),as_version=4)
    marker='NB_SNAPSHOT_B85 = "__SNAPSHOT_PLACEHOLDER__"'
    ensure(marker in doc.cells[-1].source,'SELF_EXPORT_TEMPLATE_NOT_FOUND')
    doc.cells[-1].source=doc.cells[-1].source.replace(marker,'NB_SNAPSHOT_B85 = '+repr(NB_SNAPSHOT_B85),1)
    executed=_executed_cell_sources();checks=[];missing=[]
    for c in doc.cells:
        if c.cell_type!='code':continue
        match=re.match(r'^\s*#\s*(0[1-9]|1[0-3])\s*·',c.source)
        ensure(bool(match),'SOURCE_PARITY_HEADER_MISSING')
        number=int(match.group(1));actual=executed.get(number)
        if actual is not None:
            # Replace baked content with the actual executed source. No outdated copy of cell 01/06.
            c.source=actual
            state='EXACT_RUNTIME_SOURCE_CAPTURED'
        else:
            missing.append(number);state='RUNTIME_SOURCE_UNAVAILABLE'
        checks.append({'cell_number':number,'source_state':state,
           'exported_code_sha256':hashlib.sha256(c.source.encode()).hexdigest(),
           'runtime_code_sha256':hashlib.sha256(actual.encode()).hexdigest() if actual is not None else None,
           'code_bytes':len(c.source.encode())})
    expected_numbers=list(range(1,14))
    ensure([r['cell_number'] for r in checks]==expected_numbers,'SOURCE_PARITY_CELL_ORDER_FAIL')
    # Audited configuration must match the runtime state; arbitrary post-cell patches are not silently exported.
    run_cfg=next(c.source for c in doc.cells if c.cell_type=='code' and c.source.startswith('# 01'))
    cfg_nodes=[n for n in ast.parse(run_cfg).body if isinstance(n,ast.Assign) and
               any(isinstance(t,ast.Name) and t.id=='CFG' for t in n.targets)]
    cfg_exact=False
    if len(cfg_nodes)==1:
        try:
            parsed=ast.literal_eval(cfg_nodes[0].value)
            cfg_exact=(parsed==CFG)
        except Exception:cfg_exact=False
    for c in doc.cells:
        if c.cell_type=='code':c.outputs=[];c.execution_count=None
    nbformat.validate(doc)
    dst=WORK/'ML3_FACTORS.ipynb';tmp=dst.with_suffix('.ipynb.part');nbformat.write(doc,tmp)
    tmp.replace(dst)
    source_text=dst.read_text(encoding='utf-8')
    ensure(TOKEN not in source_text if len(TOKEN)>12 else True,'NOTEBOOK_CONTAINS_GITHUB_TOKEN')
    verify=nbformat.read(dst,as_version=4)
    ensure(len(verify.cells)>=25 and sum(c.cell_type=='markdown' for c in verify.cells)>=10,'NOTEBOOK_INCOMPLETE')
    for c in verify.cells:
        if c.cell_type=='code':compile(c.source,'<ML03>','exec')
    ensure(not any(c.outputs for c in verify.cells if c.cell_type=='code'),'NOTEBOOK_OUTPUTS_NOT_CLEANED')
    report={'version':CFG['VERSION'],'executed_capture_available':bool(executed),
       'runtime_cell_count':len(executed),'missing_runtime_cells':missing,
       'configuration_matches_runtime':cfg_exact,'source_parity_verified':not missing and cfg_exact,
       'export_sha256':digest(dst),'source_cells':checks}
    (WORK/'source_parity_report.json').write_text(json.dumps(report,ensure_ascii=False,indent=2),encoding='utf-8')
    return dst,report
IPYNB,SOURCE_PARITY_REPORT=materialize_own_notebook()
mark('SOURCE_PARITY_GATE','PASS' if SOURCE_PARITY_REPORT['source_parity_verified'] else ('FAIL' if CFG['MODE']=='FULL' else 'PARTIAL'),
     f"missing executed cells={SOURCE_PARITY_REPORT['missing_runtime_cells']}; CFG exact={SOURCE_PARITY_REPORT['configuration_matches_runtime']}")
mark('PASS_NOTEBOOK','PASS',f'nbformat, syntax, Markdown and token check passed; {IPYNB.stat().st_size} bytes')

def json_safe(v):
    if isinstance(v,dict):return {str(k):json_safe(a) for k,a in v.items()}
    if isinstance(v,(tuple,list,set)):return [json_safe(a) for a in v]
    if isinstance(v,np.integer):return int(v)
    if isinstance(v,np.bool_):return bool(v)
    if isinstance(v,np.floating):v=float(v)
    if isinstance(v,float) and not np.isfinite(v):return None
    if v is pd.NA or v is pd.NaT:return None
    if isinstance(v,(datetime.datetime,pd.Timestamp)):return v.isoformat()
    return v

critical=['CELL10_PIPELINE','STAGE1_PHYSICAL_AUDIT','PRIMARY_TARGET_D01_GATE','SQL_DUCKDB_UNIT_CONTRACT','DATA_SMOKE','REAL_STAGE1_SQL_PREFLIGHT','SOURCE_PARITY_GATE','DISCOVERY_EXHAUSTIVENESS','INDEPENDENT_METRIC_RECONCILIATION','PASS_INPUT','ALL_STAGE1_DOWNLOAD','PASS_DATA','PASS_SPLITS','PASS_TRAINING',
   'PASS_VALIDATION','PASS_INTERPRETATION','ADVANCED_ENSEMBLE','ADVANCED_SHAP_QA','ADVANCED_TEMPORAL_CANARY','PASS_OUTPUTS','PASS_NOTEBOOK','TEMPORAL_CANARY',
   'MIGRATION_TARGET_SEMANTICS','FEATURE_G1','FEATURE_G2','FEATURE_G3','FEATURE_G4',
    'PASS_UNIT_AUDIT','PASS_BOUNDARY_AUDIT','TWO_YEAR_CALENDAR_QA']
# Validate the whole ordered canonical design, not a few convenient comparisons.
required_canonical_pairs=['M0 → M1','M1 → M2','M2 → M3','M3 → M4']
if len(ABLA)>0 and {'study_id','confirmatory','comparison'}.issubset(ABLA.columns):
    completed_pairs=set(ABLA.loc[ABLA.study_id.eq('CANONICAL') &
                       ABLA.confirmatory.astype(bool),'comparison'].dropna().tolist())
else:completed_pairs=set()
canonical_confirmatory=set(required_canonical_pairs).issubset(completed_pairs)
_primary_canonical=CANONICAL_YEAR_QA.loc[CANONICAL_YEAR_QA.horizon.eq(1)&CANONICAL_YEAR_QA.protocol.eq('ROLLING_TIME')]
full_calendar=(len(_primary_canonical)==len(CFG['TARGETS']) and
    _primary_canonical.target.nunique()==len(CFG['TARGETS']) and
    _primary_canonical.status.eq('FEASIBLE_COMPLETE').all() and
    _primary_canonical.missing_test_years.fillna('UNKNOWN').eq('').all())
no_fatal_qa=not any(v.get('status') in ('FAIL','PUBLICATION_FAIL')
                    for k,v in STATUS.items() if k!='PASS_GITHUB')

SCIENTIFIC_EXPLICIT_BLOCKERS=[]
if not GROUP_FEATURES.get('G5'):SCIENTIFIC_EXPLICIT_BLOCKERS.append('HISTORICAL_G5_UNAVAILABLE')
if len(H5_CALENDAR_GAPS) and not H5_CALENDAR_GAPS.status.eq('FEASIBLE_COMPLETE').all():
    SCIENTIFIC_EXPLICIT_BLOCKERS.append('FIVE_YEAR_CALENDAR_UNAVAILABLE')
if int(PANEL.boundary_year_at_risk.sum())>0:
    SCIENTIFIC_EXPLICIT_BLOCKERS.append('BOUNDARY_HISTORY_UNVERIFIED')
if STATUS.get('G1_DOMAIN_COMPLETENESS',{}).get('status')!='PASS':
    SCIENTIFIC_EXPLICIT_BLOCKERS.append('G1_WAGES_EMPLOYMENT_INCOME_NOT_VALIDATED')
if STATUS.get('HISTORICAL_PUBLICATION_DATES',{}).get('status')!='PASS':
    SCIENTIFIC_EXPLICIT_BLOCKERS.append('PUBLICATION_VINTAGE_NOT_VERIFIED')
if STATUS.get('ADVANCED_VS_BEST_SINGLE',{}).get('status')!='PASS':
    SCIENTIFIC_EXPLICIT_BLOCKERS.append('ENSEMBLE_VS_SINGLE_NOT_COMPLETE')
scientific_ready=(CFG['SCIENTIFIC_RELEASE_ALLOWED'] and CFG['MODE']=='FULL' and CFG['VERSION'].endswith('-final') and all(STATUS.get(s,{}).get('status')=='PASS' for s in critical)
  and canonical_confirmatory and full_calendar and no_fatal_qa and not SCIENTIFIC_EXPLICIT_BLOCKERS and len(METRICS)>0 and len(PREDICTIONS)>0)
if not scientific_ready:
    mark('FULL_SCIENCE_GATE','PARTIAL',
       'V4.02: empirical completeness requires documented publication vintages, G5, boundaries and eligible h5 calendar; see blockers')
else:mark('FULL_SCIENCE_GATE','PASS','Full empirical hypothesis test, all required artefacts and scientific validation')

MANIFEST=WORK/'run_manifest.json'
def write_manifest(publication=None):
    pub=publication or {'status':'NOT_PUBLISHED'}
    payload={'experiment':'ML03_FACTORS','version':CFG['VERSION'],'mode':CFG['MODE'],
       'source_repo':CFG['REPO'],'source_commit_sha':source_sha,
       'source_handoff_sha256':hashlib.sha256(raw_index).hexdigest(),
       'total_assets_stage1':len(ASSETS),'verified_inputs':source_receipts,
       'all_stage1_downloaded':bool(all(a['asset_id'] in LOCAL_ASSETS for a in ASSETS)),
       'years':{'first':min(YEARS),'last':max(YEARS)},
       'n_municipalities':int(base.entity_id.nunique()),'n_panel_rows':len(PANEL),
       'n_supervised_candidates':len(TASK_PANEL),'n_predictions':len(PREDICTIONS),
       'n_model_evaluations':len(METRICS),'n_paired_contrasts':len(ABLA),
       'n_advanced_superlearner_evaluations':len(ADVANCED_ENSEMBLES),
       'n_advanced_shap_local_rows':len(ADVANCED_SHAP_LOCAL),
       'advanced_explanation_status':STATUS.get('ADVANCED_SHAP_QA'),
       'advanced_temporal_canary':ADVANCED_TEMPORAL_CANARY.to_dict('records'),
       'advanced_shap_interaction_rows':len(ADVANCED_SHAP_INTERACTIONS),
       'advanced_vs_dev_best_single':ADVANCED_SINGLE_DUEL.to_dict('records') if len(ADVANCED_SINGLE_DUEL) else [],
       'github_colab_secret_name':CFG['GITHUB_COLAB_SECRET'],
       'publication_date_audit':PUBLICATION_AVAILABILITY_QA.to_dict('records'),
       'organization_grain_exclusions':ORGANIZATION_GRAIN_EXCLUSIONS,
       'h5_feasibility_report':H5_CALENDAR_GAPS.to_dict('records'),
       'progress_widget_count':1 if _PROGRESS_BAR.widget is not None else 0,
       'excel_shap_local_exported_rows':sum(r['rows'] for r in EXCEL_SHEET_INDEX if r['source']=='SHAP_локальные'),
       'supplemental_bdmo_receipts':BDMO_EXTERNAL_RECEIPTS,
       'supplemental_bdmo_load_audit':BDMO_EXTERNAL_QA.to_dict('records'),
       'groups':{g:{'n_features':len(ff),'model_features':ff} for g,ff in GROUP_FEATURES.items()},
       'status':'FULL_SCIENTIFIC_PASS' if scientific_ready else 'DIAGNOSTIC_FULL_RUN_SCIENTIFIC_PARTIAL',
       'technical_status':'PASS' if all(STATUS.get(k,{}).get('status')=='PASS'
             for k in ('PASS_INPUT','PASS_DATA','PASS_TRAINING','PASS_OUTPUTS','PASS_NOTEBOOK')) else 'PARTIAL',
       'scientific_status':'PASS' if scientific_ready else 'PARTIAL',
       'scientific_blockers':[k for k in critical if STATUS.get(k,{}).get('status')!='PASS'] +
           ([] if canonical_confirmatory else ['CANONICAL_M0_M4_COMPLETE_SEQUENCE_NOT_CONFIRMED']) +
           ([] if full_calendar else ['PRIMARY_TWO_YEAR_CALENDAR_INCOMPLETE']) +
           ([] if no_fatal_qa else ['FATAL_RUNTIME_QA_PRESENT']) + SCIENTIFIC_EXPLICIT_BLOCKERS,
       'run_completion_status':RUN_COMPLETION_STATUS,
       'scientific_ready':scientific_ready,'causal_status':'CAUSAL_NOT_IDENTIFIED',
       'status_details':STATUS,'task_plan':TASK_STATUSES.to_dict('records'),
       'selected_models':DEV_METRICS.to_dict('records'),
       'factor_cutoffs_by_horizon':FACTOR_CUTOFF_BY_H,
       'target_specific_test_windows':TARGET_TEST_WINDOWS,
       'stage1_physical_qa':STAGE1_PHYSICAL_AUDIT.to_dict('records'),
       'economy_manifest_audit':ECONOMY_SOURCE_MANIFEST_QA.to_dict('records'),
       'regional_context_not_municipal_g1':True,
       'regional_context_dictionary':REGIONAL_CONTEXT_META,
       'regional_context_audit':REGIONAL_CONTEXT_AUDIT.to_dict('records'),
       'regional_unit_qa':G1R_UNIT_QA.to_dict('records'),
       'feature_staleness_qc':STALENESS_QA.to_dict('records'),
       'discovery_exhaustive':DISCOVERY_EXHAUSTIVE,
       'scientific_open_limitations':[
          'G5_HISTORICAL_VINTAGE_UNVERIFIED' if not GROUP_FEATURES.get('G5') else '',
          'FIVE_YEAR_NOT_FEASIBLE' if len(H5_CALENDAR_GAPS) and H5_CALENDAR_GAPS.status.ne('FEASIBLE_COMPLETE').any() else '',
          'HISTORIC_MUNICIPAL_BOUNDARY_CROSSWALK_REQUIRED' if int(PANEL.boundary_year_at_risk.sum())>0 else '',
          'G1_INCOME_LABOUR_COVERAGE_INSUFFICIENT' if STATUS.get('G1_DOMAIN_COMPLETENESS',{}).get('status')!='PASS' else '',
          'SOURCE_PUBLICATION_DATES_MISSING' if STATUS.get('HISTORICAL_PUBLICATION_DATES',{}).get('status')!='PASS' else '',
       ],
       'data_smoke':DATA_SMOKE_REPORT, 'sql_duckdb_unit_contract':SQL_DUCKDB_CONTRACT_REPORT, 'real_sql_preflight':REAL_SQL_PREFLIGHT, 'source_parity':SOURCE_PARITY_REPORT,
       'unadapted_municipal_wide_assets':UNADAPTED_MUNICIPAL_WIDE,
       'g1_empty_cause':G1_EMPTY_CAUSE if not GROUP_FEATURES['G1'] else None,
       'target_availability':TARGET_AVAILABILITY.to_dict('records'),
       'distinct_seeds':SEED_QA.to_dict('records'),
       'test_metrics':METRICS.to_dict('records'),
       'ablation_results':ABLA.to_dict('records') if len(ABLA) else [],
       'advanced_comparisons':ADVANCED_COMPARISONS.to_dict('records') if len(ADVANCED_COMPARISONS) else [],
       'advanced_model_selection':ADVANCED_ENSEMBLES.to_dict('records') if len(ADVANCED_ENSEMBLES) else [],
       'advanced_additivity_audit':ADVANCED_SHAP_QA.to_dict('records') if len(ADVANCED_SHAP_QA) else [],
       'outputs':{f.name:{'sha256':digest(f),'bytes':f.stat().st_size} for f in (IPYNB,XLSX,ZIP)},
       'publication':pub,'elapsed_seconds':int(time.monotonic()-START),
       'interpretation':'Predictive association only. No automatically inferred municipal policy effect.'}
    MANIFEST.write_text(json.dumps(json_safe(payload),ensure_ascii=False,allow_nan=False,indent=2),encoding='utf-8')
    ensure(json.loads(MANIFEST.read_text(encoding='utf-8'))['version']==CFG['VERSION'],'MANIFEST_INVALID')
RUN_COMPLETION_STATUS='COMPUTATIONS_COMPLETE_RELEASE_GATED'
write_manifest()

# GitHub writing is intentionally unreachable before FULL scientific acceptance.
def create_or_update_git_file(remote,local,branch,content_sha=None):
    request={'message':f"ML03 {CFG['VERSION']}: {Path(remote).name}",
      'content':base64.b64encode(Path(local).read_bytes()).decode(),'branch':branch}
    if content_sha:request['sha']=content_sha
    obj=api('PUT','/contents/'+urllib.parse.quote(remote,safe='/'),json=request,expected=(200,201)).json()
    return obj['commit']['sha'],obj['content']['sha']
def verify_github_file(remote,branch,path):
    obj=api('GET',f"/contents/{urllib.parse.quote(remote,safe='/')}?ref={urllib.parse.quote(branch,safe='')}").json()
    blob=base64.b64decode(obj['content']) if obj.get('content') else base64.b64decode(
        api('GET','/git/blobs/'+obj['sha']).json()['content'])
    ensure(hashlib.sha256(blob).hexdigest()==digest(path),'GITHUB_READBACK_SHA_MISMATCH: '+Path(remote).name)
def verify_release_asset(asset_id,expected):
    url=API+'/releases/assets/'+str(asset_id)
    r=SESSION.get(url,headers={'Accept':'application/octet-stream'},stream=True,timeout=(30,1200))
    ensure(r.status_code==200,'RELEASE_READBACK_NOT_200')
    if 'json' in r.headers.get('Content-Type','').lower():
        meta=r.json();r.close();u=meta.get('browser_download_url','')
        p=urllib.parse.urlparse(u)
        ensure(p.scheme=='https' and p.hostname in ('github.com','api.github.com'),'RELEASE_ASSET_URL_INVALID')
        r=SESSION.get(u,stream=True,timeout=(30,1200))
    h=hashlib.sha256()
    try:
        ensure(r.status_code==200 and 'json' not in r.headers.get('Content-Type','').lower(),'RELEASE_ASSET_DOWNLOAD_FAILED')
        for b in r.iter_content(4*1024*1024):
            if b:h.update(b)
    finally:r.close()
    ensure(h.hexdigest()==expected,'RELEASE_ASSET_READBACK_SHA_MISMATCH')

def publish_verified():
    ensure(scientific_ready and CFG['MODE']=='FULL' and CFG['PUBLISH_FULL_PASS'] is True,'FULL_QA_REQUIRED')
    stamp=datetime.datetime.now(datetime.timezone.utc).strftime('%Y%m%dT%H%M%SZ')
    branch='ml03-results/'+CFG['VERSION'].replace('.','-')+'-'+stamp
    tag=CFG['VERSION']+'-'+stamp
    api('POST','/git/refs',json={'ref':'refs/heads/'+branch,'sha':source_sha},expected=(201,))
    npath=CFG['SEGMENT']+'/'+IPYNB.name;xpath=CFG['SEGMENT']+'/'+XLSX.name
    mpath=CFG['SEGMENT']+'/'+MANIFEST.name
    create_or_update_git_file(npath,IPYNB,branch)
    revision,_=create_or_update_git_file(xpath,XLSX,branch)
    # A draft release accepts assets before publishing a tag. Publish the tag only after
    # the manifest and release asset have been re-downloaded and verified.
    release=api('POST','/releases',json={'tag_name':tag,'target_commitish':branch,
      'name':'ML03 '+tag,'draft':True,'prerelease':False,
      'body':'Stage II ML03; scientific PASS; evidence SHA-256 '+digest(ZIP)},expected=(201,)).json()
    with open(ZIP,'rb') as f:
        upload=SESSION.post(f"https://uploads.github.com/repos/{CFG['REPO']}/releases/{release['id']}/assets",
         params={'name':ZIP.name},data=f,headers={'Content-Type':'application/zip','Accept':'application/vnd.github+json'},timeout=(30,1200))
    ensure(upload.status_code==201,'RELEASE_UPLOAD_HTTP_'+str(upload.status_code))
    asset=upload.json()
    verify_github_file(npath,branch,IPYNB);verify_github_file(xpath,branch,XLSX)
    verify_release_asset(asset['id'],digest(ZIP))
    publication={'status':'PASS_GITHUB','branch':branch,'tag':tag,
      'release_url':release['html_url'],'evidence_asset_id':asset['id'],
      'evidence_url':asset.get('browser_download_url',''),
      'notebook_url':f"https://github.com/{CFG['REPO']}/blob/{branch}/{npath}",
      'excel_url':f"https://github.com/{CFG['REPO']}/blob/{branch}/{xpath}",
      'initial_commit_sha':revision,'readback_sha_verified':True,
      'publication_phase':'draft_assets_verified_before_tag_publication'}
    write_manifest(publication)
    manifest_commit,_=create_or_update_git_file(mpath,MANIFEST,branch)
    verify_github_file(mpath,branch,MANIFEST)
    published=api('PATCH','/releases/'+str(release['id']),
       json={'draft':False,'target_commitish':branch},expected=(200,)).json()
    ensure(published.get('draft') is False,'RELEASE_REMAINS_DRAFT')
    ref=api('GET','/git/ref/tags/'+urllib.parse.quote(tag,safe='')).json()
    tagsha=ref['object']['sha']
    if ref['object']['type']=='tag':
        annotated=api('GET','/git/tags/'+tagsha).json();tagsha=annotated['object']['sha']
    ensure(tagsha==manifest_commit,'RELEASE_TAG_DOES_NOT_POINT_TO_VERIFIED_MANIFEST_COMMIT')
    verify_release_asset(asset['id'],digest(ZIP))
    mark('PASS_GITHUB','PASS','Release tag, git artifacts, manifest and evidence passed SHA readback')
    publication['manifest_commit_sha']=manifest_commit
    return publication

PUBLISHED={'status':'NOT_PUBLISHED'}
if CFG['MODE']=='FULL' and CFG['PUBLISH_FULL_PASS']:
    try:PUBLISHED=publish_verified()
    except Exception as ex:
        mark('PASS_GITHUB','PUBLICATION_FAIL',type(ex).__name__+': '+str(ex)[:130])
        PUBLISHED={'status':'PUBLICATION_FAIL','error_type':type(ex).__name__}
        write_manifest(PUBLISHED)
else:mark('PASS_GITHUB','NOT_PUBLISHED','No write without explicit switch and scientific FULL QA')
if STATUS.get('PASS_NOTEBOOK',{}).get('status')=='PASS' and MANIFEST.exists():
    RUN_COMPLETION_STATUS='COMPUTATIONS_COMPLETE_RELEASE_GATED'
    progress('FINAL',1,1,'Основные модели завершены; переход к причинному блоку')
else:
    RUN_COMPLETION_STATUS='PARTIAL'
ensure(_PROGRESS_BAR.n==850,'CORE_STAGE_PROGRESS_INCOMPLETE')
# The unified progress remains active until causal diagnostics and BOTH archives are verified.
quiet_print('CORE_STAGE_COMPLETE',CFG['VERSION'], 'comparisons',len(ABLA),'evaluations',len(METRICS))
quiet_print('Local outputs:',*[str(x) for x in (IPYNB,XLSX,ZIP,MANIFEST)],sep='\n · ')
if PUBLISHED['status']=='PASS_GITHUB':quiet_print(PUBLISHED['release_url'])
```

```python
# C01 · КОНФИГУРАЦИЯ (отдельный причинный контур, не смешанный с ML-03 прогнозированием)
CAUSAL_CFG={
    'V402_OUTPUT_DIR':str(WORK),
    'WORKDIR':str(WORK/'causal'),
    'EVIDENCE_ZIP':'',                      # пусто = найти в V4.02_OUTPUT_DIR
    'RUN_MANIFEST':'',                     # пусто = найти рядом
    'TREATMENT_REGISTER':CFG.get('CAUSAL_TREATMENT_REGISTER',''),               # заполнить путь к реальному CSV
    'VERIFIED_BOUNDARY_REGISTER':CFG.get('CAUSAL_BOUNDARY_REGISTER',''),       # путь к проверенным периодам границ
    'POLICY_ID':None,                     # если в реестре >1 policy, указать явно
    'OUTCOMES':['migration_per_1000','population_change_pct'],
    'EVENT_LEADS':3,'EVENT_LAGS':3,
    'MIN_TREATED':30,'MIN_CONTROLS':80,'MIN_REGIONS':8,
    'MIN_BASELINE_YEARS':2,
    'STRICT_BOUNDARIES':True,
    'PS_CLIP_LOW':0.02,'PS_CLIP_HIGH':0.98,'MAX_PS_CLIP_SHARE':0.1,
    'BOOTSTRAPS':299,'SEED':1729,
    'AUTO_UPLOAD_IN_FRESH_COLAB':False,
    'GITHUB_SECRET_NAME':'TRIAD-RU-Colab', # имя, не сам токен
}

progress('CAUSAL_LOAD',0,1,'Конфигурация и входные файлы')
```

```python
# C02 · ИСТОЧНИК V4.02 + КОНТРОЛЬ МАНИФЕСТА И EVIDENCE (без повторного обучения)
import os,json,re,math,zipfile,hashlib,shutil,tempfile,datetime,traceback
from pathlib import Path
import numpy as np,pandas as pd
import pyarrow.parquet as pq
from sklearn.model_selection import GroupKFold
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression,Ridge
from sklearn.pipeline import make_pipeline
from scipy.stats import norm
from matplotlib import pyplot as plt
OUT=Path(CAUSAL_CFG['WORKDIR']);OUT.mkdir(parents=True,exist_ok=True)
ROOT=Path(CAUSAL_CFG['V402_OUTPUT_DIR']); QA=[];CAUSAL_READY=False;ESTIMATES=pd.DataFrame();INFERENCE_STATUS='NOT_IDENTIFIED'
def audit(gate,status,reason='',**kwargs):
    QA.append({'gate':gate,'status':status,'reason':str(reason)[:500],**kwargs})
def sha256(path):
    h=hashlib.sha256()
    with open(path,'rb') as f:
        for block in iter(lambda:f.read(4*1024*1024),b''):h.update(block)
    return h.hexdigest()
def findpath(custom,candidates):
    if custom:return Path(custom)
    return next((x for x in candidates if x.exists()),None)
ZIP_PATH=findpath(CAUSAL_CFG['EVIDENCE_ZIP'],[ROOT/'ML3_EVIDENCE.zip',Path('ML3_EVIDENCE.zip')])
MANIFEST_PATH=findpath(CAUSAL_CFG['RUN_MANIFEST'],[ROOT/'run_manifest.json',Path('run_manifest.json')])
if ZIP_PATH is None or not ZIP_PATH.exists():
    audit('EVIDENCE','BLOCKED','ML3_EVIDENCE.zip отсутствует; необходим результат V4.02 FULL')
if MANIFEST_PATH is None or not MANIFEST_PATH.exists():
    audit('MANIFEST','BLOCKED','run_manifest.json отсутствует; необходим результат V4.02 FULL')
manifest={};PANEL_CAUSAL=pd.DataFrame();TRANSFORMS_CAUSAL=pd.DataFrame();evidence_info={}
if ZIP_PATH and ZIP_PATH.exists() and MANIFEST_PATH and MANIFEST_PATH.exists():
    manifest=json.loads(MANIFEST_PATH.read_text(encoding='utf-8'))
    if 'v4.0.2' not in str(manifest.get('version','')).lower():
        audit('VERSION','BLOCKED','Ожидаются результаты V4.02, получено '+str(manifest.get('version')))
    else:
        expect=manifest.get('outputs',{}).get(ZIP_PATH.name,{}).get('sha256')
        real=sha256(ZIP_PATH)
        if expect and expect!=real:
            audit('SOURCE_SHA','BLOCKED','SHA ZIP не совпадает с манифестом')
        else:
            audit('SOURCE_SHA','PASS','Архив V4.02 проверен',zip_sha256=real)
            with zipfile.ZipFile(ZIP_PATH) as z:
                names=set(z.namelist()); idx='evidence/methods_and_evidence_index.json'
                if idx not in names:audit('EVIDENCE_INDEX','BLOCKED','Не найден индекс Evidence')
                else:
                    info=json.loads(z.read(idx));evidence_info={x['file']:x for x in info.get('evidence_index',[])}
                    def evidence_read(name):
                        member='evidence/'+name+'.parquet'
                        if member not in names:raise FileNotFoundError(member)
                        rec=evidence_info.get(name+'.parquet')
                        if not rec:raise RuntimeError('FILE_NOT_IN_EVIDENCE_INDEX: '+name)
                        target=OUT/(name+'.parquet')
                        with z.open(member) as src,open(target,'wb') as dest:
                            shutil.copyfileobj(src,dest,length=4*1024*1024)
                        if sha256(target)!=rec['sha256']:raise RuntimeError('EVIDENCE_SHA_MISMATCH: '+name)
                        p=pq.read_table(target).to_pandas()
                        if len(p)!=rec['n_rows']:raise RuntimeError('EVIDENCE_ROW_COUNT_MISMATCH: '+name)
                        return p
                    try:
                        PANEL_CAUSAL=evidence_read('municipal_source_panel')
                        TRANSFORMS_CAUSAL=evidence_read('factor_transforms')
                        audit('PANEL','PASS',f'{len(PANEL_CAUSAL)} муниципалитет-лет; SHAs сверены')
                    except Exception as exc:
                        audit('PANEL','BLOCKED',type(exc).__name__+': '+str(exc)[:220])
# Токен Colab TRIAD-RU-Colab остаётся только в V4.02-ноутбуке; для локальных outputs не нужен.
# Если надо повторно получить оригинальный handoff, авторизуйте V4.02 указанным секретом и запустите его.

progress('CAUSAL_LOAD',1,1,'Входные Evidence и SHA проверены')
```

```python
# C03 · ИСТИННЫЙ РЕЕСТР ВОЗДЕЙСТВИЙ; НЕ СЧИТАТЬ НЕЗАРЕГИСТРИРОВАННЫЕ ЕДИНИЦЫ КОНТРОЛЕМ
SCHEMA=['entity_id','policy_id','treatment_status','adoption_year','source_url','date_precision','exposure_definition','spillover_risk']
BOUNDARY_SCHEMA=['entity_id','start_year','end_year','boundary_stable','official_source_url','verification_scope']
register_file=findpath(CAUSAL_CFG['TREATMENT_REGISTER'],[OUT/'causal_interventions.csv',ROOT/'causal_interventions.csv'])
boundary_file=findpath(CAUSAL_CFG['VERIFIED_BOUNDARY_REGISTER'],[OUT/'verified_boundary_periods.csv',ROOT/'verified_boundary_periods.csv'])
if register_file is None:
    register_file=OUT/'causal_interventions_TEMPLATE.csv'
    pd.DataFrame(columns=SCHEMA).to_csv(register_file,index=False)
    audit('TREATMENT_REGISTER','BLOCKED','Создан пустой шаблон без фиктивных вмешательств: '+str(register_file))
if boundary_file is None:
    boundary_file=OUT/'verified_boundary_periods_TEMPLATE.csv'
    pd.DataFrame(columns=BOUNDARY_SCHEMA).to_csv(boundary_file,index=False)
    audit('BOUNDARY_REGISTER','BLOCKED','Создан пустой шаблон исторических границ: '+str(boundary_file))
REG=pd.DataFrame();BVER=pd.DataFrame()
if register_file.exists() and register_file.name!='causal_interventions_TEMPLATE.csv':
    REG=pd.read_csv(register_file,dtype={'entity_id':'string','policy_id':'string'},keep_default_na=False)
    miss=set(SCHEMA)-set(REG.columns)
    if miss:audit('TREATMENT_REGISTER','BLOCKED','Нет полей '+','.join(sorted(miss)))
    elif REG.empty:audit('TREATMENT_REGISTER','BLOCKED','Реестр пуст')
    else:
        REG['entity_id']=REG.entity_id.str.strip()
        REG['policy_id']=REG.policy_id.str.strip()
        policies=sorted(REG.policy_id.dropna().unique())
        active=CAUSAL_CFG['POLICY_ID']
        if active is None and len(policies)!=1:audit('POLICY_SELECTION','BLOCKED','Более одной политики; укажите POLICY_ID: '+','.join(policies[:15]))
        else:
            active=active or policies[0]
            REG=REG.loc[REG.policy_id.eq(active)].copy()
            if REG.entity_id.duplicated().any():audit('TREATMENT_REGISTER','BLOCKED','Несколько строк для одного муниципалитета и политики')
            else:
                REG['treatment_status']=REG.treatment_status.astype(str).str.upper().str.strip()
                valid_status=REG.treatment_status.isin(['TREATED','NEVER_TREATED']).all()
                REG['adoption_year']=pd.to_numeric(REG.adoption_year,errors='coerce')
                bad=(REG.treatment_status.eq('TREATED')&REG.adoption_year.isna()) | (REG.treatment_status.eq('NEVER_TREATED')&REG.adoption_year.notna())
                valid_source=REG.source_url.astype(str).str.match(r'^https?://').all()
                valid_precision=REG.loc[REG.treatment_status.eq('TREATED'),'date_precision'].isin(['YEAR','EXACT_DATE']).all()
                if not(valid_status and not bad.any() and valid_source and valid_precision):
                    audit('TREATMENT_REGISTER','BLOCKED','Неверные статусы, даты или ссылки на документы')
                else:
                    audit('TREATMENT_REGISTER','PASS',f'Политика {active}: {len(REG)} единиц, {int(REG.treatment_status.eq("TREATED").sum())} получателей')
if boundary_file.exists() and boundary_file.name!='verified_boundary_periods_TEMPLATE.csv':
    BVER=pd.read_csv(boundary_file,dtype={'entity_id':'string'},keep_default_na=False)
    if set(BOUNDARY_SCHEMA).issubset(BVER.columns):
        for col in ('start_year','end_year'):BVER[col]=pd.to_numeric(BVER[col],errors='coerce')
        good=BVER.official_source_url.astype(str).str.match(r'^https?://')&BVER.boundary_stable.astype(str).str.upper().isin(['TRUE','1','YES'])
        good &= BVER.start_year.notna()&BVER.end_year.notna()&BVER.start_year.le(BVER.end_year)
        BVER=BVER.loc[good].copy()
        audit('BOUNDARY_REGISTER','PASS' if len(BVER) else 'BLOCKED',f'Верифицированных стабильных периодов: {len(BVER)}')
    else:audit('BOUNDARY_REGISTER','BLOCKED','Нет обязательных полей или внешнего подтверждения')

progress('CAUSAL_REG',1,1,'Реестр вмешательств и границ проверен')
```

```python
# C04 · РАСЧЁТ НАБЛЮДАЕМЫХ ИСХОДОВ, ПРЕДЛЕЧЕБНЫХ КОВАРИАТ, ВАЛИДНОСТИ ГРАНИЦ
DATA=pd.DataFrame();EVENT_PLAN=pd.DataFrame();VALID_POLICY=False;EXPOSURE_QA={}
if len(PANEL_CAUSAL) and len(REG) and any(r['gate']=='TREATMENT_REGISTER' and r['status']=='PASS' for r in QA):
    D=PANEL_CAUSAL.copy()
    required={'entity_id','observation_year','region_name','population','migration_per_1000'}
    if required-set(D):audit('PANEL_SCHEMA','BLOCKED','Не хватает столбцов '+str(required-set(D)))
    else:
        D['entity_id']=D.entity_id.astype('string').str.strip();D['observation_year']=pd.to_numeric(D.observation_year,errors='coerce')
        D=D.loc[D.observation_year.notna()].copy();D['observation_year']=D.observation_year.astype(int)
        D=D.sort_values(['entity_id','observation_year']).drop_duplicates(['entity_id','observation_year'],keep=False)
        D['population']=pd.to_numeric(D.population,errors='coerce')
        p0=D.groupby('entity_id').population.shift(1);y0=D.groupby('entity_id').observation_year.shift(1)
        D['population_change_pct']=np.where((D.observation_year-y0).eq(1)&p0.gt(0),(D.population/p0-1)*100,np.nan)
        D=D.merge(REG[['entity_id','treatment_status','adoption_year','spillover_risk']],on='entity_id',how='inner',validate='many_to_one')
        D['boundary_verified']=False
        if len(BVER):
            # A municipality is accepted only if an externally documented stable interval covers the observation year.
            verified_keys=set()
            for row in BVER.itertuples():
                for y in range(int(row.start_year),int(row.end_year)+1):verified_keys.add((str(row.entity_id),int(y)))
            D['boundary_verified']=[(str(e),int(y)) in verified_keys for e,y in zip(D.entity_id,D.observation_year)]
        elif 'boundary_year_at_risk' in D.columns:
            D['boundary_verified']=~D.boundary_year_at_risk.astype('boolean').fillna(True).astype(bool)
        if CAUSAL_CFG['STRICT_BOUNDARIES']:
            D=D[D.boundary_verified].copy()
        D['adoption_year']=D.adoption_year.fillna(9999).astype(int)
        D['treatment_status']=D.treatment_status.astype(str)
        DATA=D
        exposure_regions=D.loc[D.treatment_status.eq('TREATED'),'region_name'].nunique()
        good_treat=int(D.loc[D.treatment_status.eq('TREATED'),'entity_id'].nunique())
        good_control=int(D.loc[D.treatment_status.eq('NEVER_TREATED'),'entity_id'].nunique())
        all_regions=D.region_name.nunique()
        EXPOSURE_QA={'n_eligible_municipalities':D.entity_id.nunique(),'n_treated':good_treat,
          'n_never_control':good_control,'n_regions':int(all_regions),'n_treated_regions':int(exposure_regions),
          'n_year_rows':len(D),'strict_boundaries':CAUSAL_CFG['STRICT_BOUNDARIES']}
        VALID_POLICY=(good_treat>=CAUSAL_CFG['MIN_TREATED'] and good_control>=CAUSAL_CFG['MIN_CONTROLS'] and
                      all_regions>=CAUSAL_CFG['MIN_REGIONS'])
        audit('EXPOSURE_SUPPORT','PASS' if VALID_POLICY else 'BLOCKED',str(EXPOSURE_QA))
# Covariates selected from values observed strictly at g-1. No post-treatment feature in nuisance regressions.
COVAR_CANDIDATES=[]
if len(DATA):
    prior=['population','births','deaths']
    factor_cols=sorted(c for c in DATA if re.match(r'^G[1-4]_[0-9]+',c) and pd.api.types.is_numeric_dtype(DATA[c]))
    COVAR_CANDIDATES=[c for c in prior+factor_cols if c in DATA and DATA[c].notna().sum()>=max(50,int(0.1*len(DATA)))]
    COVAR_CANDIDATES=COVAR_CANDIDATES[:12]
    audit('PRETREATMENT_COVARIATES','PASS' if len(COVAR_CANDIDATES)>=2 else 'BLOCKED',
          ','.join(COVAR_CANDIDATES))
CAUSAL_READY=bool(VALID_POLICY and len(COVAR_CANDIDATES)>=2 and len(DATA)>0)

progress('CAUSAL_PANEL',1,1,'Панель причинного дизайна подготовлена')
```

```python
# C05 · КОГОРТНЫЙ DOUBLY ROBUST DID (NOT-YET/NEVER CONTROLS), GROUP-REGION CROSS-FITTING
from sklearn.model_selection import GroupKFold
from sklearn.linear_model import LogisticRegression,Ridge
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
C_RESULTS=[];C_DIAGNOSTICS=[]
def construct_group_period(data,cohort,year,outcome):
    ref=int(cohort)-1
    if int(year)==ref:return pd.DataFrame()
    two=data[data.observation_year.isin([ref,int(year)])].copy()
    if two.empty:return pd.DataFrame()
    base=two.loc[two.observation_year.eq(ref),['entity_id','region_name','adoption_year',outcome]+COVAR_CANDIDATES].copy()
    base=base.rename(columns={outcome:'y_base'}).drop_duplicates('entity_id')
    current=two.loc[two.observation_year.eq(year),['entity_id',outcome]].rename(columns={outcome:'y_now'}).drop_duplicates('entity_id')
    z=base.merge(current,on='entity_id',how='inner',validate='one_to_one')
    z=z.loc[z.y_base.notna()&z.y_now.notna()].copy()
    z['D']=z.adoption_year.eq(int(cohort)).astype(int)
    z=z.loc[z.D.eq(1)|z.adoption_year.gt(max(int(cohort),int(year)))].copy()  # never- or not-yet-treated
    z['dy']=z.y_now-z.y_base
    return z

def dr_did(df,seed,bootstraps=299):
    from sklearn.model_selection import GroupKFold
    if (df.D.eq(1).sum()<CAUSAL_CFG['MIN_TREATED'] or df.D.eq(0).sum()<CAUSAL_CFG['MIN_CONTROLS'] or
         df.region_name.nunique()<CAUSAL_CFG['MIN_REGIONS']):return None,{'status':'INSUFFICIENT_SUPPORT'}
    x=df[COVAR_CANDIDATES].apply(pd.to_numeric,errors='coerce')
    # Exclude cohort/year information: these are design variables, not confounders.
    x=x.loc[:,x.notna().any(axis=0)]
    if x.shape[1]<2:return None,{'status':'TOO_FEW_BASELINE_COVARIATES'}
    outcome=df.dy.to_numpy(float);treated=df.D.to_numpy(int);region=df.region_name.astype(str).to_numpy()
    prob=np.full(len(df),np.nan);mu=np.full(len(df),np.nan)
    folds=GroupKFold(n_splits=min(3,df.region_name.nunique()))
    for tr,te in folds.split(x,treated,groups=region):
        if len(np.unique(treated[tr]))<2 or int((treated[tr]==0).sum())<20:
            return None,{'status':'CROSSFIT_NOT_FEASIBLE_IN_REGION_FOLD'}
        pp=make_pipeline(SimpleImputer(strategy='median',add_indicator=False),StandardScaler(),
                         LogisticRegression(C=0.5,max_iter=1000,random_state=seed))
        mm=make_pipeline(SimpleImputer(strategy='median'),StandardScaler(),Ridge(alpha=10.0))
        pp.fit(x.iloc[tr],treated[tr]);mm.fit(x.iloc[tr][treated[tr]==0],outcome[tr][treated[tr]==0])
        prob[te]=pp.predict_proba(x.iloc[te])[:,1]
        mu[te]=mm.predict(x.iloc[te])
    if not np.isfinite(prob).all() or not np.isfinite(mu).all():return None,{'status':'NAN_NUISANCE_PREDICTION'}
    clipped=np.mean((prob<CAUSAL_CFG['PS_CLIP_LOW'])|(prob>CAUSAL_CFG['PS_CLIP_HIGH']))
    if clipped>CAUSAL_CFG['MAX_PS_CLIP_SHARE']:
        return None,{'status':'POSITIVITY_FAILURE','clipped_share':float(clipped)}
    ps=np.clip(prob,CAUSAL_CFG['PS_CLIP_LOW'],CAUSAL_CFG['PS_CLIP_HIGH'])
    odds=ps/(1-ps)
    residual=outcome-mu
    def aggregate(inds):
        d=treated[inds];v=residual[inds];w=odds[inds]
        nt=int(d.sum());nw=float(((1-d)*w).sum())
        return float(v[d==1].mean()-np.sum(v[d==0]*w[d==0])/nw) if nt and nw>0 else np.nan
    point=aggregate(np.arange(len(df)))
    rng=np.random.default_rng(seed);clusters=list(dict.fromkeys(region))
    blocks={c:np.flatnonzero(region==c) for c in clusters}
    reps=[]
    for i in range(int(bootstraps)):
        draw=rng.choice(clusters,size=len(clusters),replace=True)
        ids=np.concatenate([blocks[c] for c in draw]);v=aggregate(ids)
        if np.isfinite(v):reps.append(v)
    if len(reps)<max(50,int(0.8*bootstraps)):
        return None,{'status':'CLUSTER_BOOTSTRAP_INVALID','n_valid':len(reps)}
    lo,hi=np.percentile(reps,[2.5,97.5])
    status='CONDITIONAL_ASSOCIATION_DID'  # assumptions not machine-verified
    return {'ATT_DR':point,'ci_low':float(lo),'ci_high':float(hi),'p_bootstrap_two_sided':float(
        2*min(np.mean(np.asarray(reps)<=0),np.mean(np.asarray(reps)>=0))),
        'n_treated':int(treated.sum()),'n_controls':int((treated==0).sum()),
        'n_regions':len(clusters),'n_bootstrap':len(reps),'ps_trim_share':float(clipped),
        'inference':'region bootstrap of cross-fitted nuisance predictions (nuisance not refit)',
        'status':status}, {'status':'PASS_DIAGNOSTICS','n_covariates':x.shape[1],
            'ps_min':float(ps.min()),'ps_max':float(ps.max()),'share_trimmed':float(clipped)}

if CAUSAL_READY:
    cohorts=sorted(set(DATA.loc[DATA.treatment_status.eq('TREATED'),'adoption_year']))
    for outcome in CAUSAL_CFG['OUTCOMES']:
        for g in cohorts:
            for event_time in range(-int(CAUSAL_CFG['EVENT_LEADS']),int(CAUSAL_CFG['EVENT_LAGS'])+1):
                if event_time==-1:continue  # omitted reference period
                yr=int(g)+event_time
                sample=construct_group_period(DATA,g,yr,outcome)
                if sample.empty:continue
                est,qa=dr_did(sample,CAUSAL_CFG['SEED']+int(g)*17+event_time,
                              CAUSAL_CFG['BOOTSTRAPS'])
                C_DIAGNOSTICS.append({'outcome':outcome,'cohort':g,'year':yr,'event_time':event_time,**qa})
                if est:C_RESULTS.append({'outcome':outcome,'cohort':g,'year':yr,'event_time':event_time,**est})
    if C_RESULTS:INFERENCE_STATUS='ESTIMATED_CONDITIONAL_ON_DID_ASSUMPTIONS_NOT_CAUSALLY_PROVEN'
ESTIMATES=pd.DataFrame(C_RESULTS);CAUSAL_DIAGNOSTICS=pd.DataFrame(C_DIAGNOSTICS)
audit('CAUSAL_ESTIMATES','PASS' if len(ESTIMATES) else 'BLOCKED',
      f'{len(ESTIMATES)} group-time estimates; {INFERENCE_STATUS}')

progress('CAUSAL_EST',1,2,'Когортные оценки DR-DID')
```

```python
# C06 · СВОДКА ATT, ПРЕТРЕНДЫ, ГЕТЕРОГЕННОСТЬ И ГРАФИКИ (ТОЛЬКО ЕСЛИ ВЫПОЛНИМЫ)
EVENT_STUDY=pd.DataFrame();PRETREND_QA=pd.DataFrame();CAUSAL_GRAPHS=[]
if len(ESTIMATES):
    # Cohort-weighted mean; weights = factual n_treated per cohort, not the model seed count.
    rec=[]
    for (outcome,event_time),g in ESTIMATES.groupby(['outcome','event_time']):
        w=g.n_treated.to_numpy(float);m=float(np.average(g.ATT_DR,weights=w))
        rec.append({'outcome':outcome,'event_time':int(event_time),'ATT_weighted':m,
            'n_cohorts':len(g),'n_treated_cohort_sum':int(w.sum()),
            'aggregation':'cohort size; SE not pooled across overlapping clusters'})
    EVENT_STUDY=pd.DataFrame(rec)
    pre=[]
    for outcome,g in EVENT_STUDY.groupby('outcome'):
        leads=g.loc[g.event_time.le(-2)]
        pre.append({'outcome':outcome,'n_pre_event_points':len(leads),
          'max_abs_pre_estimate':float(leads.ATT_weighted.abs().max()) if len(leads) else np.nan,
          'status':'PRETRENDS_OBSERVABLE_NOT_PROVEN' if len(leads)>=2 else 'INSUFFICIENT_PRE_PERIODS',
          'interpretation':'A non-significant placebo does not establish parallel trends'})
    PRETREND_QA=pd.DataFrame(pre)
    for outcome,g in EVENT_STUDY.groupby('outcome'):
        g=pd.concat([g,pd.DataFrame([{'outcome':outcome,'event_time':-1,'ATT_weighted':0}])],ignore_index=True).sort_values('event_time')
        plt.figure(figsize=(9,4));plt.plot(g.event_time,g.ATT_weighted,marker='o')
        plt.axvline(-.5,linestyle='--',color='gray');plt.axhline(0,color='black',linewidth=.8)
        plt.xlabel('Годы относительно начала подтверждённого воздействия')
        plt.ylabel('Условный ATT (единицы исхода)')
        plt.title('Когортная динамика · '+('Миграция' if outcome=='migration_per_1000' else 'Изменение населения'))
        plt.tight_layout();f=OUT/('event_study_'+outcome+'.png');plt.savefig(f,dpi=170);plt.close();CAUSAL_GRAPHS.append(str(f))

progress('CAUSAL_EST',2,2,'Претренды и неоднородность')
```

```python
# C07 · ПРОТОКОЛ ГОТОВНОСТИ, РЕЗУЛЬТАТЫ, SHA АРТЕФАКТОВ. НИЧЕГО НЕ ПУБЛИКОВАТЬ В GitHub АВТОМАТИЧЕСКИ.
for filename,frame in [('causal_gate_audit.csv',pd.DataFrame(QA)),
 ('causal_group_time_att.csv',ESTIMATES),('causal_nuisance_diagnostics.csv',CAUSAL_DIAGNOSTICS if 'CAUSAL_DIAGNOSTICS' in globals() else pd.DataFrame()),
 ('causal_event_study.csv',EVENT_STUDY),('causal_pretrends.csv',PRETREND_QA)]:
    (OUT/filename).write_text(frame.to_csv(index=False),encoding='utf-8')
info={'version':'ML03_V4.02_CAUSAL_BLOCK_1','status':INFERENCE_STATUS,
 'causal_ready':bool(CAUSAL_READY),'panel_source':'municipal_source_panel.parquet',
 'input_evidence_sha256':sha256(ZIP_PATH) if ZIP_PATH and ZIP_PATH.exists() else None,
 'input_manifest_sha256':sha256(MANIFEST_PATH) if MANIFEST_PATH and MANIFEST_PATH.exists() else None,
 'intervention_sha256':sha256(register_file) if register_file and register_file.exists() else None,
 'boundary_sha256':sha256(boundary_file) if boundary_file and boundary_file.exists() else None,
 'n_cohort_time_estimates':len(ESTIMATES),'qa':QA,'exposure':EXPOSURE_QA,
 'limitations':['No validated exposure registry = no causal design',
 'Selection on unobserved confounders not testable',
 'Parallel trends cannot be proven by lead tests',
 'Spatial spillovers/SUTVA may fail',
 'Group-by-region crossfit and conditional region bootstrap do not constitute independent causal identification',
 'Municipal boundaries and official publication vintages must be reviewed'],
 'outputs':{x.name:sha256(x) for x in OUT.iterdir() if x.is_file()}}
(OUT/'CAUSAL_STATUS.json').write_text(json.dumps(info,ensure_ascii=False,indent=2,default=str),encoding='utf-8')
quiet_print('Причинный статус:',INFERENCE_STATUS,'; оценок:',len(ESTIMATES))
quiet_print('Причинный блок сохранён в:',OUT)
quiet_print('Прогностический контур не изменён')

progress('CAUSAL_EXPORT',1,1,f'Статус причинного анализа: {INFERENCE_STATUS}')
```

## 14. Итог: один быстрый научный архив и удалённый технический Evidence

Финальная ячейка формирует `ML03_V4_02_RESEARCH_FAST.zip` (компактная книга Excel, полные научные сводные таблицы, 13 графиков, QA, гипотезы и причинная диагностика). Объём ограничен 30 МиБ; локальные SHAP по наблюдениям включены только в воспроизводимую выборку, полный массив SHAP и прогнозов остаётся в Parquet на GitHub. Полный `ML3_EVIDENCE.zip` с моделями публикуется **как asset GitHub Release** с указанием отдельной ветки в `stage2/ml03_factors/runs/`. На компьютер тяжёлый ZIP не передаётся.
```python
# 21 · SINGLE FAST RESEARCH ZIP + VERIFIED PRIVATE GITHUB EVIDENCE RELEASE
# Complete reproducible source is embedded; NO heavy Parquet/model archive is downloaded.
from pathlib import Path as _P
import json as _json,hashlib as _hash,zipfile as _zf,zlib as _zl,base64 as _b64,nbformat as _nb
import re as _re,datetime as _dt,urllib.parse as _url
UNIFIED_SNAPSHOT_B85 = 'c-qXpdvhDfl`r`J{S;+&%oHGjP(6VH*k}huLK3Gbk|wEcw<#`MPZAag&;Y2GTEh)L#$$W!JH9hJH)3Mq&SUS!{x7Rrwz}P|j@VB?eTDrb`#UGIs<H|NO6}>{SkJTw6tePh^5i+cbMoZLr~gH}7RIB|>{?el{4bjJ6#t1QgZY!Q7*DUA`qPKe<m2(R2LCpU=%E|gomSfk<-^l>?nnOI$4^|tLNR)HA<E7s=hHBj?cw`7+P$q6!`41<nuewQ-+$CD|LyYC<)1FUdi|Tr&$Y`Jm!Dm}Ouu|}`Rw)2U;i@w`t|Fdpy;Rg?<;=s?_dAs^-o{_3{RiE{+)LD9Hl<P$6sDP)9~!Y<sV=F{PIt)fBX7pc=$8C_56Vxm3A%vbosBMMeXvBm(S3|&tL!a@(b<GWO6);we`u!54xIbX_qh2)GtuUXVjb5ztB+l3)K4Z@;RE@pZmwLw%O3|^Q+6R#ene2&(I6>fCumb9lrd>%l~otuPF8<*MuK0UpA<!uh8l<JpH|RvVV7Nx1rJCxa%+Z)62^*E}u078D3(P=rmL4cN&l287TAB<*U3X`b>p?E-Ixbum2V={!RnQes}qY*ME<$@kqYpMqW_0sO3*YDNs|peqA(0BgF5h`|?Ze&TxMBJXm4+(m4OfqkP5XUS7Z6)Gq%UCL(?E$E?l&=O_P&@!+K|=+)Oh6Qo5OKRnpmT-#XLyT8Iue#mwF3-$gBY7kxc?d!i2-6KMA-Ll6-gcmej7%2aF##BLfsRa<?cc`Ds(;T2Wjj7I){S~_Md-2QbU*YfP+7C8&uU|(!MYj8r>Rx0fnk$;QPnr4tMv#{a3!;Njx#2IE75JB?%IPn(U4Q!H^LWnXUtRv58+<`CDp*Bf5E|R7*T0~a@ayl08ay^kk(do_FCNAIEati}uo5xy7yiUOdL~$s$nmwL%^zvb1s9;ui_0&Yl}qK{Xi^33K9ihA?|eZW7Cib&@}`^?;;zrA+v%cFN&A8Z{;3$2c!~d_pJ~-;$HA`8MTcMUr2UcW$>}AjTgmF^$}iIT^G;Q={8yq&Y3<BKf<~XA@N+Ky2c8w#&*#L^R2>cer%bueR7!$fvKBu@OJ6frfS1tQ|IRPGOczw;LKQs{-1IBrBgllmLBBqym3swS;O|ekFE7M0%S8!^w=xy;)$+spyIVVJ8-v?xn_K;jAChDeefZ(-+W!6zi7AK^j{b|3?iK1IoJ6tI67l0WJf9CI<3Tt%AJ6gY_<S@{)rue6R?u<$H>j1Yn4(rD=kv4kIqR{bWs1fyNlo0_lG?_gZ@2GnZSkCHp!b)|{=bwA_;-R#Qpvm&tMCN`<^Kli60^O?8JE`eS3>oW8hHJie3h)<zN7WDr}+c=N58kT(bu~=HLo|civPW%>HGI~-tX(QV*W()`1?$~y~?$p;6jpkH#JdPR(^pA1)G*SvU~s5*5>}*0lhn*Y1O;8{n0FLWc?^gOyXItRj}R{1-Z-GSXtHb0TcOqYo(U=`nx+I!P<B}8E;HJIf;f3Cy#D!#HW+vsh<pHlSlD<Jbb7(O7HFO+}~U85BB=E@wV$ZZ9D07T*omY-w&;h8~V28+W62(Ov4Nk-*jSsp_aRA+Z#K#sfrVS98HqsrqLcevKCwG-?`V{K0vt{DRt}S>Bz7Li671<)7e7l5Bhuin>*WC>6O!wz48dm*0ffV;n*K7yz$}A-uoMydwAn!I2q65@%-u*_WS(}y=%6uPWhLOoe#HpxvpU*_Yc<Y^i92cFg=%3`HqxNgkFFrRvyLEVKR)P6s(Y-C~5>oS)m`E#H-q0?QCu@asmV)V<0SuUkkA&@fyDTZJFcOzCYOC+`hBbAArmH2ZKAe^sd)5%EP#~xjop}+<$*?d#~T;LamBIsI33N=0<;eU6e7K3-q|Vv$y$!ooz7mq1mt-?oqkG!P?%P{sH~2pANxuESt{a>A*A$LvQG3le6=Ycp8F9j^n{uIM)}dx!qqoxWCsQ@Jd==JHY%8w$|>TPYW~D|6uS@e{By<S__@&?;kAvk!W(Tw}yXZ343{oy`2wHz_cm}%dLnKPGx~xJ39yasAw1K>&_-q&#NN79Q6KRx4$>Iv$u19m)g9t<n6tkdxM>AIsJOqtt!2aWx2VrcF-TJLqdw`WqHe}e&JwkYv;~l@$Jc&IRBj-0=@nTIGUC5>)-NI25TfFza&M>I`S)3qJO5!@w`Q}G}y;ftL+UocJ6`tve^p4_SeA{qR_G#-rrf@T-$m>ArNS18*H{Ucq7N`?}8ao@ZQ?N`rSTp%>o6}A}a4vmDS^@DxI!E2rzwF`CB{pw>Q@IJ{qj=?d<P=xVH8F0D^M!_9lrkp>5v&&8r$zM_nS+Ql=*sz0V8y-u(kHCQ;U2vOr{y_WRrWn+Ka8Y#w}c)mp(TxBCNV5GuEGYrntu0XMR25pRP3LE4=?bO+tv++Ev3D>QqSdv$qKE<^xUqG4QJmf0pPUCBIq{dII&?ZihZv$Yp0PZyzl=|eQM^Wl;ezOi|aD7UkXnc2rAk+5>nUtX#&BnoD9^MXux0Nodi^~qw~-Z>cbzkhe_{yqrbZ%9B~bPEuMc~i6b;$3GPW*M>r4B-AqI1ir`NkEqPOK#|I$*?V;F9mDwe{XArtv|L5xyYC71JD@Bf62!2vnq0DbY>zC_V4fR?(9K|wTx=51r2knzcpC<0MLBx)+SF+wbG%%Z*Sa|qaJ*)wspVHD?yUH!sa4*`Afo&WMc@MiM<FI>(9v=6t)|?HpqZr2ZMMh=qC(hqE9+Suh=ZaPyA6L5djui&|W?(D^-jOb_W(NXewb$Sw?&hk_aepYxB-stnWUpc~Sz^rTgE9ErUhAPlA{i@V(7Ddt$w({95(i?%w7-D$PPf6hsS8s}Mu?&@wEufq$%{W$1y+SX9P7y0}7J5`f$4Z}<22m#X~!_S&tjzLY%}JPmB;-sU=}uzk?~{z3J47K^J&x?;^%yuji?N<7n8`hx7oU~_*!3pZ!+J7#s?B=kGj++N2zCr-^ZQl*k(ZM+Q*KzU%=JuGDdhTTHJ^_{)F{uWTh1~_Kt`%owM4;cI#&Ca5b0uvGM-NUNtGdkSc+_=*RAiN8t*&tyDl(c_<kIZ^^Zr#I2RF96YZv%Q9Rbh7EpG^>jz$jsNvu)e>$7$mqmp--WlSiLA_|-E^{A1ETV3ex*Fax`NC>7RMeIYC=`Cf1d0KoeDy8!VA`%6@AhWto4bUIbE-n>WctP$qUP^R&Qf;nedmGeigO<9|@DD2n-UeJW@VyheX*Wcf`HMrNmx3l*Vph1hKs=Ca<UC4?J8s(ylSnqG~dY}%me%Qa;CopN!0xviEj<(M}1}z>>hvCU7Y`3oFN6~OTd=zUiDTm?EA8EHX?(Jx^I3!Up(<bNh*)WQ=`AMt|PfyS1elUtN8_czur9R!+Jh*%R7T6qOo`lyPbw@8cNp&H~NeKV0;8kV+FbiPSMF#o4DxH|D=bp1}l?@$$;M?UDeV=gW{e2Pw-v?DTg+*M&ZYi$IanfBZXv_(~$L{?D25so;<cz2R;+*;OlZGaIH8-Ec{&c3Dp3ml50P2jR_$<ah<N4@`7RJ*#+pEhOR6*$n;B(l&`%4-{&BM@#dCe_qXXn9a7_LmE*K`_(@$hUuTh>qtFl8f4+Bg}`&;XhlX~*$|++Z`!A4l}%G@ef}py^~b``8~n<hd(2@x}HAyL^S0xny}i7Yg;WbT>i34087W9FKpGcfPoMQRX8TV&VWOdbht$;AtQfBB8;LGB*J_@_S+nY235F3%$v=TE1dG;WNbvN$yt3UV^nQzbxC~E|>?p4BRkSyT7q{pf)F566`{L&UMqM|BynjuYs)yT>OcdSs)YMml69qF9itsf?apiQ;AVhJXY;EQW<5TKcBIA_?wE2687a2X)2n`{A<#!ojSjN$$kDpA|XjydY{nJPlcbkT3$nM_V-AMlJuxvYV=E9j%BGBAp1kCMW@N)3-`Bo*Vf+$Ptz-w@iwnO!rtodN#RJWMBXj{_yyUx7tgS`!jl!?B*?q?@!7~9j#=8p<5_%)4_%1G5d>o#K?00otnvWs;$%7*PezktUOOB856Tj8ZQ}#dEE@wss%qk7h4%X2yT1uMMNY9?r(gNX_TaBzOI7HHtVGdMkyqPlEbr?+2#1M%zXHkSrM^#kGo|qxy7^&W0AB_tgnO*3>fIvW&|6VV(a69dON-qC=aA!Pdt<rFsaWU1&Q^bKjZJquUUA;oTNRO5i-;sE4_3S<`_MLwHxvVET07XkY7%xgb_=dF$2G1lyR&;0uTYU&?9VanMZ1}m>Hz@B0qInR)%#d&LY<@@_)Mu}KHK>;_U8mRpskLfX`oNm_dlqvJ=4Thl~>jtOwPxVk2-}CtEzi#Z~ZQ`2x-{$t@|4QgliwB@@IVwXfxYJct@N1A7=oE-O2cOOd1{yKG8=0^f)fp4tOphPwLfHgmh6E%57n}7f~H4@C;hv7qt7iTG;FsC0{JrEUpG(2mSTC!mT8CPSTFOBdcbmOjY+E0q2bY4{wyVZ_B#8u&d6mrW?3b+v*D)1Gw(fbPVL22r+$&9azHepZOjZw`P8@xw~wGiS+EJZ1Vq+Hkg*Ry0t^Z;03<Fc5j!&i&+8CCHWXdeI$LRcxkYCJ9T0#6HNQ-o8&&b4a*ZoTz_q!h562h{Q^%A-u;!_qb+RPeXbq*^SH2aMO4jpy|W#v)pYw@M1HUm|Lp?O5^*J8;OpN>IIXxYF7l85YzD-tt>3<LC^q|!06Vps+|+A`y}J5Bd4DvjoxEo-R2<HR<JsH?3)M~<!|}YXC6lRkLiKR9*#^f^y}I3HI}=Cm_00SYFvvf!AM2-LlUW8h{DGK8lW_~bXUEPT)ZhUQeUWAG`>(}E!mpo6&?Fd^u_^6Ai>*jTXTQd$&o#M?`Z=N1x7kas3`KH$BQY$;eZ!t{R_v-!q<1v4sa^gba<g0n%n&a;XJ6P$-X>sgS7kKBKVG@*sue;c1T=i5t?%Anm7y<NPU_CYiX$oV75ef&6=9Wb{mWhGpQqIn!bg+<fPDM^zrOw(!sWuL$rJL!bhGh?87`F}P8(AguN*z`6Tu1W&*O~~7L+e|c#MV@Hw$SdU;j#zAfCe|NT|?$3-y=xTX@lQSrUO8{$7Ip&(qEIyoFbpDKw2|=ch5v#b^BMZ!@Fz8Ha-WiK$gJDc7$*vYK}H`t>(NV!Z;wr?pIBQvH!Xn-9Zn5sbsk@Tu&MI@SOJdNiaxrlN@83M=-;UkduYq+KP@_aB7eBmCUy0{df;W70SvW!0e%Q2qeY&2KmK!s`TaV#V>&WfT&VW?!{yH2lH&bVx6s%;#sb?#-K{2_XB)WH#@5hG*QAlYrmj;qeKz>U7da(>RHzq#dKd$HVA2p3ky%h+h+@9R#-DG6KtYlF;&6e%o*Z&u=?^Y(`#Mvv#fBZYPe_PU3bHIF{uGk<n@gEz30xFSG(DFKOGBNi{@X;5&}r30%YP_`YMuQ4(88l9#l?BxtwX&`ykw<3&NoXmz|+<hFdzupKkbOPWC^j$_kl`9{l$Fz8l0vE0~bTj;PC8F|UT4?^2WY%6pe$8bGA=>%TO3ENS-6$G6qFBuw^jc&C&juktG*G6Z}pzYgH(2ni6)yhjoVQ2(xbRFY%S{<{~w%dW<PE0$D+8AkG(nl%JLkB}YZ~{N}6SEz*yiO-fT#zeJ#@=y5&+dR$k>PYAuVdPQ??y(a6WMkgy2{v{D6pJP=p~@M>4zQ1h-}MFtR(QvxYfBP3Lf#Fi-szUi_De=?`-mKamzETAa1QVuIaBho^7mnpz=y=`W@FbOe5@k^S~btEbVW9u(`LhO*p%u-P_#WytlTcG023Kx!d2ST}4=N4UL@w+SbmUT#gJ+&nD2SvnR8LHo<?h6D+4uLxW8_hf+Splk?zgItk+$eifxh!=Ncj{b)8BH?;Zi6vE=fpPk^B283EeJ7w?Ja1M<<nT+t}$NqFY93SJ;C&RO3I6|4>q`|VL<uqvJhjHMC57FM-pASh+p)6KvENaIp9Z64PLpz_2s1IlUlyEqN8fxXH&(5ZR6Zz<p>Ex6>arjks9tvfCl9!+;l^#8+F|bI3R*vec^s#yRFd9y4Xb2i*)?-i>KSu8clZX6G!6yV^A_0Zi7625AB~Zr4@Wh|aXZUam;We7+^{!T<7H&p$*~qeb&u1_&dSYIh(Qx)~z)xzhjktz-y?}TMN*<#4xeS8)j1{B^^%71TjiKPLSbwDowIYg;G#Dx6p9nRP$I5Z=EbC9yUiQ*GQ?yn*o}EwQ0rkDdeQzeyIKF9Jziu1v^#qVC$Hp8fEShNJ$y`%v=|WGyh}eU~2XpkN)5)}^!xFuJ8(5k!b!uW!jq7W>YwMc_AE}shyWijFZ}gt((RuhV3Sc*i4}gzNv+=G)F)~kl)cY}hlArKf8$UVp8$hf-qo4T0dp(QZvdeyX2KnRj)3YZi%73YTtHnk2Svce$_%AmsTb@0HjWivLpB@h9EBuv8I9#Bj5iGZ}(S#Nz-paqc*Rz{m<6Y|V<SZVaJ$XFB@99VS)#P8Rx*gE!_vh4e`SHD;+w3&nZK1Yl3`{ng6A99fL<D;&6);4p4e?6;<-H!d*Fl9pngpWr;sXuVrok$9$Ik8n@4jKyLVq4ipp8*J{m7IR^c#^*J`PUtYx?oM9!Td_^$dL@tm9ARy6hIdzSnDs{>@JOGnx?md9P<Q9XvMi`0;Vt5BZ6xU=U4}j-L)^uy>AohmY9IdDwVFQbMp`6C(0-R;z1Z@}h8UCc|+wn4QI8?IC`iX%x3oTotn7ZgEtEDM#~OF1DM~^KtDE622+W5_u{c`pPLW^6(7*uKXDPuthc+>5c2M+M`D3pUr8>2~8xV-#mUiFH0!;2i*}3H>WXP8WTpVT&DNdwl~-}KiJ*e9c23HBaX#|O{Xr?)PXaoCSl4E{~;ol+xWJrS+4(EAnPI*qnKik_j)p5mTYZ)UiK^>Rs2-xPd@5aEg6udVrH|NIO0~BSxDvsQ7c7TcQyTn9Mfzb#nb8G3hhC5kLrs}ox$j5N6-J6nU-lBhX}{b;`5<&zRpjD@V=jj56_~($x8XJFkOC%f+8M1-RCH5Vvy<eQ?~E3==YagiW~`KZGKKCSegq6isd`7h&Db04q?4O-_9bS7}hNWjD)Df??R<;0YCdnfQD^ddi>*s7IL}1ntyDb&tj1N_*h*x3okU!p3o<*cvQwmtbkH<MCfZ)E+7<A6a3AFp_bMEFdn22m%M|b=kx51ef%B8oBV9Ltd(gjml{?}_Tro<I*s;Whb?_mNbUiP?)1eyQZ$X+@?G+3x^a`fb{AOe&eV^Fz<sw!u%j<~@iF*zHd)p@?S2BaX4HxO@xTvelMx0xh-r;9_}T2o=fIPZcvPw@)y-)oA5NwZp;#U^MiU<frTn(!C9J0>qd4op?a6rF0JH_E$G3;Dir1hH$K+dI_CATs>~-j87}cZm)A3*w&kOUFvb+Qa4WYF2-_N3^5PP#GO-3`EoIM%9M*nyUNzGy|FALFdaDTt|^kVhHwY}}l?K}Iu!=u%My|wjzj{&mIQRDMCJQ>r*S#vZw)>rY?-a(Iq4_W{^oZz=wy&%}RiYB9K=!UNNdOik|JXA|YaiXEh+Ue}LF^g%ZVE&{BFnN0i7`-TwSV$A>M%jABPt&*wJyDzLweJq=weNP1PxKr5+2F5_hx3#30G|giMF#alf8`Tnr879X0Y!1R*WXxMKj?29>8u1w^?r<3d46i#|Ay}CjhM6mX+!;do~(F!otAY{A*hok6TFsMQO)ebc=B=nl>YeydJ)qf)|<oGM9gZfek0Mf!>1`xE{-6$sf#!C2M@-2iG<|Btc`t`Y9v9<XN_s>!#M2e3se*nen@4HfaOGSy(?eS#e4L9S`(k^hh4`rjxLJT(40{nH*``c;_dWv>CHu*DvGmN!3G;24{7KW%yqlJw!cX_vXO2x(Nl`dJrHrd^$JR-tlRX@D6peO!-@eInIPX_h$W&5lc$E|PaM}>7zXu!l3GJ+>>tOX6Y99gqf}coJcjnEkvUwh>!c^G_a+do<+f@|*^t9p^l5Mf*8!jfL{_r?Ya{<Oi2QB>{NhJ7=elWF;-7jWAl^7>o}UpX)`EH&coE*wPMRn2W7%`11kJwQA3qvSCoIvlo%g%i?vwcmvD^(pXJFbHh|!##jfQyj(7<vduXIl^$Yj<G&(4L{uvV|*H?m3Dx*Qx216gWH`-Qc#&Dt>QMWg$+KOAX6jFEFV0H%%Nq)1xREKxudss15^zBUz*iLfZmp8p815`<hZmunEBgMl2yU?9h#H+33{;93|v^@axK1)|;`+*|vroxOoPr^8H3k)6b_)CpbGs?(dtqe<Y82r$FMi5U}DxtOay6%I9bwab6M{MXBWxcryPf7S5eAH|2~6ios=mTyGyX~Qq%NldYS#oJos4u7WAa(l3O_u$@Ey;<7Ac?pmwo$jSgo9sYecHUKZn+f}o%UCn4rgr&1mK|;*&&XHonf_Cb_5>>A&}U7C@;>rMh26XS=*25`-9O_!PIh9y;?VA2Xt$xT=eyGhLxzSn|8aB*AUMP<`cFW34*hvajmeEV4s4VAq!ehAIac_@CU2xPOKN*l@Ox9%n(FQJE0*VgAk>JtnjYov&!<ny(nP$RPX)g(KP$ZXIE>Hc+9sD~K~t(HeWh3As=@AE+TP?HdnkLozx@GZb<JpKR*QIe?f$`CdSJC0n$xByGWblZIqimKdi+d;3~4od#fzpP%6$OQ%&}g0-J}|AkA9#Ke12wo4b3)8dUhMCc8B6mv>Kk^v8V$psK3Pz>3!Sf#tzo@-xnoav}SuYJ>2MXzwwRwX>mDm)_^){H8j`bhp8<@&+(|CwLFJAb-*ziS`FVDTHE4EX^WYj;VYhbOc3P|2LI%NFrLM|<zP;Fi0^o4^T_hUBY19OXr3UWjLpSEqNC?>4N|d*=mY%ZQax?;r@j6i{-sU7h_1_DGFo{w0<YsKzd_#cFC9?4W3|;^zAplKwHlt|M~53r_w~~)cmCZ$lbx2O{!XC@{96l@>~vf)6I=L6OgkQdM_@$OFBVwtVDn(BpR=5vJ4p4M@APzo{G}#0c*R-iPj!V8^)x$z8UV+&NWRn|0pFYyK`$|_e)&%*{ao6j&x+{;^qi4(-no1sf*13E5R!<$Cwo;d@V9>Xze{2HJT;4HuQJ`e6Z_ZCb%ps+KX1@+pLJyc&<}L9|Ep#8Gt)q<44seue<pgQMu*V0I2BPAj1HBNrvqS{{$+M1LoY5O{qnyRQ%C5^Vxl=+$RELT`HTufmUEP&o-Z}gG--sNrY;3K<U&6#tUaBb52Q7|5Lpl8$&go<pUc1+NnKr8oBHKH^O!!RZCA+5kzQP*dKN<?SS~$e{VFAXx^BT8YCeuM$fJ;?;q`Cyg+(p6r4)GkOt|og!MdbkxLUou>UC)K7h)D@4xdXU$DR*P5AgExbA3@D5cf0ZUVct_CyL2C<oW3(Aq09^ge>;Hu$BE%uaGR_9VHvZl0N)C-QL%$B@kFG4XV;u5HP>J{#%|YH4TPbKP4e0MM<|9kF|W-6gehGkR5#&Jgj7UQN7CHh<sy%opGF^jFKiM$Uhhi$HVzxP@Bc0q{?ui=go1?Smh7%$=o0HNRW!}Q9KKKLO#9ZH6)oQlTM=8*<?0PhL2(EiSGmQw)Bjq@{P_Cg}~{(6>~`^1a9aJR3unq;MEh=oXk&0K<Wf$(?V7FRdI<b^`>~6%63bzsHT*wSJ03fUPeHPo2X2M`!byOhOjS$83DRp7xccFQ@dVYB_pv%6*NI^|17R$w6E}p&d!6yvT_0{ic2f1roS5vA8G6#>OHu2>OWq=c;_eGj%A!ZUOl6;9pmG!>7G5-%r>3`lW7!ByC%NRCZl1by%RZ67`v-_<ra!4zCsSO^I6xlsRloMc+3OuzGDPtXq&6yWCXnUPU0q=Bw0-+<9XL~(8Ahu=#LsRe>_{6QKaUBYwuMzQy7Gty>2p!AFuxCd^R5@PgZ2`Pj`0ahw(}f&p(di@#?XE#x1R$`qSg#cqN$3=abW}iLozlD|oM*zln85mbmHfbGfP}R6;@hhHy;0doy?s)d*7(yZ|fnFZkI_jN_^yPA5FVGr6z>fBN8>=7UNrydcJtkMXNoZQ<wB;TS&~#b^FwJOc;fy~=hovWk!vW`)(_w026fYP*i(2CEa;sL5#Zara~xL8ZOj<O%L8P-Zo4*)aY>ZAW-@IoO`Ov9*p7#m9|zEYo-Wr14J6jJ&|8zZGM4c)kVs<-f&c!Q3&QoOL~nRkEw^g>EF+b!9gEB<`Bla<;jF^+^+-x1d*u6zTGywc8ku!)Z(|j4Q-5X0(=zj0PqXmXrAH2FaJa4u}(eK8+J_&PlDJ@mV|{kZ_p}*+n~`Dd3)v&yL^fae#BZo9h5BQ>r?dK%)2Lw<YaJ!Stp!QU&toG?SL&>P2B$^&&ZG;go&GtErH#7SEN6g)|8_#oK78btTty?>23IO4arj6<}9L<Rw!rG$$&oQk|&r8>u<vd2>;bE>;zfP?;FkWJ>-23)ATJj4t`w5`A}X=MGTPzV;NqT>ORhG~<p7`B~25MHR<~qsc6mO0;5nOFN%~w6VdhwY{E-aWG%`k907#<Y&l#>u=oY@AVFkiayX(q}I+({8`+H@N&;=NZiwi;<-N@Rrsuh$4o0Wz2J`JI#Z`|!Q&xawC~4H<X&l3(f#fBw|73=mZhaYq<ZQlD$eHqbl!-^QLj+*As0Sc6+<A&U{XYNEd>`mLLo3niqKE=sNRqTO5>cM5^jm*N)64e@Ynivdh2G@n&5Y+1^Un^OkHza=iS>cImT2#^E8#;>tBfwTRzG8N`{MySkvESZoSHA(=P=kPVMX$>FGQQ2mTocxGx+&lM_qC6(8$hK%j>(uJL9@L)Y~BVb^ny7KlENd$ocqK~kJ~M^I7nP4zZI{GDErZ7Sqak!Vul;?+arXtnrTDk&5FDYbub^C|b>f?&r}(TxlJss?z?aODfVrTKfWG=JUVcd59j<Vty_i}evQFTK1(C!u9#(jJebTUx>SjtY}q?}~-1hZEXKp2v~i<%!VwZM`cdL<eLVu*}m-ofuzK?=zH2Rr$V2e*E}+NNSCu3~Sc`7iNv?*B^dNAIc&FZ2zxd$3O9z&ODr*ea7Ng!HZJ-ya-{SBbv{(#XV0x@BFIlcxEp($1CtD()@tP7g~ken0G3tPgtUZ<C{O45U{|a@WlxXp9B{VraYU%?c3)+01aIV%&6WU==0ep8SV+5if7t<!h0aua5Nqc$w_)VqP-G)iD_qB6G1V}yr~&x=+sBQ9t~&1;zkKuXG|&C++aQ#@Lnz?ynjqlqu|!#$@ED>qmWbHjnfEEjwj+tPvlafNBproi9+=~O{-}R22{sjaM9JC%0@1<!>1fSnzm4H(!Uqor=p>YBS;t8A5_(18V%c=olHKKUwTzTsiJYF!`P+SF!6u-+nnu4&pv;eIR^hEPC=_ZD{ON7i-;F}CX+kLBq@u@Eg{QaUjClSbkjzt{emPfoDIviJ<_wn(vp&KGrO=TXb*ZE#vclDG%IKag3se=`n9GHkH?c~tT!|{Gd+F7pZm9`zSwR%fpsx@q9s%Rn4oeTHLvSAhxE>(CI@89nv1R1|J*Y?v1zoEBxt|k3{O_<>N7mFm-Uaitk5WZFFTu*vpWAE58a5wG~&b?xi7W-7&gLuCZmGWeJk->Azv-!&6K`fp6w->YyUmXheMOkwWxBe6b}6Q=XB(U-EJ?)Irtgt?mvhxFNH&!4!?ZO+jo>``1*C)Fs9=zU!|w=izf*y6Z3w)nw|NpJm$V|<n8M$KN|J(MJWxdaNFhMlu)B7SH^P@UM4wRkg}YQLX-Z;7h)XG)w5wQzx6z$wr97TBt7!+pLN*9b}EiO6q_y{eXty(6S^I1#d8yX#Ys$arDOZXiWNs4$BH_(7k~4k4>n<*{r>v>y}m}<IND8((x0sCY;S#}W$BYjd#133IHoX<aVl!EcjWS!@oLe=Qg##<z^XQ;m>7+tYqTL273ZEJJqM9JYw+1qsMA!bDZ)!+AsM{i|A@ip`t3VS^r2Rn45d*{hN5a^t_PONReQQ(#C-a#$rGOr`+OEpi7!gU(dB_0IK>y_eQoG!4wcFX7;0VgiN=*Sz=M-@)`=uAt(qt>JgYQ71<I(UW>Vt=LU7_^h6n21#i1+4F#p&MAUqykxocgVc4^a#my7j^)+k;i7U3ph!gy*1d-uhZ={&e{nIh4a4!_X2XioBF?ZGv<F*%<+j4>r?vParHAjji#5hKvZ^UGDF9?x>gLFjN!=XzIraP3VUsS5FZaP8LC4yV<i1;<+)3!?L9xs)#_Rj2J2sj#jo&93ck_L2vg`0>w%O%X8%raiq0J8d$%`IKXoIkE21#RCPU>;tlI?ritc*xKek1&&U}>bPW?<_R=P3<WQ(n5Pe}t)0(LCez_3oXzaPwJru2`%^6K2iI;e85$@P67%qnuti3cUk8P6J{m`|n>Xkb)Prk{>PTtze-EyGe?_LjUOOAEd?12*xN1tCwgS0mc~sN2`&d9LYsVC;Nx!DKgIChr!7G&P`N6e|;@O{n(Bx56K4&WTlD<fjrOSQW7pdKl2F?hlHYa!1?j9WMu4I6Y%m{)5d|;kcwy2QPyx+XPx0NNTS9h(2UQ;kSW)zSdl#a-MaR#Gf-nibNKq<<Q(yLLft@=_!>U=utrLWNnPd_)=F)^diw3?nWk1k`Hz*9HUf+c&em$HLonY3$Yr$MX^<<k>Afwg^B)8#Y0fqphRMqTOYVyu2s#P|(JYp0%|bf6VdlZy4gV(aT<d2aN#H~Sk8t}WI?;@C0FhGW=rfDf*5up(zE+1VRx?+o@x-R)Pr-Rn7qlfFe0oF27M*U=S${aBk%sU8?TQW{Yu+S5WK7prtofwgq+X+A|FXzjZkw077vZ3^SOR&9YR@x`>}2b17OJ>new4(mclPFiQ`x|V@*X=x#JVVgg~>el(Vhse4>S6}>rqsS;E?WY1AzBQiwir14&9-*uz87Ibhq0}J>A`E6HzS4cr@Ft|$DZql1WlFMdrimQgA)PT&Prs$9M|SSr+vLO&y}SYPGEf$F(V>?jzo+O_y1bR-5DM_r4700=$ORdYM=w8@QFLDl<x&cDS>_U9@&`}m@vJhmv^b4|4!lPPLa66Bn?F7WToUo6Ge3d#hQ)mG-D#Zkp5`M~D7CQoG8T{xRN@rLY~tK$GiW(cOzd03%ZJ&E<%r>Tv@{387i<K7Eu-`x|FvK=31;a96F(QX42bJP_@^{Y-DgTt)!%7kC~`zUD8Lf-IkVrF>O~J!=QB1xgn9H6dAmljYt)9iOm3o2H@QzY^&31W=BK4ZqoRUpX;?W{&Ubp6E-0ig!2zk;8W3jyOmJ(^-(KI@*xbIeY;`XXoEomCIOW>i+&$<~kDElLSxw<Fksw6oxIem~-xNviL_SiP;f@DefXh?=;{m1Jd)&LKDuS%EZN(*JC8|{7%4I#{`0>0}d(5c$F$rkd?L(a$K{LvVl4y?)bz*4!2*5Css;aV;dV}3h6|zlfQ{CO+J*HC7b~ebm8Z_PO6)ItdK<R~Kp32<;lux<O87v3)HupKpwn_;8wJ<$d!>bx6z}fF{nM0kDw#;X$Jd*<(#p9YNiVnIK4~L#T_2o;0VFVgxUP)2*0)>?~D@H&!>TssKUkMCay1ydL<MM(7*`9SZ&gb#60Ey5WtZ^CMmj>{qcd3XB9#9<m;$%^Z8$YG*AWb?Y58L4fGOXv5seW{*lXaga&cTZ1Bek3vNw1g3Tsg{r%10sGIiJZ(TVyVz7cz+GCB-UnuudVB-{p&&Tu4ZLaUNVc^U48^FV%taDGxo*O)S%6qx<3n1;jfANh&u+ttz=dg&3zWJ53SkfCW|w6{ueJB#q4^<G!oyyuYf+do6lT1(Q9v_9z9?M}Q%s=2`El!YFu5nscF5b!@p?m4ia1U-h5WXi##V2`#?lwjA=M{&}9lRx5OjAc@}KDa?vp?J2ay)lzDvu;*zIp~!(%3OP*Of~iL<O>0(}59t3&(`s{I+^@;GD3hQqup+G`d6ADy*~(FXzYry#%Y>6(D<P`s<wP=Xj4&;)x)2L+<>y&|+Sl2N{rT&^V-1={v^H}c&SwWc6Lk1gD)g`Eo*|Cw6Q)z1YcsW*o^z{NIv_T?i&@oEM|-1iiJCYWpj;m!ee`5deTH8BQ+dH@<*YesATh}y!ls|f8Axre&$C!T8Hki8=u{J)6%t*go>IAyD&r$dcW33f%yV}6U4d8V2#1nbi%$R6<y2q4p5N^BlBYb4p2||olZ*F@Ws~?Dk!3!WT3;5kxT>p9gk#C>icL28^Uu<#RqFQ7r5q%t_*^D~Dg<19NkO}8L;r1>sY#&$C7ennY8w13#BFiM<(bR|pP}z8Js9!+y{(mEWc0V3@+j)G{LnP*H^`E#*f;gOI^=o%z?K(HrRjtD`W-e_v@E>6p=BAaZ>HJPZ|>~fS=-+H0cUN~?ocrJ!pYKke9B2;7Obi?g)yBG)v|kmzPHBTMZ!J_K%QO{%@ndJ@cJo?j6n*&Ym79?W&+G&{9oZFA$eDj)I<carMg}PAk#1l)AJc6dnK0$6gPQ+o1&Ux)gsihh}yUhD(Go45|3;30*pKCS{5*BdPcnD{fuUFd0JNFp1DaaE$*|==GlQ*1|_m`JO{9n9ZBI~b((>N-lkrTr;ita8^hw##9b^&C{QYvZ~JF;b{Cp%q<!s{y53-@H4-pi@6xczRZ%0K=YV4P@MhXxqIYvt&=BMsoB)z>3dV~HH?_)UhYOuL5>Y0z`SNI7{J1cC>B(f7B_hiMw5rWd&w5mVtcStuJV}O+Ys>~Pa98o;vNy{+*<m`t84N|;95rMtMhH~SJzh+aYp=(!b;)Fs&C{ze?zsYC5al+0brmJIfv$KBsDeeufL?9lrvH0OJ2~?J)@-9;IY#~JS|#wWGNM5tX+rbvNzqCk+~1ZLq0srvB|Sb*df&imR$tAJ>w-TT`p4O6-Sk8_>R$4y;0S`w)nz!Yr~1|N9(i5$R5Z?R3ghpl2x=oN&-0%zmHLuC<StsG>3Ku??R!sk;fvS1y6TCag!6d5GNa7Fr}{-he2{5Xl-Z;!9hS9NcA0*shr-?E4*h9U1T@xHs}iAP$y|k7mtC&%Jh9ry6`j%9D?s9Nn(bOJf31s|Upb&v$WGDG<l}f+t6$OCAN8h^iC6P}rz)@~eP+RQ^6@O54ph$+H?X{3+9EEla(v+rwX~Ki-j+*;DrwFjbD<b}!i_KT*6<JXkJy8ihO*ohLeGlZM7lf*y0*2sktNf;OLv)E(S;(PrG5Uh7-%UI@vh&uIh)zGL7}bHx8ulLo7;SIMb42|F=DkMc%zWHy6>DG0D|D5pa&%h7)W_mNDt6`^LCnWUxp9gqG%@CZV1^%uhrk~HOZ*__PZP@E>bO3-1>NQo$~|K8@hP3wMjmEIul5_zVx?jFcZ>Obds;!96jk~(_SH|MZYD)!yGL_yoNa2T5$A%FBULOMePJU(hyn#JF9;_D*FE+^y3o*7vBKEEd)qj0nEe-UACP<X?J)`HGdeDSAj3x&LR!kqufbIYvGYfAj~yhFME|!5TB+(TZl_f{ragRm}}R-62qw4ukTllBi*ywqSs-HDR1g@Mb@EiSl#Z(4}d}&{CPGBhZ9r%VyRzj`NbVa$w+=~^XGyWTfbwam7h$`iRiK{1Pu7^ob!j$qgJ;|XSUNngVXplh^J8b-2}FDxl%i=G>*;%=1}_PFBEsu;=wtcy(s_e^3Q1cU~g@EpAWB!9&H#7|D*I~2CRz~23b<!ZWr2P_=vp9W&2ieG98*atNHbA<!lM=IcDxTZk(MF0~IGr&86p$pVSKOcy-JDOgO5!inFsiWL|nJ8U<jfvhEgRA!4MSZkXM!p%;P{mY<pbTzJBMA@YmnH|w#pNuE!>e3m*NyX??X4-7B!DamLUKBG?I){Fj|SM0wjbn|OIS@6Z>%bSZi09Ml@fE0G~Yk93>cGDv{lYT8PXI+wXn~%O%`t@Gr!^x8Ov!{=aBmVQ{|D=op&4rb>wz0N*(3iIo6%N6OyY4<3+*{k;yxrelw#%lU`r~0j0W~)bYXIA0JUM+bpv5tGWG-&6rM1fD)5^&)a~3tTFx=YV(^UFCjGyS;hp8~uyN?7Sl3xpZ9y&qO!EiiC`;?WYA5d18Y0B6irgurfKT3Ck>4%vJzB)WAR(s;lC~P1O=YvuFC?4tEnqK(MsB=k8i&dSTkB8y#46BvL7EC51=&;Gd`RRo6uZoJr*NeX-`TkZl)0|A_Sl0fKGQ!Fvjr-yX!tH|rBVKl;>1;T>^#DXbh@RhFxU~bjG`ssp*2dj^i)y6aP!Fi9nL5i{C+ZR#rH!W-^;GH{)t9Nocr-kw@PCEOqY0EMn-2b>zzf43>(D~4a}G+oUy4<D*sCdH4ns!og{c$Ih4<{Kp53>`Ae{4lrVq$$DLfFzQ5rhk%%)#70Se9o>S7o!*)b;unmk-e5l#RwOr8j|FaWhd41c^xiYP|WX&m)RDq_(+C<}h4SCr-56!pm87hZhx<j5NIjm`b_9ZGjFxOac+U{l^LT{-?@AF8MzvYn(3-$TE=bILkQ)nDI96%b9|QZT-puw}v=sTyK!tY!EBd~*y|6LE$5Oxt<?;NA{_E}y-J?D^igt*M@HaUkYwI+-NeBvC8Qh$<`?;>r9`ei<vQ!bkCxGL-?(#b;=W#5Ez)(K+porq}-|+CpuAz@B)%cXspp0Btp$;7Ffv4B-=(K0C^1{G%%QimG7uY~3gG8go+s^93MZx)UeOji(nX-0FYO-zwd;x_0NzUjGh6)vC6=lghz9MZ+mC9_%P@VD8p;_BQyUu5!C+jy}Fl7v&GOJ`w}ipoF$`!Fyg0-=nHul-X?&W-r4|SJ1t#@YMZnp~p9rwm;n5=!4a-3X)LzfllNqs4RHiMn2X{D}@uWziG9n@eUOWPpL{Aou2=PsZ<SlmNT55iZLMwGaE1!iv{&d3qLKlTrN;DcN#ly*Aa|hU7Agr{#G(lX;$de?H%>@YzR7<`@#L~?1uRYIKxxjB`3w3@y8qIzF0rVHY7Ph<KvW6(vlo&1vxA`FCvIYN)8IEQ@Jad(?HeBh(TfGqbi}FZTd;EU(02E?aCb3`HNdybP4sHzNt(6Qe3%*xe1<|F7=}ug<%%M=8FxBo3P5J`)D-zxc5}91I3)hSA+gj&D#7~rIu1H+|uir3MxGpw&>f`%Vv4$ZDHt^9;!z5+f*!U{EC_<R@FoMjdgxwo4J%iw#-T@f++pAAkO)uEKTWARRobp8P_lxWQ&xls?%vrCFCHBk9oxj%PCN4g<M%GhmFt9=UAX;3)a+9A(z@!6ptq4TUpx*vnsc#P?Dec;ryJP8v5JFM<8234t{$fye-dsE5rdrmg2X|eRd0)^k~U})HUE4iWTI2B@Hs1uTp*-M-6-c%%Pn(&9GMtSF4d~Ii-u{=JE7wN+-q1JDmfk33TA}5igZlG2<x7jVwZUrp0`qncJjWTmilP^U=JaeS8A*MVpLAPbfd=kdBl@8tr{O^2hTu3QePvrc;?T^{;j|w`a}bjWPbP!2^%S=chn#^SI0|4Raw4b){SWtXCc3P~pT>;wCDctyP?~Oj+hrOipAG1^9MuZcUdy+Oq`@Eg!_C%RDdMa1NcOG=ALu|7wW+y}g~iZ#+ID8xSpv8inVk=g9Mf)p$xFiNzeyX-#}yKb=zY#;V;Hbu3i9c<7wDq%5Mqf0Z&Sq514;PiN<+wMT~!t_d1ExCZ^ku`xW=53c2YXoceS3vr?J)2f;aF_&rtPeHV)wW$%<B8zhGP^!_R8_S?@5tF1yrPP_Dic4DjVI*o9@?}n0;%2I17^TrJ>TbtkaC2&9mI^lX^!S|EcG)mYwrTQfLKj1A@us|oUB=j^`G;hOCETJDJqwJP5a%ApW}anQ1jVX-s$bl!o~G4&=qYWup!`ytN730FSml@7D!u;Z=q!s0FZ8$&0xrkDyS7JH*H@5$qK(s~1I}4WfPCCSmC!v3P%b1oD8G1a+NYA8Y?R6Q9I_qihD5u56()R};g;l>5-GoY{7bMAU%?{hViQ&`U0-|9&!%OcX7|xB{#aeM!fZRApQI5SQqfkYhsZF=VJaVwddf4!q{H{x)2aGMD2vRF;q^AH3KpwVl84dQuc`I~mNhKnsx>1AynFYfeZI(e(fwTl8fA<=jCuumvU=82NzVq;;#Qbkk!KZpsIo_u8fh3U!X=gt=?oEis?2eho-Z7*JG|@XzVQ^xL5GP?vIu8A(bKrlLF&_$&jHKw!wbu|LTS@}I6jZx3Y;XpvmZC*Rj#*zn`&qE)$z1<mNe5|Gy#CPd#mG+863X}+nDpI%&Inyh0!g$&;H)>ORFiqEtL9ZUSa7OMt6HvwB=<Z;AW{U_SBWz?-hGp|JHaWle3eDQYh}P-|gR9qxC}fz2`@HzAZKyPpg<{9F|$A;_hqNW|0=@6<IoY8D(;!$rN3xm8rN=B4YjRfMVGF7OB1Wz5D%x!M*;$+Q!<!+TaRo_-(JyYEmweT#wV?OUIN1iD8ZsFWx9U3$U!Rh<b+R<jIUOJc+xFX1!|S3L-YiIIA5l=h9^~K7yqHRJs5^DjQP)NV#`CV<~z{bOr_0<Gq0LJRZ(q*s)PbI&xL302Wz;8s)M<nRGr0QJwm8@bIknl$>>?o)y8>V$X9NQN$OO7w%P#LEgnyt=b{kd*aTrCO(eAy4hu;;x?wHyf#za0!B%BBVJ4=#wl#%6t?K}$zU`*9nO2YIQG18fBpT9TbR;&l>PqbLVs|rzFP6_{Df&H$fLe_aF-5((be$9Gh-AgOUz2I4P=mfVLmdOp^=K>!^}HM^T|;ir{{95@JHb)7|x%h-t+RVLB9J!5s6~^Wl(r{Iy{~(d>)XYj|ahUIzORHVWZfm4@afzWJ3}6_}0BFdkTmqhv89A{%!g*mH;|s{-79TN&&K%281a||A3YxC2A?XQg3os2O>s$Bo?z?b>)Y2<CE#t)fG3(Kx{x%kMUck<T6TtLFSVI`&x!!O+IZ%c>q8VbM!%x7Ke+2sH~`>*c)mV6T_s=DB1*)0N}0vW%)7YO_xv04K-z=>zdi=7!Av?TJ<BobglL_L#eJ(@(PyULLK?+%a`+B%D2E%1+)DOvocdg8=Z^rZK(Fc0y9;R@L?RE$&|i}gQ8Oj7`JSpJnOvW4(_F;{w*_Dqr~pnk}PEqjfQ+Ub^fF$``Bc^Qt;%B1Fs)u-9G|29uFbWYSDCZHufbsD$T{L!6s0)S#Q#zWchDHCy&GW&i3uC&GiF1hCr8^9egzSsK2&X_7vq(fZ7~nGd6p3$BX>Zq^SS@l|SX=ZXVBT6ep`pz)^2}2Gl;T5<~<8MF7QHG!L<r9j|ACmCwV_aQAx>T2^zvBzGN)d^DuvS4%w7n~IBI?oTNNvwt29=UqA%#<`6njoc{cJKy#p7pf-0df2z3GMd;Fjpr-bV~vBFRIl;9%{zNyPc$`eS79$trja03vBJ{z<nefNO7ixwdIhMn(OpaqCSBBG-*d!4`G;n+VH)(G_{V5gof``=EOwBWx_FZ%0{80Zda3rsT8<}^s3(S+3Q6rR^SYI1O0|N~DvvW!eF90oi=KC@?O72(Gn)IonrQlvUOQUJnib!%!h%_?+De`gR2<KHHLmAfV)Ik~@$eKaoAHX<Fq(#YU9DWyRj)Vcy3ndhODvFED?H}mn73iN{H39n=n4a^_+W_#X$%lGoX*5%a!x8TRF0=HXPY0mXW?A$CUn(VEgxu|43}sqZa*DJ*S9Fd6ljWHYxQN$EBUrqlw>uL2$nSz=#*8Yf{L6tWv6n9Z1s{@DUGX<k)UEcM-6`#g68ZpW@+3I|B|RVT2xSKrxvgY4BMVspgAOwX&kNcz7h|=$J_Z-vRUjV-A3LqSZ?sIvd{1JY(r&xKDMP)$DJ+oGXHTQd<<;Vqsk8p{B^X->?40EbJ<Y^{!+B+Sv(z(xPa7WsbO4SE*U8Q4$99c7cl<vMPHO`iY`zh%S)z}C`nQdwo0ZxZ%=C<pFrq}{+2GF5w>Ga{)4r>JFq$TVUt2RuJ7l$E%c|{pNmH3rljV?czaLz4}QZsfM_LSANVjGrZ`NbjeAekNkZvtGUYGD$?Z1nq+kkGoW8q)vSoexmcuRGg1Wb$sx(31alYimVIY7EJe>(&ho<n#>?Ttuo2!*5PUvDHcFd5v%=Q`<v3Ee1fcP*Y&!Io3b06a93C5=9aWI4Ry`BAi1~DQj{@Na0U0$w8PTC_>Ks(k)N*=9ZQHtM{VD$}HqX+_zjuxO&$v#fq1~olvM{i^wmn1-;d$nG0zJToVsn@^LfaJP5Z{5-R6jSh(yzT7oL@E-_UM6Z+t%wxZH1s@v9L4Bv0ER{QVMXT~M#R@8xaR=Oib)?wlaIwld>T|mpnwqO=ABjQ;|lMOpAjM>_!FMQ;lmkye?(j}i$~xU$WXOvid&om$!bJm!C1gjIrEMD0L!4M?TeQuW0{RLe(Ylu@r+W4hm%tX1pgR3MmsqQdXoH3SECL-+}z&S`EbAYblzo<M;jHCn~>cNyL+4W=&YPtQa?JJQ%oFzA>Ib$lA_A}!ifyo%=+3^pZIZ*rOYYpM!Z8L(jv-{PX1#;Ae8Amhhc;8#2*t%gDlXpc$;)TZ%9VJ$nrXk{gGO)5*F%b$dn1nn>=%EG94Zd$35}O%1I;J&WcGN`ZOsy0o4SyyF0rw%)ghR`)Y9i#>Jz;%|+8rZAnwU{738&o;V(IgMOneKCOZy(s1<Z4LM0oc3V`)U3FZ_{P79Z^mrB7VX|GZ{V(b)AH<~Y@}!|H?r#OQ4xS873dt_igO(>5liuCg+x)@K_I_nv2MLHpCwjfAQC)GyGR0{neps-n;vd)OZ@p2ety1{Dyqz&0zD!$_aCFJJ{N<<A8I+$-r=)XLuA+Xc-<Nm4-~c(gf-Yi-{W*=S&`3|}Z{L=njT{$nF($Gm+HcT}*vbMqnS9*ShvRXK6_~~?_4LU&CcJ__<P|*e&*<)vWr+#EafhOVz!-~V-qH3Lq-(Jso@kt3m$uF0GoNt7=n3acnbB<tr1#Hfu{KGx6H{8cf^Y4cCvI~IPdCQI)zph6>oon}QX!lhOoHY0t`rLLyAT{~`ACkHqY%$d>Ptg_^eoh3AoP@YV~i^3Qa9Q-+Urv$oc#m**%%yrxFdjzR5#0mi}J0fom>0;y$@*LoHm}{OE)p@ZSK=~Wc`hOW3@4mF_8mOAE=ZhQjQ9D`+H(pzDX>|w&`uK_5CgJI%LhAO;}mm8&b*@yuQnIW5HTmTC8YYEuUmsY|DR@GfFGY%wxQc=zB&=SoPy43+$C9!IStvv_KS~!w3-nVmC?OoyN*Gk)nbHU^*YCl1#*6fQ)C;$$Sz{=*$Kwx8Z#Bq#?nLST2B4vy)*m*B1j2-gY-=Nw?FKz!XZhoMw;^oJazsBk`#VV|-TF2V&9tCK9C<)s7@VhS|iT;WMSvA!>P|o%GBUXm+)y70UXew2Ad+rQZuf+i=>J_lA^NS+T3r?>REFm~vUO$O89r`AyG@nHLMUSY?^YI7M%knXa6^>{XVXtdO}<UOkO&{r5CayvQ%dDabVOjBn(6#n)?bVzQT%mrdSNFD?XmA(KnW%x_<D+RUHiDU|<)kFauzKEBlJD=p0u_xkVCE<fj^wXc6kPtw$IulS}87&(nJ!5b4#B%R}9zI=&@yu>5R2lx6HO}b?1szbBt1-`F;qxB8zX5Vn|pZgZ6<UU_=gjoE<DfRd?Z+5-NSK?NX-xM>f|B-^cf5y3gf1>1;E}s77<c6RACtL~QxVF*u!stKuN)S>8>BB7WK1%QD%@zlKwF5KuI*D($+`w^N%ZoZrr(;El9XL_D6{K}**L*jKV&4qfUKBY&;(MlV*rw5Tg0N!+iJO;<W5-Kk%Qt+#;{>kld6t7J{kUbdUBfi;lBQwDZX6_@;f9XodS1u3+$fA~FAjV!vGbB)7}<6w39TTEoR-t}0y_@dp>Mcm5}>0>>q%r=2Kwo=ZKvg$UMr4l$MxMXa$P5#DsCMm+|Ug@(~KM^u)I##0pU795Jiz?C?(rL<e+&YGLyvdgCuY)+v)^y?7N-7HI<S^V4JaRxuzXkPTUH;c4C?BjsrRyk>@J?1R>0J&<^~>ahxc$K`Tfz*Yn#gJd7iy_1JAEPRDE6wvEZLJC^CTBfn)@F=}@ddbT@J$4%O?=b$#%v3xV>fWRG$F9-u$X+5&SByQW7d*8%}Y|l1qH|cnh-^P@N%7`343EHhrr<Jrl#|o^D5jaMev^=9@wCtABPp{)RR&1N5)ixcw<N0RWHajs05IIz>Qm5(Qu@#^a%sX+CXLT@HUTC1Pwo+$kgkI2!5<j#t9>ehbmfJ!vUDRqgzLl2*<t)=ms1noZpheqGYzu@l4KD^U)DeMjfn~S-4(jw=-$Gwnk<qpsGclZ4rIPP-lBm@RlF%`UIPI1l$3fcyD|+hK1CTxO!Z_@>k<+n#KeXCy&j8aqAy!EzFWE8!!-_*6jYW=|1RYGh>2}(n8EDW}T2Cy`X<4q{vTO&fcRVwU+8&07ne?EU6)Gj55J(aQe%ry)KpE2teY+h;uH}24vWk<iMXZU&!NeWY^IAbCLJus<4Ba?XO2)1OE+!h;j?p&!sBL>u&`x4>1>;gU*)?piT>x6OvA|kB1_;iITLE~+YAIv)!cGhl`gVvpY7qmtZsK)p$BqmGQ>?TeL23kT*KIoq_!#TR4?@F04=tzNiWRwJN038F+%l{HBQn5EZQqSd&$1HV@>EWC1FWLhh(VAR)+R`q_&#w!%W~Yftxzcf!(a#ySf&e7xn2t;+ewRLWy??{gX3b2x8lU~I{|p!>ewzQ9|w^iMW|lkc_R#*c8DqT;x@VwdsZv%#9;`r(~iTSqw;4v!JJ^N`GFHVfd`>UYZyYv4_c1G^8v^aIgsB8x`|pn&q*MCJP*<dGSye=>?E<<f;<2lIYvUu+>4U7W3{8UX$HQ^A-)r0iUKb-uy$MzOTA@y2_&L}p?T_xOA?F{OpG~l60p8$dJrBE_kq`HD`N*sfFW8agOX7Qd6?K?3`q?M6}DBb3oHZ6vJEYXAq8N5qt$L%QOkpd>G-iiC5Tx_wT>M@3Oi0{_;%tY=x;kTf)4l~FKM;0RME{i3PC0aa5oAdJ267&kib$_F;of7c-(QYI>IP`W=K5C^gvm3-E~!27q?7E84NUVf|k*7JDo6!uo58KI&D*3n^=rVVznTQ9Lx*^bA-X$P!Dzj<*$s$j}mA!sFk+k!~uE{VqId~V3`=YN0IYp*ov$q#uOVEyGcqsj?vM$4W3e09GKn5T4;4_NH)x35TgU32@Tl}KnP_vI~~`Lp@G56P#<9k;RVKb+h!bDk!!ZqRUG*a1de07vEx9*J7&vln^2qWuw$xo<N!<<L1M&-Vc>JrL0L#ND=^}g*H+hN?6l|tai|K+5kz|@#@g*ftqv_v2xyf*ZLBui1nW2s1d`u^9`&F(oybP5s`L&Z(=ce%5<7<J0=!_-Vi&Xv?J!Wr4uKi_0Sa2+B|DBfVd#QM{V4Q&=rM&uu*@KU+AVOg>6jo8Sc!<6xL~jBYC)!F+%bGmBXnKp6dUsygl23-3^<_2)Y$~|a-e5H5^$Z}!D5Ll=(#vH+fY;rhnRtvbXpx?hr|hC`M8j<KxEK1j^U^n%nBjxqR_`E3@3<MeiDO+lD3(E`@j-P>xK;m?Lc?Cgxf+7y+F%GJ2s(zB1O_V=$#EH5E+5#gjNSs#VYE!)M@CGmdf+cW=2FKwH&_{V#QJWHbgwusi)NGI)c&xNd_$P0OA}3g8+8}Xj5h0<G?VXf+8b?ydm}LqrVUhkP(;&bws9RV16+hpr`{@4X{$f5NpW}tQN$BB9nc`2a89TXK=C$aSLo=TF|aoBd)582|AK!fKq_wYTLdSwn%nI?bv|;Qpautpt4~=3@1)tcF^sh6*~!52_UwqjK~UnA7B+2jqcqufdJ#E?E||eAg%8!RDvoe2?sR=yay(RG)}<uF5qftVl}BHp)hO+<Ou6KF|d?va6a^mWrt2|Dv}lu7Tt`|7<6m|Y!kGB!aLw*2s@~GwN3*n6Y9(cqPA?nWRi`YB*0P*U3GPY7Jy&d1T;*5X23iF6m!_}tT=#{>nQz9AT6P*puQ49fLM)C+Zdz=GPsGa%6VV}AF>ym6p%deB0I5wJpDG1a-hn(7SOx}#RkCdLV^LSKy61I1N;uH-Bwmd3!{ZL0X~7&#8L`?*aNarNQ+^L6)Ht72+lU7k>fk<1dGb`p`E}^tr&<y)g3_4b`pUskg%590RX{L3cN7H1jH&{Lhm7OpbI>3KJ-EW=$WANW&+yUmaU>n2MRDW>^AgQ2Qn%O0T&|}H&{_%M1?<-4)7Ong=a&8wPK)JP~VGP+lKJ>Lv@ZoyQG!0fPsK211bp+21N+96@@ASG$3o-zz)D=fc!?s4zYG)bfeRP6jkIWxQ)y<U`QX9lmjW>4ly-u3wjvyq-q2^@FN(T#E#ID=(X##I}qEp(Xv8yHUZ=S2EgqeRw*PhkPl(o5KBFz^`-O^__XD~79kX5c|NdXlEie$r2!OVE0VTl06;;6xfWIj@KPtTEK)S2OyUC12Y~Q^2hb=0Ns$o+Hl%plzzZf8l!`9GHYO=BJ!tqAYK%L8;53UJ!)vu{RepMqT(G;0B!vDzZGHqW42HK&n30gwN<Uiwvj(v+)L$FD#uP<f2tI_kHzTEFkhCKkz%C{ZHDRdGf+7nT(9CRGfhy}@1A9gbodW6GNnm(3Mh3`?B^s-`1H-j|=^<;|7>XajGIj&d31*Gifi72c2O5k(6C0K|X3F<qIXh$nSfB#{r#eRl01|{Y0J@X3EFa{BpaaOj+VEqA=V23KS-2k5r{@5Ic8ri#x^IN|+e%bi7XWpE#UNx12TKPW2RQ=SXN35iC>jB}*$N>xU?0Os@k78H2UaH(39wD1Vul1L+h|2*<ip;K2~q+olaY-7cWo8V`z^q9({Dj8SR`(}#7KP4@}PNQ{9>!95{3W-C!}}F0TAi5?GW}m^j-^`7b<mH3FZW(_d#apO&4rM1|(R7Sx{MV0faQDkw97lT|l!2kqHHXSqDFwv9EG65D>VMY<GY)(6VDW9n-;-CKm9j!XdU#Mw1u7%z~n$b<_y}#M)S)Hf$S3Cc8il;2U5*h-Rp7$FN;=DegpW3*A@hWVi*L??D&0p#ws%1KtK)0@MXER`n6ozXMwe0vNcSj7OkrpiIMT+0Zo#*TK?{Ojwx^f<*nmBiR8C02j9mM`;~{ae?>%fMA`)ZCXZ7+YU?^4H49XGVfsD1Pjq}A(}Bywr2re7+~Vi21HOt1bGB83-t!;8JrBu9d@$Oh6J~uN!5ND5GjEf0PWl0k{ASlO&qm>G1`!xk*i_`0#HDcR2gKR57Jsr3#KM)b+4tAgis?b<HCGEMX*pj2o#fiBWA~q)sob<!4kphK+;2fbigA4(7y{c-%)wK&ENuZA!=bZVFM(v;z`0|6~hu%_|vq&GFW5;x?ISE$O8^QBVcdXk*aOj0hA&87JYF1PA3QhpbyBkb{qkQs&)s!E^tLeX1NoD;HC~n)`Hr_A_l%tk-rO;g_+^CI+#)$V3lMYtOhWS>4vHw3<Si?Lu+B=0l79k0&q|~0NEXN9l?MDsvyY)1nT-7fCa!ZA9xK^3l$A(IpmQ59025jhK)MV1VBVaf*Ep?SVh!s*!E&zy%;!>gpm!?J2pD#qJ#feI0PaarYUeuj1`9&^xD2@#%)MXpNvISdP5z88=_WZJH$po0FX?q3M2}BsPq#^8a+24y`YAn<uDwWZ_wcoE|}j)waUTbK=V)^A^VrCw$TQ%i7W^g*tUv&WOQJ>14#R1z_|d{u*yAZ8k)OfsazKtSZ5?>00i2$>%-FY$oy%yJAPuSUQM%;w1Bu#QOEJHDlws9hi+a6^FjG290LB~sq_sp%wh0=q%8xM3JkkgZ5@goa>E2X16pueM$l=8FwH<$ENAFZRfgDZ;wI#O24pqtpkuc^pe(y(K~P7k-v-Dg@Bt#DBtc2QWT?1^gs10{@uAwAj*Asy+K}tuN5FY<S>Zc64Y{M}O%rI#C36wj4t5+AL<nvL5p356_%&4>7eZ@6y}OWg7M28zDibE8(I#`*Q;}8MhNOT}g1zm8ehX3_)(Jpe3m807b%*I1<nM&_4yg#<0~E%n62MlI)|85u0O+9d6Ij9~1`QJ|fXXo}VCpEbqJrM^!w$F!dL{-=iUHI@SQp?;ShGf=Vg}O)S^@b#z;)0e&@i2V<{tnQazvH$u*<O00&+z;c0@jH2uNs92<F%h3*Kzsz`BMqg8B!DhxUlEexVW}#vHXICJc5Z%to@fJJ5zuudV?j0hSO5tWpVh8<U96V(baDfJv@X0*aW+*jDipkP;wQjG-sM0v0)7A;fKR!?!{e2>_3SjcgOr0mKH7gC>Adg7SiJ@qJYzfGjYV$=HXa4;<KjzS&BG7_bNeRQ0;M;5Ih^CxfJhPu|xyZL~xlfm~P7+Au{yyACNU8|(-%g5?1r=9*+#w?cIfrGrJ-0#q{`U>iFCa1Gox<fj>fM<P`tKu2P@=vqXBfGG<#3tk1Af)y1G@ggt;U=r+O!b=zzj1QC~X9BDiQ^htwNYKxK-XZKSh`CnFja=Xopzt_S?GDp(J+l+F*pKT#Ai=H!z60L|5I=5V#{r@kN-MNUZ$bqFJ)o-q%1~f!h3B#UeVA{i3Hjg<S~MaUYLFmM@v))WaRCwC0jZ(O9ooK%e3(EHc-jHex~er{*-)|p5Uo!h2Z&Kfdq8KPW-IQfI}U9J7$N{F1nhM@;APq;@!AldWY~0U6)!~zEH)2n&2?;mNjvm`u0vu56TnQ7-dKh(NQ@Ac3)Tc^MQ|0!*B}>sV5>OUu|p_M7@FW?AnVX=Lrxn}D}b@9%H+t!icL&d5tic+yuk{9SfbVD0k){t8M$?0VmWA#HjH?nP8SwsCvF3Uc!gadyX7TVjW%#{>~(-AO>)S#pio<}>IVxWmmoH{-m#(UY^WB`3$e0cCMWH7VP~WR@Bwc2Z6IsNc<5C!&kTTd(^2jB5d7enNdj#O2?lU!dQf`?m=-gVsC#?>K9-quFd=9i7&eGvXnVZiwAvP|U{xkVqQz~G5D&&41W+Il@@*J^0GsM64mvOefB<2(+D;fao&^BqwM?iJ@SUnRedw1q5Lg>B5Eu+d18)Uh$HioU#tP+Q=s#FaK(tVw9&mr#Y?CO%ifxlFS4;XP(3Ne%%y10I4THR!kmoLhGhmO(^U$d-xlTNfJtqbX4it|d9{nJ8RBP8KF>AErutQh`SRQ63tciBi!lG06<6uXiDew(#6vE&nH48(+aAC2zs_)mM{k#zCHD>Qthy?-749N?z1O!{$FDC!73ELis3w8(Uwb>O%;KXdHG6Z4)@~`8;7zUmLNQOxb><Tn!18b`|1hS7-C$t}IF2GO_H~`gPd;0AtRQD#Z(7ez_56JFlb;z{9oF&A@z8PE-1&{d8MMI5gDFz6nSHNxZZ?qCU)9I`PfYU4B<7g#}J!=K*VnEjs5yNkMeVog;#yRr(xc&RL{;I!zu#%;Sl&PEYOXRrJ_qJ|w5?F2hZl6*py=4mdtFyqrT)cWwM#^+G4#=RgU!=L*vkROlQbk0U{7rh*X?`E2xa>K*XSI-7?(*OBpc_g&PnQFJCGR85E*Vtr8_kp3Yc<!Z=g~A-Y)N{FYw6}_PA2~|+0buAI{Igz!)L{-Et?A$@14v~N0cjLKp&L2X^JtYN8+NNsV0lxJ-D~EE1u9x`Bx?T|Hi$YLI3*){k`qAt-<%!ILiYaS}gvMreG{Tm$xeK7vko*$j<HCo9mmD)>m9Yd%rL559;2$IiG|l<4JQiPcK`Z-3+4BiN4say|uwU*pIUmKQ+5_Jw?M3f9x(@{m^K2>CT3R*QMJf8XfVQ*%3e5X<1p=mcKjn+|e6tgWtYbMvsmD{^p(S#U%OSWP{$FH7Eng@OYf3t3A$>JL*)~&ArKtq8GJCLto1VzL2Vb^N6Hr<khSJ9PG-uXd6l*L)7=<b4q+jv4e}*lGC#nbQRZV#1|6Hv=B{<CdcQLk1;*>p@cxr;!}S-ABI{qIYl>Sl;|6O7xQvbl<%x@Oi_Rp({y#Ncuvx*4Yr2khwGCO^>Q+;DU7+896iwn13HW}7}PMVq~4vKpV1vm&Fpb~m40ZROygu$;KQ^6x{!Dj4>)%*)+^W3z=wuEpHFA?>f7}2d#|Va%OdNeUNj8nHBrOTko`!clA*5Y=y}%7G{TwJar1l}&qDt!u0@BU=uv%XTKNHne>UPXXng8wI$8{{E?%6qWHsh@XPIuJWz^~RF^)(TnUgu*wjAy+q)V?fef3rGyqQ1;)~3yQ{J4CN(7DL|B#JQw>VHj7rg1YFjw3q$J{8j8yAThD|I331vj-0zTzmJuqZ{9?eYgAILGzR0SzT|G=q1~#FJ$PSosZ@Vr!JUp<8TSzR;9(l?8eh^Z<^>ggC_%YL@xZl&aMrpvjiK+=TGtXV$i()UD08EvC1^9%%O5c&^cdC7B^>D=;29Cw%T|kQdd>q;P$AVUA_h`m1$m^PkcI5AC4yHl#pVsA5}&F-v+e<pDy^CY4ej<o6!YsLaJmELd#OjjVNKu6LAbs93PV^uTfPuCk&?y{lin#O-Gwfs4mKRs?l{4nz-JytI-u_G*4%fF<Lo2pUpGjrVZ!K$`vVccrpE!EJ_(0Ig-~fJytI*;T!Yn`NgMs>d9iUVZ@h15{HO{fIRndf^TxtNIDTzQFp!aJ}31yG}%3_7BzmS2mQ4C(8%d&FXxJ>t}@z#2hojzJe!CLufFx}^t79?UHZmh_w-aGbBCn9QK(y-c*2+Z4Sk^B$lkyUB(mz=qwHp}!h5DYpqTN+^`O^}D(5quOR-!JQTC`uCmZGS#>4nYZ{(i_k*`sqoDz>}mH`SpZCyYl%<3X#N6oAk*U>^bc^E>kzRyGS;w{83ox({6NMve&I1TPXSq`6_FS&nlamwV`wX^fe%NFtYEmphaR)H~AX7d%foKplMm=|@?4Xq8mh;;7n;lJ&Jz5DwWcb<O2H^vq&FivzrYy+$_$SQv{;HhSzN7`WC<5wy##VuA%$HXfO=PRo@e_VB>LcCPCFPYQ)QH}XylKJ$2^G=brsHJzn_Vn|4vf}CW@6aq&p#af>ye6(XsbsMel{#d){PN)Ov}dfIly97_y3TLuRgcrsZLHwM!ljs=QCDxwgqDA^nqRy^l18E2AsJmEwfF+b^MsT0{`vPuO7fzjD-wScPS{OvTyZUK#f6dM>AO8Xl~Y^?d~c+<vLe4Rvb<P+u(><f-~2&e+#va;iz8_))4!;>rlpMk-L$iBaXY5$@$BS$J{&cp;q2kyd<F@?Dc|bNWE#isi33KP+k=hG{r3mA_xgqYzhz2nm<}cThbr4&yWKzdXs~~<v-_r@Vzr!+4(E}rCea)3GoI!bqsmc2gb-3_4rc@Tk-4v4o&!pQfO%NfBVKo@p5j#sGl2vF_ee+rS}Y6FpyH;^^5v0NuIbY4k;?p4j7)T_Jhb(-^}GFL1I*JR^yJ;4a&$7+e)+m^Uc_nBKNz0f&fBH6{7HA|)uf*QUC-v9Q1-d>LdvRzu)eomMYZ8LnE-GMSs)Ba9DLHGXChxyy1<7-jc7QP0(?jbg*mHTdetsr4sqWM>kmpBSdAyBInEeUQ*guM?&GY+`chEI?=D$~mHv_ZfNJguh(k5mzK|fdclPeB9q>%bWSLiGr_V1}Isj@`Y<-}a;iEXsKS0R{G>{A}Ycgf8Ca%|5)pYH~TP$@Q_7f_;xxW8FmM&Xn_{h!y%j~^8E9{m?JH)A|t{8;q%4`6$;*3tz@&$&``RUoLc9>6ZHCG&VE#v5@A^30L&%)tQ=2E*@CO-stGnO(jiCLcC`ciR!uYd1We{Y{p)k<rSZXGBVJlNTx>+wEHOINWxk12ZG5F{vG8zK78z4D%V@gCENZ(4YFNr&%m@2+v4`sErbD<ZmkwW5kB!_k}3-QXz(*EQ_uB3p?|O&wZRRBZu>@q^-J65r{S2YrP;oS&ZcxQ;i&%D2BKJC~77hB~4@EE{Lju(5c9b~KxJr7EVJ7}W7e@S_?P#!r;<<rrXD#X4_=N9ac4B5isLv?<TBoZMR<9rX9#q?!viLdTUiJug*w-{7|70v~|q%ZbIg=~YhpNIw_Sfmhcuj$^(VK&E*8vo3X%_iA0quA&{~;z#uHDs=XZG=rE)y{pV9byigwBG2;u?fv_^bn7tR#*-zM&N6OOuMWGmdjyfVdNnkrboq%+)rotQyIo$!_>-<<eVI7IlIblb*UlUy!n3VsNfXL<re(JBm6~D}A9I;*k*o7zE4Lrfup5Px++Aj$h4l1SHkdnWx`h4o982+?N9t|k=`)G3m~F4BJH1`0`&8Vn)Ya2$-?<bO!nL6j_(E%R<2qG)aq%|LLtb$p`HQEhe5?B~lJ~eTU*5&rqz~C`ag<iNcy)rZM7YKDewD)YE#1PmVh`VDWa%E#(yZ?3&I<i$e3MzcV!osX%iiMds`u2(YL+fskqBPCR=k<(ie|F~O}K<5OzDcIENdEN=a4c%?y2e*^=|>AzF88<TP?3#yQLYrf*6I_TQ&k|yWA*e|C8`e{#<#_iOj9h-=^C}aw@QZI$D~E{cUc=puLq|<$lZR+fv+6$eQiE40d+=+brJ4T;Cv}9D0H!=zAsDC2TnehK-u6TY4OeAoGgtxW#+QGD|tTb4mzxw6bDy>bi>R=;kSPn_PP;`*Wc$7ag-S5u6YnG{$d7uQDOZ;vuCsc{KTg7I~*JFQ7N5%&Vc|xOascnEOIQJCt9J<P=@$n;-XjMwjjnOPKl@sWP&1rs7nUEZBSUmLM?4sV`F->(w`lssB^c7bf}@;TfN20T;!DBuej|jxW+)t<v3y#iS=MF27pVk+;4p?%U=IsRH>{C3(%l8~kILVg9BI;w-27f;jb}xJ2(uJfKz2I41Q+oT3xNU1*Rfp5I63p}8gV{nq6Db3|E+A~K1lUih?xnEzoS@-2HzvJ_A$`<3V8EVI{JUbj~kdnW!5i{amj+<W<|!1?r3SKK91B^(PvZ~^4-#p>l)l7+|(Np4!31P+v_5=s5y-llk{>}yi)&gm?UYLydP?F36Jla1GVJ!+cozB@SR-(xgTUkEb!f2n)d?lz7rLGZ6+sivJsF$t4;_yIcY1Cfx#F&{D?memjzoy3Dg2?S^Wlt_`R<40AwYO1<DHQjr5&g@Kg*Up|j`(@6`vSrJ%EuXXd2gturzhv(tA~GT}G6Bk}ni+SO2p}UPZrr$W<HmiOa@dgi03qEc<(`i}k->?Q)1U@GnBVMv(D+-2qoGb_<2o0DtBZV>D4%h;oS*E3V&lVHiE-HClv##&A?*aILN+>i=)!yOc651cK|IABY^EO_W%i3Xk9qBe(UC};k5^bCqT+C`+Zh$^UMQ9BUI?m}YD_FcyZj-Y)OvyMyBGAkBfJ}6`8_J725m4aAg3^lvFfW%`69s_7Ll%OaPN^S+HBfYlUmMr`EVq)ugCfl>$7&d?bM9|aZv=_mV^{pkDbD$16u1Qlzh5{`(bTm-HSYZ=b}$p$*25QAHh-j9Xbgfi?5j9>N$RWE}s;SX2RJ^Cn9gwSErAT;h*Pczs5vY$1(>Qc@jQ7`wevihw<m+3Hh<;mw&(%TgRLyOCRV*k=pD>%478dB?o)N=eqTfdVP9K3BEAvm3sX-c`Z^lGXmuCG375NugBxFPbH5${Q>9l#>wf|)E<U9hUPo~J(|6G0zG_s_FIF{(Nho5g)ad_Uvm<(v)>(`K7&p<W7*TQkB&LD7{Vfj5(x|CK@*;%kFXEUyT<;Sh=;G$PvmHcpni2sz2F)c?Y<=L{uC3lk$2~Bj?ey{vYkES(e<{^$Z_-$C10atX2+Dg>=F8g!Yn1c-&(W1jI#wCi<1OV2mKt2gl^yQ@{2ELf55L#&;j)H8DkedJwnR(p76?^vOmBCDc$sE#{fA?Y+peiaBLw=&xKzp5fICEIJ@rzJv=*x9=`{GK4pr;#Rqi5ACa1lQT#3bi}dytLH?Mq`D^<812+E=-$Rmlh8@B3e1fE5P5uf_{tHTb{WZ?(n85y6lKWS*K$M~l7yMIb`Z2)pmG~mk@!?aN|JO!W2~i?aiQr~V^v(2)5dMP&(p3OSK*0{ipEK!x!c)^{_7!hB0Ody%26zMh$aImkC#p>6kUJLC_B&iZg6@rz!aF5^2or__k`3SI02k#k`89t}I})~qU=;+$N3{89HJ`yazsFsYeh$DthWUI0bi}&`fch0?+#}AyFj9`dPjq#t%r9SGSY0kJy7^nR;6I4BTUc+wWNl)r1Lh^`vzPz?Xt73v7rw5uLIzf#3%XP-K>#WEt=+wD|By9{{@G{fAx?Ys2e~-nHO8`FlgApU=$rjJ{Ri3(!LjP=`3v`;amBMgNbqguY>jit(L#RwOz_t4`7qG<zU&;cR(_0YOt{(FA7XxWTA%N6U&s+sEZ?y{={C8?h}fzBj@hK_<a^vX#%Us-Jh!tz%F_|^Gd>kt>8U^<Ip@pMBlRZDwbQ%b8jb`Vq>oJ8_CL}-6!8QSV~>eje9iY#=!aaKyGTnoG@W1=Z9vVhoQQo)%q2aotLf*k(;14#Op|gF5*8kG52TAaCRXh<csB_OSF3?X3Bdg+pS9wnM5S7EH3_WrKz0vNSpd1@4?1UHW=LxWr>5(PNL~7j`J981>WuceI`kePBTKhV$P53);~)M$PLG>>bWZN{sUOr*as#j-xEPQPYB=gQ?%}+@mb?-UOTj;mlONwa(fKeh<P&1@PcSY-Ez~8JW9SIB8Mk~!2$$oljQ?Do2~hZ0PUbRm5M9K<ZVdq^IHotA@Ur1smyEZp##!Q!AXfR-kLqZ8ZqCi<&X`BsXfWTPWBk*z&tSox!ZGq&W9lrT{Qzyg770{;mwryG2OKN)ga$;TX=l`7s(-06n@coHNr^65Exr7i@?VM?FEcf^;_hgNW|S$BIYlr_Ow+%9mL+xBLSR~q(<L`*oSZFugvv_k2=PxBO0fr^7<nx6$KT<MpHQ3?GK?>2;9<YfiHX;0f~I25S568<c^?t`q8-7VGyB7pdCZ5#?{EWXj4>ck6TW5$qnenCJt*ErEblo<9qDzwO6v;O@Ihw;SKPQ;j|Z(-a_V!)b5`>=PSP?WA4C1PNqb@<0g%ri5c0mRLT`VpS<;W$T76U}y)Hk}=%R>h=RQ|bHN{4U;HYHGDy-hqno~%YC)vl8UOUamuMmn8yYX<3t99{FoZJs+oxT#OpIHi!yu}v5m^w%%7o?taPX%rE#TjXxIb$vq>67w{)Df?re&R?5>e=q!Zw-26IqJ%h&8v=fyD{!Qj;oG#+4vRF&?iFZ!wk;#t+*Ta8d3(3Nyd}AiuS`-#ER&m@C8klZ#;(ZB`-P}0I-7EeWCKf{L<z^Wud;fGQYZ9iG1i?q2rAcVo4~25&`1Adn#m;ukhabdkV)SG4vDSaL=^!Unm(!Dl`QAbQ@cu(cUJ8OWa5U3+YxKx6CKMlcAyNAiZR^k|3}gVRrDGBnTw8-^epqoA)zGV{hohk)Uv`f^oI=mDQEyN_l;At+HHM*@%3Rke<@W15G3QlE(l_rdq$Ri@i+jTj_V6+9}!3K4+?aDuPhuBEv@FDLf@sE`E4MC&|Y|5b|^-PUUAsMo>gwr+rvgUN;4{xJWHrNakhOiX&UUH|mVKEsAm3-zp6JyLX3EK*64m4(kPS*(^5VVXNKmHVZ)@D<RMA_sGkEdej;?{qt!E^gT=BLY#;XsbL^hh3xAJP3l@3KQ#hPRGKyIJG9+bU8^g?O)@%+BD`;t={kYh+Hr1p%WG?^x9c0_53pQ2G~S%&av^K0AKV!~nBQx*yQH$0ODCOhOPQ{-dbrL1AlDDyjhlt*Yz}7p)yvDPo2exg6VoSk_FFpG;B+CmOA2c#=5cKDhHu0T^kW|=-v!D%4zB@ZLGp7?nxZ&Z{W%A`lLU`{y@E@Vd-gfjih#DirGthf3GeOMtAX$BwqkTM9D3{JrOJ9pi}#>ExYrsCi{~IiSmhrxz&h5Rd`-6*z9Gw}c=&zE5|^~BIiG&R#{t}hlILno)9WQrtE)d~iuLzbmo}H#jlQycYiae4V-Nq_D|1~Vv}Rxx(MH_$SWuPNLDzQ*Yuq38cR_H57TrT{r$5~5z`2e-j)RzQfq>p(+JMpQ?}j3z)5Rr(3sz|iA<CdoleDpK-*;Ly6xUhbytcl9L%v_BFPAswZ#p^o#_9%zdo0)AFE4FY)}tf5&U{SzndhWq5fw0=^F;@qs1GtL0<|S&(NlOf>a*Ab=#n>$-(k_i<jbS8&-EuybcH1Cku{7y0hlT)>y!YXK%SlVJI(!=t8I|4r%--0@nR$HHKX`I76#`^&PnTL>i{F4c10^#Rjf1wf^TNTb*tOi?qJ2JVYkz071XiT?ChreL(%ZbTvQwiDNZVuhYF6OOHk#;>e`+9`mM@5yM4im&95#km2a)vECGq>9`@<NjhlX`Il{>ir|238U#j<N`nf63RSB|93Z-1ndoHU>N2PLI;~of11!3Kvy5}TR#soDkFQw*HuY7Q8X>opWLv)I6hS%23jjR<nVpnR1NJ3d3{FzIo(p6laQfX!e|A+MVpKMl+oJnKmNVFbi=ujsNQ0}ks&HtN^hSNuWP@SnIF*0h8&zD!Yz|sfI?+qvINr3IfTYdC-KK+x_r~#v=ydsY*?e`;5%Eq`A(T`HQ_=xUSbnU>Gz9pc6$xEfa4fXBvjmpFTUlstXOTH7WQbR(+lLBA(RG_*=!lU#lHb6Hk<)w|A=Rsj&zeb31&r`0OWZ&*&+83uy^z;#&!lE9Mp%3Q{zfWWqFHe~+^?0X`g3wD;r?k*66UW&fJp1a)13sf4p3~j_*K$a_ZT%J{rL#}G*{hy~ayLcjN2#?2e6(6;YI;SGl~{b!wfT-D`!4yHF3nugId{o_(?3CY(l8LOSO>0Pk|%p;SdCRNGk+sK5G4#l4@n-wsn9nn<Z~Y;OiZ#D4+i}Q#A7`I+}qxdsfe1>AHXc3Mm2dO%>e_8V7dyJilXL0Sgje^D}*QY{N~2$_3QO(K=aWNmoy}YAS?rn|2sTnACAMrtO}`efM9xOn*3j{SIQf3kE!3<yat%#I)F>%8@})caaAJK>*cP2tRnrrC<Y0qO40ghkdwjZtoPtEz31+Z0hPa^SNMvYke<mxH4ur4)Xd6joS_I7whb#cqIdxOLHOysSJs;YeJoH$D|%<WvILjH$^w2?PuQ~uf4XrM&ZP2<mDP1D!c8wgtK_*rLS<?3#^SZbrNxaq_-k_b$%e@+mddbNM8q)D^7`sZME+JFs!`}imN4=m9^&_v7;*Q4fA_)^|E8jQ=*JNLg+IX*eX}~D((Hwl^m3o>)k77EOGWKP8WOmjA7DXAAtc-n{lyi^2`0Uu=26^U$w;mMAox@Fle`h|HAJv3_`wu@aZPvqy(L-pkTDM5D<4#V&~%K>0rr3!z%zFZXLIJm54-ijey;$3LyU9W9Y%CyUsT$P$oE5xE5g4+%yfbu_Snx>#3EW)iwE1oh~gKyN7OdJww@c!L>P%*-0cI-`n^uWR#CW}jYaYrtVhn(<5~nQjuOmas7u5xitW~@u>-(!Fs%TQ-OEIaX4@6n@4%xo@<PhFK80CPxC(yafQ4=P1^#GWp@Px2y7Aty)vWJ|;;IE33)7>S;38I}!>(XKWC(_p1_2%wDLT7_%AcLqfe6KMy2r~45#VPMKiGx?@x!<cXVr(@L?KmprxO&mclPSX+nv$Q{#G4sVfCOIPd}WQezRUfWmR>pvQS3GQ3G=M(3Oa#Nhap#r=ga(O=WFubqx_L=fbxd=}_Cyy<K<V_!d14=oyCA&fsYZP{OZbuGP|P^7FYD0wbBwd2)pDz6b??S|w=iC$J=Wj*x_F=c11jH}y)cAjy_d9L$l2Fg8fg<@r-u0P@kj66amI7ZODLqI4Fe8`IAs!qAB;2szuc-U@=WG^7;sQ@__bEW|eRXf&djoo`byZvitBH$UZoM!Ijxk3Hy2c$fCK(1X0LH!^=o+9LbQ0vs-jfCcm_#VhlR06&K@h=xaI)_oc2BYjP--pjxCs!ML8VfEbK?Tn%$EfiEZ$fkovrPuTMB1=*GZ2Yj-8g%+iXY{EET*X{_0036`mY+qp{7uEFhURjKpD|erh|}upP6rbK9jfn?Cy>fT8^4UOgWdgJz1zCq>Wa_J4qo~jBmSy#4T`5iSt38yEvN+0;+yPNfgIC_A_pR}xLx6=gMR5a>87EXF~6BO(ti@8V*OM`Y+VEtoH%z#wLz5A5VQ_-(vY$Xl18S4a3774A9f3Xk9dgr|5C~Ixc~ND5Q^`B6uG`pSgYJxD$iF^e>dQEP@on^)ZIzZq7ARS?!8}Lo4;9J3*hpHrVq?d6?%=GL1Ff_nLvDndBZ<N*j4bt!YYXHl|UEjtXx&wgT1=CeIZ5Ec{S2Icso*eXWpWe6sQsRps5?Je6hw<n|$fNamBBZ${6mAGScc!eR*?fV^J76n9o$VxtU7`Dq3W&33_|=kk+NEct{znoH@5J`G>)9=YUJmOZl8CZ9an6rAh~zAn_c%Q1Cfz{}U~+yrKA;5rSanOau-S0h!IO;+@+dr?_||WO$n+1t&I5!Etqg_^j1ZKLq%$U?DORLjWFvxx+!^D|(NM{*)Y_CI>=14!gLrQMplB3oOMYafmtys5E{nf(r`l>DIwsU1WMIFxPMN<8EsRWZH;_xL$Cxk)s&A2JbZbEHUH+XPQnWd@MB5q_UBoE5iTbQ*=uas4==XPSTCF%}Oc*56FmZyu94Kz}H4^%*qg{SSy2H%oeG*sWH70z2){whSwup?Gy6>{*)SyG?WF{!yAh$_MGmS!_YXrx^{GQTElT29vu!&#EVEYW81^n74P=V$_l;dU8`)|uHYB9P5}U1Sx7w;nuU}cxq;p0&G7gbr0(NmX5Kiw_Ry=WEYL&-tp)~y!UjoC2H+~F8Si#@&LIG#{Td(8ulB3g*H)K3%2y|X)C=#x9o+l3tBWg~LJ|k`w!GE+DMVW!wU01^aHr`|;S;(56lm7pLT^-Z3UdMki?CLq#&1Pg80P-0t)7<OvDHMIe=>sSt^ZEo94B`#F0Okks~g_R=F*b=@{P6C&0F5JJKk(~DZK36EWf|Fa>J%h!gug?bfglmEv_t}Vr_K|PQR7K`Ndo1rNxcKim6EB5mwh0Dr-1`#R06XYi?NPIL=d9S8G)9#>Z*JWVDj7G1d|#har+nJ%`HZ?&oA5MwMt?yBJ_VH0wL#2`ys!&~CFB8k0=xQCTQ_t&^pHFJvJwvW;}c{0J_mdr1FY$*^N_Z>gqn@34^A($tNlFxL8*wpV7_BZC~{wq_%<>|t!8%dr}!d)QoAhdsQZ%#Gz6H`XdQNF$u7c5KI(=^1AAfQ93iCUwmrMx)e4)0sa{2xPc!B$T0A@})^+i9S>4ayj>gefDEEmpzb$WI7&QK0{uzc-<+cHEp>J#d`nGzx_3f^omabBI+Cp-3f{h+nsK=a485*Kun?BVqw$F_dtZ53E>ofq&(#ONkrhL<xCi9)Eb~vL!9#`iJUOC=$-L0r|*;!1t0|h$p18QVoL8FKbSy9MA14ez^PL^9Z9c=;76teexeKD_}M<x!y|dM<+kwch{+;-N>3<(IHU?&Cdr)W1k%Zblbt|-5faP`i0MGdS-_oJIFH`(j*`uGBBdqqHRpH-VxluFB!f)SMdmTh^&gUB8W*0=MZP(cu%X;C-52TYZhtH8dMP`V*~)3gsxu>1b~Hb=H+VXf#*eSL_W5GIfix$~HJW&nsLhqXXl<%cRlcF5a8P4NB}0l`vZkk+&w`VDm4mpq&1of>``Yy;>Wuv4h!7(L<lEY5vkc*6fqVU7<R{{%Kc)KEjrR%wQpkgIb{-mjY8G!}xxEhecXu&$k#=ppy|{7HyN|?KI0Br*bwn_!BME+*CuH8v?9DgkjsScoPxwx*XNo=OcQJb~=e?{~A7RE)_r+}TiM-q^r{Ejb^ndzP^Dd!(#YNk2V$fa@y7=?)a^E`wdUvut(<<P+FE|op5lgJY|By?xeQr4#bao3zVhdI~wG-GaaEKMOz4aj*y_2LE!hgf|g9~5>2Y!wqM%ysUQsGEhoM^;+Lb4PowTV!l`VYEVD%~STgOFRsY>2iszMR5RCO$J_HfD_(lCq)aQZAG<>Rk39sp}f&)Sai)YxZ#DA4*Kg7tDu=^<u#E0ZXKDE<n9}qf244j?N-VcOtIBs{1fGJhJY;WbdalQQU+XQ3Ow7{HJuM)fxK$p|r^7ng<=Xt)>u4{3(TDdp-C{YfMXx_TtlGCzYj1=QoP)wIXynlJJB~09G^0*Tv~kU_fUJUL}Im=jD|t@n{v{M^TyzU9Xj8!!Id_ps(Q`<;;Rt0A9RiV+|_|C@UeM%9hjg6dey_lC8Slw-|6b=~!9dl~luQfwd76a4Nc`s-=u(#0E}$jIM1ip!b$c2t=tiMf0~A)8duN2RF-`>*U8xM)1X@4UyE&soId8IJR`2(01PGF>LU9LIA=wFH@UA9yjVAdn4}|d+Fl2H_?=zsMn0s(OZBkZK%rz^ovKSvAuUv9Az&`c*&l%<)*4gNVRkBytVSmyNfF~9JNefgko!VZ**9i07rNNp^qZ03BEV#*Qpd4kaVL@o{BE#mTq*^BxewKv+5Rdg)5-@aC68r{d?hqox><!Cphe8=iTq$%kV9?Jn$x9COQkDyH|;mnl@XF&X5wS0jPqz7&+;O-SecI&(0p%K{Mruv8&S<9DHv%RK%;-*6Sp2`8CvprCudLDdS!8GL>)Vlt;lWVTl&LE(%5O_}SMQ`%+;^9L~aGiyxG{h5g38g=?~$09P2}vZEdi?A^UUTDZFxfOx&FmiOU)A1K99@r$xq=H6OD_Rwv@Pp;)E2p*E?Q&Kt+lxu}FDvTV6H23O{@6Am!XbFOGNDfw$&uDAO>+&^p*l4#~0~kB^#f$Pz5I=A=qnQensat0Dd~DG?V~G@-R10P>%3-(KtZ^ii)Syq!Hz}*%yg00UHz%QC1^3L}4ALn$;^rDgi`%2ZY%s_BZZ{s@6HRP%6uYg0`;O!+*o3{|pphK;m1Kmu;Z8hgMeS}s9u<<NMRXSrUS-KTY0W<e?%B)Ui)f?BO7UxWL|fc6pOpe!-7axZ;jO2}W1uKe`T+e5@z3N8R_Dxq5|O*PviRO+g_nDw{%++C$spAC+v=Jzzq)cA8UR(mRTyp-N<N|pS@jq?sH|)(Zrovp>CB!zzVjTCV(MpN>Gm7dUrOABRbTBY@{BSG30<Mc5K*g91&Kl?{jekb`RVZ$dZQ|bDZ*jfiwi;Ksilu57Iw^`m4g;71cU)^?(cO;G;0>dAf^^D8M6sc5QB8sU)|iewdt&3;*e+MB_~@7CD=eb31Lr+)oBXpTniciv|!L4`s!zk>`JQU-pbcEu*3}4ci^n<Il-F_>Ce>i4$zSgj@#xIGF_^&Hi#RJI&@e!#VGIM?{;S{*NMkoHr`c*UFP7OI*R@|FXH3$o0fAtVtIxtvpM|E$^6Gpt@|S->TdHRgL~P&IHauvIY*f#Xe%b)QoClWHG0s3UfnvqFk?rJ+tRR=n*+EeC|XYy)l(MgPIDMZu(@>Q)?q>GJxIgc+wXM%xw&2HU#j|-x1&q7A{m;irP-O8S_T5;hpK{2fm8%Wo!tMi&!Fape5VyR;U?iU-{MlrzJs?T07cY66~krE8fIWNO(4_SOyGzSl6N@#xB=Ja4~VP4Hr}M#$QW#Nt-M~rQzunSr`<Wg@a|;Y22?bmAne-Fk%onm=EZAKNl93q11H}XgH&zJNG<a3TzO*#J>qBmIY%qH@Ib56e@XHUo&CW}l5eR%MwuP$^pUe8GSXs4&rTl~z0>dgu>gt_Ch}*!OP8;NSG8_Ab2aw%A1GWrc>=f8)<8bND92P*q_fXW_*w^dFYrZyUi&9bQ*-uv=+<&c2f0JmG%ql>68+ShZ9f#BCkn*TBXIor)mwMeg+eGLtlk4}W7YGgSe*?20}hHRW3Rwf;5<<QE^wV-qrO>hhk_+VdbUz&{m#n9%?diqBKd-ZLPD{#P7cK(37CrFvFq;M>D`ZPw^{%qq)JyeE;-8}Cg(c5$pSYQ?SpS7$rf857Brewr@GpG%GC(NNH^}dhHJ)jItyJwlxfmbP$%Ps=i3QkLhobd4lt!WQ0U4WkC>xrQ<PU1O#HHXdp*C`l&Dt3lIW~F?Cs$fu99gq9rf!y5U~-nv#|M!S9J_iX75UUp)$X?TwWTpp3v<JNoi`dh0R+ORqE!RdJWTV!Wa~1VCLRU7(+g1P|a<|U~vW=J|$hiwRdG<W6i9q>C$E_OXt>UGShof&?$Mbw=i>cy4$&jfp!?1K-xH}As2H|YuL5k>bo1utFDfz*!ghO>2^{5fd!oRo2_Q(VTv*m$VWquY|)tgec<iTb!~{+l19JNYhVpiM{$JKeaVU<x&1pR36(C%O9lxj>co=E5DKJ+Ha3IYL8@kmN-ff!7588qaR^S>3wzYm0IJh}<(#m%J8@_JXWLgtl_sGe_;0CQ&CR{7gFw1lB*%dUiW^XLW=)m8)_WTx{!V_5*udGRNdnB|he~EK8=N=s{Pdwi87K53XQ}djWhwit(bqlayrEI=cT#7Lf+S2Pn8k!GSrWPn-il^JpCHtZl+BgO2jXg$RW&qzzz7fP;Z8hsxu7V(vg!-(u`7zUE<fQ2Lcf*d6e7nJ0au3GN`h0ReoL7jQ5C4nCu+)QJYkD6?DB9g$<r38DZ8A5SFe$dhH;HN)iNCBE*I(;y7$qx7~z6pA~Z4*i%v0V*;1waE@d}Mg!Vf6n>p*OV$F%}VYafXxZfMM_k+1)<+5s9ru(NGvB_U&UmkzO-d)mvvKh&YPbNBl&Tg%40skzsD|s^K+UX-tPtL~qL6USf&rcuw<Drq410!RIf`a^@;pv<3pZrxhLgW45vSWxLaLr*03$b7_j(Ak4;)_L$NZZD;z1S(`B*0BIGu27^VO^S?nGI?oH1ET@5{k3dj(r5=;+=D`vK4jM>*R};YW}iUEj569*;F&RZd$f4`A}az?QzDg&=!%GHKcAPTN}q|P1+^>u*M&P7d>9B-W+!|YI2?n8>`1s2dGGKbKW}}5q{1Aik8$%;ltc}5J42mxrA;K3}2ActW-fE27+eG`LXk!6sK-UA}yogc93^_xD-WN(jX&6ot8RC`~4&}7$?bOjqKtIXVrAZ^piO~vJ{im>DgK)eEdq)bYiK!jVTsH9!2ro>CXZem|3G);?<IRl=!#6qZC5PrJ+u45U!btOiA#{!sh(D3)lEfjYc_>M;lAFb)`iYq}xHKHT2-Rqw4)JdlU%{hI1ad40IY;9|{lBIq&Tw0r1@mI!^9hsGVpjmWJN++up2FXEG1-v$K30rE-qQO!i!dR9?wLa$c#kD_HU<mJdu{BlqITrfv_U*^qD~)bL!WEcAzmz0pn!TG#tMbSiR{FUu5RP-=IIFoy8rY+<uZ*G95Sd+!1JasZK2@LTei$)%<yT%R4<|0){?O0_v`VDB@u42VaD0ia5Tfl@6iWso3gNzL`O3R+vDY-5R@lKH*OMNH3YrV~0cwQ(YOS*dD>`}7%vkr^ZzY_Ia{r{1`ty#b#u%Qx6`lxluiYYv|;8S#Wxs!Bcw>qL_{O%TxmbK5(kf~&R~mhMGyg}h$flb5+s-1#M;0~yw*SCY#R%qbrO3`c=8IBY4MyD75<wBaL|IOT2B7gletU@cEkP?xv)%=sO7;>fH;=Pto(!v7*gWC6hJE1NU$#tvRdg759v6%NDK0#1Am2Thd;4`|(WKMX+%hW7<6k6Ie1zi3{uT<O#V*S@)*y3(q)V(;yl#JGx{#jIUTmcEx)fE7be>XXNv=>MT<tA&W}4kJr`54}{Jn>+R`uBOF5>AqlH3332w4pIc<Pif!*6})UZ5>Aq{FO{BdR>~OyipFB_z=jzIrTh}Th%>q=LR-F=&(zY_ES2jvzcV6@oX+@4l_btJ!i-Q#gS!yx@$)-y$LrYt2}M%|&MAxw)koOpL=jK+f2`j}3ve!1VjU7{k3`|u)<L|xhxv&1&bRDq<=dp$L54jcFTwxNff0a2rXnHE;4CPjqbDW50Rl3A$nPEyCq{^7N?Cmhc2pyOX-cLI=VhX=Vn3t&aWbMT?EldE)!69|Tj;2^I}FM1pTvnMC%7*~BX*>pSUS>n0JjUq1A<<lZ1s|faL`!9-ge{P#{VWlCb`L!(PwL1%>TKEA@5{%_eT?!$Qo6P#zZ!{7mv~L0>-kJhOETYrg^vdlfXU^;pGA$ydQ*PGwnrr*9u-z*S&sEUMScfO^o9-lCrDXOBz{oVE+YEh~~wp??s0@<76NRw|BR_mxyGdy|S{oTv=P3mw{Z_W7CpP)TKCzGnDv~mXgm#_4$N7Bip4C=`#~w=H$H$swKEGrOL^>vTvn<b{dG&&FTx+qr8h(@-D1DJ1J2>Ton1&H<y;!oK3%*-O@-X;ja63d5Cu0>B33ngr)jR=g&pPE_r5<Fdd{*G2&k4u3hY^zdsUoP<8^;XIaPBu)Ii-w$2>1Ej}2)-LXExYW3PiN(1NEPnh`zxN3AM2r2O>aD&L-AGN37$W~`@#~bAvm07cYpdm~ciu$bUG&zsdTF}NDfdD=%c}LpBPhNek=q*rQP<(<DLE1XW%=+oA?e{Q)X=O0Lyz@9z{np~G$`VL8CLkq`LvDDas%B*q`#^1hTFnvF6QpvT=S^8%m?oqK;c#MAr*C7bHY$mNas%If8#}GWy&+elcXp2D=uxc%6iUg<Cxw8y`N^ErI7)wpqinTK!p2VSUU2T$y+MB)=yF)l(i?>{6hJQ$y|B0xg#PKjry9gh&pz>}{@jz(@4eHf9_8!*O0R`Q`fzPDs#D+Ygs~YnW%M7;rt(JH;SX~pmJOr4{Eb$^0N3OY636t~{t-Mm!RdU%6mK>-p^61u`irU@Ja)vXO|I}!IoC5;Kbzc2mPned-8mtxr$c*V1w1Bc>FV7+R*40I-Hp>Zi(V3xoF|A%B9s@(w>C%*C4|h9e|vGEf<;THl$g3GY_2GIH+x!0Rg^B1$)a=~2M?EfpX++CogukQjF)W<6OslouT&DON=JAD&N#fkw1f&CU;`Y&k=bvzg==g^`hg4p#c758<Om}0Sl?iHJ)NW9qzl^ioL)^pP8NF|uR6~agMJbM?<A?d@leT@cKC5x=0T-&Eemy&CEFSbP9^|q%Eu!rHO`bT4le5Z3Z)CT6Jd{GxU=wo{^@_@I^9%F^br+8`G(73irO_)#QP(v7X1Sj)k)QS(2UIbx;aeDNFaI>KG?F-tqN)qFQha6Qbzu&qs+{KBBk)a@R~VDE1%5YqzVXk`2iK*;K3aSza^-NhE>1WM;jn|&7o^el4qUat<9;1(s<>%1;e|u>37qd$bb<wTJkePQS1!y8dIAr8hL{F-UC-q+CkdULWcXg5U$0T9pqv`#jW}2XnMB6V(YC#=Kq-KuFJm-i}P4g1|#9k*C>f{JxNR9!W?zRu@wbJTIC>k#lo?;E7mYE|6VJds(}Y>_Is^pCQDGjyA=vR8h{}>U1AZ@@43b=I?u9q<u_D9=$o_Od+Ac6Treo($a2odmESP#-2*W(8#^E#Zs*_uQz?pSGiVc6vHo=EnCWThClOX1!d-*w4QJlp$<L^H5Q;$iM5<$uUGjOxQIM?4R5Y7;nkB9N<|?L%K#?+8&PZFXob%3J=A;K>y+IV*K7q?k^7!P{Bdx)SpMB~&_Cs~MCZlicBrpI)`u1v^ld6nP!UU3&F<}Q39G_+g^mCde7$T$!)C1w-mM)vERm3UHDLC*EtgT;yWzVxAWt$K)_yynxR(?)5Fj~_IUc``yDq_+{jSZ2oB|m-WTIj~1^{e2h)~|esWlE8Jcv61WBp_4MC@cRUI-Jj;kDZii;uoM926#eRFty{Qe9|Dc(u&DDV+c&66_I1CSU$8^=p3vTgw|6dsYDJ?=?YgaED7m>(IG{N7eyYQ!ET!IiAqXt(!oq5fOMi`U(6K*S^fY|L0W^SK<PeV%r6-T%vzdrYb^hg1ZAo-CHHYi*`>6YJrNO-K(D3Wf#Fb|*FpUWl07rEMjvf`qgI4^>OrD*;ltj=X0=(flsll(4+_zG(LmHwYh%!v?7C6#sm7Oye-VWexn4SlgirxQtc&=Oai2K*<LM9X%ZgUT&A1zAV5X|L84c~Q<w@@p+q3yX4|!L*(*s}?joC~K+>P0~D#KxdTFptHBbkA}xq_$S()*;vXm((i-N!%yB5GE`Yd<Q*JQKCZ;{|;PMTn;HKs^Yteq&&yh^*F$mj@YI7gui7VQwn6cfDTT(5!~ZyEiC7!3jG!;Q)N<rN;hQ<j>5ON;9d`cY7Ad8)YHjw5TGOEvh2E@sPF<?3iH6KlZ2KnHBh$+X6S8-XeX|h%`oobfJB9lZI4;Xat*o)zuVclTs11lfmAG&RQKB=u6*=g(zK&x3>#6<wS*ao^BDk?FXSnQc)qD!+Y3yh%G^U?}98pB0(P7$-)X)tq4E~!Y=BVlE<RL0zQrp;4%JSUJIa=<EB8Zv7#bP7S#$i4oM46{s_apBo;owj)@bY?B=>fT!pj{6!&}Zd*dEvd=xc}qM0C3Z4LJ(TT9B3z~DF&W5P!0##^l}^J*oCh3=$p;UrlrLi12d^Z%^NkWN@A)A&;anoK8Gi}VJ`#>4!#Vv3Ydo?Z$8FYDx#TxjB+!i#nQtA^-=`{7I=_Zt3gf@6Y#QU-|XtW5E>qycPPuU9@mivtC8vMyKSJbwPxG6&N6WG(9L6(8&X)fbF88~PNk;4aH0arySpr+qFnY%cPN&zs-3Pu1*cpZ{>+;?^&CC^#Kv5^kPm*fr;Y4KtvKaUinI^$OQaMJwtZD<4ZDO^9<O_fQ&t1|=s=Wqx&Kbs1Je<)v9+8(e)+qlKgtR-y@y*m9x(qu_4%DRXpeU`rSx?p_q6vL9+jwag5Y!q*^B<C93obsB_N7erA?8Li21Qn#ISp;u+Jt&}cQtW?NtUl}P~ZBn+#Bp!n8iq^S@mc#vC+>H0on<Fs-J?J!BY^}^fG`^{cpGS<@G(;yELL}pZ1P@20UyH8CS^<SZDXI<}C`V+8wNkIZYN0VFm1LdW?|Nu}x7Ro6dC<bLe66O4UEz$wc56B~($)#5=_}I;WSR4VfRY>8`FsfGg55aB_9*icauMu>nOWI5k^-J+fuyn``$|aOLL*1nq;^B1zHnS+T*^RgP`QC#WL*9I^Np;p12W&$g^HfnRnbS3wPJDxOnPFilc_j5P^?0-O`wZg<gJG3hFLJ5*jlS~A`iea^7*yO`-_#^>5#4TF3dE&HTw^G-G1D}?7urP{b*BFI!>?N6VXFsD%!}>mqMT@CGtdlX%#Lp0+GO-MKCod@-?!EWg~@3ai_|Vj0H*{(+vtxhpH;X(f*Uck}Qzx9)=_Mc0psQ{gA|i;8~{ilSz53d0|gI54c1-#T_A4u3Il?Czdwi&%KSEmUs2iwD5RQMs(EiwGMXT{o$x{AI|O(2GO@(-EECx6dyTRL`bCFYc<d!JLq%|v7K<;r8v3X2zv8~GKzVV!y9OaQ-Czcj~T=&iV$>AIU$lH<ZisX)r`GiNpM3>o(LOMXO2!vNJ@3kE6ngE!wmCyNy2UxDj$F>1UnQdLO6H_mZ}A@?pi%pYB-16(0k=*a-ThzPPx9my!viM#(*8k610Aa%oT9FqwJKZJ?Un~+gy1Ckt%=<?b>65d@x*G`dU!SA*wKUUr?yxA$zTX>I;P(jpu7Zo~~pQ{!^J6ff*62XgKcq4*yV6%CZTeP9b`J$(A$I$BrD<R*1<U1TAK*pm`C})oZxhzt>{V!63tJFtwO@UV~w?)sE3*k4lz255jgVYY?Q?^ldW1rG~Ahm%5hO)LKVqn39dvOVUe`3O-1%ln6}WxOXwW5wkv`yP?OOvS4dKeiCY2OSYP5BJidFIqyAtR;K5j0lxBV&GL99Z^`Rq_<3(BWhk7?zY+<(<PBRbq07#rt|=bdIvLzzy>bj!cK@U8Qni*Cxn+69lQ|BUL88BQr+zaf{ZDKXS3@TLQqj>#+Qz=4l_?Qke84+buV1I}UAh-`wIWx9^b6IUS}>;#S*>LRKjN@y>ryGxSlm$VF+Yp8$5_W6Mh(`)P;r4qHTmXBe9Q9*UQw-59`!Z%FysM1czBqTaWa_e?eD7ihKKMSq#*H@Bl%=jnjtdhWev71XE9A%L3t}x<}nrJZ8%U@Z&Q?6PLKl%)BC#<B$%gZk4+|zEn%W>O^`}+%xWgoq@?hR7aO5p->C~jUV*U_0{kJ28PGEk!nwcevUh2HBJE>Qxk!jSjS>)o>bn+-TWwZcnGiWy5$ks6nPaV2ew&@TN}pP-&F%I3-ADs6D`gDjLhq0$((bhA>QSJ?0jw;}*iH|2K;HCl43$nJ8D6GqE`K1^Qe1k6vb^UyVD$H3cGZK#$BRZlAyCIr)o{nEIYz@6K@k5ce18H`BIOl@yJ%Cznd^APex(@!6Io>KTD@ol09@JvN#<UTiv~sOP>CXJ9%_{GuvB=clS*}__G)2v%AD;(W41wPzA;-1r_EVB)Mk-qnMd4Gc05wbbVRw01Jl6>8<9A5fOig_N?xZX=-l3bADvH3cx596bf#yLEbdxJ_sjM7#LukK4np}|*#6~LVeL-^R8`sjUvo)w<lKec4!n!=y)dvgtWXX)4$(1jZ^EY_BEP8mk)Ehncj5`AW=k`dnkW7o9yyVk0iRN-jKF*zNr#1#F)?ChHqQ8k(s9b8StL$>w{oZMPl@rz43GC%fLSzFP=^kO5u12){|E<_r0YBxB$KMwTe3(<xQ|$r<lE7DC7=KdY?BCrrAsSd*n~qi9jjbXQkglvA9uPjDn3NYW>Roe)~H@GG!xG|U0jj9K_7(ZxJSOD^Q$ZCm9_WFRH6yaU;Ydu+*;aPuP>Ev(9L+2vQ1+|^yW1R<ir+q^$z`3yN!p;{gwqqeXx(&X4<L6pQ7^PcQ445*ZYJ4G0zV9JxZMLT@nebDpuGIrVJA|B+cK5OFxJuFqkL2(C!R`$u;RdE$(0mG3>q@Ptq7TH#P<kdP6FYUGgz*9PX%y*2==1M>&h$E7xV-64W)xO}o<{bRJU4ifZj7pYR>rq6$;4MYtp8Tx>uJG}s`RQ<yKWZ74r0oK((+qs^8z@Eks^I_9DV-q8gQ9h$1zzqK5B(L9o~j#BeV<MN6Ufcd2zvX|G>+0I$KMT%0&L^}{9`A(E4XT=XS`9@n@F-N)YwKFAYpHXr|9)d9Yc)OMEKwPPe!!mI9rM$c5SC?-sRW{%*DltZP_yiW05iz+ur5w&tCcwosjxvq>N*VOw2wE1?L5EX(iSzQN2`ex{3$44Xsi!lZId9DvuI3Hzb%6kS#a^qky|dLH486`SGFR_f-0f}*TH2*y9d6Oc4R2Idmn&HD`MomA((-4yy`)(!d9b;%PzDx`jte&ck-`*QDX&{)Kg<kC_0CXJXwwN%NQDp`Q1XxaL9r1JTkU?gSx~t?;F#a*#*J3NKl>x5(tF~&LyM_W^Pp860v;MWg@IrAUo<C^!+&#}eEm1a>4M~#<QS`v&t5$#yi-at9Umt}$?@YC?4^&vdYJb*7*oeQ@WX>3vuHH%`~6nE-Ho>qBVrx`5pwnkLRh7NuK*e--z}$7KP3Mdn9Ij!pHSrFBR_k@;U2IH)D`vOz2Q!Oq_Kwri?w01wcCef+S@ry#ZOZCog?HL)RfN?65i-LpJiM+Ub6?2Z$hsQEP(i*_}c6Y;WpbCiLWx_7^Wdeyc~LUxI=Wd_xt<9I%odo-wk(jMS>?bVNgZ1CO|AC_M}XNX3jx51bdc_PhG&!XrCOW(EK7lH~RfSv(rON)ycqszagU<GVeWcx}``Mik;zbe+&5ngQ5~omNKdwN!JxhY$|r({OZ!?@=6ksyS%u*j@1q4T+#JGNd1`IKa6<zDrLZwzy+8K8CcbRm(LPlQ`?v=kJBPMpRhWN3^p?7joWwfVe~L9s2j^sGK6g%r6pgHsi@*%^{vb4wMbVK2|pwsw^47~r~TwkMFap=4HXj$XRqkL;mhs)ZkI?KSj*kJ&7-T~>@0|S%Hu3^`c>3KCG<viBL`E&5mPkAjTM{77({6bt69Wr;;~kGSZ&nspy7fBm_f3DKZQaG4aDBU{q3BL975IgnSSHV;+t>28S)>ThZ;I*Gy6I>5`VFk0hDSv`{tW7=rvd%Jz0=3WwA$G+@_{$5mTG;VWtOWYTKdwAZ{VLx2rrhSJ&$AFRq{;(s?#u)n1FpMnt9yT{B8pbJe_}>Ktr*VI<L7vfN>mQ58w+RcT%gFH37pB&&`n;KmxHYX1PTN?|q-g_43fwW4W`Un66Y`aO-7>Mv8Zlaki8G(!T4--fA#7$&2kai{+P6V3H7AGNe0M<8kU4(qT{`aRf5!~J%<b3jh40A3qVUvFZ)6&t1&vj4%Zm_p$|9q%Zi>~9ed8Dd3r&nK<Ufg$>&MV=s!d1>o}KoQf>Mi0<%X|@8GQ?0U{-_N&a*msjN&Xdyx>j2o5a`K%b)W8olPiURutzE=U(7ie^Cs-Pd3no6#J@lr)hZ<j0PlHrZBO+3&(uE}$5f)L>OMIcRxRT{@@)O(y=qC)zr873jVq9r~r)Jb%Yc<mgkvuEm%Ks9uxV+yTbrun@BIT`U6Gk6abFl^JqD`WW<=7=Q7s=}~Lu^80iV-oIw}cspQgS2X=b^?blniMe{u#yFn3B`2-on%7@(cxkO<;!!CX(4UR#!nVy)AY_`W!)5=lRvuwT1O_><KYSl1Pf|)N|><7ndvOBD;JGhE!w2$tA$C^>e&g-IWpZNhwoYwzv(3Jz5j8Zzi(edp%ra3&)rW3GSjW<Y*^)?egoh#hIBm-bjl<2$W{3wa->%K6EXiQ~(lBHQ(#D3i#IwfVOZEq;`ysxO=VEUPQV$XLkFY&d@O~Fm5q!mT5~kfhgdEx31<G3SqI33HJ7igSgp=d&A_#YKHJw)5>14(?1o_by){6VfmrnInL@O^g9}~T9H&_6nGBdZVL{?M&$3rq*;T*8Eab(M`5RV5aAm|0;fR#hW8@WhF%R#Tm@NS=;3|SLWC2+#LR;cEG{f8=e=MdD9d~(R5AM?Pge_Ec(~aNE)ofc3I<MAimc}su*!;F*k8EW41a-uJcN@U=T`01aC-2E%vIw$`B`*7`BKZ(PbR>QF*El7{;fj|!F62h{UPJI+9`FWW^3Ufx;pbJ%`~XaWQ^RaW8TzGuYPqVLQVL&W<%&9+%P~H+E+{_rWC1K!APwqo1>9VgEk$A2k-|SQFTR9{obx$O3;OQW-*A<3}}Km-P=X~RqRZgDD-)+*M9(fQFee~xnwz4-mZ!|C&{4)6Lczj>QZMQyxbNSB7f$!QfXo43jI^xZS8Kg2AKF<s)Rst#OmCQR;kp4BNuHKhLcgc%P~o*+`;rp14TLjyo4HAla09f^pKceRaTJsD=2kbTK(--Km0I`@>OZZB&j;PqbG+Z*5yH+#zJ`U;)iiYb!`fUYV~sy#v)xu=7|fh1Y<Y^o=0P{(C+PTQ6M0QQS6B+LD$V>ls+}c?F49%G{U+4RNh})sH~7BG4q0CO+>(_C)dTf@rd$UYS3#~!LSVLGrzXFzJ9yB^sZ)}<U>Hd>jc{Iz8m&2R`VC9mkfoemYTy}AXINS;ND!(46})BUI+5Q1Nksf2_wA(!8jv4QDR=SWhNDhLa3vHSkvI`L}`?lci=CGLnUeJ$mC#7O0^-J!ioO#5cA?l$(i^e$RS3bM~-UAguXn`B+nQ|G$<a_^dTiKJ2ZGg?E-+Xrlm3H4~PClgQ-t6V)D)9;s^344BpuVks@hmz!&Od6Xbb7xga>#4Pcxn$(U<_E|j*z{a(n1Qal9798~acT&>T{qD(?Hb+MkVJbO@JBtMC4IVK3XgG+T~H^iz_pP?}cYbknZI~mo5#rzLkp-K&gt|6mCKT8k_#_=|C@KH#vj=>uupqBOkf)AzkIF`e{7`FOb{5nPh{+I9}lAMMv?dU<I@{Nxh{Y5+>3=a%t<OV88yd89>B;}5TEfS9;L55=(rtOF1qu^<%Bm5~5e7z4t4+=rrHwmkHqH*;TPAzi!HGKXf7956OpLnM~qU{wepJbQ)fo!0Z@K7|spE1KKrIGy{Ng1mjV-g`6H;U5RMR*US>)=Y#$K-5}Ib9p(uKg4%pLthjJhA`SdkXey+GEN;I`MPerPQe;Een}WQf>4ADVS{SNbt0iU<y7FAtJqX<Lya2ov=Sm_SJhE@ALFl#HK;16>>X3ox}BsUU0iA!Rg{oT^iI}L;RJTr3BrJg;Ogz=dG-I*EiXLAN75UD^4erEIGw+$8hh^o>_z~4jifYN;DCN@>9G1Vm=SPu%=IRiJqi!e30E(pl26$eOIg9=HFVYU|E|Llqomx!wvEz$wfxC`w1yO%@-$5L-m&w1LZqwmFtx?6!PKIzvV<7j*L*Qxl?{A*$o8*;fym9N*PJ4E?sqHlyW4d_-i0Q?uF}%OO<*mx3n|pbTX(&l>t0pWgP>gw2zuv1MT(x1i&Wm>JFrI1;STut*zo!%>>x(D`h%vxtauXF6S%fds+i+U=Y?87RxtQR@XNc=hus^gAt;}g*8W*x(ney_uju;1ax6Ntrp@Lqh(ojGEJHk4_@!}cRLLa-A*bWY!tl}4E69D-MF*sjdogaSb}t^cd3h+)0jX4gKE78@pfzYDqK^#{X;kxM@4TPHMk>>6*poyyw$Z1;)aCcRpKamH3(73T7bjC584J&PXHHX=Yw>{h=rM}NtzKR_4TNrmuL66V&Jgf==5jJub0fPmyNG+7rq9TtwZXj;N&O_$C-wQ=Qj=OgHQ<c0Y=(VJWWc;AnuS#X;E3B+yM${3M~SAzJ%9N-4HiuX5I~zd&4exex5>yQqK~TS57<_4N<pRaIIgEqsgY4a+IPSp%@d&pQ&a$vUQH)tI#>7QfehXaHInI?!^N?OlRt{vNoP1?qdjf-pnwx_U;AzKUq@@3(skQ>2>DB9t%BW8q|=qW6%OPcjZO}S{x3-jI){An~)oQQ;8Y7RSE(SNV=^FhoBN``pZ;d%Wg%uHJZUMxg+SSXuPOD$mk%q?q3#CuT?f~S1K#sJ6?HZ!Fwm8YkUR^)H>J`UQWXTC+%wV<8EsROaS=8C<nQuTMXX`m>$kk&EP<lrh~tk6$nu%WF3#@uocQ6-71FP)~pJJsz4oihe(F3Qvdi9R{>~bK<;MtnTbh~l%YvLcRFWiLl@WPvNs=wkmV}kUKF~ORZPWb_^{`I0rVOjm`;aIe@N|E5gFWk*8&_w2UX~WUx7j4qBew6in4)z372E^lBTleU!46r`03f%Cw@qDOJQW=JS6|pc;m!+g5aofX)f*#s?u`A$tVwm-xPadY6o{P2Tc#rfzMPV9Tb6Oj@Kd~Qn6{0TAh)jdZmY%zJQ4v-!M8d<ca%1Qd<D+PKFmBd+$*qM85u=WI>i7fn9~fVE;W9QLNnx-2T={CMPGBvCAl;(^Ja>zq_}C60JP6$_Nrx=_Ix1A^cy_Y%uYP6I5!^^&9{*ZeN4~ouhnA;&~T@lq}{z^NSNg1GSL~C^JVWg?2`{(au7pSC1S;Hz<(;E{X<5rbU7~nlMB6&%M=N_Ymzks42VO5=0$$y$|=Xb_6OfN^mIvj*psEIRDWfBlPwV%RNeQl+|&}F-LUoKuJ!1hgYW?@m_}lej5GV)*J|P$+(n{e^|`O4-b-bgqdUugo!|<b0(DNf_hf2dNI!^6|Mptl+!E!aEvh=-sx9opPv2p>{IdWIX;1ZzdU_@EVPgC#%H98f8t@AV`xu>vZ&t6AH35qsFDP#DKSCwC&#DH;LqQm{p##jLaRx2VUC3k^*7?{-yFk8&riRro_&7$wC0C6<sjewq-Hf{u`#UKV6P#uy1C(A+>3^apy5f7B$q(H&MIf4P5d>5Ru8r4IT>S>v=0;w(i8cG<5*SIPmR4b$7OsF9pn~ZJQr_ceK>`4!zVMiZRPsX;`|2c#suRdd!Zg91MLJvxjaw=eBgxBSXUTy#8Dq&=tat6k)@>D0*W8BM4ZFqs@hCsbAS2)lz3{AduIA-iVmCn>pG{@5Pf8)8+q%}7+P&dZQy!qby0NBp;}PE=~<X}1imoNvu#w)fyUTnX4le&nH=4+oh(lyGD;@qt8{vbT5UT{LbrM2*p{4xYf2l5Vw`LT5aC^%yv4Qr`Qyo2Go`VqX54!zzw`d6C6tJowPsd3I;KbcX<hM^D9%s=*Mqt4Q#KQS;<Y!-dJltfX<zbc-DDH7s3YUpQ(CSOzm^rW{Gu`)HLEDOT<VrG*1%V@6s8X8WO9Iu88`4|sK+U56&zuC6^2f-{bye)axztKIciLY3fbKzuM-3s%>`rnp}E<K(cpfWYM9*1jq;|B?zGgDiIUyIcH7TkrYEfRohD6(+)|`CQTs?SlwpW|R<L<NavZkyViXBI3WHV~(8HAdE`_($$1lJcYoZBdfRtstC>%Y4uxjL)QON_HNip9m2XpcuPqmSphe+9BRLJPsRQ(5YDo459&?@y7xXOBPMzVET+2c=1)v>1@h=-5x5|44s@Qbpbe@6CiG_wC*NA1GuG9B<pC*>QJRFd;_I+LcEcL4FOB_)cYW|A7onH1(ENeHKqz)-BS*K}BSJ%d+kfg??v)K^*39m4~M_8@^RNgn9}EGzu@5)}}MB1}TZ`(!qif*H7Clx#7w<ns8ygJefmdhX<|HFWL1bO{rLVAfX|#9jHeO9?1Eh}jEs3b}>+sqK*4hhl4IEEjp!YTI`%L9UViVHlkF!37+y2LGJkz(@G)LZj8~QccbaC1zKcPnZF_&}rh2muF_$TUTeVh#zH=Sa@_K9GJ~!_~Qw;<#Pg_R)ziQ=ib$#clv))T>tk#^dHG+f4nkz<O}ix)#H0g#fqPy0*o9`o?{vTx{JY8>@(abXTSF7PWO~ONvIUCDEjw=e)`<SqfY+>WB*!sw0wT{577C?Sg`Xc6-lO>AXN;6zHqG1WMAl(LIrugLhvx70YSjtF97csaLIdg_G_5jzYw$t%wxU3A4Q4KACLneEBF|meuH@i6cF4M^!*Io!{`MMU0|3s5T?FU{x{!>qK2~0A-n3FJ`JH~kGlQDz%RceQW-oaV4mO)&WkI)g{#5H<as=Tx&0UT#xr|D{UC}y_h`uk_IV`o8v29hP6feB_a(0ZFWvW4LG>8|3~xM<i-IK@pK<rk2_H`c`v0g8Q=)}IEh+trPD)RC1s}nn=xzs;^gNl)Qvqi}c~5B!4{qgvVfa@8@T2V?#pFQ6w8>%VBSfS^sF)zH4zBNyVpSdyOyA2j0!;olI0ilzAPIoZKA{1p#Z14Zc>@M%(Fhdm{ToIQ74n8f<Bi01Xa6C@g=egM3NQQs82JbK$wS=+uA%Vig0DXi`8`5q%s8%q_YtxATL-Pi{)h#z{oV+k_V&A7om*U)iR1RAOWE84B(``<ef2J&BDKCh=-|scqtV{5^y;hKek1Pg^oOI;8#8aryec*?JV#eb9QE=Q`sF0lIgmNDccH&O+S?y3(zDm&tIcNn^-I%RS6_Q``pRpKH>R6cuDm|oZd`c-UV5#0wY8;lE`E22tgDv6%e<M_y#MzbGV;&j2h8S#)*(3cRw_&0pFjF1Pdem*$g~zxlBf01LuW)XKv8;ft-=Jo7a=~z`e<4q7$BQ*U1h^Z6x9()f(gp@#i@dfA9p)&U9~^gQSO5uwmO5+4mz$jLH?%?Nt3<SppLbfu^A9UGaek)_j`@j0Ie7_xVY@z?~pbT^&z%`o7|Q;FLNr<EM+3sa<tN&&5|VA5jkH8EdZuWgKHzjvJx__OUX-dDP<87Q-Z@)W#vwNW0n8<Q*CQFUQha0hQ-6E7#Tm-b(x&FyRG&}kCyNES~6JPJ+WloQPjh01$?Z}m&q}XqhfECS8kB6lntx45X##XeOy#F!I1SnT+^x-#_bU-KYd9?4SW6GZf7{e{M2}jq+cC=gHjU044Z&H_+z-x=y=NxadxeduCPQMJtte7^WH0?_pg;SUz@V?I&*Vzosyf&_^xXU%d7H_<_u*xe(!XKl!RJ(Xu9IP7%RTr6vT5fBFuTBf%0V~>u6&q?%^KFYJ}TAtC8BC{Jaj!p@)7Wl8kV)WacjvhPJblA9|K_pBr+P;(1Plgl-gsV<sKAz&3N<?CcbEm~!*p=!Nj+sHZe}TGle?-flC2r3-Kiy*u7b89TF}aJ`#r3&rd>xIixx{W&SQX}A(X-#}pDUpMAOku^4<Rx^@!)v{2Sk@~{HisNBj4@s^Pd^L8&Wh6Uxvbm+?$U0*~_EboyKi$U^4J0Fbc6R;}=Z2NxC<Vm%_FSzb7!E9){GiqjjdHCRYa$4L2w7l}pX1{o17l+u*hH<eA)!Me2aw6VWc~+d?J8v_=^7*(R-uURl}Q&VDwO$UJQ=-q-08wOmiR|&4m&~WP*dPg_(MFZ4?4qp0LL$x=xxM!$2^Z>-ZWGr7uM1bHEd;r*Sg<nVmx+T<@@OE6<NhnsAd2gW<znD4MRA7g0--U-ZXMO)++O>YZwE{PSaSWro3FipwW#xw<<!0gU4#!jGVqTRWAhlR9)~}((QWndL90MeWQGB5#>UiEojt`(l>Hm2*1<_g+-`)s3%GtQX;i&8JD71084!*V>nGNC5IW5bxsCVOR45jFPQ8AE;|{@8YTW_PZ1$A-f6D5aHQr}biyU4{OVC69G}d2D{$dk^Z0OYdZRvSCoy^>Gp8iNSXU#U{Ecpj^#^bu^&b>Qp~@>@d!9Uu)T8lt$&5`~JE!pN%+Xi%8~eMs%#=sSnF<XWkTRhVOwG>8H@0F<JLpEQCa<VIwfsZWCoO@M<Huz;PDkqcX1tZ0K*=YsX!;x~Amy+4`2+-ltqI3~m$qA@^cZmSN{xWOasTPeJtH9?#P&KHbI>7A;H}jhLWTBThCHEXZF4|`i9r~qEej@d&dE&W{9N$p@0=eV?P84SJR-WRqETFJdAi7_@5pvXc&){mRjV1ebk9drz9r>Be9#$2vuT8B$I>2fn|?FVqQDj?;K*>wA^b%##58_ZU<8=$4}TFkkORFnfT6X9ZvndH%ml%Ui$fx$0(-TD*mGdU$@11PKnP8xl@FqUkJ{zCcZU}X@03gzE!L`kdMtlEp1pf_>ljs0{1;!sP|~xHj}uR?uQ|l@^t)qH>VGG^xt<-9J>k*W?~av*nKbWbpQENWI6i%hAtTZ!ta|$7?2k3<2!AElFwsD{n4!{EJv{p~F&G?+2~Fu{3*VT1MNJ+i{WrGzSeBC+cJ&o`2Y+lhn88@bWV1k3`ERA_TJ(zM7r?@Ir%wXkJao`)q`uehb{dD413Y;rhQx3~%0+T%=F*IP^pGD4Wf(DE9usOahc!4O<+d`e@m7Dd6VXuQCkXFa`!zBh0Ev(=x9k;QqVW0j%XjYHExv;vtJ8n}=pX<5(La?!uXfQ7p)2UW)ow=x_*mV=ZBz(Kl&=>j{Wht#cKO^X%*qm_*tWFG`wVX3ltgNm>!9j@pnc;DBtrVYqp1$8eYU^laAh<*0D|{JfHeS+hm9BzKihhZ3Yu!~z*2vA`h%P<zj&<HpPbAwx%QP42^4lHtf{Tbc5l41tls!X*fXDDPgh+%71L=v2dyCps)d-Gcfws1T4#$OzxGgeT>%-l9q)F!hmpUzGQR{OnMjUe*mPY!w}L~F0+TzBv@tq+-A-!|;YS>+zl%wV`hy5Qhe;$iQegyNiRSA95dI?p;6Le1h+JPS!m$1n0bn3IZse!LfxahG(?`b>r-eAqe$2E~@>Ikd;nsh*Qn{rfO_wX>_06>khWaMRO~usOXCHd`rYW(IusmGzOEZ|<9ZgD!Tdd%DYIbBD#Kv!)h=xDHuaC|?0XA#BC|Zf<D);+W@`>ISF!3z#Lu3h#ti<G_Ov|vpRdOK*OL3Ih!vbgvH<3b(__=-?|G^SeSdOof%CKbY5bY3Skeq{-m)0uf1-!ehErZ<gw?x;**xV1?`fyxWFzcS?8n-fb*flE6q76}T^$Cb_#24~L@)4Z1$~xv);+m{P?tHCF7X_>d{|Sj{KXbR5%z$y>FU-tp)~fmt?|38%QXC^gWP>CfHjsh%%kT%WjFIs`;+xlUaX9d8;efuPSfL-<^aVKJF^CpuYJH|YpgiylB?9J-xYxhGL_c9h3z8x}IhNaAut%iLEF8Luo#`BAHqBu)XLr+U{aAMx@HyTf7{4Q|z2FDqR*;qhr7^W8=nMtli#h8V`tu->Zy>#wVP|9%qDVn$9t`=%1=rekU$(_PK^J}e(|;wIkGw+Aee0i3|MTgePXDWy3L5Yd;qvTrn8~G?nK!)CFN<UXPEJO6OP=27dO?mF(Dcousd#$!2i`j&iC0yh>6|K#X7Uc7DhSP&DgVR5>L$h+1L}$HH&zXfHOXARR}LOxhF_fh15Eg9I)@&)=U5aID9&Zh?DVNQv*b9&+0B}>V0{=CO2>V>R9PQ8i9boo2Xue#rH>x=R<TYR{!tt>G&gfPyjlqNY!=)=+a~D7xv_C`b#uMEvhde~KND`7l8lXR%zz*#ZK2Lv&q+o4Mj<C3;;h0M%DQQRSf7SGp%hI}Bk%=#*FhHNWM%Ev+Tyywr&!fTZKdbpx_C<|`}~!@*X-|abx{%rEQN68KWH`CPj-O654}czFldo$EHI*e7xk7-ZsN!R5jUG=?uQ$*Yo2goh8tJ2H83Kvl<O_toQ8XF`zh+3R<tY*=*RD2RrR6Q?{yE!ZN5L)j(eSl#AM^H7jJJvTlz-PT6>uG2)<yd2vkFw32llgwYd?s$!bqM@-Y>@Y5H-%OeoiY+X>&UlH7kVtc$RhEH{p{Uj+`qsv{#7XHo;%WC|qdS)qX#`l(5+t=_ILZ!T>t-jaEo^e(A{sd8_p8q5Q&&{@6{tgBjcR@YSuUa3VH-8&t{YdVRorGo`A7YQdtghdAAJ#00-twWETwKVte<a{@1b`vuU0L6={_1+G-nNYyxXs3n3+&;W43g2ieN0)A-Y78jET>)EB?-w}`gEnaG_U}^(+To}bH@$w_LzPyiw+)RbH?T*0Z|GrS3vYYBH5@|o1aek&Yfxj9k|@oWa!-<F&77#T-IRnEjaAofjxL$}BS)6b;f4~wSV2CP&762{rzLC=he1i_#EG(oW0<%JM=u{T{Ur7(%?}_WbXi7$L~*xU*wd2aOTTK`=3H-=p;Es6fvWF&w3nPeIWL^R4x6=yCpn+X%wnzL^eEcdX1@D#uWZGiVpxyYqdYsi@eosM$ITX+bYqPYIC_-=Vv8BLpbN45MzdrHxR#jXc+HmGlB=2H{wjxX<$ETLPsUmzS3Wu{?U~XFlR#+V9+0|AsL!%!%kr9*3yb+M<-Kx|;IsH?I$ak2m%+RmiOta2f&xf?7D))qeznc(3iFdSVYP@nQH)UiP)jln!d6N{EorVKF0CM?G&)22#V^m#S8i>j;(mO6NySyQ^1=Ml=7N?Nh@dVnUt7GfiMM{&$m1god%|+38)Gxdr0NT=vph3MCV8MWce`%F5vhA_VnC21YnII<!)@kStU-o8G*%hXR}Od)UqmcEea`2VY_xUI?syNZxqL>O1nmRxDK1Onj!g^!T}5(`s?3l-6q){Nb6)>mbfhueJl+{aN6e8=UIN?YdQ6c|j=mu9qQ}ID=oj8L?h-EjOX41!^@gP)kna)GXx2gG=DpH8ELc43XVGl#iNhWFw!WGCFr)Sq&nv}e`}YWE{M5mssDC#Dbg(|ThA)E$+xTNWk%9Ekh0$6`08RhC>Z1QdR!dEx_+Y06lCRxPm=Ci~3W4StkLf^i$>!wSz~{wVl;<Rh+Ps1XNvvAxm73#Jadn*62aN`Yhuzmfyo-fM)PCpsA};J<!t9LG7puzzp@T9Kc0i>#W@;M05D{CFCdh^~@kolNlcV~Hqq>pd*bS02=7vrDyE&t%{I0DCYz$ftOgX~Fzi3+|%Vs96YNwlAn9XxZjw))a)lwH?ok7|m-M#C&S%#h)*Z)u)ekA!v#C%P=jYxdUubPkGZL95;@GZeK!g{@7-naJKxqH`HTshhKV}_cXMUe6JL1);i58?-W%-357nEo*y=>1Ak4h$m}VMY6aT*g>#iKN(~3d*zO0y6BntOYH}L5VH4w1V;csBVoP!1b%uZF+m?mXFR@bCe)P9wqu^5H%MbF&w1>AuWrj@M_4#lvVUP$j>Peg{WL}DK}}{msxX@?<`NK@L5M*K}D~jq|wk%(P$FtLL_Nqgx2UW{ku*O6uaXj6_Ly^W5GpMBDt8F5E@cy1=9&y^TEqOgN-K|Knft~B$o}DPSOzg#5$CWUqx}$bzm=WSaWeIk=s$%lZZfF{Y_h3?DqTjfEDS~P)wpf?6n6$&H%%jVl6U27eCYa&Eph}_Nm?s1dZ)Wu8=n2<<*7CQawN5oA|cHfSX~jZh*$AcnSprFT!559lT=xE;F?^gIA2-jUMH}B5w=9%Z(9tS`DqWCRLre)OaFhO>+EP6s_0N;<+j>Ip>Xb0HHhm?gWA6>G*9k=A5Pwyd9}79U<o&%<W3|(DpZnsQ6M*=@fq@QZ<kY%zLj~@?^|64|X!=DWV`p)RYK>LE+k6j#O}zyVXgKjC1!R%_14R;%;ISDoc{dN(8!AZ-L5BZtr)7J0A1du9`cgqhMTN2I2j|t3q9~H|QS#TfrE3;z)3l=z2p;=;`@Ma%A67@JTQo?#@P11rv6WpB*!#G~tlK>?v)`;e&~r<+HbEjElQ39{LNF_4&2MTbL7o@9LCofYnA@`&e#oIHU;AQQYa$nRLH391-O5ZnIW-Z*virP?<BzOG|e|I>rTfFu#!$*)*%C^9Jq7%6Hl8oU_a*a?(O4YnPYc=ckU5!hJ2!X$5nb)Ej6o5E?nsBfsJu`IHqK58Y(x5t~RVhAXW~(xM<)zO_`@sH{}TFl20F?p|iE!Q@z&xJL3g5Zoxqwgkzs#|3<2bGo#Z3%knXAjXDm-Rz(pDwXy-*#{tSf&6=<0D*USAK<po@S$PhW~IEeaWi{67|<npw1LTBVYw(H)u3p(4yxTWD7q49`8)j~<_+xDWm0p$lyW&jlcxx=Coc^B>u1LtX-|$(&&e$M5qV6WetY`tn8WAMQHcH}m!ad+N2lMPesT7xFr7X}uLyKNv|-ns5$!}-wjz%9V0l9?N6QM3rZ{>md<7aMa8_y52%AIL1wZ(vJt&$Yj;+)q&lanVQvLI&Gbiap<76J1?4-rBFiWEFcUp0GwButD@}#z>v2nSD5LayKE2iiHMoKl(kq=|KkRRyyz#VLNB357~wG+zkmlu~XVLLMx1~<hvMB~Mk>ucrpjkV2rI{&HkhQ>PVE**>^2e#bYPi_CEK-H`9?iOlk`h!C)BaAX7Mn^n_lhE55^y8+-U$r!6EG49X0OlcF%e}@<t8s7WZMC}n2fjO)hOdKqAdSprX%dQ%fr-&)TqT7r!$MZrzI&l^(CF?r(e5H^(X$<pl{POhkWhr*z2MGlMLC`7m8HcSiyShA#Sbi%QvKSU`pxL*#FTM&!p8on-)__Oh`X8J+*rMSotn!#y(RBDI<D^2jK1Z5-43~L^OcN#Sy#q2V+^IeQ4EcuaN8LfIhmobF|II3-LW+rGu_6C3#D&cQ9B6R8EN-e*dMF+>b44u+29pB&p<dXSShP)ZrD2vd%IC0ud|atPQpXcU~GYO6?zF$okU)awV%o+Y-3_Hfj|v)jPW2ueO$&UYRh9R2i;s)yre>7XgL#miWnOkvRO2s%AF{}?BENeO6mfu7*Zpy<)B~{JYgtx6xEs^WF>m--(w~!U*=3%niQ4PgkQb3UZ=#SLYV3uI-5~Oe6!)Mvtg>nd}b!c{=Jgi98y)9MWbnkU?^7wYN8#bFHQ8Umg1GahB}2YSref}{8?%|sr1-z2arz>eIxQS%q>F}wNv|Gg4$z#$c4j%zl#7Zf{nr%L9O5E;&T{;c8x4ov)v$6tf+r4<gx&f&pB<^H*ZnhAT>Ed^9Yjm=Jo5as+AQCXT{i9SO@i%@1jCCOGP7rgd&&S8^Un<xu$d!Yp<!*5<OZ<A(^r4i9*p_@v*h$@V1m*FJH#%x_-}FDI8Ch7*mxS+_77{1nVLe-(<j(CM(K|O&yn6=)7^AC+UFIy+Nlxa8AGhm^+73V)dL5BraK=N&?-Dw`Z7{Wo%C0vbe(w$gu|)=Ul+IL;PpzN-8>QcQ2xOrlHejRBBU%w(hFkJ%SET+;f%9lSiVpZOs%!Qq`wji^Q)*ZF#2Q^*PyGHDG5ym_j;CRPS&CCqFZCQGZ)^g_ub^^>kMh0S;-t^-4*M1i11{Rn0Kk)Gpxwcs%MKjCSfAG+*Cqi~t#c(2E7xEMpmFPW;-d<JwGJf>zUdyy^JCC8FI8R3Kipw+C0`b~w#{Ya#sY#k~L|;sj;#K+GCVS3Gt>HjEp1TZu^Ziic<s)qKdyD0{3W-WsW6O}l~CneAH|KbUNH*QT^i+*LV#?qM}cPbIS8%1DiPfOZWel?U+vq@0nr+lOXaPGiivutcgQ7GXlWU-C_i6AxevU+T7++vFkK<j3eRsZy6}^+=HoSpI6saF!3FA?{}*jifiJi877g81gt#SzkmI2`u=A1Kq;aKn{}_9As=iyr)O^x6!S$;2dbXj=^f!-X3OJLgDbE8=)Vacp25wEJkK88k~XNv6_|$@^;*&08GG|n%8-!!w8PJ8-b)>60nZO`a)+FGd1-{Hoj$gwhUb^W_Bm~Zq|vbawmccWFN1nP8lg<JKc>B5bH9*{CLd6(Q6gpEfGr$a_HVnB&|JhdwUzGv}TQE&}V`odctBhKx_%E;J!i1r&Qb$upC_NpCAgH``nyyzH1$#0MuXS5GB`~?8Qjr=0qs&>1knXq6;mGNt1&vpDvZJRhHKO$}*jq=0X>%{PsT1u|94f)zI-wr;Yi|wKcc{y}<llKw`G0q6r-FsLl-y;M{~KKz^abBYGmUIj@ZbG)(!PlIDWg$W0jeZ>esHEd(o=WUjnXa&N0eJM`U1JQ9ucT{g92_a`%+FulF(EG7VriyC3nOfpGzpe>h3u*4!Ch;DqajEb%mWh3n?Y9!kcFeG`^@%UADe;|-bX)y&Nh>{psqFsh<#M8oL-GJ@J4qlint7ykgU6If6)m#pU2C*-gk@#V_6I&t_QK39=$0cmy7$Q4wii;~`GXSGl!bW-h-8u(E0bgzOd%y?_3{%+I?%`FQvlmiUmsI#3spVA{ISFwMKA^UW%8Gk^b*+N+I5Ev?eSURmX>r|(d=di|w9&E_@-3~6Hn2K$Rmu1F%WI1m^}bXoziW99u^h3wP$75Y_2t!fD}G6vYLtAe>L4jSE~S+ehF50;Rr03FLJN(N(l*u{RsHnUTf0%&kY*uiSTOk+6Rcut{7RaqDJN1AlU%_fg9hc6CY?HcaJ!P$oUYHECc2(Q1?Z)yXi^D%_-i`3v~U&1dooV~R>DEgtklCU>{s3Pi3u3Iswuaptuswr`zojyXT1}=qCL=<btX#4sa8^LG}^23OG*x&^R{R3XCs6qxg3kY74y(DU3i`8qU3+uv)Fk$CMKg39M094QJ?tZaX~|77$dyf!-otA9HeWgz{QI5pHM?-#BSB0)RE)YodA=U)SEN2NlIxTh~!uCeu8m)OOd6#W5A<UmwUkHs<nK+JyXZ5#D{EiR*zrVo>A8nz64VrbwIM+P}Q?j`fAK*u)KI<jWu(dD>CDvKbGuev5;gMy0uefZbXyF#?8u_Q@ibAAclPHXpjm=6&+(Vs>T)D?MzZ?xAO=`+eu776WA$F46U04NQ34j?jx;NztP{_#XJ=I6cP<*MKXGdZRdI{hG<)JxW=Yj?({<_J-+u!<Qa<sZ@3{(j!Z$aq+CGnQzX#)dWyuPXsyw85;2pE(C9N4&+|O(C9bBpPup$Yrh9P@HlON%qb71m31sHDCp<GGNzD{PB{f-n*?5t!C;6bT0)t3OBl|O5Cnh1qn!__}t-i9lq7|njtE0*sphBfpIV~C3-x+$ulUhx#00F=lb+qQZhpmAYfA5ix;4s6p0?Uv2VIgs<z?CMq;#M!rP>=?}+cS#Ov}1Cp`LLVee!|kJYCf5+w5CgqrpE>r{q}*U`a~O%o(whATvI2S2E1Wz7UAlVFdOr+^ytuW`Qd&C4w1b9zUZI+i!3&D_J@)mYE=w;s-S|IbDoiw4;D~!k`Zl_f#HQ?gl(_e8i5?c!<RVddK`FxnR<_0CR$DOB;V>#K>@gLfYdysIF2CAUig714K0%79QklPpFz#ZzDi#!elWhuFFXePRJ7eo-~tti5i(O${^UH8#9O##hnQxlLrzC9LLv{|WEOln{}LanW7(XP;2g4zU>L4!$F~<(7FKVgdP-ZeBu9K=)HQGNR&sc%sN0JQ#+IS_x)_2ZFcuYZ5`IGS{axdTch#69@VLTU+dus;lnmxO%Fx3}9=|PlX69k5>`zj>ipg9w*Jzse%?s5Fn|brK*Be*goEcLuOf}1|7xr4wJN;Lf)|Xtmic)C(hAMKM{k^M((&?j;clwy}%ZPGRTz?B@$@yrW3S`LMpaB<+(o*dFsLF^v$MR9Ka?{x#yu#VvQ$432y~Rc6BAxyxQLKxcQZbDZW(5<)q(my8uekP<NE!79tzy>l%%4B{$K|V3Jt`?vCJR!99%a(Gag~ZX!8AVQ-jjk=k3D?ZQAdnkct-iZv}#~RwJ2Gg$|#1$?Y^W6Tq13mESUAkyLgeh{2H_UV9BfBQ0_D?81}8GIfa8f20(v}Pbgo}6V50l*Ny5@UA!o&i@6GyJ)+uZa;*rDAF0eo@WT@>Hugi3S?Dp5$?r_e=(W$($~+epzO4=P1Pi%|B&pvcw9ik!fc_ZtL7$R}OpiLdExEI(LYQ7q&ZrfKJ$!}p;hJWg&gm=4<0Q6X1}%GkM<M-}W-eX&J0xB8_43~VF+3r9qQ(7A)=T@ERvN*i3S?qhSTRlfpi?_7Te^}MQ7=DaJse=;pUo0$SXvAaLWS18O!7U6vTF>#h7I!6h994JqU_rbG|Q*5UY>h1;?517yP4|eAwB+<d!t`5s;0j`l7;cKN`07-47&MTQa?3!6mY{ral(yUf&Q54sC~lK_&#Dz#17_I#tmOJXn>0Q>Fky_l}2#iX7U%T-iI`zC>dXS$^c`{yvL$s;FHsDRiV3Yi}@L-E?&HUsd%|`@ghx4l)d9^E%Ala1#AbYim;vR5kVkU-9!8dH^!He8Ef<+EjQDw@(jJdlfrjhNSkcO`#bLSI~qpYrq4wcJj__jbUf$YxWrPKFEJdyO7UA`Q6yS%uYUueKGsGhO5jS4-|N^YzTDDy@#}B28gE`{|92>jXAW(a#+!Zf%{SxMFTFPXMx!x1ePyN@Pq!|=Haor5eEqep*Dk%@x^lI7?$UU#V?{i;kYQqy^$qy5;FYf}k&<LR^yXKWmseM&#a+%@-`HGOtibOg7454Vm1THPUV<N%7B|*yg~jUHLS-!~WL?AcvLtJZ%P8J$RMt0`@FJnp#V9wF8HIv`8n6(U?Oh@><q{L7e3k1eO908DZtdc1zIO2ym+!Z{=8{@Lg)k)fn^N=AH7Hjxx!vjtrA}lgMH$3kU9ZTY7cs_{KGd&UR`m3YtA{ns&sH`_1|*Qy0F5$=-3@Q*ikt4q4x(_*YqILt!j3b&jki=!-5?#)17R6ub8I#m#JwRl#A-hjZlIHv0wt;*`SAcF`<7><*XPhLRcj};xxGOPbEw!l{a=KnNNSNM4UARL=sKKr-1G1{o=#u^Jr%aKMAOvDg(O!T0K#!E@frL@_*zinl_;+**O&b%428M}ms=CH{xTQ%!@*Gin(p+Zi>SzA>rZVvbjFS%_X0WmX9^}Xry$p=+%skwoL>Y9e)%<_rPJ-sD#)fz7+a63+eyT-hG?gH4)K^1%Lw@X8RHu|bfa2ZYQfix<rk_J^<wb$Y{~Xqp|=tgpsS6qULb#3cKxb5aS<ZU(3M8M;mRKccgay7&Ko3De+><GD>v#`nH0JQcD=SprHDi-srj|l^>wN(<YR^P4jO6ugG0Y$%_O>dVtao4<f*|K^V8;&0-CPsCq}gKx^T7&F><0kjp{qNj3WzZ?E(Wp)!qSK)iGw0y!*%lZ}gzAVv|R$;Yj3r%CSRU-&|VqRcfXj8b}O(l<R}CB)U$j>@Xi<y_maH=Ap_kkvW&7?Gr~lVh~R1N7w^8|Kc=-Z-~rP6=cW$`f#B1A7BwC>Ic_Gn8{uS4Zn<~SA4-{8WRz2dGiB~^iNamb~+?lOeyoU2<3zF!UA`z^>*Uu@25;W#rqOjApa<t7%}kn`->|Z<r|dZFpVLSY<?P&D4fac@MYr(c>^>-WnJRztE_n*-*Py84WP6$@|vA?yEPC=3Xn?xT{l~eZagqv#3X0f>tx&le&_Y6d~j=N6}=)c>YNmpiS3bjLMp8ujLEKYQt~yO0MeGpBu9qBQ&)LCSh?aPk|H%+b(;MZl*H%PxQGg}KGjM_((BDwq_cgMQQNJSf><beilf~CI8mq`1;NiE2d8v<P?cP6v@U{9u&nH|OX%*haeGxjt?@>nvn<%0Y|OzXozS{ZM|Y3;yQmiC?+Rz1j-3+vWF3JK5`t6+5-N2rIY{EaKM4oI=)}pH%&f?osq$;OSq&x&P0#9l%tj7n@*M2>#MSo;N?7TRa`A4?o{+8CDQ9hr<KDB6I@p4jEQDh?z;<LebOu6Tc>XXNk3wW-N&LIkh{igpIePV`B2RUEdlB>KZ&(MTmfw&15jp+fyi7!XdQh8$qR{VL2QfuS)RmM?2*#JU@RNut!_j_|?$ZK(htvqLQA9DetTLo7HU|%s%QZR;IKpl;H;XCMKcv$KX*kC2qz6JuZ(<DElYu6}Rl%K@D64DzWIp|u(|<kvU+I7Uc={ht|8)9)z~_H|`hUUa|A%ZFthWD@inyRT=JS$w^Kx<a&ma9G{CNWv-`{YR1#F^9U9ffN6S9|}9`aF$`s*jyRv3;*UrYLG56>9w46on3a|^w7*74S-A)w&nz_{MpU1xdL`}1%9gSU*jE3`IbS%+BT`j~Q;JVJ}g*Hk|0>Dlk3edZ|}j8GN(NLn0}W?$5<`-F-sJVR>_eEPj_oJ9m5D$*9QfI@kGgZbQzS?~Y*-#pT=o&GmM#}8~2QJPjX0FW7qYv!O`4&E>zWGjuR`0&UvF3z8rIeu84A*2w;CB*y_TDpWn_^D<xqi%mAtaj3V_A=0LMS^q10_V!P!AV!^;`Ky}&#y?c&n7KEodXb*Zr3pl<Hq9qlv&YVD&MG=R~E2@?+5t(^nZsAzGQ=gD6m4>@87U3<Ecj_13x0%e`i|N*p9+Fax^_XbpY@h48hqz)Q9@}>-EaQrl@QX`ed|2tIA_4RPvNeNS`ChzY;dAr?OljA^wSGOC^)eR{}dfFgY=19^nF!!rzUV*OT(c=w(ZBICxIIkLgE}fO!A(zY4Yfw`53poDBVpcE2zx3GQ(EecpnxnL$)-OoQ|WLb3R_74fIOzOhE`I*cJ<OQi{YrPC{yb9_SceMFWIPcPGS_UV{{aK5f@lFVmuz0P{~wL7_4oIHQ$aIX)SlwoH`3D9@2ZYLZZ^o2gru{0XQ!;xQ-pD@K_yVL7%AV~7rIty7K7HKI<1s`OT>x*m4WiC2?r!J&);lztHy|FsK%HBvjaQz{UNq2!fg`6kuL}R<PWpAIoP3k<W2F%aDRbE^J!Ijb{^QD)=0z`TN(K7`J*q!YSI=kr5AxmU~EK$)idN$A-1fK2mn}~{=t4qs{d3ye}oAv7pYiU^F5{W_d!%+*x5fP)0`1MQZW)sTn;4p`!^2!Pp{04D+af33rrhpY$S4Fg)OzvvSo-*DagiMx$hMua1Ey{F=OkSKcyV1d<u6r>Br3w{0UDT8kjfB>r@5o}a6Ed|a5=*EGWdQA1)kOf5g6`Bcs=r=dUR*+rYd+O$G+y;N&%Z8w)tj&(wtFH0*|2I#X0_DfFp1ZPARe~>TUi8XzviTkgfvp)nk4e#6tV!Qy0KEfzP9?el@*@_<)J22)+}FH)L+k$0gEfsoBHhX(vr9rts4pky+Dvku_>IvhND5?(j*Le+Zc~D0HW#d7DXNVI{eB3x$6}al!LjOm8FGbOt>^ci(+G^-)Xc=^A+?J597fgJ}hWphMaYllg<Vqx@Se0Z2uB@?9CJ}U8Gx5@<KrF-W2*S{Jc~$F}>B>Z-__+fGD$~`w$j<G{nfDLNk;%g>WZS%7ci~Y$v(|`fgu{u{=-QWd{`LF5BsCm+-EGW-Hh_y=@O91#hA9K2~a`w1B+*yxslb$lHnUx4f-ZtEaSD_6STuq{0F?x*b$-cZS3LxCa}*zz)}>Sq`w!4mR6qb*b1iE5jvRLn|Qt5av`pwBl8$xbEB0Zl_mZ>)_Ombz|#|qFFB^u!r%U(j!?NG$-l+5axbgtT=!lyww{`^_o-42+BEYq>@YcHPSlz8>_2uueq(C9qU$Q0$yvtN`b0BsHG1`Tc*AmDVp+5MX?JhLPGQXPDm;+`W^D*4^d;+8<p(*NWg%gA+_c;k0iX+ncB1!2-%J4Ax8oexkah>rx7qMl&MZJlSYIBCpZc0G8rTEs*Jn9p6?qJbcps5pr5y*^!V;YH2MnOaxWw-advHCUfrctxY|obuTN*4G%<JJ>$@wfw^!=rjk*evV`;)JPjNpofP%GXmMq{$17q4odqx26pXU9*J!xRNZ=Hu^*HGYJT&@^x&7N_`M@@s*>nxsP=yYzKu5Kh*AZ$fLRg#VQBY(uXHk*bk1&sxRqHucO`4AA3lcv_^v(nY6p!eV;w2O)e%e+2C6PHG&D$xQdfi-GCI!g<yn;U)}S}c{312I4MZeecE?N)z#5by0A!lzzqfCbvv)=<C<EU@{oOj`?mTSO(m9<84!k2MU6<56DtCB6~7l7adh?9!s-ygt$5;M*B5JJ&|TqSUjM{->Q}4jBU6z`@@~ucZnmW5HUN%y-CtO1lHOYgCrKUE$P$3^cK2v#3BpEB*T{3eYoQ_1cOX_b|<ke!?@S2LXG_l);Y-$N7?50C)?<Fh2jMFXiTxRyfmBi<Lwke!^poq_|1OF`P#T^!jc*yf+N%ktVGr6ALY+E;3y!9Mwumq;;yF;w5MD5j-BX7phDaHQ@plMY7Z^NjmeGS7V{PR$pIQymiYe&#ad{B|1PW0V?;B(4~c$+3EJ6{}ALr5ogRI0bbis*lRVXr5q<*f#>AiZ#&)_%0%nKolbkC`9Qb{jWyW$>gx%koy7(htwr^T;fPI0wmB^{y9!8=(dq5Ca{8PwW^>faaH~*($~I|u8D?I#dCt~?Qa0IB(JR5zc*0f9+E6RyZy5awml!^DeUH2hGYb5Y@IX<~FQuKads4xJ4jtlllUzM5SF$0g3^2KKJ)_Ck;gpUdvCs-REUEO1VQ8cdOad)A97-^J`b@l=e<e0NGs)O+fgNJJmTvlmaf(=um(0H)28Awbv~Yt7g7Kqp*7sxxb}trLS4;j8=j0>y{S)9g$q4h#-YPGh!lhScP6)1%NmtBS8tDu-Kcr>bdRKEhQp!}IoN_lS)<3`JoSm04b6ew$*|buR_DZD34QD1ZJx2<GW=0B(th5MrhJ#lGXtmr`B6oGz-4TWP4M$^|NsDDv{r{<Z_wF{1D_``hbnLa<XtD``lw?bw+Ur0hBy-G%LXfsRFjfypG)Rbe1r3s*70$|IGP(C;&g5j)+%^B)bMHET-2a_SJaL>v`Qv;6>MP7Ax%*L7UDaLP0A)EdC$mN-x~uE4t7_M-{rK&dOG|A(u~1em2FiTO%`shS1k>z{83SgKtd>|Yt9B!%)`n+q<T|%{vg69d#sn!?mTz_<3<f?O@IjtNN}i^P%b%Bh28U}QTPfTZ=^b3<4!KN7JP4%I8Q!1eHQQZII9sX$Z+$QzeOZ39%shjaXXL+lp*z-}`J^Wq3=}8<b#jnVyf`nfi^ny{z-gDb-KMx>w9*%G;V(a$%sUQH*pzkQk=625%_Ju3SNy~ge;$_+gg0)h+%Z&#VMXqI(yHT7D6()ZCX1)V>MZREuh;noo&=?_sV}@#b@8q1D_Xeu6;OjBJ6lldkR#n-Aj#iSF~i1v$19_jnNmw}5?pVg9$J@^Sb+sKWTgQ^Z=~i)sR6}sXm4Nn6})7$^!xMoGFk|`i=IIzxV#9dE}*<v90VqF>$o`^1%G7wzxbpAYqkfRxp|~aAw=Z2@>b+$n|xd}FeN){^t1}jXkTU0YU9->A+Sw?fdeuB!&E^V{fb8;#dTshhCq%dXlKsXN@8c8C(a^z<S7|me}@meM5Z@B!zXilyO4mdip>${=ZNxpLQv$NXB<>^VVq9JeoYdB7B;)k?I2rd-e?k4qQJK!%l1|SxC9x`l_e#7THi}Do#m@G7un^miX_twDw(4|X#~U?wYaD9m)^B^i?e)z8+MvCz{(b~?OWvw8P#KZ+NMl(q%?A?gTZwg@~CkDl4&Dt9M`#87OwB5<Ucl^qdAMHvH2QQh8mLEf!A3bGjf6>G?^NGl*vP3^J>q-#>Pi*s}Hw!6$i^h7*rK0&^Psjdr#?rau}Os;ai4JdCbl4Btl`*K`r7j+}Uc7aR`+Wxd!5pX{t#NG3-wPpk|2-%t}Dg0%7`~r-UcZE2t?|kzJazg1HpPR5_|=N)jaWcmmUkkZ&+mu<&;!$a9$%Sg0TGh)*O&l(w)G#W@QViNKi%%t>_2F;&Ux&(anOCD$pM@3b*Ssz=ZWqgh!_qR|S)gs8ntEQG^B$We%5Csh<k%K2HOSBf*P8Nx3FrHmk;-l-kf%Ao5AwFoT(tJcD^AV`Xcv7A*%O^Tgvy;LCg2?~Op(X9QRUD-f&C&;3@=D}!IDIhEmmn=-cps^qCC5j}w_Uc4R!*uz~O$%hZ*W->zb$9`VPSQX1*L!ocI<|&xr{|$COk2z3%3=xB6?Ypg`4e<NXx57a7$0?!9tOtJAK9918`CC8rxtc{ZW`+#=XgvGRg-M`37gv2CaBZgz(zdG1c-1FA0;xv;<}SiR4&Rz(^IjP1WKQhcZsLUxhMw7SMn=<qu<pqB$|6#$^Teh5Cm2E=u*ji__56FW5^8VY^9_I%T(H8<GGAP)haxcdAjLXPMqXDCxwn+ekJFSEu@)f3l&^p0pYbPsjKM-bz;^*W;`^4>{A2uA>DQSU-lCrjlfhIsbrE1RVy9N$8tLC`#^NBoR!_n!>Ht{{E5=*g^LSxKJ*H-rLdMEZHaC1`!zBcCRl-5AJh#J&54#3=pI~-R_IjWJ+mUyc_z`6@<ilIL|oOPB-biytBl36<~s>Gu#VBP4kxP<_xk928uxo0=vOBFK=H^JxLL&Ge1=_>LIve)gk`~LFVU-2#EJK>jLf<BlNz7ao;Bt!c5Rv5Pw(5NRjAfP=|sL(Vz}78zWu`L7qjh_b#G82(A2mxk)@mf^?s<kK^a+3Qx#Z6??jYm>3I=A4+iz1TXI1gl-bT~<nwn_s?R$wtz#toj{!)QjQePaq<9mIU}?Y^Ic~?Xi3fsi6q*P~*cif50waJEcjlNpD)81b?M=^7564=3&41T9y-zoN-LnEwIU3LE8;yh;88arK+#6L5GL-Ri+zg1v?g;TWV?ymfU605oFPyg-s&Ylzo=)0R+pe-T!+ZdyM#2dKK^?rC!VT|^GE_{-y@D|Zg=7oGaNYN!*re&+Fmk%R8FcD$G&ysQ&O0deYrd)EU+7SN(RDF9m}|=EL=4R^#f0bQ6LfBwoDVg6?jNd5-)J@F&51g2beDO$kGa1!w~&La4G_~lsx|7Q1z+a}n|r(Updx-8>>bp$gG$QZ7sP~V@A#}VJ_Q>1)6uluPw%6X{^$r+jMj#))#xWpnV;!>qS}pqpo>d|Jes*p&**aIs-u4zRH_PUlDmySlN88;=v)S9Acw|K7!L=jaQMovDM0QQX(^D_O{(sSF<AHJpI-k$1y~6%J~suEN(7A&(8nJJt9~b%!X^z|w~(+|0t$tJMn_%O1B+R#1AZ5BitdB-n3CF;IkBZz&<YlN6c__mt$ez-|1pCQg?pdW_n&NIFn?Me5v0qm6|C8yBvZwzXE1PZ8@%nRuHP)la0mipm8M5w9y3A&e8>Hv$w72n+(uRy#p3K|B-{<z_i%JR72g?dqI2L+sSO2x`Y#w1dGqD#pA*RApD9QZmX3yL+W0nbrfZ$(;l1why*Y5Es#r0c=?4g<s0XDonjvBGA)29hk^&U8kz>CHtk8Epf+OSUq}@U7)Z@{J;sAVSqs>3C?rj8yd-{0CDd~Wb#vb+<8}iQ*{SaFIgAU(_V4tLC|M$QDrw;Jg{%DL)I4~C6t-n^7j$bkg<tqY|M1a7b>UgdAh(ZlH5>Qqy0(JSbpPR6#3}h+c`7jdqccQ=_RRAC2ZYH%E(2Cr1B=^m4p)V$0C8Md{{H6d5lCV&JrYa?@(Vy!Op%S}N#F<)f3cn`2M;(2aqmqBAMnY(y4Bq(-ffSnYlob7oAvOKpFoHJ!CpGai4rWb9;^G$w#cATl>NuPj$1ET&e<C`pgHsZ{e@bAbuVh>#clf`){uKsy>R^ci=M^dVYZ^B6;w@t)KB%l69k)MtFlYCvVz<R9e1JHGoApmtw)ZwZu5UVc9o%*v*LFW%cMkRs>kh{RlI{ZW3^%tn1co6-BQ+4x-Pt=l070U@@))@BCOqG+QyER+h*&W3Ssz62VR_se(^NXW0ZCWS#P6Kq2rNE|Rp|}Viyt5|9VSbPCOXz(eN!qs*LUg%`&%38C)#4<-|P!2e{CVs;z4b9qn><=G)N*yb#I61<SOm7iJd)ePmhS*$X*KevmX)1PP;X1JWgO10`ndk&5on-)Y+nX$eFonlujwFa{ALpmUnVAkp0;P_LtKae07`Nq{^rIg9|5Uc9X=KO2@FRf)m+uq01TB_qR5mqCo!9)5q{{9YJNbHg-|UAauVjx~|s_9`Efn4g?gCMh(y6FhZYfZ$16!06NN_m|fNt3M|5N_ps~w<xO-AyNR!;ChDeoqJjKp&iihXpdxsQrQu#BS4Hqr9|XLRCsgp#?CN#CnS?_$OLQnvw4BVa9{XoeG!7?G7cNr6<H#C62X7S|5>;`Fa7Cnz*#uqbls^IK+sy(gDegQbvgpgibZnDD@zg@0PUd41nt55Nuq`vIp)^h_mV}&Yq*B4AxwpP6f}lx(va{7_Z0$Zxi>Ej-gITH5wJE`8v!K&$a&0YC`}VI&M!&P&otJMI3gG>U5co?tYw0X!y0*WY3?^UK-_@0H2u2hV!YuuJm;Sv+|K1N>uSnBBn_=Qs=q9aF`sFRb&;ICSbzH)@)T3BD_MCUc4R<QJI_b2!y(xi^qEkKUG+<fvj$>Q+$_WrQHwJ4_646zAB|b&}oAz|lo1yD3)Np=`=W`W%1k(`_Flrb=o#=cJhRz*l49l_|b0R$a_XI^l&f;b5Eg$zX>7;nzF2iN;wBI`_$EWSJ`|p=%Ah3BrSimf@r_rp_I{{jhyk>Q^GSAmtG=*9qtjH6~Y<yMX;9trM!S@<`&xW9Gd~J;gF#DX-vM9u~GnM7$_aaNuY{uD<e7!^Vs%JATw_kHM^NqGzKAV4h)Z&Tw<ZL}!zQ5T_TGuU^TYPu(@EIkVuS;aD&BL7?O6qsGv)i!Qb%gB~>J<ymHW*O|*aNXA&&R)DK0RMB7^Ht5w})~)3WG0kti2ycQO8GJ>>#$-#JLcL6w(ovAM-V6pGBb*$V!bK0z$xHe?E;SxO;#wCs%<xh&sLY(DmD$4sIkQ*l5E0$Zt$>^-nqtH04f8-l9$$bbBkc`{UELNK(S1flD_&bF%<S`nWVX;XldX>X;0>cf1=-3kT)*S3}b6r{2rQS#*V*|6vkf;6Rh$ex*|J88LtyGdMLIu8%?rq9Bt`I@_gM$&V$~Hnt>3fDeQvFn0|2AxQPza0py2Ha>^P7ZH6*Vn2B`<6tV~%Dwv-Viq!7NlX<4s}FLT=&ym;adO^oPek-*uvV`47{Gr8<Qa{}P%|@$Jfh{(ObMd-gv$v2rGzd8<m!NweQLa?Ck3k^L#6vZ44sIJ2jV@yKZ@h9dp@M&3pl|}1H|l$XmSLXDM5gos;8%zB#>{UfeKdNJlyzr^Kp3a(SE(Qi7wTOev!yTMC9`+_$<7f^l;6D-5yMtl_~3lRpNR+VT~J1X&5SwSVUGh@j_l$IcqJ40=Q0jO5V3DM?8MQetU4#X$N;<apB@TKJE28a*5`zPG=at7#+n$ROKn^O|OEL)y0-*e)Ow$@A}Nh3Q*gx?SxO^bO>~47<pVE<{<H_>Kfq}ioYu+%%j1ClTRw;H9uiz`18JAN!fgP|9+}c;!9AQHj&lUwfnxDwCXBVXT1JEd%boyRbjGeQ?;(GReZ6{s`t`$-mN4r-%r(;tk!ft?^lZYxmKx|{k%30c@TI*1j|MLcyY{Y*E``escw0+?N&cndnm*J>pk!JbmBjsMj&E<xImY_{Ci_+Rgpz0C=xvIf}X6TCKqTmD=lxGJ}$$6O4tLwkIC>UC5f_>l+p0Rur%Bnu1S&w#^MzSj6%p`TR&`+fU>HQl*P{VaD9r&5E*d+>93e5w#dryWNUZpAma-!I1zs6Q@0Yb=E-eRZK>upJE5oPIqSC>5m@*^h0hKjlRS!@BV3Smw6YQk2|Fd>Pc2V&pHl_&+m@rOD7H7B_nNnBY%Y6M6*=!%sxc>=B`nN4m!A97)UO7qdpr!W+2!$RxzzhF`|WEtd%I87c6agGme=OjH;Ilx7((B{OE?!`C5=ayAQO&P?~L#Iz{r+PGmy}j88)E&@W#h0tM-M(AI!h>e@jxED|8}qcZTzqYI>o5{to`%0TyFtPgxh-+2Lk4JibC#0>YvZui(V*_hZ6MaHb=NyiEFihkIS-`chm7xB)AuR9umcc;ID5o;835g@m%s1nn(dp@3e*1LPNi_E_)2B<@6T2!N143Jie<DA;Kh*9y)*)>%rU74jJ8G@3*VSv8D$Q~2eeiGPR*f$1j8c`(WH#X&>_S#{+|iMMPWCilVA!)k}RRLZLJE_Z|y01nQEIA>(im0ANbBQ<alaWs9R`mBM>xwdUM@0y)q1j4v<<&XW#bT_W_v*pU?iCZ7*Psj4<rT+BNpL5_`iPL7BJ!-C`XHVt~(j4YLscmm<hIRPQA@llA>Kg}p`x%7;_39EAL`kpe(tEFT`C-NLFa31hc;IWn)ymj9A1~cOJ94YMMW8;)5GvGj@EryLJMLr<>t*5}Op>xtXlZH98Z7KyQcA9<6Pj(b*||(_#^<7OaLM{*k<t9a3NHcXe3lk;$55UGJ*qgBTNW<(Am*I+F*kdB=C>yUag38pqp>?W;W^85di}^lTLlUSSeNFM%RdvZTmGH++OCs#oA`<YWLoTWkY1>CUgZc)=_QeSw&A`ZH!zzh@%cL~IQNR{hTRlsI|?atnJS{0+U#M01f%-xV}BS39jDEud2_2geJ;))w1cylgvc@C7irquBx84o?9{FD1Al1=hL#S10?ryUK~tHZHM@Z6G3^^?5`vx%$=C~j!ofX4087$;5`K(pi>)EMY$%d+nK4AkY_=tWoXurX_B>gN0spbiPZM|Yfc}(^G5DPyOVZowM9Q}U_*Z<!^lt_1b|$LAS%BK%KjO(1%*A;G&#u%1TtL)T8gG#cXrpD00A(+&9Xg8P;G0Gv32iQ!8jzU)pQAW<(WM`pW7-i_NWKP+Wmc+8$8rJV_c>ru8_y0&-0__r^nvUF5<-`a6{YxU2xl~$XAURQ6K;9HfY6=+XY)m`Lv}Mbzs26t!N({HiYOZroD(_UW^x=Yh!r|S0W@VHZl6VIa4G{v3acek2YQqvX#4&eaI)c@J8L#BukQ=|blU4PZ|A&fYn9cCzgAg;qX}wQt=Ot~x*Aqi=_UeO6lsI84j{GkM<+wvKK&Abl;T!?x+aVAr3H&I4N60AY0ZLK<FencilKlH(v@cT<|>zi{TQ7@leh$J`Pg2?A7%n3$+p%W%rj?Ou2)xMKg9nFQa67eXQ<pmzv=3ON=y6spi*u5=74fEP0bX)vXD4K%$Bb$9pAj#)4e|5tfEUhV_n>#Q+W)s5o%jlAC;hYq3Jwym!Wzcfh@c9t&m<^Sr*cZEDO{&%F@YRl)6y9tHguRS+uSa8?xI(665Hwq!8yJ#v!aMDM-=w%~u@uASQP^YCW1EM<7)R%DB$Sq}PcZ;73m4=dA5|@=!V4Z3NBIG$k4MQ(eH&L!lTlOa}5bQuBis{#5*)ndioA9sMsk4Sh5%$?n))5eXH8d}Xucmn0sHzJNqmyV7@}$i^XU{>k?YAz}$v@p1!Nz9$1(3cHGrusd2c%3&Fia9v-0Zg^!MoDU<!k8CHHznbeiE{-GEu4g{RcHj(+9J;me#nRaw;jIo)gS(tNk3%{6u+WwSds4y4#2Q?~5$P>s$tf_aTVh$|8H3M<anPKmmZw;d#@Y@72^+ys>6-i!$f(%jjFkF@2!c~t%|gpQwd~M3j9o7P$dk|>0hR-vxZddsVV)w;TkF#ax_R&;jzw=N)Eg*$VR=h9HmzKf8Hiov&7Qh5uyeA17F`8wjBwf{^tSc9HhBm{;?NS;f7}A|VE0(%YivuY@-18>77cNAb)Ne|)~(I<KRQ222b(!US$XbvE~J{3>8O*&IP8QeeSVwqvAPa*E>t#0=n4FK!wc8~txEUB?+mt3h)Ob9g`l9zWXa-l@N*wM$U^kk687>1TY>e?DuMX8N&1Htffk3XXTXQ_S4%;f5H`xi!u6*vd-yf}RS0!K!VG&m*6BZ}u58EfY=MOca%!oI=LK6$U}yFEh)Dbr=H~NvdL5i+=k(?2weylGdtG&2lBY;C^U5KK-C=m``i}c)ZGYFD(@#1z`d3wTOmVX=g#+ZbnOE&a(45d7r^t7?)NoUmCPp0cL=!_Mx)>4!CAA8bStFm#UKEg&(-;sC{)NulCROZ=TshRfS~7(=YpFxDY3pfF;%1w21GIcC`dhx!H$!fd?>bL<(^$YAaF|zHW)UEP5T}bmxlq8>Px}mzhZe77M$F(8si>Kqt;SC6VB;fwb{V>ir5=>V)|N>in`TZAmVgZPZH8WpOQUp0C|<j`sBR(NqIhHWhYyvCSA1F}GKW>NBl@_$My6DrISL7sRaDdH6d48HddHLj27z(HucwrSiD;6+x9>CuN?Q2gK#B%J+p|86L;-=dW>W%b2t(pZYL5xGT}Z{kn@$Ye3Y0y(N7JaR-6+erK@psWNXQus&~mEK{_C}U^z{}OJpuVYudBJ)0Sh{d5<PS{$NdvA_&dEAk)EbpG|DXs!(orV8eA)};qb}c_GY+UuYK%#^V(OMLR*W;OK3ynRGQO55~Lu{e9Z-NI(b9gB&?ZsOb!6n{UdpOAhIWFY%;Cq7MnCk>$}xbSDIX4o6|s{lv&%IM*sVbhv3sP?CxkBSPZHDH1dn>1gUD$whB%wnqZ5$hVaS_9F6^qVfiR1vhQZQwNN+ICfTWPZq;^eeVH6(`of%7yCusG#jd~RYZH|EP6!04V}Lv#O6If{(l^U`xMCT*kaOh|g0OJ0;ZNQ;=O`Vpfz4QIzmUg0GEHIC5qUARrxAI8#d)sx#rGz-rKJ2;a;{C;<(j@b>Hf$04O&cY!j|Cc?nK-o^iE0Y#h7$)F9<4h&(F)ALND>&Oc!TkUglny=g9l``E3?nacyR(SraO61us+Ptpo!)JqvShCNAHAXue&XaKyXm#=cDT0hAY6H(&x>GR{<C9_H6;zbLv@aWPb;Ix*eX#Wf-?59sQGWR<s<@Fmgm^LLoG^ypMR69ndR;(}9tF*@q^j@;tLxDZ`rOZlX>kWAgKjrpE`J5283>BB`LB<jma8Uqo7q6f{1pYdsXk)~L8UOYS3K5BaJaM{?sgVgsi%o3ezgS&Y!TDQF%ld8V!J^?Xifr+)wtK8o3C$6Q>-+_Pm4f2iax#<x^Y+mN3z3f&U_L76A7*10u+NXTGhjBQmaX4nL=3VYxkZ$+(^KRjMKFZv6c$Y62TxPGNyK)*>2<pUvAe^`OWs@cR!t17}LfH^3ns&alY6uU?ZewP&{=XPVgW{`y!^wmx{yXWw<Jx{Ix;u$|r(pNrP<Zw~Bthq|ZvHeM_JkO5L*e9XhV|zd7aTsT?`~qGUs3_PJETxVH6I9%f{m@Eq;ADCgO)`|wHD;0?9qM?Euay{ud$X5PeApaufs+odVWxXrP1n5ZWTEdCk5uz4ANfs{aO}VQa~qZ!}2VaiBjC_kRUESX%cT7&nH+yt%9cZcbWZx3GR#xHwoalZM!t*Ahas?2R^Z+V~G{o5##fsIIv-Uldy)8%oo`8lf+k%d?NBBiuPm@d{HJHDw}>vi(Z0SUYT7fihXy?p38e=I?gchaf5$_bG7@Z`@ls9$D=C{4^r`mmR9uTVAWB>bQ|jFLjRvUB)>fWpJABbevx?0I$uN3=CxITFi5#(PtGT19TvaC{3;mD!K{F3&*<Ff<2XE40yo}b`85|sjwy;4HPY=y@a;teLsK=PF>3KjsC!{s?EW!1>rQ3xid&^<o$KsoH+h*p<=$=fKW63#7QgN%n^o;!ZBOTHVBr&HkMKs-tj<!jAW-a^B@mQ!SlTD9#RUY-ApUqh>q1~c?IJ&if!&6KnL{s7nT-<$g+x=y8HO^(KL}QRxAu6u7B*^l9)^_B-FEbAx1vxo$P|NVFRa^XwHD-tE%LoSjxS1ld~tl4JzAP3C-f<uOnTt0ToTExl#I*Ms6y-E<N9ZIaY)03BjE|f2Mt#Os}|Bh;v$n$Z$Mj8UZNuM$A=Y_Bmx)dMFMDwMpqPeb*y}rZ6r!Q;Wn5Un{d>-o)vIVTEkT@j6?|;!CL^{$zfd!J6M8{AFneSmChHg*<CK-hES*zU97}WA2=NP<RAmb7>buR5*bpgf#~(PJv^(rKFSBa`$O3asE}_w0&q)q7K@k(n?(V=uU0P@1N(6E6U30&L^z#K?v*Q*6t2$h-tNk7{b}uB>yx_EIDA&$ced-b{oVRLLFnL$S~)nxwPMuT=@4ilCN7<gqe%~;1vtyK0v{tA#X78_llJjd3Purcb?r$nM!c~}gt_V+^f5i2meXiE$G!1Yc?37R0q_C|8-srN^c(Jaq5P49(*qkP8eo6KxAhs^Wez4$6sxdJs>bqsE}YHm#6Z`1{`9d~r~V)n0QYn*)Q-B*R*ODO;oPLWHQSvFObgx#i{j<15qe*6u$Npu;coa>dygAvGYXO@yL7%KJ$4qlnHQ5Rc$qN}Uw%O5nyp>n5otXBRP>iYT{H4_%mHZnGuK~xzvB5x1g$8e9=QJfJi){awG|&44_%+)!96YJmi>f-_k9EuB@rfAE&F(o(~l4&SXgZMEEl&z2J?PUrqEuF->;PK`uJryh^D8bPC%+_*B1f2Uznk}G|Qu-7c7|tXrrhK=VSIgF2XW+4yOd16!m)JI|Vba4d1pK7n$OFVv41{45;<sUT!pXI(YwqwmeqfFCb}AFusUY7UW6xopiyaDba$L%h&VJ5VBQ~7q6(ODB^e_%z$k5>Yc5PyK8y`@<PG$*%s70TMKLVD@CKohLtNO#V0Uv5LR8+5h9K?YQmynJupM40lxVDA)Iz_1(2-E*C?TNSvpZ``@<Kr+sf2g+x#qenJ~i?&r4=u!+4(1&II1dyE)N1!R=&rOFR-w5`Sn68~ewwXfqPdW|Ut-qXfavs!=p&t=2r0b_PN-&JC<<Yrh@>H`jpH>=~BPJ3DExEW~Y8(j_wt%?J<c%{uD9>bo0IV{zsg6q@8{y$^%CT7C81Mtui?s6{^;1bI*oQ^;_hK{ozWhwf7kRc=4y>DDf+#7}D5**9AI3e_)pTLAcHKj=daqi0w3wBal#PCtF}1LyT;_Sk%^CnqfbU1vw6Di?C|Xu?kADwl$DbVUMUqR>b<LX6-Zqw|wfXV}J*N+Q{(+c<><hT)_R3J)lH`RGjV6ADKHu1XMn(<J_vPal-h&aL=YK&_xaR7&tI5I=y>17?ICupo56)i?zlpkM=L2phoPARgFhdmNQAQP_Gc=U71pSfV+Ts7<wi4E%syh7aJthOBj0MP8<@G>KA+O>ulFB7<D69gotSa&A}{q{<qVzkp+s9A-jnk}-Wk8znzat<+~oSAK_wGlF?%zcI0|)-P;<wsv)lM%!n3k+K(x=4vM+Ur@wO3s}S8S()a!x!PKno(2q0jgd(w=)RN+cYU^U(X9F?qqjVU6sy<}l!igDDo<MY5{k#A)dEb4cb&#yG@72mHk3hY4r23gR;SCTedffcaBn@sd<)ajWqShq6-39rBc^DbTn2H2rM6S|d?*dsm)Qtjiqb7T`!AO(7Z~rh41)7riqu^;4~|dkcm{^zTt$pN^ccLPqUnVj2kI7Bg4X?61!uW?4l^_KSIc)0)l5jsjqS$wpSk|g=zQ3TgJ!8xh5{bcZ#h~SXb;#3CZ=^Ma<b}1KY-I78iCcF86)-3TY(C%{n41gl`xWdhT@F?TIo(vpS&8aJiN9=P-Ec%R8#S_bqXUa&lvG8hDMH$K_FHU$k7$7Rj(-*Q_Ht9A)z-hL$k%A25u<=n;!+0Ds>}WX}YJEcf=Q&R&bs}GYqnH;lq4;ALhJ*)vem`mXR`+>xQ*gxQYq?kgf=LZiuysPBd{e#84Pmq<PyLM}PStY$tZhfW0OL1w8l>_Cwp%VmUUpP|A27KHlH^xW1c%#By15Oo!;&0=qLrJ1DuIq}f{B>)7J3*R2j>>kQA8_U>TU&N+UabG$=SZlB8A6Jtf1Z{!229c9xANMiP6%To8)xiE?(S(CK`j>Dk#!p2Sw!>;n1i{{rwPjf4iPN9DR=T+4s0qwd`J8XdPxVOJq-_KcjX7B#P{Z%#bU-;xL;H}!ZxP`Z>YT;>p@9BQ+*+(k1sjVe+u?nnlO6zz^gff%~pUri|T36Zy`K|swwU<?XHyu@2meyc|*=;NZ7kq21V4=oWt$L5vs!Fbt3CGeoXJp1P7uWui`lq3EtsvPEp+2tD2too`&xq&-b^sz`))1}5A<Iixvud98JGJMq$agolHqok;ToY~1iutaDlOXpRWX4<>cc2{GG>Sq3p~rS?X)#ulrY;rd1{w(RNGo&+8$fSEf0Fk_=BBxj1$z#LMS=H@F8KRNFcft(jbGtqVxU|la`Q7F3ta*VfkR!ksnB%-!OoZmr*ahw9;TKrS*&a^Q@3&j`%gk$0U<M=zHga`Xk$ae91y@=&U%IA?h14{1JaS~{}RY^edpyV{67rJ?p{_!=QQchwpMAOznLwTOmHr-OK!)vw$cCU;G-HG?4Rz@XeA2?VVG&|B{w3PbW0kq|Dsy@qFHJAU%=H2nEZ4!4n)yr6%S^UWhue%W096e_B_p!C4x}qhO4m<r+Lv_Yngn6_$H?oAB~y4!X;-$aZ}8fF1;l<5+_&oPcMCXi4}a&8<#|@K2+vq(BUtTy-BE&q+e;h#){}e>iZu&tOV)`1UJMXA#I3S(IQe>E&GG^^t5~&_4=ifSkcR?-cqIf(7PjkCwH_{=qlWo(*4BKHZY=1fTy=C4rA+sXmAuw@F-eGag?wdZO2kcY(wJ1p+tL<Bm+`O{3n~-RCL^6SM!u^maQfFn{Xh5R|v!zan0mko>I$96zYxC#=sN$V<8`4p|ol8A1%_{FcVh{T;o>a4;y=sgjPs3sbet}kAg@csNw5J!YepHHUT7uEr=U%T`!STq-$?vi@0g6xQ!D^t2-z$d7F}ut)&Z^eSXX?2{_+k7JON+H(xJgfORHHYV^p;*8!oh>5{2kSlW?gx$8C@Fr%0-mdLe5m{+E>XL0}GLE#X^@T@DD(pocBE_^Jg_JE*;FTVEpijM_iQKM(Y=1*};c3Ry5mRa9N{l9&%2)ZICOGUgIgPzccWf@-P>0vYYn$(l7Q_@hy-KJ44ccaOC6aYIj>$9uzzaw$MB35xA1pn7}p?k<zxfJ~H-DoiGQRoum@5m&nl40=pS^sABy>r_aC*ftUgBHEDC90A+F&wQV9Mv0Jn};y$?;qB74?tuHYfpd>@EnDRkLri}TZkqECo*St>?45kLt8_HIo4M_3{TDxjA<H0!nsesi$v&C(e69(X|GFI8<+^9H%0Vr0cit$TPLL!SL=|=*$SE$__rd&byi^r3om&4GdEvE=D8G0GYLnrRVQ|wTScvk+zq^^we5@<|HL5$B`2uXEDaJ?=Rfc)pIBN`EPkOoj}!!~%ZXOuoVt>2=rgOeRg^s>ke(qT*x)&reLb&|)F8!aQ)$g@{gy+rpI!DrO8+F(V4uft*dWnx<t_HgylnZ#6+^pD3`P2M>Ex?UqHDW)Djg43wR&s@$_<}j5VB?R&8I3-O>Q1}sNZ6*=h)cA=BATK3r<Tyvy_x40@mo<q>ZJy5(*irH_K}VU(3$ri{g9^s52m^uIvu6?ix3}GR#s*?=9hSsEZHGxUH@!+|Vr%<EU^T7%6vJ5D=FNg3HdLiaIkQ&_sM0rJfZl^vV9-U)OipVSrAm`B7O#{69O**JQQG4^R{LS3YcP7J;mkw2M9-8SjWBzp5<h`V~E=t9$d`F*NDs%bP#m{2DWKR?~?}zr<9TukggoWXVO71d3o`XEcn`35ZS13R~zf+A7k9iqk+x%TRU;&1fO46ZxhIsp-o)R!F(aVoUMfnocOw+IOWCO7t-l5vb&!d7*)8_baU;C5>7tj%1Peb+5DR-f@??&s^i&hAMEo%8w8<m(Oyrnk`~mzUF88TGsOhSDlw;L6I9?PaQmGpHPBqba$qZJZ+0OBAF{Dm;FCp|LW#Xu9wN$p0hi1+evRj2a|2X?)$^*e|-IOY=a^JDZejn&fimF*Wc$k7nxx#8Fi(Xbs}oye4<vfttEM>$@|#2^mw=DUT`@cw}Cw#jYfUxiIwWMaBAUcB~d?Hrp!A96UVVkBCg$@)EtVh$ts?5idIp-S@oSB03muhC!~O)3XtjdBQ)#0L_0`fcRzmzTloAPk8R7U!xlOS5FvA!uRE@@Z0kf4h9K=K*Bq!oPESIY1=qsf5xZ6zo^hD)J)b5iO>2FIkx0EMRNLuMR=AECAG;qNo%i}3Xl*ze5{d>V`E@u^As(0Ogf)eV1_!eaQwRWvsWJJ4Y4KhXdHBx8mdIf{8XWb8QAZERDBmLN^)D3DIzPY+4v+9ZbC8+rp9(Zf?9hh?&khg5RNxr~8s)~i)Py^oh*_~Lb{tcD7CS$5aM!v1+>Ipn{i55*n(H2<cWogNYIjap&6_2quiU+VeD8i|UQ&8ltT-wC159NOC&rH`+8on}Q*1fqq~{diJ|`xBMe^QPaBO0R_FqxXcTOb#51cF=DxgH_T#mAgQ}TD4Tzd1*m}H(4pZ^<Xfu~IA*p1h}z%ceNDGU3TR3(U*lwKXhOUe!ZyVpO(tmcjoKmWuz<SBR_^Qf!5?l*sY{ZkqObOS5D`2%Hm$B_A7P?tqAcrD315aXYfO#5|`jh^dEhWHz-jdHE?LHqibA`YKO^si9z6?H~_{6|c`{s$=cbMX_7swp~tjeQcEp&LIDt^OMcCjTfB)@P>!iS~-p)qi#K4P|iu5z~Wl08j?M`TvkoIDDTI)Bl(f&Ht80_N5$^${hdiKl*2&6;0VIDg?d#6*c!;=naY*NZGHU7*_oHXP8MIn*UcK*MG-O0G)}tAn<%GX5~wPU+`7>p73?mA;)_-LZTIlNkj}`->Gfai56ec>Y>@dsl`O|NHys9FKJ~!m(;@Zou{k+>tCL((QHyPVnV(&aTIP|t-zW<iv3YM_>yTwjd6X@`#;KhVO;+NPrl;mg!M-B!WGR+hp2^1aA`?FEnr`N2hGA<EG_Y#;XZv0-T#`X0k87+$v>8j?eU72JI%RThP2Y5G%W?1%YRJpAHJg5!?FBG{Ux?1E()T`@2KanU<x(|-sHY^K$`q1(2R*_bU{}7jYEs&8z@Iqr5RJ*%q(ijGGM2R+5I)NtksE)5@E#FdnI^)*h8;>=Fl4cntRP`i-w)-(AR$lufM^$%&fxK|LyhP!E4$xIEQvJ(iXt_zQHjNrpBwAU&F-xT<%Hi#&2c4_=Bcckb)@ur5Yzx{#!wrud!=1&p>=yhjPp~>>n$>A=fWW%U8q|lO-gDeQKp#9G1DXbg@>xTU}b>s=wlj9pX)Yrg~suBf*f&f0X@_s|iL!ahdB5h)8nz6frjH)gOK5qfvi=pZ`JdB0FP4x_#|D{%GaN=DsW~*5_}SQ8A-@{WE09&@pTS=|mKz0|B=eu27tRK_4-Z&fg6Grabp(#c*kHf--giJ9M89r|@=&3OMb^J-XZN96$KrZtjr_GEMtP{e%zNaTK0Udie3_bUKc!ckc8@Ahe#2;%W6k<w50+m?(HpQZ$sjzeZ24ef=0Wn|9t|p?8bkJzVYFzjtr#!ODZq-TN!|9<Fv*+8?alU5OqX-@E&ud+(_G!I6Ia^S3dc|E(N>z^OdM=-rLIot?ei6%fB1jQS>t#dp}#Zsjq?^*WFD_6`~c`?Y5<>3iEdcVH?76Q}E3jm4Z~K_*vPjLmMKehKfh#j*$Q4(7bPIdFsDd}&;0F)Uq1y=?F8slPY&_V?8ftyAh5^nSPTq`sepa(P-$s@d9kmIT$H9$5U{_^6h=c%nTADRJvy>k~w}l+Q@H5q?&$S=vIc-BZ{$VY~C<Jf4=uX`vp$Qfx`OMw3p&Zzd<?X#(%z>G_oWm<QwO6<vn$skc5kA7UtJP$_5CZ<F@$%zt5G?NQ}cp9E16`Z9{Yg#$)~JBAH5U$nd>)%prH;Ca+HcoT@vth!7nlFB1_`X+?AHC<M7=)&pY(Yy)Q7f$P<KDLqPMS!h0@za*79O)xzWztC2bqDJ{TV8!rBB8-5gqzv4UHZZw^*jDBy7WoNWQUMcPcMQmn$1+|CgWNts%W*!oylm7&Y-F6f+)DbHS<JZOb~F<g*{WD&s7Mh3cEWg*XKP_{?m%R2>)KR=!?-~oJ!iuoMsyrL9=wygwD5CPMX+%ah0Q9QXjb?$_|u#Lp`t*N-6))Q&_kmb-3O;LvdAHG(izf#(J+42S$<|*0l>F_$8h=?NPdc3m|1s>NKIE7WIIFc2=Q3nMY|%(O#gtP+>)Wvr?|C-S;cyhacQ;WrOZU&`W2dJ?W1e_DNackY;DhfjrBr&hcn6VdEW|%bjshaXRN1)013uP_1(eGbnPCm($Jc6R||L%#}WDxxRxgPb=MiZyY=AF&rQ>jGBv_HeBa>7~F^FG<gM)oWdv%P?Wbr(sVf;{BSa4`ZK6$DOm_Bs~D@(@_fpVWSIwFq$p+GP0A?*U9}OH2#t%-#8SsI=f)P!G7d_s%cz%-u~@1W@52Cp)2j2NN&*-Kb!vGvFSL%B0NZuxQ{O%Ah5gZ`^3~{(H5#9y6P)&V)$T{@j>F^rIihNWI4qnF`O(;81gaA@JL+qge02u{-~osZ#P94brmmVNN_^y21@j==uX+bn(BIc~43aI8kIaYdi*O-B3e{2T$GMYaXywI$O>s^dEuNZkE*0hnqi{pF_IP8n{^aRLuD7&iGR|sqt+KY(T3UU7u4kN^;1V-X0(@Jp(5G(L8^ChEQ2E=c%~gJ-8)QOQX*%QQ?)#Z~v(@<zAL_`(7D&XHvOA$B54o)hUJ5aTK(fMGuwN<^$F#aeW$1mfB3eOUNe9n4tTwg}F?c@Q+64)$zPnK;U*W_3x@*sQg;Jv><yAu^xl)sTC*e8m2{5PuzAngj)kV07;Ym)ytNBV-3EC6r=vRN6gWT6NpsX{1(k;&()<l^`-Vi9f0D~~e=P3~g3C2r~NS)`mhT^hfaJ}o)*j<W3AEKDDAY-E5<I{FbQ;6tn&g1%%z5Tk_r*LHu`F+>LYmE5Dg=r}CVLKo9#%<_D&&VSsn?9M0g9HK#sktoSm<{Qq+&(!enf@(7DV=f=o!8n!*Q1DPAUbLDOBU1RWv>$j&~$m+yBJLcVi%#M2^ZLawdxE*5Tx6gTR$)Wa;uCPhEYq2MOe`)%~Y>r_n?&dKnSd3cW;*#r9<hp_=SO)Mc0>HNJl)9G1|2)>3SsoV7aR8D~C_4oFk)})s@wjul-)Fw(^pvQ2omR2zc%eCT85Z>w1)a3ID>Xk=Vq9@V$D#*}AOx`}!IMjDgk;pU)pB0m5rthmptu+pyKp7`8x#1agspakz!laFzoHzWa3R;G@06gV0?*$$p;*wd!H`bbs&g87&e0TS>AUkWX<VAS^j-yqcTi21<$z%2NOz<A>U#IZ4LQ9%Gzr_>H>N(7~my2Msfu)$*yA+b^uP0JM2Bx(vP}9}EqkS|NABa6<CU#BsSWF#d%joL5l-ZAM=&yUxG={XaSGGJfk4_QizIC9d2mZ8qakYhF@0LZf2Suz(1zH_DiSGxbPim>~Z0-aK(9^|kQW5sB`+_foHAPKx04)-@-<`1IGUzfUBCkPU;#9R>0Wsc;&Y{NlrkSAguUbW+N>XK4x6m%p@xRoc92&>JooH}NJiKucr}XQAtOs4S{+DK7=;qLKph>a!>s2QDlU%<N%3DsV|)AM@BE=}V#$4t2nSc3nVGiUHBeh4<n8s-c>D6A!84Nzd$EdFQC&7~G+{ujo1AFUscmEdAZD3la^Mu)e|6v(TlPmpVHok}OCusj~cw6HE46WaviU{J~VG;hfsPC53Sv+g6>v%!_E4Qq6wH-gr1*z9{0IAOG^c&2mFNlX9w(oG0ryg0Iq)z-LXk;iCeYQ;84T!*zc8gRt9~gh!|9c@FX~h^Hia;#b=0wp<euS4@u!#uA}5CogL+4#nX`r~`yGR=6vAv!L1IPaQ)1>`(+ed4^H{@F44@PD#w-m$a0hysuZ3^I;q$_0N2p(gZ!MNq<nB;w9-xHC3aSB-2=sUl+-oMittO>S}e{k`R|-p8LzwXcCo}fUHe|<L^D|*`rExVU5f_J>gOC-?BU2qe{7w1C8m*d-NyuaBJt;{@y3`o%-&9>!%uGZPud-x?(=vuWfRoQP|lMc~-cbZD7dU{tg0FcsBPv!Y=-#w!OtQC`E&S`K3K{E%lGIh7rSOIzjN<b*a7NtLw(rI@?2aDxjSV0d4!E2`@go$6U4>DSx?Lkfx&{;h70-%N@QG;Mz~=KlL`Y_x5O^Hi>MK$C2+7LL_K6hxe$25L2P##JD4=i-VtN_J!uc(fLiWN1IZA5Eo8)*cmc_x4v@-_tULsQ25}pXK6=NdHQK}CyYH+HpTIC(wyJ?kDIT^QvPFhdouk=(6;|OGa1LPlutk!+quf_8SD-GHM^QnY7fob<>#+|_4+5KL)1>KerJEDQ7_xv3q_WV3^skSyPyKTkqFRLyIJSWa!1&`0>@)rF=o|njmI*a3^D$J=a?uZ<8fI_EX)P7pe5a2zdvl3Dqi^*xNQlTR5S@Hxh+MV5VcU)*f@34F2R$eJ1P4~yMK&d$a9V<W>F%gnVFt%08aUQt&(5#B_@4^PX)lP{8GUSBrstJfhHl*zVJM+xY{AqsoxmT`~f|Q&j%7@xTH@I6gVGPdYY6M)K#BJYKt#Znl1Q+LU2iH#a<>WteOZ7X(<q=qUq|xP;HTNIyt$pWI5?v`D6;q#CF`I3W{FeNj>&i97=a*Fb>4F0<y!Np=N!!#LnAJYJ~RfWa;SX=rV9eh=5GW!(InrgIx+!8DX>vy1E(=$eFHE_Gymt?RC3Aq~S4=tOT#h@MZ=7n_99S*{5t(;i}B33twd_!-x5`4K3AhSZk)DVSX+8TFB{}UL;es{?gitl^0{v-)_1s_L7j6lk3~2%(oKXQb`d#{1naYHVE{gjygoTrtDLTB!{EX^wibxD>JP#Sr+ep@$9u2+DrhPq4lYf*J@cG81}UygF0-&-`1`~{5_C!x?cy0abs(H%Mb@C+LXv4LOD3L%ED2)F8Y6S8VBeP##%tP@oaanW5G+Y;ZlR-v0dlD&2iKI{Qe639XChD*+8>BAFI%t_Q`OBW|-JHhanL{mxDpLBs5@T1!Kk4wjG(O4UrhJLztQ&c1mdZhZ<H+N5?D!sgPc;Y)>-XoTlRckg!I~o;AH?IvfxRnZj#2LTEZMa0r_Q3Sk02n7ZJ01G=uE7*|H4A>MXNPb(;f9~|!28_Q^uCuO>Nq!6Y3X3MjIW%PWsry8zS-EQ9XcX~-^d);ocx4c?y6{y^Y5zs)_^Ev!^>ghbiJ0}sq9LpFp2j63s;qkbCRWc;9tng*`5=!Z*xO(r={R;LDouc>&Hr7f~@c7~>HlI9&&*t=g66n>TOG-(rNkRX_W1z&KJ#_OstK+Xs66j9EldP0KC{O|@m%~Jjl#@F%KQm^L@<_%Ahc&l~Go*e$hI%eX3gauJyP$}eXR7;Z;Ne87m?nlY2|HgG&YtF_LJcd2k8BM=5`ha3Lti_iF0l)Q($He6E+T~OC3fJGG3dB`9%FVF%IX!TID{FW?{i`o#+ouA%X2hHlq*-+q0*OIU$b|5Im-%ohHO=F2!tTj9uf}eIvwBIV_^YLrBs<t{O1z#JT&pnlTWn-x5-0>P0FA=<NlE-AM7><L)_Xy6da6J65yj2jNV80Dmg<Yp)n#L8*-eq`>^SrBhrORc4Vj=r1&b7YJL^~4az;zM{!#U-1RER$IVfN(=%klX*`1S^AykXzy?-v#f}aMKDw8aff`O9I-$<#*E#-;bB7c>|DMTJTLNq{H39NWu!BrW1jwAciYm7vE#K8u|LzC=2dhPj>@R|uq~#Y)=zS~C1<@;yqsc(v)R(wyLUFG!9W3@&^0s`~kTj9u0=lB6C9eVLXZ=PRc(HNtfa@0p!=t3t8y;f}yi&BwYch`xkzeCZsF6r37f(A$v4Up!x*}4)xYZ$42~!i%ZB}x_5SZa;D3e?n4chD-?>bM=JhwuQ+30AG$kU^vIGS9*s$^+RfTOHCV?370!!wpqupM&NggYbpJGI$u5*>-*wbf&0-X74kfhG6pY6wjA=^Fiem;Swn3D#h1+{;`2vh%<bCcFseN)hstL@ao;x;95Fh&m_C(!?ja`>U)EZ}k1%FdhTvDq&#4>Z<1z3o}&he4(U$e0)AQC*MSn9o8r9%aASvMSGr|&Xun@x3p5Y`THaPv=>mx=2`V@d978HojN{CXnqt;FQW)oKh)sE-sqpll*oxMb%mx1tVy7{ug)U`vHw1{OXdjtRPp}VX)x=nAAROt{SiFpKTmrF-54GR<~~j__)~8s^`!3=F*$9<mIJ80z+*&ArUKkO^MQ=Zr#<h^S{X^{YjsiZaYC@7<t6Jam~Y(O(Igpon;q*)&5Pz;4cAlF)RHKZ0Tj4rD^Nrc&Z{`+bvpfshBX~c+kJmH)N7~~n|WpA%y-G%r9UxZD4iv;F?2lak51y_Hu7RI3>SWcOhp7Eq7kj|Zw)DTIGo>{<m@bX3l<!bFGWbqI>fPqAR4m3o8j&`_%VdB!fF37%>7O+0a&r3s$@ZL6IsL*=jBQ3VLVgHd@QCZN6^=jGWL|rkC_jni(Uuy7F7qsqSy$92Tc2--Rn~cOen`$D1AJ+PNRMY1AiU7k(Z0oAd1X~n9{_2K>=8cW;nEh@}z1Yk#(KuPNNVlpAXe279>1o6E~`^&hf#83(mqfo8)%$y|mf`uhkl^WpexBbYiwtV_pH<&c{I)37q0->OM13HzmX@=lc=<rQd6EE;Opq_F1IvHq?f8y-_iDhz3tD1NxV*6jl5-=isAEh`5HznDP(NMxdL~q?BOW=JZGS=HmC1R!64j<*w!X3dA5z(8ewC2R`bOysD^<|I@6r3LDje+xSUNdqOS9WjOgy5a<MCBFU^kAmI`r@t@+{#&WT~@Nub|YAdcYCvTGeKQ(D@#QPHrVg=W_h=v4sH}sfoYNT7m!E2H{>T|>-RQ(rlk<AA$&6n$Oy(OYOTJ_sQHd2nddaCmR=0EQMAV$O@JS$g<j(}xX$JIhC^ek_p*~mK?)Kq0+FzC(qYw2vxsoJawm9rU>=`7%pMg6`^a-6E_dfnj!N0<)^&3K&BcZ!*R^Ay?#Nk+Em<!I%yeFe9j8EWZS$y9alm_p~y_EE3jn_fA09O0*<E5Lnch#{XsL#3TgPwF!7yCJ~m<nF~7oc4q<3(Mp7FzU-A<r)6i+is*RO0`{J|L}O;skTM6P-@ER{3?p`_aeSsHxg<?q?|QjP)dPLB7BY#-D*3UMA+ltEvQ3b529kWhkk}AQVS$p8F5CKS#Gt0KgpuqLqNrf5vNwMRf?J%tt2Oe<p*=K1Gv}LOjwr5ux7bM(lljlElg{=Td(btFwjWlDqa_Hex}cd&vm4tqovSd#ic)C!QkmCUEAr_PM3E4=f2fpvfDKh%%todm@4-KjUHOU8CxVlH<J{hTcs>CD(MVek9-kxZL?m>EU?|G8lqB3buqZnIvI%!x5yOiVw#$8WABsN{#Nav-nb3nr58hy6Sm>{Ywm^Cvg&F&B^uw^v2yz^Q<IYn_7X^&a;Lep<^0KHU&~R(5JW}{ua%>1GDy$JG^JEDnW~$YNuni=isM?F{2y3LB^3t_vrykXKrDs5#|=aYJm6q(uSM4SnODSb%I9PBj3_0XA&@NN5;#V}A?UH*tHkVN5eO735ufQ<s<aOUPZ5Qzr19)2N-n)e_bL@mnrG?Lqc!o)+VG<X{L<Rsqk9W2e_%L=Po8XTY@ua&V{do2zHxwA2kIg5wA<{4cb!dGNKV31DHr7Ah_kiw=*U>v0|8SV(@?h8!Ay`B?MV-%Xfi5^BfS%$um%;(%D0M=Oq*m(*@vU^F2%B%yRouloRQrGuR*xOJj^EvBHZd&iktM*nrxoO8q8K^YFl0EiOS$;9mWVbUlKlIMnP?emZHluA=+A*sBD`w+unz((aMABb7UE_%))%m*cl70A56uZFwd~|xUskY7#7gu&%!5L&k;%Q$rD&ow_>ZZtGzFx&Ku0!b4$nD0gg3)MzAR)Mth~b=)|-apP06FVh{%;V0Co6e^MT_r<2|+#HS^_TN$+O>8KxILdh&g!EYZ?zAXLwf{U99Xl~}j0(Kb{C0kq0E%3NK?68lTta{l>_r&}@XY@@~V=JE!4Ed@`7F|(Qz_XMiEW717EEZX=Wjk$}rsUY7=ASD<f2$3(K|V0(vW$+`bS~aX7VHCCf(gAWCtjDxEIpFO>;cBJ;5mu;;@gzRCjFP<Z|Qnrar4d$>v?9Xc;muXJLmmrZwr5sqfOYQ8ZHrPKW=#{S~?+SLBFp8(c@-Ht}@d!TOFV6&=2B<=vLu1CnLZtzzJRKIVwRCy%r;sMQsequ4H4M0(ng=UHLO#upMzkl3?uPp=my$_&|BW@*{$Upgkmv@|I2An+9wo!PMi&qYH}Y0_l)Pj&6zK&o~eWAC^^%b}fT~%{RF3x#Ko|OZ4vAwJ|8PcDp@_)08S5=y#dEjPaDkJDc#14%3sElvic=8+%(q;b*iX6i3vMvK8o(rM;?QXyO#`H#uUbtGcYp<;a=##(IWNj<iphx{)XKb=^Iu4UFr!*JckLxOX~IXJu<4$uz1cFPMUt$IzrWF2ZU>WDPU9z{SLRpVmG@A(OL%qVJ(Vf0Ae+rYmDoJEuW_^=IXN3j_3L`7Wj(sz2YTKO=9-#jxyiJ2;0?8}@_wx&z&hGB0-mdQ)c8JX`Gs={(NgOFLewGHS1WFIdex@N;bnG)X^b2YGs@>=%j7O`gs_XxoKZyylZVmd{mz$ybhMQ_2>HZ{P$X8!$}q@o3VCjVLe+NJF;qMUqA8`kTYugwDG1mufcIWzv{gZgZ>lbaxLEU^OxySe3QrH))-%`F3HIt>PX}>W<>kPU791s`lD<+UK1fI-azrQ=E~c-1lpyYxBEKJ&A~8%VL{Dr<~G9cgt&@FZ8y@(s2duV9@P56o2aIe#gS8@VMXZ4a#~8yQASsdver+fj}Ei->-tuH)wO6Fes@PMn9SMNpz<-M8_DsqMnY<W6B72d<>Euqoa(D&&p{%s<1Z0whTKXL?YwZAfJC^1Z^UmP@jbuNYK=O@GSxz*69xmh=Ea>+hARaZ0|jz1ZS@ios_yZi(ZiT1<CjJ{Zh@jN8)Q?OOwsGt*oN3N#jviemUP+vCT}}zo+kq=;|&j4r~)Z>dlhNbBR#}Ftg^Chgv?`J8U4AE3Rk_D-uf~<fTAC9sGb+g!UWwAEzatoCv06IMs3TF3xr%9hkT3pR5I`PyOs?^X<&h$=k3r7g~+c7gpH=q+1Vr&Pt=ORcXiWJm}Zr!pL~mLP=7ME4a2KK`3COn}<{@&#F=zV`@KV8P#W2Q#1DqTXND#le*XnlD=Sm^^?_>x0HIc)<SyN%17Dq5tX-}?Xpc%d7hwXvLM9<^EWc|Q-ZM&|1ZD;Te?uuOe%XUdohbO#;=sUe_8ZU2kGH5EHmcX-6v5u5P3TE40HVHgyQ1hao0WV#n7)LJhcB3Tl)MR$^D)kKBkNkD8Fsjo__vL>pHj1<MWBFO(McuZ!C>SYzrC4n!`=Yd7ehY<!dU22>|DizaGP{HQyz0Y66-78dFpjUc0pY#V>&v@?~x_8rWn~*NXOKCLd*PRZ6O=KYzBpw_iKh+y5-og8wOazjoNDZQt64t_@Ad>Zjh~PE;D2b;^w@HGvs13`Li;@TFI`py>57^WC*6yR?VLutd-=mssW<I_8}J^)E_vNdTkbEV?2SHG~u4P(()1Z?ZdLWZBk}ElfY<?v5N8YT7|}n)VdKz8sEI07;OrVDT+TWYG)+GGG8coo)n>fO($1V$Aw?RZ=FzAphiExl+MDRC+P=5b&YQrhwP>S+;PV)%SM}4`|mB1o@q3o6qiSZPX=VxzjkPJ>J@eFO0VOeMfWIh>_#KFs36k-hUAd<;@sP7Ic<rF4TyPFJ#cN^Gw~KbNvQX1{ni1+<dydw;@tsW#8GH4pa#iu#gDZuYnTThg%!<oOU+0a$YD3=DfGTtE&<kD7)!1p}F?qehp0@KiifL(-$=QCP2}U9rt8XAV>G$>$ZFSs3R}xVT8^x_ifHGDVaWEH!+N;E)D#de{_--kyuYRP)=m)BGb5*Pj{UXGEpSqEMFNczJp8m*5*@d)>2rGV^-1;Y~uQ9)b5Di_;Wflu+X8fW^$N$5*vMbR8S1YXH9d?3FO<uV0y$h3xYJRVB%}NHtbOub#q~EEf+y)I-@MI-qP`C{ixR_FE`AY#Qtic16&s|Zk+va89(_i$0M8Mg`8Uum~u-k!$Cdh#X@QnVelbkQ9FZsXX0Xyqi<HokSq$7rQi~hYWqUD%za3v1gxETcG6m|w&%jf(UId^?7Owy+V*D$7;qM{dce(J0`=!0w}I?duQwK12;|zBEQHxY3*j$G#@XzR=m=lZ?nWeH+~8GU!tlxd`4H9K<ozW4(li}Q_+)QC)Cp*n6~_G_v^awzC4G7Q8%GoP?-3mi0lBP&<M#A4R1g8Ss<V!p*qG@!S}+H4KO9X)?apx<gw5%wl&%c%t<ea@6^gphJcBeYt*oNu!RO6XusYc(RbQe=ukcyz;G<AQl@~11=TF&}*mh@Rt~)z98ja#<_I+m%h<D}99kTGj9ZQm(mto%4YG3NKWs?QMyisj1gPw}==nP114@vmw9kF>NpYp<m78)@9XFXr7*yUBv$6<@35}~NK9yh{I_xIrbO}?~6%IWrHv+5JxE&qvsVYMJw5k3_ySE{Cf=*n%ox*%;-%!@&cw?vEW|4>>eZ+4rn8fA5CPia{wr_0f(e~Tq07i1EGQNkLK3~U^00Bt^@ZRCD?TFTM<<q5hVk2U-diOagnJQta8mK^TN!5?qTUKe4l_IWT_UYfloF5&r1HqBl=$zKJPi&<w%s>FQ5nj*!YrG?7z*y^%5V6Tglrb=3Fi>yR6D^?^7BnXb%TGCgz5>~Y!XhAnS;ZTuwJ}8aHzL5J0hbAYO6ul+!%MzY1b<sMi+|yM;7#Ho*PX!T&3s7Oexo^Xx)!vDdWmoIN$wK`WmRYUq!fwglPMe22J6j-&EL0cim~<jkU{UStJ>9Q8`)F$;<n#fb<<^x({Lrc)JDzoh-Mxd4>ihXi7JG)lZ6HWa+agpsDn0i(e;989Tb`kLiEv<7ky8h0h=mUZ*IAgJsMRvOQPAhX<|r(9ua=gCU4oud0L<Z&Cm=2&i1<M*tij2;i@=h=;|O-q&E&!PE6k?iXkpqY_)_?A%=vHnLXW^6I!@F+J|&^LBd3OO88ex0J|*Hx>_{A-!-U4LhoSRfu(l5Dne&^72xz6TK<k8aK0H3fXyJ|}hgIk^NTteZl#4%X5I6;BHy5lGGa6|;hQxZdT=^Y3Eis8rF0eFw?L8`r<Bm_yyWM_7Nqcng0{qbGz9%pyI_MA(%Ud@c>$GIxG@_%o1*?xriU$1W1!Xmi1nVY(mI@i8)Fwz|4`lwCTq3;=n@N<O7w&fqmCn`^;Xt!~%D<H@em!M&N!eR69_n#mo+NOnAy7PxJze}=XOqJF&>D)sF43%gJeBY+)Si>*#3=!kc!xUO9z#lzM8>@*?OAH6hW_*CWMxPLqogrW(p|R365bleWck!UGI{>II1Y@mLt%`fnVw#YPV_}J-dsGC=P5W8J7ZhNoRdKYJ7p0rc+@^V!>u^%1c>2Ep=+g~rTdR-8y}-Cad%T61jW`c^#~6?3hbGKjD-TS5_F9gON@boIctS|SSU_-*wH$gSsd84)bdD5zwt7W=Yy%53s?t&O{!-iU{a&d#@_bf&ThE5^+cpEax;Mvbk(B7qO=mfh$f@Bl#%|GX6^{X#gc-$Hq2WeorPjRK_$iH+XFkI1c{9F%|db`)D!S~O?4^2Kjm{NPTt0X`22Y|2%e`LMQ6ETud$EjsSAV`AgzQz?J`xd)Eysdrj;=_DmJ#=y{LXqPVG*8=W%^M+^NC6>v{f#i`l(0??)Z|)RR5uIa}t2lLT<~+B!YVoM&7?1+~*sI*vF0qRDk(ioG%1Rj;r@#1Iq~ZKKaC+@@DdH?6Wq7Cv{dw_V@I-)QC2NtoDY_=Z>*RxBkWBeIxhrodfWuK70$D5e$VR`pK1bCVQs6f!idp;3rGe3v90w_0%hnOlExnDT9clx1zghd{4T)ubwNVX|Sl^&^3I>Q)8B?VJ{dazi3`hS*d@B)7`+O~RrVx8Y{*Ww<VoOI8hT3PZq$>~5+^+*%H}?@MZ!bFlG?$M`OW*&@WQBedX-QgFu!caNMGcrQSw+7X_2y@@09U_g2JzFgc&&LuFbcrQ$1p>+7-JBm3|^&*yX2`{y1AMn=aSA)KkqNYdcpvugFEOINtE8^P}O^3S(U;1=+>#q&FBHsR`hXt?{G2iw!=oD2Q%4N>>kF7@6H)Jx%!<k_e!ja8@X&|P`1$uxZ0}u%xMVi@!2a`v*Wz#9Y36T&{--LOaERiObgVlw7DgM*bjA6Ow(J2?IS2iI1c<1?gufAVt<xx1nZV)1vfNi)#N1~-7-=(w!Zq_oKOxE`Tj^ffKIi)8`iG@O}uqYeT<+@!BKP!<Kcq18mUKeJR3}Kh4vD(F?2I<|Q%MDr7;_qgqxu4e2pn~y;q8GM>QLy50n77)ONsB^SY8>ul&cXuGh&i!E<#9T)Cg~AtXx0?9C4jm@!9y3u5AWw}Zmd?M`t6h((WHk;c2$?hAiK${^a%kXiv@PqvtB;lmOxiDle#G9HV&WF_qXe{{auv*-lkMnD9u@DG)Jt<73Z>5C30P2772CSUzpImRx*;{YS}N6^Q=obukzaqgDQtp0)#m8ynG1kENy#8TMTJ_SRgs&`CxqqY1ERibUdEhItBtMby}%Z(gYZY1+}NL=-exSk+qs?(WA#ygPB!7r&_rWjF>o$yU6YlKz>KKCEMPU8bBP_bJ|RnvVGyxJhNLXo>{UG9bk))DgkOr#4bGB=-Y(5EW)|7sz4ciBspA0W46y~e-&xT*`f$ckDcLBs20kc_R#qu*8D@63z^*w$r%&lY+@rRGNw3BBfI7sWDITS^6JBkjvKofWLA>Mx&xy7$Gdx<?qU{*BnCEDV~*y6Ci;E#+tcA_h#IewXd#Hb1o${XZRNM=h!xF#@L{05-c@o~VLWVD6<(Mg3u!^YVx^QW6rGT3dR0;#$~wCoe)YoSSJ*->%j%%x0opTgl21UED^+|s7WId>K88CX_7YYFCj1<t5P)`#=2<K8z{0xVVu$TDUl8syodebr2P=eiow4MmkVS3ZdZzQ=bTkgnNRND<Nkd;kpHKRuqjo=p_dy)-!8bV}MZ74|T%m8PNk+Upg5FLFV(&O;oWZ?PxUkI^VoAx~5WS9CNp~2x(l&{?_^3s0vTyecu}CP?Z3Hic{{RD&`QF0bFr;T8>ES6WrKCww0~8p5apFlZvzV}z>>R486Ka?j^ANEG*ANNaBaxke$ZMc;sZF4S24<$U-H$c;?(Xf*5gy)P?O2_mza3)j{B!j!ON&%xfAsLaR+GZJrWKF1_?2>H?LG>Mr@i?_AAWFuvDr1IZx|*S2=%$fa|vdZd2?p^-gS0RKwm+x{ENs41}!_!pFSp!F23`kU;cSS$rvwDnZz#N+6U|DsyEtGs=$poXN3p~;g}!vfoeN42hY@ZSYLiVY4;t{p*gT1VL6pBZUQj{XPAY#H|<|}<!@zr$v2)L#&*gkQ|x-Vs0(I-1&;2QS$9kUsEW+}-hJup)LP6*Gcno&E8^_-1=bC__a^s?ofh>Oo#VNfGGQb=kPgVzmM{Vd&pV5Eryj~sXa`b^CUxwyh%iT6OJ}zR^#0<7-kC*oy|6l?;mPz=oN{YRsbfDGJZ|F`{@Gcp5cgb<_!L!tSXrPSS;m<k+4i>XQmI%s+8I@~FB0hQHbI8R%@;3{yH0o{9s)sb(~KS<CkX`O5P$hD2-5^iNe}=CW`XWj4Y)1Tb`bz8d{*1q|8%P%+%Rv`@=yWJQa&Fwub!``lfc+~w~G5CPbmr4BP%~C@}y1NWEqyKXd7ERX4z9#<F)Sm%_|_Q%EL}?5Cm)F8$b_K5J7Nvf!&2968r0{SzWDES_>>@+9zZ^!OzLlQ~g+NEw+0qkAfF361xfYAD*3^rJlWb@jZC>zts-Y5XAp-TTCb*aSGPv-sV%`soIuQ9@fGoy73GINrRH8k@}XKb9S3OSroIXizkGmp;-{pTX^AERu+Puq7IW44GPU)cWczq`O#I2Jh~>)ipu(jix(c5DgMfH^HU{3i853XQQQgdeQ`&vyC2*(=afupC$wIK5x<b*ROk<8DW@Z@>x~_bjv`=8k-S^W<aG9&S1TFQ%^(xcBRpnbo>X;Y-huLEBKLla`q~SEla_LtwlDTvPinu&E!)p?SN5(g9@~Dmt2w3?f>0o~Pq$m(BHHeg`V0e484wNkpFpe4!!d+)lSp8;HClgGuN~kr%~LkOwh3~)qhaxg<<jl?hIVb~?&>l$od-g|zPxk@m9VmAx7(6>wv~tYzX#s;g2Bx8+$*<-SJFl&GR@=&11TcOBSbWxpV>eZs0xD1@7{1x;{dH<sf2CjR_0yUQ!&tQbY~a#o|B@B?p*Ar-95w$`>3}2v`*oOA`XDE&9kpLW!n{t1iDqZ#xUBvRgF<Hx;~FLHacP+pOpk{gfh?OVD?20q9L0Fc&>DHZps273TV$3I2m0L*`v~vSMh;>oe)o$5!%i~LX=1;?(q8+{=~LYa83eptq<wz(ieWF4@UF;0=iz)IXK(sUvlN8(@q#!J@vJOEJG44y}$F#i^EVIWmI*y>-jpe2mR^0$GxyWy2Kk#d)k!0T6|q>PZu!g;5VnelT)n<`f0C2*C}!N<U#6-N?wD&WTPpGIAUmO$7Id((#-Q$dms{Kw=JPygc{OBuBkHvyqMraz=6{zG;1RPYI0q&D8mGl1qx+K{$fxnhoKxtm`Jxb=zS3}2pUOcW*;c5I3=IKVKo9a9<%4EQ0>lF`g_ePh8d<IOeyn~Odee1V~%(fIw;YtUCcv&HcbZ#XwAQ|Xr|0JiZCLnr`h$rXZCEKGWd$GLajw(&62<|#FfJ}F`H_x8UoY_k8~<vyiD-k3#6hms3R}=p<bH=`_Dl1>WLA{mHU4A;e9_#m&aR#$#3J%AgssH@C{SS&Y%alh*sBp@pDc<MUt-W!c_YcA@any*({h%RsO-mTrJL_3vc~uY;mQU&9ezf^p)@2gpf{o$BrKg_5sfZITVo+DQon;8ZTm5kO`)f@C$MAo$9u7-r`!g1X5zBwomEf7jXk7?~fU||I+$d#Dhb^+I-j4COYU$hwqVm{Ps`-u*fvr+9}~`-0aN^j|3cWhSK?lG2tl4Y6@GV+ITV|^zxG1IKXQU=0Ww{oB#RxCpTZh|9}1Z$N2BfH#c88H(xui|L*3?n{VLBtJgoe`P1uP+<fKS{03hB0UrPM^)DQF`Xzkv<JUijFQEo3lS-0jm&5A21TmOqvy|L{OpeJH(sxh-8D_mF*(qRevrd}M#?Icyb=WCb75>w#rW2)-Z!+hRI7<xd_yB7=g9Bu4K5^vKSkPefwZrGY<x`Dm1Sho<djJqHJ9|g^x?;wc!a!Q3Lx#h!Ga5#j;2`WubR~z7V|%hkC+1TbVcA>WU&?P78bd%6wR`dRs*WIz57t`67FOZ!sfM!mNqxVzjrxj|U+}`-zlii)==}==bZANMlDsIsh0c1p+1oM=<ah;}lWz6lPA{#bxVFZ+?#=&w^XlfeH($Z7{p#lT4*dA_>;L-tm#=?(^Cx)lr<+$*=VkKg^_`dM<Lh*%(k(Lc_<vFre*OBF4*Yoa`e(==_+f@Gpdd7dH!R*7wDl8YORxVCI1#nxz#C*uKY=d1y7@i4gn#ue7;7_?AT{IR<=JrDiatu54D*`IAVXU>i`H;*3C;N~bxXB149q3xV=2aX283KrVR;JYj%{PBXl%miBKK?M0a8?n>4uYAS3b03v9-lOu11)`wW{uT=acCP!(Az|^#swFryL6_?-mYX_HNe&n2kz5g15ZQ`X>f%^&Z~x5@{f2oJAE`t7R0p@@0TEVF;j>p77zxUk5r!ZVktg6_${?6K)>Xx3frn!pTm$?h<e%R*nxdheX4(z|>Dhmv{ta3sZs-%VTEB2!GA@bIkTNMU@UZ4p~*i0v6~5ue`uPp+(<4BBcr3m_bD2_hBcx2vJjn87&a+CSUQ$9~5B{oFB!}G|<RGqp_U`dKr1J;ozW`>bLivUWzADmf>F(5b#vp;z*RAiJxZ8<jJD7FA@Le*Ur#LgfZSsd(Stv6&z5@vtNIiUILpxpR8X5qTsl9F`9;yURq>$DoUuQ>n+Xndij!pOe-Ud*uFp*axr~M|9q;{h3%7*?)mUIa4D&`>zS!=Y+%&5<N}>2Il$UjPG)KTLRoEK$C&iZ6e(JkS#2n{<EQ3~4p>9yxq#JESBv6xM_*>kXl@uyI;CUw;5`=V6myy9==!nS)m(n0<ifo00wpc#Z6z{evWYagG2Q<>g`UcZQPua$JoZxMN4dIC1<4DdC-D66aUH?G_*nS&zyBvEwZX_Md@`U632v2iAB6!fp*&MUfmzEcWX7yzleE9IgjB!wiVuXmlQA_#Ugp5C6-~2lTbZW&gRt9~gh!`t%No?gxHho;V&hcNs?b1g#1cs6-Pe+wEP!ldtsWUYT}81Ia97hhd~^!esTb$)E@wL}RN+`A|Dm{_ZofA+G?t9qtZMqCp)a2_<g|4j0<teIa`Xx-qQI(tR@*<=s%=|fs#)ma93R#u;r1}M&ClP#Mn8W?wtBrTA`jYqM|qR3qmbl?eU_66dMlB~P)klS*Ew9Kx8Z;of;e;$I|#G|YXWnM>{L+nN^*?g4d;%qHmsAQr%X<QO80W1Y?May$%qNf+(VU(<)lSOaqE^XSBFaN;Gg=mrak{;LKbatgeJQ{F-YIGA*4)TH*|jyK&@}_I8GnCi#Xf7t)m&{OQ*6rSPRbF+nu!qb7(nn=UEvu2cv}2UP#AjYB-Z)jQ(2aVA`pK$#@n$mI{uQjlCUUAX|<63zr;V!EvHhl6#zJW#05S(W>|^juXB5g^rVjzMOMz<kJX?_qxFRhEw$4oQy6ag!hLY*c(>%;qu9}RIyzite>&jSbSqGA;(_sLz!~rzL$UO31vwBdB^YhwwiuwIq2H`erf(WCnCWkM@Gg@;=nxEvMr>KwfwHKK^u-eH`6ep*rQc*f!0u$`7K)$aD>F&oQ<vYxz4&8Jf98sd`l@abHzIQOqt~p?Uboj{K0uI!lW}e-UQiSPufuBpbw5XbNx)~uq9vs)>o{D%S!#|J2GIOwA5ExH(|+bK1;oZC-5Bg^X7ASLW>w1X~93&*xXC}_0DnB?^CMHcdER)JEKd{op*X2e0X#my}$bKVOKoF_{$D(O?dTEa7VKkKD_3#45SHfn?fjh*LhfWZvOe^)y<#b|NlVx(Vw91_w_$W1J0jGCHkA2ui(GmppoaNWb}D;^LsSoz>8OG1hNTTH~%kwDh)z6U$XA??@19$HY2PYtHk=D`L9_Y3)Oz@kk#m4xzJB<evNPO8~B>u{t4gwE2^p;uHm;({158u*FT2;&?s(RIjGKk{Ud0K+u-q_rWuC#`e$;8-{3b6boUpM7_WZ`qy7mD4;9o75$2Cr<llfU_)8E1zo6$*rTiuS1iJnY)+s~dzkL0FLVtem{7v^9&44S)vlh5(dvf(RFov7o!B=0&MxC2q<CuWh_#sTg?}&asPjuKsI(|&*>faz8^x2XXq2627j`Scs_dhs5G5+(XFkV3kqUvwaw)gs{va)~0Z{T}=BdJEi`-&*_TNom*kAH)|f5jz`ME~ve-@X1jDD(|f_X|VE4+Qx~rmXK^#ec#fehnoQ3FUjD3~B6N|NQl@prt>_1y|PBnXyQ}bkby|ekKd?6`CsX&p+aF00I!9Op1OjT23bBS4=)MihM&mh8E(Vh0Wp*^aYHPM)(_Ahq%hH>ZK+60*b-VkT}0j5f>LQt}3#&{8UV{XzP#49`p6mQbI0Tdbq;B!4=M%1xfG<-~AZ>17CeDtsXaDl^5VQ#~^+Swi3;vV{-RlXEAs?Jn#2)KJ@;*hn2PN-G_7dkSeC}q0#wtd_Ikl5w&tSle0>hem=QduB_o7`}Jqrwa->I;K%^+v%Y)aXqmN~?b=i4j`M7P@A+p=18$K5HU^$OIi!?&8@1iq{%7`!;S;znup)f`f)X4iAU!#c>reLJJJPC0!&o4_z!h^erVP0kz0rA0-K#k9CCCY*OE`#j+XA^Nl?yUH?R8P}k|Yo+Q9^4UuR~$KeIn!NVt<Soo<k6}CKI@QQXZSM)cE;3_<#G+(Rr`m(eYO)sh88f0I@&>_d3eez&Vd8AC!YL?o{3{+lYbx?9sK&nc@cw*eU9z8<*3bI@6W$`B5KH$SC7k`=spD6eL<w>a?8-GvrBTTSPAg-UmZOEacJTvOVbtKv-xGagAXLmd5RPx}@V9vx8f&o3iOoW2b!*IgbJdZSWEo41wwEyY)tceD)e;m}iRp#YO70;SYvNK(`fina*G(9}fGyvxo_Rt{M}8pu~90i1FsbOBlzXN67#sVcpp2Vuj1tr;~FSS=lyyHqr!?<Cl;!j8!19_(e{2<6Y|J(q)GTTxFCv9rYFxohJ6O-VnlzUU-7gqO@jWl<||%6um*qFxjL~C~?O%SXAC=Y8E(2K-kzj+}}Vzd>Ef=uNr$IYU`YjX?aDRQc9(etOWr}EqviryP~1Z!)L@C>fy)r&oU`r&W0dVC9vYfEi9u`JtdbHIR)l-$wokej5;CMYj$ocSgTfFL{6>UwcH@N3sEO9vsBV3o849>S*1m7L8PGm2lQDvJ*@KOX#|rbo06W0mb&_|YXhx9byeny743Yase}`!9aFOAp~tNmsCK_qWk`Sf@K&(bR%b~fVIVjJ69LOg0=x%xmitjjOs{8MY24-{LUmV3D!X20mrm6cyt~h&vp8(U=lyBG+Y~)CRc+m_nr67Sk0vn9)@8^wG}8^W0)A1}ly(xg1%cXEaM{Yk%HAna0j#}~h^NON4O1$?bgTa9=rVBoQFltVHcBNJxC7YIAuMFzc&<LNxmf<(w7~Kw;v@_Ky|=@S+Wsaw8lnRLhhk)~<Pyks9ECyKbhq=fU0_!0(WKoE$tCqFL|_9lvy|*-cw+#P1GL%huK-u?v;~YPoVQiy2wf>LC=ZUwD_@AT0`{MLzwmg}g_E-FV7Y=7(Oj#n-t*y~2l&rJ_-Czx|75(jM_1|S5DlTK>0;<#SaVxTNaibwHqI_#cs^Hu>poozu`;N1wd}7}%D!;%zkBy?*}r%1UOD4PMYFn4?@YyWwTYu``FfKTOZBi&Wr@~0&t8x#wXOPHABbgNknEeuRg^`Y9B5c+!4k#Mx`(@4=<rGl6xJr5mwWr4!XgcKwi;+5Nv}?x{Qw%3PRaa))P3U549kXnh=cc%!MB3dXm!o3y4^)|pTe<n14YB7N){O{%j2PpnFzP*wU0A?V|apSL&e`TQ2HY?%;*OtT-)X_*HS{4jG-$WC)QsyN|{u{GS0QsIG@!T2gxW=3fg<zsPBK0Wjyh;-8W{I9?*FCuZ0F|Q)7$1!=~xQa8?x#otpA8BQpm@VoLZ)ZTpY~yZZBu?ZeINkXS2AUmnzVp6%_UcN<;cq-NH49}D(IhqdqE9Z2!G;YE8QqGE{2se3%CZK<a$SnKT;@!mXsJ>e@<nArWolZptW5}9oB?p_K!-RVSw(aEGeKJ6XjC`4ilao5q4v&dYqdg2R_G1kVp;gTGPa67RupmvUf$w@AqhT-R7T`ef+`fiej0mGkqFerLEjwa!1rBcBpJ2sI*`9urA6PQW(|K)o&W5+niX3Q3Du-^JWyiA%_?{f1hiZL!ET*2l246l2Z@m9yTi11JRDzY@H4s0&KIG6DMrdZJ&`-4_SYO{XaO!d7*wQ1+v%2|G+ERl&^XBm+!Rq=d@Y|-JRYW6a#yUDW+68+`$6y?ejWKPFuVFFSvm69Ua)q+xJ>(1?`%$ZZ^@|xFLS-n+Xap3lsWf7QELSVz*u<Nhze_KM_$ros$P9e>zsLrR-@cJ3B(#MBe+nZ97<-j?+I13dOXV^~uGW-XksekJEgqDJ#65GntKgW<szH6QD1*s?q8N-$&V<n48BgcX+^%OQ81{$tPY;O_luhG;X3o=c3elr_>X2;cP_op1gWRngiDuI$;%zKwtl@ia(X#;~RbDE%>Oy8;ntPtZM{Sa=e(<Ck}K?U*EKz-RAIoT5<L{f7?z&GPIhzZ#MNinQ|vC_0*$XI^@>++Mw-R#DDiA{a}PR@`&f2VbAZB`~!C}<n|4bAfi-g*Z#CG(p{n34Y-E^ZE(Q~k{GJ9LjvqvJC;$5PtYaNIwS!&P~zN!oCzw*1v}@n?=lr%O@LjIYO^wNXdFU+96~soGYwV)lwNq0b6T58cm4=qG}5-LyC84NunDD~m#_5O2om_GU1HwgVMWatEV}NafiBNpd_H&3bUe;ZEQ-dT_hGLRKikqz4Grhwd!>Xc{Ha(}SpuoEN<tVoP$4E?f>ZFco8~P!z8o)`cJ`A)%zuzjx{1d-U)9(DgiPrlcTYGI00AaS>_>cbxvV$Fx#^P(c!PO5LsxgbVuZ!BMB}99O54=5Z1*OlE(7g7Rk^F|<t{z<bmI!Mr^bN|*d#Oc?YPgBfx!3j#7#Ab1n6G*^DNmBbepqJC*7;D`4sB>roVB3M_85$DAdI9e^}@&X8WV{d=IzFkAWX0jB88;1vbPoB8*P{0z3Al=rFCQ<t=7Yjudu8N72oS~Zw!G@_%1PzM{Q{_};PJeM##&edc^kEY<wek>i=N=`~Ny1m<AwnIv#o<AVA8H_{x&MGu<ZUtX+~rfi;T1FInFtd;779KF!Zsh)1^$q8n0=P7OL7--ZGn`Q%k>wpB$^%f&*7G@;fVkzETi$PB$gnMB6CP+?zIFpCSWlzjz@q{WDbReBTZR1p-;PjQG@^T+`^xyd?5f4SOkI2&HVD9qOUNmdQ)eaUT!(Q$_JMt1Gt^tM3iz^58Hj$DXn^rFnh6*1yQUfqt5wp6gx{~8tz{$ks*`)Yhwo<odZE795Q388bM~IHBZ>iDgW*M7J4oTEgF2^?@J8_LfAV~=fl83n_whFGlpo4Cg;Ogua$N@^rKPRsVAmsns&4yBWpcAMQi6tG&~1}$Od0hE;_8%anKcYsykc@-4y1D+TS$e)Kc1crees^5Ip9R3~u(>Y*4k{q6D&tESua2Sbvs0H(^G^LxC>RT%Fsq#M8FK;6^?T-JeWH{h(6z+xXwrXl1pM@hQhe$tevvdnJFLPGQMvTKSN3Y;r-74v`Dc54){6X{ef;e4d?tn-qsz5BK&hsWDrO`B{R!-%Vd#`=@ardZn;OIAyR%eK~;w+)?ZMv@UWAtd<Z@dhv086vGYsH1?qmKl%Y|0@%+(I@p=Y_z%+=|MBcf<R<QWJ*LT|HeUD9D*d;jZ3(TPfZMmSDVIoKl|X>d!xiYEw|+7jb-3$<TghpZsHnxfoJPuO&ywOztf|GGpQ3yCL^^zyt?T}1H0qbcAlMhy^92oj)!=ueOOTRE5wL<!APOLG3vi~5z1@TQ^8?q*(Jvg2NU;p(mmwyZiUvby@>ptuLgcz2XSihrlvKw%3Ads0Vu(wc#?33AM(vK~0+2F@2oy}KrWGl}8sVatNzCjy=>)gBiV8e2WacKCSS~`Ch$&+`KYIv?PZKKZs9I<**gxELuNTqCur|iH6?J3$&Ac@G!7NlN0Xv%)$cxsi<dYthL}#hQ_$m$?5h;h~X0LliNT)JJQyhP!RxR6DVV((5m=}swJ-$Z7+{*3d3x4U-di89S5?|-giQif?ase^A=Y3HkL|`lVH*+4#F_z4WFB`Uw2Ul$Qe2m&=;Zh{cI3&LWs&U(T6eS0}n0QqvHfi_*dJy10<qjMYaY^quD_g^MeBA2^_kzN@f!i?$7Gu>1UKBFU-LB^64W4wGa+9<Ik~Tv_m}3vB&HC;EvAmp$D0OBnj-v_6QX+Y#=@KQX+u3UD)DAX2Lh)6T#T8|m)d$wpCRfQ0(?Gv1Nr{sw&j<Ftl8bO&Rbt8HRwRAL#&|sO>7I1dm(2(PO36<TQ%8RR_L3yx;VyA#O!g$sFrsFDwr@I=(QXOyf;>A=P>CWdv~Z%;i~}0`7rk*wGW*2Ki63Z&AKO&5r$B;`?fRk%R_6fuE=fB!Xpc*x7q*HA?f5Lnenqj0^0W7})+H9L&3#9sk32;vX*uOMgLZPEYQ>aWzWuPGyna5aHDKe&!&UA8i=D(!rIqHIKnv2%G19Uij*QBeztNnIqw>2p7Yw5l3Fp7X*SSs{w>@oq(WaNzhW*gtdrJPdGqD_V4jp=FW23?oU>eS(;3>Jg`VFkve)5RDFPgS;A@0;XCdA+8@}Q)XR{=@PCN~n@5RJ`cMRSkGVaRS0`pOR_r?Kn`D{qtXTVipG^us#w=S>Euit3YMt2=qGfz)VU*%1(A&*z@v>2<K^Nv;}5ALZIYvVO>?bC84-Y6PGwv^F#mD@ez<U_P7mXZ799yd$C~RkBHy!e*+~TD{GwpT1DAmADZ;A*4l&YS^eX4vgbJMN*pxu&{oO)YK`lCQ#CUHAm!;N}R)8%#q^e?SZBn$qrD^3Vff@%&p6!0=KOdC@tN+O2u+_FcnBr0nD{{xy93fTxKnn?@FKU?SFj79rP=ACA$*wn`K<xa521`;G&#HvuR0N?`Fl8<&J^Ms5@Bm1%N6J*n@({aBCT!1n&8?yYj#)pO7w8xoB}quGD#^r)t@SHIpW;ja}igG#@WW;@groum48yud0;wUFzqYa9<P7JrfMu5ZX_^CoW6ZHR-9gj*e?F+Z=(*hzMO<+j)V&TDL_^E<^1l&#&{4E*b~}*6mHvD9X+I&gkd`gMX7jG)7wsz9uJQR4pTEbmXv=81bVy$U`@ZK1N7Unhgd8TW@Urbv=s^_7_qB<tci|(?Y3Z)~`ZL3m4$8ygURcZZ}c+F8SLPQ$`S;1^j7NN#jbgqTtvYb;9%EG4N<qW#EXg9~RCTx4TjBxA?N0`@NK?NmP&y`Y3eRJmlcpZV!Y4s*?Okp@keME_lNWP|KQ19@VUmqV*vwG}65))M+E#s#qk3xuYQSD*n${S#LQ}rciIe{$>>w{u22J1)=5G7z8RI$6`qiQ=b<Ts<6sIJK}D#8W$E~lEE9_$*NV)wURWyT87=$BJj;hL5F~vz(&c#?8(-1b<-uV`@@|A7-6_H<+^niUpzb;Y&=$HMlvOez$Uz$7toZ{qW)^CzZhsD&;4L}3cT@j)bEhJ;Q#0Cy_+1#u6x0+0thI0)WDm9Hwo_w9%wcaE>Z9<VGD%{4y|^jc+?}Qkb*0T<kO4Yzc!*d!`Ac+$zjCm|Bacipii>bWCEE$0#!ZaJ_Xf-0&d>C=G=3S{hb3Y`r&EK8~rCnjo}mH+XJ6I{HAmR4H5Hdn5%cj6pnpE2Nz^+D5Vd9G9G?<qkHS*{G}H<EKKApe~A!p6I`6@oDLNsf<^~0-!>5b7l7@vv7*O;Tu_=(GS)Hj*LX=g?!d@pt``HpKq*t8SG!O@m+Q;G$64m2H(!a?INU}5Kh$c6_aVuY<xt3Jn7D(AM6`R$5e%b+9m<t&P>Yv<^B8`FnZ<E-_=s~4H2m!~r<e;s4##=iIix<x?>O@0@wa2G<rgQ<)ZIm|(4r!0KzP8BuG-_F7vX|O4h}jgr>q5!ZwOrxe5sFjA8>u^aFZ`P*jW-d6BC8rcq%IjwRw#|*um)xfE{ZY{390Z6|V?JHxcZ8?+}3}C#@w6KLby-Cluxmz0ni`!6JMu`PppHJR?M&$fm<YiV&E~={mh`9(iVm=Ri~xln6QVIeP<7B9(|D##=KCUBF)aCM9CZ<idpk|5<mbv4|7P2<b`WO9eSGeOed`vlSi9IyB1Mo<dBo@q#0k!Z1jbxKHC^uG~?%ao`x|t6LO&gCzfJv6M`1v66>}zuA7b7ri3R=lp$mui-lKcK0w@^giyOUZ=dkh2|s<g55my2*~kl`FTybZ#&uVT&o9-!W~#RnO|-OTzJPCh=?2<U>2=`XwqSWdr(?Fz8E#4<`!o4Q^Sr%)DU&RFGJA-7|=P)9?%GUN9<wlX50y-B*7)b@I#p_!B*{IGLp_%AjDTfzy~Pai2*2r?}HFW7!a$?ATNP$fn+|0>Nca(F@SzWf{Z3J*>sUe5~(~7#qc20FR=DCrXFyCYo;Um5LwAxWfU&>`l+21B}+|MBx>%|pv9>-ZnU8#a>2*DK=eyrL}21zDzZIt&gvZldj!`^BYzV45(EktV!uxi0$58=-1jprx%&z`I)!&ZvtZgD^<zQpI6x#4dL6<SO91YPY_6y0jcJpKW{5;IMI3Y1a>tQrL~Onp6P%^t>(Yo>BHoR1dU3AUC2&eUdN_nTVWFaqZ-Jtyc1Se_5LSXQJ3tu>hEuWOTv+AjqFVA|x{6Fb4}FN;!lr3#Bw(Vz(CAMWn46IYjaFbz(7N3gLJmJM$GqM)aJjt1|MEZ)#(WO_p?~E_5hyqQuL%@^Z+a<C<hulkF!3R<d!$0MAt1bpahwHIzAEAadLGycUYwL0^qQ0Gl=+47U2pVnKPnCbb^U^4j609NZX5=FXdy6X)@S|QCw+*PU_f~^7vnSFvTK;3T=umd`r{xiz8qA*y!=u?!ACmB5u7W>HIB*q=ov?Uzu*|Z=D$Y$D`g+=FUJomU-EUcZtH_<8+xk`%XxaZQ|Xtz)o)%-*j01r_hY=tOy%6a!l}9DPiEe7!J~|SebFC%ety^||AGUAwXQ>~BmS?@`C*REMF;j)cTy$paM&DM{pnAaI++ZxUm1@I3yWMAvre<spZS$9IFeZ}@(zaZojZ%MNrDr9p^*!OV=-v)7t!%q<)1wGFy(iX0km#F34{ABv(T4t>l+)AQ#&07v$EC^);h1#ZuLqO>Z&}e;R&z7$$;aM6-*^#4kjGmxdif}5BJr6QQq30mY6ztHD6I>$}C)E>Z<(3@wi=d177NY>ks7kUtjC``wru}^vDH(XrYcTLj<4APILh)_@v<pX7I_cf0;9p^%btE-a{Aa-GC+muNtxv`X`Kck*kPa8s-JBAeTuqGBU1*bn0w<!>g!qtbzz{K%nulf1|XWHk4W?bEbGS)8gc7<)N4{^E>{FuldQ31~8Iqk%7nM@E}+Vh{ARP#)Ng;B9TS)cu)J~Vj+@9Wgnxmx@@19euuax@l?8$I_(fzO2BR{N7I970M79X_aU!`!`Tm}0|jUmI7(ANU!IU-i1)eTg;~p$5QPt!>=jA|80GNL<!QmkxK#+#p-?msMMVlRse%!cfPf3h44ObXm8kGzwo&;eD1Wf?g@z93WI(r=96R-Cx$^BfVEYv!JkPvo3{foLm1nqz?!saUSG;G;Y6l#9-j>1Va$Iv88j9<s@xOo)jt}~g%w0R!E#uR_*6-eNri~+W7vF>@YEO_JjQ0AF9uKQaQw<tnUQ)FDr4pZ9yEv+z7r<#}5Z3mm5TqFa?|#O`h(4kTe}PbMA@1<saw_<JCxmOA&1kXOY}5<LvWI5Z5w7*RThTYaK($AIgfOHLPA@|3eTVR%e}HkP5&AHKsQwBe=6?RoI}XC|&%w{{&oGnWR}h!^hYYSSg5`aO(W<}w39Ks!-t;;D-4HJNGX$IbGZy0bGYBILv5P;$*uWnUxbNo-w)B^xYBU4p@M{K@8ol)o5V0Cg;DFSPV9!|$==Y#|G#|7t=S3Ee0L<|1nZ|s!q!##T1a15dp+NsP2#fp~#203<lMy2EJFv+=hVaIJbxdKCwQaC8JC+xA{?P1WF@^OkrZD;dF3dy)h1H8-Y?eQrj*HYH7N<)TiiRP|2yB_l`@)0>T8$uXeU4*}#lj;-o2L~@%zjUdjB?GTutc^cs(;M1z7yefqA;MWUdLpYiPn?(0O$b&2Yc8@jqws%T~Xi>1HaquH6F$jdeLj~&O1=l?+ei^;EsB|>*T0k@{FCjcjSlYYE)r?AIby1z_v9giD2s*nI!T)z)gL;fiM7fa2(uvy4(Sxy1J}c@36FQ&sH6_;Pj`#tEnfWsZ1fCV}Ie<Bsk5L=e{az4Tl~k)&X|FBxX#kJS=Fc{XorGR4)umh7=oU^yOGa(HfG02oZ?qTo5ytv!-vDJaA6Ebtb3>#_%DVBD79`&cXG~eNe*{+VK6)bOjfU=YnEt$f5fQzFsNlHoNKxLp#Ko#s!bqYn)t{y9z|W-N`o<Plt=!c`Jh;1diHMr}qG(Cfn1V{7!E$1GTTQqQIBTHx`{n-kFRSx6lSJzq6klJ9%RRwuhef#uW5`PuU_hyzl2w5NCejYFjW{7r&nDz{rrXK+^EEvFN&RuxG0aB`x_UsGVAVI~_mBl@2&7X2=TISkr7ypjM3XROJiwY7Ll+3ryuS)CN3?wSJ`ouNSLA!H!M_3d4cGHcl1_1T&}Uj8BIOZLyuP;&nPLs$rZ-r1RiFp$;!rE#_s$n<4sU$(KzrGsnZ1!}y7dz<g-kNN|u4ISLv6k;@HRw{f!S$vU}UgE!bS$Cl%9x#Zb&A_0#-iZWKp@bB2x4SwZ>FJY4pg7{Lt#g=%z<=o0Xy*36gSxhSm@E=HwE(PFHW~I`xLD7W_ULFttr%Jpw&Bk1@(M*rEbTnHSiui`}jdseE%+a6^T|g~b&d+Z(t;Q6+%P2rPjZJq5=v$$KCATJXm%ItRuyW0C?Ygz%A#A|aF`&UH-gLhAH{IQwErz(i+ri9n*7CLg-rpQGfhn&R>st+2Z?uXolUdVk0JeZgT3|d|L`Mqw|NBc((nn~3^l=ORCsHsKIKKYoKls8X5OZ9xR>B{k#DxV-Sld9!kk+EN0B2GTB&Byu1$w5?0xJ3e1ve3r@qWtG?)dDRcXznv0<N-gQq%(rcTk%9_P?PL3{2^b4<aOCeGZ=a8cUJKaw#9chaX^B4*l?VXGpFb?jhId4@Ay{5|V)^_le^Xma`4z8vf}}3LTx{WFmBG66y5SLj!()qkxTGIubI}>f*s-PmJL&KQVTV>3BH0inDXWJuN~MhRON-u{Rw#hq!^1ek%fhPlj!>F(2*a7l~Wo9UF#<zxg?=Ujk)kUNIAH;gHW!xsg8uqWn83cyoxJPr?Zb+z;Bxn9?G1i55t=cc%qe82bL~_)<>q*U8R64&e(Ro+FNFBf$$U&>bKpmgK-kL2BzkA2k+H1c)vQ4ODq(t-4BkoMv3KeH8Ty?nan!K*==?x<#Sz2o~%e6Pj818cH&ADQSAk*s|W--D&l2H4rf2_$K`2waxx_pi99LJqIoXcatq?#OQOFo4i9C2KVXx38VTs`0YRVHF^HP7WZK`8W+1cD5#K%zWot+@t?l=3fDr?Fy(VFpZh7Q`GQG<TY%>}Js6?Sh_x?pw(|l>a2jY%DL{}wfM_gui6^d;1o-_8Y{P66H1u%)WI>AGI)t{{1{udWQf*h$z^9t6ByezX3*yy+j?tVO>|4oEuMG~u1Kk3X(ll<tzhny)R7ao~^8p2}Rd8YunYaLx>F$|ARSH<2>f0YEg)YJ&^#Fx9Odag4_|5-`EGa)=FEF-h6|z!*g<y#3K#{-Z7BE6qg+CxFdWSmr5vq2?qocpYCIm>9=R;y?^xr}IKcJU06}*^gN9ey_a)}pJk76qrLF>W)23w%7Acw+~EO4iO^W}BwG-1_&L{k>N8;1u<Yz&1-uuFk0nt)ZNfbJ$lj0g3V$Y#^o+*QInY5PWT`tTKz^Ry{LelC~(6c%Rc;!r6T4<G}?wM<$5*LZkP0LJA)j1om8Bh0%q9?Hg(=xu)mDOPvNO7ZA0552-P0f#){fd2!T48M;}UWk4faq38Gu>94Z80*vTx%74-%3UWSGs%_>#+72}3#7!hV(E!TH-eP4=#41B;3TxcO!c8_)Wm7`M!|iEjm>eL;_9AK3TF>_-Cd?9!Mddk1oSIO?3J>cK$YTS;Ma!Fiv>pYR(W?|!kb$-Ld^zJ**UMpj{=QT2U8Uf5N?;bEi@9Kfm%-(=`T=SK~$&^mK6L98)F5v0Xd&#BLct7OMlClhd-QJ&%Xw|_yhRiE5-y1Ew0~3^#<QF*ZCh&5j;jIVqX5uAFtEA1&Vd#q=QJ5guZs<5xPM6un|l3h^7Upy#)(TH68@-bN087WG7aQk)!_}IzdFCIs6H2psfad9D$%d*&6}nR)2Dj{t}hS`<Z|M%@jR6y^mnM$1__$tnmnL(RWasLhK0DdAzpW|H|q49oo&qhX0PS;D3#c|6ij^_PtQC%dZ*j;qzmizZY1YmBC{oxo*G<o6asKyYT{xd)#}D^;2m3xmL$({Epl1aKB;Wr6BJ)=y!kv9bqXyf)Cla1O~4w^s*N;$Hr!L(N<zRVHm%~E*Ioi`I!ij#uw;_{sf3bP%Tg}m)qw^ZUR@F=)eB<=TLWh$J}{;x*!-HOd2N@E|Fj7_e|AI%r%6{q<(n0_$lK^`T|z{WEkEH@|cZF2p8yz+d^|jJ%ZGA3}Az?{z3$njE1w=OL~9Z_{VFA#2#kH1S#Wu)e)Is3qhM2Pe#2l3~1PEMF06u;474Khaw!15^(o$#azHBmlyJ@rhvZZ6hfXR&9GtsQfaX2$xs->(5@(=EO$dyKR?Hk_USJ3%E!X}<q2W`fa>Q05e)VL*%YEj)?xWaNVmv`dB%>;hzATKsMB93A3JBKX=(%m+gq2Uu|rFii1<VzY>nWQpBA)7aLP|#w@7dsem5#oyum^URy{F&Fp|yN7x_>3CH`|XpCZ%4XNW0$I8==N^5C!k+0n{$k;R1dfVxFt!yPhS;rj^Qfrv>lHrS(&{LiQS==Vs?dxr`_|CBX%y<o@q3AZHi`Blmz8@&$NW+p(6=}wl=)yduGOLhtlB^)djobN8+CmiaE;{h-gePFx>n_|JnXVIT<OOGS0;x>$b!HGO@5cKVZ45?L*UN(lkM>h2h)&b&`4q{>`*@O~4S}rY8V7DVwOy;1wXi(G`!dkIh$73;?Lrz%N4#x)ZFn^SH_)C7gmh#BS@>yRh7fNT7kco1cnjkZSDhOK^^7Ku6fiWH2d3Y4B9TYF<h&!k=FCAm9!AIoK{b|`Z_&WAS2i7BwLxj%!jxY``YKn7Zc=1#y4PMZc1fFm~pxtFtF_)7=#rM9~Rx71)&0H>0V7X58z3=IvAzCTdbl#Zn-YO3TWaO}bfJiP_Z4(Yz{>?u#mByha{j)plq5G(ThxJ)FZnnSQ2s${J($~nv0Sl&o`vdr@&gB;|{dHcL34=F_i|~L4V2s+lH&rO#+*wU;Pr#B9jzX`qB_XUX-d|3M+mn?CKf@FrJ~1dc1yz~n&J}9tD1c8Z#kdd;wZMO(HG+N0ci_Wb^k&!Dzr_Lo_*;W|g&7L4)v;d7Tfqs#HZS-<+ba@>KnNO859t^YU*@_gmD}=-g8v{>Gr@%rn-9Uya-~Ri@&?u*fkN(HZITcg5l!=q5o}l~8!UQmgn+_636=|iPctd;Cq^!A3y6CFl%I2R*9EVvjjJFClNe1OxReHb=DqHke0ER{z`yTd;Mwnxd<K0(KVY1iaB;yE&6aG(v18}<2fq6HjxWV=)BYH*=Kl`Vkx!2$;4g%R!BY)Qw|wuXpk@sE`|<T#T2HcfBU%@Rb}yN2nTwX!RC7xB%9l?F$9r-0nBe+R86ul}-C;Sl=NuVMCk!+kgAJ#Y3Ag!cof)1z7@w1~V{(uv1&_#4dscq5=aqRkPil3%tSDbO7{6hB^4M>7uz@hz%O4qPWJO6qhuVjuCFSHeeDhC87=?b=4_w2`$r(acHIC<$Z{dN~8O3YdYs%}V!a%krK%!Y-`!n{qa%lFY<8l6gMv(SjaZ)(BOV<tx3`X>2qq&z<B%ud5M}^zXMVAQomf=3|*E={+^m+qGf~U!T`!f{X%~3Jo&AnklADD1%q&9!b@L|M6<w|u&E<$Pl8&;K5qWA$U18fA00Ixl3o2zj_o4;IPWP^J)&}4oL(=og{Ob6Mrk~(C);$ApPhv*|?c1}*~zk=_-;GI}uH4|DUK1G(--!Ud#a50{Q<6O6j>y0E0M?S<OCs#Bl^YF!sit{A*5FU@!Yi_Id^y=Zu1Yar3frf8*PyO|d%zUzb$y>zAY(P(eM0|4BoCRv2YwC^e+{gI_TrF(85qK^1=*tw-qhAz$r^nwhuf*RO-oEB!*$%fD+R8&8c%0-xt)|)<ELL(*YdJoq%l8;xw_paQhDsflwxA$}4Z?f}{=xhHPI=-)VvN1a9VI&0bN+DE+W_|el`eR}h!K3=z7V=JuOBDVQR2gHdN7cejuAKkyeq6deZ{e9LI^)xec`$;1g@3@K?Vy%;tGMcT|y6VH6u<pSMILY)SwmYlXq4*kd|*RAv(0L-*6CR6kc-w^719OI=STjFSO%-;9BG0uEUG&|M>#nb*q!R?s8qg6#~lm*b9&eDLB0V+q}To_SO3jx%JN)G7+?#Z?tVTg0_8#plx#_Xa#caM+=vH467e&DE{TjM%$YHvI`lzuxZYftb5&+jAZAp8NKmnd4=~g`swt(grT)i^ZZnVU7*M-@D9c$;mjmNJ3#yU+p|ZN*B#s6N1WrRYr;3MSN;d&z2*)QJG7yQ{{!^H*YHnxl?RC;q&c?rX-MvutWpZ3^XqKMe0Ixo_{U59>7=zrkO>Lr*#bon#xb5G)pAC<W8;(G;rWA!++q72&*aWXNgAlgaD6xLLF;5ft25pK1q55;;Ob$wp9TjY{|Fs_c#5G`^A+@A!*7r~@b}mceauSZ<-UQ2c(K1k_rhlfX0X@1IEA01P@Yc_(-9grN|G%S*&KwXdb`&M_W})%{{boHavw1B2zKzQMg0MDD!}@PatjEa6usPYmdK<b(6v|2B4^Ou#>+3tMSNO=*&^yT#_dU`BiMhpK?82kgtgT`Q^mZo#)>-(%i6Si4-^DqeC;ZuaJ~<S!Jj&GnK@!T_DdYOCpaLO7~6OwZ_qTgfcQUeA>U4<VLB|JVIrSH+#Cf<;=f-rx_?2Dsc2*<ixB5m;#2+^at#U%1zs@C2F<B-!a>XZOWyjQ|Aa#?1>cG?ZxK4rD_-$S<fZ>JBgg$ogkw2{7j`jMjCx}Qioj9=QSiUM{|g+91Ys(eZ(sGwuNj@{Ph5=(=NbHW@bkZX^KWeG=EVv6gUK{Wl=DP3b=cTfy!_w-<0^R~Y6J18QJv(A`*?<-krTj1;6t6CKXSl4f6dX$h#Kwo31HoeqeGT%K@j*0V`Cn8{m8!gpBzIH8*IUi(7k#xQ9Taj(V~D2ij4OwcgrYQucD*>acb7(o4_uO0v!(q#xR|F-RWl)6-=k_U0*X!-mky;>f&3HL_Ql0<X*@9H~}+=r||4`%u7EAUZ!7O96&B#2u6ulPVqb#0$rRx!SiNT_uC(12?mN)JR8QhKNK7V!uMH0d_Hmx;~HKul3$-|Zy&Io9y!{7bPNQ#a34dX%Rj=j^51;(Z(w2n5pa}$1YiAw=)b=IOQa#fco3+;zQ>{j6d<r!_HVfn0Wics{>t(4lULrK9*7TTL3e@!@5mYb74^hARyf#SK>B?}N$6LM`8SB)!+6d2*W|JJWC#TP2Q7@66n!EBn$Nrkix+DV@a-p})_ekP*|vj?4EsILItE2vFm;Hw|C%xV_9Q~*-OGnej-<g)QM(Vk;4mI5`1bD{B1{vf#_sn#V>j5AJvTKbQpu$A5h6$Au1dYXK)3VPh;pF!>J9?iB7xvnOaM5NY`{4M60ZKge)99A(ILX&*xKfB(i(^YQH>}v8IByN3@rv&M%A@XQ+#otEs}db1--$gC4xtYp95$tO8-3C?y%WRF9PotC5<c|uMmdM={^b0%^*Ctu>L>)DQpmN0#O_yPQ~k&CIfhKBASe%M0|)!1&tBl+VsH!fkzl4ZtT)`pMmkh42&X_Cj}*e#!xhYMaxDbi2uEu!aShwGKC1?OXMSDAPqDGUzoVGSv;!=0`>gM_sMi&mZnd_W)U}rmJ^5r&A-<t$*rfrqQbm-Ql`InQfA>vo#Jh)ufjcl%pAFg?Bf8cUo{_#@O-*dW4P#oF@qxv<^b4juZ=^jSo8*}y=*Z~v)4}s&}SHp4Niw4&4VrdJI}!?JO?hy`T<240?E^0#i(&>%b>1**WLISVEb3i!}f}KaM0pKC|(cO<>1u1_zR}<Dr|e+9R>Lf#@4)kN&Op^?vToLA}hT=Xk&0W-W{ST7&8_xC>E#xzJu%ZTP%pOv6#5Ge?;=>2OcI5hE2kf7wa(i9rF_-E`5b?gZ@q<$_1jZB5WucAfeiV4fkj~$m0K}vwgfcuQT2GsAC?V_z9N81gG+E-~0-w7*H3*QrVeV5l#%sc!yjTD57)v2DkYX98hqPU_TrzP&HfjVBqD7L9A|L5o+BS4qiXAe*>dLhC-|ma123*tc&w_=l2)q`E#6QeGI1gjCuHj$ZQZXO2`MKZTu0ZISMTX7I^>2^umctK)rqYV+ZsLVXMwpEG7XDQ0jN^Ao7)p_s9HlegqK{So@BP^#swXux0|L_3E|!%|*Ne5mLN9XLPJDxQY?e;$-bVTv}lWCcgn&4X+G{2{xQG+OR~~a)IIrA$||68o`w0W(y&S9w1x!<U7#EZ}BP%W=>qDcC?Z|ME`=X!ISSXs>IhQBm4W96(CFX5wfqbaPW5i+n*tE9_|R_@)&3s1`P8r=Z`qVFKDHgrW0f;xgb3S)#K=K5dE3jLJ)j@+3qb4r;!$+ya&8d$MlHT5BGmESulSBgSp695)=65=CTbPRrC})liAWqi#dhZLIX$`_zx(HiXe{W!x(;i{b2t2vcde)&;k&?i~)-3b$~30>K{O1qtR+nH0=;I9w=~ucY!R-T_Fb33OoSo$Gv)lzeImPvpB0YpXKaaB-j*s_mBxH*Fx=B)EJ@hqj2@*7zlp|vg%mu1q`_}MC0v^rpv|04E5KH58#wfVEX}9ADL9Bi$l&F>j`1V5xD8c!GS9RsDij&G4#<hi`oVPYkP?>4;N4X1oGaiNByhIDN+JPii_hu24Z2+9!iQZJpcv-9b*d6cyT?5<-9TcgsFcrqTpnE^oah5)n;LB7*Ibvn)e?)@*Es>3u`t|fBS@{7K<TL?~WHQ;=EmI*hT=Qn4DHXiG}Wl<@mttK89)i2`961$1NE4b7|br7Jy8oV0odTfW{~cB7Z~_3TDwS9L1V7Ccu&%!zFABgzd=!%Ba)VkNV$*9c26{jf?9%0h<iL`GCSgwJ|G0ScsUFahWeNZg3>(9!2vd+FXPn@;Mh5t{2;jK1Bli_rLvt{EqcTYEA@sz;*AARvXQkrwmNtt~Ul}pBy!&8XQZBLnOa|KaysH{4SU-r4nghki7S4cw;zZlaVK>W1v%RVx9)BGQ`?^M;=0*za)!B`CZ<Z4HJ1X08S9pTF6Y62mysGlIOqZH%HqfiR_(Upn)N<Mm?E{RB{O1)mI2c;R^f<vz<S}50Pi?Q!qXvkWQtO6&6D_=}SdJunOks9519tkYRmRes>WFQG>E9juMzaQ#@GSyZ(_Y)8RM3ZVIXB`M7?`c;OkUBu}GR(dvLzC2~2u1OLjn%OEko=k5a}77}~UP)HrSDO`M74bRB|iHe(^qrnVHC;|$ar0=--Mf_R(3f*h(FnkG#CjWazI^)bF09G`S4#U^58W;J^FZ3i9rNK-CzSLp!c?W~!>S*v8xT0XpIb?mR@CWnHSao6KY%p>Xt6_K$!Dteyq3PqB{+;1npe-q+c*yVg1?w=S0s6?~i7f6#skwZp0f}Gemgxx`u;^hlx%&jk1s7me1%8d;N0=t8`uH`H$c~M;-LUHJDRTxy3<wxDIh@n89p#b9k+zuY31<``Dhv4nK;dWXFD#KAG`VX}=7<D=?46{)LPj(KhH?_Izu)29efR)2n&N}-OT-2+axvg-Nw|otF_Ql?$z%0*7zg$IiC-aO!5`2t0q-O65kPDB1w+8;6j-hRS}pm#@A|e1oOC#IQc!!jyMibUhW3A-#*L6T1c#kIJPdd~sQJHOwG#FHC-h1vf1)s<AK;sfv5;{e`Za_%hCmmLapVK!DICrhF(3JR*q320Ar>?3c-Dlp%P&vPP&yk1jJk@ai|L;z2NH!GvVNZ&rt2~jiP#<U2Uz@!4;pUC<0inH^gUiI2ASz+WY~Oc879)upIPCyJrTor?OBe=vJYM&A0V}B0TuGQgI$kd=Iq<vGxW)e=p(sS7tXV#XbN2B%(k6wEei;tk;l?tUAbEbNH$JXa;MA+no0gdG5`hdzWph@I&yImTs-*#mNtHY%~<5QI}Egi*DVZ^IIIZU`WPd5cld}~GE{mvlw^=7f<sIcNyfE|i5Tw}+5zNjyJLUli!V6b4WA?k$9M;A^coJjADKv}<#*s84Hg7AiVle)XMc})C_$XdM!!AsP4Y-W-+Tp5?7`K}3>wdhE8`%E;~>KwCAetr0K<G>d&lf=AYE}{=%4^l2m%Iqd^Y4u0^Wc&hNB3vD*)M4HcKS1?G!D;#Vm9A9oOyj8j0!s$?bFkj-`Ro3m+h!;fp6B3FM(bGy@*ZS26_u<q$aR4*a7*;UZuOU@XTGZ%`u-5hT22F|^sYe-GstAP)G#sdO%n$R`BPKKl9|lL8O756qj>8QOtAp+Gon5Jdll1aU-fhIiUw+UK$s2f~Rr5e&#rvPrn;B1Ypq0CU6h!Zq8H$Ve$L8X=E>hV^(Oc?Z)CADjzt-c#GVh6SF3WQXK|lX@cHzCm~NpXGN0jY_rJW7KaA3KYg%fi3X&_Jmn8QBL@J`ZHW7#9YAWd0|aKK*!;f=DhtEe7)uLIka4dXQ18vhe*(R58>@k8)pe2>F^(TL9&03InN<z2>hivZ-bf7dDN3YQvp2L<Adn16@7`^^&tTo<pt-$b1IZ@CG`4Z1TVO~nV4M8)0}vAe6~n_)G%B{UIZ0yr1*ieN(Lv7kSHec*!}Mr>-7h&qI3ET+*$O2n;Hb~kmza4K=gb}7o%WLo(06q%BXmBpIzgP{AVw6O7@KN!>_sTJ}LQg&X34rJ|pL`V@|5TPQ#>)J8mwKK$~cQ;8Gvpbv(KgFqyz<b3AYROGzH<IbZ<fW=L-^AUZCbdYa@R5v9kJ78Xl__A{z6LN$|^stC8D)I@|pT9l&2d^v6%4RPa%XtW$IdfJEHL!-H6D=V}btuD8uFp43EM%>Ai8c(;I;Vo*0p-n|4IK=dLxD9|W`46p|t`ai}1)d4Cr{h9~4rJbO=^S*If^oJeSNF|Z$C8Hi_4sJ~Yz75{ywJ1}kN{qY3-wRgybxF;0L*?23%?-uXuG3yxpWuGe&PZ+8(8hl1;bQfSR@L51|l<1WMI}9te^!6g+w+cIpE5AoylglngswTTHWP%0D5X{6nf1FY#vy(-cZ2|;}hmh$_s@I(bI6EIOPL(CNxh}2d~osE9OFtR7}UhE!(bk2V*JWXYTNS)XYO{ip>F);>~z^+Z%T@TKwO)8t~t7V=RnxB0^gUH_3%?{b5bEx9X0Z+@X=F^>;c$doM>Jq<BJZJ@!xI)<&U<gPclqhwj-D7?4&p@^T_q2Anp7^g04>4Vm&K2~w^;JfcK|b=zpSkKJ6D1=#*|^R|T?DI{vldBm}%Z##2}dZfr{V+_j@_zZIRB_`ulk6MF|lXZCBYbRjE1f1n3{3cXJci5vBiZ)w~LU}oVs1LJc;It_QXA6)8EF`JFL>6dE8l&lua<$)prkq+Dh#%N>%F7R~d+kM+cD+FmAB(wSrBoiXm~gLeuRM+E-l$<82d`Hu)dmebRDg!2)Er_1!;BRCB8^Q4e}|-jzrci{H(nbe&8!;PhOyT)g#UT-KtZfN^zD#ZEU3AnuY7xp`b4MNWYerK^x19Cc||r(Hvm>|M|*Z<y&-3G@mQi_O>0n8GJ@1;DFNpL-=eChu1o`ta2x`k6>@qq?6u{Z&O-nvMd__Zi`ghwUVqsw1nYE#z&{EmnJR?-Pp)ZBhHb=aH7F5FuY0Md#*pz5x_s;>1<-L9(9`LXV=QO6W<76A$py7pFnk=W8e3`u_Hyx1v5i3cD?|>s&{>)s2ggB!xC><I*N7f0H+^v@dFgoIb$8bj75Rl`RvHC7uNvsH(xBlfm3J#wAI3<wm!l~ST3~KPHt1Hb$BY_W<qdGrLJCk`e;qYEeIeq!3+jQI!#AOn!|&1ustyX|0@w+z6#DVqD1ZW!hXAMxaQ-#W;|&9QZ`luMDKM{iMb9gEIfD*UJ`vsx{8u9nI)NMfv4caCQZjPQ0v3Yp(VgBH&=DBVtNJ|c5$+c1Z^Z?KQpjC$4XT{vm7^Jd?{DPa`y2J@df$*F*%Pq3&ePadjUj#IFfA7on9&DL3An5~u8!QNzuTQNKgfSkk#6OFhx10&`n~+h^^GMlb`U$J%B}nk$zF2QM1FU;4>fXJ92a-3c}>nQQ{W2=9&$zi(a1Z?;gDZ&mywkA%1f6gkRR19{Yi-)EGnQM-DXBOi1>13KAk2*M2W;UdE_EIWv!UcoZkH?5CSPJUw!*2iL5KE$w=sVcIDoj`3|A;^Z&q4BR?o(ruYgwMgPb~sIk%XsIVRioMVi#u<2U?sTje(arW+xKaTkub`MdlXS{>xBU)zkiP;SF`D`QJp1u^U5UkR4T3HEKNP;>6l;D>m#o-l#!$cJ%k7Hv&wgl8(rQjAHUgbJZM?#vhlG**^H)K9rNI@YncnN3iI6N*oA?s<x=r+(6{}~g&hGNTp2XhfHI?or_N)F@6zWo_e_P&I%b0VGLRs?qRSSxTbYma!riLe+7N9TegXKH076Gjqyf1o)tS4iY@w@@RLJFsGc1>R^XNR(Y1eG&Gc!Gx|9s%nWIX6jjf^DDNpBQqxC14NY|<x0nF*#oOGox{6F4vSw6@8lZT{Vj(n07+QG=4Y*xZr$;Y#bY>RW{Ftd54kkD$*;2w`2!F*y0ClLHLmTY{LUpb4Mw9`XVAY?Li5n<g%X<jThTc}3r@#ArQ<I}_dX#tA1x?`3H<{`Fvo!A$k~7~Mo~5qs;`6t<=Cnc@1c(iafiQS;h|XM<9DZdkqqC0sVvoK2B$KeWMqL(1KEJ)tRwI}w-6?>`Y_H;=)C!iv41i1{&3`$Y`hA5^N&$T%K<J8tlO{nwpunI>eFw32<8YnJ7nxK;lL0cj_Z?x(XuFUk99g<DCCK@h+B%Z&W(nyi;g){zh|Qm4{@P5TZ}cTu~_HGRf&*${vHQr<IvXIzi!P>z0KY*wkY!pq<U!fu{A#p5KkaWWU}dCApn;`L=>Z_mb*p-Z%I+cm%T3#c|f-s5j^hbLn<xGWQip5&56R%e5BwPm9o)1!J+74kYQca3ML%@_I~^^4-*AQq72;vCqX*R_Gm%^#R6zVjl&fLQDiTfwfPUQp|RwwAz;tuzUYWrgzyR=s=)!OW!mV$7(`)H?%8N*s>ReOL+-%sMccR>G=SlbEilXx1cPY}8@&;I;!@`Jj}th^?u4H97GQb@KzaeEGPgt}uf#QqPrdH-=DpKQ!T?s7X8|#4=^g`-f9{m#J~f#?G{y{IFNva#!KT1SG6XODojua@0TY-+S#H#vCRZk%;oOhH{0!YFhbrQ$H|)o^Ku861nT*kw%*VG^oGs{vIzG-?o0vps%}LREdB9i5fvtw$!sC-?>aCx+Y+&ukA;}RiPNs2omjm$yRv}yWV4N?>-rFQDu6pt?IK$yA3|&eP!7!0k3R8QVT9KW`s5jh_6ugab^3RD=Mi9}9gh6+Lo*BUxp$xtiJd&KEyF%`n<0J3|9I!C@f+t4!6mvi_PrNE8jfOCETG)+px#wdNJ$szaT`K%1n2Lq1-U~hz{!@#|bih7(EJ<?FvvRp39=i8yy(@<<j-XW(H1I>tgB&kc{49d~Ac!;wX~jWj6!y9B(63H>Ek~MyO6Wx&%OM{0aBVck)DXsbrl1GzP$C5p^{Pk1m*EyD1OqqYF&`*<=*ek!OGEP)x>GL1Ph!r#B}ae>b`T2AA%^+Yag)aqOowaYfO38{Zu0y=IMP={Po6&rmT-Zdz7#}>v9UP3<9Jz2=Tv8iO9NkyQI)qE^ZAwpo3U=p+ay?Ec3F>sm%R7E#XJUx)7)4Pz&q$EI3VN*5G%MDy1gk39ReqVPu5Jq7r;JO)EE`!I|;>P8XVy6=qvG;Oj`=fl(`wn=9XME=z(y29*clSQ$t2h0L@vPz_YkKW_fw{_o%U$_cp>YrrePMDS{#Vc8_L75nkvfYyxPXgd!*`&oSo5fgUXA=r^=4rxBb2$Aft%zVngcfZ>orA(!-wsW5hEbP{pNdjV{G5Rp$GV0wg=2S|1(-~kR5z8rel7#^R+(*WZ26mU-ZqXSX9+v#M{rRW|_;%Iw7<-ExI3n583y_xegqIh&g!>INv5lOtIqTt;Q-qahn7Go$Qpi{*}S;;xE^Wd-ma>6V^FszEJh_lqrRw6l4@kxXz(~gyE4`S+B<R{Y*mq0jcE*vh(n<%qQ2$%^-yN%)GfftU?9s#Z*a48y}g|TAmtgtl355#9$p#>ETH)Y_en3@dCpcuSI=<Mc1V;_$g;JAQJTML>W>dG4)&io{~12_-4WLrH@6S%coRQu&Z$R*pBWlHRs5)QXKUZ_m&Db}jEQI%{$HV{$AV9EInIH8m1q9tLLB5EQO9Enj9NCqW=Llnm#OUPvzW$+|wl1*{Prw1l-(E`zufU8Z$=qf)nL&G_bnB)!&ip-c^In>XKpM)9Pu$UY^TC{NN<e`=|(p1P}b`29kDO7YA#Zd~}J}fK&yb&>>okk)u4LQYVxnX#G9thvc_JRTdu_!vHD3V*j<vcWycTuk=<hrp1G{mEPfmp$#LVy;Z2^&1fA_oYGvv@~AVGrJr(6mCIM>)zl`AsJR@kJ3BqBC$-msiY-AEd}!)rg3gqL~*%6FW;>9hWU4%u3Uo$i%}*;=NHg5E1%vNe1@dEaY&{HNr%LWGD)BG#FLGT6P#SG1oVG2(0PA+YZ&Y;hJbJgxCNTnQ4EFut8riSggV!Hl3-9C<Z@4(IIO2?Ucc2Q#$aYiwVhmrRYpGT(>XO459i95kTdS^rC%gz(-lnTi0}U%a|%dFt(sG74{wA*0-!l4Hhm#z7Fh5K!Z_gI&#JU2E%^HWHgub<pUAX4i0{Q`u^i`vyPXuck}n)=wXAnoY=p{SrT**4Q7k{j)6b~-JRRX<ztSoj2X6^i6jAE!pIW5vTSq-ptbD+o|n!PA-q9sY6#o;NO9e4xa<k-+R4E}(TL~qF)B4D0w5Tx{O<Vq@H4uqnc*=Id;CerkNNCSKPB1Z^}~}N55{=%0<LKc9&qM<K=Zi(f+asex)qS)fm{lGW;}VG(EtL4d^!}uGX3kiL1RK{2b6rUxPZdVATv4XgW){%lxiW%7`n{x1xz2*1a$e~8#s6J{JGi(Pb#%pz${GWA{C)aNCebC4U-xA2}+NqHh_nUfJ)NpLd>pII?u^IX8Qw=o#2<qM~%%v>HvfRhye<z+>L;fWjEuf4kBK9wiyQn2V79y#*&5r0Z>_C6dk~~BEAB+af-z0gW#l-^k_1mfRv8VnC_qsP3Ds6I6*>$(`^3a5lqrFxZE3%Z?wrFwSq<ic~Gb4=Adr1oCQk9^U1VE1P#F=8qgks5fA}&1F_k*=oqYo63rPPuf;^mgNbPkfw(e$kYCvUgV+jX$sR;$m){*>^4J@SU0+NWa4lJwdiIw{IvcHmi|}ljp)7_oo#Dj<s1ythkPWta<7E*rb`#M|{*>en1|`f^;OY@Y5k3SI4AvNH$J7G`df=jJYDAqLguMVP0Im+aWY!ea?!dZ13K!u*Z!oS8F@eQWXu*&?b9nO@za|gl8w8N~i3kEw!_<FpCblTj$-za<gvSeX4+bT|h6WT5f``E=cX#?yZc!7K^10%e97@9rlsMiSYZ(eaB!uzYI5so(2y#-gPdOZ8ApsI3Yj~xZMFv+U&i$V)8z|aNBWhE0Yu=lpWkBc_jSfusMsk}PKhx<g5M`pGdPC8}65@_8sIysOS(UBiCBNfCacWL}=kZE@Qz2%F{V*g}k-?yQ4`ZwlpQyi}DObbAMj=KhhxZ4H5l}>kM3BTX<=G8!Cg02Ra54VtCqIYF(V0&ksj&##%|C$*as&&Xvor|ApQAktE$A5m!hK{+;1HEuBg!Uo;NR&H&EY^#;O7%$zr(IO5Wt^^IqpcVMfv=f0rKP+K!HTK8Wg5GL0^SKbR;1W94<QcwBT$W0Ld<!FvzIUgx;??U`CLwG3*381cCaV@cDsY8py&Z>60#wMMX&N<%|GSfbQN}!nZ?y2ICE`c&@k*wZU|tkOcFAW^Sw=1J+KQf!1E>jm57OVjtLdg9A<&f$Kti%b6GFBcfbh*gfRv9Ajzhp!13*Q_N<dlZt|Y_83NEzJ!Q1;P6u-*7gl^0?=tX-BQ*`%etlba_XK}0CY9!0fr2OG;q}*>@vi0;QNP|ctHQ6NwW`*5-nOxM6p~t<q!=hYKoZ+6ygE<X+pM+u{GKh3etnu;q93O`bLlqymh_V$ysPXuoZ9QP%wmQBtpkxTn{w*<@CF^eEk>MdqHoQgrT50AHlBJ9>6#3gZrbTNx#8QzI}UQ#W<c#^m5)8pjWn91|Hh+FQ#uAyCM;b<0xqGmO!Z&3aC5DnWZ7YMMUzq?iPlj0?lD9*MKi4sK$(Zx#(zZawSC3=$z<qhp1Hr)2F5*6ujC)oYiHx5XE|~`MheFoF2dt0kUs^xW!0_XSyccU4pv{r6w2@sHFzKVyH6$0l3vgcAZ;j4OBWV4M(7@;2h8?YqN1i*P1X_ivou_pj(6n90xig6>`;We*$=0Q?BG@cvPbqPEZ$n&~6`lQ~Ak$gRh`ocw6mFktPeCL>tRhgb!_^88S$O5-_ms&;UuQE{g`pos%`_#bUzo+K><)@6MwLQC7`yk^)iD^s?!aqk<1K!W-;BWff2ciUM;Tf5M1^%z6klWPid91!6?lH$L8?cs4~x1KzEce<D|RZch?77=<1z!HGWWJSY|ieiru2->M8Ml~JdWAHGCdBZ+dcU?c)T1o_c{mv%#NA(_!U-a@$%);d+P_7Tn-1K9u>0(Z}O3?jp-;1r?q&92NR!ro1et>Dn|qNDTJ_<D>k=cO+k;Jx6~`IETq(O~CbHW&Q=N3?m!DgjOfXGOBaeIc49FzG={3yirrT^_s<w}=tGdT9jt9%r~G_M+b8)(`C?(X;|q=7l+PO64Up5iXIynFjiTL^e60{gsHKM?rPSAo7bA-Nr(Mj4~*#txJg-^HvvJrtL$}9n!ebPcZBjXdTStnSkT-vkXmk>~TKc<N~VzUgbKjrCbdke1i>AZ!fMHcnFp9<?JSB5zwpHJ;WQz?^ht4yX!hB0wU0cyZvAUXk&4*<K5+Kl4*}Z%LG>hR!|-_+l@QsF}%Mfm0oAS|0p$0x%^sMNtZ3njk;A{T%azua*4}@C(oGk-kvQLG<&he#p_GO3bCBp1D7fLxUHvd^uyse-$hQ>NJsEx1CeO)JY4$ndA#_tizo4~bs9OV!ab8&6v8KPB-bwFBV^NLalABq%P~n+gG@%poYmvQi~08MEb<-MGhDk<urXnb4R$x#8V_vAF#(v6{iXPK^g=$5UEdyZGibzPkwOy4M)SzWg=4WVvDijyv4o~UXu&_rpF$EU)G9>WcL;1K%D!5^p~r7g5*OMdSoZ~faKTjA4f<rjS#h-B`W}uP*bi#%8Uv9FCB$e@z_<hM%juVbDZu!7u9E^j%KaN^OAt_;7n07Y*9Mv^@3;%Yik0+O*tdW><g{n^fJkP5qKP&D)`pjBMDkl`22yYvzqzx$<qC`)h*3G*-mPz2!wF4s86ymD$03Pix{%Kl@-Q?JM<{l1l)h_+g7o=v(Vb>5@}n4V*X~?uVt&P<oNLF4xZgWO%u!}?uko*ZW~Su%MK}wBu1271Op&@z!St9zNh2`qNt3~Po!~kZuyiIud51HCmnvjNg;eyuKwNmiF^ucVTaJY=IQ2S@uR?uXeUpOo8lad%`xwWL@8tZ-@eqRkheOzZD4it&zFeM6XM;o*tXR>Phz4JH4b|yT*tmCu&4s<XT%1OnnQ5<kHV%Vf+!WDN0C3D`E>EQLa^XA?W{25Sk$(V8?Z+@4Uc<`@`09mde))G!*=Te+EsKF_y$}a~X!b(N2F%3yH4mu`K{bEMl+He&l?&F&VX`oAvcTskPXpK6ekVfkPtc(A-x<_t*bBkji^qli=ATe%#iuZX;|m0P1(Owi$^c}u=*K4m`tr0k5l$#MEcy?>{l2JZx<y$=zdqCpM;Rhu6dI9Ab%4ia0Q+#}(GM}wpXl44ee-*mHlb2!aJ54Q(4B+FjONM#q{3goWR#yCK*GKhss@g7L*C($<1(v<!g?QY_7t*(X+#Iq-am43sg4NUuu?0m)Q6V*^JCGbcPu@A^N!hBw4XdQYE;(cl_h%w%fDVMAcXdSfrk1DgWcnyo&cQVIl#^S3hxWAo)3J<?JLg5N9B*`40c_|Vl-VmH^_*4_*3*Zu+kj4ocG|v-Nj-`-+l54ikR+B=*68)XVZPcP%-!&nXExeqX~Vp*Kp4w|J7_hN)-KO(%DQ#J!R4AJEo@AWVYC}?MB<8HMI?tp<%Eanikzh=du`KE(<&XI`|<+Btlvozn~F?qA6d3Qx^bbv1AJ=5gUuE5#@ZLLoN%4V{$E@d7yo<rWJ^!i%s8J2nBI4y#n#5F{QgOTYwfpZ&4e<v_ti7*&DV+C|CN}XUb3JQoc+s0xlfC%`9?><Sp&2JZ-PK!OpNgug7*dhgx<siG=$Zg{2r#4ffq-B5n*f2TyCFwRTGC$xPMPc#t_NhH#$fDiac^v?Pv2?A=&IX-v*!xk667+gP*n_2WMC)b|$s9x5N4H|vr@I7&MsPpPzJ@80DGvHaSypfcT>xu<S5lap#<{`?TxTeDIBlY8f(hC5TUAG1gn`um4!MSO2n1q;6VEEA`kBZDGqNCc(@*A4OXFd5}K1E;n&bbH3*bX01pJuMQctTN(r5?0#PjrjYX*1B$0XQVTWxNg!r$4hE`uT@I}!KcKw5^yT@p+dX&l%LC_m3nL>4vzNe<??A?&#sk*f}`hM*B4Kr_H^sWK6soC7wsY2KDwrz>`3D-+aq>rWg=*g44oONg$`zv_GfX$Nf%wAT_(CvcKl`QwB?f80;Jhe+>DIAn^ki{`;$qkyz-9}IpdRTkq|e$il#GMN?VkUHC11V&8AJXSte&)gEG8W7Rtqb)BL#btA|Eq+*I{CvV?b99jz1Vo3!z%cCWT%CDfQQ$P`hFyAxQBn)czm?Y<d4PZrBe$mUcfmvLQT>T`;hUE9zZn?|0pt76fb+UV!8z@~Adt|~2tnu^#jbyi!@sK~hHqlBrIk_`<uYu@K~stU$(z3HBM2yJxJT#gJy*G`-r-&6E{R#7Dhr_?)`Ri?>ViPX_)qpA{2(N9xq=2PhW4<U26{^Y4rviv6LkCjIzwPIvScV>f(xm?Ius}^N=c2fu>6dpoKc!n}pn#{<&Vs-q<rB<f9mivi09nX%Zqr6s9PHgkG=#r9*)*+)-`rxlv2b-3sE$$h#mUz;qO$;sCht`PdTW<`FkVN87bm(lu>1nm<L0aL+8%q7YHLBd@p2=ieYq)O~RFzIfZ4KCD%fgdFoUHV1b9Z*5>x>6d$9-lzH!YGY$*y4(J5v3P#1JK{=9u4e6U~ej1=GDaQcIM^<-#JnGTYS7dLTGe`8#IfDQi{bb<YiNV>{H@8r#}nplDCXP$1=t5<rgIFDK^B*1GK#((S~vl$MUchTha3S4}-_cPx9PPF5N2XYpORS2G&p_FmWLd7`Q%YoOPZdH402$KiOK`_!wORwyO&*{VB-QEF(3pWEa#_N;YCBfh#VnQtWvb3>t3lDhZ4e*R$;s!8{~XucM9+`D$R?MzlNI{opE^SSKvrquDRp=Dc3WGYFbmNmFr)cmHoq_S4Aw6lu)m)>;D(wTR1`o;ZIG@A*hTaRl;B7RTrwEC!j?td5#M+v`wwjCMib<UA~q6Q(tn$+FoY~tG5wb?q9QZ+r@MvKLmO}p}J#$?>l_XWB^it1DqT6IZmdF)wz@mhb^d@_1^aeZ_Wvn-+(>rhvUb?kwt)0kh}_+pla3hB&wcS@khS+bs`QyOvi^_8(nF%P;1E7Kr4%kG*bTHo2Gh7%n*4136js@D)6S960vWGZedpLVY1+DAIHS;_r$HIglofmGdDiw}DiT`ldg#fUnUGrA`wTQ?HxQO2&)`Ixeng7dw)<w}`tgS?`-e0D_=lc!d1l_}l0BjM<?c2QQ`7gZ&Nvaa_?pSRB*>C|EAPX`f6tewd!YEz$H8CK<1g=gBVGQ0OG-LN}WRrG0xZ8z415~Y@>_K@C|ql;{RpWW%TRc$y|UPPL@d@DVxOcVuIeY{ae#Tlo6ZzCL$ShVBl7z0~LP%Md*)GNB{oJNMZVbB?CHm!=ZH4;jz_s<2J+3$A|-gQx0t=fY+YhE=ORkUlFqicIE8f@y4#h^&9OsRoZ)z8sWvXS+vtCF~c?#zohyDc)K=d(nMTEzqFu+~#A?&7-DVi(LV_Bprz(VFO4ta_)?N?C_ye|Om%yH%xt+maTqA5FIQNTus^ijJ`=Zrwgg_de}%lH9h}zOguJm|4u;8|_j>mU4>y)^%f9W=h>edO}Ge)6HsHP?43Evr_BX4Xx0F>&94*%r>cqK`1oOuC@uH7|$v*7Fto!2A<9CNlrPc6OxE_rt{nfmfi+EDk_UkU2a(z416+gXB*5{W1&|1F;Gbd)b*~v>GRD;5kocZqiS?pX}%#clIKl-GTp14QkADIdCof$jgdb}M~%jmOXqY@1q)@38@02i;!3)eiYJpt>u9DQEwrA=Y!DV_JU5vg9eFlPeW}fo$Srmaw@>AhCRX=h<B+nrDy^w4WbfC;yJa+~+_yL4*=({Y1?YQktnG5eB9$B4TKeS9XDX%V;FD#U^HMwi)|n$ougv2Z_*MgPNs(Xgx8<tQWZX@*b;COHNVZ1%n|5Lx&xAM6o0!qQZP328KJJ#0+L=s}N?GmqmUwn!lDeZS=~8bWDIe?$=h(6sv^Pn&FG|PKv&BMeXp}aF;=C2o=3|LjdtEQhBGjIe#nyw$%JDq(N@FcqH0$qcbuRBX9}O1*X17wCiW<!MPUx{AOS+@=oYFXprOje+f!^=K`>mt1NT>^wR?Jdgtf<=5MCK^pZk8p(67786@#eMCk=r*<<@7dHWLId-tf{(ZZ);4v)q78>KF{T+H|~jL@7p&_t5Gml^9DCjF=3cLu0!>Je>Yvb@4eC3Zq|2s+Rlt?Q;gU`x?n9SCc$kun7Yj?nannqJ|@X=PJKUgspv;t(YDnt6@y6N-cc~D((~PxEM@1Vu{z$cRV3?uOF5b<@*963RWyf}M7S~^FP6G;GHWnTvl(N8Fu6R<w0f~ptBI8{=)Rxj+AHZSPB?3u!YpjMkBvIIX{HeHM`ycsG%E8{d)?|vTAAuH)$QZZ^+eq#Yo(!sdQA788bNDeP#(7obKgL1c8?2-vP9+|PlEPZvb@~rQj>PC_B>lZYpd>7^KsCn$lZ1}Z&q*Y{^w?fSSDM~JxkLeHHI@KRnk2-`L(`TRNtG|Qv+g_NyjBmQ9#l~N60=3uBlzwH(#ut65t-H1g!d4(ChD53}VA-Jktf6758I+ESfr5kHQ{*TIV+|vwm(Ej2~OmfK!)D_%}**E~H*J!#6ioRj%<gYs6H_H9ajxf}U+5BBmQARo&N_$V$nUH`z-%Ls7TFzH(U>$|p}`I;+kN)nu~kngtZy8DUrFmR(o1k=gnz33I!@?8%-=9!0RG?Wdb$Ns%g3>!vif_Q<*evO!HN8^4xbj7x=S>pAyWo97=0J25YGeVL(e*Yg$?)?j-fT`P8Wv!ds@SzEP1X(rRLbm^Rp-Y1tHnRiVrJ2UCct|ZlO)w!}gW8LcZ)|k28k(#P|r@=6il%K_9PPXVjCMq}kSZ3sRXz7A$zkD2co_CMNn8_Ni#bg_gqjs}h_IsX`(>k`*yBq7I@Jy=2#%Wcoj#(wN$L|`1`)YN{@EDkDYt{RqYM`hl)%U9SW8uaXTZV(}ZayE0)ySPPv|E<)UHvl|U(|0V>!iuiib`aGhPkO*uEI%es*qZQ+k;$7O}bQBUoT}_sAP+6aHwdJ&Ox+Vt5xTcs39A&TPoGn&AmO_>{^LwxaRIZO8azbwNz4WsZ~0fjiy>{X_)IJeB_gssO@}uB40F88&#;Wsiy*NYc@$mD*;zM(9ySb@d*{&+Qda){l?<;bxVthsac3N3GLWjYAt|>UC7k3vipu>OtnX0pt7Z%_9t&JXc%h?RIKDJZo5=jRcdwWk>orz+8Bqb;<T&p#%Wie;VcA9o|uhRY=^n=Ed1QA>~BWY&3(fWm`qb*-DcGz`c6f1uirnq-Q887V6D`RlTga?oX)phkA+?^Kk^Vh8xbjc6tPV$Y7`fy3THFbRVhpNuDUO|c*=CPIXyKjEK)YI^|+kdvtWC=s++7eZ<(bh_2D+tTzhS;)KnIEPEqlq&S4uA)uCoDlS_sq6H{;#c?!CkQ~iGE@k*tRq^A(1imr#!jxuD{K%@!gAJnR(tXk<wqm6OLUoEdl#j`ZA^8^iMdg5@l9t~NCTIEO&tKmj8V$%|>ovdXHPK~3XS(h>Nv!vA7=sHs-Z{3=WsU+$t(M@M9D(TqaR}5wK8<Tb?^C!KMc7P&h<?`6xu#$-pp^Dc@g;&}wEc0FWY!MpTIwrOMVPBUy%H3i4StWf^tE+Xf&otcMguPT}Ki$@zoAtWwdF<@A6$bBSAMtGywVZA<-tK$W%7dfS%)5I=phH_r_E3BpZgnz~<bAem%NJ$EQo7yjE~F0M;=bee+Z>6oUlk+0WKKP<7>wqS@kY9G<q3T|>?oz0tCV!RFcyln@%>|S6L=<`+@7h)pB?13dFyO9?I=TLi>1Cx$O_3>^2xs)_m>q~xy`%Nt#*!9R+6Kk#Xbw!dkebbX-s?f4qvewD_0!EeaUIg$6NlWkMyhwgT-cT7|Ke+xW4l_CE`gx*H1~MgW%NnRPJxvmEAa&8Mc~(yiqz%l0~gAIgBkgt4h&D*!wsBu-`Y;4+6=OH>EUc>;8$V*ilO}i%0F++gp#*;`w6IPD!${Qqo`U@5t@8u(2*4a&>WkDE9bOo|(rhwm#eY5|?f2G?O<<6M3T#o7?V*B<|7aCX4Z^L_4H6R#k?294rdyG*HMb`M#}szn8QLAfnOq&1_)pnr_O5ZC<}jCQ}u=XRdfimC1a=|Co+iBauZh+)fU*v&@ia&OK^_M4!BQEJp^pd9o%MgeOlkpk35tC7sVtE*{2dmC@aau1c~eOR=bESnTa|N;?a<8a{ERy>QM$bfZ+z%kFE<c)uE(Eei#oW0Fs0)8?MmK@%p@m~+<2WMUoJtcpZlXDtsZ)|<^klXSI1QbSlj+-7AC-7;S($UOF9W1Ot&_ak@Y(d*aEqT{UcM$(ogDo(GLR3?}EcqHz2kOkW)JEaGSPPslwjmY+Jwu-q-`R?%9-Cv~FnN4wE(Cz~Z;;|>GK3DJShK+V7U3i~N_YcF~+VAy7i|cY{Q;D?}-jv>FbxjCu+R#d;+xp!$9=5r3wT`kgH7(N9Zf2e6L>@+63vpw%j`!BibHwBgIJztAY+>!?!;TOg(CuP#qA(3B45M5=ARgsyflMwRO8eWb<|y5-s0rQ7(5#qt>us@0`X#odtZf<1OCDoWy;230`tVxOX=aKEgJG_g%IuBD2sN#y`_*+lWAkp-&+Bk6rzT=meQ~Ca3?5Y3t!`1PI&?;7sAcYpWj&eN76?R2<u(<SCbF%~pc07L^E!zolDgMUY%*EX)KFC?`A2dl?wT!z?(V*J1H@^En4C4$*$h=~&Q!Hr#L!5lP5tTisU%q`r<JugsgD#!Pl>4~ZLX<eHb;7zaz~@(O(kR<+RT;5pmwov8=Z+jZT2MAS&CCy{iOEQxBHmc-!=tT{mF)yXtM@qB<qaP5sy)se0GYT_LXwLGRzJVy?EBHlGHl2?LKOZME9=1h}7wFmC$O{5&Mz`pqLkcDzyuS<F;Xmd~TKtL?*ASb_@4Zq2vqK-HJUS9!)0}nJ3h)Ey<e2*K@1-`EvfgEtO3qNl(4%O6fP&L4!8d;}NO<z9Mx{Rq<k%C0APU%&A|Q=tk|K*>FDS#y4YG!%>b|GE!~XO+VzUgSbCl40~1dl5({4c~{T8@Z9*uBkR_g$`r^zrB{|tidHk#ky%{!Qgb&db;HwpK)mASgWB221e3MLd3vo7)1_rXI*%p9&2G5sv03Y(ea5EnJuFkI$GRoI)R$zc^z*Ydp{AZ%l4Q{vSxa;6f}<~a^mPK%Q!4-Dks7sI*Ct=~4)1gGLfW>ITBB2$Vn#%_lTdRSjIRw(u2$E)k~C}nWL4}9J=!7-ZK)TRKE#48t5mhB#P^Z<!mUUZ#Z||8?wLmY_p6AwJl5K)V>8{P2j2WrmW+CH=~z3}Yi#Djl-FiVy0nhMpuMcj)bWg$Z1}B|e-tQeb-Q-QQL^fCtCCm;E?&cos>lrSQCv*d^NpgU=$4f_dBFBP;@vi47)E5x;>JSXI3#q_Bhkip2D0o|mLd|bxf@p9`-YYIsPC9j@ke*T+HNS^*~sHe(vG*Kk)qEqH4e5nU3FLOwPz*TVAv90RE8mk%tGBvR8`s1<*8K^bxCryl7LIJlA*{<!PtyMfov2WILF=x;?bzvgmPg;FceRR{i)v0TXvXb`?*nfQ^}C^N^*H`QrK)myJYC`Zx-vatiBzrokm$|QXV}%sh!<W`*Asq51Q#rDNyT6$-%I0-wmZ;rH8r|p?hxT`;JV&7J4e=O`e*eRJpNCDXFa)_PN|Q`LvNp<+K)Q*fI<Gm+M-;B@4*<%fTY73R|lC>0{*~qA-km7Uj0vpA_#6icK>cs@&vu&fW9+rd@Jo2%R*jw)qBO#dvrVOAP$`O7(e{+FNqOqiWkrYdd+#BNdZ)U0btE-Ofq}?o7qEEVeWE9?uqV@ltZ|ur=&fyP#zn%ebb`-lW^~c(3nj_sP&AtGHinDjh#iR63GSPfG(Ki40OMPhC5TH%5;{-q+CI7njk7zdNZ}N8{<EYwN0)yrE{AGJ6NL+;ek5G>awsQc+KI)2ivsQkx+qrR`#vZ-sRRV|*iS8*=%*$(1q{65h2s*Blk{l82aDHVztct8vzrh)kTGt|h28EfbrPjqG;Rx#@7=d3=a225KqQ>ukDWSE|+DHI+7h#N(*;EGxp*nPvtxzrkiCsF-!_i8b^|r*oWFFZR;ZEL5=@b1_FimQ5R@mM4>YR(~ocoJ(+hbo-HHz-$dAwNqcNtDY(?@fDG3M5p$u#gs`*A4vnPj}8mNEip^kI-7DtrZ88tmg=K(>LxZ8cVpfk(Ducu6BydXYR9l9v8h8(&8RV}T2ZYUE!k#6ZTr?#ZW?+jrAOOy$8$?tYgjE?!eYy|So4v(YR_shn0@}|>fGclb)OpLYN0W5lqjpZwiD-_)`)B>RXxwBrfjJvW$j!g+Ev_dl+TaK&11Je0J6r`v(G0}DyPJ>p>j55U%&I$YByGWzTb;y61nC&I9W%5gcg<rH<qb=oXtq1=BkTm2JG8H?_L$Y_iV?i$(~r>>*j-Co4S9bw$XUV9((L9^{bj>Rt!iZQ(K3~byG3vwtlm6J(a@S?w%?<8_Lr`N@-T^YvN648M7(O194NlEQ$LwYsp<%CtIYC*wfu@sn-ux_YVt&BD?Gsjk%VpXWBnJrdDcKSQ%124~EKCW#1k8=JAeAS}0C!(r!mPo0#&t_Ed3SUA3ug*$^LR`$}V_+PkqDEBQraJOHHAqNL=FZ$#&tUR|VL>a7w}zqe~A{L7UsYM1!?=3cnESL7tANA0vRYpJr2yOGmUc`7<{)nX$iEr<MGyS<oGN;0<WM)g35$yL`x)pL=cL%lZ2#?`pKGBnnA;|Gh+Zto<VLBM{L4%5^`c`Od6HX?4TLPPs>{4g!e9dvnFoKFT)yEHnmjlHIRe48lFOugyUri$%G)@ZiYl7O4Fo+&IU8_y(Z7w6k#P1;U)7bCkbyR~$q?Z@0WNblU<SUNo3ct&k|#(8fquadpG&QV;L{Z-4R;3}B*eQiQnTW;OHNq1DSN87=5F?8QdPV1I{LZ{6-9`#l4zH2K4T2Jcr*mNJYd7^=OuY9vHPVzv&v{oO81rbb2p3@<HF!@}IFM@={U{IIL<Z{=lPcw;9&7!cXmho~hvs`prYPY#%NUD?dpmAE5+uVj{+oq_Mj0(L_L??CtTHhNZ?ov5pE9Ixfyeyb%uZPdteY_V_QTCR>81XmClBe{pJohLw<;EnC8=8pSvtu$;Z%VOwPHj>UzHuh1nvLVV0ohkh&7*v!sx0*4HFuUaXg8U**W4f;EX#Ir9(hVd)4FGWZWCP5_tus*-@4!C$M$6Bu_+lhZDC`#8!!(Nb9z77Nc-lLMH{n>3)-Q-Ti)-Qj=I??d(2MzL2uF3@0%^;vqHMDm1ChnI&jlb+md>1HV`ovS6e4_?|CMSE?1@T&?`OH<JE9H9>`G-GfR(d*R5Syu&Fc6Jf^gkv^R6seK%F=2bIB&$CGtBbd^AR+_?|d+HD}L(AH>TuO9<dyeV7s^(j)-Re>ED6vR{G#Hnhn?$@J5a}aSR1D&dpP+OfoOT*|3P041G2(`xs!{{dY)N~CtW<Axp@!KZ;qCK%_D<1lz;m)%s^Ye)_s-hjGZqq}KQq4jy<5B9V?R`~RR(m~RwQ*Bx)H}_cFHOm)VlDLOq}Sr*P2c(K?S)j2k+yQ`_2dj-uS}6m+PlfGs9|!W+jO;ahn?K#4HH)<nyI!{nYCBB)hZ`Gb820kIwXZ!vR80+=J~Ng9I(f9?pfP3nR*(Qc}N$Fkc2dBbOd{IeK#VFtJ_hSSW3H}wSAj)->Qz3KBH<a8>9=NRa1T6FqJAHgO*GgYvxQw5>9K4bz%}t-q@3>@wSk93~!%8zC@_q2<8`0wL!5PSy%Qz<Bsp6OR;NtliWV7`;YydY!toE29tr<y?d_8NmkXh(mk(uv<*dcW+F?w+NApET5n}H`)se`x5r)LLMqix_Xba~vhS(V=#`1Cx<Ks{q1wYuwMfWz{?It3>sjmC{5n^Z`rOHzzR#ao$W-gtHu#uWQ?Z$AO{FsW%qBiJs|_JD+1(f=MqT*np_dFv8pDpQo*&O`R(n}j5$m(6QhAe%_dN!?+qTrNYo58&*fv>uk|*DGALw_JmQh(PcC;tYija7+@Kts`Rd+J5<U@4wKCPshEtlUl9=a>kwjEQ>I?5;KB%HKAR6Eo_m+Xk$`uc|00WpN`NW=j{(yKM}!p<lEqE%niEB2M$Sx%9(TP#T$C?id&#uVUPkxnBW&TSPQDkq^#345-t7}UC5Q+?jwBw~4mNw*tERP}nxLAFaPcObG<1y|uwN^A@}45pnmUb@-NhOxD{RVG*Ev|*!24eRaLJTh<Bq_M1aG0c{|KqR5kVy%~oj-#0}*_D=L=3z@pJY^cmz1#X&UW$jta7^uXFVba8#cqF?b@r`EtkoYAgH>VOb&FfIFxgPICr)+8`=qKZ?9<xNob0G<;(jNwBj=t<W9Llw4EKKbF4c`$UDU7?a%ipnl5beEx+nX>w5+TX5m{?OsMeWec5EnFZ|ZveAhy2u8E30pU=m#oCpQl_J<tC~!MQap2Zdqyx0qzeG?OG5n+hQn+7J#Z?KeUy+v>M}?|0fvGs{}*x$o;bAW3)kL|nnqUrKD1^9QI@uE%M9g-IsazwM#F0-Y-(LvU+cz7y+HQ@VFt%xcrDiM3BlWmBK-Dv<c%rjz9fL`QnxZeOK=WHpp(ReK;<HXDqx_x}0Bh*iOxMeQ;^y7#CPbuePXoZb6781Scw-^|S}b*?BVj+Zorf09+t%SIUTp*Lvllf+@Bheo{WDEJL&Z`X(rZRKYd-oG9P#_2R;P?}r_^DAaA-OJn6A+=4tZF^EqY<`*T;@mirbK=lfKFweI5<c1cBA0XfH+@t!fA7pIp&CA0Z*tl9U?11lZ?HGWX@>0Yy3sui!R!1KQcvfwQ_PVYpvP}mHHDxRpJE8Rk3Mt_f@~!Wj%#7NCOEDSeN9s6(Of(^H{14{tu={FozmBUF+5W_-TjjR`vqnT_K?fxarRe{pN_}Z`sT!7HalsFw`%9ayWS52tMzPI{ReP3_lMqY-4o1aX6PfW=J)iBss^w$RR@ilb?V<Z^%Y#5Fo6Ij$l<^MXNHZO;E@5k_He){JCu5gztQgnD~IJ7??2gIe}FEGuCfnq<FLqjvb$x}Xf&%$hllPF+z9)*Dl`Hx<8yhr_1)p)av$w4=$^)Zt#DsC7LieHgZi`zeNT<L`u+58eX24#V9hjaZE>WtH_sDc*Wugugd5#fNyh@C=Lk0n4hQDP%Bu)_^(+Xz<U4;mD%YIYMVii1o4bp5WW7R4rE}eE&ryfzK`8tY(y^rWxAj_anQtUZTquLqz+Nq#D{VRFt-iyJPjfu1YEnF&e3-(ZrAuQKwpNSv+2kFJ((M>Gm)CJL|3hJ+%eA!6zQ*32syp?vE*<x-?MqpIn&f$ntwmW1%e28b&rz{oS9X#>bz4jPL$8W{<d0Ite|;uD0*}|}?}zN}f(!cFtf`lPE%ZvN?f1A}azxrgcHGuax9NVe?t!|*Tj|RZ1DKCyERm>x>^152jUp`NZ>90aboGmI1+2~r$-tsxS}3F~XThftNlrf6fS>gNfvUCA3)@#HLrZ$~xU}GU5Ms|`4Z{b-3}8SZ4e5Q)qBy%alubol&@JUXRX2yWyLN?bHF}~W8|$IbSj~43^Bq&4K~(9;FXX|rTF=A#z(l9%<Xy8G!zOR^ub9GXaq^g#C5TS2ap(|w`3Umav?>rQEAv?toDIf{jk0(j@8-aO!ulU0%k}zGiyaytF(%ZmyiAYZ63r?3A;yv40ijdKMCJ}dwO|zg^}aGvRyqL8T%nH=w_<A4_9Kny&gVbyP+C&o&Q|=Q;|PcM1U#yAmws%7{Di_$J8mKJO@DpD6EN9tZ`s%k-<&BRBl3)hy;iojO;DWG45EJ;WZOR86N%li;hNU&wHESmQ((d+4)-4oYIQnrwDJb|LG|s)vo&QYiR(8sM5s^4Rgmx9B)Mf<^qWu&PAx0dkE65}#&%@14voo#`A|3Mv#hsIhqHW}5fw;1T}gAQk%Dq&o_RHlWNCGkMwba_p3fIhv5L&-@w>*O?7lDTDa5+-`<bkQY}in=*%J6psnKXO)+>~&w2^S?FD7iK#&jFTdEA5tm$`hOrU-w1^h#qboW8#jZgaoQ3iBvWo#F4`K~Dx}HneY0W5g_vAl$V`X|^9S+tr5qwX1D>TRfFTEuE_k{Ad~n#R!{rbFu-pxqjBKbFhPpvxv-gJ89b6TMoaDl4`rA4ObtGoPlY3@QtATp9dz3b)&rOnnx@S=S0%)$n6TJEAJ+!FCXk*u^!}WwGX(a?zd?wy2}7Z%0#^ybNI$zk^r9zRc$7?KDN!PGq+7{lKPs7J)l(x&$+%Dwln`Z^TWvw#r1a~i^j`-K<#`f*nw%7jR(u8jXKIu_V@>P`sX)W89!$SZe6ygBkyni)Ayv9=A`_}xn>0PrFSOmS2Ln#oZRNdxNaKWIfcJRD1!HwUZrafDj7BdH;?(LvU_MC&U)MB_wpB&zW50zwZacN#C>~Lj$8R?d}nRCcRnjQ-PnhD_lfLlB%HgSV<5S=;rgnrSGmzABukFl1$X4N4#$$F7U%n&`npYQEp(?$%F~NE^s8)FQmX1Z`k4RNbC@h|K(gE*@H%ZY+egQ!UEroL@aX55sJLG2{jzb#)vt_R9e0en8@6hQ=WuR+S?>))eWNE-Eg~y->QfAyu)d73mfF6m(5F+l@}#vxZq}2jAE6yE#i41akAK1JHb_?|ZXlssfI1C$X2bg)Ltmv8x<cz2wy(Y0#X`>0UFY@&5EuSjThlsWH(<87HjIaTx8uIjlswgIy}$5b)IrrXWcZ@myIY^~=B}e}4Gn%9agQSjl#f?O?@;00kUA6mAKG^JqbkxIr;OvN!2A^fQxkfHa(6z4IHSpfD+qFB`V>O)YY;0Haz4=7C*;#vSHcn5*}fL9yvsH5q2BW1Yn%B}V`i2Du?&JxIK6$KKizx(F;0`Cw*!tYjz*W3ZcCM`H*uuDP_M1mCM7ogugy;6g;7|v_E>QG>@5G7h}`hMCNWWF%<u8sweNa^CIVg3s=u^t?a!%G4(i@gu5NBrgRC$P<X5;i8#=wdCEYGwZo`l75QWRm232|*M6;TENiTo)t@UQ{^L++eVPjb|)r~r=xSJ0!g)%P{el3NuAs>g2l_Q^iNSwX9XU};s?$j5-gYww=Kz%hGU(s6s*_iLzDAR|`(ss1Lp4;5V;}8!nkI{gAYO}{eC%D~&SY7Wm4QaOyJtw#%`*Tz71rXt&*vbr6`b{Z3_A=*o!~3FHSq(ZBntSobbTKRBj<zW$`TUj)=59Dw`Y>y-H^Z9PAScPwY4g4acbS}_kUv>Jgf`p(&!g&<gWu|y>WvZS$6MP~kdq^SpzXnM_Za<d>Tj|dl7Z1G++it;gV-hK5AOz8Z=dd>UTnXDcp=^(LK2<rd%F@oTkhC};xD|&eD(S47F|ylk5;@G9hWkz{>A&ApstMd|Mi6LA(={x5e5suo2tN9^{`!-88^SaobRLi{nfFvJFrJlGx=SZDqyAVWV73knZww6*S`09czhS5E?X$gSqxguLv8RyHEk}u@!x#6Aun3et?{e9)*tR5$!!+d9>jI3fOIjg9XwJy0g9{EbFW^``E63d729fLJzg66yKE4_ASS#-80ozX^7OkMf`*L0jMK|>8KKl$gjs&P-qO$Ph~i5t_8ha?-Qft@z`UVVh1yG#WgCa6Gq+IUawK02aBg^SkI)`FFkK_o!>G&1MO;7lwv1WdJpMr$i$3`&_4+{=UED@*TmC=-ih9#ZZyXfeJh(pm$NNtS{#aP#9{1%|xdz9<;F@01my!71QRAaHyih3n)64pI)@+u~t2d!kq1xzS%%7$m=Rt9ADKKC@!kCRXho950PM=uASjVdF*w9x-BQsqTi|0jkBw%Bpa|vj_>u3<XTTqnHhv95>9vA|<cEwWH$V;7;zJiV3#{8rA?hQ`m)HAPwKi}T2#j8BrjbkVdlZ(o%(*RB)Xe7rvyRf%4cTh=oz9?6RBk%NQh3qDvWaXxGcfZ#H_>}S<mB`4wQ=4L*&Z@`gne`sjvpNe};bEbSlX?G~4LDdB_BvFMU^DEa)|_sF72$x-UY{FY>T8D6$h&l}o8IMhBbKftE#9>xTJ{G-$0Jht2W`9Ag<RH2=rn!{?p<CU)LWhF7fBs}+M~@eYNvJ1$<4ccK1}pmy3F`Pd;4uqS(}H=`ata>!}xW2b??bJlQabaT0qy;(QC+UHdswSq$kA|Ch$qwOL45_Z5L0GApK~OdIzl!6LcovNc*VpL|tZkPm|6Q|BdkETv-`CbJ9v@x6{{|j_tZj|G|?jBPo$ii>zIMX-ts|{zU?R*r?U0?RRj|+mCf#9_&{Rx96)NR@rxQyW%`SFBWyzzAUw83*?nGOx}L2TZ;>rS{ZW*mw)eHW~hlB7|t3`!J+4s?-Jyl>%M~F_)rOL#<P52u)8kjKrb90*y*WKVTzSKrSBYxzMRKm+$3>#psJ6*8|S1b_IOVEz*ck-0PQ~~<>k}wU82nkx8|$3`FD!MzE++!!dt!8hOW&H(<xjL>RttmWj)@Y`XIWHz{!O7{JJq5aryc%4P~H}?j)O}8`kNfv9)=@PU<~m!@m13`0?V@lQShTW(b>0&U&$9lU%RP?VyQOE*RR9DiJj4n~(gKg}alrYpuM>Jtwa#1ijH0(!`@PaLQ}gzWzSDYzZC8V0dnMEnsao?Mui@T~|&Rsa9psYwNEP6T^#@{`GpP5a&_iu1r~!sBXrc#?n;-Si2a;o`GBDb52OEXLQCwE4P>nMl&da1Klx|J(0?Kc5SN7=e2STAqwB!*5hV!+BCSds#on!HTEu#hT7SK)y;5Zhr`?8?_p+wVRH1#A9mx*YgfxghZ_b#=9k+V%PzB?u?Eg8r*J?Rhwlq%z+>T1tntt;JR?=(cXLx+wMWg%kUn$>uxGT%>LYl+F8I#8b&R%J-Z!6svu|6(h;JolEMCQ-KFH-mcRjwEy51k`yI8Z9Y1m6hPxHoo_j_v14Ya4m&eIZTEd#{SX(*gyfRrI)qqPqox~G?aij@)VQ43lwe33a`V|_kdlF@$TH>K{`R~!$(x>ssBdAEjbr+P0DV*jCUyEJc3jXXnbBWq~dVosgmra7f=^CeQu_kP8zPrO-0KGuWFRvSGSW(w!y@$uDT*8pd-?^irtYJss|Bdmn1_Lmgz2Fn|E+bBhiQp_W|gAR948Ha&`gH8-i>=nb@xc)Ict$3tUaT>=*be>KATH3Mlck@{>KCtRUjp}E6{38tC&08l6_7=>(>1h?^H@|gW-{r7N_k@OYmX~aN6puCH9l^x_#jIJ@MZ;cJpZ)D`eHM@p|0o?Aq0}TNHIto#rN`f@PqWiqOqZqeM(4zGL`Lx|{9DiiwbA}ZvWu&GOB*DaxcF+W&mhW~^|B53ZEi7ASGnMjw&45DH>AsSlYOjS&1SQTYFVe#k30CVb$%XN+`Rqe;>UAN54uYS>lgg=k9+=L6r$?2chMn-+w-IGRk{9V&{m3hYZv#V*9R2chmS3NX|j(m_vxvR7z@aVpCC2)%;zzhjCCmg&K*M@96<@TxbC$GQGK0b+gY6uC+oOHHX1Cb&-TuQO*_4*G!}g(gLdMaTboo5+I;W5L_Ob=)wcC%0ls03y>^=qx)Qlugl5>55BR=vo+5X-E>_%IL~JHMkre&{qO&NonbwQP&hIq><;xP@>BYW*S6vkvFn?;ZNO~pSAur5`9K7p)dm|+y7RP9hYD@6cZZ)^6vF@!(r!JlrF9fKPH{rJ_OtCXQ)`vGMmN@;V0_>E>vQt1Q?S235U!501UGzI1HkIDAa8?kGEm?llav$N$jBL8QEFp6nwFmXa<peZ~8{Jvr*Id7=@Tv}vhP_(09UdHdW82{%ybD8p$z4zgWPUes;E(57%1F0o+77ef;r^><T5=h$=NE%M_Qb={K6hEVykqJffyN{8J+BU24X&v|=Tnl^*056k1xn+)hlw31f$?qM9m5_tD#wZCxAHEsEov>GStEq!llClnZ+-D|`&0PGX({NaT2lma`#Ws!9rrJ)+sN>>A2GAvD8icz?Y}cd?~n<ODDP+#rNNr+q$&gM0dD)(Qs1Lg29w53Oe?evs{0xPS0c*ME><ucwSMEuV%-e?;#{&$nQ0)+S_9#_Ke6{$q}<~99GSO?;N@8UM}CtDOx8Bk+=WitO1m-0iRH^(36uMkD<8W(Zu3-yV(v6pUAZ05k!$_sw(Ag`aqB4Sc0xQ4%d+O%-pPY@3m4-LH$2f&?1V3#r!uZ$ji)@ks4l#)a;qSB1u6@C-3;FEy~Wqd2!a;xmOJ)L&DyQZR%azy)t4vASy<DNdA0q>7(Bq}>#?t$p*Q+^73}^5;l)xV@cHd1EI(<xd>>T4@@TCX83^<I?Htj4*!C9~Jf)H~RZ!i@cUpb2&CS3E9O+F2&z(eUfnL<Tw=N!L*3ZP!zEmn!b5rU1>9x;JO6W`Zvy9NM{ax=n61{E`b=9VqBYq(!Z@N|tqaOfQaowot$KPel=v)K-64VQa;LWofqN^S)i}-o^ct@zb*|+;*Ui-m+8RNuxrRs2PSY(XNlSvKMzP2D5$^J|bWw;#rpmV>emGJ^vS#^vpla=z_LGyIEYmaZ(NkmSYL0J0;X?uzkn6rB>YvHOWwPCVebT3dZ985lS>~c{BXBl1TJ8(E`i2Wblb@(Z|S9(m%IXwG|^EU_l=jYhzXYm071z^!pZkG=GE>)ymebr7OH|*yk6odQS6336#-1@f7*1lskD_F3-RKZ!B$(xT>L})AC=bddUYJNVU@7AQ=^7-qEHvk8g3!Ely*@BCx)2$+Km)*G8S($I9GJ8l<GA16?*SfY6f8vS3DD9?yJ7YS=xTjP3TN8N*0;iK=c!ifP69(SSK9TV)I_i@G(7c52gXOMa;XU+<$mKXNYRYOesML=1NguA*uv&CSuoBO%FLAT1AJ$24z_lZm?|Uw%sD}IDe#bdmN0r9T)PCJ?7pp+zjBW>tZ_=c`y>w4V&dW&hx!nsx;TIDZP}igCR4=Q&V6XOeYS_KDgBs-ANvgYqP#G$z(3Se&LA)YEX6MC|)7R+VDwvP-)m?iZj2dA3WK7WzgUk#$Jum0KgD}q`EQc-PgVT7U7L9~?N!0kgu?xiV_%@||L)D}R6FqML);=sN`_mfsS>HFRUQz+&^26U2hB>Y7-{M6~UtV5kbxtc4YP*7ja_S`RPnGCxDMZx=13b%|!0SDwRgW&ccD5^KZnZID9jj&ZCTBZ)44vc!vHav2@<_qqd+Vnt-8EvK9sNBfIHyr;=Gu@fRct44JbWbi_4zVMiFmn(QsQpCcJtR;wZ1y)Jh`9YQJs)Zw7f)X%zg9fw6qUR5%J_TIsetVTAWeOiC7re(tlw8x8CB=EuM}*cs|<L#&>miOu!NU6c2~@aonZ+I8hq9+Wd7@q*tftp;P&(^G_vR%G!i@*=W2?wBjM$&>!ZGAB|_6@Spj%wm(9b`wV>bRW7YB7MrN0x9fjYvF2uAE2kUt&7VXRfD`|)k4C{40-O<UcZQ$Nk>AYNP=o4?YH?TooQ9ddo-&7z-repDc;4t^GRN}vLhw#Ndr!!kM@sRxte+3flCb?h%K)hJoYHpvM(ufy>)k=CcBV{h(aV(Q_tH1O{BtAkF9E8mXDZIyn`@O4yH83B;H|D6aqRgqqb-zxOIef||9eDVZBQ5bTIamF2yNpV7fA1SFwog_)$Zv;D>{Ry(zQz*r3M4vgA+7P?ceB9zoIo-j8o~l6`Ac9xU5HuTANYro0@knXT27l#|PPWC#3)9RgGHnd3$L`S|2CK0G`vZ5zB;mChJvm*h~Hrd3Z(&c}#w{O4fP%<B+81!^yVQ{79^Lo9wE^Qpww#C^y89NJJh~J(I|Ceb>ZSd$Sq~!P3nBg5e=odxC3WtLpn4-+qej*4AsoHU7qRtK)Ty)21Q#`;0ohvw^3R>wPeM`^37ivhG#dxtzY;KmWG4eZBqQ^<4Q(Tt`HJTpq9UQ1hA#0_;~M-W=<bgQ1%6IS(qTMA2)pCJoS4j)yfe;YWkP9(i0F!QfCg`+~vPfA?#n$oqph?_co@8qa{+<9*!zO^56p``2H7mL`J_KrDC^Ubz1HSjOIIcK?27^>*`-w%)ge@AH>ho<8Q!x+2c7Q*W~E4K8!P;=--({tK0YA@gUKimO&WE3O-DYk6RbvcHur=gY-$n|gQh3~ny7%c#VRH@OP8z!iL)RQhAjO)WTCcI2N|YG+$-ekC#Us~1-0YIK#wzH39`2hTUnrmtNNKi7v9KE-zC@6;Pa{Ur}jkWwK(Vr}?q3bW$Z#J3FKF5{-LW|N3oLe0xTraCq6r77D!1NakmxD&I5dxIxZ<2RIJsMC(X$eNGzU0zp3{iGAXC6Jz%<@gh*9h12plk?y->jSOH+`ZWoAga<vd3Ts-yC_tC((gI<K>4JVNP7<e?ttHGT5fl{cB@YEfau-F=7^WOL!^|hpZTPbUbn$VAI^Jim#c2)?xjIZ{N8cUkblmN!nvI+CZ+EzI)*(&Q#<(m`QGN$lVf0&b@T)%S&zS4)AA#-VFb0}EteJztp?+lVc4!}Z9R8d435^y8=e2PqMuf(zPX8ijca51L$;1R>^D?%f^MUPc*#q-4z=S6b`%j}1tH!Ke=Vf!FGt~(DG66OpVQ~4^Lyu7mz?`}7P(XoJ#=7jxd@v-dUg859_u&_zI};L@Ax=c;KGe-RTH(~k!61dKWSh18JNyG$y*Y-t=AxUjNr+WXK$Sgw4X0-d?&B8+I%u52Lgey5H0UFvsu`FyP+D_w1ZvWEN-rkG(P?H+;1F5@)VGgYh2H5;H7+G-WvHdhtUgeRT1s&xi>pYnfcpZ<<0XM)tdY3>Vo8L&j6$g6^w~XDWLAG5-Y&6zM^HNL|&Bk*PHcQ+B1ZK{%{(DNl<TYxuQ6>a1IuKE$=I%p_#WBvGx@9N-KB*;i=q#!*h!%M-hQ@oBp9}`LXo?L2-PPiYXDFZ`CSNAsh1p6>@Vj1Zv2Zeh2qk{XqDYS*pjK>qqjI=Y0#?t+I<_e3mwJM`q$<4Wk0{J-6mUTqoK<bnlR7_na|TLK|N1o9&u%<*StG-VQZOe>Gl|^=_{(btXmP1TbC)7Xx-X9N(r$AP%W^bxVDw?xG&O({I8)7x@TyIlzG5|7!WCJsyCBp1qhZt@X?G;QKcRU2%0ey=s%5;+Z!m9inyZz$ZdydIp(Im*!$>bGR_c6xiN`Odi*}aAixg%L*|l5O{<}Vf^^kKd2DB`$dmOsyq7`MGgKHfix){lNMiR;ZZ|aVWnKHPA_+vd^<n<BzI0B`dm@?#LW=%pkFU=VAmUxqt}acPiO0fR+v}$+(r_i&HATvK}}rrE>P>L(X+%(M6da2ZoUaXK(F~%^2r2}b#;?|FHtCg8$ilWjXE1BZ@+GLp=CX8eda3ms`Q+R8$`c|iqktC5G>I0>bQ-Ji=*#rk7sM9TqZ%&F2!%}y(}&=74#0KzBHn7*$>}M)_A(i+`nrJg1ETjNh*5-47emEw8tOgk-vyFTHI+BKd<$@&)FpVr?{-huy51g{?JvbT+b<HB(v#N#=kE1o;kx&mR<P`p5nc{^B5LVBeoQWG|tQ@S+?QC4<Rhc`Ws0|hK`1GhJ$HO_4aUAIq+wAag-n1%U5z2dwO8Pjpm^130r)lX#0F?w4O?#z2NrrfLCkT0b@QA$$sA_rdpk_9Vz2O+#u8qBG!v+M|>%4h$H8DA3a?mu|odh;Tu!X&-)VNlT40Mh&yhoR?-4pQF*N>8Yl?#SOd&*Agq~LPYNg3{V7YsmRh}}mHMK0MA4|<zv;+(Fk1X&#4OxK9b@fh%PExk&u%q|1xZl&8x>~Psv&5t6EmsG=ezWxShH`_IO3sHA1-=8J9<4@3p2`U$)6EbjIMat(Q&RCorF-O3Darftyc*i^Fw|A91!t7zMq=*=P=thkk##Hei6kSv@XlVu*vt7C-z=MwfdyquJ5W<BY+`Fp1kH#YyTTw>hF?O2V|<D^#`Vk_u~!67yU1BIk~>vWHRz<rLGxAjm;z2sn_M=e({Z!$*SIA-s#sSRy6J#+vA?S?Rme~dBvr^YMvKwe0ngFct>{^tCGHc(vA1sUYEw!0Gg}noZX#+%4$Ard|ERvZEcZqzI@rj4mW<$f}>MG+`V$6`_r%JIo2;xU&dC{<n8XveU9pnO>b}l)<ShA?k66m?R1&#kKZ3MU84E-^t-qfbSQSRd6aZ6fpud~)Z~anU?yF(;*i_*$~PZv*}|#N7OiaG_=VarWGX(Y#N=^oxX}DTU@h9XXbhT87C+#%-eZGIa96kYCR?%uzH2&mxI4`8B>qE@=8^WKH59*$gJL1+609<Or#Q>T3RurN>v^|Fyd*4-4Pz}{b}jL04QnCQ>wbd<U2P6Aa<zz6Zn1scJfR6u_u<O7SHS!;*vZJ<N#n)zI38WGe((MKgeK2kR~wl=>%w56is~a(4Z^GTqvh8Nrr6c8>rZg&1L03@LF~g>z9V3hzcdnWNb|ejBSqj%yX*saHYB$LLX9lGiTRvVm73Bcz`quI{;=V<wLBVvFuw~xf*aK>7K$Ai^H9y{cbx95J*9^0$K@;%@BbfjoPFsN?Q=w34xN(RPWooGTb`}U6xx`L`iFvms=l+yPsW!O?=S(5?yMh8hs1pPINqG|)0J3D^{jNiKem8Jsg!@l?a33dnam{mvFa=MC`v8A)7nqpyUp--*UqEhnQ>S@4<zS!aEI^#KBf!5bL=zhRd~PeiOF`FT46nCx-#9~ZotTmp2Kkc?i|m!3E1rE#dgEN<=Jf>&SBjpYpVR|*@ZP-Wn)4%y2<d}>BHMm?2jdA4MJCClmhAY{TVOjz3;Gnbu-+d?I^jsLTz@f)QVsOJq8F^V9Z@R5Fa_O2*$p0m(flex4&p0M#CyJ4DxdWIDrtYAe=svE|0-$Ou6RVBlXFvtt2VWGA`X;mgaiSFN<#dEJq(z@s|DpKZsN3{59&I(~Np#!uMD-H$A23yQAEQPE1KjAFxla&kt?+X|TOEGYp-M2-5%<Yj}3=#2U-Doa^j3v1EzVd2Cq_Vl@LMH5x>x00W>4Zx*iWu$nuAdwN(@Qmd<*Lk+u#9rr0MGs?t46PTTuEjnYaLsM6GXeQYh@R{S|x^o#0`-IP)snu^p<_~42K8clhof?Q{RW^df{EBj`$Lj_oKN@3o=E9HK6EU_!C>!IA=CYtWPh`RbGdXTd)%4!!ZHKqO2RW`AB6)|iK%3I08mQ}JS~V);!70dJ?6$n^Yu+LjbokKETjw`TXWdlEkCnn}0DP6_Ew#l~QomAdzq+F@SeTm{73fW8IF+%^k&hPCVmqTDpEDlYvGx4Kg=DnXrFeuC@HjrKJ*iH>@b&;w^10qW>1NR!whZ=uw8r<n9<A6B2duuzWVQIvm$)n`qNgfb<nP8;&-E8Nv&KB70>rcZTP=M!E{A#fgZCU_JJ*kQHiv~iQ>>+XytjJZu~@43x4r&0cxiuI+#lzpL-orlaH`!E`tvO7B(=~WXjE->Fz@r}<4z_t`(1l_!oz1%=bj)Rc8l(oL!2ff3@R_=DC6)bE3@itV(xCtqP=*HrMWfSKp6n&=+Le1xJSp@PJn7gjt*$^sr)uvZHL^g_G0~HMINgm4ei@VmQ{pw^mUdrTBCZwa>f3_l0ZM{E>wmOw<mwR9#;-3+=GqW9FIp`wCZh8zA82cRSZQ$JqdUFQ-^g|u{(mLT_dOHS#R8(4M6B~`JSd7{Yq}sd87t<vTVU?i7wOHT71O6D7bx%!Q*<IqULsW{NtVtzgtaTbd|@XC-SaS@~26?rR~z0i-N2C`QTFo7v&geKO4xrGOZNL-C0zG9FmvN1%C^We37uT5O-G8;Z)3@qb6jp$>aeaUGf*m`*wtbyv@0f)qLLbKXaRPZHr&D578znkX7^Blzn<ZFh8a<k;thc17jwpuAjbe7gB5OP5T+QXs(KcFJukbuf%sUZ&gz=_zn4CH(Y31lD?XM<^SkcE6k=HT*F<tQ%89vBBNuyu|GBoT%PP{=^DlLt@TTIX*Tp}s#+FWZtTF#VNO!uE53RSJmJ-{V1mO@)5%Mg?uX||0PB(W5av8D*K2#A_178!YAFaTuS<(Kg~!c?9q1{)%+``yZIqpcyZ7+;x#%$8)f0#3C_Zh+v(?-gVtV7F{lp|$0tbm|gco)s^tw9PgyjGkhCw{_6H}j>YH-;CRp4A*cTw@MInA2PNKwZ3{?S5)w_guDjJ&X~*o^dmD~?Ht%Y1f4tX+-=`?D@4(m?Dg-+Msf7p+uqDqw?_w_2mib;Sm&z8u;$qUr%zXEBo+tYRBZ<8CjzeNZq1*czO&_osz_pAXloiE3?5N&B{cT`ErY7a9ZJj30i=y~j;3m=|oN)^U#Ixcyot?SOKzXW0`?*=BBtIoSQt(x!51cid>%L-FB<|IDmKGn<d%EIBzXequeNQ(sSQ=%(!o^88^zb?w%ZnR&J9eSG|wyWQt7*_W8=z%i7K#EF4`qKbyHtt_T{hM}#;^VW$<er4CWZmTias*JNL0hCAQ30%OL;gQSdR&UJ*HM28qffThPcRrI$3`O5uoK11A&XM9PY=Y@N4I6IeR#}0Gq(4b1(WWdBDJOd+(uz4%dVmYdd^KlpmGr?ub@wSF?XKXlf3>>1!W$Q@@6>%BC+qRNuUS_)b~oDgdgvc++RaT~vgybo*mbesf!=O%xa>TTWF8)}`|drHPuOw`J~J4{Jm0z8t)W8>)B^`Y=a301TzBz$hfSbH7-k|WJ84fORd%{z5`1vLjyt2~K)^7>w~e+(;R3-@-g4NQ<r}_#!G7ZOTJ|)rVK%_DV=RO<H+sT1M60osJ5^qj(h|Y1^03i$aeZex1SXZ^1N()3GAS>s^d4Y!dJhJ2qfi&2^$dEe@zMJuCbuN&v-Wj<hVx=UhU4C#zBVUq7kyRJ%vG>wFQLXQK4}bQ<<y$ki{GR&>vb}3#_E?X(oKDDnEHEe@wrDq`q0@vf1T1Ff1*eI>x5ea(qu~e_gjAG!q@Kw{ADJ%wDsh&hr&AuY}Gp;V5jRa9!$i>7-&tpi^KH2Y?QO{0e1X%FGiQG^S%1AvFJ3zC->)Ly0=3GpAUp_9sL&2d^6bgovmMl%!IxwLh$2uT0TZwOBQmB*VOB(>a-e4MV}jPLBGH5e}T&}534MSuOfHo)?(dpq6y%3#_%<%mxhg1=>DP8zn{-Lx>IS^Ue%N6+tCiBk%e2OEENo5`&OQ09U*E`bqnqTEC@KZI*_TBy9aUNt5>lhtwPJ4tZiKVnK#1LJkS3;FRdb{^K64S)TP3y1zr81o!IRMsacYE2MTEm9-bz8a2N@?o5N3xaa-d_Bi!NE1~GBTU6%$)xvP83la~FF-O_@$NMB)oPVWnRlx|$nr>$(>A@TideL3kcNBEbwa6&VrH(C$ZLDHOzl}Jdh$Np46<XoOL%HR+^>j9nY5#8-Y&BH6lN-=*FP@@n5aP0|_s+JnV^gBZotj8>3kKcXN-qjiH-m3q0@agzRJ@s-~W9QHY|9stY%m&v*2?i1j%6C-RExxs6J9z|isy06-Gkert|9;(dYivToPF=4V<5N38H3mtRF<KdT$Fpt~K>cUcAIe{j{K42W?A>1Y9ehg-In-Vx8{pn9*z(LP#wyD3U5@0et(2YJapkmjf9m`j8hc?nM2hI#=E5-ePHn(H>!IQ_UYmG4&ijk<x9JNv<>6GpJ!ia5>(wDV<NeU+Io9v8!lxcJt)pg~fZ?ukuhpv6JkF&3wU`l=#{KeD8^`d@HRv4d)yAQ6|CagyiF)pQ#Qv7M%~@aKz1Ce?ckX(Bz`Ez@bxMgFRDq*Hw7E|TZQi3bA)M4<b$ssY^!I-6K-*d`e@}zql28`+h0t-F!P~q)osT%{$wx_c{$_K!V+$T)=b@3ea1<0MU^ssz$$B*Z0V!S&th><F!(6M$Hze39=BnlhK31u<k5fdTL2zD_Yz-WxS;wvLk4iTf&lgR;mE=h6UgNQAvwm{wtd@4AQ32TQIh>^QDpbAV$=@~=%lgz8QqzK{moTCP26>V_>pcHl7W8s5Px=67e8Q^CoAw<=a<XCChqT$>qw&;zAJGfTscXN#(&p#_kGtG#Al-j|y$R|aD&F*Sxfdwoi8*(=D`!Yfs<*)Q+AFTKE&H|Czl^Ehbf}C8!ON3rbuW5bOTY2~+^z&jcWV{XMfD!<=b9vT?Ry^{1<Pu-W<+#mDi_iGFvv~v-o<FZ`>M!BjrTgcS;BxhURSBTYu}7l3J%}?{M#WRIQZk$ZL7jpXJ4uqve4hjaKx<X|J^kGyX|7nV#~j<`sPp2vOn9k2`l(S?#XR4rnzQM*Kojy$JxH?qC{PrkGh&&&nDdQ!`D~4Ba!2$BgSF$;QJ&=X4jteh1d8u%`B<*2A;3cTSMkthvRv2Il`(_vDz#AQ7RFm_pm0yAi7CcTGb8RIELwgPub(O*e*xkT5Vjrjx4D4IGQ8gfoOQ*TeUbP7N>>=pchrAnI%yl-Kx_UTA@qs=I`4tQY~-hc5m{>3+?mMRCMk3x<?-=JKWE)_unl$_J30UY5|G7<q`7Bv{Rf~EC%7M{`xlZb#EVWlii*8at)4qRAaU|N<-EiB`46F#4R73WE0y14Wc4Gg~=WLn5Q268%<VkWLT**B9o*D^-^?&E*y`U0}46vI8XK`;q4%3f~@i`sntnk$jnz(OZ8^iyRdJ@!c^#8OX%{~XFSVqTu<#{Gp?aR%<9K^=WGX#Hj86XbArlA-IJGmjj8>CF%t*vG#R*KRaglN{um%ViT(EW`|YBqf~~6FWw&6Zn&6qm9hWm1l&Ja}v{3?@^Z9g1?rWXu*ZKI^L;382bA4s>v?{s5TlhdZ$1$VEe&rLcMB?gA4#|q8da3a0^F~x6vl)bG1airCDWf%JOU1iB8^mx}aML>hPv5=D%BwzYRIta3o?<Qp%t%8Nyq#{tCO|OP%8TL-=k5Kd0*NWk9;agLH*3rbVed~Y2;Jy>c*sui?)j~exH1v(Fu5Ls-@M*}uPIvx<utf4kLW4ye76-34WsV!_ukBxsvC1?3^5<XkGov_pyv@LEHWK%*FEmWph0DHu>ZW_+VLL!92gYRa?k;Q`nuM>I!pG0%^dm6HNSAs8Hmg8qABDXn`u4DWT8XH-ONf&v;#HB;LshIZA88jgbI7pLp9EkQ8Z}24i0*z-^#byN|ra~GO=9~0<^=@Sq=MJVXVHNOpEurw%$ZAeSiXI`r3^k;K`4<H?+xG9LCWnvxan|%Yw_=deZ;gcx+8yhKupwhdTXXf&e@D>~Bt15;^4K{&JrqBm()zU7r3C+F{(pZ;RS!7(UiOWie}Yds3stz6$NKoTaMPz&jFJ^-G1H2E@Sq|IsxF#1yN?4VD5XEyw3GZ1radI@>(fP$dPb%b#<6WyS6Yz+NSvAE5)81<e<Y9@5!(nRmP8$;P+MtmB)%Vfl#w?vNH!>XWW6ljX<1?bfqOZPyVW3p(pr>Hg^^)e1##V)%Di%pWP;N&r<|X&YHmtmA&enU0$~No&2s<W(nGdBf9_?_`@#6hM#KL{N<yvzO=jf#tdbhB?wq(;qsC`9`^xNiOlSW7?oHUG?io+_;#dZT+2tlR{tgdwFvWjlWTMvCi~<S1R>L{`u`j)B*Qu%n#PL$3_u1TJ3q1mL2b;+vIfL?vA?n=Fy$BzaqcIL&a?;>cFx4kB&w>Uf{So)o;<Blk1m@KrFI#y5J@~T)#BRHC=r<Y)0P}N2KNdq|t<K%qD~Kf*V4^^N6s&SDS4Tx?Q&5pC5-ua#x3^&X5}#n<YCUgw-jz<MqK0Xp)aEIq0-B(>i``HBuGBHt-IK&S(gUH9T0cv1Jv(VoHs(+G^Xc+n@o6#g(anPw@J^_kpK$LvO(EJt(cmaR0OL4eMx)p%c}@1bkH9x;4n<-Q#)fCKIr=#19Qh9tFgrMYq<Pf@Jt#@|t+FxA4ECz*zsG$!{}`dNR>ADhzd_XVz$oo`J1+0|56iGv%6x?GFty8DvyVC85L*R?Wxqoq}1&XaK{Eu?bYScAOfI8{>*5va9c@rl4o7+x9;%@A4V_e|PKCtMpE_<Li;ElV+A5YwnnTU$9`=d}8~}dqr&^2@dnEK|D8O$c5S6p-M0$qc&JzUAHcfope=T8Moe}%epc5nkfGf`J=V3O#h6dHHFxC0=?h0-pEAzmVV;e*RV6+0Zcm^|E|0vAi&?aY~?#|2cxLHH=p^*Z{l@)#jKl%1O7f<{Xi(h3Ut4KNQ`|Jt|SV2bm;gT%wEaXj<W7K{+xx)q;q9P*<gfSSLQ0H2K+U9%+{L$-ou&6;Uem&FXHfE)4SFUE*sJJ?s(S~PT3J9b5yFzYV!#fJf^QL7OV^uVt^8r(^Z>4E0!|=kV#LF@B1UVTe3}Qlq%4=v?^0|sccW_`*}e$d5G+wOxv$PqoMb|_^E?-zV?+kHyF8V*+;aq?dJRYcl?Mg^L#U48rr5$YGz?=^c=;Tve&Xp@JMLU^85O;aGs}v-KObvsD(XZ1<;$1Lv|s3ypZ1`M5p=bkAUOou}+=Ect217o_N^ZCLekldGA;BYW-Nehas<>w0fS~wI4t<*Jk>-j(%p)Cwa7aee53Z<6f-{x`WT>&_7i>z)VcSL$mvwvALJSGa30ZZ;5yuUzY#nl?XZOD5JmIBk8m6CSKcH-IwvJ`4N+^|0Wl&&4*;#_m2bIWFGLwX|4I}{p3``YeCzg;N8RMcZ22ZvT^uY!>F0KY&fKvLwjsN*9g0bU!?XOSbtv&wo-Dzw4nNnwo!}T(-w6cPY?lh_XR5(PuZ)sE$iEtxcuIZs`~@q7!s)(i8AKZ&jC!%_3T@j-Gu$vT8-zUyp0Vp?XNiMe0tyGK<SPf_K$=ro$;ny^E5Bkrs!(iMwA!XM17OIamO2j?d9zz+6=y{R}3*Za>CcGzaPx!?NlYG#>_6q_)a4g?5aTJ_<SezXglQaky;-tS#a6cmJphochBU4=42)=xmEu(ZBB!#-dI4N@86%+)d|*R>MexSgA?mPozxg1s`0aA1NBELH%@wPm``P(aWtNzCWdVzw25gb)$aBBa~Z_LdQl)Ed)xutXRB~zWP3c!jP~_5TxL@wXUxsiru_SL$+i5v{>Ia5l0WK?=Bg)Y@S-xP3`G{UE0y8*F^FJ0>Ew#(vh(KP^f=7Yn(=o}eK;gWyi}|<b$zv7+v1O;ZZ@&|XnWfqTn=E|1pd;kVJ-Ydo2J{@aXGy9M#gI~7$Nhfo|^*JbbixS^Sosn+7k>>#ji$wQMkvB4{Rt|adehff!Cw&jH=@fP2i>qAj!L&9iqu&0n2oQ2NJPxW#zGKHI)$*;EhwNG3F&ru5f}PV^9(+Z-Y5DU&HlHDHa3ib{0mrqAHZGBF^^m-;@Unm=b34@tl8qSf4s=!~L=?Yr!4HCue-q<yX)822<+S54&RM56RiyRUTfUJ&x7n%ke@scTJtR!EK>2`q;R{Jcjle4CY6^QpT0T%x_?y(rKtY8y`qs$w!{2Ur}rD=(AQ6AI<B{8xfa1Vz}$XmrAx8Y9I7a-Cjt084W6?mSN4s$XGUKQ+94^IB^lUEB~#58kd>fhW_l`$d1x9Jq5<8RK%>ekB!mu=a1_y5j|@D9%$%$PMzE>!;ZBnRu2j;#cO9i+rZxnIWk(kuc1v#iUTO3#Fupe_<p6vqJJT}^74CMUF)k2xn=GOF1bs1y{~wY4z=P5_JL>QJS!(w<LFJrTS}4s;^PC&+e1<8`9qY_gc`_gLgcK8!^jQ-w=Ho=1$DYEjPBGGu=i$Vl45VNTKocV%dGQnWrEBR<p|DkVb{6%bkihi=lbj)Md#YQ927<I&tjwr(~LrDbVG7UNutXO6e&`nD5*aC^M78~OgVe+wbsvYPE1wCGrBiA9OD&^FvHWz3=m~77*(+teBPnP<S~J<;dIR}oNMd$UQeF~ET6f0NU7&E4y5T4#kMyPSxyF}8a`i^-$kw$HdwSdtQmGLQ-}VQyThOl`BTP$x7>P5p&Mgh5hBr93#NY`HRfURWoloxc&5$tK%qY~Xf{Ok(bx-Y8?QeAy8RbtvW_>bsHc|v%Vq#M8<9fD``COHVds+3Bej}b`a;72US}k*9Sjsl?xJ~|EJ#Yer#Lv7!h4i>Fvs9N*mp%v;HPUO4x2a757-s+OW^x$-vx+se43zi)87Y2hNw|R>DD3mIjQoqY|8+>{ni{$YdQUPuutoE${_obCskp*BI(ya()mAPwlvmsz4aOSYe($FRNIui88_w(uASn2!JhYJa@Ouu_zv*aLpiJ)v)hOkx!)Cb#MEZ&2zs3BsY)Mh)Z!!*gexW<O*aqqw9@-TE<W+D9t@NWWqVlkrwv#&`TVdMzn*oX!E|)biw)ekF6)%f?SL)Qc}*-lsxUPvzG#0k+9bo%1fH4=lQnlf!p8&MJOOtE;H&QfqJjP~z5oNc8IQn32K|GC&3C(WV;r)u;dH9jYt$Tq5IkMSvs}`>Ir(n4d-iH8i<S6>jn{(&pEkejeV%|;dxtiO!r3s_ZZ>+Zw)WG#@t<!9KX=A*Kk1pU#aL#_Ju6EM3aFKJL>@+7U8$PfZxG<3@wPu+nIQ$UZJ%M)zwh3YhrT5TeGChll|ED-Zz{I}Dr9D!>4r9i=x2k?KuzUkaTNX8{V70qEo&SK%?_rpzk6ceW9WIpdI{8iO>y~rK5oh3@wV&p&2^&{XNPGe>b1vDv0LG1%y_%-kE1R9lt8sf#K|K*XRZ|=YDI1Gn^Wg{ULPHY>;oLJ54WkUmV<_Myf?6Nm@!W#M3{SlPQH(9MvgaKy}xx^#HO5nRVbA}4sSg};z6Q}(!J&i8RKNjhN?uLI&6on;;XKX)U?@tkAgklu1TPSX!N%|J<ZhR_t<z}+K1B`x8a%`AMi^XqxKLYHM-Yn)h3D~*u{Q4bZs@KkpzF*WQM6|$7@Mp=htvF`QxwGx39^?{63M>sRcYMW583&dM_|UX{c-QfYwif-!qS271pic9(>a<cBl@6rr9kYS8eys#Y%>BRSc`5Z`#F%q1eo3_fE4rhE6y#dEeg-J57#l78spgM#LXp=IZ;GCf5=tyW|#n+=jj^lkfiz_7DChMOVqkDmG8;uF+|kwf%17%lUAPA-ffdgrgun&*`TY3ibOG%Q2W@Xn(Q*20(de_HA4b-J_afmx)Lnxjy3^DLs&fiM)CHLbx$2J#oDw=<XzZ>s3~u?EUPp&@g&RlhVoJUcP*qXP;NGRerC?zx>q}WwNp;7;a{?Keak^BnOSO*FjM5IdCOE6;ip5fl28Y{b0=P8p7?>^~1hD?arOYBx}xxy*yd7z94i#{BdqDK3p_};qmyOjb4hXz`pl>hYgMUtvl=}(5%zrIonKLllpo)(N7L{9f!N#`WC+*wmuL>aZN@ebd1E4CCCIl5qP&S<wHt6()_c-;dxe{&WvqkpNwVrR!^YD&j(I83mONL`6DzjJa+ZpbZ2ao@nt&9VR*G$a8C{kw>7nMyG@|OUsxC2dMF-Vze8Fgb@&sL#tk?%x7CU7CG9`2tV^0NeT|MxRc7V5aXn3kG&4`=A;3i5YV*U477YB}ya+e1)#iY{%)<QQWLTb&t>smt>fYp3n`uMF#o?uEO|=gOAndn)H_5(gWS`icQp53Tz+BCBPf1_b@94W9b|4Xb0D-GiJ@OH_Xpx=Ej_&Wx5q5^lM|)ZsVCH4TfMO^PuJgA3ymR>&H(cPj^NKaKG{Y727jS>II-~Ys9gKrp_uH4u*XnY)iDmr)){Kp{VjaA4z+7&+5-D^3TP=p{(`{~B(tYN3Ne-x<mrJi^HA9y4sKa&Ug@2+jhz!8fd}na!)sN<#0|)&{r=DbEuh$8v>OsiYS92RYnSA~+%GcRUwD(USN7X9PSfNrh93rxX3HND>xa8`=XS>ZBI->pEt!lB!b2|9zV%@kd4&Jkokjk;ol6;1GjW%qYDq?Umz{ecbdUtTVo75iWvs1mK;AX(7P~g9!u};wIc@3W(AgE|y60sB)!1;1hPQvM;)-0pq)nCza=F5*{zb1mm$Lo|=xxHZL?a8>$@M-qYPL96NTKK~a*Z_zGtkrF0>tJ+v*1%TISFa+`uAw`IYrgL`e3_PK@Xd*%zV<j3gE~cYW`lP_1<Jq7$N?12E5RLqwrj?7-&Y}evi3xwwI$%;&uP=t+qft*O?~bA-Q^W%wD-=Zanh+eS9yG;aZ^xj@#h@p9*r@Ng*F1WHR|xT$(B5h_Vl~#+{RUA_-@bndZSu;?den^kE8Lsb=Ae$Yr7pJoB?+vXYOwdg6&dA@GAqo31VLP@J<%9^Ymcejx<;-(%Q?dE|ij%fOiH$v>sLG{6<|bQrS-A&yyUCo2PBmLgJ!FT)fA)NYW+juAWoSLIncsEsh4zQ;__;P$O-B#*C6mPsn=U@Xgn*KS9>IRHp_tL}g>wK33;fZ7g%bqD}IM`AFOLImS0`7ufcWKc>Ec|K@G;Y<=xH)BJr<Y@abwDIMPc{3qJ{W585~4I=-|FgH{taaC!qE%~vTwb9w^Zer0(u8)2^-^e;IhkXr}A2FzG6bVP*^6Vtcquzx8NeMPP{d>%(=|L!$m&)rfq4M($sf2%d>DIYiMsWYq*?3T@kYtB$R-{1qL|sDGg1I|!wsjmO*kDH7cf}`r8b~Y|mx+dww)`ASaliAb4a<u)Y4*7#zCg68?6=pxsUZ-4AlKbS+`G7w@QQv8t;cco*zBkI{W2=8xTb>!r0T2rUkiy8e?1P>jgLQYrqY)c1pZZ-bn-~rg3Kw8Tf=-SO>=L}T4LHgpLDr17wXyCYek3xQ@!q<NYF4R4qqe^IPkjH7F!c+d|G1OW6Lh*xtelr&`3WXy04T5zIvTur81E=QD2YfPfMj<TE36E&}*m$_9Cv81vqoF{Dlykw0S-~1x~g48cL503T<~<!SmX{#nqSk&ZlPbBl1aXHu6h#9PfV@0^2Us;+R$FiDtE4B`^w!J`5o+YOU~h@0szjIA#WYpZYdE@MG>E)teoRS=hdcY^pBLHwSPd1|c`S;tNrKReQ?2v=M!)$>;%O*pM}PSq`n2lbLuu^o{v?eLc3HQl(v%`~YetEAf6BuIExFuin;gF}C2^@@gD2<QXyrn!x73Klpem6p7ZL;U8XTrA%<M8Z}ZrEGTQwzA1)OdTeaQ;S*0Pa@h{s1+zSjJ`#>M^01D*m+y<5CEKzyIBbrLqzLZ?roWdTp_ep%;8?EA_1k?rG*dt};%<AXzZZ0;77hOLjkWDs{o|3dp^Y+pZKLXLn#Oq;Owis^c&yDzP=k#T*5-4rVRax#CEcp!dK|@C{ZWr@cDN_B{G2t<>nHTQ!k7kQpPWmha<S%TS0WjXD8vYi%{xPWtK)i(9=h?Uu(%Cz2)nlyu(>Nc`NAt6sD`0An`AA8+21?(+_bW#KhL;xbx@R1a0S)6@bvG*WwN*rqLR42>vHjs>X|Jd1musq%B>k2u@I)_7MN6=+;rVE<^f)|YnXJb|NQ@7N9ZN6xjM7s^xpS6fnjckrv_PyYUxXhtLIzG*JCD)mTzqw13USPHwxhk;Wov-FM9K$5nQbedDZdyCHEiS;Tvq8ZsA)#0GyKf<JZ@EP@RDxZZ9YMS>FkJmb-SeNv){6NE6)>`q^SD(YY|#+U4C#o?`q3PUL74C(9XXIjZ#?)yO1M7j;0^mYkDum#v_Ked4;=mbJh>BhEd!jo$t5MRfFd%N3gl>f&o~*19og#OPal*W&x<_o5JV`dPN|+pAW#X4Cvol^V<WAKW@Ed#qQl(eCmzo#)!}572D<*zlM5v|GU9k6BY5<liQ7G`;jb!vV2Q+LseeQ&&2YtMZ%t!wh%dRhy#q$7SOnf2W%Qt=JV^%S&23>F*ikthg@uEI`oA7^otyeYH1V+=d6+Wc!NC&AIV>qrsZ^T^G?rt~4#7w@>e){Jz@e^+V+Tms<^6yQCU}JHKBXTDA3kf1e(r67=A5K7-Z#>-kbtbSS&PmrAxcy6JtZh|ehyh4-Gujb6`VGGeT2vM8*m(dL`X{jbfQ{utd`)B6%DoqV3Dl>ZJ#^{4L6^-|53#nzaBe#@Ko8MVERej97x{am!0D+gkc@4Is}tM$8n{nxA>P%z)o!Mkn4Fagm0G>WI^x>5<)-Fib-h80%-o7Jit(@FDOc-`@EFmgZp*@_2)dw}DUIePA$gfhi1jQu($oA>G6yk5FrcX`^UM73B=-roBD`*riWoes@+3jE*H^_y~s9gGV%3D_t1i`zWz>$pv%?r?do6FG6K0<q?+(9CDe$!!>uA$S*}W<ItL2Q<~co7OlkMB1{*9Ghj*enXt;K}>`cytL)hzE-!s!dPiYh`g{0nI?AC3WXkE^kNyVePt)JH|AG;R&14&$b8_;M>K=0=N+w4_xg&LFWwn^xLu~3ZCknWbz!RCVRqLUDWb6EPOqY9bg6piQraA=4Z{8Z2OF2(;c-uid$OAL&zG3hJ*Hr#D}OLF<d{tdRbhiV?unNMy;0?N5p#R%&*Wovs-4Npd-Cx;H<};VPSS_=D_seM)HWMm=^hT9O}GC2D-yrHPjqZi3jLD#Zrr2a=62ABgyW6)Th1#(&8B>te!W)hzvNzBU+wY)6j1oH+I*+H3}g{zK990ZClKMZDl~#W-ot98>NpGIrnsGcYb$a>)#SmP95NP;DOCu5XuT$n3^}!ae+XB#76#m3=R0n&0J4qOSRKku8N=sps&Cf<Ys%4V`Rb!jUR0&o>UC~3w*omXbD{m7Q0+G{g3RHO<7e9%u^G?ODs&k((VTGZe7YZXx0myCMV?u}K7aALwtm+TUb8b6j|U&M`rIxS>^AEU6RR$_$h6d(n{-a!^y=1qX>?G9Moo4ib3ckYu|HzS^Q97IHPY?uf0LJ-QoJcs-xYNj>h@p`0x8|?rs;YFt_iz0>+D>DC4;+?54ubCww*4zopjv+lL3$iinlZZIXfcHdP#jKEdPBHXuFq=z2D{(&!nfV`<IiOg%zu2PO3c8&Tse{vI7Q7GNxMo=r@>&qkC_?6?!QT7iDaA_C;`fvO<&227Yc^`z!TWfJps4t<GAinoU>2#S07H^k&y{y1yk53a7V+R!FxBVE60Y?!lBODU$Zl9x)=L0Is-spZRO0n;vR1dujbfU#26`$Qij&Z=O(6Ka0uv3q_f*aw4LJE}p&X>gAk>Tc_Pb2eZ!ObM85?fM0%Nf|bRFgE`f<d4)H-W~Dyp?p5Hbv_iYaLyK-R5#U_{eH7)fuf&e1@uvehV1JSPJ1h-^tftBP$2q%WH<-BvoNlMoX7a}ZBY4AujrqkqIp%1!gLtr3YzKH%>IRhqK7G3;aJ#ReAeef0Cu`O=3hLj_2|MQ&y>qt~96sZp`ki!1^*wSWe&+GB-i~mgLD4)QR`1hi-s5(=yPOYju+WrE>LV@5UTW3zr&uw_;ls6^LkV_4J5a;Usot}=@9sI&wo>IU`IfHX_To%QKHW;RU7&uBwkiiL;0eq_d+0uui6zbD-_8gVf!>?BkGSUYdN@t$S!=l=)B|%9j*r(0L^Wabc`WzD<^6iL_M@V|5BiH9eb(n!dp@TfWaU<<-LbBdS!;6@%8Ey2+DP5fdB)w{a<V*9(hrfx8Z{%Ks_A}S>hA7dc~#P6?$MFE^WLy>rR(c(`V7OXI-f+t`)Xcoq<89S@LN{@Bc$noxSP(n-mg8K=zeD{m)iT^A-(ixO>9X6BQ`sPpYXLp;YT;e3oy`o({YoE%`-|9@c=S<(tRQJOAfC=QTun66!X|&CHh%!Qa&B-zc0C2%VdA@|9SF;hTfL=Yi2Z%g&&hr3y;6;qgh^))@YA^c$E1~8j%s42CpmszUjkyQ+(&L%~YH&Znk+-DrI9I{{FWmKnCUR5~AHz_X9KrDXvc@od#>w=K1z#5p^)`M{q=;pV}-Y0K73rbDQk_p;<kO&Sn2w1aS3u85kR-@aBUt-jbtTLoj6kb#HHWs4wcZ*Kui_`90t_udp|W-c+gx+ZK8ca``^4JP6t@BzM!`{1-T`mZQPta0%bH0v{cl1KnCh58GLE35<De)JAtCYGF!NkHg6kG2&0?)P^l5*Handv-R3*9v@lyKIOFs^Ana|p~arR*8tVkX8bkg+cd4Ly5c2JcSG+OV0gm2-I^18&-MO5gB!+BxU%LnuZ_@pGr0+SyW6~BqX)Y+`y&E6{fQvL<7PE7#elDiT$F=gl^GyANtvPjpa%nm>bA&+JNZZqyRmrHUO@rAK6euKP&2qMeI~6vmn>1vLAR%hWlXgWYe<93cQdlbZm|vj$WqOVX~7E2nf2vw%W2y()g&*(jyd91l#l%t<g_|>n^Gc3-OXo~j*6_miHyrMb$By@vm@t6{R&m}2K#B}v)kr9`(j#Ud>-vbl}0OTXwmybmQHQIywpW~v+eD_mAHa5NsSRt^{}(vAN^oJhMoSbvRLuoJFTyq3)%Whr^lzxr_7Q*A#?YAC-Y6EQR{2&Fw`jLdB}s)vwJ~t@zkX+&uR@?9v4cXatjkV_u^3>BB-{<4sP(QcSx4LXU;nloaLGPg5TG5*V$T;u)buRfK%iwJF>6wd@qufQtNjju>bTwFyA#gCu4s5$+b^Qz}|v=x5v`Btj$h0@(Axd+sH57S`PPX&V;JQ-4%r_wKH2)MpS9~Az6EUxXBQqq4Rjn8^y9q>!%(qROb0bbmME#yuX`c*d1)TJ*CVzJUUmJKQ<g3iZb7vdA)Y%mR4nJ*DG8D+dbAzbad?Z#_i)DC^X!9wYF-668PkTZbLPjm&v1fjI2T%mQ7yqNm2o;bO?*>(fQ!X&+DQVQ>}rOs~46Vxs^K)mOeAspz$HtvC6YoeNEmm=}R0h$;0$5=RM@r={cn^A>xz<7fJBPOuI&;EkS%Ae7)kKh=Xb3jI3_E(M0)Py|TDOr}mJ=4-KCu=Va#F<wIQV%hh&JySB-q6-`z;GO0{|5mij3y|Zn?@kRJu&O>0Zn|)*zST%paU>9iV&$RMd6DGasyjqKKRi7YO8N~bqcmXbKl{N*OSWNW_sM&f?PnDf?Idgv!(mETm!r5#!*dAMd*8%Ltq}9BowtaSRUBa4Y@a(iZkFWHFe~Hnf(aH=j+Z|DMirxD@zHbhv%II-g84~t)nN`PkPuKtjzf3U>JB~LG6Yv409bM2j*jX+p>$;#@62UGF7k#c~5u7%ZljY>^`FpT_P<T$IPcl=aw-4gbdS0LO!4?1Lq0iBWEjnd66#nY)8f@;(!>@i7)wVN^j!bOyDrqQ}ivX_yN;vO1c<#`<3p+i|_Q^5Z^uyp$Q~tl({I(y2LnT7@ZB%J@QM0j)`Y3B-hFZT?O9q;?jOrb|^V>gM?Uh`6aQgahHXNQNc6VpxYY;N-b@>E5-q6_oSc2C-Kl*LM>Pqo2p0d$mJ9_RlXx|R^j+yjta@*#VZ@mSyHQoI658<H`Y``9{+f52#dA$k4Rin4<<mTz7X*U>3n!DRW@sV?lZV;dve~x1kX`cnMxny?SVtwkmch*+*F+N!1tan6Xz4N?INuA<F++L-VH<r)z*lP@lLBkMNJ>|7Z`($lCQ+`zF$enhhzslT$<eU$4b~!njl{tyS^bej^=8Dko?^oMdc{kJ)+1<gg{^IA_e8cvR+5q~Z!^?h9H`HZFUm|t(+|~Ta;ne*z-QnETH+BB~Ys;&SF0t+D`3E#Bm0J{QgVC+|J8iMKv0q%sX=l>inqFc_;_bl<jkIyU6#O;@K|lkTwZEvJm1NuACV!J!T|4Z3?XQ;e6?;bB)rWq20b3!dt~yb=EVXxWq$sTXleDxDn(AZ*A9odcx$Z>{b+6U2)^+IK#93;!NdA>l!Ej(s2V>s#E!sXEXRD81VWmx!+UgU9Iu`|fv5LO<CAF^NAUAzG&W?F*pPRk;i&Zf{wW#j()v8RE-Dxap_&}Hp`c1mi-S~mHrZN8~HA$UpwY#o0csNPBPMJc%QnANvNpMi|t93~9-E(WXA}h)DaOLC&!l(Q`x^+9r=>1e%Ijs(*)05UGbBzu^+gg2Rn*h8goqlhw&j%}p;2QH;vLAM_Ba{ZDi(MAV(lKMbORp+Rs%3l&SUi7w(4v3pEws;GoRx$@iyKiG74=hcz5SpK2j@}SP>tOsn_jN%Vzchzk4~o?-?s7{y}g!yohOL_P-NenyjV{NqHcr7ZoY<69$bd&m$a|sNxwxs%m1f(q=)KaK=u#41lay1qn4D&(5c$adwwx00WG9&0*Dg%Vq6^r<1O$!)8GR|mb<ox*YH{4eVHKBojS-bs+S?xK1W-Su5DSTKL27;o#GEGZsM0p_ie2*kM;#^v+V>zBN!yjlyS`)@oSP*r@?qv@Iq(j({AsbO;{I*e5CJ`o&^}!OAQ!Bj>hb@p_|oQLvE|99ckd`<~4b>xXCnfnPSES+l6)h;NPRUNJj#DT|V1&>kub527U`>dAgm>u#r2yb(4CRk;WG@JEEBfDvDEN2;qzKq5C(`q=0J?UyJP2ODc<C7@*-Yw9OXU`2zTaKnJ@~Ha}%7pAW|)#UI{zhbN|7^|Seq`;S$h8wPWq-ulO9il+I|n9)S}l2>9n+nrBpu&x@0__>%9{?1(R`a5pPg|f!^3DP1iMD0A+k(x!C=6W`{h~^d^j=GB}W8Z}{?;S#VP$mz6w~osmt874eecvnA>4d)+d5%Xze4>G;`9hc!k}P!4z0tOg0~H=ZC_8fKN3T3@PN<kRVrV4{tXIdyY!Z*Prts6D>Mpk?7&9?}-<8!~NTG=pOdWCRNu6tD-P@;1gNa6?B0NP8Wb^(~F<j)M`K@_)9b4v2KilJREYK{<T~wy&f<4lmj80;PEv@tF)2rve2qOOC>sAqBYH)pRtJ`t~F;1iyPFW23q3P9+ei)p<QIHCclaC7rb^2h|7515&m($J&FXdE~J8|#@`}}K@-npRsNr8SjrLTAOL@qq1-a_6T;rVrY)i+OmtRE`{LUjR_kMAjxl<N)dthDsO!`?5*l=uE}`-)%~X@$E9b_c?&&k}p|Y5dImyLi27Q_-&ItL#sJM{f<;WC`AO7q&=J%RdQAE98%SJIzBqgconkPx^5Um8Y>b_=fl!CKqW9zrQ~L)8)KdbQr~xIzlT&>t&qw#lwv{3Y6RIc+T9PTQIh6pxuGq-VKA9F;9#)@b}@9Rio73SMRh+NdJlTtT8=I?hWCgVBL>it<56U@9z+1-vh7xO1*ZrgKoIj+toUs6?<1~aObR~3hWdu;eG*Z))PpNOj)V{y+>R65GhozW5esBB8ro>BT+5Tv8wayQt1cfm~b1I1ujCr@${|bY`I4s>fhMPP1)R>-x$?c2g?C>N)Jj6+U}RR_6$a|t(l!R<Cdz5XOQk69JJ$HrbDur62Ya~A;I&dxr1|_6Ow+%8FpuX-SqdvQP~gDPZMt&VA!2GSHFcXboSH1xAn%=yNO1B)|@L6{k!>%#@`s`1^Hbr!24}<9&1lT%r=zPUiWYDO37iVb<&|G9)^qV)r_x;#!SXNyd1x!TP<!yJI!AMhYy2J&G=6`(3^anTxRBYAM~>DKGU%8`Cb(tGwY<!vekU3U!^gEo!Zs+2-v(ocV8NyNiNM&T=JJ2kE>56=*v!jVBRad`vw^t9GZmq#9?YnN7U@z%h2~%qc6NWvD>P=cz;LuYc>xr@px%O_s3dYiquNmFKTF{uuN?s*hmkE3@{S+E%DripC6{TxQEsEVUIVese`GjZ564aZ9FTiHw{*Si0@2e2-nKDzgi_@@$f^f0k(FI_;W<3gEP`<8LO%<$BiA_!0$d0<G;}U3#dDLyM*L2Rlu)+l&9jiD-<`JHNipi(5kiQjYR80^egwk(_hv{4nP#*7{6vm$t!PmZGdx|TGUolUF~D@^9PD)-QyNB;f&rVq(FcXF}&8F_iN{R?QVY?S`D_~h(Lp{1~QZ*q2J`?-TgzO&xmUmpI(}t3iS_m+oZfg4x+V!>BsP3jlPJsDQCZC#l|(YHtFqFv&*gcZWw)#UO&zI^2%_`G<~RF=ipB}FZGJxVW~@+kDqw}?-siU1WNF@54DG1*xl_IdD2JRdGoQjrYYCmV2Mg#$J5z+hTF$&{@~wu1Knb$`Oea3(?3c%bT2QF$Wj7cd#&)M+Y<FW?Ljw@yqBtoYH`=)JH0u`H?cDMK;2a?DWO=gp3T8L(|(yxtkY?AJ8yO`qj_Tlxb5fhF|m<U(Ypg2wDj#hoxOI6T|e9V3UnBv7th<zrtJhCS(J0%PI*)w-vQ?vTxLY?(y^*`PSJ=)_>2u(AHE%GJw%=<M^suPhxPY2s@242WL@xEA3jqQuL`TlyuDH{Aw!MU4D)d7*D@*stLb1y)@!ybqw=ILQ|9&>j-!Kf`iP92<F@n6hP&-NPl}pSkI&1|lpJ*l(5(KTNgqp79C$<T@pi39@`WVoL3k+W3a%7JXVuw2Tw&aom3igHlP{h;5-3#o`^?s_uRe(LD>!)ajoayWoD?%k{lmBC&;^wK975`tE7hH=m8}*axB7;g-6nmQr)#ZC{;pb>sh8cI%E(vnSp9}szg`2zcW9~ARbJYyiQ!*<B%s%A*qxuolQzc9UtF!&`5MuiH!M(C#l!ilSy0TQfb*67p=pW7ahU4sG~SlJK)<%0thkyTH|!?#94Sl6h8R~#{6lLpEt=_%AVG6Jodm*bFnj@BcWtJ%p)#0U;`F6f|8Iw@uhDB>(MnJt?xxSF(%k#5w8ML0_^ZpJALp8m3mPu;rZrsMfwR*t2GibgS5tedakfuhw$k(WFR%BtxuWP!x~2_y4Wf4|IHu~~H`{rsZuglm|I96FycuE7J^S?jKy&nN?FR$>|2E)UMeeOa<YkY0Zle*3e*x2F!ov1CN+oKJk8S_h9j5gRn-9-ntF^mG)K(oSiA0PKZAW76K*F%}AHQdA-5B;(f<@7HZD`wje)Tyl)_&CsZ#QROZ1dp3iWOsfn*vh!I^$x|bSpkLsWsM^A_7+LhIW*_hUNEq=yUGH2F+8nz*^PG@YH+lB-p;WO?DH`>-9K%ad%*<v`wGiPX@~?RN?)A-&XwdEl=m!a>_P(h_G#L!g~7%p}~4n^86B6mo|psE$6a9lenZZeta?(;Upe?4(aLh&Ek10-A_89e;sf|-(R*Z!cdPZ?<l_;Z!YPtcD76;w_nI!^a3Q74jJeVyj}4548>@)n9uy9#Mz7;FTOA3)#@Fm3v$<QQ~Y}TYq4tW5e(ZVqE4L+YVwbp2))6l@+KM5)a-gK-&EF7qpcG=oT3P(<Nv>j<sH)x$^E34`<snO&2rdQCOENv+{mv{B^X<WzNhi?;q7X{Tj{+`C}yD#fc96p5p+k!ky9+DW-{<v1E?n`jj?ZEm3(yd*ll-)gv9;<Y&L5#`WSrpSih>z<Fe7?4$_w7-BNE#?)~;B7OQx5Xx+8@BEFLY`OrE&yVp;}pZ#GtyRDS13nS>R3mg{6z&9<KYF#_GiurmNlru{3RTanWPw#w2XT19;-`ATsBxOhF!=2v7kT2>sYIEDp?%GP3)9=k(Z5PrSoc}Jr#t$vh2}6^qzPqZstgd4xKvHRRh<fS3{(M$y<%%{QRfz=nM{=~Su(6#e2~;r#BXyyKt8xB%p@22|)MnEXfj2fM^?MBhI4Y?&jUyJnsngBQzdVXN$(L8ZmO4>?Bt7p&V>Ok=ulaP=ZS(iVYAe$AOQ~?LGkmWyO?kiF^P(Z5>$!GadBZ77Cbq%;$yl)5Ye+BK-HBdrJ=d4q82YL%Lbk@2ZIN||$@?o37K9|nQrC;72;B|>a$kuQ_HM%E^Q1qscpwz5{@>r#3t_Mtmq5XsD)v;k=zzUhr!T7%v?U)co1HuZsvvLjn^n7ET`6MTGb?}C;u&oR{$%2Y?aaxq(a>MilfGug*L@{)j8z5dxKFX#?6M{&!P8z8y%oPvs`W<<{|dkeZ(LJEf8(|ufWxXH31GSOYVGWAxsHekPjIMNeAY1617DpR8sQIib^Iyv=j=QIfk#JEa|JvNtYPiM)cMdlT_ao3g^HD>Ro~)<H$4XfPD#LOV}5L7^E&o`sjixj#Hn=X_3!$_+VZy!!8+g=Q_tu7?`SW2V`SeOmh%=@#m1y8PtLX9Xgk4M{L24~cl|#s1&-Nqu=kJAh0pEB*G%onXRa-%i8J$VjX(QzC&8t~v*hJ@!bWMm;edY%v)%-Q*<4b3UPo^Wa0l#f5qV{fLP}_OAMmtG(W?wO;&xEIG34|)eGCd#woMUxDqzzo?m+d&joJ1Y>%BB?so9UJ6^mmzm!J8H{<<Y5+ZQiwj5ig;n%Ud_F}V(L!eVNb$1EDZls58=P0D7s+^>8lHq`BZ9zYxN(l6k_+KoARcWn!S!n-9pYfqbvD;edI6RFm-8VuuW9X#NS=OvxK-~$0;_g&e1&0T=4EHUOU%~)vA(fh=%2>Zu_KjQgI{N7hq5>?eESed=udKYy?KJLq)=Uj#N%f|m0__Eux63WGO+%He)_5PQsh&G8#vg?2~lV@dYdyg%N^ZlEjRBNC*JEte5`{IHQ%<q4>0(3wpkPVgVRq6_seh?aHJh0EH5;xkKc~jcs$zCryQ>H;2-qu{PI@^xhqp8;8aCjXiBl|n;zJTK6O}ps%>7GiIuFv{-w<d|kpvrkEylR~PxFXs=K6TRH&7A<73Z7oKZ?ZHNomz-EnZUAMl8B2~tedYW8?%59>r?NtJXzJF!M0Zj)|+nUH8+FSG*>5^%gy%m5Zm5#?wH6A#*g<`)^F^8;=U)M=Gq65gRa;sQjq-MRr!hXs(QUEr)0g3x{$kw+%dUp&xslP#Udx){z+c->6oJnafy>^<L37rXH(GqRhAPcfT}6oh~JR;*+p7@Vj5I;QI^oeg(rNNE}*9tpO4bMtY5-STL8YV7k@gq>#SNigS|K|J|!g<v(I26VQh-*pH~}T|56~rWnw!YbVkQ>c-P(A{Plbx>)EZRx|+QI^$vsb$50C%wo~-G45szON=rW6TUYNrw?-aN12?mC)2#TrVL|l<@Xq3$r)T8Zqt@-t(J}wI#bk7pHy^ebybfAQfcxDIf14*y`aOlZu-oyIsrn{%AyU{)g$GVY|8wm8$bNGJJd-I@+zst6@Tc3<qETf}%x}-72}F*$;mr~B)0aneytr@r&EWq2^UwvSR%BcBM|LgZ5_qR=8KK)>obk>*{`yr|-v#jI^+yg2flQ9nW0du*ZU*w$c0C#K-D~HY_)UTc2eZ2Ast<|1mknQ#Px6Ld?Y&*3X$HF1aP>)*qMS`(|23(^pVXqR$8K-^fcqIy;<e?-i|bsu@u%yTy?jIh_uhn!o_3qVPib9TBUxsDSVyohWSRfYhV2`&p?sxFFC2Xy!)uwBQ)^kgjm|o31Aye&@x7)wU7B7zFs8EzO!Q-axS?^FLY-51Y<Y*mj<~iSJ?o#=zV`z<Ui6DH&3}4SYe&pn=U(q+H4x*!l~9^Cf*9CW+~%YA2&O*vW8k)z&6<mG-RUDn`!$6d>mgwjeD!{lytX7SoMsJc%!P4ZF*=IG^qymJXfT~3G#T$od_NvKj+&x9=)LdLD|WA`TlT!2K<V4M<r6(;G9&LW@2cY>)d=ds2yvX5i!)n1O#M>K{PC|N&kp|ad;`oaf^-6mN@ug!O12oz@+OFAu}7SDrZT*hEDv6m31YFY!*i0O<ib0`WPP~fTJ5y-2B+lq>URMyJ{{V<V3|~)v)LKQ*UGEIsoHr|h#&fLjfU%Sbxl5XESx8As*2KTT|c-|wT8veq@1Q%-mdE1&PdFm#RQ`3NK22KPMgc)l*>G2<mo_LQ2MYLJ?Lm<E~Bc#*C0~0zxq^jAe6Q`%5NtG#Y-qp<PmQldxKvE>&%<qe9L>mfUyi?Iogh2zWS<;FM(>nOk>jN4AaUh>4`0MWG|lia30ez-;5D{pj9UiG<s#<@hD%KEwPEK<*>zEfc1W?3bT>_tW>iRqxR81Q$ZpE{9(T{Yk_`c{wAd!(@U<};d<Xau#xv02?O&d4C{*qHt$1ty*7DltswYI5eXo^zX((uG0YI(kj6DET=^1=a;?x-Iywdp)ZA{M`=1GI%loh;F6K}x+Z-{6Rsj^b;pHcePZRpmm|QY?+ddtq2mO2RWZmv;x>)O}q1PwN4YukY1`rDy0qhT0l`>u{<>R}pSd9-ro!Y&&bf`AJt420CI7{?E4&V5!y|72Gg#4(;JP#HsdbF>jX9=MwSGmUwhvq%`{Z9V4q-`u}m&LM(>$C3c6YIk;8QjGWT-&C%JnnNW@P3=&k~3@5c-9v#lxL(W!!Gtqb~7wlUBI#K^v^hvW2aG_M1wLsocGSc+*Z7;_SuiNNa+7RY;AH6qKu@Zr{!FTBTY421GWY=SEU6S>wAoES)=x}b$3aPZQl*}6C9r2ql(kW0iCk;Z%A>Y-v4_sN{Yfab{#LTfg9u>Kp~tv@ILn)Su^VK-sLj{{}OIsUAt0LAKQKI&<}{pTR(<F@FfXj+p3BIxR9ml@Bo0OAWa7)t(VAbdcOT7D;F`wb4rzv#qcv;i+JUR`OL$7NUuj8Pd|LcY6<aU_|fbAr6(?q%5Pqk#9fw48^j0K2WbBnFSuwW3fLVt{A-?!&x2G(AB#orilF>~`>Zzw9Mz`Bxip$?v)Zt)!GokP$S>*WqR;gddOfR6(E!6w{ne=c(MNfqgLGb3+q%6Z=~Fx@oBpK@t!}_$8dQPtE<TWEv7n7XVARWLl)mNrV2M%grq%kLQ@-s>tC?COY+Wscw;r`-e@nsk<Q?%d=Y*Jkr(i`}fwM<_`b)KUcGXj-yE+bcA9~Yl4)z!3dZ>@j(CvEZ@mppxi#mAHXo%nSQLMh@nb6(g%eRVvaz9}&(gsSzp|mGkhv8Z@ZdGpejmX0Snm)66#fptXThP~cqM9zF-}|rl$E<&R*xc!h0P4I=&vc<}GyCT52{8>}mAd_&--f&G{cO}}&XL6S*DD)&THbh*ZVMx*cK+C)p5;BXTXma0CSL?7iTbTJs-LzUA7r`n6<M-t>6XHm=k#zy4}$WD&ct^`ZG@wKKn6|hdlDBbeDU;YvfI^o>uJ`LRPCQmXB#!Wz45M=7x3DF240Y`6R*Bi=A#mvQ|_dOQ(QXE#|b^{kDTrNX_Zitz;bcm>3aLi9cwOGiS>;(&|{sdq<)*E*!1vBhe9ks0#}>OMC<FFL>UgHM<G9>mw(yME71Jj3vlhMoNjxMh#tU7!wh0|=K8Nx>&}ga3X0;|ixJL=s-p)UNvq#@z}+i;_WzvgPU-^%Y5eNf9Q;fHy(8Pfnj#G&hAZpA^0`{ZF`jqj*)4Fq6ShFrMZyCMPhS0{hSl!ZFgaymP&M?}n16QNkMYY_J^52P1<!>ibXTpQlk~X?Z>#G^-w^cE&94qB)BUp-^xcA}R6{=o28~JmC4QNEQ~Yc)g*^B7w?_6DkNh_R98YS$X_ch!0k_4iX=^oenIo5qxk|Fl2Ywt6ZugJ-fM1D5jX&4hoXj|wmKAw&*qKK6Rc*uzyj9~MwG+L<zI*HO<Y*$J;trQKN_%*|%<8@mP-MHE>=u^Z$Di-b1(W2{w_5`*klY^H@_MBlACY|*4mOQ(T9sRr)N%Z5L@WY%MzFHh%Blxn+OWM^zDmHfNhI@x#mu|1pVQ&TPujg%?_SQmXV9i5B)-<ZUbs3vgGX(jM7tv8rJlP4DaA+i=KR_|1A?Be=GS@<S#r#Yk8OmqN3|KQkCnx$n!Mj{%AfytgTItanYH`~)N7eTy$!zWYfZ)V5wCTit}?dEaAD^BRqk+8s8Az0;>u+xtjU}6s5Ka|onxh&;_cD4&5-)|Q2FaHTAabbjAMC3s?h~iNnX_N-HMstmOuh1$4rE7Ja_!i`%aI5>9?-^7*Wm2A8VU`IKO^v{pr*`qYb!=lJ`!U)n-D@+D6s5oZo8QgD-hLIaN3F{fv#O%OgZ!_w)Nc`rNGvYOL#z`9t|7O3TpwYU*as?kG)&1_mr7-`QD1I?lUzyiH#x<+viqTwlGEosuFLUfEN2SYyrv4ckg(w%v~CcSEl^EP}C;kTogGR4sjQ6{NvUZ1Bq=S)Oq(tfgd+MtdVCpAg}+k1W3QX7vpWJ|tTmgdSHHTd(314)fi;2ikjdto8s%5GO7(U(UYCXWl@qeuhk|bIR*?*!l_uJ{zeyxVHOF<KFZHVX~jBzF#sa8%%I+S0FlixL;BxnwM9~&=dUIWZcK9?(<FRPjnn*eFj?hN!?(+#qH>JaLay;>Z@5TTvuDSM|;6JKcf1jDucZmCX<J>HdAJ6#ew?E<ubVRN5=_#;JpOiZ-NOBn#;S`0h;Np&#kwI%Lf-=deD^?Q4auH$(AGMf5wv$qyg+7YQTUca6i>XI2`eQxN+_Nl#ZD39(ICF@_mUDaoMadk|zPnUSvS!Nowpo0+Tj$=Tk<klvI~bENu)g5@2;scfpLq%SxW1XcjleYhGy7^U4IrQof#ODi8qPqBD3ZyNU~*N498mFS3`FY?MgL)X!}Lv69gb-YdY0cbWI)jm1}BDS(gZUJn-K*vQDmyxx<f@vUg58?>8k5uros9tt$%-F-W04wdC0sK8LSSV9Y;`@!IHpkj9Z$A;`MBbw%Jgatct=|{s>u)5#y<Ls?CD!xGT(WXu=hS5Ys)agvbH>vr<+^1ICs5;_}`p%bv@^fl!-y2)nm1)^<9tF4uGe~>2oam2wgUu%Y&}|)Dz3u?(KKqGcmnyV;UG@{QcJ_bO#qV;O3dQkzs`$~KMDO(mIHj>)@kDfP11_LspQ^ay_SzD=A*RKMBD+s^r+g%J*<`6>ba)s;mp{)3uN;PYca@_1zDW1xk6R7MDdf1~S|^!r%!~^7$p4}}MRxKQG#gDYu)$oK%acbF&ZoATOxGxMT2bBBxg{MM?*K2eWmg4FL?6OPf!7k){8NJnl!aFbhLW+oxcm$Dd1`;?uh+A@*PkxC52m{M5@$JeNa9<{*FL&%GFBaveI|)S9LJkaB9`3lJn#Xe^@WY?|4+)XTC(1v|5J+~9<%nl7rv!*dHOC-P(SG@i~6`%t^KVsuQ$W|?V7FOZH*3z$lgC>AgF8LQ3s0GumabMijr;~&F>WYRHw}A37>Cjtn|VO>NlP6i{}TRVTK0vH~G#dgS*P-b$aF1Ms>d!CEcDER}_*6-SP7In)yDqYo_a~$kPNj50m|9MPiExXDZh@UvNojQL_PeIl0ZaP}g-=k$UE7-7~NFqq<w|r^^8!c{~iXI-g}oH!bNqo;KoTew}g<h5Xnx(vNjR?_V)FH2!gPZcWRvVHo}`22C_(5-Af>Qba|TwILNHsc0#x-~K)Cw{PRNIIr_K7-V>lVUi_&`|Q@L575{?EH$vSokZC9Dp1v4Dh_j|BKH2q7Q710E&%BlN&|}}P_|CbpIYaz<2ag$=Hj~do<Xv6Ew7V~deA%2QEl`(U#}?rzzWo;+(c_)Y<Dy7LSu<=9k{c8QC@7Zwh*J=@0E7Dw>r<XsOkLm>CacE{5=ZZAEUgOPgWyNuX4F^T0*KJrXm!h^WgbAx%(5iBB5!ke@IDz^}*kZl^_Yay%ceDov9x8{M&P8QwF0-WoYKL?Ang1JT|}wV|?7;%*z?CNX)wfToeLEu^ErKop?<VK~s3txjsRh_g5sIv-?V_iK}aq83sX#jvAFC`5>Z&R|GffqA_i+<H~ET|3Kse_JSH5F2>&mf~l^UFC0QUOsn`*)XtG#<Wr(IX<QpKZat`|SZ_2XT9EVedGNRCJx~9@I1Ns}?co7!tWl-71?KSk=Y_$5g|}Zd?Q79AJhB;j)itKd3{(+!9hj|Mdfl+p5tp6S)px<uMR`ttBgnVawxAv}UwsY+>l@rhGavNu$Q+)ZLrPDdCTn<Y_|&~*N@ojBdRJ6vZKMTsn*MDq@7myZ<h(1jFxk3kW57Q0EleG(JtO3=_S>I&;#F=TuQWTHO$A6BwqaQf6MFxN5yHNQe=c$8a^>GN(4qIIU(ffTro`7by^kts!F@|UE~-2kN;`Vh2GCi3?(6=p&;~bfP+RP8zs1>sC#}x)UL8vv!~LbF@$5>L;s)~SQ)0HD2<rYT|K#BG$A=b7ofHmBJqmyLhpWYp*T~bIedI47Uxa2y*%aHxNjrxD);`_yE3`#z{Dt?o+|X{-=ICFcU{cC04^^AUFT+m{?$bBE`r-r_706FSho9$2+%X+fN}xS4TC5r)S*qngeZKelIE1X(Ut`i0cd7elvzPtjsK-OZVW`<Hxjf9uU^ZpPy+LIP?v5{fR}rj^zo;D9&9m6E?B4UEJ(+^rKPqHqr*GXzx`#jVo;q|^B;MhoB3`Psr^C-<PW{&V$#^sAGQDoVC!xu26``54brY+I;ptrw#G$d#Xz!E3;sdx!Gk;@}N!VNxOGGfKaZrwOi+x>6k6-b!N5eyJbFH=gQME-JuCo;a>RR>La&aR@j)-nl8<xDuJL_;T9j;|g8Fh!)J)0ugP><G>z9}A+>{ct4e*cK9`T5Tth&bF_WQFY>{l-B-|A=F46l)KgMk0qhlTLj8R9Y7=Ovuby-x<1tS8ogl<DKyH;D}ZhK${KoGRMbI(+wPWw*$HNmtM6TVwQkqHQ^Xos<U<XO$(}<becVVFm}ET_dMxSc6A|utde2MV>muxhp$HN@efEJQGwf)<9s@&0jFEcAN*@GHf4c3g2|=Z5nHE)_@V4^aI1b7+)BQAsh|%=@w>-9=k2Kcau(iO@;pgwSh@S4Gv=`x&@6OW&4OBQOPSa3QiaxyOK=hm4a6F#Q*DEu2rh5v>Y*3i)EFJ?E%lWCjNmLdW+Wst7Y70ieUbV&Ng+n{q^9114wEJWJ*-hY%XN9uCFjjWt8+f~=_&K+s;~lfYwe1*9J-ecOpMt-qT}8**+!LC);)SHK>I7V95K>K|5sTFx8FoqCWmeRn#zd(z@J_cP-hLWETS@GzB)mVR<*r%5`wS=nGk>o_zOEtkf%655tcMi1ysE6&Cc5??7ZjPAq82z-{*W@?IO)#{9DP^*7x6oY*i~`W7)X$KPSvF<@kqtW!Q+2-zU5|iP#=*pA_!&(-k=p`0d=1v2olv4wi>Q=W&_P4GIkB!~46`L;5q=Rv(qYz^-j86t$9Z!LXpV$Ns6Ixqp4A;I<fyyKHx{UyFm*iL1(^a@q>t3mp(mKtA*`_g)1qS#vj=9DJ)vDX8J?79X|lc!t7gp$%Ut%!Hpsa9wP?Xpb*@7fg_(#R2s>YB9b}XPENGq|Oc;Za(NRwVb5hx{8zdx%0EVOnThxIza!xI%eHETiqqMB&m|MRWYinC+W3_VBpzp2^4rcwPx=Svz;(MlSZ;XTyy=tt+2E8p`oCRI-y;7xwp(p4=;Ll)BUJza{JSn7tg1)jCi}`2T;)}di!?ci||uwoL2u0Kqdbxhx9J>H^<&L&sLC8MbOUEJyoj*Yzn~desd)Hi-6J5v91WqPWkC*6Z71Wl$*D>hgqfajy!Y-onn3=xAu=i%hq3KYP%blX0-|YmT&POv;ZO=0UaHgGS_J*(wUfR&ZZqY@VARrFK8nRDSkE0(M)_s4UxZTL#9g{BuYt^4dpELBdF$5MjLZO_Y=YX<+9n0`*gCD$&S0eqZg!j*xSUs;drt;^+uJsuJqXlg|ho<p;U5$11aWZpzDj*9~`?w*Es82BpU#=1@VsHk!N+kbUW?Wt6B@f5Er>;r1%3CY~M~qX0Fx?{hD@2cx+f>dCf9nzrB36z>U^2K<GeTCL~qfR_nBX>0Dkh!fzJcKFjS)AdO23sC<_J8i>z;&9~odQWuCNsgA3rN%x83HyBo&rUukO&`ZAF7`+G)!9r=rx+YikfHwkf347C{Gv#|ouhLncFG_oK_}XJ1<VXOtpTr`#beB+jxESQ1naBMaZ~)Kxz2{-9U$$g+wY(Gi9ZxMl-NHWZ-D!$k_TFmR!=m}q;dvL@&&gXc#+#RxkEqI&az<5JrK;_Ui|jVTcKfGIj~#L?)z{azkzQ-r@VtPx+Vl@CFhYCa=r&!JT&>gfDuUpF%CqSmwG+C#b{3>^e=2M1aF#u2rmrXC*L^u)$mL+>(adbM(<^uU!`@gY%PY#}hL1xAdD69ZHAHGzqU`T;?9IQ?6P;JH$BKjpb2kkw|7zsjbd^z0m9E9F^}e|P{yhm`1e&;PZvE9?h_AhFTRk*z{S3-R+IiH*_sVyx-<|ICZY5vsx9p#5U_S-U?L+1n{Amk5g&HT-FK1?W!^Yh)&|i38Z#VM2a+{6}h9CWjgs7_Dbct(xqi(&WwqSUSu^hNE^A&u%)_zTo@^Ue)+rtyC22S@geSGuQ{6Yl+vr$8envs@RK(B3+>i?<`0yf#zSME5*oV0iI1=se6x7}#pY_9t&8a-^+kI92O&m6gV^ssSl_jsRuANs6ks*2+0tn<eVg>*J*ChE=iny(9>qoTCzP18<_H=GKz_oUr<e%=_wOB)aJ*;yMNlI<#PYF@XF&HUGL&+Nw|A{wgWCK@Nkuje1awbw%5Gap-l9@hj(b4UbYMKWLB2pi#pZ@2ZhYUKv)4`qPsPAS_R-Ucx6<xbhe<b}~aNL(0H(|T<*!zPMw33#jvh|d1U6YIn3W5;y~EvvQXJ&#hU<5qX5wLO0IBBv2{HQCnFn_ljFpF^kpD-VfLG>M@8*~?9Gi({a<>iwa!oOFdEcpWORVu&Z});PQd&AC3gRmH<=G;z)H0_9~-{>-Y4(TuNRqA284@HXeFzY&IuEfWA^HG@CTmcn$d?I+<jif8*bQVQD!-t=s(V?wcwe<oWw*bvu~Ehv$g<>BTcSATSvrW1m@RMpwUoi%{xbj%Y+kvi{1;B^w-(|5zl2Qf6{aO<oG>Wq?Br{%ZDchdu$F8UZwRqIC<U1MfcXj)F6-oW!P0jp^HNa6L~8orqHFnf)a<+y8|f3^J^7mjzBWx5-VY6g6cj{Lv^j2`+@EbH2R=RK$}T~6+3)i`~$mY3fz)l0w|Q@b5XHQc?j%}DfZQ?Rf!Lyg|}W2oALWT<}~Fmn`VEZpoVXnOL6FLGJtQE}%=8=#2$czA!!c+391y2wo=&?A`?jb=06+lxh4@BVRsgI`C#UUY(+(F{=)J3A9}@SNFJ=o2#xhm$k)b@G|}a+<H>(|>E-+3abMff62sMPp_xv}S*N2!?X@?%n4dr5V?gyl+=VDormvf1a%vqS|+&W2;*%YJbk6_2-+c-U6@p54#EEr&<(mZW9Arg}2)J`G5-fxkk~6{^?t4yWQ8XwPL+1mjlH$2;NlnEat$PUa)^I-qnv`o2+dfHZFgyUWEBC*~w{dL&=u$<vO5&a_U@(r~I3{zDK!{oFIExTcmBMxpJsURP_<9b0ss$Jh}al)@jtzUs#j&fh9bA-zNCB=x65IS+BxwIA|RrMVS~Gh)R4vf!X*R--E~rpDo-K=^m^S_ixU8&)4y?4O{%OqIW4xRT3EK3n0^n;QjJQ);@)Jh8=F^`QI9xJNrw<ccXc0J!_^IC;Vc|My*x7JdPV}Z6qFn^J^PEt*9aO$0DcaM<6z0gF`Od=3K)}=e69PM+fKKgpRjAs9W2O@Kxc>={MD|$`F<btT(t(1Ajk=)hMe&)#@hdak???*5g;_iyov;L+MCDLK4t<+#XP)O!O7p(?k47JvW~9wg-?zY)4B4g!=Pk(t8W?6}gmWQC<{0rSQk*HpJVK-<M|z{~p6aaSptm7UoqRpyUgum;?Tj)M<YFoPvyH9-hD3OydN1Ov9^k^mz8!ZDOF`)oO$^o^Wr_sX8S7(u!UO!>=pR?jw^UmYpX7Ioj5nug%7uzv}onp}00f_x=d)uwPBP`O;;vgxK^3&%fp7O`6*5z_%)Hf_%KX#gi7OK8^}`PQM=%q<*R$s+c!#`?uJ7UXT})DB+d_&H(#Hg6ID3xGq9MRXkpL;$qbowf5<Vh|zf)H~pT&9{u2abXphs{kD(w!+eUx$`CE4R}cq>NLPrYOm82nH+s`H?kdE+%FA-1CH%D@iID*BRHGPbtaqF>A9w>PKO(cb)))P}4=IzE84^_{*FVseNyOGbXt7H02PVSx+PfJ6dj4exI*vE4e%vR|9{teb!I_vY9BzIIk)OnmMnFFLSoJujl~t>{%*V!T;OWp(IMFL-IG3pR+UhKVM-A<)wTH%u&p&LaHJ&<N`*?2$pEHL)R`8(0C*D&a?UeghmW6I!t+*tkQmNkZw5?8(O?0^!lcRcJ)8l+OQBD3FZU*UNBwy`u@6)esmF8@ij~~PAs5YtsF!!NHXC8=9wSF2AT*hnntYu1|#vB8U8ZNBSzCdZsh2Lm{W$_7Di6F6++ZaVUNIv^sm~t~8TjXK4j9ckq&RsJK>X4JgJ#Cvg_ZW}9H-D}(Z^c=*$ZP$gRI9b@=Cuo6Jl;HcMGtH&YwvYBg7BiqlxDxG(wAQh_CuSQOet%(4gSi+{fxL;UH&^6CS0$m-nXRvUR5MrlXVTB=Fj{Pn<9&sGVCo<$-RDPB%l{tWHu3%m!^dPaFXfk1ll$Yp^M$-P6?xf<i6Ba?}Jls9*WxI+I&^+!uEB!#rZSirl>u8HC`g#Rhyz)Js5>PZTMkc5)etcw*&CU%mF=kK4d~FHZXNGz_phh-=9F0$TsC0qv3qD!|J;)YDx|CO$XF$OU@f|a5`6{fj_yNdetgMzTZAl1p<20fwf|g`|&-jY>$|a>I1h@BFdgQ3ak2GINjH43QzSx>g%|huT}brkIkdi-5dGu-t7#*H436*VyYxu-mKn-zl!>vtQ=2E+f@D8nK*l1zV@sOBX~fq0)9jBgs(Yr@&g&&@Sl}?69V4{Y<;J#)BOu|`GwFWy~;CKj5lZty@WdNxE%3(ND}P-sI!Pk7cT4D*W7QWuk%rzN3H2YK|>5HzEDfih)Hm!<rqjM<2vlmS4X~MHDn1H_|I9<(1BUDQeM|X_iuK*wnDh88T|DPE@w;OvhPYNGT%+RRP%tT>G&|ISsSX`SVGZF7q#jEU46IIqT{3+;h;MCs^vLB_}>0hE0<X(Smk$ymf=fq!CuE|JTbWL#-8q%xG?+lgw^EhPF*6-oA@T!uC!=W0o<!FbE=}!N~GU>XFRLd*%Wzivs<y=Tg=D~oo)X0{mJ3jDK=rP(me0==I?TvD#2?CjPqy#eXiAVwgkF7OC4`-1Ad?}k7uM=1IMFjfcvttrkG-IRhO^gU27|CVY`{1s;ZXge)q{9$9-**&Ld|^cxks4U!2#@4u2|rYsB7dq)_`mLr(Y7?ZBh3S3Wo7YqGJG#wUH<ULM){U4Q25I2o=}mXzEWa_4^;@0_9=vOLON#fZ`4xYEMRAIH}}1}r@Y&);%{B8-^lp|AR4rFqhhz1W<TtyZ<9Hg}44uAR>BI5L)%^DTY1cKy?8M+}suz8;b5LEkDfqOa1ero2BJhGchYcdpSk5YCL29T^?HDVb~f_V8Gb2FBAb?@o)oJ_C4M)LFuL^`KFWUzJevWR5zvbDud2^Y3u>mZh!cjq5cs#ucbHoM&aHdh3D@<ctr~$@xyd&9dK4H{)>i<&^sswPi+aJ$F;?q&M5I+I(xm`=La%(tf4lpF7GamliIrnbYnk!}cqZ*u5Om@vSpOE&TNiYjKljdzq|{hxVunm?VbndSDBxG>+>-S<@(+kXpUbG^|gU_R%7%nOXYl`R>oj$1t-Y;)tyGh|IR0!zDufX;|eMlkoDiN}&&(!OH{pqr;#*2#9OttT5JY4{?==&E9|j=PqC6qVKrLq0YeWhi{McZrv)e)wHrkPq%nlaf5fx#f?jI?JMvGaoU$eJ6Q9_y%1Te+B$@uaKp5KIA17m%qLeQS~786OQh?vKdTXy>O=fO%^jk0vx&T&=X0~<d1O|7vzdSt*LI7D?To(uuE=-;|FQgOe|$c~6Rh0mM{d{N=H~i1^L^R$EoyY6)|+ACjkAI8I1Jlj_12C3{?v!pt{-gdBEP=dTdMDkhRbfe?j|`OCSmyi-@`M}_Z!&d47d2|Hb2go4gd&lm9*v{aJJt_Z?W<n0BN$x{P}p9_-w^7%ECx+ZTCT@AC76A|2Vim=!{3vpBOZtdk+#kF0AY^*-4kv=yh4Vif5G=7tOf6EDqH$Na5?6s32E|7#Xp{EKm>WhWkqdMsojzZQ_J!Ku$@#)))sXl;|GxUp47VdKDZH2T=aY-)t#n9shmHXRIw+nr=g`dK@k)0YCc<Pt6QCna$sE7wrQvCvav&7@15E(ztcl8~c<ED!moBP^9q-X>vGO?c$f!=8cq;y*pjc@>%6)-k$#cOj@d}sPm{FEN!*yPAU*_&CrgCw#D_PI`NxxySeV2KK0IS4n|I!ccLLY2h9^dHP{aEF;qur)IfFK*@~lb*JeX@5)A0zSDiO|ldiT8SK_w%TTeygvfC0eGwB^p+wbjx0-YgxY>R(@x5ggh?Uc7z)34)e0c(ixQ+`2y-$z@1Us+bQ&79MRsF<|M&My7{L#h##mNu4av_EtlkFa)m=P#9i?clB*-J7qSwpolG3oWx%-Bv3O16Sx8to@{#SSwdoN5B=Jr@gWZ&e8T>JVL$35&IbW(45odWs^fQqBGMS4DB(*_p6*Rjdg$XY`)wN1{hL9Cv<hE55w@6pI-%uotD5ysFWl^soNsv^yNf<=bNd~Z|xvJ-?<l(h-EKd8f|6Y=qW?yhRknwaCQ>(dFPmt{pmweB6Yjp{;lzX*8I*j#Yt0*-xpHqCF?r9^HOCS&e!JnHeegajj2>IOgYtMpXX-FWxo=4v}`k~Ez#y%s)5DWw>rq-u-E|p&RW>N+Z%i$`GakJb`ic{7T$%eYh_iaX{Jt1>bh=F{hB&}Y%aOgzT57Qq~TF%)yVlY`XV>!*XobRtS-!ITy-TNfw|+a0c0%<0{j80VXLES;SP&2{qB3WCGuq~P71hqTgpZIRB6n8TcA(J-6cFN%=fqF%2T5=9RO95`XYTxt~GUhdc5r#t5>fIwBskK59X}lMAJc}u8F5)9G|kUe*JE;J+Q(^JzL=zItgHFR`;+maJbCC=C;>Opz6U3p}9e9CO^#T?H}&h@7K=yX9O|pl^2)?W!zYf_FWOHx!2@GNaRlwlD&@IDH|Cebm!&~B-u|N_VLC?rvj+V0?OTNTKRs>fuy$0&rsE3R^lf{i*j8(<N7+GTXg%k-kKH8EVDtiD{X3%qdnft_;LR}5lGoq2zNiKnLo)p6suJ0ez%Lw@w}9Or%MyO7V0s@l*<18p-zHgwvQ`t#eM`{k-%rna*s!v^=HQ&(kqVg&=1R&Kb&pK=2?r%noNI_-emIZ1{Ndp<us4VMYF~c>>9o^GrQe>2W^_`cfL(i+!=I7Z8q8J^*$To*~4St{;BzEhOL6UlQs5wm0qW#lu3URK<gC;KpH9Tu0=PgS*o5VzsxFMVLIOE$nFBvm%P&Y+)@!&l&d^?m(x9difwG~Q&Puj&L?x!`qNW(0YVDa{K^|Ttv-}J38(5kipNpUH-e)ZzH1|K+L8T<7G5<y?1<s{>-_Z+*&EnKX-qfw`oUt}tzqvz&Rj-)uN#*qYn)cbI1>4uI!r)G>EIN{xx^*^_HUL|KlFh4zPy;;LH??p`PjS?{p)z&ne7WU7-2Fh9$d9IZ8fcdM}}8X9=&Q%JBqvuw<>;QeAq3foq;grIeLWKC)Us&Fe}dw<Lm&0L|v?=NrQsInY^c2X=pF3kN8oO>~hM1*Yn@4!Cfx+FZ?}&c?@c**s1cu)n)cN4exoE0yIS~o;K5pQ5Dtmq+9k-=+Z_zyZ4EnqS}}gySGeA-VPb8utQz%K89a-He=9JPMkmK#&oNdguB;MXRmSlSEG15@rhZ@@k7NqapU1t#yf`jxTD1dfH^L&4*Upy=J)l(9a09o3vRVT&0-hbc59yB^*QfOw`}yddHdeI^WCFu2hFSJ@V$D(j}E@hsp<Wi@@f`UFY!nmOGk3W1K;gjRo7S#-8c*2o_tA>q6gSyh~DDhvR@mU@7>!5L{vlSu}kh1vDN0&U9lKsiU*c%=a^4;($u8Mk>e;dy}aKqXo+`_wXmJ1%K9V`)fQp4HnGrbljDl!N|MViUR)Wm*U5K!FO>O;G^f|oL!~IYGT!f4wi%BfoBB>Wy{DsE>zdK#{k3%C&a^UHK?3&DSv_WFdr4=u^#xnC$Bwh@^v(|hbUPx^ue~QwNlbo5uNcGh6eFd>!)?mHBB+pI9_CH%t!K*5G@31DMNXM?zLw>^nBDdZa$S=(BZc{(lIoQ;^t5ydxyvBe(syCxt|j>_cL8tOoh9>o`J7%fyqg8_4olCkd}Vcdg<@qqPgHM5oojG|K7EJ#PCNUsnM1)dZ5uCVU!%+kES8b7cm5$4UGKCPt+^pqjj7Fg!|2kV`hL3X1hBJ*AJTYfP4a!sFx-bW8$hQcoOLq-6dq=G_cL$>yp5@@v;>B)Q+si=<{lz(f3=}It|`>um;L|yZ?m<*wgu*_Kg@~i`X>%L9xI~8B-iM7KT^>70Xsti<R6!<PGK#4s3E^}JiZm=<q(C{A)!Bm13fRf-$r3Nll$4+M85=HJk{-d=p599ceLm9#=edx`+KyM;cqryYR+cmmA+iR+OzoNch+*qPd{RFKXi_0yRv#UEcfSFc`GKfe)yseXA&uhRzviXzL|s;KvGw~PV8Xe#b<-rSa;a5L4DF~=LFRD&GlvHB0Chkbsv5Yy4$Z4mTT)^L0w9%VQ#_eu9#>xZEmPp%qzi9KT~Dh>_tc%EKdu#80+K{119Z&*{)$+512tsQ--I0P=nBH(r!LjN9GQ#-I@OFY_a3ex;j1QSi@vXVWN%tPobA|s@{0b71`=5Q1!7!L96LCXIFBL_q^?UwSfQNg_4_q=T+zuU4~QgcX>qTUZ?Ui`6T<6g*P?+2JXmEe}Xr*oqf0Am(}&7l^OzWbfki*cEr+Un%9lh`kj+WpQ^wUcDh=eisOxJ*s~^EhoCp~VcpG&n(e0ENq)v?dK2aOz}fW^?PbIBX{aZ)PvtwrRO_NV#sD7e7fzQyt-eVZH%B&tjz1bytDeiy`}i{w+hCi07Fdt{1^6(VdUY_}t)g8cn96`heJ1HWsR1Ie7M~G|-E#VpYR^W>m2%+7ZU<i$fwkgATiXjDEYa;<`_A`lKYwt~W#vxwB%+z!h64>8D2DJ7D&RossF{8$QxJ^R^*nFbmOvamwAC%K_wd-RTo3c>WmRcGFQG0_%q<RI87A$@hYwncO=;lcjSJP@Ut@af^xfRG`cPJtDie7}YR!Oa&L8b7DVv<ZJ=uEaRXwz~Bc_QOvUZx28>ltA4e9j!TaNYXY8xt?PXdKKIV|Ve&F%P~#umIi=0N}~_cT6H$Im=ozu2(9&-;dJnQyQ?$SUl8eW#8H&kgFR)9z2m<uLoB58<}y84V;Uh_V+v<Q{ggPr{i?M<w6Xc5S3x%iwRTrd5Ka{sywmyFTrwov$XcKgRvyhtp;He1I=x_qNQlOD19EEU{$m5>=D;Xwc!ad;8Dk&8B6R>gGCWplM0wi(MmW+Am)<n`^;?-+w4}_4W6qF}nD6<~IL!uXy<ZB~bVA<)z&MLhdM=t;<9cZ$QI07M)~wQU?{?=r9Wa3&zQP=Q71+iwy=+>qacBRIdugtT=gBryM<eL{i1U%w+8A2*qKWr*3Vx?aXNT(Qf<X=cd%GCRHgO0_8P!Sh;zJ$T~0Xio89j3u3oXipK;aZZPLx&*@iWl<GF!Ig1I6;&Imb^t;7qKRC$NTi!}7^LSl|yZI#=P}Lgp@de$yb&ceeI-fbCGSl^Qd3@Ukw=WZpWdEvdsTYX;stQ4PGTtsO%YLw%-*0z;gBRo@k5xpUO;^(NkkY4yn{FRkb~4f^tkM0sI@WG;z1mz1E@YFgboqBHcV~TVQrMzdYiwB@d2%@a7ldZs{E+00*Ei`V<SF4WTD=5x+QUbW5#47}0p;}Zd>C|Aa!uWeTa_Hn`YU9+a6Y}(<t*V<aT-cWRv=j4K$UOb7{fvBxZ8KTPek!hf<BSI$*RqJ^XdLff5lPp;G4+zJL%(JbLm%s?65<nmXPFN>&2XJ?e~5h?oL7EG9PK^sQ6}9URArdb3Y&*OW};};$rlT#T&8ta<yBgpkx#!q%4YY3M@?7hgI6L54MyXhXB#4y&LLid@QQ`2vL`D`=@WWo3(%!%2qoshpWo>8Z}Oj;R-S>GqfYHkXU2QuM~PT{sF6>{^~HGwqMPW;-IsK4FqnJCxV#V{Rn!2G^*{$^RZxN-$RaoQhg8ZGVNVQFSLK@mQOP;$BQ<{SJxZYG=?qBnf+$dr)&uWG5$*<;xk6JThutGj{>(>{biQTR!is_d*9tsSTf^!?zBS{QZ~Y!NiV6Y^7XM-qi5c#FO2v;q+8qDT9t$SYP4#`#o`71`)9dD^hljldV+cm#s2Vf3Lg@GpteJD;sL090ZM-|vFab<)-=0AtISJ)%aq2i-owtsaO{tz-*J@ag94s!!#^9ntZni%Ln_WuM)oDxl$iC6HS<c{GQs@%8I+dreXk<_aa_r^e`!A(bfi(cu)kVY>&v0@bVzl?wLio!pD+hZR`^t5uE)ko^0)r8zHjR+@|DG?cE4%^WG>*h@A=H;#jWG*rwO7zv%`1$YxV)M%=S-vx8~ct*|57{I@GCGn}hAXu`|5e5Gf?G=uh8vY6S1uyVr>8K<ARk^+QthH2JbrdL^EMOPpM8Ier3Hhf#1U&5NwJ$a8zg^bd91T~h+TXT$x3nCkki-575a=w-0AM-R3S`})8=_-((L!M8S?trqvZXR4TXD3QNwr@Yv;<I1l5W#fce$YGApdoxFDll`jiGvTf=uJi|r@E5$#ZJXWgvgy4$krEzpZSP%Xa;FWp*VlKrMZ04g3^%kcvbuj8))RSunhv&GsWPh>-8WG!2B*qh-cHXRCc|@sTIqiX_Q$UhH>b9b!Rd2Z2L2Ya>?T-$E!v$#^u<Qwz8^d(IAkWPH#jLaBfC?H8&TT;5wuEd=dV?zo(|(7t_9%<h4#N9Px`w$u09;ZRi5q6?vAnJOKZ*k3O{7XtOfOT)WRcgN#%4E*;V%`4<7hn*P~OUV$nZuuN$-G<8_)fYviebVpZzG-RYfRj5Jj@yNB1Kt<gF>BCFWv<GLKSR=K_4i|+4FH@r##`<K>7?+n%Hl@TX3;<je)(@LfJv3t$<D7D34`Ed4%RK(A5zFN&J7Yha&({Ab5l1Ciy`I<WQbDPmoK_6Orw5~@wMu{!ZJYkz^UKzgUqssRvyr-h^qsH2}<|G$gi=#RYb(h~$CHuNrv=arRee6@o`JR(a7X;mK_6CQ5j2maO=H(c;+E|#f2<R?pIMislYn5R+X?)|XwiVdD6kBzYTvnu2t-x~j%}(=mkcYuT(Dcdpeh*uoBh|aBbl)+og15VJ1+|OI?Y*n4yxg_mpe3!C)dS7LD&Q&!FdtXl2Q6zYcy3;)r`g=~hYz#AgX{K>dr+0E3wz3NG5(spyU^|n<#-(fDUY<Aq<ypzXT6P|kv85WJ^G`p(*Cukb*i|Q>=SH%vzeRHD=Ng4=?dG8c=`co$^z54c9WLj*Bm8b7uzL5<4R!4;f+38n>K9@zw5CNUWv*z;Bxtp=cQ9Ibobk<x)}IP2Ua@i&5jR`WahcY1IV*%7)SnBN{8FDACvse(*jRzSDc8l^5^DM>q1r?{jJ2!<*ghb%Ol72;bu$CJ70F$nC1dwpWxNdC^juXq>*xT$l^GVg!y6(k!s`BO1`O!mlwR1)E;OgF<}n@CSha!qHJgTs|DS0`?+Gh7FwNS4{M%$^@=@s%^+*6vsi}J+ppQ_F#lA@Q|0GqXz*78)yZ&4cF8QQt16VUc|HN4hPULfr8By=9us^ci9zRz8hhq!pYoc?@cc}qm}zvr^@7IUVT$vBoXmTzeI3IrZ0SyBNUb*G7R#4dYR1}dt`}l>NM8<^zZeFn?=2Dp;*p)1bEg+2a~k=5t5w}O2P$2kgpNMyH$l!ih?n^8R})S;HBg*c?Q58i+z5Dd*WJ#c(jE_5c)d*msLh>6Q=ut4%}st2?Ae@b#zReS_r%Vp<l76=S!qig>R~%-&|FLU^}9K|EwY^10!5F$L1R!t!O>gRXNzDte;UH&?Qx^DeVG11C1{o6HjO>d+dRC=hSSp@&+!33TMxu+CwHjQ^nYU!NT+bkJ;3c>IJv<fdo#{M#&wS7JhEr6zn+lz`j8o`si+HU#+F~Z+tLv(Gk(HIkD{Xuj@j15@5{P^vWuGwbn;*C^sbS9qlP8-OeTIODC_awc6l<UW$w(@H&YA8cZu$TT1AmIGS}*Lq;DLlj@dUOK@#2Nko@=u`ufYnF#dDtK&&_=J%v)f&W3hZYm>%!jydy8G>7A$cWABnX8Lru6{I}PV;c`QM>@C`^p)#}Gzc}G53}8eWUjG0f4W0E(+6?4qoCZY=WSpRY_?ZRi)phE`89{f<7_@W?@Y?2m(Oz`!KVba4$v-#>Q&4fpr=t_K|GFY%2{jY2|vuN<G3t@?b7<Xe4W@ABB1)mU~*5y6i;_6E{SjT67K*v?eoi}`1_c5ANnWZy*rXgoIM(O`JKV&F?U3r?r;4>&&tNxq`EkDWa<D%y}P*t?&}ZK>Nbu9(*Sy6GJM?z<(ULZar`XP->@?3)*8Rzqa;w<#3$2n?NB<`QDKCAp|%}YE`DQCOQLhT&2JnVKN7w;@2$*nwV1T7U0%MzFZVcJUwkC<A7Qj9g8_*X|9*O`=fn@L*G!llqO#_tqXFscssoXJ+-B=&3y?FW7Wg55Y7GL)b|1QMbcbsnDEd)dd{)_<K6doXcIq+g#tTWzpPlD@+&+KV^iNBx>beU&+@0P^k%r|UmlvC_(38zB`$bhrDl8Y5PIL#i%FJ_NG&`-(yE7d5FWV)O`+j=j06ON=^R_eS->>!}OL3?jE-Wuu_aX+Z4?g91Cxu2WI????IYiNtSUiAVfi|8!KY}%;h%1bgYMb6Ghjoo?cw{2occr=6HWL!RCpW-FckW=I?V}pp!d3R!2;M?^Ljr8H-MwzONrKR<$VdCo<X|kJGh_%vr_?G&`n;T-C$SZxCv5?|zUpuIcOZ-YlfP#*tcL-g&cfhoMFPKUKYBOyDC|H7J>0*ihq=+Cn}OU-JCqvYQEjXlzzM?7_05@(m$wj5clkORRlEL%yJBH43!#Si7jXF)bk4U3)z9U1H(R}2ZdIqaW8|zS^V#q3oVkJ&_}7I~l^?g$sF;b%4?{2a1HcRdc(9N+{Q;;v0VHDcf%I4lpKjglfqmh#+%W^BQ*(HqlZ`)^FF^G{MBx3vj8?Pf=b6rfnS#z#LXLpvX8)8JVpCnMSgiXQc}8div&#+^l+B{KSrEoC%u4Kp)E_$>1HwqWE}8wSn|Dw&xACsOd*{xu*c#pT8QX2pT_RFhi~5=#By3h-gZ!yHLjH}$U;g}C)2Y<DcS#tw+!Ys;A0sXmcQbFo1@z{I*8kr%+MVv*57CBoSMc_Z#tAzqzC|#6G+`$m$~V7L%M)wu(m}fu?jD-4Q!&O{+sb(83F%`omOhJge|RVTE<MKbMIMdR)s~KkqMd)`Xu9q_?(jhshi&z70sVeY%K8t)Rj)OxT=yn9FTKq_3?cPT(HIW0Ioiph>mO&d+4J9|IDk>x_U{jjTfRB)WosWA&Ia}wqVGNn<Los&T2OVRDc<-SRzIDWd>p-My)djFh|k2_hnhQZ-h|p*1fA?@UeE2{sC!lbL>NvP%hkm3jGAd|qZQ1$r(t!ly*zFK2my_QA6E`S{T*@CM*-cIwQUU|L#qt-_PoidP*lU!FC<Bo!re(Tw>O`E)ma`5SGx#rihEi{_zJ>W&9ywBE6b=mu&4SH-O!a<z5nIxddUj224{0~`mfdw^yYGDxQ`ILN?gv)8FMi&f%}~sRrWjW9Vgjz0k8;LTmJ^F{J9eR4fvRd&nO6g%5>njl6K<G;i5pnrP5HJm8^ClCov-solaDrzyn>M8@rY<O5sTU_Kz?NqQgj<IulrY*39?&b*%t9bEyQIG+znK$N&BS9(A$LdXty%Rn(_cTX7^`&|+*xHOcR`7$(655+xZ8TluktEpJm`-i)mg`!~ZOba83yFH!qGZb)rHc_z$kb-(Cscjy;UjhHj{1<IoHL1TWMQ+l_2i@<W)RFSwE8VT5&cv34KkDq2zV46vBk5L0*-a+Ge9KMJ~lI&lLpnHRWO*Y}N1;;O%5(dx0%VAM-xr&uYt)qS4Li0$J%CxlNnxEaCCD?MMO9ys@PZk(R=7N4wZPxDWN>-P^pDAPUTD*8+wO8JP_9|^>O}$bnd8)kc`|)<pFN}w^CTt07SMKE^u;-WS=;HTU3q82+qoCItgY`Oge>tN0QtW%ktV<r>_|Jz`S1Gn!;k6NVo-W?Gdn3+7t<rHwV&6D~(>l__7Oi$Hnj9kbyV&dN@5S?aw6eZlwPs$v8;OTrE49$xhgS`mZ=|k?)|a2r)gYdWMPF+2p9}tHCzZu6kLlxwzc_0EV}_O1pFm*p{4B=n-eni(D)j9e2({_#$luGb8-iA8QH}Cir~(^>_+bT#qw2+}Gw^uAK1#E+uB|HE{jQviduT!G56vy~6*Sf}(^PzxkT5pwh&O8YbcU4|an?fFQN1~tQ|w<AoNF%pIv1)Jirc`&`2qNFmZO86=I=!oEN%xryxb;`Y`V40BGVSKu)<jT+3u_9W=b|EMR1k!IT*X!nsgc^eYPQ$uQ+!ZNHkULW}OOya2{pX`nkho4%TeCR{!*FDgXoU@k*n_dsCY*n38`w@=fZ>(E8Xo!L|A6FuSiCyGp+6wgw22g~fSWB@V;DuooJC)+NL`Xh@u#fq<Hp;^k^9!@4l(jw(o_0ovXPG};F)?o>}RYWg<bx`(rqgl$c`(})l3vvqHcc8AHNtUxDxe+l4Ejc6mIJh;}UCi_eyVj|e>iL(uTc-VYKy_J0y8u;QY%M5_<N8@f`m-YySBX|Ct1&EcMwqm&U8}?$ATXOgIo-XrD<!1Q#G|vYLK0C+2-UG~Uv+CcvM1oNnQcgD%!ksp|=n;;STDhK5VVkw&X?GFm)b<mzrw5(sUwOb@;<nn2FfpFFBo27|UPF%OH8OJZ#(u;O(m#Dww<O`(qXaxmBWL&moRKiRB1Tp@Kfj^>T{vPnX`XJkfdCFR`|8JxdWZ<sa&zh&VOl&!b%^ro?Rg|{OI#1;Kp^I;+M-S+!aUbbtZ?`d4r$zv>drer9Pl!VU-`0=4l|Ex+6{N7&u1?`wFq#$4$dgmSWo>$^e9fd_XXQc5bX~<df(2N0q(RqYNd~&NO_Z1w)F!CI(%nRZw^XfC<KxB<BykQnr!W#lx18Rpc|v!U|Y}Y;4N0}a4FXZ!lHn7_u1b44T}R&`@r^%rWVsz(mT^`zKmZ8vPP1Z)V&;ApA$9#g^B@#5rQs}2NE88z!MZ;jg1$1U_Cb4akNzFW``dwLj~BEd<#3T$4Z__zGU8_NU^`Gtvj(oYOLZ&=g@3-F$czoX6r0o&^OEBy(1Yxr>{L(P-_$2bKpUlR)0i*7PI=B>Ywkkh4fXMwnV;gyR*IZ?TZHB8~Cvt)=u4%JDl?yAw)F#l|8(s1Wz`9X~ntT!@0)%WJQ9>)3!7y=Cp8MI_h_ze81)Cmb-kH@uv!}HYcd^9sJ_#^E$#+4I186bv4cQ`5&5=#|Jv29&0zIR_fbxnxk~mAKcr+rpRyKI7?UFo+)Ot{S;F^%6WJH^`Xmmpc%7b4>lpev9kONZHg1?7;7h#kL(SCljU+ZEElVY=ngV7-iZU_ry;L&*QL%BRFPmfwiC)~v_fIAu83*$T}@Bj8y+kX=kbRQ4TeEmg1&xDo0r~aEA0&T7cV0e>TS<PdL5cTvq>i=n~R-c*W1gz1{PZr>DzC<SQ=hd>AT?%GDn!J*ASVlhtZ6yi5hiV)u$b%7ISjSxtqEqMW^oG=~3JWrH^c{slDC-{cX_}bE&|LLwu@(ba}q%)C6VuUB>H6=D2_hHU9Bk??aIv7hngD?Wf3mp>V2LJ5n(?`xbB8-7B$x19NzVlk*PTk&VG+TL#l=8|W=Dwb@_ZuSjcL@8Q9N_P6^inwZmu+=xGSBBOuOM>jxocNK|Ffe;B>G|~>+@%hOb?olm%Z0eUT$ZI%$eNs-^M*(p3yC(|QeIkA*to*J_P1=aDnCJvus?*_$tTZO(Mr~bqS`{w4#kt4(l@PqhlGW+y`eF^8=c+`|I5ul8&$5Fp(dXf+b(wxK8-2o$%w5gh@Bso4ilDII(c`QIQ&Q5tSgSR<?p9~y;;b#U?Dl9ifoR+ST&Q{8AYx2inqv2S>+f#khc8Q+(rhoPdGr38+fmPTe${-!AjNFu^Eavd^VFnKA9mBvK?HQZsN`dn*f*D#JbUer#SQ%YdE#bt0ha`PJiEQv9h|UvQB4Og_l==oc{g?yie>5p)G~SslU{fO(jw<4{|#X0P*SZAb538lYWD5Lt12N6BCV3|+;j%5jHLZG691T0Egv|idI}|PrxGXIJZFW=*6VW*T)qE66;!QMLTJ*=Y|07m3;%v@JH3f6ZKdxCSO>LfBbjL7&1~hvYs>vU)VDHyez5bc2?1a16&EmV*Au!srSu7LU6$M_U|Q?Ry1jZg8*oon@z=8Y9N#O@BJR|XtdgwHw*FM*%4l;~r%a0)<6RQF_J;>Y9B!?&K6?E}(YZAr3r11=vow`aO*2Rb6Om9(rIJGz9VJmj4kdi{_y69xaKUcxwbpND_dNZ_ebD=YZ|>PR>jQG;7SmxPilq5v8dp!(&S!ekw4!)}tC`&YlcwKG#NMYQe*8A8*=(NIf9=C_-v7-{+{3@6$J45D8<!R7bWpj)YKWqo)d)auyNGOIIK?HW)s2++<QxD2>AU=fo}aC=9Uq1yX%!5H8_*4)hGMg|sVT8Pf`f0beP=RyZO(=7xv@Vg$U)gnul(+BoHy`Jh4(kSf-J8~`UXIA^Y$KBVyh~Qy0>aTN>+6PwbQs<D2LPfa6{jXxfsu9f<q^a<m*wPA9zsrB3OM+b;C*X848Y;^9qnMmotvowsLSb%mhPkbNzBZ17K0r`h#(MU^R}@x=G?q;y1GtUk3F8yf#*tH4fC~(0{lxGuw|3$;BIe#$y7Xt+AW<uqlB62Nhp!^(l=R%6Ne8<|B;rn#~8+Xr=v1+TLw+{jfmH+o18>!DFHJ*nQmlLQOZm@e{WS)qRhk%gI38*9yL1Tb!_^{qgfej*E5U5hzf_*lLHC)`-{)DQ*ErZ^9XvMC-G1dTL4|gZWYU?e>Rqpis&FL8E3`w2}U=oNU8;PUXYQT93E{$E&X&V4xiWK1XVE5uQ|!Fw739&$cV#F~-TC5>Nm#f8Imxy!(29Uv?&8)M5&xHD7z@lhL*4*H&ggZzQRVk%V>8J@mhgt$qvk-;(LVpX^r0-wpTn=7Mpq>hP5teP4e~FNRI%4MKkN2)|>HHuvTE2G3dwIb<u#lJ%Lqd3hpFh`R5wTeBVJm3<}cw=D@Z&j)_V4R=;+drUf$3+ZMUuL_%G8MSn!7s0*yQ4kbyLEpk%rJ^&Iac<`iXa{W5Hh_|C#ccCc>(-1~XlnPK=mF}j>IpiUNFbcs_;)lIGs1yJlU&Danl9_-)hxiP?PFD#SEnDPQk<y{LR-9m8KZH5$#L#7ym&|~j3!pd?oY#S?zAIUy7=Jbm~^?mz(Vcm1=SZ6K0bf&7-QoZyGY+M&Ty))Hge?_OTkg=_YL?}J&aBhSRXDib`Q5={V!IVN(KY!)p;EwOZ^KYBe0=b=%Ns%fk+mKePjh~p;rXqIx+=SG!ZMbAD(4#r}?3#v-sV@o9MhA)$65pu1!yfnC34(gXe!%r0;nVLxpSfyMunUF{Jb1JWWQ6HH+`slb1Nzb8ll;6xKu#n9-Qgh-s9*FrnsnzZY0;T`=2%r{C<uW=9sBVxq9p3GvQJ$6;Dunh(p2BEOv>*Xgbp)JkHFud_9!eB+;R4tVJ^^$b};)uk0-=xhacuk)|J@Wwg=*9XyJubAr~?i$$h+N+&!SgdT^g&0N8$K4NATHM5I0~hRw;`88XU`raqo}<&WeIx=cGYjf#VIAqBr(-H}M*uP6yX6->M*C<IF*F@A_3#ere{wf&IFAlrpI%vadM>|qi_bbyaYGa<rvM#d8#`}v>{xpM-w?`V77n~2(?M3QpEk(XaB`Y5%<1=+Whwgu^(+>x)EH6xoM8qBfETlvFt~?sRPAgkj*DO;aaYGn>2id^*TL|twcSsC=JoUR+*jiJ>ycbPFPJ!NPo_TCCq1c!mcndDRaGiXhkNk`xwf=A_bZ>tVfQlL;`6&yhn4Tjx$OLV|1iyMc*stGsD7)c9`r~BTTTT&lpN>$C-qVt0&DZfMkmF-^XoF|@z$wbvkF)LbIvQpCZha*qdpxx&53r9rV8{|vumfe;ox$SE$fd@tGP$sWBhevx6$V{_5HH=w&QsEa>Uw|y5Xa27UKmwFMreN+j71BR!ofXv1fWBQUw<Ac(4wkb?q3KT#!k_UG>sg8^`m;t;n@tGIxynL~T9%Wf`^_w(G#y=n;>qLy=zHqCz0f%03m}X}}$b2igRL13S37$Ds!&>jQD_Zd<^~9|fXg^<aY^1ndX*Hb_yC4_ujM*HLz8gLCFG!^KH&(!9pe`A!$7(4fY93Q54!Xb31Co_gTQrpA@Ktxkzn2deo=K|^Zu6-DgN5jVOR1Kt`x;LM{sta;&E*}+BECQ29^cVTCwKI0@Zl57kTS-RrOu|9gur@UpXVMaVz9*W7>m>fR4fo^EYhP!>LyUy;foUO{x?%~9blFp4Oq$Q8~2wsotiS_k1mv^*owI0@icr75B+}e#hupfLy>bafW`GA`>kNWVZW5HoW)>!RUBSy1{T$QUa`Y5$bIvdM<4X&KF=p)jX$_b-1|8XyA&AkpLa-EM2*xrY&gHH+DuH?HDU;$l>tuz$;MpdnMO{CaF)&REg$RVCTqf*~pE~}r$4Tj<U6blAdX2c>Q9>h3%J>2yE$X+++eK{7JW&PW^3bRhzyKlf|<}%hy{8|k{5z)M2X{APNgp9i8A#vK$&7wLmh{$2C27x`E$mj$K^Jz%5=CxK+opkS}yZL=+8rl_qGkLl5Eh$)s?@}F;sWk86U|;Xo5N(KTFFU?tAo@^l$niK|9)7?>2p&v=D9Vlk*CgLIl9zFkvi5n3O^S_4f`f7xY3Td5x^O&mepAiW&3ldK*y5{pkItT-q7P&`ZXef=>TbXZgXh!}AHkDk!+BAA1bfrl6eF<9JZ!ZWc|7mO3u%-7nH5#Rs;i7!B7^SyZFX~B={Q*MplNE&L!%~T)S9wNG>_e5&dTrmqO`%yq%)@cMH`(@H{#CJcH{COwvkD`?re|UYT33YIQ9VAs@u-Y9#K}0fC;%*Oc2lNdjlf7b@5nivf{;%Y$F|3g_hi`U?1i0gV#?Nj}+S84y{E`iWM2YdUQP@>X|eU(zYnd??JifmwWt-cF8*+u#~r~A$LOQ#~+0h&WHN-4A}Pfjz_KdN`KDB`>=*)>7l<*$E%+};5i08V&jm5wc4l?d)QSq4{5@0_dFTpY%`<MTGx0sR^u>hzq`Wx%7Sy?w%|J5FXro@^8D6?)S1mzK>&PCldZ5Pu@@<X*IPQ|Zsqlh8jkx@;7j%O7Yd%^hTa&~*KgY3<JD-V+%T)SKh}=As|bGt?X}*9ImRM@s0{2$4@tJe&Wp`({WJCkm*~}G2S4ECn6bSFCAAyIRXboJvg-qFZMEA?ol$<)+ICIQpn=k7R~$Iydf{xp-4)q*)DU<h3h8=tT7&79dWsk6=nb0hK?DLpygoceqfM~TCUkJYfA@PtMpXqWEkk0!mI8u@vzo`V+goVOX3z4o9xCyl5zXB5nyzjZY$I&Gjlb~*pwX{grAO0MxLe?6(0{b19Yj(^S*v6|oXNF`3Swj3&Jnv|j{L{gB%QISmu|jW`Uj9|^;WZ`f$bg{8=sS0fk!#<ZFN42ctf>LB<H-3T$F~b7dm-o0o={V(fQ-urf5Gv!^_a0J(T^o{-N_*^EHv#*&g|9?n&e$LiNDUr}Oe;&8yW)_sIJ3tN6m^Le!S;Zodchirbpdo!LBf`)#axL3e*PF|wY}Tx(6;ZkdC{16(~-s7W-6pXO72J}+{0HqCw2^akd-O16<}H7S@!a9XtGAwak8?}CeAxY;kE>1Rbu?oRh9Of;a?g59d<hCg!yF#JlzkH=HzL=3v7%J0r6|F31u=NJ5GuSr7;p$}Mxx?5{aEw6PV8R`aqDrRADC>xS(tXlnjU#;1LE>r(9;&|^e`$nz${4Tuk@}*W}Ek@;n>=`_8sYE?WEG7u{FkKeC4fs(lxb)V`9w$A*h|s%(l!UrBJLpu`P!;-NLHbty-@&tp;Ljf%JpETIIstp%Ht#bU(!NR>&340lIIhmNDt)b(xBYbTM~f+Yn5@G}m~@)$no4czUsycN+jIQ8#)C~|Qp@{{{l_k=5k}lvSG7v>NN3p_YEK#c|2wV+eW}I!H0s*)GO)gLmF^7ONfvb2)VS0jee>kq*;=nN3P2mwb2;L`13d<)L#x%CTn+o?i60dujV*Vmvt6=I7mMa6;9q^Ly#wwJnX$E_6Vt3em7N-mH|ffJsk&VPpptEQbK`1Z>?<nS%!FS&wJYG@4c2LU4BiuEvN>}#l^dN3JN`DnXhR$D5|75tTj2iS39@{rES@tcV@Fap(4rNwjlR7ZKI~Q`Biq3rrxzabW{m7w))3}nw+`~&HcHs1MxItD18)J$PO<<gV21lPHax0>SFd`U?Bf0Hc#!pzuQ;GL)5pJSy1iYSFMo2h?W1Sxp>4jtEvBR4WLGGED&wfs|JP2p#v_=DnRrmhqu#_nvbB9L2ZN#qe)|n396IHv)!>dBeqD7z3^4a_sps?2a<tgYb0Dt=i$d-#uE3`wxtALm46i|o4o+|{x8gC~g$x5+b<2o7ZPjaGLDzz0N8A1PRB6vNtJqA}Nfyj2po_sh+>u1(9(hVgH_7HIsV}94v1-dO-T$CP_&XZXOvi!Wde-Ir=j=W);RQ2+wE4Z}^hT3&3a(nO5hIPC4s)ieMprrXxc0T2(E)L)2TH5P49xq^ufNAiqe%Y#Oxa<|-Yi5;$o_1}9bTfPW1a)g<Nd~e<8UrP#K~xx%Ps9^b~zPy6+?3pVtePyb5_YD$F6@+wWy6qsDl$d6^%~!=CzquZFGAb&+fpH@G8es)B$j^3nFz|qaGeJ1`}}d(qfRu4@WZ|OV8Tt&Z3Ua`y;1Kb8U39ILtKv_+|54R2JUw1;nM>nVw}?m*-7`*Jhn;I=$g_v0h#};KQ2SV0Sv_mFFO=D!{Tf1%Y92f1CAe^c|gJo5fI6sd#X+!OXu&))AyKU#c+BL&fM^O))d#^f%Al%+=F7j&#`G)Yi$SWqmid0!lj=oSj~m8Mkc(2;8=BiwT|$KxF@>F_$8+pjhGuu^UQFWNOvn*{io<H0km2_HM)6s{1H!-%j?#YQH>aTmClJ^wYVAX{J+GW9ET`Kg~VT)!U{~CFabWO|-vQ9=ZixnO=xNfN^bcz&%0dEd4ozKj`S`O&Uaab~}Dzi+jn{_ExhN&7(p+Lb0}mU|ZpKtGjz?w)f~~YONTyQp1HsCS1SaG`E&?WyG~>r(;@P<s-5F%QM63Iz@o1K1Cw&RVx4#j?HiQ=2y>;2C>^Caf1b2Gw{l1b=Ca+=E!(or5d-juh2&f=*aapQM<4cSL*PoeL6kY6}l_le(~z?)9Nz)vY5}*@8Bl>yDg}WzlNMW*st?+|0Qr^-xkkI`Our@x&71mCSEdY?cLakQU7N}vYU86tQ#vqaWNWg;ynWX#`kS;sPX7ENOle0uF8>5-S(Y^zo)Z<b04#P`|Qk7x|-?Rc6U)3H21FGn{RJbpR+i8eV%CV`_Mk4_#cZi&GmwUdY>oXErtuBLu}67d|Af7z;wgw6C^gZJ$KJ5aZs_h)b4le$kexk_}o@b(nu$~<DW4ykK=4O-KzIrq15JqnEu)OwXar?nV=uK?&9>3V9cwsb2COIc#*4O_??Q!xm~UH2d}H`hjUexC;FaidqEWMf5Mnau1=kmqhVvmJll(|zo4Eyqij7YXX&GCVD#VyVt%}Ld))bXZo{?hM|uBohF+TU&z4{dHHm~SroHP&oWP#z5iyyP&Yi+YTyEUP+j96fG|kJngM3p7uI#&)><Y1pey!w{MGb7l@cl@id+5Bufph)B%3XS?xenj#UP_D+25ReYYflG#YTQek$Duknbl+@y7I?~oTKfM_xjhlDH7-|X<I|yx!gL%gR!+a68N*%N=cZ<*I$tLHSEWuKQ*2*EG_cvJ2L`%o<!aVY_YWP>l6kj*f)k^s$&Wi=AMH;UGN;wU{dt|Kd}|Qgm$M5k86P-<Ra0Zf(==u!cJs#^{u7#R(38-(zNyK>thNWkc>AEfS`1Hn1FAWAugB@<QT?<}PXim%hTka<_YZ@#koV7)KF+}~+z$_%qx)e!=tbv3TXrTo^s=UT+pQ#SODqL>{luMetJT4g)iPT9y-ew?&$s2cBgo0nW1&3>7Bz0$nMhJx>(!(#XTtqhFB+wrI?f}n%K2@V^VF#4yP5N?!>v#^(BqJB#-0vi{iSpKsqADm`G%-;`qa2Vd?JFK`S~6h@sAH@=h}DOg~+ypkML}=(l|R9E{^tP(n`nygtpZExOWmm(^KDi_qRoHFkN;o6)~BEU*^|PCyV2zHd@oy?GsR(W@X5|*W&S^Ikk%M91(BkHvXfoCXY5=tk6dbdpStgXWJUI^J@er{M|hgE1U1sW^>)T)@$|(SOEMAuvlkWd@Eql^($w3I=t*Q9%|l*G;T}R)-`@swyQy9E_Srvu+i&c<h^R0&5VDGY9|~}Nt5d_v_Z6o{?V=zc?}Mn=p&<P3mdt)WpBK6+dDLw!}X1*t=^Zzhpu{yd+NpscAv(5%-c<E@stBucZ9=x%kKDFSBa0j_0mN}u#iA2zq|b?+;-|M>pTL0+uK;7%ChQrURAR{7V`1<Gz*8{y#nyn_qMs+zZ{ZCFN6Ln^@ojjPKauC82ITqKhGcpS1#&jJ@W|7rkne*@>UK*XN>xjwxUMm_O=kp!fV;OAIwMV@xot-`d>mVOLt1{d+}(MJl?fM`FqW1`)?>E6>`TLM2}w=$sq=<-k+7sOw+nF!rP;)ZB&5GRXiQ^rgpKK(J;UI(YNuvpr^F-N4WLOQI=eeCe3vuG`crnQbE}GeV2;QD4W7k|9N!PcWKDi2Cxqu>hzd>wf172^5e^bIJMV}5$p~$Z)i<`@zE<Sb%1ngZ)Yi2r?fLcufSFxcTP8w&lBS}oI*&nlOZ0j@28FA{QZ@kza)^AJE91_7ye17+sLp+WVyC?0F0qIxnCePr#@HBHIgmQY%_HLs{t(1veq16?w%lSw?n#x^yVl;l=t(gUvPBbhS5BS4)cdw9~R}Pqnz}qU##l8_Jg`{{TuNrYsO&7oR0fGck7XtuS*SgAD8T)Lvg)R`8sk^_GCyoeLw5mIy*x{vD<csDUgU2tXzRj=7rpkmFnw0l4|IuF$On(_^=<Pb#uXM!H-)S5wm$eqva{wKU3s<Y6c*6UjctHXgb`#7p-m(_t2jG2l|Z$f_+C4GQ&<#EgbH%OJ6!{s?S^NcF&sWslW!$)5GS*&x1T5yVWB)3oz6F4QkRF+|h%Lyc(B5pU5HV>&C$Bs3got^prbi`B4=UEgaeS9gD}h6S76`!^!usP7iZ;8^>XLb<KNvP%|5key`gSCSQXQbF$Z--bVPkf$HVN|3RZK*XK;!wz0i9fPT&kI=02~ydTWu6AOpj+SFm%kA7R|S^YTs@KHX>4z8|ur866m{Y*$O{I9UMbg+v~PifV`1K*l9ey;PX_c64{v)F@T>1^c9J#OBIu0F6SefhYSwNW=mJqY(rotWi<`$gYx<jd(-=LIwqjk<cMSC59|@u|(uAN5YUPbYo#V+P8-VZZhV`iaRQwvu@75BRx`Mt$o6++d@61iG>a;W0+KxfkA-9Uv}VuhOyX3*KyC{o(e?^|rpW(Bn75*^S2|xK@SM)qvC*R6Osp<JJ0^B-Wj!kMhZ2m15+4)14e<3sK!H1__HfYiocutsRRo?X-%@>ktlk%=Xz+E!tpbzh4I$I6P_cIdci}lY8K9IeneSmAN^I_3gMXq^4gJncH%hp_6kfrMlON=cgH&-pB&AfEBxy8uPQ_=Ia;k?w*oh>~H=(-+nIMV=1Fe5pGW=9MpU5Ul!209ZzWRTsbWAC*^W_*)L#mlgzIrRHv`a%kG*tsB~kr&<T)hve}Uy7AlF`$+FV3@(L~6&b-&lQ0|;(i%zk9Paov;fbu4^pJ^euQv2KLV<VkV?^Lg3xptKvf0Whac_Q|ow)L!@w^RxQEB4h;kN^&_hCnSOu`+;ib+?ajQ5XYZyEwB3^3}h7Nz@zT=G+Vdf7q3gH%NL#jR)fQ`??pr=M(9)h4#?sl^5mcbd#m7RFhVlX#%Il+@RVdHMk3X>PN@4yE}ASK!z9L9U|E|+9xNI%4d1ijpMw;IrZB-_6_Ro^m@OCZ=RDiDH$?c;KE(0$C9!Qh1=8^LJNqn{AuGa=k?QJA6=G2u)U~u-B&iY_m>R<pxe_MvyEuS2)o50neLJ`0qM3!hsJR}+C<A-EA~l0YR_JOU$)ZA)&A6#k+T4o7I5^@Vvou2>An(g6hGTyLzXCj*8KHIt}I)O*cUk;kVm|~q0_^_mDFAu&$qN`$;TuRb$V4UG$<XbMG|d;KT_P4{ZD<duC2OkzuNry5Jn6qRQcs+H}Vo{<C^D185(P=&5gXa(LKoVkqK518?M3U=_U3~g40sH^0+S65Etg~=~B>tdFpkumh65puDz1JRTh7s-Kp-bq~Mm!;<dK8z3WOpv<m&kp9eGOj=`O9|8-TjHtts^8u|9o+P;MiDXlOI(tRZ{U=0F8Y8=F^M{axLP5U=eA5(XiHS<aHc#q@Qc+W|F^rwx4)tF8vh%KDhFg12hA;YzyF;wa+mGO6qyF>UEjaw7P>&2a*9n@qoLlkTu%9Ba5e8!2puXY|uwI(4AhJSg_-}X+2^z@pdbe8Ad_kNq{P4>uijy0|*b)*$S*X9Alw+rzA#0UK^S|fE>PVRl3xep%r)W31wPMZ!7Y6^VE7uk+=q)j-Ji=Nb)9u>SG^y57=dteTTraGvblf2Ou*PU9wdOwi!+hGF6;^v;*oIZ6C-*|JmUNq;B^^vo|;91_vwrRgcWqn{f&cQp+-1g=^aGdzn^<J29s~!3UF2b|TKGR0WeUXZY<^}Z|k)d=A`i#XzRbVS2c|<VS>LcD<8ZPmMYwHtt#yuVN$x_^WLGw9`&S+!c&K5Nu8SVrqk@5WZ7&h;;J;%%2w6asg4_|%!>Os;NMN0*CvV&efaqp5L^l*CLai8tI)dyh2Ib+xM@&lL8kZSF~><s)(=i6=*?R{72jgNO({zldU_Tc`Z!5Ck4x!5nwj^z>UCNP!h$)X9585O7plB$~HaVy(ECI+eRKrx9HoE?ttW;4U$6n*1gXJW0dMCZ%2J@9y56?^BrkF0Z1zVUG2;nR7}y;>+Nva6~K!I_~-X!<^>rw-L1VSDl*bh-G20Snn!7}k&j2AdUdHQ!vv)k0d*1FDLc!lsU_xVfnf9|!h)xl}q`J6nBwpx0xre@|u;JSbK<_9s*v?@k(c#`z-{>7!p%e?s+^eLTy_>q~aiRoH?iJxlknY`R{Cs%3V55t3=tH`^pl`AmHg8nS?Vei28fj}vvq<9EISTf5;GZ9iMJ)pf){7ssl@BK>;jUdnH;r@fY^pP$RWI%cECc^Gv5<h$=Ekde~5*T2WjX=&89MhoY7On-IvyWr6WDnn~k?mdPDcmLPege$z;#M5je_ZXiMs;>2J>+~lHzP%)vlV-0Eyf;x?D9vOSr5sdtOgT?y>JyST{WfQ<R@c@fxNLWQy3Zn~?znN=;X(J*ysDEa0LS%@o(rwqiR}QSms0c{T)?C!(vP#1L>v458oa8_&j$P%-PhI~v=YSXTTwSa-V8Q>3h%P&{kqEf`C=B})Wqa@T)o0wz8fyhSJ?Xrr$OAT4DPMD12VMo>XU85kFD#DpG!}vq5iRAG8^szp76xup5nx=C5tioToBfTyW$|Xh|ugSy#NY|tOJeSrnQQ-$d0=k*uA-ruadED)ST=DFK+hlv4c*X8-2vShEKbigRS!&K&G4W4$2~dzZU7~^XKdL?;xzfwDY^+hHnlV_wMKPp1AL@E~N{Nv_BmgIl5~<NsQCUtYFEbU10&pZ9%Xe)|TP{L6u5eeIX*a?y0Lto<=Zy|4SkxwqoR!s(V(10hG>rgoo<cyJN1e;Q_F(+DAvij5*(js%Z70bpVq>c4(jAX+u)<{vKU0a!!nHQgUMP7(eY{{zCGv(Pde^PyB(!ubhqlq<IuptY!Et*qIPEx<4hYZd$5l4DfwLsIWw@Gx!4I!T2ytl5+?DOC{D-5Xj^AWgC7|1JK9&r$F$*MAta?_Mn!tTIlMgj?RFKKW=|2`K|mwf^urS37F{QeX6@nL#G+E<xNh~E)CVXK<lpDC;YV4sxyutFt_~fm*(2Rz|JOoidu-z9fA&@4Z-c3;k)|YiFF^ZkN6s!KiT$IkF;h*dsnUTaZz7S)q`(lGb2CDU$j0yciPuSJH7Uy4+TQjGB}Vz(inCg=rOWwk{Yc^i&<^0mOX}-Rf%bfBELWmZtm~+<7~5-UpE6Gf1R=5G#Tl`M-SN(am7k&>DK$}o#`C+-mHbLS;uN}CEs}P+(6Zh{};m?<NaZrdADI*TNN!CV(v4+0WG<I(E$4laMHajhkq;*S~ESLPEZNI&35ylDY-)Z|4T@|pv5@JPY&RuLN-C7la7D5W{;2S36>azZ@(R?*N)=x>~m2TecX$Rdp{<q_Alea2e|4lnx}lxw)0b~!94Oyt@Bj+qiIL7Ci!eu5rcZi=4Rt(fP^RKHyf=Cya5g;Q|>t1xBIdDrritMzXLw0J;}wZO3K=i-!D3!Y*t;pzclk+Kb#qTJsxZi6q!bqZzTB%=_Pxso3&3<%f$#a2TiKmJ#G_@7<c$47QO4s&A1-MxVQYZ&#$wcbv5X9e}nhSu>=tHs0Iu7+HNId<J;X@WB=FPM8rOes4Keop?<U{YcF>)^pCNe4iKj9b>|(6CaGqJs5I-#R)N1SGCn&GUt@!q4H>pLungngrWi=gNO6X15%xEp&u+*r*Q;)k><y|$UPg-|#IwpaeqTCBO~cXvUjmr9E9PKhH=SDDn*gkqA4?qCC0_IfA2aOG7txisGJxKK3N}hku*Qu10VGz6Osy!jU9i3FkfDLmx?I%;qgL|$bcVA|Q?tN*%~Pq_zW=%mRw30r$~zKACcROGtcMLR7%0$5bnC$DSDCp@(f-9*{-M?S8*{(s&#h{tyBudY&P$<CgjAEx7Qa=G2fwR{t#WmQdc=>#3#K-u1>LaY{0Jfln6K0GiUqi<{!Opxh1-25L58Lk6m`cL{%F26$zB-oASv`%W*D~fMD!N#zdlnMn01b>I=?&z|BqVRRIP!nl+{chwRvfhj`0&zFMbp}`DKrxT&4L+l{rr)_twvYfsOV4yna>eQ@ca=$^2}?${!ADV4XL~estgU@iQTe*zu`vzANo*EK|O1CwC}#<DYses9`tVt?hF~9*@r5c|bmE*M#~AOwoUuulHtHUzkh}K5GwYfy8btiPD6bldo1m@Rfk!q$l*E8)vNg&<Zdmrk_4uw%>LHPc*s_c6ZltV}@nA(#V}H_^R*3>i|r7RO8b@OEr@XQOvM)?d;w5CRAnP`s+}jTXL9zx0>JSHp6yS+YBz}<FhaFNTYQ)Srvmhy*WX;pO|5~#}wVTetQOun7wksX8cOOW*i}OZ%3xs;bl!2?0#zxe0o@$`9>;Q>h5S9@<!EOHUA{zGl$f{?Zr;Q5+2Nuaq}locl>P8N60&c?fZ56nb%#~<BV{}AF5_7k(_!hYSoGE{aFwN65j*nJM?jP^?kKtYZ@;!%&i0us}12+?{}=EnIbEwCNT3|SxVH7jD=NYa@)GyUUgLfN($>=8636$q|ZM)jX7IwxYy*`=jZ2bwP^OQReZnK&)fLo(HH)yOw-XtngBv7{`u)`?X_4-gkGS}FJn%AERIlpxU2+i!TEnAt62s;kM|y{0|)x?-a*R8G%GmG@}th&+O4*x)u-53?~Y9Sa7?vM$oxKvx59JvZfmWeK9R%=pC<PgV!q4=X{`L#1xBCfY_x{``AN&T)1lPj@t>m*7JNU~{R@TnRjirgi+d-o(KVyjLy%ryP_81T7DEj?pp`Z8XSP&hrP*0fwdBpZW(sQ^Tu{?%<0cP3TKf5}{`8jov{t&e5R}ck8pyykLuW<^I~AB<1O#lh@y`Ni1f{Ub*mu)wUC&&*BO)?dc;J|-w{<0<qsa~l7Rf<Jp$$3>p2-HOFBa^kK3RuKz2=;ErwaLh#dV9FH&8IToa4vwHF8USmL|q|#h=m|yRc{XsT<U2N4}QUZ5*~fq~W(VovG63jkG<IV2^99zU==c2Qs+|7odfZ*RpJFC$cAn_{ZSg$Lg5>KAGKehWm3f?1sYpP%Ws@!QB3}ew_kplq}oHXD{E0`G=J4H2b~Uxq|bBtO2ZW7zCH=<Ty<$QFPe+2~?)u)Cbtto6_~x`fO&JkcBDO6=YC2>s=dCX6pI*YW7!dtK3`y5um#w?aMc+Er9~n*wFfi<XC6q(!%+Tf{&{gXV!R%Wkod{1fl9Z!!(RM^E?>z-eIq{Fm}!Q`A(AZ<LW%eDPkS$Wc96VuV}x?=dW}@?Z0teeYz!Wn>EPHBW`%a!|_rgUz4H3Jb`Hiax1m1;Cxr(${eUHDwjXW8WW+&8Ki&d-nXyX1}JLidP{UIV~X3on$&nS4l7{$cuW#je`23$UFIJ8x!*tA9X0D*TkgooMw(Y)$Aa#?C8UQeiT=Qz3;#1zTb+z;uK^U*JhHCC)6@{L_WpT<I2D8J^-Yytlcr5IsW;#Mc$M#ke}Cdjy%-RPoOYbf+F8T(Q`NCY@y+X8z>YsTXm2!bwqpC^)ilVpH)AWV>{*W-hR}P%(d_v>YD6;o>W>{88m{$;^GQ1}e<1odJxcbyn|({|aakS@#feibQRkDp*g;#@`%$8P+jcSEu&w)T4(PqMQ_0sYarc|B;RqaT@aMes_=)E^4cV?MKepv#&<<$8T4nHrU%u><G`Hld_@45s<^eG)%U9c+eSx=1F72iGFs$Wv@u>EO<J0P)v+khMpV3`~x@O`Ni2&7u!~Tz(w2MSqqsj+$<qcn{Xvkn%)8>8JlJ4Z|i~JB`E!Ix=s8RT;!iW=&10fuo%={Vks$QP_9A8nX*!NIq1$fii8_K1iJ7o1-VZoP>!R`D|$!6Q?AQ?c-Y<yDDm>mAK%pKw^{CvGjwdS6|;}!&YUjO7fr7@hz)2d$I#{;|Kjo`NKoY$)jyNP!jZo}3%e_!w5#A%73I`Q?WTiLj&I*6{&5A{6lSxP%VL5_YV3^ThI;%+`0KNz@sySDmwXWX;HIaFCeKk0nui2Wc?pQCf_)8KP;@pfxz(U4ruTF|CMp{sMF@3g7=>IV;APka?!0a}hX?Pm5J&6b@?k+AMgu`Y|?grkq=%4yQ2cw^eCTYeYtrw2{;hq9`8^Bo>Op=aYJzIRMJ0pwSp&fKnWB2Lx}q@HH;uP~Fwan&oPX7AA2rM-EDNaN49E`4HXljyU}U`~dYjAh_=ALL47A53@hBre`JYnl>nRUcN>h<+`V;0YAVxzKVj8*0;gD`u$yP8RcfZ2yQ@OJxmPEq{~xrb6m3ePxJ;-qDVsv45R3YaO+ARLB)En0J}E8r+R0xPfH0?WX<Sayl6{7#Ag-_d@KrcN)abkfuz}mhw@CGo+~DXn-g$sK<Zh1EHc}-u-I=%Tu^tlWwQ0VY_TgHaFd0ztLs;z+m`T7viP%Vx`t2dOLMy!pp!uTzp#C*JptiuS&;CY`pg8eDD%B{nch!pA2`fcPTXQC<bbmt%_P;Qf21c9SBe4<_(8$RrY=Y_V=z8o7+5@=AZHe6E(lL9HaH)XFl_bZMj8<juT}5tA=&uihtgn6Li9*Ct%_w{bM#iR)+0-oNVn<h-^#?(|1ki<j?W`yP=lC^twz2oj4aKUkIX<zijb3jRnpS$p0rjU};Tz+~EHCQP(Rk(3-#XBL%0<QmZ)F-ryuPMi=Gk58H1Qa)|CQu0;go3*6m`wQQIvpSU<MuD>%%?bWtkXNU)yIprSdWJq2X<Et)kWYR;eL+vm<#+nZF`YS?DUes1SH9n&qx9}UQN-D+nA0+o~%7B*WzI~R@{Vlwu<`mii;4JsKPTx7p%>>|EG-4?9b(h|NHP!$iEq}icBj>Xo-goerp?%5?#U$t`M~NNBIbE&xOJ&#v9D(e(1+OZVMfW!Y$M5onEySew9vKE1R(fmjP`l6Zd6gdn2QKIx^Wosbe6PN_d|$U??n0LLiv!n|#pu2W^8=1GW~b&P@&vHPZJ*{Ns;*G`U{Z5e@;Y)64M)-8G}dv)KO}VCNBD=<dL+KjjtAt0We76tC2&?hpR0cpL<DGkwvZ!xXd{h&yGJb>fLw19lPG8pcy@X~TLpq59DFnA=5fM^Jpx_nPi6qAzeW&5;zC3BQ2ogGPIWDTn+j*^V2ORA`Fyb!qIksv$my$TGjlc>+UBRI)z9j1%yZQ!sqwPcSKDsvuCtcrB>g#;zjnsFS1XWf{n&iA+m+24^p^J8exj+-jRccM|0uZGzAdr7oX$wwi54@uKLK#T*+{GVuX=25XPKWQ?Za4GsC+Mud2SkhQzwUM)+Sm*e&(Fgpn4yUmX3hSvyn^G4}Zm3pOGRtw)pzbS0=6Ic0OM-EJBJ;@z@*QirHj7yS2GHN`94;%kP`mlv<2QPYc!OTWxA_ta|9@Q?)^>O)KtfUZSV=z2y?J<Zl8ZKCuu<8sY@KhW(|yZddsP9b+T(wCi^St}xt?VE!nH>#o_g#gMz`YsJf1Z0LF*kLIwpdz#BclrT&YGVsDX?d^JYS*}mpIwu&R=_4Im4fI!qcT5TR>&zbP&wgW7*2=?Di?i{k>)y{3y{usdIIq?BFrKfzeVu3>2gAp%VK>5H!Xn74?iX?R^m_sw-L8**&NMIGRyi8f`MBOW%s{<f>)r}u`nyY}H1C;|NLW^vbe=g3b%fs$xa>F08bd=Y=}fC>#394^7J_%!dD`SpTf1{#Up9~1#_n^WT+31~TKwM^+56wBC6a?hp{&fyYu-}7hK-tSz>4p7-TOc@UWfZrczRT-br|JYu>wBe`tm+QmP{yu{^phrYBj!!Ra%>nS+HVFd=iaF1(@U8KvI?eEBSivq-z?FY<WFhVC%%k7>Kz|R<_UG9*kYoDtOJdZ`S!~jL0_bxKz2#jYb~-pbc+mDm7(W(tb;Iu+0`poZ0g`_ug59x&v~xl%lO--Y$cxZLF=lvpoQ}k6v5%R+1T0tMMOImrug=gHvOxUY{GVSqa=R;`b)Z4D3HnYUks{Mss0)_{x5JZu!DH<U{tl`M%V9N4fuUl)pwD+Xxiu?{PBCWBGq?6MH?U(+cm!>=Nkan7iU{gj5+fWva$+Yvyd)9tCGVil?zuefIoi`@_AXa;eh2I5<EPXAw5KJPn@-2utBLd^>iPM+3>{)39|OsV+qa8?~J#?`MZl<W50lX@ow=4&B@j7d3B#JU4^p6o&d9)Jpk3>Td=g^1dG$hn0QiKowgf8F_(^F)y%(?Uk6kiUe4{{r#cmx9*jsZRt1EH-`hItiYq*R*Mm;Xx*joq3PH|cbRTJu2K{!y{-ULOu#4L&Ifb6^MZem-h<cGg|!vd_2IltBN{yDL;Cf*@J*H`l4fVelb>D(ST#*T?vF+2)SRo;K&s4uco{41m-N&}KgvpwM3M6@y6~-Ro}HTdfrO#%89~TkBQd-0=HVyjS|+_G@--<|bngpA*2XBgQ+Lb1ztFRn)&cq5g6dVmm&Ds6+ckWcsEi`F^~7stXaV!fz-R*}Mv$cU*$bj!iHpNWGb}Y<I##jPgiQkQthq1tIKCjEUcJjF0xqcswtnfmtXVX{hT8p+(1OdL``&h!s{{05Bx2rn&x>i2m@jq*_O?B`VPQqaQFsJE`zm}qj!vUq1Mq)$_!Fk;YxkQI$JT{jK)ch2s~t4_nVY*^wpt-Cc6l0sZf<K8_V|8vWMmlrrB$tSXi3R?x@V;5@~qJ6fqWDJU#CY7^E&ol`_j7WeoH#^lalSqI=iR8{AV?+RpD*)d~ADoo#E&|%;G&n4nD}QI&M!jwb5vJRlv}pjrSb>CMCf!3R)R27&Qp=)$97?7cM&F?qIERZjPqF%V&P~d}#E$SaK&0h~l+ECvLsh8yAmzGWwcBhIKko49(S-TQ2Ah9`g+&SL)JzzdsS1&2ppKdJ|^%q~zog@SxD=4lnj!O|Jn{U?o-(l<J$`2gwZg+s<6^T^P8GJF_VT8Q)?wJRk$_){%blb2lR+taGY%8?T^<#ob-3G~)i26)c0CPH~}=I8OBS5mVfb2Gt%|e>*$kFLNfVO(z5~(>p7}id$`xJNjnb=`JE)j6_(ZHGkFt?0JrGO+esz#?$Yk1T?0@uu{d@Oehig5eu4Pc9z*e+v`8N+S#U~e6S_b{fLF@uSQ)Q`+~DO<bVVL`(2mwX6>p4ZFzU}c}r`=7%s@=*GuyF%5aPPo4gNZK+|(*RAsV=n_lm%E>NpJyMKA*ii>V!hg%^ZnTyWO(&Up>Xf8~NS*MC>q&4BwJ~~-ntIF5$L0TWS#Z%Dv`sgc8*PPssdp0yLsKITl&*)5Z2Rn({H5qBQlV-;T{jN$6|Mkaz1D3+_7AHL9uG$EvU>}}f>4%|aEZ*6zxi#!H?$y-X&d6L`e$Dc9?m^pt*w%`wOix@5r?B8Q?WNgT-l&hP@=J!((iXmk^)A24Bz>&5k6nL))bvBSBijqTgyh|mw=26Le?o>{Dt-o=45(giD<<HPFMIXAzAbJs<FTE>ptY(h^O=UOS!GHTXH8TUlKjmcWzaHld0hQ##D1k~W{L+;=hE1v3V#@{M_ZT`uJrsYoH*#4d-`C!X<n#LR>sc?Jf5T3$~!!XY|$0+Q(yn9>OiXR&Rjh-2GHx!hhOKbNkWH}tJ|#+WnEYMNUJ{&s;RzY85;}x<$7UWSt|%u5><6BdzdbSM<#;D-A~(^(r37!fLe3)ojpN?8t=<$Y#B~I7nbPBRSuu9rc7$ADXVNpaaQU6-Bic##0YERP#O#5)!?;T&W{;zhh#bU$7J|zI%%*jTMBHudhB_ew1;X<rQ&aa<45qj=P7=%qG4?d9%u*|wc5=?r8+|RnLOYsiVrMLMi`%qPS}baXGnC<u3u<X{pKoE{gGjddwKzo$@Qrp1eurk%_FF6+-b2%z>Ysy)(-rZlB$)Q6ScA}v}}oD-`?!mbJq)x9p`AU0oQxHOSUpqTx;#D-5VugV$Eh3(JIopMzzj6jjeuL_YvB=TE`xHpXCG7xhZD6;ZAEloE~rK(3y}8hT!^rXVz~Y6PGcr7y3%=%X+%qWt$W7n=yv6)rl$DcPH+=>}KIA%p&+g1n*F<Y=PxK9l`OASpgHPyNUKU40+wB%!LpjO673OUL?(ZBXV=Fom*Qhk%>34uNd}>rg$2y*B+R>S&D~aUrpRi^E~Sit8IQVA*^amwc~s`GM|(WZK=jGUx;!sMNqEMPK_rYSdOyJdcFqS>-2ZFmn`C(Ror+>J&djyycU(RvstXdb?Fp<^-A)x*sNdeNnVa~_VNEttW>@&+UD!mzaOq2^C<;$dva3985*=Uht=o3quph;>^i@zXk>)Fm<*TsU-yk$@V-oX!z;@Vgit}8&r2GxwQlP=?+zbsQ9gFihFk=E>OS@vcG<1g`R2B$pGkAQ9PWe@`FJ;Zd7qW8@?EL!?`i_y#>$QQN5MX6#!%nD{*tboC9Jr^E#LILwV4y52IkW1;>PA2xTE~R&AJWEGO^W9EH5Mo?z8dui5u~*Kp6OPahQ0Io#0(p?)Rg|BI)f)X(A0?U+x)I&6;j(m-Ie5Y>IA5FhOVB{YTNcH6IH_arm<|(iGE-BxO{JO65=};-Vx{4$+ASpZ)#6*IcmoTEF!?G<zSt!1W2x*$qi<v*j*n2Nak=*|L-E_4=n&#*$z4+1>1OlE$a`!|Wrhzq+e@e|LBky@mqjQ=AzTiXnwW1=h2nfpv_BJ1Ft)RRBh%ghMr2^691}wi>V7ceka8PkMU3x-D5>GcV5VW`}V)VswecN2hcElxa^~0T`Ri@FLtpHhlv}=}oQ&wL8s#j6Yhn@d`ORPOFi9>jIm;X2{Ea0?FseC`Gdq4V?HD>e_woIs+CLL+DiI<{IViW*o5bpZN3?6$*U-YLB9EnUax#(%`XPFfU5hY<M*O&gKQ5w{x;zvo9UQcKWC!?)U<G6gN=+OqbH!wXMJ1M>DZHzAaA00=w*g*`=kvc^x01KAmfkku{8)Ar_a4Qk$gh7b?=z&PH+9NCWSm#E;*cWECy_e)>W80Y=~koSZ0gGtDO;&#ayPwLtb0X(Bu)_gFvdYOimyX(zy}yNU=iD)%gULmtY~5YGm8YWF@}U@*2z@a^vbwLPIqUWyKj>mZPiHF_{Y194G%n;r7I;+xD=r|0~2N3y$oPSkeyqhI;Ky85u;WF#&6?iSQ=Hx<9mvjN3OOpdrS9QEKOlba2l!=`XsY@7K1dk^20QSVWk)!JDT4A4KIBg@@4@3XJM5fy=lK<}_ecjaw4ACIMI3@(01R6O-bZZKRcybPu!eT)y)Ul)61uTkSsW_-mWDd%!gPkUU?6K~f`z-aOZci&^?EbF#Ugw5;03Po@TU~8|3lk-OG3_cW^CR5vkRW~N|RtK)dXPr9S+M|LGQhRlHx4?E<?A4Ur#eZLIEnh3`%@!-So&7KQEs>}(TSqlyhW2aonkH!7n?7F2v&BQwcE_R;mM0n`;zO;SgVTH7^=>Y^e;xV9aBXSaYfM&0D%grIkvt>j9vDx*=Ci61(d*7e#4WS>bDa&>Mk%+P>Q&xE!+i8B-j6;}U`LQTbeNgyt07Xl&*2AZ{k^G<0G;+)y@srhI_Hj<_qoL9MM#2g@bA)s7S?ey5ycDiZmK=qOky#YP`0a&wz-<T0$ZSV`k-2OYJJRan(yfR9e!pxN_ylcvyvnjG0TN=l$zG6-g;?!tPNZdRrxG;ZflR-M`Q$Q((=*5$`;Y3Q_}$MmIAMz@$-6!eQB`!@=~6fYtme`i`zYXxx&r`$JGE@s>HHC`~A%mSQGm8!KuE%Gr_JsLVt~@*OsI8Ay})XWW$(@#!G=4jBdd=n`)J1!+a3qIaMcu&ZxJZU0;!!-zTlcrC16)qP!gpZlYps&@-ls$9=Qi{(siCp8Czd_N~N+Nd1L9C}}Y{P;bE)q$t?_B_dM{m@`nKNpti`Y!^|R`W>E6%a8T_>SJbTZ29gZ1(-XOG0mj0bMz?+sRNo`CjrOG+tXcUnO|R5aX~D^)o)F~C+C?}dn)Z!%UtNNyIVt9mhbe~jybA)fp&FZkOK;vWO>+2+oj#bc1rCwv%5|(DsS&q@bt;^++$iVRuKGph5QOct=8cc{NNyWW6XQ<=)~;ZIBY!eXWzY4^g$+lvGL33I*W7F9z?OfGqK0BErh+n$I5$WUU*(WVYf6VW;|zi^w?^noA~*8`quKR)~5AY3D-TEr2_+T=D3Pn<n4TTTDEy%B<GkliC|iF9=dx$1IQgS>k;G>?wo`&g;1Cmi;+ELkl$}a8wq>w%l$lI#?5<h)A{-F|3U5389&pVp~sjcc3=3*x{;FfGi-^~^c9e<ngv?_)R^cL%?;TRU)R|DBK1jHXm^gcY~e!uw~R|VUcS%wdfMU9(_hV^t26bE-fl+Mf42oca-PGB)T_QdrSU;fy-PpS2XdKIfe1g>M{vQk9+SG6a8C_SzYA&sBQ8(?`Y5`=&A4$o?%^kc%%^p4ZQ3Smb);S#pO$7jebL_w5FQ2LMo-p6d9LutUfi|39aMP@|BUl0pYdSbN}Bj_{-+PdVek8%+mE<(Pb2PW6)%FgjpCi8@7co6h?;UUnY0(5eXq7pugEMrh4GFwwg9ie<mQ(IEes)(FwX66>-X88y=!T|jSRg1WtcbCV>4QwnrDUVA6r(hjRE<p*SgrwXrXvW1Gx~V^b{|^=Y4Vh9GUkFH;yVm6#-H1yiFzeEv#L)f)6%~J#p#X7aOfng`~oPCsAz##f>#QsFqO+J6^WMY9AP!)H9SD{J1Y5%2mT@htW~dtIny)!r8@^=|Y#skP3u}42~R0Ho-FqlJgUc)@F{Jb$H~B24nDC<(ouKGf@|)H-wb^uB`7`6+NGsL33!AeuwKglDn6<)#b;5T0<U(>-K8Rw{^Do|0lt09rw;hpBZn${FUnceR-%*E&UqP`fEJ->Wgd*W}CoYHMBAIq%N#8c;2hat7+c#58ot5@<q0b73Nhd{TD`PH=qkOn0TD1n2;;+*Dw1lLaKOe-aN7FdB#(Y$r>g!t()#{2r`S!rBCr~HS@7e2{fA>uT${TNvOB0#*IRXQ0&1Acre<YSm9g&?#;y$V7n0(mSytPPkt-ZryXI=81+`#h@bndN?CioSZZpFIwS97k^PEZZmN}?iFS)|SN;N<Lf>xKwb5noT&F*)*Xr;YOHa!~J?}-)MQgv`_R;sNzq^@)9|Lc1Ufa<peLhXPXUVUe`x}NY?^9#C(+zo!o7l3&+*TPlCjKH#+GY4)c(_^W9n;Btm519)jEh<N(8YVbnF-D84;i5%hR~gzzPAX1eF>gc!s3zdJCr(9Co$+AZx=bfox8?u{lmgMbA^j$-RCQ-MYEpQEQdpO5##q?vglbzCBG!=ZnDS9Cp@1HZ%~~A{e|c^q_e~H`^5<Z-<9!X!N{CMjpkFy^E3Fg8)oscFh3V}J0m4Lb8AcgJO|Ec)UTp|cH3N&nj1VjU1O9iYLIKsI2<}{tH(3gFzE^NmXGe%I)m<^gS|KgGOG&T3$Rv+sqW%ee&=iZu)n{SWt&XY-o11HW;XMJ=l@gh-_iL0?-p``Y9j!7JV9?@c891u?6u}^&N}eT-f}sFKK=5E>$_Fs9DjuL#l@1W@@AEhn~}I_0j&smCGG0MHdjG@sID5F_QB~pcVN&ezHy+nu5?l#Kb&=RjM@$;_ilXu_i6!F7X{tTm+~d7d!dQC1Ko6QGO)td*M%s$^5I6!`Xh$>W@fz5z6izjLpP`(<@dmbn;gotC9=g#bz<=q1z*6*?7)i9aA?~Z@q?Te6Z(s@H}9cdtoSggpU#7)?37G=S>51Tt3Q%AYX<X;7c*!jd_A?UeE#Gl^e}Xfz2|<THP7A}eyZ5Lo{j1LwZ7F#Jo=&YKiX_BCO~dq-I*^%r4<nK&jWb7QR8~Pu?pB}zt_c}!lYAhNmJX_{d2oeJXUo|$y)}c(Y>*5tNUWo<4I5Pba%+UXa2}}>GVmc!1{2PWz!&YF=5Q=Uv4gp7zKNekZT;<#S9SatckwCJt=+j41qTGVPRRlX6e+=q^eEpGF+%fUz^MFdTHOgH43-|EKxSki_5BaEI?|&NlCSVUiaezaV_g2YmFP<;aIF#HfL~x0N-z3UvzBs=%`|?D;Z1i9-bynI2%piza^`EyB<D&5!bofnU<FhRBa*~WuOhsqjJ>Wp11NjRA(EH^*RFK3psPR;DwIVR?G8TZP0et%83O#`TOlY!ATz{3Ld*h<oDh?)9=Dpe=;iE!r^Rw%dZ@9g`!KV!Q2;i%J*Mv;<$sFpP9$LEDD28gRe`_bGGbE9t}16c3)tx8~4dpFU#e%f8Vp57zy+9I0?VT<{XQITWyiCy%upEXOtr=x7NhLOG7A5JS{hXC_SVh_<#oRq1}P4&rnh*x^8#mYGau+yf6a|eLMyhW9Gxuz=a4pPO^t~uh35L+ZAdJ_u@lj1ZvnAuE$X_U(P;HW8po{^>=LB&jZ2HA9+lG{cSV1M8Q>i%-r{oyV9R-Mg6SZPG11{RX3KF-<BEMWRjzHm>=zwJs;rMs1*p59L787gGdzM&Jgg(mB&{-xhsH>XjEN-m#y5-YWr1G(3vA2Uzcc8Y1i$$fN%BYgcXaP8Oy>75l+Vz{+vT82Riik^arlJIYcQqs<AWdn@%_F8WSKM&-8g?geSD|dc2&X)!?o-I%YTOi-2okB0Hn+(u8p^3Cm;0PKoB#&Xq^r?u7;Z{WJKq)?+%sA{96CurOr~sDGA8l@?9gAgk?%N(hWJJ>v3y9J<PR207`Zk*NOWVn3E0)I_xKzFZ{yb=xPkO4;scwVFDc4O{lS953*-r;}>;dVV%19h6#}R{i&#po8u5Co)+F+;64YR7k5ky>cy^!fG_{{d(VbLU)Y(d_3M()>0#}J$~C-q8Dv(Vza>T`S0!5dGg!|<8!^J8gsJ?VZPRCZ;C)k?|<$&2PD{C-pq*~-8}7CSG3hYBQx|uJa!PUCu<1h>|C=uAXqXxRk1&ZtrNl=^|@XTO6?E|_=$L{QcsG$!rCi|&fDMhZAuhn3{x#w2$yGf+*Ue0-JM;BcUkYpe}WNj)J?si^?9iju3c$x|BP1^EAKp~;Jp!qSpD#rpPKHcY4s4$t{pz8qOBR%R)x*s6hVG?m1-;Rk6xmps8CV}TYpq?dmUQz!_P8t!h+nV(zLHnMlg*n=RLoJXXUi^%_3ButBKp^O|Spk_vGXb@SV1~T%P-bBgi3QwhX{5?GfQ#xuXm2w&*wTYaR@MpN8O8cG~x6NTo4$8e89+zproOa^v+gH78A=xM&MPdhkF#7ddXzIQy+X0b)kAehl1naiOd+CQf~#*tFj_I}i{R*%6_}HYMd;G{nh#I?g9^h8G9?up3Wr)$og3m)iS4lzVo$6}jBzQ@EiWedMsj9N^&2p7YfsU8!no-5LefUhWsvb+7m(WvzD9jNgcJp?@7uZA8EY<*l#ltw+7H*$t?_F6+P>zt;(Z9&`?l#C%#LKPp9jA)>$RTTpX;<6W6~0y$WKggXHqSe;ewQ5h0MErR7H?X7Od>0kP}cc|?&<2*F$%Y#6LA9Ds#m&`z)1B~cQwE>jK+d6fNzOx$U$MLV+hw5ir^d|><rn`NuF?Xc6uslQCRS3+J7@;`Ij#0Vt$Jy5qTE=^i)vjL|ZrwLFZTb06;cIs+_`a*&VJZww+?eCAS-U((R-l;&VDnnuH8{I^r`zbd%HQ;&KX>ez4-maeEnSy?*{SXR_Da|`tMV1>J3OiPvY7W00mlf}TKigs_-s#sy%&R({6<fES?rDFINb5u^+>bRu=8&B<eyWvKi;-XLB`unEQd^=_g{88%slZ9jTIrYwTU-48Gaw(bQTB~KlF~oVP(|PKVeS1f2&J_I`t-7<8i(zf?y~}{{s7voKX#DN`N+42K#VysWqW?x3N@&V{cz+*F2!hMDDsa-3(=utT{K?(y56O)l6?xn+#E{F+{!@AJNsJMf{$+S>KZEoy2=rzKb~Mcw;D+>G|~CqU_LjL!y1~*fq=T)5})-OyyVFkw)qdxWZj-bIZ$foy`Vg7@o;5eHkrb$TbxGlcDbw9X{~&#9?2?^R#bU+_usk1xu}4pX$k_dp!aZv=6L?^LlV^+L+64dy*I>+i5cx^0YW~)|IZcc{EU?vb7p#imZbV_HR87CfQV#>f8wQ=9(B*S3Rt{o~w6WR*wYGllsHbEP9v-BY&;k9J!H67w`v7<+VKmw=o7c=QU?%mS6I;7E_3Nb4r{LbG25D;qWvZPMY&Np&--joQep+i4|!dDK=6Is4mM$mzEc0bA214PXvYeIe5wai49gEzZiYSuRsUH7Fa0a?4iP;l{xR3e3~Zr{*yN$2`vn|Du)%Q>9zhi+4Igjgv%3R{VvK1@J#YOz+ljwWnRHazyDY^0$HB%SF5T?#h;6Y9)-hfbgy)Vm|*RW99tOXcgLp8^K$CkEo7Y>e`((LYd+*REY3+k;TUj(eV+)B|A)Bw$w@|z+K7n#N66eaTXo-WNP4SqDj%V&4&5muq^n<zaK4Kf#iLxg5*r<eyBiDUa~-`KpFVTQn`Aa0IW2Sp`N#G!ad&sP;?>TwkD|v=?M$dn^7*54T$}7{6}2|gP8%F0o9vW0LvfPn<)QkXy2hws?!yEAw-qA~@Xj8^@xh_e>{eIpGyNnQ^=l87@3Re<p5BK|vvu&9#wz-Bvd_Io?au0Zxt89iEU5sRu!&J)3Vb@W^r+RTo_jF~Ew)fjgz|kIc7}k`R$P05@4b%*PuaJv?I#SHjFnr~JB013-^Wg~1)Ej&_ittMC*3`}wvq&=z5RLH-kP9UJQVldQtpxM*;m9**Vg7Ya%vC#G`>zohhXFfl~WCzwP13mD|;yw_Ft&mkPtJ*UQAD-(u=-^oMn&lvlkx_cC^rqc$*&XF07+q%I!Pnt73q12PDkWvD*RTRU?BJ=IZ0co`28pr9R_iBJM%pf_W$&+TWDma{U|7s^4OAYx}r^Gr!nslP7+MHt+qPI9A~5n;km0)s>z-wKn~T!1U}9XPz`ORR<knu%RUmtiCg1*qhG2-n}DY?d_&hu8$QE`Kf(gCQLBevq<%x$lZZkm%W?8J+c5Gql?tXoUK*w@$!J4kPHId*YNs<%t#;9>5(vw^~nMV8r<8hksS>w-m7;uXzpS`EsZ4wB>g^x636)cXqob6^R#Oxq5mRRx9MtTPP{)qnq4ChtA7T^)ig`1s~T{7U9KdwMy0l2M}(ccpw*4G`$_b%R;~mnB_{#yiK}DH8tG2iyH@8-78%VMpoji$DiG$kN%~609p?EuJ_TZ+Ppo>?4H~n~%sTCQp!3Nq<NG$NP0==j{1Mvux<hfTJAUkWc?gGZzXGfER&xQ;pIYO>N)AW`OmC|1J^cE?Qg%90L~^AllGkKrcr;mms<2i$W_q1PR*$6j8K8@<_X=EZ)?Fpq#)P)3=3ecsu!AE^UGKc^3W0x0`Uw<xQG2-;SD)(clm@OfpJGS@!*kui;KOhi>4Pf_s$P5k`HZWT?@B0&d2!Tt%p}~r)(kH}$&5V>`wvGx^j((xNwzekj}zp!+CQp#Z?#L{bm@1?{=#i-H={NvXjiA7{sa!@9RI#tEtIDegZKJSZ}7dE+4*X6vyxTO3bjZT#z%BuNba4q>b<LI{^PI}M>Gd_q<Lyir%(+Y?SRMdAuYz$eq*}Hi&?2v|0@5u+1mK8gk_@Ad9`%}S|XrPawk@@6D16$E2R}j(pH1J6n1oZP94|d<Dt>a%fCHLHx<3p+9lWPayoVFR}<q-%wP4_cM;rIyk84<73!?G>wQ?$&^~@Bn+7<XtX=E%R6vbQ|3-E`-JHc*BZ%U#yi1OvQ&DNV^KmYaP&k}_(m1P+*06z_CYiU+Q@e{TQL6!S>K!JA@R+g(fjDc5e+~LOswjobw3$__&SpKNviUt=j(+oFXA6@8WXTVW2m3_ph6JWc{AC{OPlow(O531RaStmHyZV2hQZuKAPSnbb=kY7`JZ@HRw-E5HN29J;TRC{a)zaCqt3Fuz#piZ$K~iH-J5bkVN$B~TkB%qgyljx&^%Z$#)BY00#X&n^9`kAYxH8()duo*!eC|RpLSO7Hx(qvhL-_p;jE?_~_@J5mrVC_y4GcT*UG;lIYA1Sy^WsPl_%$jLv{FjDS+|`<&=IRv--K~?lG<3#w^E%QudgB@9GT=hSdK@xPNhma0t77=Y1HdX8pn=2s-)-awz$&eEUrT^aEN^FZ-JJK&)fhi0%|i%(0Ow=7PP*J{ZT6mo}eFit-t!}*jt9x%A{#;#_n@ph&93}+DBs!A}hC>5A;A>$IXd$*do>6`p=o89<(1ti#_hGy0uEup@X}_EHq@|HCTD7UXZ`(`Dr37zDRFN(w$SnZ{W0U(i!<8Dfc|(7Zj16uzi2bnz!q``Sz-Zfk6=)_K6`bt_ZAkW|nb>TY^HR_T{x=`vsKANjRRncN4ky8&@L9g6loY>-X2_PHw59g1N`^?w;Sk#g*PS`Mq-Sd3VS)+Q((1R|CO&xAxov@^-gNp(8~?hla6WdGoxo=q10fPv|T5jIfE(;JezKYHyln&5E*Vjc1+e@SQlU%uZqQcYQS{sh=ZY?Zwx3cCI(nNwZ(T5B=Nuit%~zneu(NO%KdUKV7V(%D#HKjfo4o;U^6AGK(QoTJB;vRQ+?ynT2m<(AhVy8!C3s>p<cf<Bi{3<Zl>z%Z?B(hQ&?~uk3HrK;04Cu|Mb8ib(t0y5rwuULO^W7J_mpy~UKF`S#8J;H;b7N@vbCR;ky8*5&k)oEQ8tH+QT0l(j~!qJ6F2xnh3vI!%aGZ#VfnRl`y4C0}dCXIk1?lyzZ5GTnGQx(BQauxWE9T+Avxt>Qwu{2=^<>x;~HSpi#FGg^3fv>b+}+&=^@I-*k4V}&*kP_5~E4~lzfHh{?AuwLe;8^%j!>RrV>zY*|@Yfp0I@M%9_<Dy^hOZj&GYJ)sI{k&&e7F+F=`v}ca{Vj*xz)$r7fiE8QW1D?c0Y_}j;`UQUj|o<LcglvsYeU<EnpbIhu{uxqHf$uXV?&9NH#)Y?-EeSN&$?900BUcKI?l7s?-=|cjmJsWFR<m%?N)Y7<9VkK`J%<;KqcqD#Nc)u*ISjy+9tp4YdB}x5B>p+G3C#v5scl!^~1hp%OxkZTAVIap6C^H8^J5hH@K~IW6)_#n|JuQ0VdO!j{4na2QB0Q?k@UhlS$1>9lcG$EYsM<_fVd<^ACEb#_y-)4Q^lGln)<zYzv>-HW7=ZN%)S+9<>45>zmE^P?Aq<KwWElbo$EOzM6GQ?uBn}RTS#5Ks6t3^n8yNJ(zjqD}R2Q>7UEQ`=Q=$BanNu%#IDoOZ)l*STktl3251Duoa=jc7%^gv3eOQ0huf`<$apcyP8W*v+)y76t35Fh2peWHufz*Y;L*Z4Q!h&R#+6h@!2n+vKx>Mv^wQx!|vq$V@8+pcI^~|_P5IHWCRYHb#gp*M(v<DfK6iD)GHNLXfYwW%%+YC<R0vg!_^d1HasETD5U0HJ*Q|rIMm6^aXM=f-@7$xGt<318+`|_wOp6)hBAPkFSU0T6vLi9bSc>1@-qXsyB}@nVq^W1XJ?=u=e|QN1av4*>{+xX{o9UaEA@}lsuQE}-=2;Ri7E;+(kUKz?5vi!ot_sj{WUVqu{0mQ6m_w30SwwUSPn@f=JttBSEc6Swg9S(+sijJ<tvV8Av$?K5N`9me_U`O4((boon4GMvEbw)s85!}(le2o&*3epIR_6arni2t$q3qmZf8laSgCr9mQq%2P_8-4AegB19j;~T7IM)`YG^L{Nt&sCAYW?lr8=n>>ynStTX8kEKmRj>PF0dC44G5@QgEp6*X#BK8CCk~dX!Z*(FL2;$7VB>D=y8hvV*oEEtrm(*o;4g-j$81^2)&bN+EiWt<dhqm4UQX@Kq@trY9B(J0#=Qq#fIm8uSa8-kB)?RK^jnuGQb%|C=2Pi%FK9L%<Nj6#y$ya#V^sKnWK6c>U=k10Ez<bcU~M)|(G-KK?CQOr{x?fSYU>GIBbdMXe+<xEp@^IG#QC4uBTKwl%`}V6izP!g68NDMRX)Uvu0&q-F0IW*56rK2XHLQfEw_+Ltr_YgXag$eQi8`$~!ZnY+9c%j4;FyctO*jEz1}ikAHhcZI;JdS#K+4|fji<juN}1bW=Mto@qGc|!{Ony;hOKp}LCZGR=<41l8Ny&CQ=h5uiR#0~0W>(Lr(0zR&aLKBgwb$dZA`-i{!u4X!SbKPh~4D0Le<}ybgjr~P{n&C((XNNZ4u?B(gCs%Brmy~Me%(hy8^+8oRaHDZ~Erk2HpAwvb0i$#4c^bl6%S-}w{LW^Ohh6uZU_;CGDbW*v^r(B?HNv9S&A80H^AS*fw=OUDl!sr;qs^((92??h;x-azhA+;IW<+k(1~z1Jk*s~>7SB$sslJ1|e&_cAnv^rM^Fgk$(->&4aSNZElvPz<rOAtWFHmvTu{`UC+IysX^NN<&d$r4#y(Zo%;@IK?u-Rj<3)RLqjXUGJbZ?&8c-oE6yTN?rbvs#ftZfMf*whLQPMC7u9SgVIV<9;q{U7I%KUTgd9SEvr%(2~$SVJ44a56p5EoX(jp^;i~v@{#lt+d@fb;QeFZ;S?|_TjG8pL)7fIH@|CUF|VPJAZ`iL6_g~Ry%-!QaxDs(aqJD4t@{Toa}D8(*s3bd(max{K&^4w9X`Fta|(And(m%VjH;lI18OnYFv;R49bVLST}y!0Qc455a0K=ovO5ys)_{|icm_cUKwt0mHI8`t5)-`QR5m~j~g{jF~Qx3P_Ah87RVYSHG#}Hb#niEP#({|vUKeAIO(T!R%+P0q8iXWx)`hIul~^nt2ZQ`p{ExS<tHRdj}gauwfAmYZ4<9k^E6#j4dHsE*LNhUjHw!Wh_|f0Ncusd&Arz0#3UHBBiHKK{<Yk9_dO|riiu)#aSkd=e+o0z^Tn!dv4zdQ&er?YLmw6Y>rO_(lX883t929<q$@SCf_8U((qliu#88d#G%jVlGhO9CKUra?C+~&pEBO#BBP#!l-vg$(5A+ku%|ChktE>^`z4#3<n(Mye)Po<${4c-9rZfaPE1ZMuM!Nl-|F24ex%AtE&S5#WHjF;^V0oJPF4wpNOonIFtLuMXA=H1ay<l6M--p2_mD&kT`Y;dv$?qK~YTp)8tT$k*`2)i0rvEJ*VzF_%0)FU4EiVwz9=n?`YVgcwb-ck3!Lw;Ul;NUNOF_Fn_YF-REPFWF0D$K`&hHawdOWUocXg<MIZ<QL+H>ba4b&oAXFnLW@9XyS#51-O9<JTRGs1;RfYr<Hw)qIE^lsou>0yclX*TBETes+T2q}QB&Hx5_CwI%5h`9Nlv8uEWjE4|Il@`09u{VDELcyU0o8X~=@UuZ}vWPl2D(O^Re({(rhRFxA@`x+&JF(sxva3HB_F6E_?ULJfNJ_q)rscqj;;bvpR>*A>AgPF^@%?hC-HOSsdM&VQ`6A2wh?v`l{KQJl+Mlwye*Z2lq+lYRpKDpU2&MbTxF!-xw1#u~_d7Wb@}p9F3*d}Y13UCpQK`neUW53=9Gx|LGGjJLg%0dRY1!ral!hC3J>W<>Ypd?{w}v%XIo@wkH?AQNp})H=kH1LBTpn#2#;KJE_B8RJO}VY_9(O+t`qs~@pVh03Ov&Fb`mFAJ_5V3kNyjWULMwc9rC#&mo(b&b$T>MzaCWiswBSyPF?l|HHcvz_l;Mpn=_|Lh@Cyy>q~U58E3B`_{JeF=Z`29eI@RUA5BL&JR#{T3H(V&eQ;^$?+e$jFF{l0vUyheYdC$!Tn`zr-PQ%sslbi?6DGFgTe~}Uyr}5$icAAFQMA3Ql^Rz>>1qF;(>-uy)Sz~-?l4p3+xbb_;`bC}6X^vQi`tz$u47l$bTc0cNDA%Xc<>k&>hbGx!v0GdGZEYy<D-;f2WB(76s^IH$^|yI&zSyh#q3G|{;WJx0cbSH~QbBMsZlJ6-!EpI~wK(SDF?2H9w?(n5-D5|>l!qMmeCmZdp%A?BOu(kaverIb2$Ts^UJSN`{Eh@2o=%GW_!sy3+KN#(!v!N>2iP2Fw@!5J30U}^C*Cl1cn+HL?M{28f=z!E?yo1u!s$>i7UpyxYm3h6q#a*`OHT%U$+XY>RoNz7#?{LTv@(9&G~a#J{N7s+9>(x6n!y5Bndx$}JTfaMU>>bh{-*wIJvGSdABf5&YkXLAK7*6XndW&AYDb{<g0;0~s7^YQ_swX(cL&!w(-3D9ehF%vx&V*#3H}xqs4Bt0KAfF<?ROpZkWt*IOYZ?3ub<6Fk=|7IdF>=7%WTQ&eR6DXZ5M5e&fbqsz_dXlHiVC$zPABt^S{{aRi@{<c%=cQ`Z`F5&WRFNn+5plxR<MQ|GgYh#7~jq^No(ysL>)Zj=Z2!TIslg8%O;q0px3-{;qT+{m7fwifJ64+)3$eqCA}-?XVuEwYk>Vmg;Bx)w=1~n6!126Xzw~yKf!$ut??NNcZC2XLiz&y*}=f{dENuYPdP!>z7P?%<vu2?3`D@`TXT)%LRC+Bw{dHREAk&biC*BIG!dx@t8g*bHArnv#hBd=zEPBVZ+CFteu_-kr(7{0$(nti+DMfjIoDzne;rZE<)@&@}c;8d+2d(W%2E?*jN7F-~i0d*MTYWiK9<&vBmgJ|NiiWQBbXBxBZoCp`(*2YbG@)mKT{;%fRqbb|Ao$<>A~x&&6ikyRxF$`wXc|+{br$yor>DQUwa)rG?991*BGdKb@A9srx{q9UdhF6};MuD_gFgS9|a{LK4)s^QZl2zTWnXR@RY2S~lc%$A>#@H%77v{pK*CuD)gNR#|K)yg+qm|7k(A`hn|d9)*AXFzHEU^M#N5Hl8_;<Zo1t5+NljAI4p)*LDo&i%Dw{X~JSxRXTeAggXt3hq&pK=vJ0rzGup{Og=BC#6cr*BKK!>O1qk=Z;07D!8%BDiblx6*^Rktlni_0*aL1zeL1A9a!k+D_}p1F4^DN==r_tT)?DsF%W{8fzUAV{H_ytEc=B}F@^-*}I2|o<O!y-zK5)+Alc|~Pqw%2@OPK2g6u`f>{oZ7WRxLjy?7qLKg88-Dg{wki!f0{r3fH)YX-my&zxtZr^2=>LONROU582eyALI4q%^nKR*`yXF#Z9APYGkjaOG6B`1=#IeSIeDw8C{e~r1uWjjeN;@rq0*WzOa4P^XLu3?u75}NK<sG2vWV)ybIajiFUq`p8MaYFw^sztdPeXssh9qI(MrxFcF~J(_uH8Z1|Q6Xp}dAT?>qa#_@<FlQ%y{CRy;?F~|pX?|P#xErYV!T{Z@{_Q9SQPC_?7^@_Z(9b*dCHI^C`%6K0@M?hR}dT-tuKmF3|bgawoduqU=T{&o+<Azwpr=wn*x)_;TURWoAzMQq@xZjddMW7|_xh*Mu;Mv=;CZV_@F;Wya7PC5?!Ef^aSzm@<2(;e>ad`AuHsHq!tZ#RVpKFmzc`)$dBflDCC!r9PyQHyQudjBAYu@d0$M<}2l5~3WnMH}4wDutT=>4&G_nKSk_<Bt?e;t{|zP6rG8#&QOE2dH|e!co)+?VbY>~W}^eYrhgXwtF`G0i^BJjOXTh{l`j{Z{8AW%-;BUe%<VPG;#L&RcS8(tx8QW6EQ#@`pX|^9Aqa`klGHKnMEtS6r<G&HM3ueQ+P8$w&3h^9JaX5Q^9GcinI<h<<|F?-6wS9aj9nwyV3ZZ~Us|T*8aW==RYztq1>9Ka}nXcg{f?)Pi7MmA!b<UE!<>UIlAa@l?D#LXV;9RXFbMYN6E*@yFybqHb_9>Go#UORH?<N^b75p|`%%U6-j;U}ZHu(cwKPO2iuN7+c9(rj!6R)jN+1&H5>pn$s;Xe`B7cPs|Kd-pNX8ZcKJ1=w|0#F2?iw@;c;rSQ}U4r3x?o27jn0>#J~6#eC)1jABjklY@`9<mv>c@8;SVX-o&LoQhYdHoJ`;_2ryr0llGM*#G~e)C}<`)!GhIri7<QO&TAbLwSL%BxQdnxM_D64H3)UHP^$GW0~rS)V9aev5MPh^Lkx2%j>?pzQNZnn9Q6eD{o32v?u`l#@r2G4o=JQp*~fmZagt#p4y&`b$)ybL+qOSp0avB%H!s^VJ|zk7>lN@6|cnspCvMapw+E}%Sy)T)|Xw;>{&O-4wh~&0$n#pn6IGjJDD-!&qJ}x{k1Oo;z)(~A5~qBOCh0u!;W{%AWqs)FAh$d`+V8qbfSH0y2_z)Tu|I{Vr^P(<ww0{{36{y4oTfVeZ>%E(JYsdr`DDZDW^}O&SEmXziD(rIYJswqBXs0_sQm-X2m?dxv&61_p+MO*rxFejFC0&t)Cm-R|ht>vCXT_op$+jJz_L&xG3JWcBi-ojOeoUfS<dY#^?0LHVZDirOs;;!kf^*=)p-XHvxB=Qpl{QY0q?+)oxRKS~)%b5Z=`5kAd<1zHjw&%yd7{ACmr%4ZFVt(_MdF`AXT*9t6inZ>v&fix=_v*%8p(|Ixm|4?5{^S3d&g>dA|5H@giUjkYpLapSKCl4>V78~A-rC|mudr9IoCp#WsgctRe3b&ovWWhSm;%LLLh=1<_PaHU?qn*_Lzf7h|jkQ1@xHb<i^KG32sU2;%GV&2OwH(FA9W!0QjFd;UqsZhBbYF2a+n8nq$dk$MPWW_`1#cNu{NFgRN3;p)2yI0+l!P9z-Cmq@rSKeoaZYIzN+X!9b(93h}P_@kcE!b(5E|Z!EhuWCU^H3&4`RZ9{`$6P{jvHdDAur|QYla2xt+85GA`%&^f!D{43o@C5wt$>R{qyuJP*QrjInHCu-sKZUPgZ?NgC}0T0S4@1{Z0n02-q^~J$VS$G_W_D%|GD`-|kQ^<=N}qShw3KN$sWk$@Cicb!Evy53cts94c8wzuMH&MuYLbrq4HRjTx=q{rl4iM?w8XptmFS)ZF*!HWfCrtli2Sj^DGJQ}ENpuF%|d?I$M>A1Q`!$6=Q<BxQSlu>@(a2DSEU-dm3IY>zV6G+oh(Lw!{n5uL|$Z=84Hve#V5!(ZOgVHm+Sz0H2S+Vsy~@#3%P-h-G@YlHlJBV@TBy{S}G8e~^XmV!|?R_h^B9}b6<ce^1G>uWHGEkXX-n{OED&EJDiEH3qlFdx))hS)*tSrR;Tc2V!@OQgD(d|TVkwU6Dn7{9gSaIw)XW?8&XXI*7g?jUK6rFNM3D<xo^%1BD__8ARKeouF~R3{aWPNbIc(j~RxP3__HH`+}fGYc!D5UbPs*Z$e%vUF9UhH~rfR>Hh;6gK%fGK}C@1?Zb*{)rN}7H~%Jhejqs^jRPg0K0#sUVI4d4Q>>M{uo9<W;W29U$c{Bl*KREA;^LoX_4adZ?QBeyKq8Z9tul$CCWg3+7%I@vOkm#ik@HjdafhP73mLmL`^@HrZp7ri&&gUQ}3<bscVg>aeIVqCc~wTV%_yGpLLKOA%sqM<zMo;MfocmoPtE{G1<+j1-7>2&c26B8l5j2FEGH1wK`w+tKX@d?X-MXiGQN#*I$T&auWrO_3fYxi@76-TYg=cWub?)q4cvk*B>s`ft?U+OsL6mOPTrgm!oF@9PERN#m%DS;j%l=I?^V!tcLZ9Lt4I_7codCv+x(JSFtHn-!A^Zj)iD=!}j&_gsLYWHkG~IBemzNzN`r#`IqtK-aEi*W6J=I=U3_N^VxHRHM^g8U!2Ls8n2zeVTH5abLPriHoa(dkn2BcXad=#*l$~MXGGuLs}_6-zcZ(f%m&EdUW?j4XXrd^_xT>QhUAQ@4sUu7#jj!G($^t+QsG;3>_{zAdI{WX@sI-+-@4oyX&npcx%YxTK8fn{0#&_gMju?RZ6EJ}>h%xn|Dl=#bFFRV>Dq(myV-%s&))0Z9n$>l!u>FeF(l|)y9V4m`yuZRb#g1Tqtvg}fEK=;&e<5AI-A)QzGk;-Yd!pfu~pieO<rMo!v4(Z%P^6SVJ+@)eRj?gJH7wmz?GyBu4747Le^ep{{gmX4ky09>8SGxO2sLZWT$<5zjjqwCGw+AsD%i{k94`7s88aIJFkUU4t%B-Rc}SD+C{4X2YJulLtK4YEPL40TS-LuJTcHiw|pi}!-o$)BW{o(CJtxQ^LT{+OuNzOXEwK)4}_i}855e0iC}qdn&>{%++(E9GMBoJY-0)#z1T2Eht|bWR)%A){mV@vpahfS{)GYGBUKL}`dLj`X+}h+8nFKJ1M*#a*;QiMHje*~Tf?Q0ea+i~-LuJlWbLZPaBzGR>~9TN0Us^88zX<JU&@u=V?u2NJASRT^GhA5PTp&5N2imp{&`KhWJSK!bPF&bJSDY5akH%g0UpABb-gK5@HyJ9uJg4S1OD@P(I<yyQ*ED6NE|o3@cN@qXR|aHnaXx%)E1GzE*(|;9ZlN>T#geDQHA#{Kqtk%#fE_ulIjq-Y})e5t-Hs2DnyMCK6E=&$JlMp<4}Ci<oQJfjexnDo0`HO)pRm{XfvG;Ec%nco!n)P1G9C@_l+^Vfd*Hm359=CrIv-KW;f)R;BS6wXbg!Epo{HuFSO40c<mrgkFPQhIT~9lyR876R98QcJ?`&uO{1to(NI4%xnP}=M*-Aqjf2}_?&t7@5TXNQlg;AYHb(TBaz_mr_pO)u`36VjY)_5su(?!PRf|yC-M;)HF2kYQ{&a&`vDq7py~f2br4p%eke<CUwN2>$)p@;E6Nj9}UUY1f@oxC;RPQgvnecGc+;bia4wYiaIn^6h>lI08g4hZJtZRrx%QbvexR64m=lj#zUY-LZ#=A_!O=*pMLVkQ5L|eOSWx%g0&@y+p0uyKo9Omtg6IHIa`c#Yf7pWKO`|xEfEmkFn6_8YF;nkyh=&$R@B4le7oB_@rP?%chUAep`6wB3m8ZWnll~Wt6CWStZZ^vW>-#>G_P3$g*m3Icrp;EGnf~~phIIC{`=+D>J`qKHT=sq0|7@b&AD!B=!xA}gwR>^xK@#>tt1lZiJVjkNR?@aypON{*-rCwij;O?tN8#}GGo?ZF1dw8%Lj!pt<EC@&H9JML8yD46__$(HtgVvrhc>E(zs)g+K0`Y!k<ce9pssx_)GDs@{G&D;W+E+Md=R-yy5_a9Yi*C`KJa%WePsSQ^^#jo7`j<-eGql=vX;qFM3d(VFb+{$^X+CnmP_A3jS;4%rBUL~9-4+pm#ZT+C?B?inEvLGXHLPv3x&Ab(hwXSWYl}v{2gOsm;Fnwe_@2Fc>bBS;;w#-Jr!M0NSGB^t%CQye%VCXTR#kgax2J<P`zZZNxV^xQUJu!Pkola*75AGpFJpuL)%(DDd7D6F53B!bsZx7Ljd(7c;qW%BV)Iis;VPIC|2CtMJE(J<u$O!7RZI>Z<^Jg`IR*Wj2+e7B6V?v1VQcfEe_r(uN)aTty}!ywWCIk41Gt=T?;~tCSkBV64}|-B7f*%<{LyJVoklf?w2N3og3<9V1@=A$`?6&q`R+MgTA#(mhHPA6X9(5rE!A@GzC(N)ENXkMkC5BC_M0cmf?Aig)stS{G=&NbT;IeX`y=Xhy;haKJ#MlAcG6|o11DQ;RfD+Itbf>B=<F!vfagvpLob;vo^{f4nlFxPbWG>iRNvCkgP?IP?fzZ`u{&EQQMG@?KQ>f*S<P+v!uUzVxC<T`o#W|x@7*8yQ4%#}nu(OHe>EFlz-AX?>TD&3q`sWaw}F-Q%3(8n0?bWl&Z?~`%bq2YYs-jQ{ISv5IRGb@;w<Q+Z!s`0HjQsyEhfv51DfvkQCqz$oi*Mzpb|yLnk&vjZFgK=o}_S`)OIJntvzA2*;5wH@UOsVtar?Pewoolfkm4of<63ch*YN74<%6segd1=xs9|!+!`m__eV;IP}^|GY2Q3v9}K$A0sWB<&?7I#sO#UvQRnz~6to@!EcKcS{h3uKLPB<(_!t*?>(IPMzgIV=h7<N5P3P8hE*Hk(&tlM!F+-?Kw3QG=X?vgz*(Ig4A$<1t`5(MTP7E{4TEF|cvevpc%h#ZDN06o{B=k-;P&~Z>dp+gOz@@^Ds>5xQIsK@oUUY~d)4v8}W%%l)14_%kW^f=@_7grg8rsaq_E3#DQy0Eg?wq=>Uovw2$euKI3&i3wxp`G?I8y;_YCUwXES)q1+c$r8pA1uS!ZjDq5z;g90n_WOAq0UKS$RkEKq^42G~}-U?>6JmP5rm?I88RssMD+l=<CjPeO7o?(d}xxuw?0!05(%R9!B?X%w6Tcp65^cxpM!tmu;L3)9ALo!gAd93fXTx5~86|)z@iFSCAO6V(mJ`!`e%MjT`^Qd7fccYXx=bwdF28(mhkTSFtz?OFhI(g&VqlvpG5WVqERJ96fFujRC7QKS=WK?=N%`Q5~%hzkXgDD7K(|tnK~Den}@#$KP90^^t<;goJzV9hleRI&d>uWGrz)-lOT`97(jvd@JwZ1?^umK0Inm3SSjK@j`0cp%@Vlr3C3a!%oy6_=1@W##UR;J!PG%b1&Bq)tTE3Yo%^V!fJf36w@lXEIWhto(zBLQd1}JsIfZr(#p<Ii`_<pV{Y3!NFA(N`{hjNHH}b~7hoH&U+eXv6GneH0(BL3vTJ=3M4KOQ9z_}5LJVWd{Q2H<ux{GbF*M@QciAaaHLYzY$_-6k;lXobXg&5WVcuT_zAE?3gZbW!-K~GR3z9Gy3H|}|qZu!yk83!Nti!9>uAr{^U71M#v#kf{TJT#+bQ(;TQ?&tzSg4uAx={T2P7SHzH_+rEBSb%_XxxC3uy(zqlWkw#KK(Ll!tBj1`?8cKfW4V^+-Zz$QJj@wE#{d#@BGkc;6&4#nZ4ET8^@(S3iA$e-BAL|fs`OPv{U$A&0OuV#GLkJs6VPH?OAP)5e@wa4o<J;nh+r?_@n*iwWKT8iYNf};z%V+M{05ABG&k6*CHT%Ml3me6_Xdjq9|>eW|O!(H;WG_L*shw=zOw+d4Py$*FMP|iO+*pQL;kP?|v&*C7Q}nWeW9dU%~dWT#K`OayQ>qp#>}d&SA1oE*aONP#1d94?BG;*)jBlK6ExnQ#O_t)#zi-G6GNAuDXOH+Eg~zQeXdR&+Ie$d`A`X9oI6bxvT=<$=`g6wwiPr^L7`&pp9G``bxrTt#A9hNTcR*4TIeN9s++UYPGv#m$58omk{CcjEiAa6^<OYamW7pq`reysUCnx9kB81`b%`jc6W?Y<gau6vW3OP_bu(Joe|RZ88Cc(;2VJj>-RVmN<&KA<Aht7pQ&Av+&&uYPwvR7EGti0kJqw{1tY)Jzu@{NnAejQIwyZqx^v#l+&L=MNvv1CW}UDcCe6fc{CIi9__{3)*0u8lY7cf7;g60O&DOPR-)MIK@YH>yhZ5{QK4^ViZ`GyDCdV&rv2lQDp1Kv=z8&|iU;g9%EWr)pKE1C081nn-3L5)(@%jN_CB6S1R!7sO9Mr9kwW+hqFEL<(<`k{D7C%@ZLUybG&=5xBPAR{&2`gJTPPT?`b5{}H%iHF6Y6)!K>O6evIPJmR;%HD4u(yjBH_aulCWF79N!u9cRMg&X85eY9Zk@FgwQeWj{LhieBNx7q9yb?aX@>gbSLfZK+!N(Eoa>jj&jk_IE~Bl!qgW|zt<wiG*ZO^Ktu*)A{{67U>Ks~}e|Yl}rH$*QXzT}n#p^iR&6KitK?gm3v)P;u-o^&97XyKfv2#o#rk2n7K{0YcN%TkPL?cZSFMgJL&FnGix8<ESa=0NNcem3`eb*@Vv=rUl5Z5uIeNLAw%`b0`{rJi3&9Rox{5IKU>&ra_*h(`ZWMR-a4awgmctUxPIO3h~w|>-u20fd3CtK}!eYBqoel(4|ex-6~HE(i+Vl(D;;!@$~$MxUR7xQ!P*L<9v;xrmvZ^U`7x<3NIZ?*8yK7HQn!Y6~F8c+q&fbl&0J&m_sO|TP07ig;nj@t5KdILC#90T3%b}!Phs}+&@Cy+^}m~=^`>=sygn2nMn#9Sq-^HMey;JI;)pHmG#G)WT?NX6~pAg|eY?+ax%7>bmwPgr2+M@y}8FQ47Z<Y88eKh2d+)6uNJe$(a5*{@rX82RPtcqT-3LO`9zaM_ve|JZ(JMhS%UwrZ_6hl0)$w6k9(ug)0$m*!z@P^n*jOMTq^3#|v*D#yR2;h;S)dOk+^yK(WfX`fW4SZ92BH|ILvNR;d*pbhd-tt`lFA+igu-+z5;)9Pi%s&g%vgD<>&jXWrjcAXkmytEg@`_)6!uXMKX=Yg-yv~t1*>fO6-oA;4@?EK9W=35a)N)>JkL_Ptu?K<M-loE(FWPWHgs_jkf5F{m^j~DCLX<v$TO8c|y^;pHudkmLv4(m#5W8QfH6Exs%-AB>l%i6KHpF_TQ2KJ(G$8Y&>P$Na$Yaq5C%zNWGukzv08IJ0GcGJOcQ~P9j>oa;7%|-8Z|A`0B>#Dzc<2cVT<Eb$F44Ca*v#RIR=A*}-&f;P+%X@5}Yq4%JEYE3&8n<+=c>{o<(Wg2+F1}NHW)-Ls8kUd#M%n8!Y5Tn@d=CCS3L8fwcQ&@I$*qh3)$%f!DSZm_hE_EI9I0BX_7-*Olq55fDP<Wj>dx&9CUitbll^F=Jlk4jZy9@vbnY*(QJ(8kBDRBVyjWVjZ2wCQEZ8W((0d<uQImvD(Xk9`bj+WuL2@+bTQzJ=HsKvZ&22F7MZ0+<nc9j9Yr_<2$9Cn^)dz1n4+HjyTy-BIen;i=nYU_(>g4S1TJ!d`&xnRnJLeBIDA$SIPQhaqw}#)J-R?NT?)vHNwX@6<Hx7E<`UXsLYU}TdKf8Cqb?3akQN|rDFGIR26-=}>&!6G9Yt{saKF+Fp6rDAS^zvDPJ>p)fOZvGug`U0crt8+|$jfD?ccwc^yk6~^lg-rD-+ne)D3|MW!-l<^-0oZhU_e3LZ6H3f*JoLZk++yNlH%M(Csr^88xP6L;=^}UCFc8=JJuK%8Q^&})}UtFthCP392pe{>qR5u(~wU4{9u>yvcL9@xfyfDsIVtNS3ZZM!*%hPEry@dt{|{{UvVm%-m<8kE1WZ^UNOklQdj!D5%p6)`%LSvuBdwo*A#$QBdB|*!?|Q1qy*IF?*a0(VPq<rmE^dp1j3_G-`BXfew5bNZohArOLIK_j>oP^39H*AnJ4GYz8kf=;c7f`&e->+z2V!MDywdxPR$lk9RvAJza70<qh=yyRK0Ec;DNGmY}tMW)iKf7F-7whkN%=;tor<EXW56<Xn6+f4ZpG6Tcd0md!R^9`=Ajzw$~Vst=5s`qQe0>%qkLy<dYj}h*F_4^KSKCysHeD^2YekeIewff4Wyw55W+Nptj{)8jTn6<l;QjL4N|;b4t|}{9?~Au*YD+&n<Mi+aEr9g*t{D9YTA2tV-CzxYo-xt>5awh30u;)p1bbj`vf8l6!9FPfFW2G3|DncCMA_cI@n#b)`YQKf7v@Pvh4Twup9M=+;c|fIH;B2dD19+}JvPtnUExR}9GV2!#uZwV#O8iT-?Y%c@iCibn|Y{sYTFTrpMud4jE4Nx*y)*NfgUzE`4c@Mk$AuP%Z2;&Ar#e&hYYC+8q5pH43^lXhL*vNTlTDAJw#TVbV;irJ5T;H8tnzd`ER={jmPMtz#z$GyOECy;ff_XU(Ku6GmxoYtTDY-+>bfG)!+UHNjKk;FYBOR*1p++ASdTeER@dC3&*O-+I$K5Y!#Y8GK{P@*o2!bC^c#RQpB<kAC`xVk(YO*uI<<Yxv0?*W-Th51F9FQX!oP<`;&Cb!|#9%3&=df>H5sH7YG-qP=f%e4}TURybU=`;eJBL1yD!v|KLdvydPnt?KR`hQ4`zvar<Xw`wiNciH={8m%0Yp%&}mr}^`H3qntu-91lauvFlBX-mK^&@=vDjwAZE7KWVtSk`s^&6SK<`|>o*uWmEaPWE!$K%RfZU&8eXLLD+olYL@t0w#|y-YKl69;WZ_{moG=4PEbwMA$aE1`dwGu=^ta$wpM7<QBy!CBy42Hr0^V{=otRtM5n;2xAg7m9`}#|Q*g$}xDedre1P@0-UC=cg-O$uDZE2e%DnRh3gTVBq!-x@}!k1O7jBU_a|ZNSun<ImA@pycZI}?VhULeFC@FUMpwUmD}N&Xo|7@qTRVOM9$~KZ&&e>ty^odWWM}LLd&K;lc2QB@JO{d;yHXg&DDd6nhs=*PM`j@GkSV+$JUGXV>53&B@!Hsllu5`OsdVm8TO=VipAaU3Se&Q>gtYaME<1XEoEPg$LjNtdXK=p5uRef@L_eMf9s}ylR-8g@&#&Q;Suw&+Wb@HK8CK{S9rN>-Gv=Uocaf>{+I;`wTbKcXd|K>t!oe~=T)megk!6rDvi+}T#3$u-RUwtbp(t%jf2?nOKi!z&hz9gHs!Q!%sC3|H@7qJbA9mu`L*UKcV{kD{fIo90Z@t?5RT=w<W5`nwDNF~#{E84*er$i*%wmQRqATcZL@jZG$2Q%?1-;b$$KrBm+9tuV-$tbKchd3jxn{uX0!7RuD%yqwKI?#8vfZ`Q=J~1cCmVWuf)}R<S%r>dK6RX%;vRzxOu_Osd8Q#;J$&^u}3=mO7qwI^7L@vD!cN7qp*gREZ>!1{9e`+vDeLV`atiDbDm$)eExuDwih288gpkfe?@X95*yzOHe1CwpC9uX&+pQ<H?v_c0^`jdJXShs%|`fU&6(g+Al!^Hqj>iLba5P_$-%ynjyrqLH{#>@iwMOHuO;6}=arlG{acHhi@Z*cDIK}_KzD_r`AZ+eENS9lIjPH)>SqCQb1|m|Y%yzmS}xzkLX&9^=+jO|D0!*jc~}1pdjMGI%-8Bn?liGuR<r$qm6{8dKa^nv>X6sbtHB@Zxx6a!;nXT?lZHqfwycAwyw6XpUbR1vXU98nyq>}3`KXzh=<S-XQPuqjH??S}`LJu{R!!?JI#-UF`4p618h_Rig^r-P-wC!`CUW@3QIYGOh3$6stNZ!n+<e1IO{WNg><())16mu;nLnP`%b{iTn&O-E{c6xJ-V1u<Wx`rSe>9$!&;BWAp<RTfvjA(9-2s|{(=CJnMADGvSC02f<<Q|<Y5QTtTFO*yuZE%Rvgg8Xs6;Eo(oFmDLV`&j8-~cB1~KvIb~`Fo6J@*a5Uzpo_L9oo%i?Y{TgmV6_W!VZsXo1+M6%rw==bGjb~_ImhwRvpreTJJ6?)tORs{n+Mog-mt!lD85nsSb*4LCz=8X>k;z#F=?g~iP9}o;7S6kjE*5RBCxIwF-PkpuLKa?zY^5!WiGpG6MnC136rVk7DVz&X%+6rL+`qZrXIS#IGd&T}TuTARR>txdS94-B$vi;Vk_{D+F@2cLudi2isaMPPA*Gcb3!@@_{+{h??;#T+dS<`p@wfVZf`}%l3e;{C+Npv7;Inz#OUKxr}SG;bao=eO9adUM!uYll;vJLgz(IRl?7_;@N*|4bdqkf*@Cf|ZPpxajc?YDCs9P#Dm`R!GK5>?bvf3K_kp&h}&#+`Sz2ls)|mHv2D|5IIV)c#Q>-bEO1@=SDRa@!5mP4%~H$@!&+Wop@$ZaPaP#Nu9sn*4&q!1Eni3vJFq4!-O$-$$pKKM&U2zYTKWDc`&fe)fX#=8)1>h}hr7BhH;RKF9d_u9ek|rlUUs;YYaROaE%GkTprQqH!0k@q?MNtBLbVg#b{pdwG;~#I9g6_P(c4^JpKNdjCQxmFgakz2#!DSTmgp5caKi<-GG^b=~pbotWyh`Kmu;7{&la)~;S#I<*sS#Otp>N$C3Pe~hrT8SeT@D*bA_H@{Z|^*NHmaOA^ZhM$w0jc&@T=AOlR)~L$08d7Wr4bKpp$r(kDojD2M)}iNlZ$Leu6jL#lL|FF9*3EDWVwmv2eNXOl&{KwQbOiR?ISFeun;!79n#>@+Jx-zt>@7z*?ht63%VK)FmDkj%tpujXI{U>o6IJ)gXYGP>PL#e34g+ajd`Wef73dKNdlxa(Ah!a}PwwJi${HK?yLryOjOKIiO`0ASd@oW|HAJL?$wnsd4f}YxJ0RB>evjNMD?&?pma8usrN3e4(w!!*NZ;PN03+2z-A56mrq}K>{xe}&9^YQC*(0tHBj$!xu4tt@v>ILsjbP)mXgr$Sc3VuDY#1BUB0Mi(v8~>3Z$)V%r-w*hwcF_Psl#aiKBmQW`D1>BSeJE*85@7|8RFai``Y?`)%O8skK<H`YjZ=fjjK*L6KS&_HoT{(#b5T3UE<gJ*EK86CCT8&S2G4e#f|~3b5NOKW*pheFtr!YgFBl`&0D*chrV|mV9d(6*22Af+8(FN62_H$KO==+!is-!POH5T?GJGo>;}jZU2_3++f1?rNb;gy{lsqAfwfNH0@h#IJ{Lpx<=tOMzh^j!_q8iSXDU+B?<oWc!&?u{x8)<r0}WTIyT7A4SFa0rxD-#FaZxXV%kz2_lM-DSP`_O+l|b!HDmv5&;sD;ip1(7YMQB6$8*rLy8NVhY?OTC$k6}diJK25MB+*^YebJrPX;R(q?bRqw-TGc<omFZ0^%-?{I@jv=cisu2&=<Qk!aZ%B;W$?5a5KF<)!CjF76pSCb1Gd;-+N%R6pG+DYUId|&#{cO_}6Dl$}KH%a`bkWmglqI{EgNxKN+&ji+9hD@lokml}<zBL$>y09@FK=8s7da5Nz-BR(0Lwmi2SWw^Jg+e#SafH<uTb|3wlrHkEv0k<_+hm#gJF4^H2c#kQv^VvLc7c&*;sPw@2ze?Yvj+RVH7AcMwrWpzLDU~=2?cn2F*Wk)Txf^%{6*<)^1_sMRS;!zmBQyUYy`QeLv>yrvrfBW={0Kaj)tB8~SdSy)SeS$PzTD~<>+nyfzt!UC0K#R#=2Y8}IHP^-UjH2C>;NMv<l}<a`7}A4%tJ3$YtHN%O{qZQ~Zg8=P2SKjI{fd6t^Z5m94TS7V6y+wb_tui*G@I+5GUyL|{eB+r_w~w(?j7`?682vGUGo!dz0tkfH)up?mSS1ILhI;0eXqe>=}?6FaUyvMc=mP!%?zTBf#P%9Zt4~|lgs{<Yu7ZUxk+Bar%X#u?rf4dM^?7bsnEu~<7xCx)->PIuCFuJMzwN2B)}<?nw{xq(?n|B%J^4o{`LN-hR=vX=}i=nYPd9;<~A~ZFVmz`!t~~Xugegn1oyWZ!F5a73nOg*XbZFc@H=VD&yT}{9h!gcg&8cmd2nnAC0?)2nEc48$#;H>k(EGC7Pv=_?o_OWH#2%_gYPh;iOAwBZAq2$hQo3mzku~CWA`sNCmcNb8E`NjhZ7`OT-R44?6l?94va`?6lBV=mXp-uylqR!*7lCAzf|+Coq+QnQ!ie!*W9UeV|q_V3n-a`_Arepc)igYAHEp^+Msp6=)FyElD2mX?RBB*7EvC%=Zm?luv-wiL4EiJnPf10Z7<#KLYlk{a@J2e06*v8$tpgW3&3voE!hxD-4Xv?H<oX@e?<-Ya4lB40Xv-zS{3OrK9I(zy^stIcB(IBqP1YOW!K;r;axR;Jx$f=RL%l;s5s3>t61QTE@9Vernw_O#dImR|E$o7dbdP7A{%xot8XwJ3Fo(G@hhtqhe>Tc^ze7GjHNcYU-alkbwzuvyE#Ng?JLomcY6m>S9iaZ)7rY9c>aPtJtm7PJmBt6{-lkP_YB@lx?n;)P@!D|)X}|<(x<)N+SSDk+AV9(wSkG&Ej@FX-V4Y3o_C8aT!*mz+6trYNt5oU%OsuG{?;0wKX2s4Y3Nn#C%F^0j*Bl8yFb@|idD+a*lUa!|H~5!$nx*DM9stPV6-fc2=NBE_5A!ct1IDh-+VI0>VP$Uz9aEgYn8X9>SS`EO6R+lNE<jps!N%){i`tF(f!kRspF-60?jJJHKOv~>l<x$XR?jW&o8;faCyxh*@D4pgZ0yJgn2Gkfc5V!R6^#RgWc>)-ih%ou8ZXIIyb~!F6f7;(i+0QB;<@-ZQgR>(Q)DJ0Nn@u@q_yg+RGP%_J#C5nf!vPH4vdNusEeV%TMY`bDpgF6`bRR7NUbS1A4CFVJ5q2AGX$+TT1-Jf4uu#e9jLQT;G>7r}34eG-QR*Qw`U(K1&>NbGi@F-@5Wf!=}Ao*i^IrPT$v}>>k=mbJ&=iF6;QV-x}P!)1e$0xA47NWNK|%RCnvfzOohv%cMBDYHC*K75yC!#X)~OFuS05V;jf`k^J5CGkTWqksHh_t9a)MAMj8W_k^)N(Z{0FyH=4Z<#z8d2ks_9K)YV_STG=hP3(R`<8i3cy_JOyY7^=(A9}kl3FxS%mWT2DJ@qL~?_6mV#k!fh4rvrzmz{Tyx{c#0vl%_dh;dC1DJmdPDF4j7@^4#Ccl}<Ko6`DX>xH+fRsrv@-}$UD$~_;yjD@lN6JFh5MCJE+Rt-m#N})5WEPxf4j!fmLw|?>Qf^ECg`F;94U<fkgrxi0fH@myq1HE^~YeZZk?Y7C??M>as^N6nP>PG-XI1)Rm{+r!iYHPZ7y02)MS+|Jyd>S#bc6(pn<fGIkuH?~MQ|d--Y<=Bu<Zj!tcg5>K432$S^yd&TQzyfRs0jbItb2~Ktu!_@Yq5HEx##Y=>RsNu0Xn<ZCi;CBPw!_K?7lY2@$nqw5qv7jANqWkH9rCE=AZJ0qwrv|$Z)Xfp{Tq1&1bZUUnVCFTp^2oxMfdMXh)C>)LI@X^*27=(B9q3&r|Leg?;9PiKo=53FI#3*HWTVL2^%k{)^V4J+<0g;3o(Cy5E|3=$u*1LhH_*-y6xx-;?eBt2=3J<(k*z_uYN$dC6`tkA_n3mo{(%?EB>do&Zl)P8~>gv!pX??6}>sU~p+$pC*s$3Z3#8yd$>lg{RHRa|3?23O$eYTQ^#f=grxxJ-sYD+!xd3n%k_hEiVf_1B=^8KeH$MLBCnTdg&c8uYVw(_x2-s4`?1ZC)y+yZMu{y?YerX+GEoB3*|c|&Z(DF_xK-{x0-~x%1A|{o`;c19?n?@g)`1bW?kYD@VC10>5%nSV@^h6!+O_S<IjrTA4k$?qb<PY_A~R8$#4Q5u&dc?Y}ZF~7045H9`d)3jvp$;#bEPK3%R=OUTw5Jog*`u-HT(mSf?WJ%EkBd_~E!{(|a+?-R}K*65DTg)_J&l>b5$+c0(w<HY?-Q6UD8>4)W!wLaoIm*UK)y!MjUtH#?MgfQ?0Vf&>MN8bOf1%!`t_FA|TQ>GHIHL%Mu9$G43C0}kqtbkwww@2f<wN39we|Le4%DG7zIFCS~%?TaR=^Az69^)<$uC;FnI9;7*Yo{{CB&75mM%jS)FMyySqH<RIf1cIl6U8chJ!KCr$GJ}aluLe+ilKa<vROskgy;<8A$z^mfp6BJadfr&IphgVN*!ib-mbZ(rZ7vp@f49#`g#=Gjwb2*^<8P8L+BFktm0<7CEBgQTaiT+cKz!P#%jR+<{-AnCjg8*SUw-VHIs;q%>ePDowJB{YIi%cb8xIPAF=n32{k#V(wFFQaTiBC3-+O2C$Y#fj1s)c!ZfEqkt_;W6J<^GF+O0gAf_9ew7RY1eLevm0#k&emPnGtP-obsXpL0%YIGX#vZDs}EUPc2m24gcSh-#CLNZ+sR%z`(1%czb?{_i#IDU;@e(xK(K32&=g^b<4>wLD^^<EGOfxU4J0Ws(Ee(e!ykT@jGnCpwdIiCR{HPa@*Tyn6^5P-89j;L&zdM<vNx_Fe_1zNo5LIn|PBoJ=QqOwPxRjV`{97*37XIHMi1pIOH)zi-DzVjs*o4TSg$+%o}+7lA=sqcNK}N`76k;A6^;k90+k_F#cLz!^mx2g|#6e_T7ej)r`jt=Vu1^s2+e??G6mgQ>VWEzLD+Y;Z?=Yw4PM-JO9moA4e`0Tc?oj3<psJb%pQ{5^x}Tz9BkuHc6D-cR8^f8;n6Ax1v9^`;S`gR-2IIZlR;hUN(Gvj(1?hj_96*O}c?``2hZ)u0{`D!3;%upYiR+3M8KyyiffOxGtFqsR`XGY6`LlHW~Jm&!4@W`y6J97qd$zmjC1Qyod}*pVn%F~Q&MN%?r{TlErgyELihV;N`D%d=;!p3q+jKfar)0@k|?3IN|^hC?HMCuG=nKrPL=k*Ok1xPKMCZNO_T&vw&lFd@D7NU_#Ppqt2W_&U^>pXAfCSH%NczwM;t@@U(gPysbDF$(P807Kq+dO!Pf^OXQbMGBOm*fe<i#%!vLGNXbDCVZtPTCYBK9wD3bz4nMFyPkSHl64@0E1fYf9vJc|v_vRvS=0TUjyv~BSr)Hh@3<}a`<)=Cyh6y+D>97NV~U|_V1BpxQQ-`ruKllr<19<1jzHMc8#)3OyHzqE>7ZK|;01Um8>K@t+e-Hipv23#R-2KC<Y*Dn-$%+#s=K)PB<0yW(y*)Zy_~ozsc9H_io-PMqp$SP!E$B#ei+Vhv_=;3ag`2C-P$ps2AeafJv?1OkC?X1aRWTRSK_F3ZK}X_vb1*WXxF^~RiNy%p!@rRa4^dzL%lTzPX56@t#stdHIY1d57Wg)>PK7F4fJe`c3t&VQ*6@Yy8@!z&!+>r)kr0FvvIPs5bHsYp8LzEK^Q%=`LoSsFX-3M*GZA~rup5srl`DJZ)&!Tb$3@p5UN-A^w>T#!^MVmmg@_nhgN0v)9@!T+oHd{(gj5b9h($*LI~Et%QkwRv<Pnd$W$vvKLH8Rk4WEJvDQt{)*pV;9g?aNJXA3IlXQ5{Y_3{tHK{)QUXdCbc>>;RrCeQZzVZGt_sSZM=Vb~(i*9o~;E*>e&1<mzxNOGmXq$}SWWM0PlPap!*x!w1hgJ;~)Ml<o+G~p<;~KrVV&~;x`pe!8;4l5rM!V~d+No-j<cC7qur(XCcK3@Cy<N-z#<kh%mWkfGro3y&Gc?PG*4!~~^PEfotiSyB_FMAWblLL)+f+q<>jt3?2y^}?54qJI4H}hUS5XhIcVp1@dhmHD9D!=}qItm7@1#(@I#^@W<!r8!hXM(6YCTzVS&;PkSrE<4i(Z1OHN#u19rW7Ar`^k^NBdt-ZqEQW-_GyX7liR}QTfXc^m!O=Z<ezfE<oa`g=FoGIOe<xTm2V)KNUS#v$Cp1jj5${VC%)7P<(}{WJpOhf`^cSMJGw)T^5UFqe9cHy{lmlr+=%6GSk0h)$_D^X|lF;ybW->0~KOA!qBlllQ!<{B+s-8w#@rWrq@qIw{&kVKmoifN?=+W<>8~(%%Va6;nes=dwICFZ0d`3>;X0&_v3v7(wgJEKYLiydT%HJ1UwF~=rJJhYoP!D6c)7(vvTD^z_1i-JqV&tXTUiy(?w@=Q=ZxAlfme8-|<)Tw8W~HLS>iw>$kMG=rz};J}-dJuMcj<jeRVS9ONvvUTQ;WxSPb5A!Wv$8W0H2JG?lUd%tX0VPYdjeiFT~tyaJ&{<z&5znzNV))OQ>r!6`?-nGWKzNNa?>S)I|B+El@^)9%Z1{W!8&aQ5{#<YI0TV>~UZ5>|ylR)bPRV1zb2pqDpP6@HNyP0n!{tSii(jVWx_@LfCw%Z3yitI#T)G^yXKITcBHLqVCcue!+dUf|b11&vcsN-(?{2d0jYjS{Eu?~o*v%HDm?P74XI+aUPX~|HhsEiJEb=UTp$%i;?j+^CLs?$FVuBh{q^Z=bz-KjqDL2zTY<}PL)X}DQB&~z#Y2w(Y~>LtF8JE;afF6?c0p4Q;b@(DJj>GxJz-Gb%b<yExJr`(=iwsJmL(yf&Qb%XazlZ_Shn#^~LuYdb~CH%&>?G-^jPQvs+>VrDe{>(ou=Ves-#TltM6~gx@Jaoj>a-DzwP`@Lga^v-e=9AZ-uZ@%mX1&($CbHtyYZJ#Hpd1;;#;3oKTPVXQ#3nnN&Na$!{zPu>)oW|CfO^p!^T^?9ViHFAf%X&e05=Cu?E>ek>7>?jmzx=-i=h1&$?^MQ0^a;6uCC+Ll>1)oizZWTrCYNPM(wRR@y4P~P5iLnNG)zR4tfyGygG3#+R_nkKRI(DYKR%oKr*5$_ttU6&MN)GGPShcclvGKl3;#D`ztOce=r5<t^E0YI9qIbe5|Y<GTwJQG->nq*&GRk*)6HV^Z~C1Fx`V|`0mMy(3rYx&n+uI-c|bcCyL$5lxXbP-Rd!{o@w}5>N{p?)QbT_cpRd;W0+aw<HNOFo|I2z{`l@LL*|uVI$wK?uMerkwwSfxglj+M{Dr2D^3LVhxc9NkU0mGn^WLb7dVn}@c^6=iz}a%szXJ11ewyBz+8-wTo<NlK5%xsrXbW7d-=U~}9}MeFZn(#_xwv{u&Q&5MS82Z(!}H-nc>zbn1+S}RR_Qayxn@L)xWVG<s#Ovz*u`Hz=BHIok9Mx|rig{-+a%w`B{@LIdz;vM{QUNZM{Tn_&0=szV$o!FIGKyZ%`g^9Y33$q@GQRBD{(oz+o!>;%Xg0JMZT=B-mgk4G0RDr-}*e}u}s~a8lPozK{6I`7uJkCKs;NVL_zj~ZD?4%^2>ACH{~hDMl^?2yDKY}C3}G43Q@G(L%DIX?$hTwpSQa!ll83b`nmknVj9W4C*UFaF6=0XtCH(@Ni?nfkrk)bUh&2Efj)O9(oPF2e7w9LW4<w2+k-AO5_EMQjJ4BeQsVc`1Flj?@214S>;$0r6udM7t^xeLJ-=SueK;n^!1mO|Py$?NyE~1{D!)dsy+TE2uWOCk>{Fvz$-De2fG@&4TP}5vOV*eU^IMT|dx|;Dpi<mz;OAz229|rzo0wdsmAyubdFCT>L;dva3?aPk<=bGc(mGcw-3C5~HTO<2eS3oRPtEz|RP9^9TKMj7M%CI(^TGU)Wso~qjhdfq5jf^)_epL{H%nL7V3TGI`$nI*a6P~Lt>#sn))&WKj~$&)uhr;o_2u3KKLL!7l6>174ujEZyf7)+z*1HC`&B1EdjM#;_gOrRWz8;0+}Hz$+5oZh1aN>{^_}#=sB-y+pV5Bq@;Mzve$$xR3$Amlx$dA7zKoX2tLC+NrYe8j5^f(s=>Nvo4V0<z@rxfsVI#j>=jgjJ%FE#fKvX&kFYA@NFRs!3Lz=$amBVpp7;O!<6PxtQJ3D)}Kf~Z9S%(2M4!j8edgQ`wpXz%c0WSh~bGX$Y-n0YKDVxkHsqOi)*W8x$LM~r9*S~V7fQrpSYdQ?`ax>9=qvC#yaZHbw$3{Qnk~W@nXK&^ZE}OTrw+iDaF@A2qsAhwx1+_c2{Q>bI8ljyIx#xdX&a{Q+6nUCsP<-Jd3R?AuCY44Ibv^?{_Ct<g+4Mhq-Z|_)ROfL2sF#|*08_^g+ufTqWdW1o-n|viV!nXi?-BSM53(2G_Z;yu0D`g!(uc|XzHFUHHsht_qYdA6mV6!#k5hUac_RuF+odc{@v^(xm7>9L5BhCENeKsZSrQodG|5^8c?B0-nN8~WIh)=_6yoyZdFgG15ALgkP{4PzyB1v@bFrz`S0=vfbO*7IYg=hkONP+zu{_2O9$)F@#UCY-8EQ}S5PLa{aW*}E+!p#PXxw!)sa1wJUO6}JWfq9OZ5>{l&fD?lfDl&KgKZD0(lFv&5#@{uX^$wg+34J&rIK71jedPn%O_}iPTU`44{Qh3EjF9Hi2VzhCvx-mMvY3$jDeRr;1@HbB4yOh9thOXqT(01xm{)UTN+Ik#J!=pM&;sq*w$080^dNMy)*~uB7C%(J-#ddiHOK-r9*m0%`Jz9cE7XepTyf^UidEBD;M2eeru2%4wr#O8Y!?kubZQsC;FKaetRnyH?RSZz%06AzEs?=A7|~TjJ^$z1!>e-(N;3)E%Ai#QH+_|jH0u5pc}NP%zUVCW%q8pmUrfq!hd6ZEmoY(p6f?SOnY_`isxQ>+9EBTUYE<dxV$s~<uj^Bed65u%L>b@HBahb+ATNDJuFSf!K&l``_GQ2R(L1nMxmH+&I7pA^BYI_AvL-@QB=ZDeZ?(e>?`k~JX7x7>~vh{{pA55h64L=u9byxKi}kgzw)X)$8f?IT0%8oaC1Q(g#u~kUvRlI&9H4=Gj*lzUk|$6ZAPfUQo$}xHs=u>#u?jxRcC%Q_R;Q4CGH7j{VI=JPWf0au`Bv&47V6VOJjdBZ1GO8`3O{sXG?Tg%QqQXgRZ`ME}ok_qz2%WUesMvIBY7tBiWKws5nwvRZ@=MdxjSqVq+IjnLt<#0u>wYGg5U$q4l{AZ-UGG_bQvcmwOAYrh)I4*L0*mXE15Cs>NK7mXBvuYSf0uOMl;bYjdLYWhdIzSnt>%k1X8ukDr|49`8R4%plMpMK*90m-v~=JiqrHy@3yue|JLo&Om?1BXe_(^h2$~ulo1)xmy5EF*Lpk&r?UIHdo1XPP%OY7}wKj|4gal;qD8%jzazWJd$T4`2Fn?B@&gMD6PynULQI=Scmv^F<d*7e&V*rM_Egg1#}NMs{C$S%y?ufj^RFwj=Rv+<uAjLU|OA0yAj_wM})~b%h`#36~lP<n(hs^8#~{tqr+_-{?;#tHYse5qr<$tFB)HO*EB}JY`P#?#fKBn5p1;BG1!wiLAer08s49;+X~sICILj&;FPL$WWCwFa7dIIarMVyYYAv;fzp`vOOIaEL=gL`>p!;m(7#RHnbkz-@?!5#*Ezp(9R1~!nOpK~F*2Hpy}fwlDVD^;@X%D7^C$7S{KEXFsOI8u!AjxHx0Qmo=X7m!F6Mvmq|LsXa=In9ci&!)%`aP=XO&WIu@?(-`09N+!vO?j=O0UL!@%Q|!&2*#x~K};w@XcN_iv7XdUJ&o|Ejs%o#)Sb2%a^Z{ZT18zc-leDO_9k)XW4FGA)_YX9%|VwVx944&IC0uP=AgU&-kG7tpcy{m<ysRN05@z42GUga<yal&=b+H6L|xA-y`QEni65jhc>taJ3l-6ONO1-f+h72sz0S>GLH{?FX45tMkENurW+G+$lGxG`8U&6roX#My8)jagvK2lJ(%Y(++_o{kJtgaVbdRF7TY_;_SFG4f^}(^#pIal3l^3RuK-~EIYmDyOW;~WN}76MTk{=VVZ@**W|GHaLb%G9i{75qEXX&iLy(_yHn2|T$|I$w?1Hmj#BGPx#j`4yB@8-EEt2P;ceA)tx>nP8ck9H>qOn!^*M$1+uP|f*Teo(4_Y=8S?&FglyYvS#`0t>tCEJVE3f>ohCvPglyEQIFeQ?<S}|PQeyY~2YxTN%_(H=nm<>PFY%)ZJfwre_)>yL>`KfNud0jM+^JW+rR#lFOiH#7oYaDB}hP7np0<_Rv@;Q}z!(uZ74_BRHtouG|N!yv;$7Z4jBfDBI)nHj00C_gP{1vLSHeNne5Opdi(0DS*j0IDzIsL*;u@>K-fxWBDl{g6KVTHYlk2A!f#<)m&Zliu@Ey3W$u0Gi=!_n1Umm@DJyV&ydmX37M`b%)Gce#1uyLbd*G&Qo%*6CuSwoBhq@i3d4YAWtLIt_&9-x#>{@B(H7tMPR3ZNGC=nIBhwj7SN<1q9yi;H;=siMIYY(&zG-G3j5*cBH*(BCE%q<Ic9_3~w*O%BgFO>arMmC&dHOAJ;6fDXJ=Dz~1`9^h#?bnNR51xDn8!7Ch6ZqKnxv8-U&Mu`+8Hq7m%+*kmZkb~vuslc}1M&}dPkCdcFXh$7(zzOaI}_u}&cN~#Px7&5f6eEAhOUj<F2`WFuabepE^{IYzaNp!m4-vn7^o1P^W?NdI^%^MELvQgJBw<;4}A@~l$b1`Z^&wGAb$+xER+-KMI`FiZ`_m>jKgCP)$LUDaJhOPTKA1}S#wCn2S>Gw#?hr2A+$m?7z4$iYx2F(obmz|w|LW|->s%7~3EH>I|@uS?bM1M)2qPG(@zH0iTD_sHf4hDO8`|At+J;+8MBWmy0>yB7?41M<tDH~m~o@dLDhIv#t-A9o%Q>kIU@LNj|C*^>6j(p(!6xEkhxu%mPlL2=NoLno7?reC+UbbV>K7(__dql4H9T5-V@Nas2$q7=4<NYNwE>~{ERIB}QSSZsCUUeITW;PM@A&ApJC87w^KHCQRrs4a}Ox}nqu7gf2_i_9-k9teBu?IEh<L9rng5_DEj-ERT8Pg5x-{qV^t){OtZ(|sp4y1O@LpeGx<G1met!Yet4r+e=3AvrX0Cl&Hpz)@BVqV$|7drLoc$0vvQ5mYGr>K89em=_^?Jur=xBITwsw3oPZGG7Bba)6}q;fy){5#&}w0z88^!Iay<2#1BfCq%yoMC3O8OHloc4g{uuUCbJ;Y+Tx2t80{z#=d1aN|0)z2B>~+`zZx6L^0lel~e82p)2X$8cvTmGukk$3q7?kJBTgGLwjem*dvUxXz;va=;cdBUlN;jZt}sy{%9l+Z#_Hv{TUZuM=&f!KTE_kY@Mwn%3X5aWf)FMcV!z+JYSZ8Rt!zd2Gx0I^9|28$x4v#!TY<Cg7TstWMUfr!@!>Zn&o&^k@&|&u|TVP7_O3F^^O}S&XlzTko;E?_2fZpR~p6+nC+qSAW*>8pVjRFR-|$X={EQKzd%e@G$!CDV|%O^r27l1#fFBTweY<ZJEw~E!2)<d|-JVO(7xXCwF+9tIk%N$*^iw8w;R!HI^09m~65pjcPjxWl7)-5Lfduf={1nay&sm`-V@+wb1^;=bavR3`k)jn~i7Jvo&j8>!Y}%IJ3Gy)jPnHX2#TaCfLVU3u-h=G{D_PgV*u<U~CRT!5B?n-FXD2?4f<f_vcaPVZOHgh<DHokI^DIW%*6HRQFt45!Md5^2C0(VS_D^tX}~-`GB{Ybs>I}ec_Yei!$WceTq}t(3_Ee45sApv)a9V^91uQ*T$+;d(fx>Y*>+PXd`t<A)f}b?^YlHXID>lE(7%vOVh!2fPS6%cm2c0Xve=ESJy>Lqx5xEsHLs-J-JSs{l_tp7rFYjvW9TzY;KY(f6J{S3Ukiw|E<0$$70=gcnRbMbzo8&-tBJvcJxBO_%h@64Nl0%r#<90*rD-<&YkOWM{V~0#d)oz&mZ7!;dAYIHn#`#m5a^3bZqrQ0)XCNi>>m_p?8?um!Km;5=z`hb-MkCF@8t2U&2az3bo<hV`{ttIe$#g)D!??PCU91u|C}9cHp+zyKL+5@YoS`;z458;>01*e9-I-=ZjW{nINO7gJUnltUjNTVD$D8x3sqSWe-)^yNc6oYDK@S*>W35HL1IM%=VhRSSY`j-UWmsV>C8GzMD}Ip_}3`pI`O|n(pA=ng7v7#jEWPwblyB7M~yYdC+NNIPmuC!}$E;;fA#4QU&R3NS(%Uu!p$t9DvyuCCs{8=8wJ1@5Aeg%>*%9jk7DZ-&(WwT>C@TS>dCon~0yclFf!{dr}*9*FASyM8!jXxZnz>q8oSaQurdn`VacKTrr2d)mT0@MYSAmov6;5J^QUU*ecpxNq@YO9-sZ))d%a?swD?ChggmSO+z>9!|pdS_Zkd_^wNVCR;FGCU-dtV&b9YkI1a<F#fl}fMRJ&kbkK<;hdiLtSw$%#y!QX^|Fq2(-_Pf|@5@Y8+_a@}g*eYnkDjo4KbG_RH!DkDtkn3yzFgxWY^OH$KF?psdS-I9kyXw3^%L-fu(w#xfUnSeg7u11rkPd{ho!vv&fZ7%bj|CHuQ}Ue+3@iKdJt(wXaN_1#4mw$du=+LvEA3IV($9@<R>e-d%(QEr5Qcy*IT$!tl@am6Tg}(0l!M_P|p)<IDoxt?is3o{5XnF$zZQ?4$H<p)h||HiF2d3IRcwn@A)9T)}#;krH>iHVpe-UYE;yc*Wj}6pYk)tjaWGDBE9dZx?8syb%v@32)bCR27HH=7D?BC&&Q!ZcgN?;IJYFWd(C8|ppfzy@4Pz}^<tozuWw69nDCet2-vgxX3%?8Gu5CK!QQW4#)=u*t?ljbsjZvk-Xz(5dX@WOwa1s&Unaa4u5!8L+jET^xB?iTWsMa0=Nd>yvvHws#EzNOhaJ-!3-;kD1hY@0q_3w*Jp{J(3c9R|+$ZlRDgu)Gy(R>=z@(S2Y&ZJ5PTX$TTyu8U*}!(!$uhtN8qtTl@7HJVw~qovNqo6`BOiZua+j}!%g4$1(J((m&3;yn)ORWgxO`|JE4KX+eHk}@%i+%G(x)EB9RZ7cJ=~4e-%U-|LE<BM2Y}n9fZiV7JT79Xzb6yDKoFt3rZx@#S*^E=O~n!seCL25+Y>H2I)jEIhg9Dy;4q_-o5TTo^|}MQ#@i2-(|a}Y@coms6d^lac6Hbf%B>*{&`p6Bd;iP)>hSZqO^VM)DD<!GS5xIUZnlKi?mE18!~S|xJxw$J1tW4-K{(;|T-BNA(5jSOBprYErQCB$MFwWQ93P@^3w?8mxk?tf&-U6BA8xkSH>y)q8<oZV6s$D?KEC5LL(t#V&NO1Mqslk6SRC=r?Irk#A-15YbH7{t%JkiBW<Q(qh0SVt9`;twD27Dw)Zbiq<H^U~$Hl7WtJjYrt#}O{QvIm9VV}8YoF25AFL~t89@(wQFL0h;IgO89QD0BzD`4sOoUJowOM+;9bZ(==GK$v5tXaE-wF%?6rFI~`s(736vj-A4cB{VspsfK;t=H1v-b_QGN&06;bpvHtRfpK^E}1v&ih*piM!hx*ik5}Oba6=94WO=+70x0MW%mwGEE!#|gJ;ja2IWSOx|bdGyEc~9i9x3$6;+eLY;+3WZ>=vA+ay!FZ9DCyl{Y_qZm4|+3QqejTYz@8yj6$iOWj!<TJzr5-LSW)dziIPfX1?5Lj{4oH^pDT0J{U)g{*DsH{?C6jNTAPT31iZka))abxvF=nq!i?!f7s|WOJZ@gRt&@tNY&6ke!|&_Mz(R`+^31u?>OX)tGls<JB#e)_ud_&mpmE({MKEG4&kU0>H$MHTt;?f9Ln1n~iV$yv?DpN{(f#8I5*@xF4^(=*C_R?uzKKIBRcWtGdzj;5=Pw0-_erFx^$}gHUGo6(pWa`7Sj+ezzmDHF_>3Wxv}uE7>=jEX#Ou+tlsLrZ$ilI^At27KGN{tSoi^`s|svE||*j=%pUjnw$u7!0HA=;Mae(<=X=<ORGf+(7be4uy>)d>GF0y!L?g}BFPkK*Xk76nCz(w>LkdkS2XLvUeorUt|po*sN<Y2TC;$ayavf59xIXptbV?_1V=SAmf(2Cy`cSwMB~E-#1OlYU{!M7uDhi4j6Ak#)PM%tY)l~%*jhl42r89+rxJnpxz(f`1IWj8@Vj_=S6-u|H2PN6d2@>so70sG{XYJ>W}V+r8tyu5$H>&f0b1TCBxY{bUyJ*s<EF%Tf4Ea0l#h77X@i@CanRfI(%j3>R@hs${tPP(mx+El2K*lp8|!c}EACYRuJ~&Cnbxa^59*y#6kH8s?f4LP(W~A+?>nyqh04(m!s7Dz_I+o73j{o?0vkQoS4Q=Dtu>0xqjUJh!STzdl&#F?yWx~pz2N|Fjo)C}t+N)?`2QoJcctZ}Cl{EmF`ucewgYm8QI?i2?A1DBt7@-)_$+3Vk+Ii@1NVuLQpRqbPaW@v{-JiJ+H|``81g6U9%ZsN*jnw0X*7^re$W5#QTz1fkXoxWiON=PT-jP%sA#+D{CYh%bDAG|Pl1;HB<G5?CgOQgQHf&YftB;D23{>rh&S=ngWJa!*U3L^S`~#J6=pzN%y7d!c<Kg=>G=<OexV5???-WRQ$Nowa{zkV{&33b;OvtN7Y=K-{k4Y~*LZduw1L({a;dZ%CA)Gxv1Oj_t^?H%-n%XXCz|`LFT4sw8ktv8gjmmeFg9J|g?Mz<vTg|-`27)%TD;#$*y;kn8k}&l5%h}lkv^B)PUyE!lc4*A(`eG3e^&Q#J=*CNscIjPdp$=-^px47X=`dr4=isSmS}r*3EJC%By@Z8D8bg)DOcX@(HasR%ngc?@Md$Yy1P*Q)%|z={9QEf_dnnmGjz}&8pL*Z?Zo19v}4RTvg*Zpa|TUb?339L2V3?#7)VQfzPh?*(Ks?ic<2<;>WekH8!Hl4)6#XltB!Z}<1SmTGwka!3tO!IVJ3uio>giSMcQ27^o<q>7WFQJgD@BW0yBQkdQXeIdDq=oL5GSm_IL{YsjX!>kYU=ngzO$fV|cH$W%uRFjhiD&=lm2Z*$@GqmD#zXt}y>kT}GL#EIaGT?e`hD9?r?#S>m}1ksq7e^_8>syVmr5Y2IRR=6tWtaMuQszo~He)W6)EnEirJuKvIm%OP-Xe5B0*8|?r+1i$+J$7OY&8}0ht54FGv(U4-J*NgO*IqkTE<MFIwV@x(3u#OO`YOm4ptUu@z+@k9a&vvugkT%I;jiIxIn5#<8%D0qjhPnRIa#`Av!>NuU1yiZ9s>F}&0k0r*4~DOGbVS%_Gbc+zkLlrz`7`iszeL#8<2ik8U+qV|lU<VGvdeaqZi?dYEjzuK)3ipNcp4h!elJS(vAb(jfHiUTb7*(~kP>?7Yaqcq3D)md#1;<i&mnZWg7cEFPlY(00<;J>d_t81H{ix>cIpk(&qN`q>Y>swYY)yD(5$Jmd4bLH9?G`9IGYUh#fRAPe{!#|ZD1%j@XN6r*XPOmHj3)<fECuk(K)0|T;@k{VfVZKXQ@u-gb~*>;t|VEw7hFo8@y)U`r}Z?t8>UWk0-s?ewhtT2!E8}u$K(%(P<H$uiy)jkPIK&c|6h2r5-z&(k=$j7kfeQY`X3nJKpjSqcK)d>A^7&lH(KEtn_3T2)kbne_ZuF_MVqV=Gc#D)u8Hhu2VOl^8D+rNzS&2w!QoE<HIuT)e;sqThVc>0Ey}{q&yawUEpqJcvY)~r{Y1UkYEr()UxXJ%2!)^quX>=Z8mbVzv%|Y`TPhRw{@@PKmOV`dv#~Bp0rSOE-kK*Mle)MK<)fkALCD}K2o5YG|Mm``fa3p<_jb+^BHn;jUpyEof-7OzYYWYmS&RsE1F%msL|T5!7j&Ib%zP9aDKQd>l!j?GP{NX%eSV0#H-qhHM~|hdl}~Mx*B8KnvWjF6C#={+ABl|I?2MVJk)&MT|J2Ub}E&<Q{|XlK}>Qg^NY!lpJf-JruKb8`SjXeVeIO;Ym8tbxGXl%HUGLm4&J3wTQ=wOo*P0nx>KGnboUxN+QL<rJ#23HOgsE=3#p&pX=Z`$K<97={eX3U(qpmy+|%~66<z4b5BXwPDv8&vz`>F{es}(ev}D#Zd{(T_wDeZC_hM}O?cHy%C;1~Lezr%@TNMOS!u~7M%WHE}|IC5-`WT67Ml}7+c6H-stZf~&d$=79Prdy2diu<#S?M2*_<6kD54ANPZvt`4P3sfbk^q14-Gi%Uiz_u%ae&&YgX{7A#I4s)KBOPL!{F2peu^e_Z~9>Ys{uUlIxd+Mbkx}9m0AtoHjQE4rE=stbGiBvo+eoX;;;8K*T7Ia+9T1Wq%oaU*JAe|`k;NT3Hg1cId5q%(Zrm5%-8aVI<_0{lI=Aqa*<d_BmFhJDm{eW3_C5~0rk9ohQQ%Oo<?XFcn;#~U3k?hEjTxcMcnP4GrBg{%hU3F<2bnTyDXZ;Y%G4i<MoF`@BA6}+N2k4`5lNp8+r{HB3`-H=T?c{+OOBI-1eY0<cg&_-Ci(o9x~E!NO%yB93`Vd_N*#pW#@1JfA{Nrm@RKA+z+#+*=XwSNM)K2Y|P}Cl7fpfcxWz*!)3e}AZHEndP__K4$hgbeI|p;z%B)LZMjvTrm~&H0qrSXF0xh+xCkmk8P&sdwu3`B<U^!ODE9gY#I0racq&KsVKvOrZ!*Cf+0cOvwx8;L<-ji7Xtti*!|2bi2lc96e>qK@<Er%;_&Q@Eh2)oq^l@on<8#tdLy32F;;*^n{57y+2(QZC)&0E>;6<!yJK?TCVW(Tm#D@Hih)Oc^o2`v1#lmCM>)oKotr6zYI($c8a1Ju{HHeKW{;P~fJqy_u=B5L^_+9Ozry#O80dfd}Q|h|#ij%S1RV!N6*qZJ4*{-tzOeUFW%!aQ{TImHB*Yx=NP};HOaogtA>Fw5V+vCHt_O0lMv5)%o*Lv_r?Pk~?z~w#P(6n@3pVV{T8PqYe;a;49<P{5oeLXaTc!>nt2D@iIamP7^F2?CF-khk{GVTuBw`lycPumvjc)@TK<bR>{G#^L589@<w+QyA#P=dzibsuQ~+Z+!YIG=2S{1WEgP_81Mr=d{Ldmwg6CEL!rh9($V4Hc={mY=P|&H)cSY}Vw5J1W^BfJBquD!@Y5wLo`1NmZ`^I|_jAfM?yc=Al>x^cTiIP3g=wM!Odc3(^>=AtgdTq4mOgp7PC|;C-?e9&rEsjrq-@up82AMHhDJTK-jOcc){_un$A2({Wst3mVO&wI4b^(w<OX@IAZPj8$W<g}5pA;m{g7ufcsgxsl`Tnz6c8wOKCUrOQ-0m4d%BQ+*;W4SPJGzxn}~4*8aO_*Ji~NBrzY+w>>!M6KR%Gh5V{Z!IB^NV?vo0C2YeN4U`SM8cP00jmCV2|`<ECj1B+{G6aRBK4^WKq9p<{B7J%KPjrmx$M92a67Bq?dh4S4|h}gAb-^Pg<>mzJTGFn^Opu?Pf%XW;uB8z(-;4Bc*yAW!QU|f|HjJvTcyhq>PZUE-nbw1S4~e%DqMh_3O2B)vwt5Y{SKgp>CxO=Zb*5S?Pe$clbm32y@t=*{gS$9Mlhqu*^d*UJPj{6(H+;vEfa`1>i9Y^&jrdn<aoIeRc$Z_arbxZ52GnxISPyCvphX3TkY0NAKM4EX(gX-?FCA4pHtT6bZL@I?|8ID44A98vbNac!<Btfnq0WOHxG$Y0jk@4vCQD;@;R<7`hE--^1aK=xbCNm)&_NeseJ1nJE*V9^s|puG(JLCPT3x6yVC>f#P|<ULf&)5>h1EA>hv5oDghF$f^o$gsE@k$vYNQ(A>|jUpN5|G(dNuwRr9XgWk77<za(#ToZgwbmwpGkgM*HE9U9Yc0<HFj)9Cq2P<HXNMa2{Q___M613K}6<Fa;f4YKoxW2y^q{_{<Si@LeUCg0x6@|cdHZ}8r<(WtI`csF98QDqziX?|TLOaA0<g^fVB{y!(-<AgTs#!(uji$&)tjf%ac@eTPD2H^5+$&z}@R!R!bpIzpa)KeJ<KJH647Cl@iGmTUl<np|}(P5$YnL&eJKh5Id-DZ!oE1kcQw}<5x*57~|b19q;Wx?oWxplK?-y$kkuk1$e`Z3lbi2~U7VIPA7FFp1a_se%Ev=kuD-+rHNjXu5c!g+y}R&K1hotA8@YjQY3IDTGyC+2}Ul}F57?#0$GJ$2nmgH?WJAosmb<2f%!)7pIg<)WT<UIex4I|KF~IN85K4(h*{`nO=@bc2k3?|Y@gLhWv1Vawilu0q^N{{Pi96IU;_AR&&~v|E`4Vhgb?zv7xWwCGwiQl(e-VTT{|Ak*0dyEQ4#H<aH5>uH8%SU_RhMjcw-(pA`&y+@jg=$~_Uj_+<B7e>?k4&^&HLZGS2&5fjD65%}B{|<|na&deU($dvU-msubriYlQeh;(ho1Zh&{o(jfTfdo*Uy=D?GPV@1@unzqNxzqy#&h`+wTHvQ#|1;RnGiT1+iblA?o$bC-IRydL?tfvSs()Icccy#*{&;G2R=M>$(yYLlzwY3CEx7GMbL*5217J%ti^4=mLk3ClshyCFgrZL<-Pi$pjjh2<$mMiRR*8M;a$F6b%I%20{p?VYuU-~&a~Fr@7lW-d3a;DV!uqi-fqU>w{v&E9`D8qr~GrZsMa^ZEct9c!ZNJ2{yqY+h!erD<<VRt{pVL5u1}*NYh8mk>wn`dQk}c=rnq{PUN>jYz@C?tMb?kopkHY|uLCqJCu{!IY$$o<l*?6#=us>K>JIRfTZ1fA-h5tN-lyHKR5$&z^BUdhvHew{iO;>n%MGKphkIk?;3!d|IaeWtcpUuM+!Xi`^(K|{c2Rd8-DRZHAI(AZ{uDc%LB75+oq)!E9qcgM>`psH#9vbIB@Ch#FNf_K(Lv3gOjVQQS6R=%_B?;&c%z13+cLL5D@vBC6mP@<a+z)OS44_erGI1WszF}7E!^jK{D4hxq!{@3zV1v9T&KP8k$J7o09v|!-kfGebfn<-RF|t=$Z$P<-65S`|99!P+S|qk7{#LHo<GWsUN@>sDp{T7X}0!We%?{54V$}UehpD&@M^9~z2k^+xH!5j^`x9|x$cS&N_$lvt;VX0>IWZy%_=<=_;4V-YFb*t+t$5V$u`f>eO_0?Q3~(ApH7>MF91C`F}1b|pFaXL;?jwZmG(ZCYLH00RFpxYyn2r5cM0_5@L(=`GkL>|nC@vdj0O`$z};e7sb?#L%_<nY#(qdEaUfGX*E4X6pVgMO@@oCp4Vu>|Xn9C$m&fy<9AooqGyY$tCs9iIeVc|n-(}|!eK=*K@%#=N<T~2~jaF+MVrrlNb;Z}coDLq;F-Btf`aN8*G^$M|)nqgHbeBjV9n8k{Fc)}Szby94IdWe5%@9Ab@Vm}XLTY5x?{FUuW8&$S48*@+h^)}AV~N#Xw5r;Uq3!c+JN|5D2hU)~#6^S`c-wgN-1g_WSk^+-rq2(xXVyc&Th=)Cg`&{*(fNEB)7#<BOv>)8B}Uj>-z0I~cd2@)Yv6SRA|3ozKBIVqb>&>H2>h=ow|vtHC+PFZn>P6TTaCTBp&!d|vIsu{de&MkqfcqtsxbGxZuhZ+KG=2amcV<$#g40dr+H882KloQ%;%96q$WPsbS7TW`%ZZ03gq|Gjn%d4tkN*@A1nOGjN$9l7f&<P+C8g3zu~4?nI{?k4Pl+YCrWbeS1%v8psq#S&wJppZbRp|)?|XbqYdly=cxtFF*pd#!C}^Ffwpoz*i3)C;XC=9vl%(4lf!=Oc{_ch3bi#!Sfe*I-DBg!Y9?Z0<(ejqS~}He7oS6-D?O{=P*z*WZS6IITcWmvbfNF(0$HRl+O}`s(Tm*s5O=SkfA({7dwp*nTnD{+Ge*YR4&y^>;yv1XUZ*OM_X8@(5G`KcQ|_i?A@U2RDGjt^rIyrM)M2_-cYTZ*{WiR_XZ?P<ZlV@WjTjtR$zaP&GWL0$bUyb|B+ISv1S{L_YvL#^DNNQCTPtG{)!T1LhA_PmeaVHx=XMLl9|l@t!_M^y3f9jKX%w64cN?A!)>&5F<3(HJd&fumuFTh9=c?sAGf|qTu6BH1%Bs&^Mhfcf>LBc`<p;XFJO_O5Q%N@iQ)q9f6#g{|;*)@b`(^7ruLs&jp~u*XTkiNP1<2WqitG9*fhed?YwBFBynZvX*=G}>a^CLc4v%B#gEZQMoS1Ft2O#Ug39(O04bodJZtZ6-h>R+2pPeP$p6o;ot%?knGpi*$bn3xB$mVhmY*q>WvhJ=myg!;(XEGuLFs$4j?LHGh1O9fJbT0eBqDXSSwiiiSVO82NQ&Hvix#XnRrFIQ!sa_I%?2@xHNA^3Vv(D`+Tg;e^Z*aHQTBM!*DR9YbqEN;P3g^Y!3C9boK(;8o?^>4dSLT#c61TRKMR&?I0jQ+ftLdbHyeAl#otdjQmpQ`ljmKa^saR}@&Vyd-Lrn$QL<y<x;L^G$-vF_kJ~%ZC5E}Ltu~f}#9h5+%MzeF`X3_}uKf?ww`nY-ro6o_=Ujb+*(}H%?>R0bi{;&(0B;;$maZzmV1p2|q!F9XyiPO<kF7<_{ubL<2dZNyiDbTQn)t1XqsTT06dExXg!B~Cua7G0($U|q9eX@6EL`$mqz%u}ypPzhXHWc__@nYN7{%#y^9oHo)!R<3>j`!M;RZa$lt$!BF)c~poN91?VsQe~*O}`<-AK2B|&h4nZ`gj+93HjFEJ*;D=>Ri;=;`Uaf*bXv#QXCnSOPxM9<o4i|pvbRotI31)&L1R!e%?CmU;lK4x86h@AFDJ;S6y`h-e%V(2V~f$%F5sFFi57`t<Mcc8K={uqM|+bA?6Dizda;BaD=tmbEUO;qMMCNia$6q_ZOmO#>|U`ii12~oUdnEi4su0w=Uo(nqhTPctr5*T&FB7OuF}1BW^_b?NTdy(Gv4V<O30#YUfe0x@z<M-7OrnT@U_>@l;SA&-WrTdeQj&?U7K@t|?Kd)*kCuM7^y3t`SEL5!l=7n<AkKMgCd6fSWC9kr7BNRDGIQKt=a%9a+wW{$|ZAAC1F2dcCn<vplpr-mlB9j+IBJVRy1N2KA@fD(Pl8%+Q-zL)li(l9WQtD14xoMc;b@*y1zADQJ`0Jl3|gsPbT#oB;rNf?V$5Qp}zw;TOHnNx$a$4J#iwME4{#(Q%t;F(aHtZ`Z|vB9MJ3XWP%)lucHwQ_>1|w+T<&epO#Nz30FkIV7^(uVv2EDTkLI_se|$Pc^6~*2pRL(7W3KkR7r3J!;5wtv#)u*8Cn`tIZjwNK;16VKweg_nX<0c5N(voKLUt)(!LT%Nv!hA*V&_qsBz}Ait$ORV^S}c}xvhJR;(!6qE0PtRLTJJShnD<!dwL)oyVJBJks0)E3>QIR4ZPFw{GLZ;FQBXHRl9BEZgTOfZMsdE#*hQ#?-=OMc2-iCKxp(#ZF$pj*3ge-9?VZ#9<AJXbQQDCvotMi4DF``RC!RdGn@cdTnSp6f}DH(3PjX{<@i)#k6NKG*|RKdm;i!V^bI1rk>5OB<uh^hC~z$LMx&*(O0#J0$ndgRw=M#;!RHgJ&-G7W~%ssO$K<0uTsS4sMtJjLDupnWU!61JBR1odDIH(1wb54|S`~-Pzuj$4N_n-_O0fK<V?@q)J+WoikjvZA||H2eLyay86}NPfvgc!u7VFGfb8Hn<gRQuX|Z+9jHDiv1}7dB2i9nMZwSM!{tzIWW>7TsBzh;4cmS7G*xXfxz+Bi=%qjGC&2Vf`;cRU4tF7F-0q=$;)?6rAZ53o!R?^O<W2dG;x7Yr5dP$j?q=B3%Qw%_PS2l#Bxa<m`e~+vjD76UtF6DiDZg<J_a87Z--H!WH}I*0zk?$-y<G3f_Tr0g`uA+S_LCU{{&BU~v@cyCI8XR`<+Z%imLbj0*=||;$@JITroJW**FmCM?QT^jF5}wdPg>KkvT4ACLUCa7qC}v*wV&5~UC7nstDu8kK{SqawoB#f8g}Ye&}6On4iVeass9L(GmA#Q>|}J(`Jg^{u6>sL#Y&wx7a+JD^GdqnsyP!Y!Xf0N@ew&G_OHHQ);CY3=@CZjVjlYpQGU#T;Z5_~@%K;9Zr3b?%@87P-sk-lY?HU2Y@2%}+RldWRgOGr;vOHi!B^W|{Jy0+mnC^K|E_HK+fWaGQy{dexAwQ`7Y6j*M7QxAgc)SyO}0Uab)5EfGS^&=vrsIY&sS3?*38`^Ej*s-E|rqTT1kt7#P}IE-XT8POu_c|?%d2N_iCJea}cR}lS-Vs<Dm*pgXOl5RfqdsO*F~0CeIPpb(iVF6!Jh<;`!gajyg9GB9_)WY+oYZlcu~;?JZq)VvTaGa8>bX%uF@P?q<u@xVJt$`h1?wkE<+ZqKph><5HR)7Nb~C8#te*_pCdTM^kRM^-a_%Fvw2Ca-Cu;Y&2T$ry~EL8aIKTShM|HJos1BJOHbFRO-KKz@g~SoUnWH`=A-zLz%gz3WHNJgOkzsE%^sgbGBYOZPnm#pSS-LPk-E#>PM>>=i8n!S7YdoAXFkBi3;()@#agr-%v^1501&0e!V>y_IFbL5d{;R2T(5GsuLbEA;RH#t^dJxn&yyo5W4&cNo5AA_MQUTy|2_NExAVD+n_wd^*g%{s9?P1<%GnP{XSdhTkY=*wKc?}Rz2Z$Wk&At4yIjO2zT)fxeeAP;4-mSaB<{AHG%V9TffNpD`={V!9zk0RbbjxVRHYJt5tTel_t`$faVLgwe>P(+8kD+W`I1HY4>sg!-NpR5!I=}?S-|*PUGQzpU(JrP*yaki@O?q$d8P%98646Ifl_WO`de5w;HYnd-`vFpgKZqt!=O3?bzg1o6O?0JM7f=KiLf7WYW1>Y4w>;aV3?3g~~Jt@Bwufi^jB`K;V9mDooUrK5!p)DV;4F-*eO-wK7mX+gq8`!WuKZh8?wEDeJ4dF*f95voWpAY|F(?ys7*9VUtJOtted~@;cjjT9E|Gd-3jh>qIKPm%s4ab%p8Br3XDY1xvKi9M!q*%|?12of;4BFn-hXL1!#_CqsF!CJ|qT%%hprrH_{*Bj_j3W=B#%kv(=0`e;guLXS)oX*s&-o8yN=_-m`k4d0&JGN!=~G~+kyBQ{4S6WOa`w@^^wXMVviHxW|%`;S!m-ya-0&*&BfkC^e^=&Dw!GvS>shhgUz46A?UV%A~*-EM{WdNP^mN`5k$vDGL+1#Br6*cNq%lOAWfSBzIui~g#-+#%+9kleoA(WjVg=Br~~-`IavxHbOFt^<G5-?_co>emYz57V{48J&mRU3O35k-Oy?2|eopaVn)~Q1<f2jjPo)ZK#vun86I<!ffmGh@cL|=J$9}-A!d~^8L3gM^t0uHltu8V$KZ;*zQB%2HG;vzmU{+aWVYTVcVq6o1t(%N4kohKJhSNR^n7%wcBFyB33mkD6hrOJfVuX7?68Tq8#b?aUAe?<GD`s8`0JzX)h{uxoCfwaltLvOJyx;*#N*B2S|SXkS{pz)ls3lLf4^Jt1Y2;S9J7K(%n%<PqaE^cIALO9yIb1ZM+<?vz%*&0}{*rA3Yn_3_R>_Yq@bqaxvJH(z0C7MYur43>=2<^_rSrgmm<xup2cjj0>nXS@Tavzo$<cZbwz&%-g9mpC}vtcJDRZsHqW<!9GV?+gerKDR%KrhraVBZfT>|nCRWjXD-Dr?+=5)1^vEyglDW94+Cd96b~Iw^CJ)DycD`l4Sf53yPKM)antMH_#BRgJ))Sptb0;&w0@YPgHY+We9QHA4Ie$ffHSe54@A?T9`q!h7(40k?X<+d+MzbZ1EwD9ABajTUpwfF&*EThxTq=6H*2dO{b<${^f0)}&Udej7FD}7Bz8AXF8cl6Aeute4o#=XnygXnimV7{1g`trP7*<B3;N*Q$BqtKgEhn|LVL@d2ar=-ezulA%1LyZ8E9`;`}*9%tuV@(Qov%HICkE}DC#5A0nzV2xHj62ylzV|E+OP~*XHW}*<F+&H8Yy~8()#o;a`GNv+_%*T>n0P*Or4L=o|Ad&Os-JhM>b0e7jK_o+YN%d+GXmoE*f5q98|+B>0n7sN8AUkn-|cXQ0X`B3UU^&wX`1IjT&vk<_m{$mQVUx%tWn-c>%4<+hq{f-WcUcsx0>AL_N7H$Ogd-^3nVHg6EfZqx^nzP2|Tlc<k<c<fJ|emGq*s#Pvl{fEcZt7n`-2J`t9hN^@0RlY3FA7V-B-pzt$ZVwRGJzUbb-Rb__5gqJ4+q<5EigyW2!q@FP`uuG_mvYz3Nc?eV)RHbK;YP}|^#{A^FgqIN*d5N!6RK@i?+HBcuKxD%T}Af=Su{d$@qFI13I;HG9=`TYry!ur!r@lFX}9Nf3EfpA(-<?W5h6A=y{AS`0StPtzr474w#&p&SMYh_CQf{QT&P3TV6%XY-ywgIx`PDmVaw(4$@R(E>pU)`6TU68ck$b``oy$Vw^v)im7kyP8t|`5e?wf*CAl8w*Lrni{;rQ?@d_?r_uDmFkZ<XQ|GO><`6t=g-~v%6w!*>5#RB*zQX$%3=(>M!9lx?(eDTd%>)nD+r;>^%>vbJLq<@*}o7OLkx+Yw-i3NI`<`AVCrx8cC=-wvh-*p}p4efH8-<YlSl5PxtyeP#iYSp-5{b+W$A0i0-y;mF+#l_P0Vrbx`SXAPeXL=Ig%>3o`_};W51aTSZNJ_0Lw4+)RWXsWt^J=}~<zao8V)OX2H4i9dy!)UqD}Q;F4e1`f$Ioj6#-U0Q|NocQI^K6++Zc<d#)pvZoNKao9p3cW1$QMmGFx){LMSh8<&FW?pF9Qg)hV2GjJWaQ&DrT$=aE*giVKA}?w)VJ1bkPSajy<H%3jc$_&y|7pNlJ|-lP00E~rzhoIpFxSuH--MN_S}x$gZseqz*zEP|S|0#y_A<jvTjh8uVY>416|&EUaN?z86iFYN)G_vy7CwUsKnIna8AbXaR+&3>uU8OIcgSD@heR}?n=Q<GjZnS26btF~I6kbYI`?t$HUa?u63B1&V_)9!<=rKLmR`IaYlDyokt(hC~;Zsc~;#d)I;cd+uFOmsc?i?=o(0S{wBjFpc2?CsY^R;k6GiK@Oe9*Q;~X*SyCa6Wf3db!)!BL63w1ezD2a;>f#V-!Nm)4iq9*}~|Y4@KB2=u?-%sj_tU?2RBd34Og{BTdt+@i0#uOo4aW%bCxDL^~19S9L&+O>9*B0&2-0@p_S9i)-yE9@|#O8&~1%UYLzfy5Dh_*HzX|%elx=`-6Q!-0<Lc*iSl>8r#>}O%E_yEqE}5hqv1EC3DnT=nlP}$WPr0;@c4+{zB$8yL|=-zM34st|a@PBw;F5>e?*MLv%FQElrm_f4BF9XRv+0z)X4?PT$aclvKIZWY8aCS_zlZyh4E`<k<c3HEf-R?T@F{(S1>A;-%1+3>Fu2+2-1D^-oRpQ?$T=ncj(}y?pW)nB=Q6)VB6v-L+cDWv9GF{58za=-xLiJ!iJgWclOA5n3EUD*-Qn`098q?J#C)p;#8L``OAV)mG(Xb7;)dPPl3Wm3vWgQ7+)%gf<$Z7ixC4d}Z*O9mVk@w%Q$?r@-2Qvd`VSL8HX&MBW7%TMrg{?1@nO)^$;TM2|EY?Ov6W{>Z6salDvZt-0f`y&SnVoA%N6kN45ntqGqf%n!2`+^jRdm?*2`su1}R)GhwTbE&v**=ZXil^Y&kc!UGjPhV1dGflZQxOyj9Jcmr0W)xHDJN~8Lov>(eXwm8)=2vQ}=sOptxe#@zB(iI~GHM;Q2;=bH`cRw0!EDxL*J9QI2X}s_d@fl%E%QzHJPuI33QWYxh-{1r+^hM`Z{~Ll9%?k4KpbinXJ{T$gx&_P&M4oX-|nL-NErc2QQn4eP+8K2Zrsk5<#+1WgA<&cXYy*zo4t}>;-?$ao0Z%1WD>>k9o{^wT1>cy%dFCWbGbdb5dC7dJy!9{I=$u|N|1%zaLCFqN4DFIZNH?!Jo<XayV?&v_NDVP)eUG8^ImiS(<VJv-F;HAV-kaNxy#pi`YPLJLfxDEbU|Ue-uxRGo0BnAmN%(e4_Xz<dgfQ@Z4z8*?go5YKNj@#tmHstz!L2Mhj=0QF1p!Dz<)>x>hPB)&fRe>noTW>6j+RF;5-#ovbP09P>=Y6&BI7_x7OcD&krx<u7V(HW5yVU`}^rA;VM4#A_^p4KN?W)FPsLKhVBofd8ZC4@zH!_`X!zMP6h>6)x8=!u`1IA&}O{N^7J(F=2(w$7xtVJsL5$MDT9l$aRFF$7d!arHJ5A-Et-{&^JLSQWVJ!^K6&WXb;aV|8Ft5qGdSPpVzuAs_jXD-rXFK19ylX0*^yGy%7$WI|6Nz_t~|Azltt6Ydf0R!gN{I4vF+uhb2`z&->rS7AM;&V7mo0TkCxk!<8?2o!5w3j3ekjdA^B3_Tk~o^BJ_HrHrBmgkN<Sf5J_LaMVq*919tXYw(8*n*-Vi$+-}2*pKsBN&g!sSIO(Th$(PB4pw_5DXb-im)~KlwysB?5l7X<^0QXxAy}YU6iiNS)uFa{!^-la6_7~YZZK3hUSh_QBmO*<ByJcteZEn%ewzd*0GoM@!kj7ZLz=QLV4YJO2i;Pz|c-(&da`e#lWp_|~pdXetwW)QC^I)9<ufKBtoZero&mI_+So~BA-WwZ%373=i=@H<|k;BHqX@@q9^WJN`PKqI)sj<P;EUR4^`N?s5X+lL^KWr~mkU7}5)p&tNlbCy5*~R3M%p{xac{}|))L-J8Lg?4FYHlv$6xV~ZRQoG4jn&E?H~ag*EQIyyTl#?#8Pz`v$7*)((Of_57E)sI=-B_e56*N}>qk+KPqq4jY(8&%q`b7@NToq(_F*$=vsFXuryGEQEAm{at1Q0xgh#NL4f5A3)8yj!)w?s<kSs{FMwg0|B1bO(3YFe5H7J_ju7^QY?+sMqF>7UzJ{8hXs&AlreTbuT1`}z2s09jbwHSePB>8V<IH_e4iH<Pp1SQTj=6tB`V*2u#*Dvb%T?C!xu$KL>4E8IlCejxBq0+Hc>ptpkqr(_-n+<6pOdaVktkTahuIT$!)9%!7GCUtgHrpEwQ0_1@V@+D~v&HOPyG<Lqz4C$lof-bg>n}H=jV<VVmfAIkUvgge4*|ob6FcT(b$nYa%X#s;?y%~?e)Rj^@bT`>bf2Fl?QYCv%%XU2t!J$}o0RWpvtu97WVS_MQyM+1#$q74<YU*nOD`S`Sv&|!0&wImP<bBZ8iX#LtYcId<G`LmXO+*kv)`jX*+mbb>3kQ|w!fT~TS?xz#kez`te`78Y5<cnLyaX^T-fNh6GGeMZFY@bO>RK@jC%?lgrg`oVQ+j7A=ci$$-Y)!xzel6KhsCwe$CQ$XyBJS^OnTaj5qC%FJR#XI6scn(?+L~`>lQAm)i7g-ui89#FZ6{%p)G^xB1p`a2~Qmi#w<F=Adi{$~R}LH_IN(TK&s1(U9kg)L*x;^hOVhY3w-27$Q+`ACBvjXX8GI+RpPcc?Vc}S{^1(uxSM|SJzhhZu&cCt@$T>+D>->_t!MClhRqYXF0yHb~a5@?iTi)N)@~B5%QxZwN|I6e?#y43H*2Z7yn2e`Ml>WOmYT|m(1fi-EH$;z@~%Qd6DOx=B2tGZN~A+**<n|w8bg4cFlKKIO?w#r%x#Z41KN5deuwWq8`Wi$lTknqe*3-Y{a&P5GwQmEeS_#cUQN$&7{L0!^_kQnrHvvGS~H|tqbU6fG_>qqIuDbG2!EnORI9a0rb4{p2Akv?D+fB_9?*#l+PVt5*$8cHu^JgkeX{q3$9NdKdykpZ1ZlJBQSkk>+*7C^5tql;aW?5yx+1!Yvb8Ud2~5$xMkPx$+wvN3bhN}*(d1d*1m>q&0FtjI;wG9MIq|qwX-MnD;I0+U22T-hEI+*f0E-Gf)-qJkvU!UGKhXlYmEXseuTGn1n;dfC}P^<{I}-YZ|wn7iY$=ZV@&pe8-9EY;NJtBeh=`@9l80m7~|5T*?_Nx73H@%Nw=cCqcu`m)SmW@Iq%LF{JF!Iur1mxQ{GeJ;G(xU{W4!(n%8wUR4PaPvZPUPSH_!`VY`d*;V?1WerXvE#>vssE<GItPx-61$lt35K2$#+6uCO8e0*bEQ8w5g^%HMB;gpkQe9c{La@jbHj;~rvY@7M*LCI4@gAUv7BuuAtbUYuXUo?0<cTb^b6XxT48#0xOI39RU2rh}A#<y4YJLVV3ihS>X&Y=zT{ZWu@km=#oQd@yKAJ1yJ>}M+5x2~%#_$)WcZGZ#q#W;6D!fliNZ+^6Bk?Gw+HQrr|*K55Mq2u3vB#Hj@#>2ybq?)Jey3BK!{Cckq!TSB5<aImQ<OTq~`4@^(u!Vo+{s)L?rWgg96FM!KkA7w)SWz9hD0wb-=w~(ji_K1l7mlGbZ`CjQ(f>Xk-1xBSv6J>`lkU6LZkeM9?(-Xa2;8IU=DnC++`s0%=+;Ml@hgtB<<!6`Ytj+IRJ1h~x{bOIF|gkKxVlecyGnt530S#B*G=j#yB#M^bbSvmzjxQGj@x32wVq4(%$!Ewzhcw7?H7Ed46s0a(ZBbwhlGO_+=-gIm9wl4w<B%4PA?BnX=Jq;Q+@k!a7PubUv2Es6mv1Q4R@~JhisQ$FL^`RGdgxUrd60{&$|Uc#bt%n$&HsaGha%0pRIZ$>aZ2O9>gYKmF#q8OCA0#7bL`5GeX04g;!FbRawKmZJ`+E@!n~f=41$MuX6msy8f$(-J`0R!y6hb>TFY&fnqV=O_z}M*dbots^cD`otv}UgIaK}tRXlz#GBrDT$H70LpDr?R&|f1VvSuo94smGVvb_KqkH!|Q(tt3O!y+YCIDWi{bXlBb&l-ma#{+Z%Zl|ab;q*t@bTA}o1nS8TI|Aod-o375=8UjYk$T=weLVy!p(+%dhkoWD?F#T33ign6A{C$x`y)ov!Hv=NKV}vJF<<@RNdxor3tyt$E^9qyxRSuUxy~YGG;Homy;5dc-UU01L?P(j6iIc*iZs4k9cxg1ju&-2on|TK52{}6utcP1}l3}Ir=70^9pUMg5xuaC*tzGJgunv6JVobf^;v(cy$c1%6K-U2TJc|8OvV#6)<acSFm<;III(Ts8z*XT$zUOg^msX+aAE^7)Yv-f3X14yM4q|slDKG`#T-r5T0__oj2FJ*A%uZXnx+3oX@Qf6RN3w0mt{)?*UEo&PsT_S8NeR;j?V?;rfO)+Q-jYoZOm8%_doixtsv6TES!XlZN>F<c1M)(DZ9C&ePyRz=hD5n#<yuV{dUm)aH+(;kTLcIvTRf;q4IdQ;@CgisHxV+;-Dh$HS+RewCYbMOOTACFp&q(z;v_2vvJzl)(c;8;GZKW4YP4B0lHFM_!cP!}jLR8q{~MK}iFPF5BU>lgvr<1@5-i?E$s}0d8(u5>CG7ZO<4(a@Y8#e_HB^?0Mn8Jg+WIZUR_7gwfz`EvDvC$wV%tCx?fA-E|nMJ!w@Y{VgCOtC_Rhysp22(kO;OeBC!|zq60)(O4i`AA~eL17E)0Dma9?_LJ(|8G%iCzzYVCnw-tF)$cg0&kAcZX{4<eq{Hzs9!;C;0nxfw?atZjHC!$CapzBQ0xyK_nYK#}ZaCP;;FDAFBK{Ejp8Q!;&h!>q`SUox0O7vhQ}|6o9$S6ftVUb&F50H}w&7KWogfrMk?|VuuIFu|IlkCrw1l4<O@A57JYI-=t;zmn&tyqg8a)7Ru;*u(BRI0!(AoCd88s%?Eqxzc23Wg+=nA(3)|P$v?wVJ=SJW(pcQUuPA4cO5DFqo8>wf^C|1tNr-EAbt!RYtrSBzGo215=YfG;8mY9xUGNJ0bw2mmQkTxc`^1}`GsI0KLbDe8FbIC<`MPHb=N`#AAS94C45!8!NZYp=atuh-|~7f^p8zvNa`_YCF@B-eKCxw2Qn%yd_GS65e8RaaNF@6?9bXVQ~5XGe3rXDg{fxwn-r?bT<}>GNktooDNX*T=n5cQBjVd$qi;Raa_ePtRv&!*|As{PX>b*4)Lrer9d+-d<v1zmaat=Po;Q&C%?+a<X`TrK+vub*Z?&{cLu9_F%KNcV}aKM|-M;kMBGg)E*ynGR9tXbH7@;|77$K6zrYxtB31b2Z{UJr!QCQdUL0p%PYIH*_Yv^r?t|?9gv+zuhuro7YBvWV7Ga{J<>PkvhT)w`Ma0NJC7503bFqDaH;gNS)bdzueHz9t7n~=qfYkj*5krz{W6p7zwRYg7W2}}c73m&-`z91^|i&fvD2NEXRqd-otGXeyGPB;+`FZPH_dEpCmh>-y7Bnk%>JNWlU6!M?U|RUw->RQxzm}w<CWOM&7;SyJ4?qk_0in3;l8nyFqaQPd)=LTxt)ByQ;arCJ6mramS5%uuST)$<=uA=tNG2xJDOHJKgm8-lqb*bJ)h~_kDiB;g`JDRz02<F*G6cNh^~(FnKzxR7F#x|<7EDR|JB@`W^Q4*{Pty{5x$p->20IkdTTz~Sl!%Tcs%o_AA9=n?e=MIc4?rlJ<6PxljG%ay_DE4w(pITZ>!H<zJ9m*W~Z?<lfOT^aBtKsL>Hr5!^^?=-gc|A@={$}f3<Qy-(2Zmwv(M_%a?PH2Z@zZZFe;4)%UAUPBM3u%(yTUu01@ux2lxxJdd8AAI-em*7g3gm;0N^1@(L()|oxO4DFtWH)`X?xV;dL#`dE6zO?yV>lmTO3*&|Evs@w@HW$;cPdDGLoE*I9C+^%UoV+WJU(LLD*EZ|XaP?85vA<MXc-PGwo$nT<(9-GFa(!99W9;np-lz)=H3@p$+wu9+-QwV6oLMz=Eq!n=xBg=N{*%2Y{lPGN1{;pFe)f3WTYS+kKg=BL&JWZ($)_P;XixfYPaCTTZ+6p<%3E{gC#Of_r}xh<inUB?f3Ch9?QaZ@`{U*E!JBmR?$Jzc`+To8`@Au?+3P-uKJ8pS&!-l~F>~wf42H%{#@1GcYZvXov&PoNV!pRq9qp7endJ7<z1~W8_3ioFttWSLYpI#Ht2<I^ag<n`j~!(i&GeJe?w#d><b~4K?-ZWj&7|AU^!)4{z>R)<?@jgj^7`vuQ+rh#cTP5UU*|voN)-xsr1clcjZmRh9q5;j&WcM9XP!$5P*V!pQsU*=?7c_pXFDlvX|cPOnt!}`TzY<QX6v~7wy<|z2ATPJ($rgb?%eBaJzJVfEuSfGpT1qqg?DF9Ufqcvq&JGMR@MC8LAP_KwD{y=^m0epJAPQ&h|Zd$-s$@N`|9>w@?hiD9&DBUtycfp+^gE&c=#?c_vYZ$kx{E2fi^gE|J@5^!B95q3%#dW`^A$t!{fo5_2=5FzA`+zyRh=4bnoTMMy&hl=%_v;<#*1rsky|~PC1uRm+utM&dk-N<<-=n-!Ba|Hq?_wBV_K@x9ifNyS{P$Y&$mSWkSWX_4Dn`zJ79+(YH3T&Bw7!W{_!x>+OS+Whqmbo5NF@<<6rgFFK3wmIvd)(cSDy`fjdvY^=Rq+>q|3bG3U)x%>9f{MoRkyjIiW*Uz%*i_O=MLL0sD+Tx?^Mt3m3ac@m}F*B^KJXtQ6mO@+2Zm|)|uZQ1`U#&dbkd)Ui&hq-xXL@!u`>6Rare7TN6ARn=(zEJ{ns{2g*ng?NoO^nfY`>g4+M3zyrj$kXa<jg=RBOH*rr#c{%|Bo1pY_9+c}?AYH1lvTo$5T<c$w|oo6iii-peQTkok14aU7kW*<O6S`L5otUPL!)sW~Z@Za*36y_40}tCx@V>g}@wJ=s}1T`vt@>RJ@0sZltOo<-jtcl*GwpGw)KEa=72aAWYgyOM3aE!JOc_3wgGc=B*7d#AX#^lDJQpV{m*>Mx`3Ubc#l)asps@WYM8%xS*)eDzN4`0=Rt{AqUmV)M;*eh@a+@)xnJK8%*yo4L?h?(E6w&gK4&xu~eC>kG?T^-bY2KTMq^W-bmkH}~J2g`Zx&(KF$~a_mIXtH*acZ;N-+d&_0@Uh36KZ7{d0M-SE#XQ7L|;(l&5+E<@E%IVGfZ;}_U8l$cA)4`i}Zwvdqxzekmv64MsIOsJ-PmP&uew5vAw~lsa?hnV^#@xl?z3zQ&sd(@72~2S3+1=rG`uTA+*Ss@0%Qo|+`PYU0ot4$$t74(IxVI$jEp%QkZM-Y(UFdff$0y3t#_Y3a(zDlz+;*Wi0-<_8qbIe^i_K2$K-!(Td;dkF)Hv(SpY5KW&%W#AQ|<HF)2*lb1K25^JRR&8ms-y@?ypt1-=?3H^p$M;X!|_6(K{RMNy&T3t?`>@C+4Gv!`<4+qvF$*#*Dd=3lCG{wo$uyQa-!bDIP3r9WR_dugz|5ci-mbGSZXmo9IFKculJ3<`(vk^<jT$<Bp`1Rxj^0?|04h!rff;U~l93V7U5t`TRgTS00sjXXlOmZuiyGH|f$sZR>t_e<^=wZcRTf9F6WI-|W9zZR%UnQFB;V^`m>6ch?FFN6$94%r}cqGxJA@-Fx?&kE5sN<C*q+KU`m5()Q1^)$@^FSlW7(?iS|C_jbx#=EB9Rr>*%Nb8quBb(RWW_7~sYnN2^;rKH(jE4?!Fa&y$VKTO`KCGVXoYtK{L3zsV=r}HQ89_pp_)ZTtHdR*JN)9%dAn`b*uW=i?_J4#Mh`h~@V*3PheTF>tthmW$`r|pBRn#ye(_fGbYUKZZ1CPSO=c9Wgn-A?P_qbJ$>b>m6v=wv2YTrF&;GtJu4`Y>u7zkbzROAYEv=NtJ#WB9Ioa{p*YUpfxA2P0|w=zea$TMq9$>?$wb-F<RW9iOjA^`(28hBEi?-iwpW@|Lle8|n8K(x)wRFIt>!OGlUai^sW>XR-X=S?K8O&Uq?xcKrC{`J0!Cl`U=Q*w|U#TFS=iwexCn_w?Z8mENy!sG-J2=wRp3V6?gN?sV38v^#t7a51ym4)@+|JxmyPFW$_mb1VI4Y4yqTm1LvQd3^e$k$by#yz}VYi^bvE=G{lS9^P1fv)8$}vwi$9QBQ9THAN|e@611LCJI&c&At6*?9OWJoxb)u@$lhrUmNYE4%XDghqLb*vkRG-i(YiAu)d<syxnbXtgbGauQQ7;7ZM9iZDFsVZgjfF-GkJ{+;gMbSm=aKYqi*;+QM*kEx$5&@$kj#!S-@23_E?Pa1xu{Zkgf!?&HPU%Ti}$_CR0kYyHkl_SHpw{>9GrYcqAvyti#W>w<KZI!_<B9-Wk*&F-8GmtO82E$+>ymvd*!&Aay&_Ad`o^K<F(i|SGrR!4HS^)6j&6!z}lFJ(9Pj~Ctzjd$~nS2M$>>r3SqTaU`k7oCmRqlX8rT)1jfqYLA1=ILlud;3N;!kq(iaCBPGYHQKm){3-ocja9YG(o){d(!Bqn=9p+)ZVkRVexW4v^{R=;l29ZclUQd_c&@L-YsXhb6aP<PU7D3v$vP!+}ie1D)jR0+s4`Q<0t*{(<kMV`RzB+X1LcBKkR)V{|^EFd@%V{*BgSWpXs&HP_L^sz0*<6{>Z55W=M)pis4^(g>p+5SlM>3DP#)cVXNO0j&!4^cZ62k9QKW|&^Co3l<W68V?i5138OvKhV2twV1R{2-w;NoZiMxE8yXd`Q8VIhYiNu;t>d5eKxyhjwLPGQL(&>=Q|Z9Jq4prcpVh9`)0(;wY4--Bp^9%+Lpzn+5cRW~J{SsV`lsFR#SMKpGJ3+rrMsM=o1@N9@xI1k4E3|2(C%4d59)G5==GtD+3uM`tyj}SSVuO8h7_;$dw|H1?yuF=hP9Sr=n+%bj9M#Xh_62U(44)Ce}$rlvG8i;ZR{`_UZ_a;>OX%g$_Q2pY@dL|*(eU%y<v!6L@=e$AQY3Nih^~$V<uFvO4H~D;OItM>$J~xwSU@EVW4`o-#_wEPQ71KdR0Ie)QEt<TQ@`Jopv=+*K7Um02(nvRn62NE=H;=OZc*`hjQy`DVHdew(@0lePt=4_vp1GNfBN`91R=c6;YBkQ$5iQ6Q)pE^q`=t89<(j=z%6E!iOKIlUIdoBAMRGXH)5-s($!^=pmGzIWqJRgN4%#<B1T}9)@F;2qSA*ETyv>YI?trFP7DEdIw;Z(`pW&Z{&A#DN*v_#0NtOdI<;{bhMh)z8o+XaSqBDUX)34%t!HToh#E;{93K0*N#l(u(H-|n?SQo_~|G6nhOcRRL?%D!|_1Bqlh?R;%x|?e%yxMYaBfRG|%S?NU_6cU4qHU?-rA3wU8)g$_MIJI)PKYlPQ%lxlJ(uMsL)u0^tNKmo=?69BCcJLSY10Qiv4~UVp4S@<rqXxqeR%z!+n&MK;w!RXak)2<Xy#L*cYNY!Of8kgS$DLyZ7Tp+6edHR7iI!C2@wh^q?G*n{Z9NN{Tw0u=@|xDfEHp{|JO{Y0{?7I$;y%uZV6xJ)Jr<=tXBB?inu$DB4uPFE3aFwlE-3vp|_Ax{&#xx`)~lTEB=(*dn8VznS$h(vG<pcrTW$|4)2O7O)v{UZnVOw`W?eS_8n5UXmowE3lnV!Wl9E#y_~Ib>mN!jB>?{jA<@>aYVzQy?27p!=?VEnqea!C-@_h_Fn*rZh{<RmVf!6yqJe7tr@hAA=a^HADf!p2;>=L+ds5P)v?3N>U(aJTwkn$gdEWAXdrdsG=xd(B37POlPxdzKF|m1E$=&ChrP~5f>Nwy+*q^GBn(?h3?24GVVbugkze~?V6w&)%MWPKq?vZ&0(0<!~hHHCVbNgm1Y}oGX!D9nt#uF0g!5~(Nub{N!ln3V1pi}r*NhxXuZ0<)*{d^V`$=n4vBYRd5T_)LQAlZ)^*^)J!+(fjcDdDGSCcD5Aohek!rs_#wHz-MNh_Z30Pmvo`mfNuR!>t^~NEmNJ15JT34rP!Z6aVL;sT-n<5i)un)Zm^c{GeDOLPk*3W=Gl?@Fj()|&7NTg>COH!0r0A~o6_wWx5yRKjdJ23x-)=~8ntuujm7_CI$#VBYeaCa}nVJR@Mpl!n|4_-U4KVctK?0ql->>4Cf+GuMLE>iblX8S<D3NC3kEL(Vjfwqy)hoDujKw%D*7y07$192z20Ba~&&KFA&kd=E?ac$TgC{Q4Rl&6}bMx%We64^&2tf45a*=?MHz)lUwP|MQ@ys_mGfam$jc`b^Q)EbI<^b{hFgW`lGv<32!a(+9VWAbv(3CyA*`Yt_=%`?_lHbx+JfrOP_&*!(*WIk6;WO5~SGgIE$U012Jw<Qo1+l{eX8pV-ZCq(kzVGme!Hs8`?Wqyg)i#h6s-1CgGS{-C!aY84as<Eh}>zQ0K4+3Mloc8uNu=-b^G?1@>zuA0C79Z_oqYIBkSwyOvG7mTd$Rcby4erL2MkU9<^1J22Zn;Eaa59@t<kCJt7|03Ou5uxA^owy=o`>RIx>(BOb7F<{lnJZa0Mrs))lRf_N2_*VfF$Hu1p@Zp$=-|Ir1ix%!$}>HizKSEC~TShbpTA)7NmCBhGJD0Cw#+}K46KphhvqH*VgMGodAY8e6~T@Rs5~7eaV&^xKMzr&xX%?%r;w(o*H6I!QB&v^vD|%@GY?w&lW1aiPT5kff-_DWJVLJX4cwm#t7whuMS(LGB3$DtZS}D;`(CHCZ98fgWS68mH~=sRGyUSdgY`#F0hbCgxCay3kbWAD3wH{%^={1)25`tcuGhl=>w}18Sdm$X%WO1abq_Nuc-{QP%LK>AftF^tMTCn)~wn>I?au85!{(z(yLsqVe10he5O>ro(L<KAAaCN7h*=W)~1CixglEIP!`N_Z>XKg!Vc$OOdRMR>0FY7?bw7ZbWOOR`9paI<Wu-ZHDR_c1tx7u(<}x^U8<=y^w5ddM&K4@PUL#KHk53Wj5#zyM^b!JUz3O@9myJTr3esk7Zljb>w$)bqd`ZPQL2?qeJI%t9Qq(LxRvYS^#&2p5cMW#Fcu`#7*C`dt0CV~?MlSOZc(ez>1$|vlj0`|{l2jhm10sI@*M22*=YB0hd;5K_X`{V<^*+(K|PX72!IazNh@Dl%Z+XnRgfGb)(;s#l;K|}0<wkO0hm9t&uB0d%tB6rNU8=)xLpHX{7^*oAr@5&nL;|7$srpkl@pukm|EC6C}om~teV(OWy(nO#mr8kc%YUO#m#hCO+{lC<0w7Ps;S-Nc4{3I&rF#FiXzBS@F|rjC)CmoNEGl;OanL$9_ti}>5Xh=a|;Uwa->wIl+441cA%#Bw-URha%L}`OXJ!TGZ>FO4S(gz>Q1^`%p_rfL&M2THj^mh2FR8vhzPKe!b2i!x1Gwr$Yt}1lvM;fY5gu0vYB$pdV*nOa%kyrGc5LTFOkipJZ+WJMHp4tdYee?C34AhN=@fV=^fO@9q&q8iGun(;e3VWuY3`(luYCR*A`qZ4Oov3PUNfL?&nTsvq*ihW(|6g$dxlmgtd`QpdCWpj5#0XoevAnheZ!`js}`?n@I4@I@;}!RIIn<7kTv{ohU*tvuXT<!*;NUyTTrq&q*0QZS^}kES|b<;69-B`aPsp0hAWd&}3SWn$XZsQKvo8du=rKpu`4S6R;zvP2A9rM{R>dAPxmKur?^@utP<K|N85Xg&iFIPK^D=!SBqo-#GT21@?Q<*;_7DC)TqG>2Z`;#D!=MNA<DVu0ti-U~ozHjWIo19TP8MxtgOYa8Rlpsr5UfZqM|xbga?QQE+2@waa(<wL|)0wHDEjLn2IlK8H(Hl!W_JRaff|Z+Ozh2&jl4g_7A%a&@;7sT=)4PlJtW*e8pIQ|l&MRZv$QWhPj6wn(L>&5Sg`kq5W~t~xN<ur-e-l@%KcjMAPnZ28~fxym`nYJj<zbl?AdVDyK5m>5EBF`oslxT&ITO?2w-jS))48qh&qGZa#B{CxtQ9Y2sf#MWO)tOOVs{584BAkrK4+Q%b3^k+aqNn>QvXkaCmGQfn?iP99&I-St%09hgz2r5H$tZRlDX|y{XAP;djx19r?3gECA3#wzjdVRH_4PlCpHN{erPRIZ(vi<?-g4n=?NMIwo>sgo;;#@&iE$<&D(wra+;PK_%k|o#R)_QbDVSI_{x4}|YDVc$-B(nkgNO~XkfFv-r^=v-54eFH3md_8SHjDY)g35Vgi4A3QNtEK(0kTZ4;#Q_q&KHU9HkVx1TFJ*}ww5Mrb2CvWO@tEfms`!Xhu7SyyBo0YdKmchKBFL40EIORvf2oj$C#)^)rKmtm7@qM{Sq+FKcyq~&cYCQ0Nnub04llFW7J_78L)Z5Eq@1gavK>;h`6|v+%(7uMf_8`293M`>7u0K30D4K2L%%{Pq4ErE-YNzPX|>v7XViKOZV0Ze-FeK=sD}=&DgYsl|9($(<P88fJ35W^B3))B`{$ttN~^V2{8A7f(-%yXrnb1tx{u!@kvKB#dH>yLs|tHF#iGuSD_^oq^H{NZ8J&)y2B06VNd|Hj>r}uE+z+BO#ko|K78U%CK>g)-0k9KWX1^GC+U-Gr{PQAW~MePxDKaO2ssIplS`J+g0DbWO=Xfu2KZX?u{kep<Njmf5TQ<`9)JXI8xxqSA7Gc+TZ}~K<AUBDv<;H$tbM+xo2J0!0u$}Sqb@WKBH&<D?LbL7xHtj+qBV6>7B-h;fwP!4o=d=x^iI3ku7dE`S`sXrtqG>C3ssN?j&#GNW0M4P-4ICA*nkOebhb4f^g&29(Om;}LKz9c>NHS=G1MDcZOFPa!Jfbk&}gR%+<!nU&uqYI;-+XkuIOsLb_1GV)+V<>fLWfb5+MhI$O`V|^6*`8A`uAk!lKhSS{)g~DQ&^x@R$My+89*fx^jW`G##V@9W8+(T0<>kIUs4O4YEt-RL4>aAP8<*LQ5Afinbc8ZJFu%-ENyKd1Bl;muzX%A+-Sc{YGQjk)ffTs%VNkyWtE{9QJXUq2*?1;z?*zk1-=CPzH!#5{;Y8U@Cdcn0EIJy{5MZLo)!VhK>USOGQf&A{Q;2WXeFMq8*CVA}SLV<QlMyQ4OaiCngr&g`jSea3;ntMBpUm5Dl`u5boaC;f_Xcv@`zZGN3+?9>|@w_5f&h*v3;(+|$w1B!c8TB+bJXO#x^R1oWQXQH}m7KQbi&KnjFyj)06#+9nhI>!kMaTIED(Tjf;|D6FV<4GGqQ>flzv{RV1d=UgMrQO{z0tfx%t==X*O;DVPRWiAk;RzE?jOr7<+1E}rkhQA4)reC@VTuU3Mz?3v6+T6AQn)o0vHV{~n9j(VKnbvX++^a2H!?n;rWFGYKpb+K-#5OTL%`O4zpo%jS5`oS^PdBfJOG8C_pP|ugP8~4WPZHB7*oJsws&)NDt?Fbz?KSa4V*-ka3#nZYLH!-I&cuR>OUzDv7{sKSAOei4KsOz&s+xMufEC))x)yWeXJ^Sgh{S3soh+uyfxOPag;k_{Rm0<XG42$|xMZElqA3hi(pzD>Z!}>yIA@TXhSu%@b<{c|lhJ6txCuMVOV+?9nzcYsNM?7DPnQC~wU*R|u9@v>yE8fW5aT|T<5qnGow|wMYU*jbj*f`@I4Op6p%#kyO;Djgm`oHStbpHJBN}+VQ-j6sSVy)a3ZpMnG>k)$9vq$(Fzht!^;_vQ3h1fyKK8<}&NoOFBx1b!{neLOpI`mr>i6&e_UaGszjgI#;MjmIfbfF5BJ`kI@AhrpFUHqXJ9$7e5OogLPBEQKXA0%gt$^URV9|yf%xwvR9ItR2Xu?j=!`Z|$K)&u84T<f|9mj`8<C4(SlpCnVd39Xts355SmMAfiqnI#ipDB3-Nq!=f$|N>(`BE8*Qq|&a&Ru<~6gkvuEhaMCjoXcRWb>oxviTIV0@)LoMZ#r7_L++(2IA{2wUXpH0t-Uo1V#QKO~K#oz}doY{P1WpHcJ~3+zQ^&bVaTr%mpGWbXcJ??O-j2@zjZ#Qr(HFJBzj|2i)@cF3H_Y!OLxAiaV&JN!PdLR*ZNf_92DO;_Pr0xCydT!QyZ?V9v9)Bj57`o?VoHN(t7r+E?(=y2^DP=f;h+&-;i>noY^|6HBB3oX+z`7HitbbS{l#Vz&eXKn!P6tgnFLsVf<F*P+(#*wTb74mI;gMPWmXla$8Zr*F9dn0kjfu@TDwu&MMOKeo9g1XQZ}hFTj9L9jQ~>R9E5TrtkH5H-15&Tnj}>j&yq5W66Qs%j4O8sUkon5Uox&{n}uN6Ki<cmbP2{zWMmHz>6aT4R%_{8%IW_)p7jHzfGf>vzYt7H6@0JjTlJ99X`bmS%Ws#&76NAS<>C(KiA8045T>+5{bjJb-RdlZ)RZ3>MRyK$a*2=0Qx`r(?IoZGiA$`RacKRM{|L6hp%MW~|6go^Fd3zRls1=#ZsBJ+9I8Wyu<NYdH0`S?hz^HwJN|1*;5iXaGG2oK#-Fbo&Fnr*_(1y5vD3+F=k+5;vDrhh5OJ)j8~*!VMN_zEg;S&aMvAE_vK@tFb6tYkZz&c<$*_49~rB=!Sk+V^nuwR+E_mjMX;wCX4w}=|v*D4I<R@-As`W`nHDV>l@yTS!P-gxnaerG>?emOs=%M0X!XfyD0p*Hs$MkVI6;a&Azs)z3`W@6;JO)kg@^_il&;~{t;5ReUi^TX=EVh<BnP%)sE^_wlJAeU4zB$a0#%y0VdU8kAkfU>Vksb=(L+H;tuE=;B*R*1-Hu)01fcx3F!5xr`5Fq?KU>k1HD4$h8!`zn@gk;g)-=V6JmP-+hULq9aBLm9^yIJ$kdVTpnV_Ru3?d<gO?sCgE7l3D%i{1+KsqB9;sONCgjxqZq1U0?OqLbH(jseo}GrF-IkxxM%dMdM!N=8ztWzpRXa@GWY?XbIvp-wqjRH`<r*E^*6$sM`3}V%R~pQH2{hK*DE5xxw`!EBLrhd5Tc+QpbDL%-2B6zCs?}k0Y@eWx&G$*K&E{4%3|92I7P&_%=!1<2+ATmS^ms<1O2}_LG<*(fs5OvM7-9~H``OaI{4!GrR0kbK4lxc7vZ!~o0b2Y_6ng61qDUT2T@ZqXs5XS8Ffh1cIf&B8cm&*H$YF{FZW=t%1z6SnTASOodz~>|>(M|?1oZ>Ow9Xj1-Ovp*&)W<^81$ispqn%rFxwHY(c6-_rg(TxoZh%~Tnh5`gr@taYESEZRqs#kO-v0UbTef4=Niv&>Cm>qSIAZ0Wrw$=$V`roBAsmpDjtf6uXdR$Q)!~DS!ZdRbeK5-;yIBBJmk15Y_^A6qbeO=ptENikHzq=7D;z7Q==Y?u+~B!WudAA(dh!(1)L`-S`F_R;OdFE&qQia6}kXg!e~IIwI*yy4fNtR^lpEs%VZ^zs|M6<$^4QS+K&Q{Oz=nt2BKVuT@V&Q@P}3L;Rkr^AzZkTdl?rl3fi#6n~`XFU4Hn1kB&v&rWm(wZ9ZJ&SBj`U^-p4RObanIxwMIElR6HS>0iIcrej2L-upvb<3J9MPwr~{>KjEHwDCX~?HdnxLGyt)H!?bSW<qyABF7^bnT1EBd_{Z!djgWE!rPLqn@UP(J{pzhqcP|`7MS9h2u-6iE}ra&72Z*qzhr2#-nyrmdD5%g9_yT6KC=VBrFS;%al#HO$}o|WkzU+lpqEcTQc;3R@-|sXDEQ$A!Go@L`c-8z?R-|<!zI2BubAj&z2;2aSJy8`P6jAG0Q5Y-HgP7Y6~d&|H)o@PyvK85hLBu`Ws1Uo5wc~={0&lEYCVzM#tjKY*PTQ;xfK`0Ip6w_0w>jg%@@YWB(;!r3}#c7cUECOltdvjCqCd6%?C`zgF&O`vjxCQfJzBuby&NEXaFL&V9DXNy$dlx+y!<RZ7{I*um^o;d>F>isajXOl-Uo3lxW<;U~5%E3sE@+bU)>$pTSimy2_|1xIjdUliEl_4Gg}Be%EA?j(~CdH6pyPj7XAf8pGxBK!<4uT?_Uc-SD~j&|y7npNhdLV9kisI(-uu%i2i6FAQE4t>wC5pG>NN9fIozH&9&50Up>*yV3NU#e;}hYw2CQ@zNR&2PX5k7(`lqbBIL2EUs*2M}Rdz%Rti+#|vqKHaY-RcZ=DeEzN4!Gu2;d>RO6t6-c*<wJ>$UX{>xA`BkmkPNsLzmIE{S#-#%Xj`CWSwSl+)RKDIri?gxlJo`(Ua&r*qtr~Ad*d`jPdM{SBX{V0c^b^cK+|0sSz}|rb@{8%$AV{t>TlVD>ADQ#kea7+|-F>+G13QG^wr&tk*iAWl`8+YPK6tP1p;Nx%^VyH|`ll05e*Ehk<c!E@SVOtmXyBWWICC)5ovD{+wq|x_N-w>j+4iD{-A;5N%+)MlZXOpU>t>-C0kR2;(j5FXN4@e;tk(1hhSN$inIK=X_Y&CuOp)#P7odnVfJ5S|2S~u!Bfbxc9v_WMmsAjwy=$z8`v>b&DZRNv7Ik3f;vBmsgG^!VY+4!AL8yqwDQQ}HTZ;DAnJzn04`rtGutEC7nUgk3-Lk4oheBguN>d<^yFx;!8(L#1@STi{xu(E2G)oxfD}b)Q;dDlX0((P_^YDVlk7>Y~_qKo~Ecu)$-);(<5L(&^%G^3eF9}aPQ>di1;JjD7834saE!I|SQw#KppOnKd*|L2w$q#yN3ABE3nN=cJBM>2|GU15ZuSHM@d1GRnsaQS`c-tF5l+bc7WHJK6T61a>xYpMMm_pFaYPtY?C=9F~n&ZriskmMu4ixby+yt6+v63trRrC}zg+>6I9Sz8*%G#Uol6?q-QrZ;b4<4}3reouKfCum(t2(@+E+<0I1^;y@qR-GJ7lHyb5OyzI>!PH?unS#f`ihLkKBeIhg<RW_TOc}b5AfPbkgRmltMh)((O!3XTM>y3+PiJWCR@gCf%d$ua=D-~`OaU(3!5;Z@Nz+u7NAYA70-#17QVC=Q0&ZE3|DqA<N%SDO_S`PFv;hxXS?$`{Pdho>|S*c*Hx3PFyA7ux`47=+$NRQu-l<mcvG0uziNA)#hpF%LA1*gNGf`R{1~vh`R!B&!$JKHs&~^nY9S9Uxcd9xg61hu&t{$>e=6Lwsc`ZBUz<+0CjN*v`$w~Oy{grY@V(>8a^P8;%MWPKA{NG&o8@k3dloXzuOb(9+SNSz|8$2?Ut}kK+Q<fR?XE>TscdbV!8=#L%IG>*hE+R=#LX5)sG!Y*1V6{_!8JG+V+*hh@-2pM3v3gZy2D(0OcYc2xTUOEJ!F*4nvhT`<jqX<w;c+fkSZoN%Kn?&4aKXM!@CDCol}e~xP&cpLU0GMZVOa7y>TdlZVw|0FqLpg-ctx~dXfA~+#<k)(1T@(@sNC-<{f5jTm4$&U9qd(&>@amUQ6-PqK(pWViS~{G>d;w$Y*k8wVbyvCrpmSllh&UOj-QPabCxLZ=2C_Kky=#mT-OqzzF(!x5?LAc8gWEFv0G8qf8-Kyg{7K=?_H@I#g!6^SlkTr8&R^=}xwDw{jb=?<rUKuhW&{e$9B&reABjKAs>7P5N1o0V8N~yf?*Uy3B~a-6~0)syZj41q!(dygiJ^7NV5^g*gyMe(0iZ7{DoM9mRc3UHajA7_;3>@I3_g;M`LL&wU1tL`@K>F!U14JJ2HDP^q=Nmk!8uj$Xf0FY|cr6HmXxf{<<fNk40wLw}&lzsNZPtqSfhiZ{=4=-n2B|A|*WdH>PXAFjT<`h>Y-3Rhpi!>_MCz54ytXTsHQuYLhvK7IeS_x}@KL5Vft>W}yg|9uk+3Rj=Q8+iWg{oh=D29OZe7w^9zT>X;ydcOaL?{>(o2533qIqEV#=k{&wDay)HR0ez$)9fnx<a=-+8x}cq)T3o%V+9s{S8Ml73~h!7U>LcgBQTFIq2AV-Jy0jwHFgDMJzw4uFcv11WU;#fbh+bE8}`eAfroO=?%92kJ;zCc=scm7Cm!^o(=lFl`rb=Zn#<y$hrmxK<Dxu!h%5Ds1(P@nIB}dv6cyRj2a5RNhdtpxe=B&X++H__7BN>8_dU?KvmcJ)PH~wy1;VVrUzZ5|;(yS^W-EBl2XPi81O;B!`UmYgzMNmEEiJ7?SNTgcv7&|rpDr%hD`q8u7l(Yd?+USnh;a2?MvZ=2{ngc%0ucEh&?gyQ0>KOa^;iFw5EDXB^c(R0^Y>pDV)G6>T>U!$_&SvT6u$7)@Cm{E!~1W%|2qN6AL@Pf{;vVvA8lV}`t(O?>Qj6r;QE4RSbu-K`Vv+cfE3>U6%>FL@dYkQflA}|k8sUh{TlxLI<838djWe4-G|!WgrA=PgtWC$(v2_}`43nF)QjIj&#>3Cv)BZ^`W&I)_g^!>(9Cb(_jj=YU<khxsNGMo-S=M;pu`vO@IRr+ud`vW4$aQ8;jmsa20;*L)SpqIFA4Kg@$;?s-xRKX1!aE0K0?>O4_zlV^cjBp19n{&;N6!nO}~aNd>1F+6UK-*`%0Gf!q`2|e&8i_>`UnV-%$D2Zg<VsvcM4e2jqR901OK-OvPt36`xIbql<Xht1n>~IMmOtK8pm230EIu8Jow#-@g+O=};dV_!VFXhD^BlObEUIHbDRM>UZcPFGOP!W5>UtE&|rjbN;u8dn6wqPwbuWn){)ai|Is}uE?TSV5yugz7`yxZO<?JD`5CzMv{L-$be(u1ccxB-(>Ru-T6Cc?4$SJMo2V^*!VZze<afj?D<FVzrW!t!omtoKN1@f2&?lgsPU;apMPY;%nAEj@bYs6_YHW3So%HAH2nJ|Bcu1<irmO9&`!a*f+vC<eC>@o9VZ*9Ej)a<yu5l18{w=I`AB67V-b?k7_!K#f4ut1)sL_K_38&#e}DBaS3kY_&ei{h-~SIV5TH7?mGa9Dzw=X&O$><DugQ!K0Zn|BO;+JY!cI0EU9i^JuZWa>3$WOVqeTNvec@nKeBxo7Q7$||M)nq?^W3jqO=XH0i`1alTwZlh#FwrT54az+&-@^!_i$Y$(<+b=$a$h194zB5BVA$%f+7&>{T)#<@H^sKj-Gj%Kw-h!s-`zFy#-*D-zFma!UNd9c<2riKSP?fz?c=r-}-N9O#y=?RyJ9!_~Uc7Iyl?o(h{&O?+jUbnQz_tJLWIq?AKja;{iPSTUTFL``Bm1N`HYb9IV(Ip(YRV`vEXfF;1>{)8@cl?w`0e5g_o#G43lQd$}4+4{=FPlRyyYPFfP<7^bYr<BSb-LybkFQ5f!^KNxYxxmpWD#Hs^SnEfP*(+5Iw3P3H$)^lQ$9zBJ3GC4NqAWO#=K*-!7&(vJGn9oA_75;|#VB?D*0P7%+V-P6v2ZloP-a;i5<;vC^W^jxF)-ACBVtzJ*_ZMG?@hB`3*?zT^+1$c+t1IY|w69w4K}Rm8@k7iD3>1?xw3{ekhgMfHTsV#~wmiS;Et=RZ=PfHS3}*u*>n(O8FZg`G@>*cm_0(Kq2f6^uE0aiti@V_@UiE`zdlyBQFW!H9!ts#kNGkap$#MepGH6$wC^LKsu+URcQfdA%Xm@2>m`7S5aqD-m62B$hMkf7Ao6N(N<><rc%Hp59$;{Ix^Mk8@zWNz#H2)4;%@3~rk?~1k&f0x`2K&xWu6}s+AHvm-U<1N3|9bU(s`DMFELf}=Hl<(U*33k{Pq<WO%fVk>{Sq+tH^dhJV@}~8Z?O)AzDWmAvZ2dev`3z|2WXv?`JujSwg4#|885xkdppz`x-jc#tZ{_<G5TE$fQ%Id4j)}we2w---N6_Q7`=f!0R2nD13Y0+_Z1wBHN)tiqUG;+qz|dov8O@vh^`w)JdUYW2R8A7-}XkoQ=d`~%8iD)RlL;38%G&>XH2Ovz<Bumn$AKoS`~*l5Tjng%U08BcG^Aky{E2P6|?;&dM4M3IxPS+k{9uQ=d{M#10BIw)w|jeutxsUTc_3@jOpOVw2JroM%Pmc=6%rV4;_;>7~>1VwlhBi@N9)#lqPb>p)F2Sq&4^+=^oYF28KdLuQBF9LUs_fe?(7Chf3gCet2F%2H*<M6Bx>vmqKrd1LiEc20}w-2c7wqK|PX!#kgT;UG&z;<Tlbp5?w6wgXdD0+hu~BibG9|-p<ruSPvQ+%|RfUojp3$jHWOC*z?3BHhh!9zf*nNEXDze3=$K!*l+1Dz7&mCY{-`$H!to;6Xc<Hlef*}bd^pQu038Q=NG*5h~@)ye+hN8Zndt(8!T?I(+Nhdg8bodcfo4G4vQ|I7WcRs?Oq+b6{?NQVZSR+0viN`>ttoD5RQdOwcutcetPI^Vh@=BM`&lZHBOSrk97hBWLve6C~qlt_nhSD5`io7ArdW%fo*+>&@nG{(Ena?8c}V7{PgBr;;`MEKh6nn&;P;IAPn3Z9iGV0MR5%J4r|+bKg~+XR)jr#5Xjh7O*F82N6EDDk4PjX;{h{DtT9^$Zb58|{tN{IpCZY3J~E%X`|iS*cL1)BH;lr@sB;uP-2Jx<!)igfxLl)aSSp)tFT(=aEmzve7k5fi3fPCq?OvnrN*CmuWi)2`rq*Sao|L;42JQX__}R0H*E!q!T{KSsOObiow-1@3Jqe@%S>E|F9qdi_)8tcYSzDbm3xd;cV3yGCV;MVw>=$gtWPuygEHUXzehd=aEy?GX5uA*KouMjjAlB^?`vk$BDm)l*oGY*^CqbT?y4F!R>$fTg=Y?S%U`K8A^Nxa*r+71RZ1THfpl7%cif40ZNwOq+)Hcz|^lNT2eg6@fwU}=CS4=BJH8^Mh*W_*q>H|Mjy!wu15ar9vtMrN}xbjFwG|nlZ5NUtpGH&E$d$55g@iroHr#z2O52A>(#G-ee(~R+GpKC(9ex}&&9uGh>X!Yx+3b<Dt%eJncd4p7W8^HUT^)qX6P;+q)CuY$&-1Poqv|Ex*gsfA45Xc&ZKR<`5u?)!9g@?RmTI{|d-*Iq(b0#G~e<dGxIy)2*!(t`w*pcxQqi;7-MES5HU0&ZfiVt=zYM0#O<2MF!6}md5TAIv`IEPi|L`$g<Z-#+s&IpWqW(=)u>H?Bbu0Pzs+qG#t_6)JI(#>!Uy{5P@$Fe=SyC>bMf7meq$NN-daoSmT#ElG(3y9%tTx5htzMgvq)7oI}A__?65B`K6vUL|B$&{l97n^C+WO~P-%i<}QSL?ttZ!x*#S4<je1T#}1&GoBqLI@Ti;L%!h*ixL5?5(iLIV|gygq*W>XmRCAAdYRkdhc}H61@XuaLl(-yw+tkjn^8Ph8((G&+#yEVD*^Ckh-DA#rz93ncu|_;a|PqmTSfJC8sCgasF}1NAA@wPA)0i*PVNM=QgD0m#jRXdC-iZq4i9R5^vrHD(A}7Wk;@q3p=bEZ%n!jI3A<J71%h#>|kUKonK7cSnd6c+pn(aM6f?dE&i3eZto-9Pooy>6H)`Mb$;IE6mnO%`k7_k6UYuIxXpreP)6d`cP%;0bxO&1^Ds(XboTX2G9G^lBKNPM)VBq1bgB2>w9df!`Q9I(!f&wsh^t%nIpZHlO8L@f9dinO+@4RTnODD}t@9U-6CNHWQN}}M3&Lk4OX3N{r(7VJSiNlvJoEuK;IGly*OzF4=l&PgQ3g?g0EK>a^=0HM7)6<%$Hy+ir+rEcji9*Z6_EZr45G1UZE?P?$NpS{=z_x_`U4PVeth-wtA7x#etPwNZt?u7aP=_=_5bVYd+_2P=+(zi@*jXseth+P0bcw&JpSzJJMiCs&^svm9Rl$89PUryU$m%xOvY4f0qT7p0RDh%tN;7z=U&U}_pg5B41B$n-bpBjq8-|ijwtZNgZ;)EQ^PLZ9%c3)S=8!%TA37YL$t$mGGAYEAsqmhZgXu1!?MxkXb^SCgO=W!gLbEbfmA41htpBeJ|YV28fAjB8rDG$<v;BhASP`y@CHa580&<uvhFeW7w&m$!rrPo+r~gMCf=9A4L8`35pc9tkq=RNu%a%}ZS@m&tJ#dy%o9;&){d#)oxH~$$n-8;bKqumW#T?;fy3afAiF#JbA2I?y~TwZ%Vy!bH0?y+RGippUa1|A@0Zg!z!M}Wj1ov@fEpoJGjy;1jG0ZPKqE=uVW>mN_-D&Y1N960G!{>?9o`uNKaKGETA;hf6i*Xti*a>Va%0ss+f%UWrikmO_ba2;v)XJNyBfV2$3f@1Ume9x@0-7&L)p)nLyhBv*TSl+gHEhM54<L(byzW;chhOxZC!GILg@k1^4>_)Ue7_+ClZ7G6kMk?a2F0f7Y9CPw0lh#FScJ^%163BKz^!?Iv9pDaAq)J1F7_UX2bX!I%20C@!7DjM*NnG=Es&#8M+R93NtCr!RyGIe!CYkDGX-_!39{Apdsu9#nriOYd>{5c5(U<Z)dLpfX*GHqEO>@9AiK!{?{G^sUcK%G6Afv{d|30D%M7;3d4rGs-0QI4x=1k2qgbb#mV_6{tT6AIg`;P)UqgBVcxC2fcg}(P|iV>(%E#f>@zfem(MG@jiD&UoWa;^e@5I7R_2-|9R7w&e0KNZPI9nXKa9tgD$-XpoiQBnarF%wt%uP{a1C5)h-*y$;>Njt9Aa>!3Mtjo6i<slZ5pYM*!?#2U1!0}zRD{3h}@<u)qp#ekb@^z1uO*9x%O0*>gpHV4bwFR)0C+j%w9y;KFN;JxA(P`AP6X(E-pt#g8`bZuP1-&4+XyGkb=Qu>a!4+tC*X68rZe%&QjrtOKup9TPCd_w}qXoH|oN^Sfj<~Dj|~}zlo_gb>n&qVpXdvp}?MFq7|!~(tTeWFfv@?lJHiTUf(^~ASk{5kiguCW0sa20Zb%b-{jee;^|u4llX%N;*AJq@NJpiwY;uJiccM=SksL@MljbbLpp}7yTwE^YhcJpQg#Lmm?a)(kS;7cFyeS&_v0q}>g|^|S46p!j>GMg+(>`T4`g7m{+_v|PJBpF8yCpK?{ir5Uyv^Q_3O<ew=ia3*K^C3bM;RyCh`m6g7HO+WQ5xVR6*;`r*tGp%t#h4&`W;Wgn#$}pK%O-!{RPpIt>wAFh)5+yhi?YOJ43zT=Kf5;Ojw>CI8v-ms`5MM*g~;sD*}A5UzZ(w4KTfJ1{NXo1Lz>-(dkzf>m58jzA8fl$W{L?{C4>n*DxVq22_j$G2h@_o=o=-JZHf*vx$wPuelp-a)z~R;Kh!xUbM$MJ6b)*dJ@w!iM{;qjX0Pxe5j}M8rUakcVNj)5U>nv>Q7#H?9~2XGJ)*`2#GpkD03~DvjjXH}QaGsv~0<pa6a5yjtLNaQzbX$P2rT%Rs63c1jo*+3~#p+sun~;_%IywPbnDuE@U?727{BQ(nOB3(Nm5IB1K{%m*E_tb9Mc#a#`OcpSQ)|8(>(Zu(nVSYD`Vv4ua^^tVW+KXkE28{S7(KSS%?cR;TG;nn|r^*#9Y{i`2c{ru|Rn0fHuu6_b9e{%IP>>2+==DqKcq3++|zweS!@86ge?_;w5{QwPuvOrHigs1O8k)JRtqj2^2)GFEjCNqKITfJH_^C$+Une%?3cuFdh<}HV`{X)Kkc|e~hygnMflL1@rz>S1u%YvQRIAq>rcn>RJdt%zQ@3xjW$=vWf;bomE#o2D)Dz;KHx!Aib_JJ)MCz{iW$%0I?P$f<;ZBY*L<Gu{IYRy_AJv(Kq2&{)}NAMqRzyBuk1N)8^w~PE9mgry8r7FvU<N|$5PmPpY?@c*08qTh*UCt%TucZ$C=<)73sff48EjFJN^Z|LUzp};8Elxb84*?bVNI*K7(jcneD&9)e&CucXGyoDP>gB4f@7-L3#DhSYfX{$~j{G>Be7ptr22mIdW%a0zLym4q4loh3)ou(!F==f)wX_d~R-GIA@4Bs^^ahdBmTu^w08$8#qY_WBI*3M^!%$R`U88+4h=ya*EG#6<&fU{`Q|1ov)YOrS9J=S>HfIu^^M#6xPOlCit^TPZwtGDtO7rY&x`KD`$12QGzZXowhm(T_JH$tY&3~Xl#@@K6i0v5o$&qbqjHVkF`bn>|SWalw=?VJTP&ZhNR7YYwCVUCZ4oR8x(3^`QJvrt?pji%X-<GG3^sz5}uQw%%(LY5#OtKTY)lyb?GjyFYh1JFshYo26muU#y-X{{9BwO!nk!>y}#a;aZg}5FH5hJ8>d=6RFu{2FJo-$d71WC7RQ^I;@bJS^Frf#1DM|3bw*-z|JCCH$K9)W>j=_!h1%F+emU#aKsQd2_RTkIT@$mv=S7CBGNOOENT49|g7>NGL>FZooa^-@V4I(lSvbh<R1@|$T=9%BdZ_bU96Zs6^K2CUSz=BofGC>Nyyy&T!0C6fwFd*`~z-r<xYDw8aE!TOF$&=e%Z+roG3=#CtGQ@~R|{(k4?20T`%+c$&R3@0dZ4F0;IyJShRr&?({eR&5{K?w&jiysXFqbp6Y7@SxEI-{TP_?LhQUV%R$B;I)zk=!vFXEW|XUeb`JwM<KNlIt!d$s}*phUz43sTjXu?FG0Gx}@{a1Oty<M+MF+fh^&zbLjrE_HDt6KOK00nQ4M4KPc;(VP(2TZIeK0w}5#ZoC>vbd|sUzS+|3HB$v}pvVQ_drJE5}#vy?&0x|KA?}@Z;;xu5O#G%j($0%ex2H^|ai$mb~V#p_Q)HX5RIGzRai%h0tJ5A>IDA80dMH!xb>B4F5fz9}mvKsBap&Yt3jcVH%wosm_L;LW<v$<Glkoq0_T&c#4N3^X#D@+raSZ+2CW8uY$dKe9_R^~*x#`G}|L@}~1GDbAOTOz4SmN4q=Va<0`;~%J_Z5qv-qknYO#qJe0%E4N)J~I?PRxlaqQj~V%Xe2gkqf7GctiJZWFAn3e`AXpNbfK7@sGL+S!!o6~%{X3w(7Smr1qLp5>!`~Iz<<Y`-SVPzuP#EjbmqW5SRV;7u0+k(7+bg>95c84lQW!0U}TYBvzQ`Rzhj1JGWFmW(nu-N0e<*zxP4l$F4tF<{%niOk{#mU@8Q3nUj2-_Tcz^5>)8XLm|x#5l?A}65X$At;e&Mf0d0GN#rY(eWpW8K{v}Y}q3}kgT+ZY+|H5}O|G&O3`Db77ywzn=OtoIx&6Z1at5Ka~cXTsF%GLl(bZCrVGbAOIoAEJTvL@I1Eoek0s{+h@t=|nu&(I*Lh82~<F`qexq<fUl6^xW;UALR&1VX;Nr+tj#tR3QCel?kyvH_NCgRV?DNrvVD7<X~FMavl6U%?tEqT#>3#z9DfmL=J(2fIyH9u=5lt}^H4D8YTEK2VZpL+?2~gf2LsxHwkvyLjn0lQ<0wp2*5018>aUI69^ax`t;BpmJT0di~R{)}eDYNM>5#y}O-lpx@J{m<G<3a~j-vgrUejp1>E-L$7(uQ~OW^5e-1NRUtr%gTk-Th9<42nAPR27gTc?!d=z@ZbHM7-7B<4khC1;R}sTwor^?z-Ph&%nEFiP3tIyZyuo@tEuMSbQ0r}2^9>m@SIMv^4uL0V15=p?_0O~7|5Uci*CVuIp@#_`kERRDFnX(WbXsYLxZF1aWP*AlSQ~mYU0Aj_1w|@xL?Lu0qO;{?k6v)5)EkHXg^rP#$03*5@diudU39$fjOx0;DCGeX3EgnQe6BE_IGUF*YB&hTb=_<?djd(_pud?i9mQkzS5$B?v4OXjXGE!S`OKCA@n}FhYnw_`@`o=y!`t8k{V-N}G(QbJ<$PY<NWYkbFLB6s9xQRb0l13P#~dd>`HfCr8%k@O(iNI}`y;NmH8z!sXHuPqo_2Z37t64U#!^gI8$uDT4qTiYJ=eOUX`ei_4-{=G6Y382R1;X;&{VpxCoRa+h)wdrF@&KZ4@s>dpb>%-I+OX}WtKKtVc>!+dt6k8e)3j$&BLJq8pES`Ts_EKobMIIT1QYII>nqcWqL^p10l?npi;FLj=GW0ro;d(c`#615>dz{v*4sb@0w#p?CSNl2D+D4uj92n+(1AtV<vssEgOY`b$!ntzZ@4zD$&T23_MsxyQzXFNV0*~FS=xf(G_HXLH3ny_p6N}pvNRbhP2M0r9s1yXizf)ph6^rNSpR(qar(Jb*<a=d_HtfZpGTGgT4hd2aDoxs8<wdcfeA)srXK_o6su~aEVxeZM8e1n$RU)(^<#IK#(5+skTkJdZ34}74Bzb-M4ps?vD7!Y0M^<$f>#AObPSuVXn$lhB;4q2(Z>^4+g+W0Crc`dLe4p#rmB=k}A?$D)2a{0EJAd1l&01ZMZ7=V=4t1Z>Nl{kS}G*nLSLdiNS7ni|De!TfrO12IfFRRvF?iy?0xIZ@_@@BMnEMI$rsjfqFQ09T?xhd=QF-2d3|6*KF5ET1Vl;7w(SS1t<VY6BZ$!I|C!H2m0RB6^mf8L}l$nu>kWIrwaChJ@~8lnEkfS`P*9WRAER8L7(f=?5V~1rp^N^esWk>lvst)ixdV#Vl{djt<0XXf29Y#Q%V!oV;A6ir&fMjwr_amqYph(CK!M+8as#KyN$O~Mz4u8$T(BPa(aTdA>vPKAQ+(6x~b6pa9n&56wC0?B*b?Nf8Mj&5lHN{IBs+uZP@Gg&UK^Diz`Y^n%Iym@R#0s2Esb5__wmQ`(SAzKT3h~Y-pznj3rWQ_1iU&{dsFLX1G)kBi0*9<{5Rofqm>fu+AvG-$QMqhqygtBR#BDCgZ^tfqR}frT}UNAHg`e&<V1ifoVnIIq|Q3>@60N9=WV%E3-G?VA>v=-9;Y^)eg31=hQ+gIB7u4hD~jHrl-@FTWv(g0Od&xMyyjg`EVZgyw&B%QpKMz3B7?ukfBqVGRq2{D3$Wb49{W=4WlHgnXm*%jWLN+cePpz_Q0@p(iEs|T!@KsSxptiIFUbau(sL(C6TkPPG6!|t#-3zy=%4M-9WW*r-G8I0t>(c23Eq4$@y6ny)6EYE6dajLmTsH0f9%J_VF=%l!7D)=fMUgNM#S4f8bPb>pee-unIWDt6-@1`GuRx;fp<H$NT~=Oz23v8_e{BnxG7;6pA6vjS_+lS?FVK&c11eL54t+ylqeC8#v8x$bQQ%{6kqO$k^(MLUkz%-V7-YtCPrZW}OO%@rgb$MUcax!^B!+hDC_eEHRwGT62DYm4_b)NrbQNgkS=8NOy#NhcVc*PMO|i+xN-kCs9tY<2(Fu(-EQPj5SCJ;5#NVDd$#f$F8hQ@lkCOn`Rx1Pj?j1@luqESMnzuI*b(S4nQ>ta9LpniGa@-@3KNJ0oVG4Pg+!za4e4Nq}?0o$VK{HMs`3`j79}!Yn%>mY|xh8l;$R2r<XQ0l*pAS22GqXxhan)mdtC%!GLkgp*4zY9Ky|^e5@(lkisgvge!`cS(#(QnT;(wMkL=Z3WeaJ5Jx!9$~7Z#*?uR{<Twh%IR62hm}p6iH)WBiH4capCw&y-j{eTh9&4A^qYr~?*hc8TT)P1F(!UAY4aNOI`K<VTn1+0^$+5i+_qyGk0!xU1S}j9x_5kw%0M$FLZB7XTXtyj`u>+l^c27~qC&w80!5FV@x8pq#vp+~TorCcZ5t9m+0xg6vu>rtfyNf({O}Ox4>++V)!xZN%CeQP@slVc(=>5^ng;X(|;t4_jqDNKLg=no_TUq&YkE$NhQPn@w-T&{wV7?=O%woI#{*3<qF?{*CjPAVn=KFMa|L^I&<?H(&!o=MdV|xE1JccOr>JP8}0}sLkiZbv+EcNl#&jC7qLC4@9UH#mFa>i*(<=xbQ*AbYVo90sJsp0v=CuI|O!B*gz_k<GSUK3-IR$us32O$huGm1%T0(&$Cg^syF{}OC@2w67YcbJWlk{QzB8EPg1O_(n`VJ%Q=2w6;Evcx5sHa5G*w&UcX%{pA%p%Szkohr?U-C>_`vCFkW6b3<l)nG?#Mwp~3^MNa;!AfH7I!5N?S>&DK7!*mu7Lvtycbd$D1W#K!69Stq-xu~t1@V$Ka5`PmBw&`VgZ>P0&SwQwvDSoAn&xhRUTSqgBcV`0m&FbY5Dz3R(Ny$u+o>c9*zp-$^}uPjFvezj0&n6ZYausiayyvk6egIij6^0##tSFm`5+%a=*M;Q*2BwaDg#7{3P(OQG{vO|?T1>`WHmWMwe=dT7jaH_1d4_vE~Z-ioPl?m41$}Qh}4_|l+^rsDP7!4tn-jte6Ml?p7=|+4lMwPT1Xc&c@Xu2+aniN7}wAuoCJ_cFO0_v6t1HOWP=)cwJ+5BI#UOAa(~6-r<l~T(*b4xMBA(T$!97XgBGQ~G^sSP31%7y9X4fGdxtkI;c(2mYNMAGsUia&Fbg1y!@#^cosR8J7(w>d&M?wwH!&v>{6qGR(5k#BNo)88Ln4QqL@hU6gd}5v8;ms9p9WhyJ0atA7!-c|VGhS25sTrlDA)R(K2&Xj_A9dHw6hi!j#}TU9n6eI6;IpsVM~duOljw=qgC}zNQ6xf?;(AQ0-I8drH|M(X)99!U9QB7hgZM1eOdfTV=T{?Pk3~SPp^K*lh$}{G&%b)w)*fd6pG18NA)}0!eRS$T$-eNp8$|WGx~)z4dSpp1TNwcYM}ey7C5Y5Tz$bWYGR(FIVk!giWqOli>Fm+9d0Q>*H!qcguVZ_?5ZidxW&UOQb4EQPwsL}?I_TL+R?_)6i5IK7Bi32o;e)V$ErBTyMa;k@MZ~3xd~ey5MHPu*9UDSwj5=ztqk-oQNm)H=;_yxzJg>pPm<w3v9MG;W5lm{QXLlOmaf^;1v2&vFQ5JGe?#=BrPYN-V`b^jl@gapO8h61{7|Yx*QOr|<lXc$x(a4r`jS~pssDBLFC_Z0cu^ocQoOA1kT2BtK}7rxfS{TYfyBUnr!Z1K;jyOv32J|ru9I08%}B)jSlDb2w?<V2^f6Wc5q)CS{~4uDhv0?m6Ytbg*w+T~JmjH=AuD7G8bMXkJy;+j7k5N|PUrS8sYMz!vL;m<@|qLzQnt%kHYC-N291^Wt4S2J*Z-RQ^P8Q1RqNnIWLh?E$v3!*jzz%xH`zBFJ;W`J_9#}!u`x+rn@b2+mxW6y!V(EmIAX%96{yD<nn(5pDt1Wa6u;CC_sjBXt1O9)m7E22J2fUwLcM!cHRsT+A>pO!u^1<hMyFW|ZDttCkY6x{7K1|_+R-*$z}r1oA(J4pJVa&O8zES@IYu;1Az8fr@a^Q99{8?6(A3CnqfAe2ZW1cU+n75(l;DfC<qt!2Lt~T6+3tqiA(I4O_$mLb`x_dv6bH66Z5^LUvW^8R9&2L!d14X^PK-O8xLqJS?QVO>69~o+G3BM@fK|6IWpHd5V^9@)eLM9bIit9#sAPKx!T(1++X39+qQcZk-3lA$tX*)6l^A)SLQSCM8q}7aP#u$EK-Eo8%L5H_MGN*lH1k1ex4f4Sy4o1@+jhrYoP0aN)p3|x&q+6GwAm>i^tMiM^8lXDx?b;FPB(!5A^B*+xNr+aH+GI2Tt!1W?wz~o9cxtzlWPgotQ8@^nO;^n0ZInt)fBp*sv#;c0k(dsd-8<a5y9gZXB4}^qCsb7M@*}M{t%d0Lt^eeZIk7YmM$HZbK-;+ujc7XSxx{aSELBeb<nxZ5b`>iCE-*xv(|1i{zzfWdP8Mi=0}?fy6#-ZG+n7GNw#ojL~{oGh>Nn1U5aZH!_=2opT}8F%*o353QbvAO5IuSzKdbgJ^{FZqHBTiJZwBXryX3)lhus}AN|ad5)uAwt#kYA9tXOcH^i%do*?EnnJnYe%z}SX^38@;Z`A75#jlopbCFU|qPP36;sQ^IfN}rce~X1FVAFu&@;4}hCOz5-Ed?trAck0V<fQyM<)8h72kBN+)4)88|LP3YN{WCh+2Z@ZA}RI@j3>dfWWMH`xeA+kF<nY0ipec?BT*`27K_)Qmf#$JK>?*cpJ*4KlM3)FOvQNyNVO6MSQb_1NIzvkd>#3yzRm;pk)k6Du>~HffdxoJy+{^-w+j?__K&2fVWNU>+1WQI1YxT>zi<*Mj#Yvw`3cOx{|OlV0_S8B41#hR2n$hxl%emg`9kVKCqe!#(4bhnT<GrmznjFYK!~3}4S30*LFc?-d?~f#M?CBVQtBsGrVN%Z6w`42lBVPH_up_K@OP9s03-eob+E2{=1q!7iBr9?gyCI*=Y^uX7)$_<2s~~NgZ|o^tmHM%R={&2<M>#6-(d4ZVTL~CP0Y>`sy-JeFS;e-^Bm{1vk1x+_?Hqk3hp$KEGFD16rAjHmOSsZiJQaxgKh`NkzEEd9~h&a`QSC_9KR4ON__vfn8E_}alkqB^*1oXzo&uO`53J9%~qaM>^LF;X$Q#>@Xj)w+5@z+llrO2UVUF$nI13>%KtG@g99dEvI1C?<4N*b<dM!f`gF4Noq0&Fe#@dqNB+_(=E_37R$Hn4nJVTyWd$u|a+}$-fZRf`cn3|woL~i-6$-`79&A_w%a}LG(Z;iQyD#D4A><GR1K1<#qZ;1q6nH4EwkcqY{CZuluL-%l(9*S&v4zM7%%IP+RbyFX?3k0mm92VOGB%78t`>Zm)+hlEexht}JjoQjz~5F{IC^p3kk74;=N%ax+8-{`Q}yAZ+^aGlXnN7B`WiAA5Mz=`j-Km7d1Q1ju@&e!CVK}jCe?RyESPU8mnf9B@?~{>WeEh^4?j>RuL{{jGQE}0rqV@K{qTcn4)rP;TGSe!0Ttvz9R$TA-wpkGzox(tC?1v>g0V*GI@Y2M9S0k!t}NlJx*iJZUW7y@axjkYVwUVAYo^L|OvRg%pr`9~v<v<TBld){<q6gqKytO}*CM!_%)@Z3!sFk`c0)`ynE>BO6t@BHPNsy!@3qSX!BT==M!1_8?}t2XVB#FEg+NS7vN!GqW)x_FcGEY;%+0O|l3aT*9=7_ukQCvcP;&)NYebkHi*Hp!JM{*J@r2U10ecTf7{|dVIhTHOxM`3#8?PJe5>Dwu%r@vCXn&zR6)0O@<HFEYROda_4?i?#@8Vyf=wU3pT6r5gjD{B~Z<F}ATzNYWc!ID0`CCy2;K4q0-MBT5!!EPchhkDif%2XsBG7S$h=etYdJ%M5RC5tk@R-?0V&E7P7*>J!-!k4BC%2%;scao|ky+vtkcK`XLzZx8E1gJ9r*`}UXiz2?TVpXdm>Cv+2O}FkAdW1Ilg0?|02A{uz=O4J>=}V?0jFelg$=S8qH2rSl=0?vBaCR&+C99g-O~nUi>OP`Mz9DBG7p?EYefyiU0|e?2GzE5C<FvMLH34<WzA*|Br;p566c_QXrIt)fo6K5n8|q_HYUcYiotS^*d{IDp`c^ZM<;%!9hX`r8rqGl+=o^tEk0|f9do8240463MK}hKxN8xQAfX_6co#lp^EzyZ7wvd-zJ4isJ!?UqTa+quP$D_K$O@LgiLV6mX<B9qYs*KT^D#A<&Spz$zKCR+$tCl}Vmet4V#}7GgRz7{;wj#2K*NhC@ge?&+nZ%M3PDvbFwFeMa)dinC(Ln2YLm)S0~gbUVm`H-#DF)0ajz=+8zBn>wm^w_q~LuD=r>f(Z>MwiO>TOU9>>_-ZuSjX4yII_B%ZS;Ubd1IQ%Uri+oq&POzHNtdaUuGje0DqjcslvDh^21b*wopDXPPi;;<N7xd$jMT}7rvV2bhin3Vwn9#|98UCTsECb4akoSX@Hm)99>t~Jy+9DwW0DK1DxvM4W8ILNKb)>9!-%#;t5slwBwKA+D0Y-xXDsiFy_S7<e-xX3CVu|AH<Vj`Ew9+Wf5L{`NaS<mOUgZMCq3qrRCEWYtWEvJ)PIRa{@IwChR%LUX2*fLd5pv$g=oY(Z}8rz!#nCp8GG^d6TiwOCijyHqp#+hJtHDKK^<J?9G+_rWE!c82FyCBpEbW9)%cu$T@tuDu5+&=I_r(c6T1l8<_GZz52g;1QI3#WlH8dDLL3M|EF*MM|)(inZEB#XcZ&_3KoVIM^;;9H%(!G0ro0EwGqo=&W1-NtKRB%Yh8YAU@a#^En@0)9~i+=3d^(WwUjC$)$7i9k-xmC`%w*|htdCAu`JaAK-eI<J6b(9s=LnHHy}cGFpR<%g5XR-7=?Q7@A#qgO{3JGRVKc9kPnGsa?k1BeRk^mx<B4SB^XZjJ^6@}}I$r_xz9mETFw4O!q7iOn<sURBwN8s14NI6DoGE|{?>z2gq+YBJ*~3~yvH7GuBBWZ%Z1Y2HFWhU=LutZ6swdG?jCpw;*hgS#eSBSBZX03opj_ExK-`TYcMhd)nvt9Z^YVvFpw61eU_H@YMHEW@rvF(%L*W(QhWQVS_3M&JqFK2$Twv~!Y@X?G*bY9fmTvVh&#is~@XtCSg-`dQdr&jOdAD;0^oL?+9z;_LD%tYct(6t4(x4dNky7~fh_fyJaT5|O$I8jCE>N7bQb9;u-8z;;khl(zYb_Yf@X=+2mSu5d?oj9NSnnxD6}Cd?=aRq((7mOvs0Jk#6Fd24a8(_o5q2V4SHmzUn=7uCiHqxZmQ7!fCd5f^;}pLZ-QPX!zG7*;x8ME7R`;jaJC>VDGzX_cL8!vL6EvsuJmC3aJpGIVJf&q-kcJkQeH0ISZ@L!p@7$YwUT%D&F6s2KDQHa~m+Hgp;^bZ<SilZVY8h&4yYtu%H|=T-qlhN1J|jUbeooQM&{IqRyQEG|T8V^LeM>lyu~-qTr`%*(XuEa(Kj5K~*@L7#4d!<20u6!PV*bSYEf@mz}tcb+d}JRPSy=Q!+5<_VoH7V||n)hsNiC_AJ1$BO3`_c~9pbmT|!?Ci-UNufvb35rd8V*&}LYlftX<OaNm!p1<Q9YSSJ^4?XP%>%82bd7mG@gf4HthP1}`p8*JVyBSBSoO{(96EEc#6>TAC=Cr*{93WZ6Ua}?8DAn#IS@#aYO=J4;=A2ECA4&DsW@`ChlOjSv+&^0J<iL!6HYtycKYse%9y6YqXAASmh6<7=+5AErE|I|P}X*buYy;YLsvg905?u7?TfO<d(-PUQ8iK1C@@N#FePFjHr&eTRt-C$rVjdjEm_fFf2eg>xd2jOmoPPQw+f^$c4B^AD>b^uWtXtS!^@&aG^PPNwY|yN8xA_Nm$uJ*M0r78w6Zn2(T5TF9Va>%eB5z_Gfh@QYcA}S?0`r(?HflZ6UQ<B8_3%o{RBNT%~7`tTvHf2f<SaI%_kGe!YDuiU#0V7v!&}U1)a>fm@0KJKC?q1z2FUzwe&@GO6dss8Jbq5jed_AXYmK-k;jm!8*S(*y@VffqSJ2nD83uMq9>mnm)y3Qk=K#@6E-p(>7A*IHo=v4cXnV~h@5rIGlx!fsz9$pOr^pouHM!D|APfgZj1_c;xN7ttE`=x%4wuVPTwj`Qiw_fg$F^ednWRk827E)CVPwtQ>__1*4~=%td!4%P2FgN3~;Us=oG5g1zWNNVIbnat!-)zy8aqyaXPW{qdBgz29|Un<U|gsc()dWzNT)-l6|uu+5Cxg%j)D~4a^GbpKwq9+f-)G|5lSH_+541^qqLt+@dRpJsFL{GKJoiO?@~SkW*_v2~wn;P7k+e42Y=Q?xJH%cOa|vVZ=JtupcIEpTLm%?06(7;OF!#xzH-KLy2-ur0eI^XUraGrg`%?aI`E142wE!V#elBNK{0ba#V%kzo+74D0tTJ!Q9EzyVK~GdLqG^nO}amTnVs!1L~j*-e;64;x3=!lYnqWHX|eBrN#S@Yvh4xBz4ZZ@Q1s0t8N6FtyvBc*M<^+d2CK0u%S~^rWK1+fdRH8X$`+d8o)FZ;-s|0D`H|W?3*IHEFZIdw>RjI2qW#(r`?uk@Y>iUGU1SD_VGfw+-N2lkl6CV{9;PvEo&B!Rv3c@Hi}oWK^UIa0ARz~^AUol#Hnq|hH?Sk{yeHR!<i$jwNqFj6p)oMJ^^>gxfMYnF%&>KZ$+Mx<8uhN4(Wys4^KJOQ|GRSZzh%K%KKQMw**ioBcZ*w;F+TqxYhXA9=GhEh`_)`h~HD{cS)g02!{<CRKveB@e2o%XPq_Ndk#q8BB+{a742dmHl@>LmWf*om;l`DTn(`d?g1-MVf{VC8i!1}tI+OMp^2klyP<BIvV$uzinqglv0ew_$8(g@Lj$P6EXwXvmC1PA2P&#K=j@MjrrQ*;J<hSr-Q9x)t@(&h7_r+dbSTa)Uj%u1Xte2`0RQ#4Z7^RBSH9??-tDua_&#WVeUQ;6x1jMrWjNGoE!dXH6;hbyys;*9LA@2K226DeAmEb;b%y;B)^^$3cJhEFvOJWw6ldU^JUT-siY&$3q(j`STuc}89uLBaK~QyF@TbcTj*sMj%<nf8pn*&x>vnJ)>a|v{jj}m+Enoo?<7ixEx`Dq$bxeM&e*5bUkUgS1)R){jhhY#vQYvjlpod&eJ<1<2YEX$9Y)9xInGi`7kQix#%w``G*d}5=7otpbKRqVM#q6wm_9kB7Vk%R@dtbgdTYSy1&_nrDnzFXwAqq^f6Bsu_71PgmGqB3pISW+?J})6(a&|)EIru;_;<l`8+8?_v(?yG$!{NkWp++fa1_n9>og{-J2Kf_s!Wyn1i>_fkpvMQMU9h5765V^mRJ0K&+Mo@nPGi{W2hH~I0W|XfKbQ}&hA5GVJS1TflQ%l%%Q!@(&U-pWqLtzlts=etX$W87!#S{>$Y@xj>?{p@6B1_*X1X)=^32xE&P?g07XfS}iij&X3>qRxd**|a#prw(^T<GBYk&d%nO^fU)@M)m^d@b?JdnU922U0%JWY)FK$PZ0=Gzrm+pnY|O`?D363i(vt~hhV;)#Ko8`fv!6!&{~t}KtInRnPU8-;wyIUOLX;2Z!8f(MF9(lGfHiE|83q-cwAe0eFe!%4lagyy3$*?R+u>qUJ-9Sta>A+*8{KRA*A72=%hc!{wV_lXm0_fLQgXtwVvu;UlmI=lJPfd-rkdd|dxVNaXcdFRZM=iN78zJ7^*gt9;hzP`?#CqHM&Ov$;A=K%AOa=$)SLZEGjqYDDvQW5^^Yu|G_mEmCw`5xGOSg`nHL=Xe-nU*G}OpLq5=|bIY_Z4X_5NgEj!e&x%Ov?gj_emCjdjYZ1<$=2Mz^m5L+Ff&MpSKdQr(j1NyNd7ZlSBZ;wG{Zv$x?vT&-Xs~1=}9lXoh&Eg(|$+661Vf_-s|Il4~~TDk4)3MVu_W0aPE$V%^XhLorT5sSnP;(0QlCIBr{Bbo<vK=cHanU&8C@AVwSQCi#wN9jYH5GBX2UXwAAAVMhgK)~L=H%_iKpkNH;!n$T>cA&c(3c(e!57D-x1tb^cE6(y9wOlm)HH#G<S9?V23UBZL%2<8<CiIhMw{@?-hrH>c2VL7%&RZ!8o4@edQ?%|Mw<|cFZz{&tdkB1^=<UqztvYr`un;LIJcTMGDGHbSjL%_)~pTs#4rAzUW6bXe5F9<ktON;<_Atqs4g~<z-#{+ByL(Os-=K(^J#RTOs@_ah!)%oD(h~r$!xUuy|LnX8jm18SWUwld~6Ih6iuyw2A#3>3AiBsmm*ealF>=v^5M2f-~l5-D3p&x*^EyAi|mkgpk-W?`ulgLF8JVl8m?_?7Uk{wea?9Q=%q4>i3=02x)0H!=74&WejigGw6fo2vUjk>|FAC%K{{JxVY14q6k-(GO-$iep&ocA2iBm7r%_24&l4n}9*c^wGIDInRT%xIQVG(B*7mafAzA_QUn3HZQ!i0g^uwn~Rt)9}K$Dlj0_mB<b<X^wppDOjjuR-CUz(H|Pf1=4AE;<o*GVB`QkV0UF$%|p8U6@rx{9nM!Go&G5Xc2yLYlHnRSY#EcmPa8KAK|#)K4nbyQyNtR~%<pjTCov$);G{fXV#Rbaohg7&o=Y(-=X2QuYJWF35!AixyQtUt2Ht8U^^^(8WTe9z%S;({=&W31Azloe0v4L78)%Ld=9i*7>uAdDlO)ewR6Q)58FH@EZy3@nZ+A??Li?b-AL7Qz_RK&lrpPHA{kp}t&8l0&Ziik~WRYlxG7CHCehM^JmS8zZu8PJD-B2x`cq-siF0J{Q0@}OZ@@u=^*OKMYrILg^`F$Cnz~H7yCnJolPwwmz-h;f&>}u1T&Zy>}<qLKj`S+Us%61onMPO#pQ8|RK0GAV#4gQuadeOzAzvfJx>%DcxZc{n)<bKYxwm92i#s!^gvISxf@(oglB}H-xk7mRzutnwTS-IUT=b{9N%xr9WMceI?u9c{zaweNqQ^mxF&u~acb!3Yb`KpHHyT-pcS7k-9+jcI@K$u7>d^UMqTYPhfZC-&57M}k9wstK&aU4PT&aW7e!|ZNGW6Z;mmT*{L?I3|6uzAQa8a8IZ$P2q!7KaccB}7qBY{|L!=tNNtxkipHtzw(s0RJfcRaN(N_sp;)vV@G8nXaCGOm%hjS6`W}C2a>q#??l6%Go?+Vwy0fSa)eeFHWsoJXruFH6zcjlywDr{<=b^lefY*dk-8>lNzXHKyPgJI9Ifr*Vhb{Y_Et<$Xv|RU9L2qKKA%VVZ-bJ?Qmt{3EG24Xw#uJ@eh+=)3**U@14t55XayISr`B>YRN!;fYxC-PH5kiB=3h+$?2=Eeo>v68e1R2!x*NhAFGYx-hv9*7mqBFc_P{@00Lzi;7MzXiE+V)2Kx9N*YP-PNWHfC1ItZaFJIODbKRQfXKE@B)J#{iJjypcUm@HTNvU2l-47>Be@(<XPY|R>O)i2jOvJ$)tpsXto0Rbl$L4woHKAs-$rC)P14JcXL!ZOp`@Mty{jCFcehC9>S|^({1l8(o(y}O>YCA%09Bo0-qNUNBf$t*7of-_%6zu#QjBFwF+W^wK9oK=c@32q7Hy?`p`$KFI;CmDOmJ9)t9wa?L-edy10CiVus5$%Xu?qq76H5|ZI1=^b<d;+xI_O2fA_b^J8P#0LDHNGbpAsBC@i=PjDMd}L0PtZjkddLsR*Fc$2zxa&k^0+39uq2QcCkj8It9VVt|t$%WDv!!Ou{P(CO56#!8tJwNu!#EprNwDxtQt={jr5rFew)b`0uDY)iddb2xy_Jn&}vTZ7<xkj}~1HlMbEqIG#bmGl2#J2ZqxHz;lY^FrahEfoLDg=?S2nbFr;-^>o%=T~ZZ#5*xg}ry=}8MM}c{qTh*%z?ZnoRz9|sOZDXZ`S~yBf1LkGN%o#4dHhrH?q=wVu-`a*FYw*F6f#dy;S|43I{zhts@+)ZEPuq^yL;-#-e-j8zjG)&F}@8hP6D3<fg|{Z1mWU`d=i8K6GFh478(R5wx9**#9yi$QiNb`?b9frIP;$`o~no5YUTl+F(vk+sEdCSRmS6T6u2#TfyOCQ&@@wrw+p2iTZ99m7<fak_i^o-ORm7JpzQ#)<R9WUcpT?-*l!;DgHM<cdF%z?eJEsI7-Wb_v}8k3pq{BzIMn|Iv*f96I;m8!tW%%<MXkUg&?{=6?T4E7`sxht$kW7#(t-RrmIVU(H3^4*j*fwONj!4j6RE-FEBHn^&h+Fn;szWz{4l|WV@Pd81!n%+bihtQUsx?NjO5tN@E_-!MiJ)lkAY79sYdN7Uh5~C*?6@l8IK-`e8Y?m!ab2`;d6%ug{(`b3i@1xEQ2cn8v^=)XdO=FP)}7!-y>~SRkNs~%QO*1=iOOY$p-IbcuztbiLq@6^}7T*>5{vah1|Oj|6Ev(u_Bhj3NsfHHZ2WuvfhD7m2%{P(5NRuG6Fd9ZoGf6wY77oop|gtW-_=?d<4v#sDACbdls@%lcTNp&QA9xMC!hX!mmNJ$?^LLyr+j2R+9Pp%|(QSnVap_!&ZCdM$~Mp`jwa)lMIK!_UFjk{TxYXBcY9WIfY*Uf701z4w8pD!G#@HwV1Avp&F~?bT1Wx0kezp5DhKHBjXgLD^!qUAs!-$Bcj@nQ89s!@vu}bl2>wU^2lF*BVN?JQ`53lf97??6QYOC^Iw1Hnm~H%gtSgev*CK>`q=m+{gd4^hE~60?CYP;)$8+X9L~D9u(Z(7JI{i;uQb(iXG*X2)d6f&I$`vuNmO9fM8h^l(zzt$xZp^L>s8;_{=t_>lyO8)6Kw+QcwIC4L%=6W{VMIdRul~{l8<7TdrF6t${KX+N}F4qBRyVFo-lNWDbtz#UXs_)Iup>7n^nYN46-@C!<uMaP<i+$3V65Om;0!;+E}AVYS_(<x@*b-a5!^GTrVP4G@vQ0n%TmGDZr?LIh32Q#f3zfnWXgr?<%}t1^vZKfJ8yVEdUmOsM&46Rd<;--JG?w%-0te5L$&u@a5G;4(*(6cXQwVZ^w1&Gtt%R?A^a=GO}X!G27#-*6^t4BriiB=edv1uu^Ai7t2#KAbWvnzKD0(0CjT0hd~p5OX;(<WVW^%@6B-PLqh>z?o#}6&TMtdXAqRaaA$8L9tECll6F!xjNI9r!6!TlsAGuz?&7;hxW!bt9zqz4H0QPPN@aX38ZYMw6x%X?G^$y5n)Z{HorDU*{Ga8jbcMLqvT_}7xGUU%2Wy`+menvU)@GCW<<D-zS{0>~1M#b7d24c#1_A86-RW$%36I`1j7{vh1kjoOmo3THW9>dD*(+C>auwsGQG&;Fe0QS2atU!3=pE{CTvi>b#M_;I&<t+T`-(GF_&uhLvryw-oBeb;IZTBk0T{O&QZ99Q^g*k)*J|%~TleHjFuJ+3-)%R)A>S})y;g|jCzudguXb~9`=Gf^x0nD(Zs_Ov;>4(?6*DCgFkqT$+{e?Wfde!T37aoa5eYFns5>7By%>U9og3FDY?d*-Sm4e$-{25_rRKbHLPbW=>xtHP2@`p!`+cB>Mg;JV)d{c79L1tCM!AR4fKkoJr5>8J|3(I{PETE(8ae(K9yuzr'

def _save_complete_notebook():
    doc=_nb.reads(_zl.decompress(_b64.b85decode(UNIFIED_SNAPSHOT_B85.encode())).decode('utf-8'),as_version=4)
    needle='UNIFIED_SNAPSHOT_B85 = "__SNAPSHOT_PLACEHOLDER__"'
    ensure(needle in doc.cells[-1].source,'UNIFIED_EXPORT_SNAPSHOT_MARKER_MISSING')
    doc.cells[-1].source=doc.cells[-1].source.replace(needle,'UNIFIED_SNAPSHOT_B85 = '+repr(UNIFIED_SNAPSHOT_B85),1)
    try:history=list(get_ipython().history_manager.input_hist_raw)
    except Exception:history=[]
    executed={}
    for src in history:
        if not isinstance(src,str):continue
        m=_re.match(r'^\s*#\s*(0[1-9]|1[0-3]|C0[1-7]|21)\s*·',src)
        if m:executed[m.group(1)]=src
    parity=[];keys=[]
    for c in doc.cells:
        if c.cell_type!='code':continue
        m=_re.match(r'^\s*#\s*(0[1-9]|1[0-3]|C0[1-7]|21)\s*·',c.source)
        ensure(m is not None,'UNIFIED_SOURCE_HEADER_MISSING')
        key=m.group(1);keys.append(key)
        if key in executed and key!='21':c.source=executed[key]
        # For final cell, self-contained snapshot is authoritative and cleaned of cell outputs.
        compile(c.source,'ML03_UNIFIED_'+key,'exec')
        parity.append({'cell':key,'runtime_source_present':key in executed,
            'sha256':_hash.sha256(c.source.encode('utf-8')).hexdigest()})
        c.outputs=[];c.execution_count=None
    expect=[f'{i:02d}' for i in range(1,14)]+[f'C{i:02d}' for i in range(1,8)]+['21']
    ensure(keys==expect,'UNIFIED_SOURCE_21_CELLS_ORDER_INCORRECT')
    doc.metadata.pop('widgets',None)
    _nb.validate(doc)
    dest=WORK/'ML03_V4_02_FULL_FAST_GITHUB_REPRODUCER.ipynb'
    _nb.write(doc,dest)
    ensure(not (TOKEN and len(TOKEN)>12 and TOKEN in dest.read_text('utf-8')),'GITHUB_TOKEN_LEAK_IN_EXPORT')
    qa=WORK/'unified_source_parity.json'
    qa.write_text(_json.dumps({'version':CFG['VERSION'],'cells':21,'sources':parity,
               'notebook_sha256':digest(dest)},ensure_ascii=False,indent=2),encoding='utf-8')
    return dest,qa

UNIFIED_IPYNB,UNIFIED_PARITY=_save_complete_notebook()
ensure(XLSX.is_file() and XLSX.stat().st_size>0,'ANALYTICAL_WORKBOOK_MISSING')
ensure(ZIP.is_file() and ZIP.stat().st_size>0,'TECHNICAL_EVIDENCE_NOT_CREATED')
ensure(MANIFEST.is_file(),'RUN_MANIFEST_MISSING')
ensure((OUT/'CAUSAL_STATUS.json').is_file(),'CAUSAL_STATUS_MISSING')

# 1. One curated, small scientific package: no raw panel, full test predictions or full local SHAP.
RESEARCH_DATA={
 '01_test_metrics':METRICS,'02_confirmatory_exploratory_comparisons':ABLA,
 '03_model_selection_DEV':DEV_METRICS,'04_ensemble_test':ADVANCED_ENSEMBLES,
 '05_ensemble_vs_best_single':ADVANCED_SINGLE_DUEL,
 '06_ensemble_intervals':ADVANCED_INTERVALS,'07_factor_dictionary':FEATURE_DICT,
 '08_factor_support':MODEL_DOMAIN_COVERAGE,'09_global_SHAP':ADVANCED_SHAP,
 '10_SHAP_region_CI':ADVANCED_SHAP_CI,'11_SHAP_stability':ADVANCED_SHAP_STABILITY,
 '12_SHAP_interactions':ADVANCED_SHAP_INTERACTIONS,'13_SHAP_additivity_QA':ADVANCED_SHAP_QA,
 '14_grouped_permutations':ADVANCED_GROUP_PERM,'15_PDP':ADVANCED_PDP,'16_ICE':ADVANCED_ICE,
 '17_ALE':ALE,'18_publication_dates_QA':PUBLICATION_AVAILABILITY_QA,
 '19_five_year_calendar':H5_CALENDAR_GAPS,'20_task_plan':TASK_STATUSES,
 '21_municipality_boundaries':BOUNDARY_QA,'22_source_and_units':SEMANTIC_QA,
 '23_model_group_coverage':DOMAIN_COVERAGE,'24_future_canary':CANARY_QA,
 '25_ensemble_future_canary':ADVANCED_TEMPORAL_CANARY,
 '26_source_grain_audit':SOURCE_GRAIN_AUDIT,'27_real_SQL_QA':REAL_SQL_PREFLIGHT_QA,
 '28_extended_source_recovery':BDMO_EXTERNAL_RECOVERY,
 '29_correlation_sensitivity':SENSITIVITY,'30_size_heterogeneity':SIZE_SENSITIVITY,
 '31_hypotheses':HYPOTHESIS_REGISTRY,'32_regional_errors':REGIONAL_ERRORS,
 '33_research_status':pd.DataFrame([{'gate':k,**v} for k,v in STATUS.items()]),
 '34_shap_local_sample':ADVANCED_SHAP_LOCAL.head(int(CFG['RESEARCH_LOCAL_SHAP_SAMPLE'])),
}
RS=WORK/'research_fast';RS.mkdir(exist_ok=True)
SCIENTIFIC_CSV=[];RESEARCH_INDEX=[]
for name,frame in RESEARCH_DATA.items():
    if frame is None or not isinstance(frame,pd.DataFrame):continue
    path=RS/(name+'.csv')
    frame.to_csv(path,index=False,encoding='utf-8-sig')
    SCIENTIFIC_CSV.append(path)
    RESEARCH_INDEX.append({'name':name,'rows_exported':len(frame),
      'rows_total':len(ADVANCED_SHAP_LOCAL) if name=='34_shap_local_sample' else len(frame),
      'cols':len(frame.columns),'sha256':digest(path)})
progress('ARCHIVE_RESEARCH',3,4,f'{len(SCIENTIFIC_CSV)} научных таблиц')

# Compact Excel workbook: all relevant summary tables, not large local-explanation sheets.
from openpyxl import Workbook as _Wb
from openpyxl.styles import Font as _Font,PatternFill as _Fill,Alignment as _Align
import numpy as _np
lite=RS/'ML03_RESEARCH_SUMMARY.xlsx'
w=_Wb(write_only=True)
for name,frame in RESEARCH_DATA.items():
    if frame is None or not isinstance(frame,pd.DataFrame):continue
    label=name[:31];ws=w.create_sheet(label)
    # Excel supports primitive scalar values; JSON-serialize nested dictionaries.
    def safe(v):
        if v is None:return None
        if isinstance(v,(_np.integer,)):return int(v)
        if isinstance(v,(_np.floating,)):v=float(v)
        if isinstance(v,float):return v if _np.isfinite(v) else None
        if isinstance(v,(dict,list,tuple,set)):return _json.dumps(v,ensure_ascii=False,default=str)
        if isinstance(v,(pd.Timestamp,_dt.datetime,_dt.date)):return v.isoformat()
        try:
            if pd.isna(v):return None
        except Exception:pass
        if isinstance(v,str) and v.startswith(('=','+','-','@')):return "'"+v
        return str(v)[:32767] if not isinstance(v,(str,int,bool)) else v
    from openpyxl.cell import WriteOnlyCell as _Cell
    hdr=[]
    for col in frame.columns:
        cell=_Cell(ws,value=str(col));cell.font=_Font(name='Aptos',size=10,color='FFFFFF',bold=True)
        cell.fill=_Fill('solid',fgColor='17324D');cell.alignment=_Align(wrap_text=True)
        hdr.append(cell)
    ws.append(hdr)
    for row in frame.itertuples(index=False,name=None):ws.append([safe(v) for v in row])
w.save(lite)
from openpyxl import load_workbook as _load
verify=_load(lite,read_only=True,data_only=True)
try:
    ensure(len(verify.sheetnames)==len([f for f in RESEARCH_DATA.values() if isinstance(f,pd.DataFrame)]),'LITE_XLSX_SHEET_COUNT')
    for item in RESEARCH_INDEX:
        r=verify[item['name'][:31]].iter_rows(min_row=1,max_row=1,values_only=True)
        ensure(next(r,None) is not None,'LITE_XLSX_HEADER_MISSING_'+item['name'])
finally:verify.close()

# 2. Publish heavyweight `ML3_EVIDENCE.zip` as private GitHub Release asset.
# Remote repository reference is attached under stage2/ml03_factors/runs; main branch remains untouched.
REMOTE_PUBLISH={'status':'NOT_STARTED','repository':CFG['REPO'],
 'scientific_status':'PASS' if scientific_ready else 'PARTIAL',
 'technical_archive_name':ZIP.name,'technical_archive_bytes':ZIP.stat().st_size,
 'technical_archive_sha256':digest(ZIP),'verification':'not_started'}

def _publish_heavy_evidence():
    if not CFG.get('GITHUB_TECHNICAL_PUBLISH',True):
        return {**REMOTE_PUBLISH,'status':'DISABLED'}
    ensure(CFG['MODE']=='FULL','REMOTE_PUBLICATION_REQUIRES_FULL_MODE')
    ensure(bool(TOKEN),'GITHUB_TOKEN_NOT_AVAILABLE')
    # Immutable source ref controls Stage I retrieval; Git branch is based on current main.
    head=api('GET','/git/ref/heads/main').json()['object']['sha']
    stamp=_dt.datetime.now(_dt.timezone.utc).strftime('%Y%m%dT%H%M%SZ')
    branch='ml03-results/v402-'+stamp;tag='ml03-v402-results-'+stamp
    segment=CFG['SEGMENT'].strip('/')+'/runs/'+stamp
    progress('ARCHIVE_TECH',1,5,'GitHub: создание изолированной ветки')
    api('POST','/git/refs',json={'ref':'refs/heads/'+branch,'sha':head},expected=(201,))
    create_or_update_git_file(segment+'/'+UNIFIED_IPYNB.name,UNIFIED_IPYNB,branch)
    create_or_update_git_file(segment+'/run_manifest.json',MANIFEST,branch)
    create_or_update_git_file(segment+'/source_parity.json',UNIFIED_PARITY,branch)
    progress('ARCHIVE_TECH',2,5,'GitHub: ноутбук, методика и реестр')
    body=('ML-03 V4.02 — reproducible computed technical evidence. '
          'Scientific status: '+('PASS' if scientific_ready else 'PARTIAL')+
          '. Parquet/models are research evidence, NOT approved scientific claims. '
          'SHA256: '+digest(ZIP))
    release=api('POST','/releases',json={'tag_name':tag,'target_commitish':branch,
      'name':'ML-03 V4.02 · '+stamp,'body':body,'draft':True,
      'prerelease':not scientific_ready},expected=(201,)).json()
    # Stream original already-tested ZIP; do not duplicate locally or base64 encode a giant file.
    with open(ZIP,'rb') as stream:
        response=SESSION.post('https://uploads.github.com/repos/'+CFG['REPO']+'/releases/'+str(release['id'])+'/assets',
         params={'name':'ML03_V4_02_TECHNICAL_EVIDENCE_'+stamp+'.zip'},data=stream,
         headers={'Content-Type':'application/zip','Accept':'application/vnd.github+json'},timeout=(30,1800))
    ensure(response.status_code==201,'GITHUB_TECH_ASSET_UPLOAD_FAILED_'+str(response.status_code))
    asset=response.json()
    ensure(asset.get('state') in ('uploaded',None),'GITHUB_ASSET_NOT_UPLOADED')
    ensure(int(asset.get('size',-1))==ZIP.stat().st_size,'GITHUB_ASSET_BYTES_MISMATCH')
    progress('ARCHIVE_TECH',3,5,'GitHub: технический архив загружен')
    meta=api('GET','/releases/assets/'+str(asset['id'])).json()
    ensure(int(meta.get('size',-1))==ZIP.stat().st_size,'GITHUB_ASSET_READBACK_SIZE_MISMATCH')
    digest_remote=str(meta.get('digest') or asset.get('digest') or '')
    if digest_remote.startswith('sha256:'):
        ensure(digest_remote.partition(':')[2].lower()==digest(ZIP),'GITHUB_ASSET_SHA_READBACK_MISMATCH')
        strength='SHA256_FROM_GITHUB'
    else:strength='UPLOAD_RECEIPT_AND_SIZE_ONLY_SHA_UNAVAILABLE'
    # Record remote object and SHA under repository segment without inserting 250MB into Git history.
    index={**REMOTE_PUBLISH,'status':'ASSET_VERIFIED','branch':branch,'tag':tag,
           'release_url':release['html_url'],'asset_id':asset['id'],
           'asset_url':asset.get('browser_download_url'),'verification':strength,
           'scientific_ready':bool(scientific_ready),'source_sha':source_sha}
    stampfile=RS/'github_evidence_index.json'
    stampfile.write_text(_json.dumps(index,ensure_ascii=False,indent=2),encoding='utf-8')
    create_or_update_git_file(segment+'/github_evidence_index.json',stampfile,branch)
    verify_github_file(segment+'/github_evidence_index.json',branch,stampfile)
    progress('ARCHIVE_TECH',4,5,'GitHub: реестр и удалённая проверка')
    published=api('PATCH','/releases/'+str(release['id']),json={'draft':False},expected=(200,)).json()
    ensure(published.get('draft') is False,'GITHUB_RELEASE_STILL_DRAFT')
    index['status']='PUBLISHED';index['release_url']=published['html_url']
    index['git_segment_url']='https://github.com/'+CFG['REPO']+'/tree/'+branch+'/'+segment
    progress('ARCHIVE_TECH',5,5,'GitHub: публикация завершена')
    return index

try:
    REMOTE_PUBLISH=_publish_heavy_evidence()
    mark('GITHUB_TECH_STORAGE','PASS' if REMOTE_PUBLISH['status']=='PUBLISHED' else 'NOT_PUBLISHED',
         REMOTE_PUBLISH.get('release_url','Technical publication not enabled'))
except Exception as ex:
    REMOTE_PUBLISH={**REMOTE_PUBLISH,'status':'UPLOAD_FAILED','error':type(ex).__name__+': '+str(ex)[:350],
                    'recovery':'Technical source files remain in Colab WORK; no scientific PASS is implied'}
    mark('GITHUB_TECH_STORAGE','UPLOAD_FAILED',REMOTE_PUBLISH['error'])
    progress('ARCHIVE_TECH',1,1,'GitHub: ошибка, технические файлы остались в Colab')

# 3. Create exactly one download archive with curated scientific products and truthful publication status.
remote_record=RS/'GITHUB_PUBLICATION_STATUS.json'
remote_record.write_text(_json.dumps(REMOTE_PUBLISH,ensure_ascii=False,indent=2,default=str),encoding='utf-8')
science_data={
 'version':CFG['VERSION'],'source_repo':CFG['REPO'],'source_commit_sha':source_sha,
 'scientific_ready':bool(scientific_ready),'full_science_gate':STATUS.get('FULL_SCIENCE_GATE',{}),
 'municipalities':int(base.entity_id.nunique()),'municipality_years':len(PANEL),
 'prediction_rows_in_remote_evidence':len(PREDICTIONS),'model_metric_rows':len(METRICS),
 'paired_comparisons':len(ABLA),'ensemble_models':len(ADVANCED_ENSEMBLES),
 'shap_global_rows':len(ADVANCED_SHAP),'shap_local_total':len(ADVANCED_SHAP_LOCAL),
 'shap_local_in_download':min(len(ADVANCED_SHAP_LOCAL),int(CFG['RESEARCH_LOCAL_SHAP_SAMPLE'])),
 'publication':REMOTE_PUBLISH,'table_index':RESEARCH_INDEX,
 'notes':'Scientific summaries only; full test predictions and local SHAP in private GitHub evidence release.'}
science_manifest=RS/'SCIENCE_MANIFEST.json'
science_manifest.write_text(_json.dumps(science_data,ensure_ascii=False,indent=2,default=str),encoding='utf-8')
readme=RS/'README_НАУЧНЫЕ_РЕЗУЛЬТАТЫ.md'
readme.write_text('# ML-03 V4.02 — материалы для научного анализа\n\n'
 'Содержит компактную книгу Excel, полные итоговые статистические таблицы CSV, объяснения SHAP (агрегаты), '
 'визуализации и независимые QA/причинные протоколы.\n\n'
 'Полные наблюдения, прогнозы, локальные SHAP и сохранённые модели **не входят** в скачиваемый архив; '
 'они находятся в техническом Evidence GitHub Release (адрес в `GITHUB_PUBLICATION_STATUS.json`).\n'
 'Если статус `UPLOAD_FAILED`, публикация не прошла: технические данные остались в Colab, их нельзя считать загруженными в GitHub.\n'
 'Научный PASS нельзя выводить из успешной упаковки.\n',encoding='utf-8')
FAST_ZIP=WORK/'ML03_V4_02_RESEARCH_FAST.zip'
figs=sorted(p for p in (WORK/'charts').glob('*.png') if p.is_file())
causal=sorted(p for p in OUT.rglob('*') if p.is_file() and p.suffix.lower() in ('.csv','.json','.png','.md'))
with _zf.ZipFile(FAST_ZIP,'w',compression=_zf.ZIP_DEFLATED,compresslevel=6,allowZip64=True) as z:
    z.write(lite,'tables/'+lite.name)
    for p in SCIENTIFIC_CSV:z.write(p,'tables/csv/'+p.name)
    for p in figs:z.write(p,'charts/'+p.name)
    for p in causal:z.write(p,'causal/'+p.relative_to(OUT).as_posix())
    for p in (remote_record,science_manifest,readme,UNIFIED_PARITY):z.write(p,p.name)
ensure(FAST_ZIP.stat().st_size<CFG['RESEARCH_ARCHIVE_MAX_MB']*1024*1024,
       'SCIENTIFIC_ZIP_EXCEEDS_FAST_LIMIT: '+str(FAST_ZIP.stat().st_size))
with _zf.ZipFile(FAST_ZIP) as z:
    required={'tables/'+lite.name,'SCIENCE_MANIFEST.json','GITHUB_PUBLICATION_STATUS.json',readme.name}
    ensure(required.issubset(z.namelist()),'SCIENCE_ZIP_REQUIRED_FILES_MISSING')
    ensure(z.testzip() is None,'SCIENCE_ZIP_CRC_FAIL')
    ensure(len([x for x in z.namelist() if x.startswith('charts/') and x.endswith('.png')])==len(figs),
           'SCIENCE_ZIP_CHARTS_INCOMPLETE')
report={'version':CFG['VERSION'],'fast_archive':FAST_ZIP.name,'bytes':FAST_ZIP.stat().st_size,
        'sha256':digest(FAST_ZIP),'tables':len(SCIENTIFIC_CSV),'charts':len(figs),
        'technical_remote':REMOTE_PUBLISH,'scientific_ready':bool(scientific_ready)}
(WORK/'ML03_DELIVERY_REPORT.json').write_text(_json.dumps(report,ensure_ascii=False,indent=2),encoding='utf-8')
progress('DELIVERY',1,1,'Один научный архив проверен, технический сохранён удалённо или статус ошибки зафиксирован')
ensure(_PROGRESS_BAR.n==1000,'UNIFIED_PROGRESS_NOT_COMPLETE')
_PROGRESS_BAR.close()
print('ML-03 V4.02 | Научный архив:',FAST_ZIP.name,
      '| размер, МиБ:',round(FAST_ZIP.stat().st_size/2**20,2),
      '| GitHub технический:',REMOTE_PUBLISH['status'])
if REMOTE_PUBLISH.get('release_url'):print('GitHub:',REMOTE_PUBLISH['release_url'])
if CFG.get('AUTO_COLAB_DOWNLOAD') and CFG['MODE']=='FULL':
    try:
        from google.colab import files as _files
        _files.download(str(FAST_ZIP))  # ONLY small curated scientific archive
    except ImportError:quiet_print('Not a Google Colab runtime; archive remains at',FAST_ZIP)
```
